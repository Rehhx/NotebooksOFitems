```python
# divergence_rsi_5m_all_parquet_simple.py
# Runs a simple RSI-based divergence strategy on 5-minute candles across ALL parquet files in PARQUET_DIR.
# Reads parquet schema (EventAt, Open, High, Low, Close), converts to New York time,
# resamples to 5-minute bars, scans for divergences on RSI between price pivot highs/lows,
# enters next bar after confirmation, and simulates a 1:2 R:R with stop at the pivot and target at 2R.
# Prints per-symbol stats and saves trades.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\\Users\\pcagm\\OneDrive\\Desktop\\downloads\\parquet\\parquet"  # <— your folder
TZ_LOCAL = "America/New_York"

REG_START = "09:30"   # session start
REG_END   = "16:00"   # session end

# Strategy params
RSI_LEN = 14
PIV_LEFT = 3     # bars to the left to confirm a pivot
PIV_RIGHT = 3    # bars to the right to confirm a pivot
MIN_RSI_DIFF = 1.0  # minimal RSI difference to count as divergence
TICK_SIZE = 0.01    # used to offset stops slightly beyond pivots
RR = 2.0            # risk:reward target multiple

SAVE_DIR = Path(".")

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= INDICATORS =========
def rsi(close: pd.Series, length: int = 14) -> pd.Series:
    delta = close.diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    roll_up = up.ewm(alpha=1/length, adjust=False).mean()
    roll_down = down.ewm(alpha=1/length, adjust=False).mean()
    rs = roll_up / roll_down
    out = 100 - (100 / (1 + rs))
    return out

# ========= PIVOT DETECTION =========
def pivot_high(high: pd.Series, left: int, right: int) -> pd.Series:
    return (high == high.rolling(left+1, min_periods=left+1).max()) \
        & (high == high[::-1].rolling(right+1, min_periods=right+1).max()[::-1])

def pivot_low(low: pd.Series, left: int, right: int) -> pd.Series:
    return (low == low.rolling(left+1, min_periods=left+1).min()) \
        & (low == low[::-1].rolling(right+1, min_periods=right+1).min()[::-1])

# ========= SIMULATION HELPERS =========
def simulate_path(df_after_entry: pd.DataFrame, side: str, entry: float, stop: float, target: float):
    for idx, row in df_after_entry.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    last_idx = df_after_entry.index[-1]
    return "EOD", float(df_after_entry.loc[last_idx, "Close"]), last_idx


def pnl_one_share(side: str, entry: float, exit_px: float) -> float:
    return (exit_px - entry) if side == "long" else (entry - exit_px)

# ========= DIVERGENCE SCAN =========
def find_divergences(df5: pd.DataFrame) -> pd.DataFrame:
    ph = pivot_high(df5["High"], PIV_LEFT, PIV_RIGHT)
    pl = pivot_low(df5["Low"],  PIV_LEFT, PIV_RIGHT)

    ph_conf = ph.shift(PIV_RIGHT).fillna(False)
    pl_conf = pl.shift(PIV_RIGHT).fillna(False)

    piv_highs = df5.loc[ph_conf, ["High", "RSI"]]
    piv_lows  = df5.loc[pl_conf, ["Low",  "RSI"]]

    signals = []

    last_price, last_rsi = None, None
    for idx, row in piv_highs.iterrows():
        price, r = float(row["High"]), float(row["RSI"])
        if last_price is not None and last_rsi is not None:
            if price > last_price and r + MIN_RSI_DIFF <= last_rsi:
                pos = df5.index.get_loc(idx)
                if pos + 1 < len(df5.index):
                    entry_bar = df5.index[pos + 1]
                    signals.append({"Time": entry_bar, "Side": "short", "RefPivotTime": idx})
        last_price, last_rsi = price, r

    last_price, last_rsi = None, None
    for idx, row in piv_lows.iterrows():
        price, r = float(row["Low"]), float(row["RSI"])
        if last_price is not None and last_rsi is not None:
            if price < last_price and r - MIN_RSI_DIFF >= last_rsi:
                pos = df5.index.get_loc(idx)
                if pos + 1 < len(df5.index):
                    entry_bar = df5.index[pos + 1]
                    signals.append({"Time": entry_bar, "Side": "long", "RefPivotTime": idx})
        last_price, last_rsi = price, r

    return pd.DataFrame(signals)

# ========= CORE BACKTEST =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")

    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    df = df.set_index("EventAt")
    ohlc = {"Open": "first", "High": "max", "Low":  "min", "Close":"last"}
    df5 = df.resample("5T", label="right", closed="right").agg(ohlc).dropna()

    tstr = df5.index.tz_convert(TZ_LOCAL).strftime("%H:%M")
    session = (tstr >= REG_START) & (tstr <= REG_END)
    df5 = df5.loc[session].copy()
    if df5.empty:
        return pd.DataFrame([])

    df5["RSI"] = rsi(df5["Close"], RSI_LEN)

    sig = find_divergences(df5)
    if sig.empty:
        return pd.DataFrame([])

    records = []
    for _, s in sig.iterrows():
        entry_idx = s["Time"]
        if entry_idx not in df5.index:
            continue
        entry_px = float(df5.loc[entry_idx, "Open"])

        ref_time = s["RefPivotTime"]
        if s["Side"] == "long":
            stop_px = float(df5.loc[ref_time, "Low"]) - TICK_SIZE
            risk = entry_px - stop_px
            if risk <= 0:
                continue
            target_px = entry_px + RR * risk
        else:
            stop_px = float(df5.loc[ref_time, "High"]) + TICK_SIZE
            risk = stop_px - entry_px
            if risk <= 0:
                continue
            target_px = entry_px - RR * risk

        after_entry = df5.loc[df5.index > entry_idx]
        if after_entry.empty:
            continue

        outcome, exit_px, exit_idx = simulate_path(after_entry, s["Side"], entry_px, stop_px, target_px)
        pnl = pnl_one_share(s["Side"], entry_px, exit_px)

        records.append({
            "Date": entry_idx.date(),
            "Symbol": symbol,
            "Side": s["Side"],
            "EntryTime": entry_idx,
            "Entry": entry_px,
            "Stop": stop_px,
            "Target_1to2": target_px,
            "Outcome_1to2": outcome,
            "ExitPx_1to2": exit_px,
            "ExitTime_1to2": exit_idx,
            "PnL_1to2": pnl,
        })

    return pd.DataFrame(records)

# ========= BASIC METRICS =========
def calc_basic_metrics(trades_df):
    pnl = trades_df["PnL_1to2"].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= RUN ALL FILES =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    SAVE_DIR.mkdir(parents=True, exist_ok=True)

    all_trades = []
    for f in files:
        symbol = f.stem.upper()
        try:
            trades = backtest_one_symbol(f, symbol)
        except Exception as e:
            print(f"[SKIP] {symbol}: {e}")
            continue

        if trades.empty:
            print(f"=== RSI Divergence 5m — {symbol} ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            continue

        all_trades.append(trades)

        total_trades, total_pnl, winrate, profit_factor = calc_basic_metrics(trades)

        save_path = SAVE_DIR / f"div_rsi_trades_{symbol}_one_share_RR{int(RR)}.csv"
        trades.to_csv(save_path, index=False)

        print(f"=== RSI Divergence 5m — {symbol} ===")
        print(f"Trades: {total_trades} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {profit_factor:.3f}")
        print(f"Saved trades → {save_path.resolve()}")

    if not all_trades:
        print("\nNo trades generated for any symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    o_trades, o_pnl, o_winrate, o_pf = calc_basic_metrics(overall)

    save_all = SAVE_DIR / f"div_rsi_trades_ALL_one_share_RR{int(RR)}.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== RSI Divergence 5m — OVERALL ===")
    print(f"Trades: {o_trades} | Total PnL: ${o_pnl:,.2f} | Win%: {o_winrate:.2f}% | Profit Factor: {o_pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

if __name__ == "__main__":
    main()
```

```python
# divergence_rsi_5m_all_parquet_simple.py
# Runs a simple RSI-based divergence strategy on 5-minute candles across ALL parquet files in PARQUET_DIR.
# Reads parquet schema (EventAt, Open, High, Low, Close), converts to New York time,
# resamples to 5-minute bars, scans for divergences on RSI between price pivot highs/lows,
# enters next bar after confirmation, and simulates a 1:2 R:R with stop at the pivot and target at 2R.
# Prints per-symbol stats and saves trades.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\\Users\\pcagm\\OneDrive\\Desktop\\downloads\\parquet\\parquet"  # <— your folder
TZ_LOCAL = "America/New_York"

REG_START = "09:30"   # session start
REG_END   = "16:00"   # session end

# Strategy params
RSI_LEN = 14
PIV_LEFT = 2     # bars to the left to confirm a pivot
PIV_RIGHT = 2    # bars to the right to confirm a pivot
MIN_RSI_DIFF = 0.3  # loosened to avoid directional bias  # minimal RSI difference to count as divergence
TICK_SIZE = 0.01    # used to offset stops slightly beyond pivots
RR = 2.0            # risk:reward target multiple

SAVE_DIR = Path(".")

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= INDICATORS =========
def rsi(close: pd.Series, length: int = 14) -> pd.Series:
    delta = close.diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    roll_up = up.ewm(alpha=1/length, adjust=False).mean()
    roll_down = down.ewm(alpha=1/length, adjust=False).mean()
    rs = roll_up / roll_down
    out = 100 - (100 / (1 + rs))
    return out

# ========= PIVOT DETECTION =========
def pivot_high(high: pd.Series, left: int, right: int) -> pd.Series:
    return (high == high.rolling(left+1, min_periods=left+1).max()) \
        & (high == high[::-1].rolling(right+1, min_periods=right+1).max()[::-1])

def pivot_low(low: pd.Series, left: int, right: int) -> pd.Series:
    return (low == low.rolling(left+1, min_periods=left+1).min()) \
        & (low == low[::-1].rolling(right+1, min_periods=right+1).min()[::-1])

# ========= SIMULATION HELPERS =========
def simulate_path(df_after_entry: pd.DataFrame, side: str, entry: float, stop: float, target: float):
    for idx, row in df_after_entry.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    last_idx = df_after_entry.index[-1]
    return "EOD", float(df_after_entry.loc[last_idx, "Close"]), last_idx


def pnl_one_share(side: str, entry: float, exit_px: float) -> float:
    return (exit_px - entry) if side == "long" else (entry - exit_px)

# Trigger fill helper: enter on break of signal bar extreme (reduces side bias)
ENTRY_BUFFER = 0.0  # add buffer if needed, e.g., 0.01

def find_trigger_fill(df_after: pd.DataFrame, side: str, trigger: float):
    for idx, row in df_after.iterrows():
        if side == "long" and row["High"] >= trigger:
            # filled at trigger (market/stop order assumption)
            return trigger, idx
        if side == "short" and row["Low"] <= trigger:
            return trigger, idx
    return None, None

# ========= DIVERGENCE SCAN =========
def find_divergences(df5: pd.DataFrame) -> pd.DataFrame:
    ph = pivot_high(df5["High"], PIV_LEFT, PIV_RIGHT)
    pl = pivot_low(df5["Low"],  PIV_LEFT, PIV_RIGHT)

    ph_conf = ph.shift(PIV_RIGHT).fillna(False)
    pl_conf = pl.shift(PIV_RIGHT).fillna(False)

    piv_highs = df5.loc[ph_conf, ["High", "RSI"]]
    piv_lows  = df5.loc[pl_conf, ["Low",  "RSI"]]

    signals = []

    last_price, last_rsi = None, None
    for idx, row in piv_highs.iterrows():
        price, r = float(row["High"]), float(row["RSI"])
        if last_price is not None and last_rsi is not None:
            if price > last_price and r + MIN_RSI_DIFF <= last_rsi:
                pos = df5.index.get_loc(idx)
                if pos + 1 < len(df5.index):
                    entry_bar = df5.index[pos + 1]
                    signals.append({"Time": entry_bar, "Side": "short", "RefPivotTime": idx})
        last_price, last_rsi = price, r

    last_price, last_rsi = None, None
    for idx, row in piv_lows.iterrows():
        price, r = float(row["Low"]), float(row["RSI"])
        if last_price is not None and last_rsi is not None:
            if price < last_price and r - MIN_RSI_DIFF >= last_rsi:
                pos = df5.index.get_loc(idx)
                if pos + 1 < len(df5.index):
                    entry_bar = df5.index[pos + 1]
                    signals.append({"Time": entry_bar, "Side": "long", "RefPivotTime": idx})
        last_price, last_rsi = price, r

    return pd.DataFrame(signals)

# ========= CORE BACKTEST =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")

    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    df = df.set_index("EventAt")
    ohlc = {"Open": "first", "High": "max", "Low":  "min", "Close":"last"}
    df5 = df.resample("5T", label="right", closed="right").agg(ohlc).dropna()

    tstr = df5.index.tz_convert(TZ_LOCAL).strftime("%H:%M")
    session = (tstr >= REG_START) & (tstr <= REG_END)
    df5 = df5.loc[session].copy()
    if df5.empty:
        return pd.DataFrame([])

    df5["RSI"] = rsi(df5["Close"], RSI_LEN)

    sig = find_divergences(df5)
    if sig.empty:
        return pd.DataFrame([])

    records = []
    for _, s in sig.iterrows():
        signal_time = s["Time"]
        if signal_time not in df5.index:
            continue
        # define trigger on break of signal bar extreme (more symmetric)
        sig_bar = df5.loc[signal_time]
        if s["Side"] == "long":
            trigger = float(sig_bar["High"]) + ENTRY_BUFFER
        else:
            trigger = float(sig_bar["Low"]) - ENTRY_BUFFER

        after = df5.loc[df5.index > signal_time]
        if after.empty:
            continue

        fill_px, fill_time = find_trigger_fill(after, s["Side"], trigger)
        if fill_px is None:
            # Fallback: close-confirmation on the very next bar to reduce directional bias
            next_idx = after.index[0]
            sig_close = float(sig_bar["Close"])
            next_close = float(after.iloc[0]["Close"])
            if s["Side"] == "long" and next_close > sig_close:
                fill_px, fill_time = next_close, next_idx
            elif s["Side"] == "short" and next_close < sig_close:
                fill_px, fill_time = next_close, next_idx
            else:
                continue

        ref_time = s["RefPivotTime"]
        if s["Side"] == "long":
            stop_px = float(df5.loc[ref_time, "Low"]) - TICK_SIZE
            risk = fill_px - stop_px
            if risk <= 0:
                continue
            target_px = fill_px + RR * risk
        else:
            stop_px = float(df5.loc[ref_time, "High"]) + TICK_SIZE
            risk = stop_px - fill_px
            if risk <= 0:
                continue
            target_px = fill_px - RR * risk

        after_entry = df5.loc[df5.index > fill_time]
        if after_entry.empty:
            continue

        outcome, exit_px, exit_idx = simulate_path(after_entry, s["Side"], fill_px, stop_px, target_px)
        pnl = pnl_one_share(s["Side"], fill_px, exit_px)

        records.append({
            "Date": fill_time.date(),
            "Symbol": symbol,
            "Side": s["Side"],
            "EntryTime": fill_time,
            "Entry": fill_px,
            "Stop": stop_px,
            "Target_1to2": target_px,
            "Outcome_1to2": outcome,
            "ExitPx_1to2": exit_px,
            "ExitTime_1to2": exit_idx,
            "PnL_1to2": pnl,
        })

    return pd.DataFrame(records)

# ========= BASIC METRICS =========
def calc_basic_metrics(trades_df):
    pnl = trades_df["PnL_1to2"].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= RUN ALL FILES =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    SAVE_DIR.mkdir(parents=True, exist_ok=True)

    all_trades = []
    for f in files:
        symbol = f.stem.upper()
        try:
            trades = backtest_one_symbol(f, symbol)
        except Exception as e:
            print(f"[SKIP] {symbol}: {e}")
            continue

        if trades.empty:
            print(f"=== RSI Divergence 5m — {symbol} ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            continue

        all_trades.append(trades)

        total_trades, total_pnl, winrate, profit_factor = calc_basic_metrics(trades)

        save_path = SAVE_DIR / f"div_rsi_trades_{symbol}_one_share_RR{int(RR)}.csv"
        trades.to_csv(save_path, index=False)

        print(f"=== RSI Divergence 5m — {symbol} ===")
        print(f"Trades: {total_trades} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {profit_factor:.3f}")
        print(f"Saved trades → {save_path.resolve()}")

    if not all_trades:
        print("\nNo trades generated for any symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    o_trades, o_pnl, o_winrate, o_pf = calc_basic_metrics(overall)

    save_all = SAVE_DIR / f"div_rsi_trades_ALL_one_share_RR{int(RR)}.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== RSI Divergence 5m — OVERALL ===")
    print(f"Trades: {o_trades} | Total PnL: ${o_pnl:,.2f} | Win%: {o_winrate:.2f}% | Profit Factor: {o_pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

if __name__ == "__main__":
    main()
```
