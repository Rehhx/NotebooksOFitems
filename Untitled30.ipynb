```python
# aapl_orb_15m_retest_from_parquet.py
# Backtests the 15-minute ORB (retest) on a local AAPL parquet file (history since 2016).

import warnings
warnings.filterwarnings("ignore")

from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
from pandas.api.types import is_datetime64_any_dtype, is_datetime64tz_dtype

# ========= USER SETTINGS =========
PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"  # <-- your file
TICKER = "AAPL"

# If EventAt in your parquet is naive timestamps representing **UTC**, keep True (most vendors).
# If EventAt is already in **America/New_York**, set this to False.
ASSUME_EVENTAT_IS_UTC = True

LOCAL_TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB parameters
MAX_RETEST_MIN = 120         # wait up to 2h for retest
RETEST_CONFIRM_CLOSE = True  # retest bar must close back in breakout direction
R_MULTIPLES = [1.0, 2.0]     # targets at 1R, 2R

# Risk params
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Outputs
TRADES_CSV = "aapl_orb_trades.csv"
EQUITY_CSV = "aapl_orb_equity.csv"
EXCEL_OUT  = "aapl_orb_results.xlsx"

# ========= LOAD & CLEAN =========
def load_aapl_parquet(path: str) -> pd.DataFrame:
    # Requires pyarrow or fastparquet installed
    df = pd.read_parquet(path, engine="pyarrow")
    # Standardize column names we need
    # Expecting columns like: symbol, Interval, EventAt, Open, High, Low, Close, Volume, ...
    cols_lower = {c: c.strip() for c in df.columns}
    df = df.rename(columns=cols_lower)

    # Decode symbol if stored as bytes (e.g., b'AAPL')
    if "symbol" in df.columns:
        df["symbol"] = df["symbol"].apply(
            lambda x: x.decode("utf-8") if isinstance(x, (bytes, bytearray)) else str(x)
        ).str.strip().str.replace("^b['\"]?|['\"]?$", "", regex=True)
    else:
        df["symbol"] = TICKER

    # Ensure EventAt is datetime
    if "EventAt" not in df.columns:
        raise ValueError("Expected 'EventAt' column in the parquet.")

    if not is_datetime64_any_dtype(df["EventAt"]):
        df["EventAt"] = pd.to_datetime(df["EventAt"], utc=False, errors="coerce")

    # Timezone handling
    if is_datetime64tz_dtype(df["EventAt"]):
        # Already tz-aware -> convert to local tz
        df["EventAt"] = df["EventAt"].dt.tz_convert(LOCAL_TZ)
    else:
        # Naive timestamps; assume UTC (typical). If your file is already ET, flip the flag above.
        if ASSUME_EVENTAT_IS_UTC:
            df["EventAt"] = df["EventAt"].dt.tz_localize("UTC").dt.tz_convert(LOCAL_TZ)
        else:
            df["EventAt"] = df["EventAt"].dt.tz_localize(LOCAL_TZ)

    # Keep AAPL only (in case file has other symbols)
    df = df[df["symbol"] == TICKER].copy()

    # Sort, set index
    df = df.sort_values("EventAt").set_index("EventAt")
    # Keep core columns
    keep = ["Open", "High", "Low", "Close", "Volume", "symbol"]
    df = df[[c for c in keep if c in df.columns]]
    return df

# ========= BAR BUILDERS =========
def to_15m(df: pd.DataFrame) -> pd.DataFrame:
    """
    Build 15-minute OHLCV bars from raw intraday (e.g., 1-minute) data.
    Anchored to the clock (09:30, 09:45, ...).
    """
    o = df["Open"].resample("15T", label="left", closed="left").first()
    h = df["High"].resample("15T", label="left", closed="left").max()
    l = df["Low"].resample("15T", label="left", closed="left").min()
    c = df["Close"].resample("15T", label="left", closed="left").last()
    v = df["Volume"].resample("15T", label="left", closed="left").sum()
    out = pd.concat({"Open": o, "High": h, "Low": l, "Close": c, "Volume": v}, axis=1)
    out = out.dropna(how="any")
    out["Ticker"] = TICKER
    # Regular hours only
    out = out.between_time(REG_SESSION_START, REG_SESSION_END)
    return out

def session_key(index_like) -> pd.Series:
    """One key per trading day in local tz."""
    return pd.to_datetime(pd.Index(index_like).date)

def compute_opening_range(df15: pd.DataFrame) -> pd.DataFrame:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    # First 15m bar per session is the opening range in 15m world
    first_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH", "ORL"]] = df.groupby("Session")[["ORH", "ORL"]].ffill()
    return df

# ========= SLIPPAGE =========
def apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps / 10000.0)
    return price * factor if side == "buy" else price / factor

# ========= BACKTEST =========
def backtest_orb_retest(
    df15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout close after the opening bar
        after_open = sdf.iloc[1:].copy()
        long_break  = after_open[after_open["Close"] > orh].head(1)
        short_break = after_open[after_open["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            if long_break.index[0] < short_break.index[0]:
                direction, breakout_row = "long", long_break.iloc[0]
            else:
                direction, breakout_row = "short", short_break.iloc[0]
        elif not long_break.empty:
            direction, breakout_row = "long", long_break.iloc[0]
        else:
            direction, breakout_row = "short", short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if window.empty:
            continue

        level = orh if direction == "long" else orl
        touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1

        targets = [(entry + r * risk_per_share) if direction == "long" else (entry - r * risk_per_share)
                   for r in r_targets]

        # Walk forward for exits
        run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in run.iterrows():
            hi, lo = row["High"], row["Low"]
            # Stop first
            if lo <= stop <= hi:
                px = apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets
            hit = False
            for i, tp in enumerate(targets):
                if lo <= tp <= hi:
                    px = apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = run.iloc[-1]
            px = apply_slippage(last["Close"], slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _, _, px in exits) - fees

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "Direction": direction,
            "ORH": orh, "ORL": orl,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": targets,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        equity = pd.DataFrame(columns=["Session", "CumPnL_$"])
        return trade_log, equity

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$": "CumPnL_$"}))
    return trade_log, equity

# ========= EXCEL SAFETY =========
def make_excel_compatible(df: pd.DataFrame) -> pd.DataFrame:
    """Return a copy with tz-aware datetimes converted to naive (LOCAL_TZ)."""
    out = df.copy()
    for col in out.columns:
        if is_datetime64tz_dtype(out[col]):
            out[col] = out[col].dt.tz_convert(LOCAL_TZ).dt.tz_localize(None)
    return out

# ========= MAIN =========
def main():
    print("Loading parquet...")
    raw = load_aapl_parquet(PATH)
    if raw.empty:
        print("No rows found in the parquet after filtering for AAPL.")
        return

    print("Building 15m bars...")
    df15 = to_15m(raw)

    print("Computing opening ranges & backtesting...")
    df15 = compute_opening_range(df15)
    trades, equity = backtest_orb_retest(df15)

    if trades.empty:
        print("No trades generated with current settings.")
        return

    # Save CSVs
    out_trades = trades.copy()
    out_trades["Targets"] = out_trades["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
    out_trades["Exits"]   = out_trades["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
    out_trades.to_csv(TRADES_CSV, index=False)
    equity.to_csv(EQUITY_CSV, index=False)
    print(f"Saved: {TRADES_CSV}, {EQUITY_CSV}")

    # Per-period summary (optional)
    trades["Win"] = trades["PnL_$"] > 0
    by_year = trades.groupby(pd.to_datetime(trades["Session"]).dt.year)["PnL_$"].agg(["count", "sum", "mean"])
    print("\n=== Yearly summary (count/sum/mean) ===")
    print(by_year.to_string())

    # Overall metrics
    total_trades = len(trades)
    total_pnl = trades["PnL_$"].sum()
    win_rate = (trades["PnL_$"] > 0).mean() * 100.0
    avg_r = trades["R_multiple"].mean()
    median_r = trades["R_multiple"].median()
    avg_win = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
    avg_loss = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()  # negative
    pf = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())

    print("\n=== Overall Totals ===")
    print(f"Trades: {total_trades} | Total PnL ($): {total_pnl:,.2f} | Win%: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Avg Win ($): {0.0 if pd.isna(avg_win) else avg_win:,.2f} | Avg Loss ($): {0.0 if pd.isna(avg_loss) else avg_loss:,.2f}")
    print(f"Profit Factor: {pf:.3f}")

    # ---- Excel workbook (timezone-safe) ----
    trades_x = make_excel_compatible(trades)
    equity_x = make_excel_compatible(equity)
    summary = pd.DataFrame({
        "Metric": ["TotalTrades", "TotalPnL", "WinRate_%", "AvgR", "MedianR", "AvgWin_$", "AvgLoss_$", "ProfitFactor"],
        "Value":  [total_trades, total_pnl, win_rate, avg_r, median_r, avg_win, avg_loss, pf]
    })

    with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
        summary.to_excel(writer, sheet_name="Summary", index=False)
        trades_x.to_excel(writer, sheet_name="Trades", index=False)
        equity_x.to_excel(writer, sheet_name="Equity", index=False)
    print(f"Wrote Excel workbook: {EXCEL_OUT}")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import numpy as np

# ==== CONFIG ====
PARQUET_PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"

# If you already have a trade PnL column (e.g., "PnL_$"), set it here.
# Leave as None to compute a proxy PnL from price moves.
PNL_COL = None            # e.g., "PnL_$"  (keep None to auto-compute)
USE_PCT_RETURNS = False   # if no PnL col: False -> dollar change (Close.diff); True -> % return

# Column name candidates (auto-detected if exact names differ)
TICKER_COL_CANDIDATES = ["Ticker", "Symbol", "ticker", "symbol"]
DATE_COL_CANDIDATES   = [
    "Date", "date", "Timestamp", "timestamp",
    "ExitDate", "TradeDate", "EventAt", "Datetime", "datetime",
]
PRICE_COL_CANDIDATES  = ["Close", "Adj Close", "close", "adj_close"]
R_COL_CANDIDATES      = ["R", "r", "R_multiple", "Rmultiple"]  # optional

# ===== Helpers =====
def _find_col(df, candidates, required=True):
    for c in candidates:
        if c in df.columns:
            return c
    if required:
        raise ValueError(f"Could not find any of {candidates} in columns: {list(df.columns)}")
    return None

def _summarize_group(g, pnl_col, r_col=None):
    pnl = pd.to_numeric(g[pnl_col], errors="coerce").dropna()

    trades = len(pnl)
    wins_mask = pnl > 0
    losses_mask = pnl < 0
    wins = int(wins_mask.sum())
    losses = int(losses_mask.sum())

    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()  # negative or 0

    # Profit Factor
    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf  # no losses
    else:
        pf = np.nan

    out = {
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": wins,
        "Losses": losses,
        "WinRate": (wins / trades * 100.0) if trades else np.nan,
        "AvgWin": pnl[wins_mask].mean() if wins else np.nan,
        "AvgLoss": pnl[losses_mask].mean() if losses else np.nan,
        "PF": pf,
    }

    # Optional R-multiple stats
    if r_col and r_col in g.columns:
        rvals = pd.to_numeric(g[r_col], errors="coerce").dropna()
        out["AvgR"] = rvals.mean() if len(rvals) else np.nan
        out["MedianR"] = rvals.median() if len(rvals) else np.nan
    else:
        out["AvgR"] = np.nan
        out["MedianR"] = np.nan

    return pd.Series(out)

# ===== Load =====
# Try pyarrow first (recommended), then fastparquet
engine_err = None
for eng in ("pyarrow", "fastparquet"):
    try:
        df = pd.read_parquet(PARQUET_PATH, engine=eng)
        parquet_engine = eng
        break
    except Exception as e:
        engine_err = e
else:
    raise SystemExit(f"Failed to read Parquet. Install pyarrow or fastparquet. Last error: {engine_err}")

# Auto-detect key columns
TICKER_COL = _find_col(df, TICKER_COL_CANDIDATES)
DATE_COL   = _find_col(df, DATE_COL_CANDIDATES)

if PNL_COL is None:
    PRICE_COL = _find_col(df, PRICE_COL_CANDIDATES)

# Ensure datetime (tz-naive for grouping)
dt = pd.to_datetime(df[DATE_COL], errors="coerce", utc=True)
df[DATE_COL] = dt.dt.tz_localize(None)

# Sort so diff/pct_change make sense
df = df.sort_values([TICKER_COL, DATE_COL])

# If no explicit PnL column, compute proxy from price moves
if PNL_COL is None:
    if USE_PCT_RETURNS:
        df["__PnL__"] = df.groupby(TICKER_COL, dropna=False)[PRICE_COL].pct_change()
    else:
        df["__PnL__"] = df.groupby(TICKER_COL, dropna=False)[PRICE_COL].diff()
    PNL_COL = "__PnL__"

# Optional R column (if present)
R_COL = None
for c in R_COL_CANDIDATES:
    if c in df.columns:
        R_COL = c
        break

# ===== Overall per ticker =====
overall = (
    df.groupby(TICKER_COL, dropna=False)
      .apply(lambda g: _summarize_group(g, pnl_col=PNL_COL, r_col=R_COL))
      .reset_index()
)

# ===== Per ticker, per month (robust Grouper) =====
monthly = (
    df.groupby([TICKER_COL, pd.Grouper(key=DATE_COL, freq="MS")], dropna=False)
      .apply(lambda g: _summarize_group(g, pnl_col=PNL_COL, r_col=R_COL))
      .reset_index()
      .rename(columns={DATE_COL: "Month"})
)

# Pretty rounding for display
to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
for tbl in (overall, monthly):
    for c in to_round:
        if c in tbl.columns:
            tbl[c] = tbl[c].astype(float).round(4)

print("Parquet engine:", parquet_engine)
print("\n=== Overall per Ticker ===")
print(overall.to_string(index=False))

print("\n=== Per Ticker, Per Month ===")
print(monthly.sort_values(["Month", TICKER_COL]).tail(30).to_string(index=False))

# ===== Optional: save results =====
# overall.to_excel(r"C:\Users\pcagm\OneDrive\Desktop\downloads\summary_overall.xlsx", index=False)
# monthly.to_excel(r"C:\Users\pcagm\OneDrive\Desktop\downloads\summary_monthly.xlsx", index=False)

# Save as XML if desired (pandas >= 1.3)
# overall.to_xml(r"C:\Users\pcagm\OneDrive\Desktop\downloads\summary_overall.xml", index=False, root_name="Overall", row_name="Row")
# monthly.to_xml(r"C:\Users\pcagm\OneDrive\Desktop\downloads\summary_monthly.xml", index=False, root_name="Monthly", row_name="Row")
```

```python
# aapl_orb_15m_retest_from_parquet.py
# Backtests a 15-minute Opening Range Breakout (with retest) on a local AAPL parquet file.

import warnings
warnings.filterwarnings("ignore")

from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
from pandas.api.types import is_datetime64_any_dtype, is_datetime64tz_dtype

# ========= USER SETTINGS =========
PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"  # <-- your file
TICKER = "AAPL"

# If EventAt in your parquet is naive timestamps representing **UTC**, keep True (most vendors).
# If EventAt is already in **America/New_York**, set this to False.
ASSUME_EVENTAT_IS_UTC = True

LOCAL_TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB parameters
MAX_RETEST_MIN = 120         # wait up to 2h for retest
RETEST_CONFIRM_CLOSE = True  # retest bar must close back in breakout direction
R_MULTIPLES = [1.0, 2.0]     # targets at 1R, 2R

# Risk params
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Outputs
TRADES_CSV = "aapl_orb_trades.csv"
EQUITY_CSV = "aapl_orb_equity.csv"
EXCEL_OUT  = "aapl_orb_results.xlsx"

# ========= LOAD & CLEAN =========
def load_aapl_parquet(path: str) -> pd.DataFrame:
    # Requires pyarrow or fastparquet installed
    # Try pyarrow then fastparquet so it's more robust on your machine.
    last_err = None
    for eng in ("pyarrow", "fastparquet"):
        try:
            df = pd.read_parquet(path, engine=eng)
            break
        except Exception as e:
            last_err = e
            df = None
    if df is None:
        raise SystemExit(f"Failed to read Parquet. Install pyarrow or fastparquet. Last error: {last_err}")

    # Expecting columns like: symbol, Interval, EventAt, Open, High, Low, Close, Volume, ...
    # Keep original case; just make sure we have the names we need.
    if "symbol" in df.columns:
        df["symbol"] = df["symbol"].apply(
            lambda x: x.decode("utf-8") if isinstance(x, (bytes, bytearray)) else str(x)
        ).str.strip().str.replace("^b['\"]?|['\"]?$", "", regex=True)
    else:
        df["symbol"] = TICKER

    if "EventAt" not in df.columns:
        raise ValueError("Expected 'EventAt' column in the parquet.")

    # Ensure EventAt is datetime
    if not is_datetime64_any_dtype(df["EventAt"]):
        df["EventAt"] = pd.to_datetime(df["EventAt"], utc=False, errors="coerce")

    # Timezone handling → convert to LOCAL_TZ
    if is_datetime64tz_dtype(df["EventAt"]):
        df["EventAt"] = df["EventAt"].dt.tz_convert(LOCAL_TZ)
    else:
        if ASSUME_EVENTAT_IS_UTC:
            df["EventAt"] = df["EventAt"].dt.tz_localize("UTC").dt.tz_convert(LOCAL_TZ)
        else:
            df["EventAt"] = df["EventAt"].dt.tz_localize(LOCAL_TZ)

    # Keep AAPL only (in case file has other symbols)
    df = df[df["symbol"] == TICKER].copy()

    # Sort, set index
    df = df.sort_values("EventAt").set_index("EventAt")

    # Keep core columns
    keep = ["Open", "High", "Low", "Close", "Volume", "symbol"]
    keep = [c for c in keep if c in df.columns]
    if not {"Open", "High", "Low", "Close"}.issubset(set(keep)):
        missing = {"Open", "High", "Low", "Close"} - set(keep)
        raise ValueError(f"Missing required OHLC columns: {missing}")
    df = df[keep]
    return df

# ========= BAR BUILDERS =========
def to_15m(df: pd.DataFrame) -> pd.DataFrame:
    """
    Build 15-minute OHLCV bars from raw intraday (e.g., 1-minute) data.
    Anchored to the clock (09:30, 09:45, ...). If your file is already 15m,
    resampling will just align to clean 15m bins.
    """
    o = df["Open"].resample("15T", label="left", closed="left").first()
    h = df["High"].resample("15T", label="left", closed="left").max()
    l = df["Low"].resample("15T", label="left", closed="left").min()
    c = df["Close"].resample("15T", label="left", closed="left").last()
    v = df["Volume"].resample("15T", label="left", closed="left").sum() if "Volume" in df.columns else None

    parts = {"Open": o, "High": h, "Low": l, "Close": c}
    if v is not None:
        parts["Volume"] = v
    out = pd.concat(parts, axis=1).dropna(how="any")
    out["Ticker"] = TICKER

    # Regular hours only (LOCAL_TZ-aware index)
    out = out.between_time(REG_SESSION_START, REG_SESSION_END)
    return out

def session_key(index_like) -> pd.Series:
    """One key per trading day in local tz (naive date)."""
    return pd.to_datetime(pd.Index(index_like).date)

def compute_opening_range(df15: pd.DataFrame) -> pd.DataFrame:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    # First 15m bar per session is the opening range
    first_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH", "ORL"]] = df.groupby("Session")[["ORH", "ORL"]].ffill()
    return df

# ========= SLIPPAGE =========
def apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps / 10000.0)
    return price * factor if side == "buy" else price / factor

# ========= BACKTEST =========
def backtest_orb_retest(
    df15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout close after the opening bar
        after_open = sdf.iloc[1:].copy()
        long_break  = after_open[after_open["Close"] > orh].head(1)
        short_break = after_open[after_open["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            if long_break.index[0] < short_break.index[0]:
                direction, breakout_row = "long", long_break.iloc[0]
            else:
                direction, breakout_row = "short", short_break.iloc[0]
        elif not long_break.empty:
            direction, breakout_row = "long", long_break.iloc[0]
        else:
            direction, breakout_row = "short", short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if window.empty:
            continue

        level = orh if direction == "long" else orl
        touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1

        targets = [(entry + r * risk_per_share) if direction == "long" else (entry - r * risk_per_share)
                   for r in r_targets]

        # Walk forward for exits
        run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in run.iterrows():
            hi, lo = row["High"], row["Low"]
            # Stop first
            if lo <= stop <= hi:
                px = apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets (take first target hit)
            hit = False
            for i, tp in enumerate(targets):
                if lo <= tp <= hi:
                    px = apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = run.iloc[-1]
            px = apply_slippage(last["Close"], slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _, _, px in exits) - fees

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "Direction": direction,
            "ORH": orh, "ORL": orl,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": targets,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        equity = pd.DataFrame(columns=["Session", "CumPnL_$"])
        return trade_log, equity

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$": "CumPnL_$"}))
    return trade_log, equity

# ========= EXCEL SAFETY =========
def make_excel_compatible(df: pd.DataFrame) -> pd.DataFrame:
    """Return a copy with tz-aware datetimes converted to naive (LOCAL_TZ)."""
    out = df.copy()
    for col in out.columns:
        if is_datetime64tz_dtype(out[col]):
            out[col] = out[col].dt.tz_convert(LOCAL_TZ).dt.tz_localize(None)
    return out

# ========= SUMMARY (overall + monthly) =========
def _agg_summary(g: pd.DataFrame) -> pd.Series:
    pnl = pd.to_numeric(g["PnL_$"], errors="coerce").dropna()
    trades = len(pnl)
    wins_mask = pnl > 0
    losses_mask = pnl < 0
    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()

    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf
    else:
        pf = np.nan

    out = {
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": int(wins_mask.sum()),
        "Losses": int(losses_mask.sum()),
        "WinRate": (wins_mask.mean() * 100.0) if trades else np.nan,
        "AvgWin": pnl[wins_mask].mean() if wins_mask.any() else np.nan,
        "AvgLoss": pnl[losses_mask].mean() if losses_mask.any() else np.nan,
        "AvgR": g["R_multiple"].mean(),
        "MedianR": g["R_multiple"].median(),
        "PF": pf,
    }
    return pd.Series(out)

def build_summaries(trades: pd.DataFrame) -> Tuple[pd.DataFrame, pd.DataFrame]:
    overall = _agg_summary(trades).to_frame().T
    # monthly by Session date
    ses = pd.to_datetime(trades["Session"])
    month_key = ses.dt.to_period("M").astype(str)  # 'YYYY-MM'
    monthly = trades.copy()
    monthly["Month"] = month_key
    monthly = (monthly.groupby("Month")
                      .apply(_agg_summary)
                      .reset_index()
                      .sort_values("Month"))
    # nice rounding
    to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
    for tbl in (overall, monthly):
        for c in to_round:
            if c in tbl.columns:
                tbl[c] = tbl[c].astype(float).round(4)
    return overall, monthly

# ========= MAIN =========
def main():
    print("Loading parquet...")
    raw = load_aapl_parquet(PATH)
    if raw.empty:
        print("No rows found in the parquet after filtering for AAPL.")
        return

    print("Building 15m bars...")
    df15 = to_15m(raw)

    print("Computing opening ranges & backtesting...")
    df15 = compute_opening_range(df15)
    trades, equity = backtest_orb_retest(df15)

    if trades.empty:
        print("No trades generated with current settings.")
        return

    # Save CSVs (human-readable exits/targets)
    out_trades = trades.copy()
    out_trades["Targets"] = out_trades["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
    out_trades["Exits"]   = out_trades["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
    out_trades.to_csv(TRADES_CSV, index=False)
    equity.to_csv(EQUITY_CSV, index=False)
    print(f"Saved: {TRADES_CSV}, {EQUITY_CSV}")

    # Per-period summary (yearly to console)
    trades["Win"] = trades["PnL_$"] > 0
    by_year = trades.groupby(pd.to_datetime(trades["Session"]).dt.year)["PnL_$"].agg(["count", "sum", "mean"])
    print("\n=== Yearly summary (count/sum/mean) ===")
    print(by_year.to_string())

    # Overall totals to console
    total_trades = len(trades)
    total_pnl = trades["PnL_$"].sum()
    win_rate = (trades["PnL_$"] > 0).mean() * 100.0
    avg_r = trades["R_multiple"].mean()
    median_r = trades["R_multiple"].median()
    avg_win = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
    avg_loss = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()  # negative
    pf = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())

    print("\n=== Overall Totals ===")
    print(f"Trades: {total_trades} | Total PnL ($): {total_pnl:,.2f} | Win%: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Avg Win ($): {0.0 if pd.isna(avg_win) else avg_win:,.2f} | Avg Loss ($): {0.0 if pd.isna(avg_loss) else avg_loss:,.2f}")
    print(f"Profit Factor: {pf:.3f}")

    # Build requested summaries (overall + monthly)
    overall_tbl, monthly_tbl = build_summaries(trades)

    # ---- Excel workbook (timezone-safe) ----
    trades_x = make_excel_compatible(trades)
    equity_x = make_excel_compatible(equity)

    with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
        # key metrics sheet
        summary = pd.DataFrame({
            "Metric": ["TotalTrades", "TotalPnL", "WinRate_%", "AvgR", "MedianR", "AvgWin_$", "AvgLoss_$", "ProfitFactor"],
            "Value":  [total_trades, total_pnl, win_rate, avg_r, median_r, avg_win, avg_loss, pf]
        })
        summary.to_excel(writer, sheet_name="Summary", index=False)

        # trades & equity
        trades_x.to_excel(writer, sheet_name="Trades", index=False)
        equity_x.to_excel(writer, sheet_name="Equity", index=False)

        # your requested tables
        overall_tbl.to_excel(writer, sheet_name="Stats_Overall", index=False)
        monthly_tbl.to_excel(writer, sheet_name="Stats_Monthly", index=False)

    print(f"Wrote Excel workbook: {EXCEL_OUT}")

if __name__ == "__main__":
    main()
```

```python
# --- add/replace these parts in your file ------------------------------------

# (keep your imports)
import numpy as np
import pandas as pd

# (keep everything above build_summaries)

def _agg_summary(g: pd.DataFrame) -> pd.Series:
    pnl = pd.to_numeric(g["PnL_$"], errors="coerce").dropna()
    trades = len(pnl)
    wins_mask = pnl > 0
    losses_mask = pnl < 0
    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()

    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf
    else:
        pf = np.nan

    return pd.Series({
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": int(wins_mask.sum()),
        "Losses": int(losses_mask.sum()),
        "WinRate": (wins_mask.mean() * 100.0) if trades else np.nan,
        "AvgWin": pnl[wins_mask].mean() if wins_mask.any() else np.nan,
        "AvgLoss": pnl[losses_mask].mean() if losses_mask.any() else np.nan,
        "AvgR": g["R_multiple"].mean(),
        "MedianR": g["R_multiple"].median(),
        "PF": pf,
    })

MONTH_ABBR = {1:"Jan",2:"Feb",3:"Mar",4:"Apr",5:"May",6:"Jun",
              7:"Jul",8:"Aug",9:"Sep",10:"Oct",11:"Nov",12:"Dec"}

def _calendar_matrix(df_monthly_by_year: pd.DataFrame, value_col: str) -> pd.DataFrame:
    """Pivot Year x Month for a metric (e.g., PnL_Sum)."""
    mat = (df_monthly_by_year
           .pivot(index="Year", columns="MonthNum", values=value_col)
           .reindex(columns=range(1, 13)))
    mat.columns = [MONTH_ABBR[m] for m in mat.columns]
    return mat.sort_index()

def build_summaries(trades: pd.DataFrame):
    # Overall (single row)
    overall = _agg_summary(trades).to_frame().T

    # Monthly across all years (YYYY-MM)
    ses = pd.to_datetime(trades["Session"])
    trades["_MonthStr"] = ses.dt.to_period("M").astype(str)  # 'YYYY-MM'
    monthly_all = (trades.groupby("_MonthStr").apply(_agg_summary)
                          .reset_index()
                          .rename(columns={"_MonthStr": "Month"})
                          .sort_values("Month"))

    # Monthly by Year (long form)
    trades["_Year"] = ses.dt.year
    trades["_MonthNum"] = ses.dt.month
    trades["_MonthName"] = trades["_MonthNum"].map(MONTH_ABBR)
    monthly_by_year = (trades.groupby(["_Year", "_MonthNum", "_MonthName"])
                              .apply(_agg_summary)
                              .reset_index()
                              .rename(columns={"_Year":"Year","_MonthNum":"MonthNum","_MonthName":"Month"}))
    monthly_by_year = monthly_by_year.sort_values(["Year","MonthNum"])

    # Calendar matrices (wide)
    cal_pnl     = _calendar_matrix(monthly_by_year, "PnL_Sum")
    cal_winrate = _calendar_matrix(monthly_by_year, "WinRate")
    cal_trades  = _calendar_matrix(monthly_by_year, "Trades")
    cal_avgwin  = _calendar_matrix(monthly_by_year, "AvgWin")
    cal_avgloss = _calendar_matrix(monthly_by_year, "AvgLoss")

    # Nicely round numeric columns
    to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
    for tbl in (overall, monthly_all, monthly_by_year):
        for c in to_round:
            if c in tbl.columns:
                tbl[c] = tbl[c].astype(float).round(4)
    for mat in (cal_pnl, cal_winrate, cal_trades, cal_avgwin, cal_avgloss):
        mat[:] = np.round(mat.astype(float), 4)

    return {
        "overall": overall,
        "monthly_all": monthly_all,
        "monthly_by_year": monthly_by_year,
        "cal_pnl": cal_pnl,
        "cal_winrate": cal_winrate,
        "cal_trades": cal_trades,
        "cal_avgwin": cal_avgwin,
        "cal_avgloss": cal_avgloss,
    }

# --- in main(), replace your current summary build + Excel write with this ----
# After: trades, equity = backtest_orb_retest(df15)
summ = build_summaries(trades)

with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
    # Summary sheets you already had
    summary = pd.DataFrame({
        "Metric": ["TotalTrades", "TotalPnL", "WinRate_%", "AvgR", "MedianR",
                   "AvgWin_$", "AvgLoss_$", "ProfitFactor"],
        "Value":  [len(trades), trades["PnL_$"].sum(),
                   (trades["PnL_$"] > 0).mean() * 100.0,
                   trades["R_multiple"].mean(),
                   trades["R_multiple"].median(),
                   trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean(),
                   trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean(),
                   (trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() /
                    abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum()))
                  ]
    })
    summary.to_excel(writer, sheet_name="Summary", index=False)

    make_excel_compatible(trades).to_excel(writer, sheet_name="Trades", index=False)
    make_excel_compatible(equity).to_excel(writer, sheet_name="Equity", index=False)

    # New detailed sheets
    summ["overall"].to_excel(writer, sheet_name="Stats_Overall", index=False)
    summ["monthly_all"].to_excel(writer, sheet_name="Stats_Monthly_All", index=False)
    summ["monthly_by_year"].to_excel(writer, sheet_name="Stats_Monthly_ByYear", index=False)

    # Calendar matrices (Year x Month)
    summ["cal_pnl"].to_excel(writer, sheet_name="Calendar_PnL")
    summ["cal_winrate"].to_excel(writer, sheet_name="Calendar_WinRate")
    summ["cal_trades"].to_excel(writer, sheet_name="Calendar_Trades")
    summ["cal_avgwin"].to_excel(writer, sheet_name="Calendar_AvgWin")
    summ["cal_avgloss"].to_excel(writer, sheet_name="Calendar_AvgLoss")

print(f"Wrote Excel workbook: {EXCEL_OUT}")
# ----------------------------------------------------------------------------- 
```

```python
# aapl_orb_15m_retest_from_parquet.py
# Backtests a 15-minute Opening Range Breakout (with retest) on a local AAPL parquet file.
# Adds monthly-by-year stats and calendar matrices (PnL, WinRate, Trades, AvgWin, AvgLoss).

import warnings
warnings.filterwarnings("ignore")

from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
from pandas.api.types import is_datetime64_any_dtype, is_datetime64tz_dtype

# ========= USER SETTINGS =========
PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"  # <-- your file
TICKER = "AAPL"

# If EventAt in your parquet is naive timestamps representing **UTC**, keep True.
# If EventAt is already in **America/New_York**, set this to False.
ASSUME_EVENTAT_IS_UTC = True

LOCAL_TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB parameters
MAX_RETEST_MIN = 120         # wait up to 2h for retest
RETEST_CONFIRM_CLOSE = True  # retest bar must close back in breakout direction
R_MULTIPLES = [1.0, 2.0]     # targets at 1R, 2R

# Risk params
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Outputs
TRADES_CSV = "aapl_orb_trades.csv"
EQUITY_CSV = "aapl_orb_equity.csv"
EXCEL_OUT  = "aapl_orb_results.xlsx"

# ========= LOAD & CLEAN =========
def load_aapl_parquet(path: str) -> pd.DataFrame:
    # Try pyarrow then fastparquet
    last_err = None
    for eng in ("pyarrow", "fastparquet"):
        try:
            df = pd.read_parquet(path, engine=eng)
            break
        except Exception as e:
            last_err = e
            df = None
    if df is None:
        raise SystemExit(f"Failed to read Parquet. Install pyarrow or fastparquet. Last error: {last_err}")

    # symbol
    if "symbol" in df.columns:
        df["symbol"] = df["symbol"].apply(
            lambda x: x.decode("utf-8") if isinstance(x, (bytes, bytearray)) else str(x)
        ).str.strip().str.replace("^b['\"]?|['\"]?$", "", regex=True)
    else:
        df["symbol"] = TICKER

    # EventAt
    if "EventAt" not in df.columns:
        raise ValueError("Expected 'EventAt' column in the parquet.")
    if not is_datetime64_any_dtype(df["EventAt"]):
        df["EventAt"] = pd.to_datetime(df["EventAt"], utc=False, errors="coerce")

    # Timezone -> LOCAL_TZ
    if is_datetime64tz_dtype(df["EventAt"]):
        df["EventAt"] = df["EventAt"].dt.tz_convert(LOCAL_TZ)
    else:
        if ASSUME_EVENTAT_IS_UTC:
            df["EventAt"] = df["EventAt"].dt.tz_localize("UTC").dt.tz_convert(LOCAL_TZ)
        else:
            df["EventAt"] = df["EventAt"].dt.tz_localize(LOCAL_TZ)

    # Filter AAPL
    df = df[df["symbol"] == TICKER].copy()

    # Sort and set index
    df = df.sort_values("EventAt").set_index("EventAt")

    # Require OHLC
    need = {"Open", "High", "Low", "Close"}
    if not need.issubset(df.columns):
        raise ValueError(f"Missing required OHLC columns: {need - set(df.columns)}")

    keep = [c for c in ["Open", "High", "Low", "Close", "Volume", "symbol"] if c in df.columns]
    return df[keep]

# ========= BAR BUILDERS =========
def to_15m(df: pd.DataFrame) -> pd.DataFrame:
    o = df["Open"].resample("15T", label="left", closed="left").first()
    h = df["High"].resample("15T", label="left", closed="left").max()
    l = df["Low"].resample("15T", label="left", closed="left").min()
    c = df["Close"].resample("15T", label="left", closed="left").last()
    v = df["Volume"].resample("15T", label="left", closed="left").sum() if "Volume" in df.columns else None

    parts = {"Open": o, "High": h, "Low": l, "Close": c}
    if v is not None:
        parts["Volume"] = v
    out = pd.concat(parts, axis=1).dropna(how="any")
    out["Ticker"] = TICKER

    # RTH only
    out = out.between_time(REG_SESSION_START, REG_SESSION_END)
    return out

def session_key(index_like) -> pd.Series:
    """One key per trading day (naive date in LOCAL_TZ)."""
    return pd.to_datetime(pd.Index(index_like).date)

def compute_opening_range(df15: pd.DataFrame) -> pd.DataFrame:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    # Opening 15m bar defines ORH/ORL in 15m frame
    first_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH", "ORL"]] = df.groupby("Session")[["ORH", "ORL"]].ffill()
    return df

# ========= SLIPPAGE =========
def apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps / 10000.0)
    return price * factor if side == "buy" else price / factor

# ========= BACKTEST =========
def backtest_orb_retest(
    df15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout CLOSE after open bar
        after_open = sdf.iloc[1:].copy()
        long_break  = after_open[after_open["Close"] > orh].head(1)
        short_break = after_open[after_open["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            if long_break.index[0] < short_break.index[0]:
                direction, breakout_row = "long", long_break.iloc[0]
            else:
                direction, breakout_row = "short", short_break.iloc[0]
        elif not long_break.empty:
            direction, breakout_row = "long", long_break.iloc[0]
        else:
            direction, breakout_row = "short", short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if window.empty:
            continue

        level = orh if direction == "long" else orl
        touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1

        targets = [(entry + r * risk_per_share) if direction == "long" else (entry - r * risk_per_share)
                   for r in r_targets]

        # Walk forward for exits
        run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in run.iterrows():
            hi, lo = row["High"], row["Low"]
            # Stop first
            if lo <= stop <= hi:
                px = apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets (first target hit)
            hit = False
            for i, tp in enumerate(targets):
                if lo <= tp <= hi:
                    px = apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = run.iloc[-1]
            px = apply_slippage(last["Close"], slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _, _, px in exits) - fees

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "Direction": direction,
            "ORH": orh, "ORL": orl,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": targets,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        equity = pd.DataFrame(columns=["Session", "CumPnL_$"])
        return trade_log, equity

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$": "CumPnL_$"}))
    return trade_log, equity

# ========= EXCEL SAFETY =========
def make_excel_compatible(df: pd.DataFrame) -> pd.DataFrame:
    """Return a copy with tz-aware datetimes converted to naive (LOCAL_TZ)."""
    out = df.copy()
    for col in out.columns:
        if is_datetime64tz_dtype(out[col]):
            out[col] = out[col].dt.tz_convert(LOCAL_TZ).dt.tz_localize(None)
    return out

# ========= SUMMARY HELPERS =========
def _agg_summary(g: pd.DataFrame) -> pd.Series:
    pnl = pd.to_numeric(g["PnL_$"], errors="coerce").dropna()
    trades = len(pnl)
    wins_mask = pnl > 0
    losses_mask = pnl < 0
    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()

    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf
    else:
        pf = np.nan

    return pd.Series({
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": int(wins_mask.sum()),
        "Losses": int(losses_mask.sum()),
        "WinRate": (wins_mask.mean() * 100.0) if trades else np.nan,
        "AvgWin": pnl[wins_mask].mean() if wins_mask.any() else np.nan,
        "AvgLoss": pnl[losses_mask].mean() if losses_mask.any() else np.nan,
        "AvgR": g["R_multiple"].mean(),
        "MedianR": g["R_multiple"].median(),
        "PF": pf,
    })

MONTH_ABBR = {1:"Jan",2:"Feb",3:"Mar",4:"Apr",5:"May",6:"Jun",
              7:"Jul",8:"Aug",9:"Sep",10:"Oct",11:"Nov",12:"Dec"}

def _calendar_matrix(df_monthly_by_year: pd.DataFrame, value_col: str) -> pd.DataFrame:
    """Pivot Year x Month for a metric (e.g., PnL_Sum)."""
    mat = (df_monthly_by_year
           .pivot(index="Year", columns="MonthNum", values=value_col)
           .reindex(columns=range(1, 13)))
    mat.columns = [MONTH_ABBR[m] for m in mat.columns]
    return mat.sort_index()

def build_summaries(trades: pd.DataFrame):
    # Overall (single row)
    overall = _agg_summary(trades).to_frame().T

    # Monthly across all years (YYYY-MM)
    ses = pd.to_datetime(trades["Session"])
    trades = trades.copy()
    trades["_MonthStr"] = ses.dt.to_period("M").astype(str)  # 'YYYY-MM'
    monthly_all = (trades.groupby("_MonthStr").apply(_agg_summary)
                          .reset_index()
                          .rename(columns={"_MonthStr": "Month"})
                          .sort_values("Month"))

    # Monthly by Year (long form)
    trades["_Year"] = ses.dt.year
    trades["_MonthNum"] = ses.dt.month
    trades["_MonthName"] = trades["_MonthNum"].map(MONTH_ABBR)
    monthly_by_year = (trades.groupby(["_Year", "_MonthNum", "_MonthName"])
                              .apply(_agg_summary)
                              .reset_index()
                              .rename(columns={"_Year":"Year","_MonthNum":"MonthNum","_MonthName":"Month"}))
    monthly_by_year = monthly_by_year.sort_values(["Year","MonthNum"])

    # Calendar matrices (wide)
    cal_pnl     = _calendar_matrix(monthly_by_year, "PnL_Sum")
    cal_winrate = _calendar_matrix(monthly_by_year, "WinRate")
    cal_trades  = _calendar_matrix(monthly_by_year, "Trades")
    cal_avgwin  = _calendar_matrix(monthly_by_year, "AvgWin")
    cal_avgloss = _calendar_matrix(monthly_by_year, "AvgLoss")

    # Round
    to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
    for tbl in (overall, monthly_all, monthly_by_year):
        for c in to_round:
            if c in tbl.columns:
                tbl[c] = tbl[c].astype(float).round(4)
    for mat in (cal_pnl, cal_winrate, cal_trades, cal_avgwin, cal_avgloss):
        mat[:] = np.round(mat.astype(float), 4)

    return {
        "overall": overall,
        "monthly_all": monthly_all,
        "monthly_by_year": monthly_by_year,
        "cal_pnl": cal_pnl,
        "cal_winrate": cal_winrate,
        "cal_trades": cal_trades,
        "cal_avgwin": cal_avgwin,
        "cal_avgloss": cal_avgloss,
    }

# ========= MAIN =========
def main():
    print("Loading parquet...")
    raw = load_aapl_parquet(PATH)
    if raw.empty:
        print("No rows found in the parquet after filtering for AAPL.")
        return

    print("Building 15m bars...")
    df15 = to_15m(raw)

    print("Computing opening ranges & backtesting...")
    df15 = compute_opening_range(df15)
    trades, equity = backtest_orb_retest(df15)
    if trades.empty:
        print("No trades generated with current settings.")
        return

    # Save CSVs (human-readable exits/targets)
    out_trades = trades.copy()
    out_trades["Targets"] = out_trades["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
    out_trades["Exits"]   = out_trades["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
    out_trades.to_csv(TRADES_CSV, index=False)
    equity.to_csv(EQUITY_CSV, index=False)
    print(f"Saved: {TRADES_CSV}, {EQUITY_CSV}")

    # Console yearly
    trades["Win"] = trades["PnL_$"] > 0
    by_year = trades.groupby(pd.to_datetime(trades["Session"]).dt.year)["PnL_$"].agg(["count", "sum", "mean"])
    print("\n=== Yearly summary (count/sum/mean) ===")
    print(by_year.to_string())

    # Console overall
    total_trades = len(trades)
    total_pnl = trades["PnL_$"].sum()
    win_rate = (trades["PnL_$"] > 0).mean() * 100.0
    avg_r = trades["R_multiple"].mean()
    median_r = trades["R_multiple"].median()
    avg_win = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
    avg_loss = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()
    pf = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())

    print("\n=== Overall Totals ===")
    print(f"Trades: {total_trades} | Total PnL ($): {total_pnl:,.2f} | Win%: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Avg Win ($): {0.0 if pd.isna(avg_win) else avg_win:,.2f} | Avg Loss ($): {0.0 if pd.isna(avg_loss) else avg_loss:,.2f}")
    print(f"Profit Factor: {pf:.3f}")

    # Build new summaries
    summ = build_summaries(trades)

    # Excel workbook
    with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
        summary = pd.DataFrame({
            "Metric": ["TotalTrades", "TotalPnL", "WinRate_%", "AvgR", "MedianR",
                       "AvgWin_$", "AvgLoss_$", "ProfitFactor"],
            "Value":  [total_trades, total_pnl, win_rate, avg_r, median_r, avg_win, avg_loss, pf]
        })
        summary.to_excel(writer, sheet_name="Summary", index=False)

        make_excel_compatible(trades).to_excel(writer, sheet_name="Trades", index=False)
        make_excel_compatible(equity).to_excel(writer, sheet_name="Equity", index=False)

        summ["overall"].to_excel(writer, sheet_name="Stats_Overall", index=False)
        summ["monthly_all"].to_excel(writer, sheet_name="Stats_Monthly_All", index=False)
        summ["monthly_by_year"].to_excel(writer, sheet_name="Stats_Monthly_ByYear", index=False)

        # Calendar matrices (Year x Month)
        summ["cal_pnl"].to_excel(writer, sheet_name="Calendar_PnL")
        summ["cal_winrate"].to_excel(writer, sheet_name="Calendar_WinRate")
        summ["cal_trades"].to_excel(writer, sheet_name="Calendar_Trades")
        summ["cal_avgwin"].to_excel(writer, sheet_name="Calendar_AvgWin")
        summ["cal_avgloss"].to_excel(writer, sheet_name="Calendar_AvgLoss")

    print(f"Wrote Excel workbook: {EXCEL_OUT}")

if __name__ == "__main__":
    main()
```

```python
# ======== WIN-RATE FILTERS (toggles) ========
USE_VWAP_FILTER = True
USE_TREND_FILTER = True
EMA_FAST = 20
EMA_SLOW = 50

USE_GAP_FILTER = True
GAP_MIN = 0.005   # 0.5%
GAP_MAX = 0.03    # 3%

USE_SPY_CONFIRM = False
SPY_PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"  # set if USE_SPY_CONFIRM=True

# Retest bar quality: limit opposite wick as fraction of range (e.g., <= 40%)
USE_RETEST_WICK_FILTER = True
RETEST_MAX_OPP_WICK_FRAC = 0.40

# Risk management to improve “win%”
MOVE_STOP_TO_BE_AT_R = 0.5   # None to disable; move stop to entry at +0.5R
COUNT_BE_AS_WIN = True       # count breakeven exits as wins in stats


def add_session_vwap(df15: pd.DataFrame) -> pd.DataFrame:
    """Session VWAP using 15m bars (approx): cum(sum(price*vol))/cum(sum(vol)) per day."""
    df = df15.copy()
    df["Session"] = session_key(df.index)
    # typical price ~ (H+L+C)/3; volume may be absent
    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0
    if "Volume" in df.columns:
        df["vwap_num"] = tp * df["Volume"]
        df["vwap_den"] = df["Volume"]
    else:
        # fallback: equal weights
        df["vwap_num"] = tp
        df["vwap_den"] = 1.0
    df["VWAP"] = (df.groupby("Session")["vwap_num"].cumsum() /
                  df.groupby("Session")["vwap_den"].cumsum())
    df = df.drop(columns=["vwap_num","vwap_den"])
    return df

def add_ema_trend(df15: pd.DataFrame, fast=20, slow=50) -> pd.DataFrame:
    df = df15.copy()
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return df

def add_gap_feature(df15: pd.DataFrame) -> pd.DataFrame:
    """Gap % vs prior session close at session open bar."""
    df = df15.copy()
    df["Session"] = session_key(df.index)
    first_idx = df.groupby("Session").head(1).index
    prior_close = df.groupby("Session")["Close"].shift(1)  # last bar prev session
    # Fill prior_close for first bar with prior session last close
    df["PrevClose"] = prior_close
    df["PrevClose"] = df["PrevClose"].ffill()
    df.loc[first_idx, "GapPct"] = (df.loc[first_idx, "Open"] / df.loc[first_idx, "PrevClose"] - 1.0)
    df["GapPct"] = df.groupby("Session")["GapPct"].ffill()
    return df

def wick_opposite_fraction(row, direction: str) -> float:
    """Opposite wick / true range on the RETEST bar."""
    h, l, o, c = row["High"], row["Low"], row["Open"], row["Close"]
    rng = max(h - l, 1e-12)
    if direction == "long":
        # opposite is upper wick when we need acceptance upward? Actually for retest confirmation,
        # we prefer small *opposite* wick (down) → but we confirm close >= level.
        # Use upper wick as 'opposite' for conservative breakout bars; for retest quality,
        # we’ll measure the *wrong-way* wick relative to direction:
        opp = h - max(o, c)  # upper wick
    else:
        opp = min(o, c) - l  # lower wick
    return float(max(opp, 0.0) / rng)

```

```python
# aapl_orb_15m_retest_from_parquet.py
# ORB 15m "retest" backtest on AAPL parquet + win-rate filters and monthly stats.

import warnings
warnings.filterwarnings("ignore")

from datetime import timedelta
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
from pandas.api.types import is_datetime64_any_dtype, is_datetime64tz_dtype

# ========= USER SETTINGS =========
PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"  # <-- your file
TICKER = "AAPL"

# If EventAt in your parquet is naive timestamps representing **UTC**, keep True.
# If EventAt is already in **America/New_York**, set this to False.
ASSUME_EVENTAT_IS_UTC = True

LOCAL_TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB parameters
MAX_RETEST_MIN = 120         # wait up to 2h for retest
RETEST_CONFIRM_CLOSE = True  # retest bar must close back in breakout direction
R_MULTIPLES = [1.0, 2.0]     # targets at 1R, 2R

# Risk params
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# ======== WIN-RATE FILTERS & TUNING ========
USE_VWAP_FILTER = True
USE_TREND_FILTER = True
EMA_FAST = 20
EMA_SLOW = 50

USE_GAP_FILTER = True
GAP_MIN = 0.00    # widened band to be less restrictive
GAP_MAX = 0.05

USE_SPY_CONFIRM = False
SPY_PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\SPY.parquet"  # set if USE_SPY_CONFIRM=True

# Retest bar quality: limit *opposite* wick as fraction of bar range (<= 60%)
USE_RETEST_WICK_FILTER = True
RETEST_MAX_OPP_WICK_FRAC = 0.60

# Intrabar modeling when a bar hits BOTH stop & any target
# "pessimistic" | "optimistic" | "open_to_close"
INTRABAR_MODEL = "optimistic"

# VWAP slope requirement toggle (default off)
VWAP_REQUIRE_SLOPE = False

# Risk management
MOVE_STOP_TO_BE_AT_R = None   # None disables BE stop; set e.g. 0.5 to re-enable
COUNT_BE_AS_WIN = True        # count breakeven as win in stats

# Outputs
TRADES_CSV = "aapl_orb_trades.csv"
EQUITY_CSV = "aapl_orb_equity.csv"
EXCEL_OUT  = "aapl_orb_results.xlsx"

# ========= LOAD & CLEAN =========
def load_aapl_parquet(path: str) -> pd.DataFrame:
    # Try pyarrow then fastparquet
    last_err = None
    df = None
    for eng in ("pyarrow", "fastparquet"):
        try:
            df = pd.read_parquet(path, engine=eng)
            break
        except Exception as e:
            last_err = e
    if df is None:
        raise SystemExit(f"Failed to read Parquet. Install pyarrow or fastparquet. Last error: {last_err}")

    # symbol
    if "symbol" in df.columns:
        df["symbol"] = df["symbol"].apply(
            lambda x: x.decode("utf-8") if isinstance(x, (bytes, bytearray)) else str(x)
        ).str.strip().str.replace("^b['\"]?|['\"]?$", "", regex=True)
    else:
        df["symbol"] = TICKER

    # EventAt
    if "EventAt" not in df.columns:
        raise ValueError("Expected 'EventAt' column in the parquet.")
    if not is_datetime64_any_dtype(df["EventAt"]):
        df["EventAt"] = pd.to_datetime(df["EventAt"], utc=False, errors="coerce")

    # Timezone -> LOCAL_TZ
    if is_datetime64tz_dtype(df["EventAt"]):
        df["EventAt"] = df["EventAt"].dt.tz_convert(LOCAL_TZ)
    else:
        if ASSUME_EVENTAT_IS_UTC:
            df["EventAt"] = df["EventAt"].dt.tz_localize("UTC").dt.tz_convert(LOCAL_TZ)
        else:
            df["EventAt"] = df["EventAt"].dt.tz_localize(LOCAL_TZ)

    # Filter symbol
    df = df[df["symbol"] == TICKER].copy()

    # Sort and index
    df = df.sort_values("EventAt").set_index("EventAt")

    # Require OHLC
    need = {"Open", "High", "Low", "Close"}
    if not need.issubset(df.columns):
        raise ValueError(f"Missing required OHLC columns: {need - set(df.columns)}")

    keep = [c for c in ["Open", "High", "Low", "Close", "Volume", "symbol"] if c in df.columns]
    return df[keep]

# ========= BAR BUILDERS =========
def to_15m(df: pd.DataFrame) -> pd.DataFrame:
    o = df["Open"].resample("15T", label="left", closed="left").first()
    h = df["High"].resample("15T", label="left", closed="left").max()
    l = df["Low"].resample("15T", label="left", closed="left").min()
    c = df["Close"].resample("15T", label="left", closed="left").last()
    parts = {"Open": o, "High": h, "Low": l, "Close": c}
    if "Volume" in df.columns:
        parts["Volume"] = df["Volume"].resample("15T", label="left", closed="left").sum()
    out = pd.concat(parts, axis=1).dropna(how="any")
    out["Ticker"] = TICKER
    # RTH only
    out = out.between_time(REG_SESSION_START, REG_SESSION_END)
    return out

def session_key(index_like) -> pd.Series:
    """One key per trading day (naive date in LOCAL_TZ)."""
    return pd.to_datetime(pd.Index(index_like).date)

def compute_opening_range(df15: pd.DataFrame) -> pd.DataFrame:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    first_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH", "ORL"]] = df.groupby("Session")[["ORH", "ORL"]].ffill()
    return df

# ========= INDICATORS / FEATURES =========
def add_session_vwap(df15: pd.DataFrame) -> pd.DataFrame:
    """
    Session VWAP (15m approximation).
    If Volume exists: VWAP = cum(typical*vol)/cum(vol) within session.
    If no Volume: fallback to within-session running mean of typical price.
    """
    df = df15.copy()
    if "Session" not in df.columns:
        df["Session"] = session_key(df.index)

    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0

    if "Volume" in df.columns:
        cum_num = (tp * df["Volume"]).groupby(df["Session"]).cumsum()
        cum_den = df["Volume"].groupby(df["Session"]).cumsum()
        df["VWAP"] = cum_num / cum_den
    else:
        df["VWAP"] = (
            tp.groupby(df["Session"])
              .expanding()
              .mean()
              .reset_index(level=0, drop=True)
        )
    return df

def add_ema_trend(df15: pd.DataFrame, fast=20, slow=50) -> pd.DataFrame:
    df = df15.copy()
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return df

def add_gap_feature(df15: pd.DataFrame) -> pd.DataFrame:
    """Compute gap % at session open vs prior session close; ffill within session."""
    df = df15.copy()
    df["Session"] = session_key(df.index)
    last_close_per_session = df.groupby("Session")["Close"].last()
    prev_close_map = last_close_per_session.shift(1)
    df["PrevClose"] = df["Session"].map(prev_close_map)
    first_idx = df.groupby("Session").head(1).index
    df["GapPct"] = np.nan
    mask = df.index.isin(first_idx) & df["PrevClose"].notna()
    df.loc[mask, "GapPct"] = (df.loc[mask, "Open"] / df.loc[mask, "PrevClose"] - 1.0)
    df["GapPct"] = df.groupby("Session")["GapPct"].ffill()
    return df

def wick_opposite_fraction(row: pd.Series, direction: str) -> float:
    """Opposite wick / true range on the bar (for longs: upper wick; shorts: lower wick)."""
    h, l, o, c = row["High"], row["Low"], row["Open"], row["Close"]
    rng = max(h - l, 1e-12)
    if direction == "long":
        opp = h - max(o, c)  # upper wick height
    else:
        opp = min(o, c) - l  # lower wick height
    return float(max(opp, 0.0) / rng)

# ========= SLIPPAGE =========
def apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps / 10000.0)
    return price * factor if side == "buy" else price / factor

# ========= OPTIONAL: SPY LOADER =========
def load_spy_if_needed() -> Optional[pd.DataFrame]:
    if not USE_SPY_CONFIRM:
        return None
    spy = load_aapl_parquet(SPY_PATH).rename(columns={"symbol": "SPY"})
    spy15 = to_15m(spy)
    spy15 = compute_opening_range(spy15)
    spy15 = add_session_vwap(spy15)
    spy15["SPY_VWAP"] = spy15["VWAP"]
    spy15 = spy15[["SPY_VWAP", "Close"]].rename(columns={"Close": "SPY_Close"})
    return spy15

# ========= BACKTEST =========
def backtest_orb_retest(
    df15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE,
    spy15: Optional[pd.DataFrame] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout CLOSE after the opening bar
        after_open = sdf.iloc[1:].copy()
        long_break  = after_open[after_open["Close"] > orh].head(1)
        short_break = after_open[after_open["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            if long_break.index[0] < short_break.index[0]:
                direction, breakout_row = "long", long_break.iloc[0]
            else:
                direction, breakout_row = "short", short_break.iloc[0]
        elif not long_break.empty:
            direction, breakout_row = "long", long_break.iloc[0]
        else:
            direction, breakout_row = "short", short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if window.empty:
            continue

        level = orh if direction == "long" else orl
        touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        # --- Filters to improve win-rate ---
        # VWAP filter: price on correct side + (optional) slope
        if USE_VWAP_FILTER and "VWAP" in sdf.columns:
            vwap_series = sdf.loc[:retest_time, "VWAP"]
            vwap_now = float(vwap_series.iloc[-1])
            slope_ok = True
            if VWAP_REQUIRE_SLOPE and len(vwap_series) >= 2:
                vwap_prev = float(vwap_series.iloc[-2])
                slope_ok = (vwap_now >= vwap_prev) if direction == "long" else (vwap_now <= vwap_prev)
            if direction == "long":
                if not (retest_bar["Close"] > vwap_now and slope_ok):
                    continue
            else:
                if not (retest_bar["Close"] < vwap_now and slope_ok):
                    continue

        # Trend filter (EMAs)
        if USE_TREND_FILTER and (f"EMA{EMA_FAST}" in sdf.columns) and (f"EMA{EMA_SLOW}" in sdf.columns):
            emaf = float(retest_bar[f"EMA{EMA_FAST}"]); emas = float(retest_bar[f"EMA{EMA_SLOW}"])
            if direction == "long" and not (emaf > emas):
                continue
            if direction == "short" and not (emaf < emas):
                continue

        # Gap filter (use session's first bar gap)
        if USE_GAP_FILTER and "GapPct" in sdf.columns:
            gap = float(sdf["GapPct"].dropna().iloc[0]) if sdf["GapPct"].notna().any() else 0.0
            if not (GAP_MIN <= abs(gap) <= GAP_MAX):
                continue

        # SPY confirmation
        if USE_SPY_CONFIRM and (spy15 is not None):
            if retest_time in spy15.index:
                spy_row = spy15.loc[retest_time]
                if direction == "long" and not (spy_row["SPY_Close"] > spy_row["SPY_VWAP"]):
                    continue
                if direction == "short" and not (spy_row["SPY_Close"] < spy_row["SPY_VWAP"]):
                    continue

        # Retest wick quality (opposite wick fraction <= threshold)
        if USE_RETEST_WICK_FILTER:
            opp_frac = wick_opposite_fraction(retest_bar, direction)
            if opp_frac > RETEST_MAX_OPP_WICK_FRAC:
                continue
        # --- end filters ---

        entry = apply_slippage(level, SLIPPAGE_BPS, "buy" if direction == "long" else "sell")
        stop = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = POSITION_SIZE_DOLLARS / entry
        side_mult = 1 if direction == "long" else -1

        targets = [(entry + r * risk_per_share) if direction == "long" else (entry - r * risk_per_share)
                   for r in r_targets]

        # Walk forward for exits
        run = sdf[sdf.index >= retest_time].copy()
        exits = []
        be_armed = (MOVE_STOP_TO_BE_AT_R is not None)
        be_level = entry
        rps = risk_per_share

        for ts, row in run.iterrows():
            hi, lo, o, cl = row["High"], row["Low"], row["Open"], row["Close"]

            # Arm BE once price reaches +X R
            if be_armed:
                if direction == "long" and hi >= entry + (MOVE_STOP_TO_BE_AT_R * rps):
                    stop = max(stop, be_level); be_armed = False
                elif direction == "short" and lo <= entry - (MOVE_STOP_TO_BE_AT_R * rps):
                    stop = min(stop, be_level); be_armed = False

            # Compute hits this bar
            stop_hit = (lo <= stop <= hi)
            tp_hit_idxs = [i for i, tp in enumerate(targets) if lo <= tp <= hi]

            if stop_hit and tp_hit_idxs:
                if INTRABAR_MODEL == "optimistic":
                    tp_idx = tp_hit_idxs[0]
                    px = apply_slippage(targets[tp_idx], SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{tp_idx+1}", ts, px))
                    break
                elif INTRABAR_MODEL == "open_to_close":
                    high_first = (cl >= o)
                    if (direction == "long" and high_first) or (direction == "short" and not high_first):
                        tp_idx = tp_hit_idxs[0]
                        px = apply_slippage(targets[tp_idx], SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                        exits.append((f"tp{tp_idx+1}", ts, px))
                    else:
                        px = apply_slippage(stop, SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                        label = "breakeven" if abs(px - be_level) < 1e-8 else "stop"
                        exits.append((label, ts, px))
                    break
                else:  # pessimistic
                    px = apply_slippage(stop, SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                    label = "breakeven" if abs(px - be_level) < 1e-8 else "stop"
                    exits.append((label, ts, px))
                    break

            # Only stop hits
            if stop_hit:
                px = apply_slippage(stop, SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                label = "breakeven" if abs(px - be_level) < 1e-8 else "stop"
                exits.append((label, ts, px))
                break

            # Only targets hit (take the first)
            if tp_hit_idxs:
                tp_idx = tp_hit_idxs[0]
                px = apply_slippage(targets[tp_idx], SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                exits.append((f"tp{tp_idx+1}", ts, px))
                break

        # If nothing hit all day, exit EOD
        if not exits:
            last = run.iloc[-1]
            px = apply_slippage(last["Close"], SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
            exits.append(("eod", run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _, _, px in exits) - FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "Direction": direction,
            "ORH": orh, "ORL": orl,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": targets,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        equity = pd.DataFrame(columns=["Session", "CumPnL_$"])
        return trade_log, equity

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$": "CumPnL_$"}))
    return trade_log, equity

# ========= EXCEL SAFETY =========
def make_excel_compatible(df: pd.DataFrame) -> pd.DataFrame:
    """Return a copy with tz-aware datetimes converted to naive (LOCAL_TZ)."""
    out = df.copy()
    for col in out.columns:
        if is_datetime64tz_dtype(out[col]):
            out[col] = out[col].dt.tz_convert(LOCAL_TZ).dt.tz_localize(None)
    return out

# ========= SUMMARY HELPERS =========
def _agg_summary(g: pd.DataFrame) -> pd.Series:
    pnl = pd.to_numeric(g["PnL_$"], errors="coerce").dropna()
    trades = len(pnl)
    wins_mask = pnl > 0 if not COUNT_BE_AS_WIN else pnl >= 0
    losses_mask = pnl < 0
    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()

    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf
    else:
        pf = np.nan

    return pd.Series({
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": int(wins_mask.sum()),
        "Losses": int(losses_mask.sum()),
        "WinRate": (wins_mask.mean() * 100.0) if trades else np.nan,
        "AvgWin": pnl[pnl > 0].mean() if (pnl > 0).any() else np.nan,
        "AvgLoss": pnl[pnl < 0].mean() if (pnl < 0).any() else np.nan,
        "AvgR": g["R_multiple"].mean(),
        "MedianR": g["R_multiple"].median(),
        "PF": pf,
    })

MONTH_ABBR = {1:"Jan",2:"Feb",3:"Mar",4:"Apr",5:"May",6:"Jun",
              7:"Jul",8:"Aug",9:"Sep",10:"Oct",11:"Nov",12:"Dec"}

def _calendar_matrix(df_monthly_by_year: pd.DataFrame, value_col: str) -> pd.DataFrame:
    """Pivot Year x Month for a metric (e.g., PnL_Sum)."""
    mat = (df_monthly_by_year
           .pivot(index="Year", columns="MonthNum", values=value_col)
           .reindex(columns=range(1, 13)))
    mat.columns = [MONTH_ABBR[m] for m in mat.columns]
    return mat.sort_index()

def build_summaries(trades: pd.DataFrame):
    # Overall (single row)
    overall = _agg_summary(trades).to_frame().T

    # Monthly across all years (YYYY-MM)
    ses = pd.to_datetime(trades["Session"])
    trades = trades.copy()
    trades["_MonthStr"] = ses.dt.to_period("M").astype(str)  # 'YYYY-MM'
    monthly_all = (trades.groupby("_MonthStr").apply(_agg_summary)
                          .reset_index()
                          .rename(columns={"_MonthStr": "Month"})
                          .sort_values("Month"))

    # Monthly by Year (long form)
    trades["_Year"] = ses.dt.year
    trades["_MonthNum"] = ses.dt.month
    trades["_MonthName"] = trades["_MonthNum"].map(MONTH_ABBR)
    monthly_by_year = (trades.groupby(["_Year", "_MonthNum", "_MonthName"])
                              .apply(_agg_summary)
                              .reset_index()
                              .rename(columns={"_Year":"Year","_MonthNum":"MonthNum","_MonthName":"Month"}))
    monthly_by_year = monthly_by_year.sort_values(["Year","MonthNum"])

    # Calendar matrices (wide)
    cal_pnl     = _calendar_matrix(monthly_by_year, "PnL_Sum")
    cal_winrate = _calendar_matrix(monthly_by_year, "WinRate")
    cal_trades  = _calendar_matrix(monthly_by_year, "Trades")
    cal_avgwin  = _calendar_matrix(monthly_by_year, "AvgWin")
    cal_avgloss = _calendar_matrix(monthly_by_year, "AvgLoss")

    # Round
    to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
    for tbl in (overall, monthly_all, monthly_by_year):
        for c in to_round:
            if c in tbl.columns:
                tbl[c] = tbl[c].astype(float).round(4)
    for mat in (cal_pnl, cal_winrate, cal_trades, cal_avgwin, cal_avgloss):
        mat[:] = np.round(mat.astype(float), 4)

    return {
        "overall": overall,
        "monthly_all": monthly_all,
        "monthly_by_year": monthly_by_year,
        "cal_pnl": cal_pnl,
        "cal_winrate": cal_winrate,
        "cal_trades": cal_trades,
        "cal_avgwin": cal_avgwin,
        "cal_avgloss": cal_avgloss,
    }

# ========= MAIN =========
def main():
    print("Loading parquet...")
    raw = load_aapl_parquet(PATH)
    if raw.empty:
        print("No rows found in the parquet after filtering for AAPL.")
        return

    print("Building 15m bars...")
    df15 = to_15m(raw)

    print("Computing opening ranges & adding filters...")
    df15 = compute_opening_range(df15)
    if USE_VWAP_FILTER:
        df15 = add_session_vwap(df15)
    if USE_TREND_FILTER:
        df15 = add_ema_trend(df15, EMA_FAST, EMA_SLOW)
    if USE_GAP_FILTER:
        df15 = add_gap_feature(df15)

    spy15 = load_spy_if_needed()

    print("Backtesting...")
    trades, equity = backtest_orb_retest(df15, spy15=spy15)
    if trades.empty:
        print("No trades generated with current settings.")
        return

    # Save CSVs (human-readable exits/targets)
    out_trades = trades.copy()
    out_trades["Targets"] = out_trades["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
    out_trades["Exits"]   = out_trades["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
    out_trades.to_csv(TRADES_CSV, index=False)
    equity.to_csv(EQUITY_CSV, index=False)
    print(f"Saved: {TRADES_CSV}, {EQUITY_CSV}")

    # Console yearly
    trades["Win"] = (trades["PnL_$"] > 0) if not COUNT_BE_AS_WIN else (trades["PnL_$"] >= 0)
    by_year = trades.groupby(pd.to_datetime(trades["Session"]).dt.year)["PnL_$"].agg(["count", "sum", "mean"])
    print("\n=== Yearly summary (count/sum/mean) ===")
    print(by_year.to_string())

    # Console overall
    total_trades = len(trades)
    total_pnl = trades["PnL_$"].sum()
    win_rate = trades["Win"].mean() * 100.0
    avg_r = trades["R_multiple"].mean()
    median_r = trades["R_multiple"].median()
    avg_win = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
    avg_loss = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()
    pf_denom = abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())
    pf = (trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / pf_denom) if pf_denom > 0 else np.inf

    print("\n=== Overall Totals ===")
    print(f"Trades: {total_trades} | Total PnL ($): {total_pnl:,.2f} | Win%: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Avg Win ($): {0.0 if pd.isna(avg_win) else avg_win:,.2f} | Avg Loss ($): {0.0 if pd.isna(avg_loss) else avg_loss:,.2f}")
    print(f"Profit Factor: {pf:.3f}")

    # Build new summaries and Excel workbook
    summ = build_summaries(trades)

    with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
        summary = pd.DataFrame({
            "Metric": ["TotalTrades", "TotalPnL", "WinRate_%", "AvgR", "MedianR",
                       "AvgWin_$", "AvgLoss_$", "ProfitFactor"],
            "Value":  [total_trades, total_pnl, win_rate, avg_r, median_r, avg_win, avg_loss, pf]
        })
        summary.to_excel(writer, sheet_name="Summary", index=False)

        make_excel_compatible(trades).to_excel(writer, sheet_name="Trades", index=False)
        make_excel_compatible(equity).to_excel(writer, sheet_name="Equity", index=False)

        summ["overall"].to_excel(writer, sheet_name="Stats_Overall", index=False)
        summ["monthly_all"].to_excel(writer, sheet_name="Stats_Monthly_All", index=False)
        summ["monthly_by_year"].to_excel(writer, sheet_name="Stats_Monthly_ByYear", index=False)

        # Calendar matrices (Year x Month)
        summ["cal_pnl"].to_excel(writer, sheet_name="Calendar_PnL")
        summ["cal_winrate"].to_excel(writer, sheet_name="Calendar_WinRate")
        summ["cal_trades"].to_excel(writer, sheet_name="Calendar_Trades")
        summ["cal_avgwin"].to_excel(writer, sheet_name="Calendar_AvgWin")
        summ["cal_avgloss"].to_excel(writer, sheet_name="Calendar_AvgLoss")

    print(f"Wrote Excel workbook: {EXCEL_OUT}")

if __name__ == "__main__":
    main()
```

```python
# aapl_orb_15m_retest_from_parquet.py
# 15m ORB (retest) backtest on a local AAPL parquet file.
# Win-rate filters: VWAP, EMA trend (15m), GAP, SPY confirm, retest-bar quality,
# Daily trend bias (EMA 20/50 on daily), OR width vs daily ATR, RSI(14), ADX(14),
# Opening-volume quantile, Day-of-week; Risk: BE stop + optional ATR(15m) trailing.
# Monthly-by-year stats + calendar matrices (PnL, WinRate, Trades, AvgWin, AvgLoss).

import warnings
warnings.filterwarnings("ignore")

from datetime import timedelta
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
from pandas.api.types import is_datetime64_any_dtype, is_datetime64tz_dtype

# ========= USER SETTINGS =========
PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"   # AAPL file
TICKER = "AAPL"

# If EventAt in your parquet is naive timestamps representing **UTC**, keep True.
# If EventAt is already in **America/New_York**, set this to False.
ASSUME_EVENTAT_IS_UTC = True

LOCAL_TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB parameters
MAX_RETEST_MIN = 120         # wait up to 2h for retest
RETEST_CONFIRM_CLOSE = True  # retest bar must close back in breakout direction
R_MULTIPLES = [1.0, 2.0]     # targets at 1R, 2R

# Risk params
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# ======== EXISTING WIN-RATE FILTERS ========
USE_VWAP_FILTER = True
USE_TREND_FILTER = True         # 15m EMAs
EMA_FAST = 20
EMA_SLOW = 50

USE_GAP_FILTER = True
GAP_MIN = 0.005   # 0.5%
GAP_MAX = 0.03    # 3%

USE_SPY_CONFIRM = False
SPY_PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\SPY.parquet"  # set if USE_SPY_CONFIRM=True

USE_RETEST_WICK_FILTER = True
RETEST_MAX_OPP_WICK_FRAC = 0.40

# ======== NEW WIN-RATE FILTERS ========
# Daily trend bias (on daily closes)
USE_DAILY_TREND_FILTER = True
DAILY_EMA_FAST = 20
DAILY_EMA_SLOW = 50

# Opening-Range width relative to daily ATR (filters chop / blowouts)
USE_OR_WIDTH_ATR_FILTER = True
OR_ATR_MIN = 0.15    # OR width >= 15% of daily ATR
OR_ATR_MAX = 1.50    # OR width <= 150% of daily ATR

# RSI (15m)
USE_RSI_FILTER = True
RSI_LEN = 14
RSI_THRESH_LONG = 50.0
RSI_THRESH_SHORT = 50.0

# ADX (15m)
USE_ADX_FILTER = False
ADX_LEN = 14
ADX_MIN = 18.0

# Opening 15m volume filter (first bar vs rolling quantile)
USE_OPENING_VOL_FILTER = True
OPENING_VOL_LOOKBACK = 60     # sessions
OPENING_VOL_QUANTILE = 0.50   # median; try 0.6–0.7 to be stricter

# Day-of-week filter (Mon=0 .. Fri=5). Disabled by default.
USE_DOW_FILTER = False
ALLOWED_DOW = {1, 2, 3}       # Tue–Thu

# ======== RISK MGMT ========
MOVE_STOP_TO_BE_AT_R = 0.5     # None to disable; move stop to entry at +0.5R
COUNT_BE_AS_WIN = True         # count breakeven exits as wins in stats

# Optional ATR(15m) trailing stop (k * ATR below/above price)
USE_ATR_TRAIL = False
ATR15_LEN = 14
ATR15_MULT = 2.0

# Outputs
TRADES_CSV = "aapl_orb_trades.csv"
EQUITY_CSV = "aapl_orb_equity.csv"
EXCEL_OUT  = "aapl_orb_results.xlsx"

# ========= LOAD & CLEAN =========
def load_symbol_parquet(path: str, symbol_to_keep: str) -> pd.DataFrame:
    last_err = None
    for eng in ("pyarrow", "fastparquet"):
        try:
            df = pd.read_parquet(path, engine=eng)
            break
        except Exception as e:
            last_err = e
            df = None
    if df is None:
        raise SystemExit(f"Failed to read Parquet. Install pyarrow or fastparquet. Last error: {last_err}")

    if "symbol" in df.columns:
        df["symbol"] = df["symbol"].apply(
            lambda x: x.decode("utf-8") if isinstance(x, (bytes, bytearray)) else str(x)
        ).str.strip().str.replace("^b['\"]?|['\"]?$", "", regex=True)
    else:
        df["symbol"] = symbol_to_keep

    if "EventAt" not in df.columns:
        raise ValueError("Expected 'EventAt' column in the parquet.")
    if not is_datetime64_any_dtype(df["EventAt"]):
        df["EventAt"] = pd.to_datetime(df["EventAt"], utc=False, errors="coerce")

    if is_datetime64tz_dtype(df["EventAt"]):
        df["EventAt"] = df["EventAt"].dt.tz_convert(LOCAL_TZ)
    else:
        if ASSUME_EVENTAT_IS_UTC:
            df["EventAt"] = df["EventAt"].dt.tz_localize("UTC").dt.tz_convert(LOCAL_TZ)
        else:
            df["EventAt"] = df["EventAt"].dt.tz_localize(LOCAL_TZ)

    df = df[df["symbol"] == symbol_to_keep].copy()
    df = df.sort_values("EventAt").set_index("EventAt")

    need = {"Open", "High", "Low", "Close"}
    if not need.issubset(df.columns):
        raise ValueError(f"Missing required OHLC columns: {need - set(df.columns)}")

    keep = [c for c in ["Open", "High", "Low", "Close", "Volume", "symbol"] if c in df.columns]
    return df[keep]

# ========= BAR BUILDERS =========
def to_15m(df: pd.DataFrame) -> pd.DataFrame:
    o = df["Open"].resample("15T", label="left", closed="left").first()
    h = df["High"].resample("15T", label="left", closed="left").max()
    l = df["Low"].resample("15T", label="left", closed="left").min()
    c = df["Close"].resample("15T", label="left", closed="left").last()
    v = df["Volume"].resample("15T", label="left", closed="left").sum() if "Volume" in df.columns else None

    parts = {"Open": o, "High": h, "Low": l, "Close": c}
    if v is not None:
        parts["Volume"] = v
    out = pd.concat(parts, axis=1).dropna(how="any")
    out["Ticker"] = TICKER
    out = out.between_time(REG_SESSION_START, REG_SESSION_END)
    return out

def session_key(index_like) -> pd.Series:
    return pd.to_datetime(pd.Index(index_like).date)

def compute_opening_range(df15: pd.DataFrame) -> pd.DataFrame:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    first_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH", "ORL"]] = df.groupby("Session")[["ORH", "ORL"]].ffill()
    return df

# ========= INDICATORS =========
def add_session_vwap(df15: pd.DataFrame) -> pd.DataFrame:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0
    if "Volume" in df.columns:
        df["vwap_num"] = tp * df["Volume"]
        df["vwap_den"] = df["Volume"].replace(0, np.nan)
    else:
        df["vwap_num"] = tp
        df["vwap_den"] = 1.0
    df["VWAP"] = (df.groupby("Session")["vwap_num"].cumsum() /
                  df.groupby("Session")["vwap_den"].cumsum())
    return df.drop(columns=["vwap_num","vwap_den"])

def add_ema_trend(df15: pd.DataFrame, fast=20, slow=50) -> pd.DataFrame:
    df = df15.copy()
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return df

def rsi(series: pd.Series, length=14) -> pd.Series:
    delta = series.diff()
    up = delta.clip(lower=0)
    dn = -delta.clip(upper=0)
    avg_gain = up.ewm(alpha=1/length, adjust=False).mean()
    avg_loss = dn.ewm(alpha=1/length, adjust=False).mean()
    rs = avg_gain / (avg_loss.replace(0, np.nan))
    return 100 - (100 / (1 + rs))

def add_rsi(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = df15.copy()
    df[f"RSI{length}"] = rsi(df["Close"], length)
    return df

def true_range(high, low, prev_close):
    return np.maximum.reduce([
        (high - low).values,
        np.abs(high - prev_close).values,
        np.abs(low - prev_close).values
    ])

def add_atr_15m(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = df15.copy()
    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    df[f"ATR15_{length}"] = tr.ewm(alpha=1/length, adjust=False).mean()
    return df

def add_adx(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = df15.copy()
    up_move = df["High"].diff()
    down_move = df["Low"].diff().abs()
    plus_dm = np.where((up_move > down_move) & (up_move > 0), up_move, 0.0)
    minus_dm = np.where((down_move > up_move) & (df["Low"].diff() < 0), down_move, 0.0)

    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    atr = tr.ewm(alpha=1/length, adjust=False).mean()

    plus_di = 100 * pd.Series(plus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    minus_di = 100 * pd.Series(minus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    dx = (abs(plus_di - minus_di) / (plus_di + minus_di).replace(0, np.nan)) * 100
    adx = dx.ewm(alpha=1/length, adjust=False).mean()
    df[f"ADX{length}"] = adx
    return df

# ======== DAILY FEATURES (for bias & ATR & opening volume) ========
def build_daily_from_15m(df15: pd.DataFrame) -> pd.DataFrame:
    """Aggregate 15m to daily session OHLCV using 'Session' key."""
    d = df15.copy()
    d["Session"] = session_key(d.index)
    daily = d.groupby("Session").agg(
        Open=("Open","first"),
        High=("High","max"),
        Low=("Low","min"),
        Close=("Close","last"),
        Volume=("Volume","sum") if "Volume" in d.columns else ("Close","size")
    )
    return daily

def add_daily_bias_and_atr(df15: pd.DataFrame, ema_fast=20, ema_slow=50, atr_len=14) -> pd.DataFrame:
    """
    Build daily features (EMA bias, ATR, opening 15m volume threshold) and merge ONLY those
    back onto the 15m frame to avoid column overlap with OHLC.
    """
    # 1) Build daily bars from 15m
    dly = build_daily_from_15m(df15)  # columns: Open, High, Low, Close, Volume (if present)

    # 2) Daily EMA bias
    dly[f"EMA_D{ema_fast}"] = dly["Close"].ewm(span=ema_fast, adjust=False).mean()
    dly[f"EMA_D{ema_slow}"] = dly["Close"].ewm(span=ema_slow, adjust=False).mean()

    # 3) Daily ATR
    prev_close = dly["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (dly["High"] - dly["Low"]).values,
        np.abs(dly["High"] - prev_close).values,
        np.abs(dly["Low"]  - prev_close).values
    ]), index=dly.index)
    atr_col = f"ATR_D{atr_len}"
    dly[atr_col] = tr.ewm(alpha=1/atr_len, adjust=False).mean()

    # 4) Opening 15m volume + rolling quantile threshold (if Volume exists)
    if "Volume" in df15.columns:
        open15_vol = df15.groupby(session_key(df15.index))["Volume"].first()
        dly["Open15_Vol"] = open15_vol
        dly["Open15_Vol_Thresh"] = (
            dly["Open15_Vol"]
            .rolling(window=OPENING_VOL_LOOKBACK, min_periods=10)
            .quantile(OPENING_VOL_QUANTILE)
        )

    # 5) Merge ONLY feature columns (no OHLC) back to 15m
    feature_cols = [
        f"EMA_D{ema_fast}",
        f"EMA_D{ema_slow}",
        atr_col,
        "Open15_Vol",
        "Open15_Vol_Thresh",
    ]
    feature_cols = [c for c in feature_cols if c in dly.columns]

    out = df15.copy()
    out["Session"] = session_key(out.index)
    out = out.merge(dly[feature_cols], left_on="Session", right_index=True, how="left")
    return out

# ========= SLIPPAGE =========
def apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps / 10000.0)
    return price * factor if side == "buy" else price / factor

def wick_opposite_fraction(row: pd.Series, direction: str) -> float:
    h, l, o, c = float(row["High"]), float(row["Low"]), float(row.get("Open", np.nan)), float(row["Close"])
    rng = max(h - l, 1e-12)
    if direction == "long":
        opp = min(o, c) - l  # lower wick adverse for long
    else:
        opp = h - max(o, c)  # upper wick adverse for short
    return float(max(opp, 0.0) / rng)

# ========= BACKTEST =========
def backtest_orb_retest(
    df15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE,
    spy15: Optional[pd.DataFrame] = None
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout CLOSE after the opening bar
        after_open = sdf.iloc[1:].copy()
        long_break  = after_open[after_open["Close"] > orh].head(1)
        short_break = after_open[after_open["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            if long_break.index[0] < short_break.index[0]:
                direction, breakout_row = "long", long_break.iloc[0]
            else:
                direction, breakout_row = "short", short_break.iloc[0]
        elif not long_break.empty:
            direction, breakout_row = "long", long_break.iloc[0]
        else:
            direction, breakout_row = "short", short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if window.empty:
            continue

        level = orh if direction == "long" else orl
        touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        # --- Win-rate filters ---
        # DOW filter
        if USE_DOW_FILTER and retest_time.weekday() not in ALLOWED_DOW:
            continue

        # VWAP slope/side filter
        if USE_VWAP_FILTER and "VWAP" in sdf.columns:
            v_prev = sdf.loc[:retest_time, "VWAP"].tail(3).dropna().values
            rising  = (len(v_prev) < 2) or np.all(np.diff(v_prev) >= 0)
            falling = (len(v_prev) < 2) or np.all(np.diff(v_prev) <= 0)
            if direction == "long":
                if not (retest_bar["Close"] > retest_bar["VWAP"] and rising):
                    continue
            else:
                if not (retest_bar["Close"] < retest_bar["VWAP"] and falling):
                    continue

        # 15m EMA trend filter
        if USE_TREND_FILTER and (f"EMA{EMA_FAST}" in sdf.columns) and (f"EMA{EMA_SLOW}" in sdf.columns):
            emaf = retest_bar[f"EMA{EMA_FAST}"]; emas = retest_bar[f"EMA{EMA_SLOW}"]
            if direction == "long" and not (emaf > emas):
                continue
            if direction == "short" and not (emaf < emas):
                continue

        # Daily trend bias
        if USE_DAILY_TREND_FILTER and f"EMA_D{DAILY_EMA_FAST}" in sdf.columns and f"EMA_D{DAILY_EMA_SLOW}" in sdf.columns:
            dfast = sdf[f"EMA_D{DAILY_EMA_FAST}"].iloc[0]
            dslow = sdf[f"EMA_D{DAILY_EMA_SLOW}"].iloc[0]
            if direction == "long" and not (dfast > dslow):
                continue
            if direction == "short" and not (dfast < dslow):
                continue

        # OR width vs daily ATR
        if USE_OR_WIDTH_ATR_FILTER and f"ATR_D14" in sdf.columns:
            or_width = float(orh - orl)
            atr_d = float(sdf["ATR_D14"].iloc[0])
            if not (atr_d > 0):
                continue
            frac = or_width / atr_d
            if not (OR_ATR_MIN <= frac <= OR_ATR_MAX):
                continue

        # RSI filter (15m)
        if USE_RSI_FILTER and f"RSI{RSI_LEN}" in sdf.columns:
            rsi_val = float(retest_bar[f"RSI{RSI_LEN}"])
            if direction == "long" and not (rsi_val >= RSI_THRESH_LONG):
                continue
            if direction == "short" and not (rsi_val <= 100 - RSI_THRESH_SHORT):  # symmetric
                continue

        # ADX trend-strength filter (15m)
        if USE_ADX_FILTER and f"ADX{ADX_LEN}" in sdf.columns:
            adx_val = float(retest_bar[f"ADX{ADX_LEN}"])
            if not (adx_val >= ADX_MIN):
                continue

        # Gap filter
        if USE_GAP_FILTER and "GapPct" in sdf.columns:
            gap = sdf.loc[sdf.index.min(), "GapPct"]
            if pd.notna(gap) and not (GAP_MIN <= abs(float(gap)) <= GAP_MAX):
                continue

        # Opening 15m volume filter
        if USE_OPENING_VOL_FILTER and "Open15_Vol_Thresh" in sdf.columns and "Open15_Vol" in sdf.columns:
            thresh = sdf["Open15_Vol_Thresh"].iloc[0]
            firstv = sdf["Open15_Vol"].iloc[0]
            if pd.notna(thresh) and pd.notna(firstv):
                if not (firstv >= thresh):
                    continue

        # SPY confirmation
        if USE_SPY_CONFIRM and spy15 is not None and retest_time in spy15.index:
            spy_row = spy15.loc[retest_time]
            if direction == "long" and not (spy_row["SPY_Close"] > spy_row["SPY_VWAP"]):
                continue
            if direction == "short" and not (spy_row["SPY_Close"] < spy_row["SPY_VWAP"]):
                continue

        # Retest wick quality
        if USE_RETEST_WICK_FILTER:
            opp_frac = wick_opposite_fraction(retest_bar, direction)
            if opp_frac > RETEST_MAX_OPP_WICK_FRAC:
                continue
        # --- end filters ---

        entry = apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1

        targets = [(entry + r * risk_per_share) if direction == "long" else (entry - r * risk_per_share)
                   for r in r_targets]

        # Walk forward for exits
        run = sdf[sdf.index >= retest_time].copy()
        exits = []
        be_armed = (MOVE_STOP_TO_BE_AT_R is not None)
        be_level = entry
        rps = risk_per_share

        for ts, row in run.iterrows():
            hi, lo = row["High"], row["Low"]

            # Arm BE at +X R
            if be_armed:
                if direction == "long" and hi >= entry + (MOVE_STOP_TO_BE_AT_R * rps):
                    stop = max(stop, be_level)
                    be_armed = False
                elif direction == "short" and lo <= entry - (MOVE_STOP_TO_BE_AT_R * rps):
                    stop = min(stop, be_level)
                    be_armed = False

            # ATR(15m) trailing stop (optional)
            if USE_ATR_TRAIL and f"ATR15_{ATR15_LEN}" in sdf.columns:
                atr_now = float(row[f"ATR15_{ATR15_LEN}"])
                if direction == "long":
                    trail = hi - ATR15_MULT * atr_now
                    stop = max(stop, trail)
                else:
                    trail = lo + ATR15_MULT * atr_now
                    stop = min(stop, trail)

            # Stop first
            if lo <= stop <= hi:
                px = apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                label = "breakeven" if abs(px - be_level) < 1e-10 else "stop"
                exits.append((label, ts, px))
                break

            # Targets (first hit ends trade)
            hit = False
            for i, tp in enumerate(targets):
                if lo <= tp <= hi:
                    px = apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = run.iloc[-1]
            px = apply_slippage(last["Close"], slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _, _, px in exits) - fees

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "Direction": direction,
            "ORH": orh, "ORL": orl,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": targets,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        equity = pd.DataFrame(columns=["Session", "CumPnL_$"])
        return trade_log, equity

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$": "CumPnL_$"}))
    return trade_log, equity

# ========= EXCEL SAFETY =========
def make_excel_compatible(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    for col in out.columns:
        if is_datetime64tz_dtype(out[col]):
            out[col] = out[col].dt.tz_convert(LOCAL_TZ).dt.tz_localize(None)
    return out

# ========= SUMMARY HELPERS =========
def _agg_summary(g: pd.DataFrame) -> pd.Series:
    pnl = pd.to_numeric(g["PnL_$"], errors="coerce").dropna()
    trades = len(pnl)
    wins_mask = pnl > 0
    losses_mask = pnl < 0
    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()

    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf
    else:
        pf = np.nan

    return pd.Series({
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": int(wins_mask.sum()),
        "Losses": int(losses_mask.sum()),
        "WinRate": (wins_mask.mean() * 100.0) if trades else np.nan,
        "AvgWin": pnl[wins_mask].mean() if wins_mask.any() else np.nan,
        "AvgLoss": pnl[losses_mask].mean() if losses_mask.any() else np.nan,
        "AvgR": g["R_multiple"].mean(),
        "MedianR": g["R_multiple"].median(),
        "PF": pf,
    })

MONTH_ABBR = {1:"Jan",2:"Feb",3:"Mar",4:"Apr",5:"May",6:"Jun",
              7:"Jul",8:"Aug",9:"Sep",10:"Oct",11:"Nov",12:"Dec"}

def _calendar_matrix(df_monthly_by_year: pd.DataFrame, value_col: str) -> pd.DataFrame:
    mat = (df_monthly_by_year
           .pivot(index="Year", columns="MonthNum", values=value_col)
           .reindex(columns=range(1, 13)))
    mat.columns = [MONTH_ABBR[m] for m in mat.columns]
    return mat.sort_index()

def build_summaries(trades: pd.DataFrame):
    overall = _agg_summary(trades).to_frame().T

    ses = pd.to_datetime(trades["Session"])
    trades = trades.copy()
    trades["_MonthStr"] = ses.dt.to_period("M").astype(str)  # 'YYYY-MM'
    monthly_all = (trades.groupby("_MonthStr").apply(_agg_summary)
                          .reset_index()
                          .rename(columns={"_MonthStr": "Month"})
                          .sort_values("Month"))

    trades["_Year"] = ses.dt.year
    trades["_MonthNum"] = ses.dt.month
    trades["_MonthName"] = trades["_MonthNum"].map(MONTH_ABBR)
    monthly_by_year = (trades.groupby(["_Year", "_MonthNum", "_MonthName"])
                              .apply(_agg_summary)
                              .reset_index()
                              .rename(columns={"_Year":"Year","_MonthNum":"MonthNum","_MonthName":"Month"}))
    monthly_by_year = monthly_by_year.sort_values(["Year","MonthNum"])

    cal_pnl     = _calendar_matrix(monthly_by_year, "PnL_Sum")
    cal_winrate = _calendar_matrix(monthly_by_year, "WinRate")
    cal_trades  = _calendar_matrix(monthly_by_year, "Trades")
    cal_avgwin  = _calendar_matrix(monthly_by_year, "AvgWin")
    cal_avgloss = _calendar_matrix(monthly_by_year, "AvgLoss")

    to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
    for tbl in (overall, monthly_all, monthly_by_year):
        for c in to_round:
            if c in tbl.columns:
                tbl[c] = tbl[c].astype(float).round(4)
    for mat in (cal_pnl, cal_winrate, cal_trades, cal_avgwin, cal_avgloss):
        mat[:] = np.round(mat.astype(float), 4)

    return {
        "overall": overall,
        "monthly_all": monthly_all,
        "monthly_by_year": monthly_by_year,
        "cal_pnl": cal_pnl,
        "cal_winrate": cal_winrate,
        "cal_trades": cal_trades,
        "cal_avgwin": cal_avgwin,
        "cal_avgloss": cal_avgloss,
    }

# ========= MAIN =========
def main():
    print("Loading parquet...")
    raw = load_symbol_parquet(PATH, TICKER)
    if raw.empty:
        print(f"No rows found in the parquet after filtering for {TICKER}.")
        return

    print("Building 15m bars...")
    df15 = to_15m(raw)

    print("Computing opening ranges & adding features...")
    df15 = compute_opening_range(df15)
    df15 = add_session_vwap(df15) if USE_VWAP_FILTER else df15
    df15 = add_ema_trend(df15, EMA_FAST, EMA_SLOW) if USE_TREND_FILTER else df15

    # Daily bias + ATR + opening-vol threshold merged down to 15m
    df15 = add_daily_bias_and_atr(df15, DAILY_EMA_FAST, DAILY_EMA_SLOW, 14)

    # Gap feature (uses daily prev close inside)
    # Build GapPct based on first 15m vs prev day's close
    df15["Session"] = session_key(df15.index)
    first_idx = df15.groupby("Session").head(1).index
    prev_close_series = df15.groupby("Session")["Close"].last().shift(1)
    prev_close_map = df15["Session"].map(prev_close_series)
    df15.loc[first_idx, "GapPct"] = (df15.loc[first_idx, "Open"] / prev_close_map.loc[first_idx] - 1.0)
    df15["GapPct"] = df15.groupby("Session")["GapPct"].ffill()

    # RSI / ADX / ATR15 for trailing
    df15 = add_rsi(df15, RSI_LEN) if USE_RSI_FILTER else df15
    df15 = add_adx(df15, ADX_LEN) if USE_ADX_FILTER else df15
    df15 = add_atr_15m(df15, ATR15_LEN) if USE_ATR_TRAIL else df15

    spy15 = None
    if USE_SPY_CONFIRM:
        spy_raw = load_symbol_parquet(SPY_PATH, "SPY")
        spy15 = to_15m(spy_raw)
        spy15 = compute_opening_range(spy15)
        spy15 = add_session_vwap(spy15)
        spy15 = spy15.rename(columns={"VWAP":"SPY_VWAP", "Close":"SPY_Close"})
        spy15 = spy15[["SPY_VWAP", "SPY_Close"]]

    print("Backtesting...")
    trades, equity = backtest_orb_retest(df15, spy15=spy15)
    if trades.empty:
        print("No trades generated with current settings.")
        return

    # Save CSVs
    out_trades = trades.copy()
    out_trades["Targets"] = out_trades["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
    out_trades["Exits"]   = out_trades["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
    out_trades.to_csv(TRADES_CSV, index=False)
    equity.to_csv(EQUITY_CSV, index=False)
    print(f"Saved: {TRADES_CSV}, {EQUITY_CSV}")

    # Console yearly
    trades["Win"] = (trades["PnL_$"] >= 0) if COUNT_BE_AS_WIN else (trades["PnL_$"] > 0)
    by_year = trades.groupby(pd.to_datetime(trades["Session"]).dt.year)["PnL_$"].agg(["count", "sum", "mean"])
    print("\n=== Yearly summary (count/sum/mean) ===")
    print(by_year.to_string())

    # Console overall
    total_trades = len(trades)
    total_pnl = trades["PnL_$"].sum()
    win_rate = trades["Win"].mean() * 100.0
    avg_r = trades["R_multiple"].mean()
    median_r = trades["R_multiple"].median()
    avg_win = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
    avg_loss = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()
    pos_sum = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum()
    neg_sum = trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum()
    pf = (pos_sum / abs(neg_sum)) if neg_sum < 0 else np.inf

    print("\n=== Overall Totals ===")
    print(f"Trades: {total_trades} | Total PnL ($): {total_pnl:,.2f} | Win%: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Avg Win ($): {0.0 if pd.isna(avg_win) else avg_win:,.2f} | Avg Loss ($): {0.0 if pd.isna(avg_loss) else avg_loss:,.2f}")
    print(f"Profit Factor: {pf:.3f}")

    # Build summaries
    summ = build_summaries(trades)

    # Excel workbook
    with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
        summary = pd.DataFrame({
            "Metric": ["TotalTrades", "TotalPnL", "WinRate_%", "AvgR", "MedianR",
                       "AvgWin_$", "AvgLoss_$", "ProfitFactor"],
            "Value":  [total_trades, total_pnl, win_rate, avg_r, median_r, avg_win, avg_loss, pf]
        })
        summary.to_excel(writer, sheet_name="Summary", index=False)

        make_excel_compatible(trades).to_excel(writer, sheet_name="Trades", index=False)
        make_excel_compatible(equity).to_excel(writer, sheet_name="Equity", index=False)

        summ["overall"].to_excel(writer, sheet_name="Stats_Overall", index=False)
        summ["monthly_all"].to_excel(writer, sheet_name="Stats_Monthly_All", index=False)
        summ["monthly_by_year"].to_excel(writer, sheet_name="Stats_Monthly_ByYear", index=False)

        summ["cal_pnl"].to_excel(writer, sheet_name="Calendar_PnL")
        summ["cal_winrate"].to_excel(writer, sheet_name="Calendar_WinRate")
        summ["cal_trades"].to_excel(writer, sheet_name="Calendar_Trades")
        summ["cal_avgwin"].to_excel(writer, sheet_name="Calendar_AvgWin")
        summ["cal_avgloss"].to_excel(writer, sheet_name="Calendar_AvgLoss")

    print(f"Wrote Excel workbook: {EXCEL_OUT}")

if __name__ == "__main__":
    main()
```

```python
# aapl_orb_15m_retest_from_parquet.py
# 15m ORB (retest) backtest on a local AAPL parquet file
# Filters: VWAP, EMA trend (15m), GAP, SPY confirm, retest-bar quality,
# Daily trend bias (EMA 20/50 on daily), OR width vs daily ATR, RSI(14), ADX(14),
# Opening-volume quantile, Day-of-week; Risk: BE stop + optional ATR(15m) trailing.
# Win-rate tactics: scale-out (TP1/TP2), gap-direction alignment, time windows,
# momentum continuation fallback (if no retest appears).
# Exports monthly-by-year stats + calendar matrices (PnL, WinRate, Trades, AvgWin, AvgLoss).

import warnings
warnings.filterwarnings("ignore")

from datetime import timedelta, time as dtime
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
from pandas.api.types import is_datetime64_any_dtype, is_datetime64tz_dtype

# ========= USER SETTINGS =========
PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"   # AAPL file
TICKER = "AAPL"

# If EventAt in your parquet is naive timestamps representing **UTC**, keep True.
# If EventAt is already in **America/New_York**, set this to False.
ASSUME_EVENTAT_IS_UTC = True

LOCAL_TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB parameters
MAX_RETEST_MIN = 120         # wait up to 2h for retest
RETEST_CONFIRM_CLOSE = True  # retest bar must close back in breakout direction
R_MULTIPLES = [1.0, 2.0]     # targets at 1R, 2R (used if scale-out disabled)

# Risk params
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# ======== EXISTING WIN-RATE FILTERS ========
USE_VWAP_FILTER = True
USE_TREND_FILTER = True         # 15m EMAs
EMA_FAST = 20
EMA_SLOW = 50

USE_GAP_FILTER = True
GAP_MIN = 0.005   # 0.5%
GAP_MAX = 0.03    # 3%

USE_SPY_CONFIRM = False
SPY_PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\SPY.parquet"  # set if USE_SPY_CONFIRM=True

USE_RETEST_WICK_FILTER = True
RETEST_MAX_OPP_WICK_FRAC = 0.40

# ======== NEW WIN-RATE FILTERS ========
# Daily trend bias (on daily closes)
USE_DAILY_TREND_FILTER = True
DAILY_EMA_FAST = 20
DAILY_EMA_SLOW = 50
DAILY_ATR_LEN  = 14  # for OR/ATR filter and daily ATR feature

# Opening-Range width relative to daily ATR (filters chop / blowouts)
USE_OR_WIDTH_ATR_FILTER = True
OR_ATR_MIN = 0.15    # OR width >= 15% of daily ATR
OR_ATR_MAX = 1.50    # OR width <= 150% of daily ATR

# RSI (15m)
USE_RSI_FILTER = True
RSI_LEN = 14
RSI_THRESH_LONG = 50.0
RSI_THRESH_SHORT = 50.0

# ADX (15m)
USE_ADX_FILTER = False
ADX_LEN = 14
ADX_MIN = 18.0

# Opening 15m volume filter (first bar vs rolling quantile)
USE_OPENING_VOL_FILTER = True
OPENING_VOL_LOOKBACK = 60     # sessions
OPENING_VOL_QUANTILE = 0.50   # median; try 0.6–0.7 to be stricter

# Day-of-week filter (Mon=0 .. Fri=4). Disabled by default.
USE_DOW_FILTER = False
ALLOWED_DOW = {1, 2, 3}       # Tue–Thu

# ======== RISK MGMT ========
MOVE_STOP_TO_BE_AT_R = 0.5     # None to disable; move stop to entry at +0.5R
COUNT_BE_AS_WIN = True         # count breakeven exits as wins in stats

# Optional ATR(15m) trailing stop (k * ATR below/above price)
USE_ATR_TRAIL = False
ATR15_LEN = 14
ATR15_MULT = 2.0

# ======== WIN-RATE TACTICS ========
# Scale-out: take partial at TP1, rest at TP2 (then optional trail/eod)
USE_SCALE_OUT = True
SCALE_SPLIT = (0.5, 0.5)     # 50% at TP1, 50% runs to TP2
TP1_R = 0.75
TP2_R = 1.50

# Align trade direction with gap direction
ALIGN_WITH_GAP_DIR = True     # long only if gap >= 0; short only if gap <= 0

# Time windows (ET)
BREAKOUT_DEADLINE = "10:30"   # breakout must occur by this time
RETEST_DEADLINE   = "11:30"   # retest must complete by this time
CONTINUATION_CUTOFF = "11:00" # continuation entries allowed up to this time

# Momentum continuation (fallback if no retest trade appears)
USE_CONTINUATION = True       # enters at breakout close w/ same stop (OR opposite)

# Outputs
TRADES_CSV = "aapl_orb_trades.csv"
EQUITY_CSV = "aapl_orb_equity.csv"
EXCEL_OUT  = "aapl_orb_results.xlsx"

# ========= LOAD & CLEAN =========
def load_symbol_parquet(path: str, symbol_to_keep: str) -> pd.DataFrame:
    last_err = None
    for eng in ("pyarrow", "fastparquet"):
        try:
            df = pd.read_parquet(path, engine=eng)
            break
        except Exception as e:
            last_err = e
            df = None
    if df is None:
        raise SystemExit(f"Failed to read Parquet. Install pyarrow or fastparquet. Last error: {last_err}")

    if "symbol" in df.columns:
        df["symbol"] = df["symbol"].apply(
            lambda x: x.decode("utf-8") if isinstance(x, (bytes, bytearray)) else str(x)
        ).str.strip().str.replace("^b['\"]?|['\"]?$", "", regex=True)
    else:
        df["symbol"] = symbol_to_keep

    if "EventAt" not in df.columns:
        raise ValueError("Expected 'EventAt' column in the parquet.")
    if not is_datetime64_any_dtype(df["EventAt"]):
        df["EventAt"] = pd.to_datetime(df["EventAt"], utc=False, errors="coerce")

    if is_datetime64tz_dtype(df["EventAt"]):
        df["EventAt"] = df["EventAt"].dt.tz_convert(LOCAL_TZ)
    else:
        if ASSUME_EVENTAT_IS_UTC:
            df["EventAt"] = df["EventAt"].dt.tz_localize("UTC").dt.tz_convert(LOCAL_TZ)
        else:
            df["EventAt"] = df["EventAt"].dt.tz_localize(LOCAL_TZ)

    df = df[df["symbol"] == symbol_to_keep].copy()
    df = df.sort_values("EventAt").set_index("EventAt")

    need = {"Open", "High", "Low", "Close"}
    if not need.issubset(df.columns):
        raise ValueError(f"Missing required OHLC columns: {need - set(df.columns)}")

    keep = [c for c in ["Open", "High", "Low", "Close", "Volume", "symbol"] if c in df.columns]
    return df[keep]

# ========= BAR BUILDERS =========
def to_15m(df: pd.DataFrame) -> pd.DataFrame:
    o = df["Open"].resample("15T", label="left", closed="left").first()
    h = df["High"].resample("15T", label="left", closed="left").max()
    l = df["Low"].resample("15T", label="left", closed="left").min()
    c = df["Close"].resample("15T", label="left", closed="left").last()
    v = df["Volume"].resample("15T", label="left", closed="left").sum() if "Volume" in df.columns else None

    parts = {"Open": o, "High": h, "Low": l, "Close": c}
    if v is not None:
        parts["Volume"] = v
    out = pd.concat(parts, axis=1).dropna(how="any")
    out["Ticker"] = TICKER
    out = out.between_time(REG_SESSION_START, REG_SESSION_END)
    return out

def session_key(index_like) -> pd.Series:
    return pd.to_datetime(pd.Index(index_like).date)

def compute_opening_range(df15: pd.DataFrame) -> pd.DataFrame:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    first_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH", "ORL"]] = df.groupby("Session")[["ORH", "ORL"]].ffill()
    return df

# ========= INDICATORS =========
def add_session_vwap(df15: pd.DataFrame) -> pd.DataFrame:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0
    if "Volume" in df.columns:
        df["vwap_num"] = tp * df["Volume"]
        df["vwap_den"] = df["Volume"].replace(0, np.nan)
    else:
        df["vwap_num"] = tp
        df["vwap_den"] = 1.0
    df["VWAP"] = (df.groupby("Session")["vwap_num"].cumsum() /
                  df.groupby("Session")["vwap_den"].cumsum())
    return df.drop(columns=["vwap_num","vwap_den"])

def add_ema_trend(df15: pd.DataFrame, fast=20, slow=50) -> pd.DataFrame:
    df = df15.copy()
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return df

def rsi(series: pd.Series, length=14) -> pd.Series:
    delta = series.diff()
    up = delta.clip(lower=0)
    dn = -delta.clip(upper=0)
    avg_gain = up.ewm(alpha=1/length, adjust=False).mean()
    avg_loss = dn.ewm(alpha=1/length, adjust=False).mean()
    rs = avg_gain / (avg_loss.replace(0, np.nan))
    return 100 - (100 / (1 + rs))

def add_rsi(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = df15.copy()
    df[f"RSI{length}"] = rsi(df["Close"], length)
    return df

def true_range(high, low, prev_close):
    return np.maximum.reduce([
        (high - low).values,
        np.abs(high - prev_close).values,
        np.abs(low - prev_close).values
    ])

def add_atr_15m(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = df15.copy()
    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    df[f"ATR15_{length}"] = tr.ewm(alpha=1/length, adjust=False).mean()
    return df

def add_adx(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = df15.copy()
    up_move = df["High"].diff()
    dn_move = -df["Low"].diff()
    plus_dm  = np.where((up_move > dn_move) & (up_move > 0), up_move, 0.0)
    minus_dm = np.where((dn_move > up_move) & (dn_move > 0), dn_move, 0.0)

    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    atr = tr.ewm(alpha=1/length, adjust=False).mean()

    plus_di = 100 * pd.Series(plus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    minus_di = 100 * pd.Series(minus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    dx = (abs(plus_di - minus_di) / (plus_di + minus_di).replace(0, np.nan)) * 100
    adx = dx.ewm(alpha=1/length, adjust=False).mean()
    df[f"ADX{length}"] = adx
    return df

# ======== DAILY FEATURES (for bias & ATR & opening volume) ========
def build_daily_from_15m(df15: pd.DataFrame) -> pd.DataFrame:
    """Aggregate 15m to daily session OHLCV using 'Session' key."""
    d = df15.copy()
    d["Session"] = session_key(d.index)
    daily = d.groupby("Session").agg(
        Open=("Open","first"),
        High=("High","max"),
        Low=("Low","min"),
        Close=("Close","last"),
        Volume=("Volume","sum") if "Volume" in d.columns else ("Close","size")
    )
    return daily

def add_daily_bias_and_atr(df15: pd.DataFrame, ema_fast=20, ema_slow=50, atr_len=14) -> pd.DataFrame:
    """
    Build daily features (EMA bias, ATR, opening 15m volume threshold) and merge ONLY those
    back onto the 15m frame to avoid column overlap with OHLC.
    """
    dly = build_daily_from_15m(df15)  # Open,High,Low,Close,Volume

    # Daily EMA bias
    dly[f"EMA_D{ema_fast}"] = dly["Close"].ewm(span=ema_fast, adjust=False).mean()
    dly[f"EMA_D{ema_slow}"] = dly["Close"].ewm(span=ema_slow, adjust=False).mean()

    # Daily ATR
    prev_close = dly["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (dly["High"] - dly["Low"]).values,
        np.abs(dly["High"] - prev_close).values,
        np.abs(dly["Low"]  - prev_close).values
    ]), index=dly.index)
    atr_col = f"ATR_D{atr_len}"
    dly[atr_col] = tr.ewm(alpha=1/atr_len, adjust=False).mean()

    # Opening 15m volume + rolling quantile threshold (if Volume exists)
    if "Volume" in df15.columns:
        open15_vol = df15.groupby(session_key(df15.index))["Volume"].first()
        dly["Open15_Vol"] = open15_vol
        dly["Open15_Vol_Thresh"] = (
            dly["Open15_Vol"]
            .rolling(window=OPENING_VOL_LOOKBACK, min_periods=10)
            .quantile(OPENING_VOL_QUANTILE)
        )

    # Merge ONLY feature columns
    feature_cols = [
        f"EMA_D{ema_fast}",
        f"EMA_D{ema_slow}",
        atr_col,
        "Open15_Vol",
        "Open15_Vol_Thresh",
    ]
    feature_cols = [c for c in feature_cols if c in dly.columns]

    out = df15.copy()
    out["Session"] = session_key(out.index)
    out = out.merge(dly[feature_cols], left_on="Session", right_index=True, how="left")
    return out

# ========= UTILS =========
def apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps / 10000.0)
    return price * factor if side == "buy" else price / factor

def wick_opposite_fraction(row: pd.Series, direction: str) -> float:
    h, l, o, c = float(row["High"]), float(row["Low"]), float(row.get("Open", np.nan)), float(row["Close"])
    rng = max(h - l, 1e-12)
    if direction == "long":
        opp = min(o, c) - l  # lower wick adverse for long
    else:
        opp = h - max(o, c)  # upper wick adverse for short
    return float(max(opp, 0.0) / rng)

def _to_clock(ts):  # naive local clock from tz-aware or naive ts
    return dtime(ts.hour, ts.minute)

def _clock_le(ts, hhmm: str) -> bool:
    h, m = map(int, hhmm.split(":"))
    return _to_clock(ts) <= dtime(h, m)

# ========= BACKTEST =========
def backtest_orb_retest(
    df15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE,
    spy15: Optional[pd.DataFrame] = None
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df15.copy()
    df["Session"] = session_key(df.index)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout CLOSE after the opening bar
        after_open = sdf.iloc[1:].copy()
        long_break  = after_open[after_open["Close"] > orh].head(1)
        short_break = after_open[after_open["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            # no breakout this day
            continue

        # Which side broke first?
        if not long_break.empty and not short_break.empty:
            if long_break.index[0] < short_break.index[0]:
                direction, breakout_row = "long", long_break.iloc[0]
            else:
                direction, breakout_row = "short", short_break.iloc[0]
        elif not long_break.empty:
            direction, breakout_row = "long", long_break.iloc[0]
        else:
            direction, breakout_row = "short", short_break.iloc[0]

        btime = breakout_row.name  # breakout bar close time

        # Look for retest within window
        cutoff = btime + timedelta(minutes=max_retest_min)
        window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        level = orh if direction == "long" else orl
        touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)

        # Choose entry bar/time:
        # - default: retest bar (if exists)
        # - else if continuation enabled: enter at breakout close (btime)
        # - else: skip
        retest_used = True
        if touch.empty:
            if USE_CONTINUATION and _clock_le(btime, CONTINUATION_CUTOFF):
                retest_used = False
                retest_time = btime
                retest_bar = breakout_row
            else:
                continue
        else:
            retest_bar = touch.iloc[0]
            retest_time = retest_bar.name

        # Retest confirmation (close must re-accept beyond level) — only for retests
        if retest_used and retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        # --- Win-rate filters -------------------------------------------------
        # Day-of-week
        if USE_DOW_FILTER and retest_time.weekday() not in ALLOWED_DOW:
            continue

        # VWAP slope/side
        if USE_VWAP_FILTER and "VWAP" in sdf.columns:
            v_prev = sdf.loc[:retest_time, "VWAP"].tail(3).dropna().values
            rising  = (len(v_prev) < 2) or np.all(np.diff(v_prev) >= 0)
            falling = (len(v_prev) < 2) or np.all(np.diff(v_prev) <= 0)
            if direction == "long":
                if not (retest_bar["Close"] > retest_bar["VWAP"] and rising):
                    continue
            else:
                if not (retest_bar["Close"] < retest_bar["VWAP"] and falling):
                    continue

        # 15m EMA trend
        if USE_TREND_FILTER and (f"EMA{EMA_FAST}" in sdf.columns) and (f"EMA{EMA_SLOW}" in sdf.columns):
            emaf = retest_bar[f"EMA{EMA_FAST}"]; emas = retest_bar[f"EMA{EMA_SLOW}"]
            if direction == "long" and not (emaf > emas):
                continue
            if direction == "short" and not (emaf < emas):
                continue

        # Daily trend bias
        if USE_DAILY_TREND_FILTER and f"EMA_D{DAILY_EMA_FAST}" in sdf.columns and f"EMA_D{DAILY_EMA_SLOW}" in sdf.columns:
            dfast = sdf[f"EMA_D{DAILY_EMA_FAST}"].iloc[0]
            dslow = sdf[f"EMA_D{DAILY_EMA_SLOW}"].iloc[0]
            if direction == "long" and not (dfast > dslow):
                continue
            if direction == "short" and not (dfast < dslow):
                continue

        # OR width vs daily ATR
        atr_col = f"ATR_D{DAILY_ATR_LEN}"
        if USE_OR_WIDTH_ATR_FILTER and atr_col in sdf.columns:
            or_width = float(orh - orl)
            atr_d = float(sdf[atr_col].iloc[0])
            if not (atr_d > 0):
                continue
            frac = or_width / atr_d
            if not (OR_ATR_MIN <= frac <= OR_ATR_MAX):
                continue

        # RSI filter (15m)
        if USE_RSI_FILTER and f"RSI{RSI_LEN}" in sdf.columns:
            rsi_val = float(retest_bar[f"RSI{RSI_LEN}"])
            if direction == "long" and not (rsi_val >= RSI_THRESH_LONG):
                continue
            if direction == "short" and not (rsi_val <= 100 - RSI_THRESH_SHORT):
                continue

        # ADX trend-strength filter (15m)
        if USE_ADX_FILTER and f"ADX{ADX_LEN}" in sdf.columns:
            adx_val = float(retest_bar[f"ADX{ADX_LEN}"])
            if not (adx_val >= ADX_MIN):
                continue

        # Gap filter (magnitude)
        if USE_GAP_FILTER and "GapPct" in sdf.columns:
            gap = sdf.loc[sdf.index.min(), "GapPct"]
            if pd.notna(gap) and not (GAP_MIN <= abs(float(gap)) <= GAP_MAX):
                continue

        # Gap-direction alignment
        if ALIGN_WITH_GAP_DIR and "GapPct" in sdf.columns:
            g = float(sdf["GapPct"].iloc[0]) if pd.notna(sdf["GapPct"].iloc[0]) else 0.0
            if (direction == "long" and g < 0) or (direction == "short" and g > 0):
                continue

        # SPY confirmation
        if USE_SPY_CONFIRM and spy15 is not None and retest_time in spy15.index:
            spy_row = spy15.loc[retest_time]
            if direction == "long" and not (spy_row["SPY_Close"] > spy_row["SPY_VWAP"]):
                continue
            if direction == "short" and not (spy_row["SPY_Close"] < spy_row["SPY_VWAP"]):
                continue

        # Retest wick quality (use the entry bar in either case)
        if USE_RETEST_WICK_FILTER:
            opp_frac = wick_opposite_fraction(retest_bar, direction)
            if opp_frac > RETEST_MAX_OPP_WICK_FRAC:
                continue

        # Time windows
        if not _clock_le(btime, BREAKOUT_DEADLINE):
            continue
        if retest_used and (not _clock_le(retest_time, RETEST_DEADLINE)):
            continue
        if (not retest_used) and (not _clock_le(retest_time, CONTINUATION_CUTOFF)):
            continue
        # ---------------------------------------------------------------------

        # Entry, stop, risk
        entry = apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1

        # Targets for scale-out or single exit
        if USE_SCALE_OUT:
            targets_r = [TP1_R, TP2_R]
        else:
            targets_r = r_targets
        targets = [(entry + r * risk_per_share) if direction == "long" else (entry - r * risk_per_share)
                   for r in targets_r]

        # === Walk forward with scale-out + BE stop ===========================
        run = sdf[sdf.index >= retest_time].copy()
        qty_left = qty
        exits = []
        be_armed = (MOVE_STOP_TO_BE_AT_R is not None)
        be_level = entry
        rps = risk_per_share
        next_tp_idx = 0  # which target to aim for next

        # position sizing per leg
        if USE_SCALE_OUT and len(targets) >= 2:
            leg_sizes = [qty * SCALE_SPLIT[0], qty * SCALE_SPLIT[1]]
        else:
            leg_sizes = [qty]

        for ts, row in run.iterrows():
            hi, lo = row["High"], row["Low"]

            # Arm BE at +X R
            if be_armed:
                if direction == "long" and hi >= entry + (MOVE_STOP_TO_BE_AT_R * rps):
                    stop = max(stop, be_level)
                    be_armed = False
                elif direction == "short" and lo <= entry - (MOVE_STOP_TO_BE_AT_R * rps):
                    stop = min(stop, be_level)
                    be_armed = False

            # ATR(15m) trailing stop (optional)
            if USE_ATR_TRAIL and f"ATR15_{ATR15_LEN}" in sdf.columns:
                atr_now = float(row[f"ATR15_{ATR15_LEN}"])
                if direction == "long":
                    trail = hi - ATR15_MULT * atr_now
                    stop = max(stop, trail)
                else:
                    trail = lo + ATR15_MULT * atr_now
                    stop = min(stop, trail)

            # Stop first
            if lo <= stop <= hi:
                px = apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                label = "breakeven" if abs(px - be_level) < 1e-10 else "stop"
                exits.append((label, ts, px, qty_left))
                qty_left = 0
                break

            # Hit next target?
            if next_tp_idx < len(targets):
                tp = targets[next_tp_idx]
                if lo <= tp <= hi:
                    px = apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    fill_qty = leg_sizes[next_tp_idx] if next_tp_idx < len(leg_sizes) else qty_left
                    exits.append((f"tp{next_tp_idx+1}", ts, px, fill_qty))
                    qty_left -= fill_qty
                    next_tp_idx += 1

                    # After TP1, move stop to BE to protect remainder
                    if USE_SCALE_OUT and next_tp_idx == 1 and MOVE_STOP_TO_BE_AT_R is not None:
                        stop = be_level

                    if qty_left <= 1e-9:
                        break

            # Optional: if still no TP1 by continuation cutoff, tighten to BE
            if USE_CONTINUATION and not _clock_le(ts, CONTINUATION_CUTOFF) and next_tp_idx == 0:
                stop = max(stop, be_level) if direction == "long" else min(stop, be_level)

        # If still open at end-of-day
        if qty_left > 1e-9:
            last = run.iloc[-1]
            px = apply_slippage(last["Close"], slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", run.index[-1], px, qty_left))
            qty_left = 0

        # PnL from exits
        cash_pnl = sum((px - entry) * side_mult * q for (_, _, px, q) in exits) - fees

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "Direction": direction,
            "ORH": orh, "ORL": orl,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": targets,
            "Exits": [(lab, ts, px) for (lab, ts, px, _) in exits],
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * max(qty, 1e-12))
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        equity = pd.DataFrame(columns=["Session", "CumPnL_$"])
        return trade_log, equity

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$": "CumPnL_$"}))
    return trade_log, equity

# ========= EXCEL SAFETY =========
def make_excel_compatible(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    for col in out.columns:
        if is_datetime64tz_dtype(out[col]):
            out[col] = out[col].dt.tz_convert(LOCAL_TZ).dt.tz_localize(None)
    return out

# ========= SUMMARY HELPERS =========
def _agg_summary(g: pd.DataFrame) -> pd.Series:
    pnl = pd.to_numeric(g["PnL_$"], errors="coerce").dropna()
    trades = len(pnl)
    wins_mask = pnl > 0
    losses_mask = pnl < 0
    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()

    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf
    else:
        pf = np.nan

    return pd.Series({
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": int(wins_mask.sum()),
        "Losses": int(losses_mask.sum()),
        "WinRate": (wins_mask.mean() * 100.0) if trades else np.nan,
        "AvgWin": pnl[wins_mask].mean() if wins_mask.any() else np.nan,
        "AvgLoss": pnl[losses_mask].mean() if losses_mask.any() else np.nan,
        "AvgR": g["R_multiple"].mean(),
        "MedianR": g["R_multiple"].median(),
        "PF": pf,
    })

MONTH_ABBR = {1:"Jan",2:"Feb",3:"Mar",4:"Apr",5:"May",6:"Jun",
              7:"Jul",8:"Aug",9:"Sep",10:"Oct",11:"Nov",12:"Dec"}

def _calendar_matrix(df_monthly_by_year: pd.DataFrame, value_col: str) -> pd.DataFrame:
    mat = (df_monthly_by_year
           .pivot(index="Year", columns="MonthNum", values=value_col)
           .reindex(columns=range(1, 13)))
    mat.columns = [MONTH_ABBR[m] for m in mat.columns]
    return mat.sort_index()

def build_summaries(trades: pd.DataFrame):
    overall = _agg_summary(trades).to_frame().T

    ses = pd.to_datetime(trades["Session"])
    trades = trades.copy()
    trades["_MonthStr"] = ses.dt.to_period("M").astype(str)  # 'YYYY-MM'
    monthly_all = (trades.groupby("_MonthStr").apply(_agg_summary)
                          .reset_index()
                          .rename(columns={"_MonthStr": "Month"})
                          .sort_values("Month"))

    trades["_Year"] = ses.dt.year
    trades["_MonthNum"] = ses.dt.month
    trades["_MonthName"] = trades["_MonthNum"].map(MONTH_ABBR)
    monthly_by_year = (trades.groupby(["_Year", "_MonthNum", "_MonthName"])
                              .apply(_agg_summary)
                              .reset_index()
                              .rename(columns={"_Year":"Year","_MonthNum":"MonthNum","_MonthName":"Month"}))
    monthly_by_year = monthly_by_year.sort_values(["Year","MonthNum"])

    cal_pnl     = _calendar_matrix(monthly_by_year, "PnL_Sum")
    cal_winrate = _calendar_matrix(monthly_by_year, "WinRate")
    cal_trades  = _calendar_matrix(monthly_by_year, "Trades")
    cal_avgwin  = _calendar_matrix(monthly_by_year, "AvgWin")
    cal_avgloss = _calendar_matrix(monthly_by_year, "AvgLoss")

    to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
    for tbl in (overall, monthly_all, monthly_by_year):
        for c in to_round:
            if c in tbl.columns:
                tbl[c] = tbl[c].astype(float).round(4)
    for mat in (cal_pnl, cal_winrate, cal_trades, cal_avgwin, cal_avgloss):
        mat[:] = np.round(mat.astype(float), 4)

    return {
        "overall": overall,
        "monthly_all": monthly_all,
        "monthly_by_year": monthly_by_year,
        "cal_pnl": cal_pnl,
        "cal_winrate": cal_winrate,
        "cal_trades": cal_trades,
        "cal_avgwin": cal_avgwin,
        "cal_avgloss": cal_avgloss,
    }

# ========= MAIN =========
def main():
    print("Loading parquet...")
    raw = load_symbol_parquet(PATH, TICKER)
    if raw.empty:
        print(f"No rows found in the parquet after filtering for {TICKER}.")
        return

    print("Building 15m bars...")
    df15 = to_15m(raw)

    print("Computing opening ranges & adding features...")
    df15 = compute_opening_range(df15)
    df15 = add_session_vwap(df15) if USE_VWAP_FILTER else df15
    df15 = add_ema_trend(df15, EMA_FAST, EMA_SLOW) if USE_TREND_FILTER else df15

    # Daily bias + ATR + opening-vol threshold merged down to 15m
    df15 = add_daily_bias_and_atr(df15, DAILY_EMA_FAST, DAILY_EMA_SLOW, DAILY_ATR_LEN)

    # GapPct (first 15m vs previous day's close)
    df15["Session"] = session_key(df15.index)
    first_idx = df15.groupby("Session").head(1).index
    prev_close_series = df15.groupby("Session")["Close"].last().shift(1)
    prev_close_map = df15["Session"].map(prev_close_series)
    df15.loc[first_idx, "GapPct"] = (df15.loc[first_idx, "Open"] / prev_close_map.loc[first_idx] - 1.0)
    df15["GapPct"] = df15.groupby("Session")["GapPct"].ffill()

    # RSI / ADX / ATR15 for trailing
    df15 = add_rsi(df15, RSI_LEN) if USE_RSI_FILTER else df15
    df15 = add_adx(df15, ADX_LEN) if USE_ADX_FILTER else df15
    df15 = add_atr_15m(df15, ATR15_LEN) if USE_ATR_TRAIL else df15

    spy15 = None
    if USE_SPY_CONFIRM:
        spy_raw = load_symbol_parquet(SPY_PATH, "SPY")
        spy15 = to_15m(spy_raw)
        spy15 = compute_opening_range(spy15)
        spy15 = add_session_vwap(spy15)
        spy15 = spy15.rename(columns={"VWAP":"SPY_VWAP", "Close":"SPY_Close"})
        spy15 = spy15[["SPY_VWAP", "SPY_Close"]]

    print("Backtesting...")
    trades, equity = backtest_orb_retest(df15, spy15=spy15)
    if trades.empty:
        print("No trades generated with current settings.")
        return

    # Save CSVs (human-readable exits/targets)
    out_trades = trades.copy()
    out_trades["Targets"] = out_trades["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
    out_trades["Exits"]   = out_trades["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
    out_trades.to_csv(TRADES_CSV, index=False)
    equity.to_csv(EQUITY_CSV, index=False)
    print(f"Saved: {TRADES_CSV}, {EQUITY_CSV}")

    # Console yearly
    trades["Win"] = (trades["PnL_$"] >= 0) if COUNT_BE_AS_WIN else (trades["PnL_$"] > 0)
    by_year = trades.groupby(pd.to_datetime(trades["Session"]).dt.year)["PnL_$"].agg(["count", "sum", "mean"])
    print("\n=== Yearly summary (count/sum/mean) ===")
    print(by_year.to_string())

    # Console overall
    total_trades = len(trades)
    total_pnl = trades["PnL_$"].sum()
    win_rate = trades["Win"].mean() * 100.0
    avg_r = trades["R_multiple"].mean()
    median_r = trades["R_multiple"].median()
    avg_win = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
    avg_loss = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()
    pos_sum = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum()
    neg_sum = trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum()
    pf = (pos_sum / abs(neg_sum)) if neg_sum < 0 else np.inf

    print("\n=== Overall Totals ===")
    print(f"Trades: {total_trades} | Total PnL ($): {total_pnl:,.2f} | Win%: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Avg Win ($): {0.0 if pd.isna(avg_win) else avg_win:,.2f} | Avg Loss ($): {0.0 if pd.isna(avg_loss) else avg_loss:,.2f}")
    print(f"Profit Factor: {pf:.3f}")

    # Build summaries
    summ = build_summaries(trades)

    # Excel workbook
    with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
        summary = pd.DataFrame({
            "Metric": ["TotalTrades", "TotalPnL", "WinRate_%", "AvgR", "MedianR",
                       "AvgWin_$", "AvgLoss_$", "ProfitFactor"],
            "Value":  [total_trades, total_pnl, win_rate, avg_r, median_r, avg_win, avg_loss, pf]
        })
        summary.to_excel(writer, sheet_name="Summary", index=False)

        make_excel_compatible(trades).to_excel(writer, sheet_name="Trades", index=False)
        make_excel_compatible(equity).to_excel(writer, sheet_name="Equity", index=False)

        summ["overall"].to_excel(writer, sheet_name="Stats_Overall", index=False)
        summ["monthly_all"].to_excel(writer, sheet_name="Stats_Monthly_All", index=False)
        summ["monthly_by_year"].to_excel(writer, sheet_name="Stats_Monthly_ByYear", index=False)

        # Calendar matrices (Year x Month)
        summ["cal_pnl"].to_excel(writer, sheet_name="Calendar_PnL")
        summ["cal_winrate"].to_excel(writer, sheet_name="Calendar_WinRate")
        summ["cal_trades"].to_excel(writer, sheet_name="Calendar_Trades")
        summ["cal_avgwin"].to_excel(writer, sheet_name="Calendar_AvgWin")
        summ["cal_avgloss"].to_excel(writer, sheet_name="Calendar_AvgLoss")

    print(f"Wrote Excel workbook: {EXCEL_OUT}")

if __name__ == "__main__":
    main()
```

```python
# aapl_orb_15m_retest_from_parquet.py
# Robust 15m ORB (retest) backtest with win-rate tactics & summaries.
# Fixes: duplicate-column guard, hardened opening-range writer.

import warnings
warnings.filterwarnings("ignore")

from datetime import timedelta, time as dtime
from typing import List, Tuple, Optional, Dict

import numpy as np
import pandas as pd
from pandas.api.types import is_datetime64_any_dtype, is_datetime64tz_dtype

# ========= USER SETTINGS =========
# Single-file defaults (use MULTI_INPUTS below for multiple tickers)
PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"
TICKER = "AAPL"

# Optional multi-ticker input: symbol -> parquet path
# Leave {} to use single PATH/TICKER above.
MULTI_INPUTS: Dict[str, str] = {
    # "AAPL": r"C:\...\AAPL.parquet",
    # "MSFT": r"C:\...\MSFT.parquet",
}

# If EventAt in your parquet is naive UTC, keep True. If already in ET, set False.
ASSUME_EVENTAT_IS_UTC = True

LOCAL_TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB parameters
MAX_RETEST_MIN = 120         # wait up to 2h for retest
RETEST_CONFIRM_CLOSE = True  # retest bar must close back in breakout direction
R_MULTIPLES = [1.0, 2.0]     # targets at 1R, 2R (when scale-out disabled)

# Risk params
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# ======== WIN-RATE FILTERS (intraday) ========
USE_VWAP_FILTER = True
USE_TREND_FILTER = True
EMA_FAST = 20
EMA_SLOW = 50

USE_GAP_FILTER = True
GAP_MIN = 0.005   # 0.5%
GAP_MAX = 0.03    # 3%

USE_SPY_CONFIRM = False
SPY_PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\SPY.parquet"  # if USE_SPY_CONFIRM=True

USE_RETEST_WICK_FILTER = True
RETEST_MAX_OPP_WICK_FRAC = 0.40

# ======== NEW WIN-RATE FILTERS (daily & extras) ========
# Daily trend bias
USE_DAILY_TREND_FILTER = True
DAILY_EMA_FAST = 20
DAILY_EMA_SLOW = 50
DAILY_ATR_LEN  = 14  # used for OR/ATR filter

# OR width vs daily ATR (skip choppy/tiny or blowout opens)
USE_OR_WIDTH_ATR_FILTER = True
OR_ATR_MIN = 0.15
OR_ATR_MAX = 1.50

# RSI/ADX (15m)
USE_RSI_FILTER = True
RSI_LEN = 14
RSI_THRESH_LONG = 50.0
RSI_THRESH_SHORT = 50.0    # symmetric: short requires RSI <= 50

USE_ADX_FILTER = False
ADX_LEN = 14
ADX_MIN = 18.0

# Opening 15m volume filter (first bar >= rolling quantile)
USE_OPENING_VOL_FILTER = True
OPENING_VOL_LOOKBACK = 60
OPENING_VOL_QUANTILE = 0.50

# Day-of-week filter (Mon=0..Fri=4)
USE_DOW_FILTER = False
ALLOWED_DOW = {1,2,3}  # Tue–Thu

# ======== RISK MGMT ========
MOVE_STOP_TO_BE_AT_R = 0.5   # None to disable BE move
COUNT_BE_AS_WIN = True

# Optional ATR(15m) trailing (on remainder)
USE_ATR_TRAIL = False
ATR15_LEN = 14
ATR15_MULT = 2.0

# ======== WIN-RATE TACTICS ========
# Scale-out partials
USE_SCALE_OUT = True
SCALE_SPLIT = (0.5, 0.5)     # % qty at TP1 / TP2
TP1_R = 0.75
TP2_R = 1.50

# Direction align with gap
ALIGN_WITH_GAP_DIR = True

# Time windows (ET clock on bar timestamps)
BREAKOUT_DEADLINE = "10:30"
RETEST_DEADLINE   = "11:30"
CONTINUATION_CUTOFF = "11:00"

# Momentum continuation if no retest happens
USE_CONTINUATION = True

# Outputs
TRADES_CSV = "aapl_orb_trades.csv"   # for multi, will write *_ALL.csv too
EQUITY_CSV = "aapl_orb_equity.csv"
EXCEL_OUT  = "aapl_orb_results.xlsx"

# ========= CORE UTILS =========
def ensure_unique_columns(df: pd.DataFrame) -> pd.DataFrame:
    """Drop duplicate-named columns (keep the first occurrence)."""
    if getattr(df.columns, "duplicated", None) is not None and df.columns.duplicated().any():
        df = df.loc[:, ~df.columns.duplicated()].copy()
    return df

def load_symbol_parquet(path: str, symbol_to_keep: str) -> pd.DataFrame:
    last_err = None
    for eng in ("pyarrow", "fastparquet"):
        try:
            df = pd.read_parquet(path, engine=eng)
            break
        except Exception as e:
            last_err = e
            df = None
    if df is None:
        raise SystemExit(f"Failed to read Parquet. Install pyarrow or fastparquet. Last error: {last_err}")

    if "symbol" in df.columns:
        df["symbol"] = df["symbol"].apply(
            lambda x: x.decode("utf-8") if isinstance(x, (bytes, bytearray)) else str(x)
        ).str.strip().str.replace("^b['\"]?|['\"]?$", "", regex=True)
    else:
        df["symbol"] = symbol_to_keep

    if "EventAt" not in df.columns:
        raise ValueError("Expected 'EventAt' column in the parquet.")
    if not is_datetime64_any_dtype(df["EventAt"]):
        df["EventAt"] = pd.to_datetime(df["EventAt"], utc=False, errors="coerce")

    if is_datetime64tz_dtype(df["EventAt"]):
        df["EventAt"] = df["EventAt"].dt.tz_convert(LOCAL_TZ)
    else:
        if ASSUME_EVENTAT_IS_UTC:
            df["EventAt"] = df["EventAt"].dt.tz_localize("UTC").dt.tz_convert(LOCAL_TZ)
        else:
            df["EventAt"] = df["EventAt"].dt.tz_localize(LOCAL_TZ)

    df = df[df["symbol"] == symbol_to_keep].copy()
    df = df.sort_values("EventAt").set_index("EventAt")
    need = {"Open", "High", "Low", "Close"}
    if not need.issubset(df.columns):
        raise ValueError(f"Missing required OHLC columns: {need - set(df.columns)}")

    keep = [c for c in ["Open", "High", "Low", "Close", "Volume", "symbol"] if c in df.columns]
    return ensure_unique_columns(df[keep])

def to_15m(df: pd.DataFrame) -> pd.DataFrame:
    o = df["Open"].resample("15T", label="left", closed="left").first()
    h = df["High"].resample("15T", label="left", closed="left").max()
    l = df["Low"].resample("15T", label="left", closed="left").min()
    c = df["Close"].resample("15T", label="left", closed="left").last()
    parts = {"Open": o, "High": h, "Low": l, "Close": c}
    if "Volume" in df.columns:
        parts["Volume"] = df["Volume"].resample("15T", label="left", closed="left").sum()
    out = pd.concat(parts, axis=1).dropna(how="any")
    out["Ticker"] = df["symbol"].iloc[0] if "symbol" in df.columns else TICKER
    out = out.between_time(REG_SESSION_START, REG_SESSION_END)
    return ensure_unique_columns(out)

def session_key(index_like) -> pd.Series:
    return pd.to_datetime(pd.Index(index_like).date)

def compute_opening_range(df15: pd.DataFrame) -> pd.DataFrame:
    """Safe OR writer that removes duplicates and prior OR columns."""
    df = ensure_unique_columns(df15.copy())

    # Remove any pre-existing OR columns
    for c in ("ORH", "ORL"):
        if (df.columns == c).sum() > 0:
            df = df.drop(columns=[c])

    missing = [c for c in ("High", "Low") if c not in df.columns]
    if missing:
        raise ValueError(f"compute_opening_range: missing required columns {missing}")

    if not df.index.is_monotonic_increasing:
        df = df.sort_index()

    df["Session"] = session_key(df.index)
    first_idx = df.groupby("Session").head(1).index

    df["ORH"] = np.nan
    df["ORL"] = np.nan
    # Use aligned .values to avoid index alignment quirks
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float).values
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float).values
    df[["ORH", "ORL"]] = df.groupby("Session")[["ORH", "ORL"]].ffill()
    return df

# ========= INDICATORS =========
def add_session_vwap(df15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df15.copy())
    df["Session"] = session_key(df.index)
    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0
    if "Volume" in df.columns:
        df["vwap_num"] = tp * df["Volume"]
        df["vwap_den"] = df["Volume"].replace(0, np.nan)
    else:
        df["vwap_num"] = tp
        df["vwap_den"] = 1.0
    df["VWAP"] = (df.groupby("Session")["vwap_num"].cumsum() /
                  df.groupby("Session")["vwap_den"].cumsum())
    return ensure_unique_columns(df.drop(columns=["vwap_num","vwap_den"]))

def add_ema_trend(df15: pd.DataFrame, fast=20, slow=50) -> pd.DataFrame:
    df = ensure_unique_columns(df15.copy())
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return ensure_unique_columns(df)

def rsi(series: pd.Series, length=14) -> pd.Series:
    delta = series.diff()
    up = delta.clip(lower=0)
    dn = -delta.clip(upper=0)
    avg_gain = up.ewm(alpha=1/length, adjust=False).mean()
    avg_loss = dn.ewm(alpha=1/length, adjust=False).mean()
    rs = avg_gain / (avg_loss.replace(0, np.nan))
    return 100 - (100 / (1 + rs))

def add_rsi(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df15.copy())
    df[f"RSI{length}"] = rsi(df["Close"], length)
    return ensure_unique_columns(df)

def true_range(high, low, prev_close):
    return np.maximum.reduce([
        (high - low).values,
        np.abs(high - prev_close).values,
        np.abs(low - prev_close).values
    ])

def add_atr_15m(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df15.copy())
    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    df[f"ATR15_{length}"] = tr.ewm(alpha=1/length, adjust=False).mean()
    return ensure_unique_columns(df)

def add_adx(df15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df15.copy())
    up_move = df["High"].diff()
    dn_move = -df["Low"].diff()
    plus_dm  = np.where((up_move > dn_move) & (up_move > 0), up_move, 0.0)
    minus_dm = np.where((dn_move > up_move) & (dn_move > 0), dn_move, 0.0)

    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    atr = tr.ewm(alpha=1/length, adjust=False).mean()

    plus_di = 100 * pd.Series(plus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    minus_di = 100 * pd.Series(minus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    dx = (abs(plus_di - minus_di) / (plus_di + minus_di).replace(0, np.nan)) * 100
    adx = dx.ewm(alpha=1/length, adjust=False).mean()
    df[f"ADX{length}"] = adx
    return ensure_unique_columns(df)

# ======== DAILY FEATURES (bias, ATR, opening vol) ========
def build_daily_from_15m(df15: pd.DataFrame) -> pd.DataFrame:
    d = ensure_unique_columns(df15.copy())
    d["Session"] = session_key(d.index)
    daily = d.groupby("Session").agg(
        Open=("Open","first"),
        High=("High","max"),
        Low=("Low","min"),
        Close=("Close","last"),
        Volume=("Volume","sum") if "Volume" in d.columns else ("Close","size")
    )
    return daily

def add_daily_bias_and_atr(df15: pd.DataFrame, ema_fast=20, ema_slow=50, atr_len=14) -> pd.DataFrame:
    dly = build_daily_from_15m(df15)
    # Daily EMAs
    dly[f"EMA_D{ema_fast}"] = dly["Close"].ewm(span=ema_fast, adjust=False).mean()
    dly[f"EMA_D{ema_slow}"] = dly["Close"].ewm(span=ema_slow, adjust=False).mean()
    # Daily ATR
    prev_close = dly["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (dly["High"] - dly["Low"]).values,
        np.abs(dly["High"] - prev_close).values,
        np.abs(dly["Low"]  - prev_close).values
    ]), index=dly.index)
    atr_col = f"ATR_D{atr_len}"
    dly[atr_col] = tr.ewm(alpha=1/atr_len, adjust=False).mean()

    # Opening 15m volume + rolling quantile
    if "Volume" in df15.columns:
        open15_vol = df15.groupby(session_key(df15.index))["Volume"].first()
        dly["Open15_Vol"] = open15_vol
        dly["Open15_Vol_Thresh"] = (
            dly["Open15_Vol"].rolling(window=OPENING_VOL_LOOKBACK, min_periods=10)
            .quantile(OPENING_VOL_QUANTILE)
        )

    feature_cols = [f"EMA_D{ema_fast}", f"EMA_D{ema_slow}", atr_col, "Open15_Vol", "Open15_Vol_Thresh"]
    feature_cols = [c for c in feature_cols if c in dly.columns]

    out = ensure_unique_columns(df15.copy())
    out["Session"] = session_key(out.index)
    out = out.merge(dly[feature_cols], left_on="Session", right_index=True, how="left")
    return ensure_unique_columns(out)

# ========= MISC =========
def apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps / 10000.0)
    return price * factor if side == "buy" else price / factor

def wick_opposite_fraction(row: pd.Series, direction: str) -> float:
    h, l, o, c = float(row["High"]), float(row["Low"]), float(row.get("Open", np.nan)), float(row["Close"])
    rng = max(h - l, 1e-12)
    if direction == "long":
        opp = min(o, c) - l  # adverse (lower) wick for longs
    else:
        opp = h - max(o, c)  # adverse (upper) wick for shorts
    return float(max(opp, 0.0) / rng)

def _to_clock(ts):  # naive clock tuple from timestamp
    return dtime(ts.hour, ts.minute)

def _clock_le(ts, hhmm: str) -> bool:
    h, m = map(int, hhmm.split(":"))
    return _to_clock(ts) <= dtime(h, m)

# ========= BACKTEST (per-DataFrame) =========
def backtest_orb_retest_for_df(
    df15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE,
    spy15: Optional[pd.DataFrame] = None,
    ticker: Optional[str] = None
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """Build features safely, then run ORB-retest + tactics on a single 15m DF."""
    df15 = ensure_unique_columns(df15)
    df = compute_opening_range(df15)

    if USE_VWAP_FILTER:
        df = add_session_vwap(df)
    if USE_TREND_FILTER:
        df = add_ema_trend(df, EMA_FAST, EMA_SLOW)

    # Daily features (bias, ATR, opening vol)
    df = add_daily_bias_and_atr(df, DAILY_EMA_FAST, DAILY_EMA_SLOW, DAILY_ATR_LEN)

    # Gap: first 15m vs previous session close
    df["Session"] = session_key(df.index)
    first_idx = df.groupby("Session").head(1).index
    prev_close_series = df.groupby("Session")["Close"].last().shift(1)
    prev_close_map = df["Session"].map(prev_close_series)
    df.loc[first_idx, "GapPct"] = (df.loc[first_idx, "Open"] / prev_close_map.loc[first_idx] - 1.0)
    df["GapPct"] = df.groupby("Session")["GapPct"].ffill()

    # RSI / ADX / ATR15 trailing
    if USE_RSI_FILTER:
        df = add_rsi(df, RSI_LEN)
    if USE_ADX_FILTER:
        df = add_adx(df, ADX_LEN)
    if USE_ATR_TRAIL:
        df = add_atr_15m(df, ATR15_LEN)

    df = ensure_unique_columns(df)

    # Backtest core
    sessions = df["Session"].unique()
    trades = []

    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh, orl = sdf["ORH"].iloc[0], sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        after_open = sdf.iloc[1:].copy()
        long_break  = after_open[after_open["Close"] > orh].head(1)
        short_break = after_open[after_open["Close"] < orl].head(1)
        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            if long_break.index[0] < short_break.index[0]:
                direction, breakout_row = "long", long_break.iloc[0]
            else:
                direction, breakout_row = "short", short_break.iloc[0]
        elif not long_break.empty:
            direction, breakout_row = "long", long_break.iloc[0]
        else:
            direction, breakout_row = "short", short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        level = orh if direction == "long" else orl
        touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)

        # Entry bar/time decision
        retest_used = True
        if touch.empty:
            if USE_CONTINUATION and _clock_le(btime, CONTINUATION_CUTOFF):
                retest_used = False
                retest_time = btime
                retest_bar = breakout_row
            else:
                continue
        else:
            retest_bar = touch.iloc[0]
            retest_time = retest_bar.name

        # Retest close confirmation (only for retest mode)
        if retest_used and retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        # Filters
        if USE_DOW_FILTER and retest_time.weekday() not in ALLOWED_DOW:
            continue

        if USE_VWAP_FILTER and "VWAP" in sdf.columns:
            v_prev = sdf.loc[:retest_time, "VWAP"].tail(3).dropna().values
            rising  = (len(v_prev) < 2) or np.all(np.diff(v_prev) >= 0)
            falling = (len(v_prev) < 2) or np.all(np.diff(v_prev) <= 0)
            if direction == "long":
                if not (retest_bar["Close"] > retest_bar["VWAP"] and rising):
                    continue
            else:
                if not (retest_bar["Close"] < retest_bar["VWAP"] and falling):
                    continue

        if USE_TREND_FILTER and (f"EMA{EMA_FAST}" in sdf.columns) and (f"EMA{EMA_SLOW}" in sdf.columns):
            emaf = retest_bar[f"EMA{EMA_FAST}"]; emas = retest_bar[f"EMA{EMA_SLOW}"]
            if direction == "long" and not (emaf > emas): continue
            if direction == "short" and not (emaf < emas): continue

        if USE_DAILY_TREND_FILTER and f"EMA_D{DAILY_EMA_FAST}" in sdf.columns and f"EMA_D{DAILY_EMA_SLOW}" in sdf.columns:
            dfast = sdf[f"EMA_D{DAILY_EMA_FAST}"].iloc[0]
            dslow = sdf[f"EMA_D{DAILY_EMA_SLOW}"].iloc[0]
            if direction == "long" and not (dfast > dslow): continue
            if direction == "short" and not (dfast < dslow): continue

        atr_col = f"ATR_D{DAILY_ATR_LEN}"
        if USE_OR_WIDTH_ATR_FILTER and atr_col in sdf.columns:
            or_width = float(orh - orl)
            atr_d = float(sdf[atr_col].iloc[0])
            if not (atr_d > 0): continue
            frac = or_width / atr_d
            if not (OR_ATR_MIN <= frac <= OR_ATR_MAX): continue

        if USE_RSI_FILTER and f"RSI{RSI_LEN}" in sdf.columns:
            rsi_val = float(retest_bar[f"RSI{RSI_LEN}"])
            if direction == "long" and not (rsi_val >= RSI_THRESH_LONG): continue
            if direction == "short" and not (rsi_val <= 100 - RSI_THRESH_SHORT): continue

        if USE_ADX_FILTER and f"ADX{ADX_LEN}" in sdf.columns:
            adx_val = float(retest_bar[f"ADX{ADX_LEN}"])
            if not (adx_val >= ADX_MIN): continue

        if USE_GAP_FILTER and "GapPct" in sdf.columns:
            gap = sdf.loc[sdf.index.min(), "GapPct"]
            if pd.notna(gap) and not (GAP_MIN <= abs(float(gap)) <= GAP_MAX):
                continue

        if ALIGN_WITH_GAP_DIR and "GapPct" in sdf.columns:
            g = float(sdf["GapPct"].iloc[0]) if pd.notna(sdf["GapPct"].iloc[0]) else 0.0
            if (direction == "long" and g < 0) or (direction == "short" and g > 0):
                continue

        if USE_SPY_CONFIRM and spy15 is not None and retest_time in spy15.index:
            spy_row = spy15.loc[retest_time]
            if direction == "long" and not (spy_row["SPY_Close"] > spy_row["SPY_VWAP"]): continue
            if direction == "short" and not (spy_row["SPY_Close"] < spy_row["SPY_VWAP"]): continue

        if USE_RETEST_WICK_FILTER:
            opp_frac = wick_opposite_fraction(retest_bar, direction)
            if opp_frac > RETEST_MAX_OPP_WICK_FRAC: continue

        # Time windows
        if not _clock_le(btime, BREAKOUT_DEADLINE): continue
        if retest_used and (not _clock_le(retest_time, RETEST_DEADLINE)): continue
        if (not retest_used) and (not _clock_le(retest_time, CONTINUATION_CUTOFF)): continue

        # Entry/Stop/Risk
        entry = apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop = orl if direction == "long" else orh
        rps = abs(entry - stop)
        if rps <= 1e-12: continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1

        # Targets
        if USE_SCALE_OUT:
            targets_r = [TP1_R, TP2_R]
        else:
            targets_r = r_targets
        targets = [(entry + r * rps) if direction == "long" else (entry - r * rps) for r in targets_r]

        # Walk-forward exits
        run = sdf[sdf.index >= retest_time].copy()
        qty_left = qty
        exits = []
        be_armed = (MOVE_STOP_TO_BE_AT_R is not None)
        be_level = entry
        next_tp_idx = 0

        if USE_SCALE_OUT and len(targets) >= 2:
            leg_sizes = [qty * SCALE_SPLIT[0], qty * SCALE_SPLIT[1]]
        else:
            leg_sizes = [qty]

        for ts, row in run.iterrows():
            hi, lo = row["High"], row["Low"]

            # Arm BE after +X R
            if be_armed:
                if direction == "long" and hi >= entry + (MOVE_STOP_TO_BE_AT_R * rps):
                    stop = max(stop, be_level); be_armed = False
                elif direction == "short" and lo <= entry - (MOVE_STOP_TO_BE_AT_R * rps):
                    stop = min(stop, be_level); be_armed = False

            # ATR(15m) trailing (optional)
            if USE_ATR_TRAIL and f"ATR15_{ATR15_LEN}" in sdf.columns:
                atr_now = float(row[f"ATR15_{ATR15_LEN}"])
                if direction == "long":
                    stop = max(stop, hi - ATR15_MULT * atr_now)
                else:
                    stop = min(stop, lo + ATR15_MULT * atr_now)

            # Stop first
            if lo <= stop <= hi:
                px = apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                label = "breakeven" if abs(px - be_level) < 1e-10 else "stop"
                exits.append((label, ts, px, qty_left))
                qty_left = 0
                break

            # Targets
            if next_tp_idx < len(targets):
                tp = targets[next_tp_idx]
                if lo <= tp <= hi:
                    px = apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    fill_qty = leg_sizes[next_tp_idx] if next_tp_idx < len(leg_sizes) else qty_left
                    exits.append((f"tp{next_tp_idx+1}", ts, px, fill_qty))
                    qty_left -= fill_qty
                    next_tp_idx += 1
                    if USE_SCALE_OUT and next_tp_idx == 1 and MOVE_STOP_TO_BE_AT_R is not None:
                        stop = be_level
                    if qty_left <= 1e-9:
                        break

            # Tighten if no TP1 by continuation cutoff
            if USE_CONTINUATION and not _clock_le(ts, CONTINUATION_CUTOFF) and next_tp_idx == 0:
                stop = max(stop, be_level) if direction == "long" else min(stop, be_level)

        # EOD exit
        if qty_left > 1e-9:
            last = run.iloc[-1]
            px = apply_slippage(last["Close"], slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", run.index[-1], px, qty_left))
            qty_left = 0

        cash_pnl = sum((px - entry) * side_mult * q for (_, _, px, q) in exits) - fees

        trades.append({
            "Ticker": ticker if ticker else (sdf["Ticker"].iloc[0] if "Ticker" in sdf.columns else TICKER),
            "Session": ses,
            "Direction": direction,
            "ORH": orh, "ORL": orl,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": targets,
            "Exits": [(lab, ts, px) for (lab, ts, px, _) in exits],
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (rps * max(qty, 1e-12))
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        return trade_log, pd.DataFrame(columns=["Session", "CumPnL_$"])

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$": "CumPnL_$"}))
    return trade_log, equity

# ========= EXCEL SAFETY =========
def make_excel_compatible(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    for col in out.columns:
        if is_datetime64tz_dtype(out[col]):
            out[col] = out[col].dt.tz_convert(LOCAL_TZ).dt.tz_localize(None)
    return out

# ========= SUMMARY HELPERS =========
def _agg_summary(g: pd.DataFrame) -> pd.Series:
    pnl = pd.to_numeric(g["PnL_$"], errors="coerce").dropna()
    trades = len(pnl)
    wins_mask = pnl > 0
    losses_mask = pnl < 0
    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()

    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf
    else:
        pf = np.nan

    return pd.Series({
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": int(wins_mask.sum()),
        "Losses": int(losses_mask.sum()),
        "WinRate": (wins_mask.mean() * 100.0) if trades else np.nan,
        "AvgWin": pnl[wins_mask].mean() if wins_mask.any() else np.nan,
        "AvgLoss": pnl[losses_mask].mean() if losses_mask.any() else np.nan,
        "AvgR": g["R_multiple"].mean(),
        "MedianR": g["R_multiple"].median(),
        "PF": pf,
    })

MONTH_ABBR = {1:"Jan",2:"Feb",3:"Mar",4:"Apr",5:"May",6:"Jun",
              7:"Jul",8:"Aug",9:"Sep",10:"Oct",11:"Nov",12:"Dec"}

def _calendar_matrix(df_monthly_by_year: pd.DataFrame, value_col: str) -> pd.DataFrame:
    mat = (df_monthly_by_year
           .pivot(index="Year", columns="MonthNum", values=value_col)
           .reindex(columns=range(1, 13)))
    mat.columns = [MONTH_ABBR[m] for m in mat.columns]
    return mat.sort_index()

def build_summaries(trades: pd.DataFrame):
    overall = _agg_summary(trades).to_frame().T

    ses = pd.to_datetime(trades["Session"])
    trades = trades.copy()
    trades["_MonthStr"] = ses.dt.to_period("M").astype(str)
    monthly_all = (trades.groupby("_MonthStr").apply(_agg_summary)
                          .reset_index()
                          .rename(columns={"_MonthStr": "Month"})
                          .sort_values("Month"))

    trades["_Year"] = ses.dt.year
    trades["_MonthNum"] = ses.dt.month
    trades["_MonthName"] = trades["_MonthNum"].map(MONTH_ABBR)
    monthly_by_year = (trades.groupby(["_Year", "_MonthNum", "_MonthName"])
                              .apply(_agg_summary)
                              .reset_index()
                              .rename(columns={"_Year":"Year","_MonthNum":"MonthNum","_MonthName":"Month"}))
    monthly_by_year = monthly_by_year.sort_values(["Year","MonthNum"])

    cal_pnl     = _calendar_matrix(monthly_by_year, "PnL_Sum")
    cal_winrate = _calendar_matrix(monthly_by_year, "WinRate")
    cal_trades  = _calendar_matrix(monthly_by_year, "Trades")
    cal_avgwin  = _calendar_matrix(monthly_by_year, "AvgWin")
    cal_avgloss = _calendar_matrix(monthly_by_year, "AvgLoss")

    to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
    for tbl in (overall, monthly_all, monthly_by_year):
        for c in to_round:
            if c in tbl.columns:
                tbl[c] = tbl[c].astype(float).round(4)
    for mat in (cal_pnl, cal_winrate, cal_trades, cal_avgwin, cal_avgloss):
        mat[:] = np.round(mat.astype(float), 4)

    return {
        "overall": overall,
        "monthly_all": monthly_all,
        "monthly_by_year": monthly_by_year,
        "cal_pnl": cal_pnl,
        "cal_winrate": cal_winrate,
        "cal_trades": cal_trades,
        "cal_avgwin": cal_avgwin,
        "cal_avgloss": cal_avgloss,
    }

# ========= MAIN =========
def main():
    # Build input map
    data_map: Dict[str, pd.DataFrame] = {}
    if MULTI_INPUTS:
        for tkr, path in MULTI_INPUTS.items():
            raw = load_symbol_parquet(path, tkr)
            data_map[tkr] = to_15m(raw)
    else:
        raw = load_symbol_parquet(PATH, TICKER)
        data_map[TICKER] = to_15m(raw)

    # SPY confirm (optional)
    spy15 = None
    if USE_SPY_CONFIRM:
        spy_raw = load_symbol_parquet(SPY_PATH, "SPY")
        spy15 = to_15m(spy_raw)
        spy15 = compute_opening_range(spy15)
        spy15 = add_session_vwap(spy15)
        spy15 = spy15.rename(columns={"VWAP":"SPY_VWAP", "Close":"SPY_Close"})
        spy15 = spy15[["SPY_VWAP", "SPY_Close"]]

    all_trades, all_equity = [], []

    for i, (tkr, df15) in enumerate(data_map.items(), 1):
        print(f"\n=== Backtesting {tkr} ({i}/{len(data_map)}) ===")
        trades, equity = backtest_orb_retest_for_df(df15, spy15=spy15, ticker=tkr)
        if not trades.empty:
            all_trades.append(trades)
            eq = equity.copy()
            eq["Ticker"] = tkr
            all_equity.append(eq)

    if not all_trades:
        print("No trades generated with current settings.")
        return

    trades = pd.concat(all_trades, ignore_index=True)
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    # Save CSVs
    out_trades = trades.copy()
    out_trades["Targets"] = out_trades["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
    out_trades["Exits"]   = out_trades["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
    out_trades.to_csv(TRADES_CSV if len(data_map)==1 else "trades_ALL.csv", index=False)
    if not equity.empty:
        equity.to_csv(EQUITY_CSV if len(data_map)==1 else "equity_ALL.csv", index=False)
    print(f"Saved CSV(s).")

    # Console summaries per year (all tickers combined)
    wins = (trades["PnL_$"] >= 0) if COUNT_BE_AS_WIN else (trades["PnL_$"] > 0)
    trades["Win"] = wins
    by_year = trades.groupby(pd.to_datetime(trades["Session"]).dt.year)["PnL_$"].agg(["count", "sum", "mean"])
    print("\n=== Yearly summary (count/sum/mean) ===")
    print(by_year.to_string())

    # Overall totals
    total_trades = len(trades)
    total_pnl = trades["PnL_$"].sum()
    win_rate = wins.mean() * 100.0
    avg_r = trades["R_multiple"].mean()
    median_r = trades["R_multiple"].median()
    avg_win = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
    avg_loss = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()
    pos_sum = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum()
    neg_sum = trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum()
    pf = (pos_sum / abs(neg_sum)) if neg_sum < 0 else np.inf

    print("\n=== Overall Totals ===")
    print(f"Trades: {total_trades} | Total PnL ($): {total_pnl:,.2f} | Win%: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Avg Win ($): {0.0 if pd.isna(avg_win) else avg_win:,.2f} | Avg Loss ($): {0.0 if pd.isna(avg_loss) else avg_loss:,.2f}")
    print(f"Profit Factor: {pf:.3f}")

    # Build summaries
    summ = build_summaries(trades)

    # Excel workbook
    with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
        summary = pd.DataFrame({
            "Metric": ["TotalTrades", "TotalPnL", "WinRate_%", "AvgR", "MedianR",
                       "AvgWin_$", "AvgLoss_$", "ProfitFactor"],
            "Value":  [total_trades, total_pnl, win_rate, avg_r, median_r, avg_win, avg_loss, pf]
        })
        summary.to_excel(writer, sheet_name="Summary", index=False)

        make_excel_compatible(trades).to_excel(writer, sheet_name="Trades", index=False)
        if not equity.empty:
            make_excel_compatible(equity).to_excel(writer, sheet_name="Equity", index=False)

        summ["overall"].to_excel(writer, sheet_name="Stats_Overall", index=False)
        summ["monthly_all"].to_excel(writer, sheet_name="Stats_Monthly_All", index=False)
        summ["monthly_by_year"].to_excel(writer, sheet_name="Stats_Monthly_ByYear", index=False)

        summ["cal_pnl"].to_excel(writer, sheet_name="Calendar_PnL")
        summ["cal_winrate"].to_excel(writer, sheet_name="Calendar_WinRate")
        summ["cal_trades"].to_excel(writer, sheet_name="Calendar_Trades")
        summ["cal_avgwin"].to_excel(writer, sheet_name="Calendar_AvgWin")
        summ["cal_avgloss"].to_excel(writer, sheet_name="Calendar_AvgLoss")

    print(f"Wrote Excel workbook: {EXCEL_OUT}")

if __name__ == "__main__":
    main()
```
