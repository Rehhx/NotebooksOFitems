```python
import pandas as pd

# Load the CSV
file_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(file_path)

# 1. Map internal_industry to sector
def map_sector(industry):
    if pd.isna(industry):
        return 'Unknown'
    elif 'Financial' in industry or 'Insurance' in industry:
        return 'Financial Services and Insurance'
    elif 'Legal' in industry:
        return 'Legal Services'
    elif 'Technology' in industry or 'Media' in industry or 'Advertising' in industry or 'Information' in industry:
        return 'Tech, Media, and Info'
    else:
        return 'Other'

df['industry_sector'] = df['internal_industry'].apply(map_sector)

# 2. Map internal_class to building quality
def map_quality(cls):
    return 'High Quality' if cls == 'A' else 'Other'

df['building_quality'] = df['internal_class'].apply(map_quality)

# 3. Filter for recent years only and drop NA
df = df[df['year'].isin([2020, 2021, 2022, 2023, 2024])]
df = df.dropna(subset=['CBD_suburban', 'industry_sector', 'building_quality', 'monthsigned'])

# 4. Create a proper datetime column (year_month)
df['year_month'] = pd.to_datetime(df['year'].astype(str) + '-' + df['monthsigned'].astype(int).astype(str).str.zfill(2))

# 5. Group by: sector, quality, CBD/suburban, and month
grouped = df.groupby(['year_month', 'industry_sector', 'building_quality', 'CBD_suburban']).size().reset_index(name='lease_count')

# 6. Pivot for CBD vs Suburban
pivot = grouped.pivot_table(index=['year_month', 'industry_sector', 'building_quality'],
                            columns='CBD_suburban',
                            values='lease_count',
                            fill_value=0).reset_index()

# 7. Ensure both CBD and Suburban columns exist
if 'CBD' not in pivot.columns:
    pivot['CBD'] = 0
if 'Suburban' not in pivot.columns:
    pivot['Suburban'] = 0

# 8. Sort and calculate month-over-month % change
pivot = pivot.sort_values(by=['industry_sector', 'building_quality', 'year_month'])
pivot['CBD_%_change'] = pivot.groupby(['industry_sector', 'building_quality'])['CBD'].pct_change() * 100
pivot['Suburban_%_change'] = pivot.groupby(['industry_sector', 'building_quality'])['Suburban'].pct_change() * 100

# 9. Round and display
pivot = pivot.round(2)
print(pivot)

# 10. Save to file
pivot.to_csv(r"C:\Users\pcagm\Downloads\Monthly_Change_Sector_Quality_CBDSuburban.csv", index=False)

# Step 11: Melt the DataFrame to long format for aggregation
melted = pivot.melt(
    id_vars=['industry_sector', 'building_quality', 'year_month'],
    value_vars=['CBD_%_change', 'Suburban_%_change'],
    var_name='lease_type',
    value_name='pct_change'
)

# Step 12: Clean lease_type to just say CBD or Suburban
melted['lease_type'] = melted['lease_type'].str.replace('_%_change', '')

# Step 13: Drop NaNs (first months with no prior % change)
melted = melted.dropna(subset=['pct_change'])

# Step 14: Group by sector and lease_type and sum % change
summary = melted.groupby(['industry_sector', 'lease_type'])['pct_change'].sum().reset_index()

# Step 15: Round and show result
summary['pct_change'] = summary['pct_change'].round(2)
print(summary)

# Step 16: Save final summary
summary.to_csv(r"C:\Users\pcagm\Downloads\Total_Sector_Percentage_Change_By_CBDSuburban.csv", index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
file_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(file_path)

# --- CLEANING ---

# Drop rows missing key fields
df = df.dropna(subset=['leasedSF', 'quarter', 'year', 'internal_industry', 'internal_class'])

# Convert quarter like "Q1" → 1
df['quarter'] = df['quarter'].astype(str).str.extract(r'Q([1-4])')[0].astype(int)

# Standardize numeric types
df['year'] = df['year'].astype(int)
df['leasedSF'] = df['leasedSF'].astype(int)

# Normalize internal_class to A, O, Other
df['internal_class'] = df['internal_class'].astype(str)
df['internal_class_clean'] = df['internal_class'].where(df['internal_class'].isin(['A', 'O']), 'Other')

# Convert internal_industry to string
df['internal_industry'] = df['internal_industry'].astype(str)

# Optional: Filter top industries
selected_industries = [
    'Financial Services and Insurance',
    'Technology, Advertising, Media, and Information',
    'Legal Services'
]
df = df[df['internal_industry'].isin(selected_industries)]

# --- LABELING ---

# Create quarter label and sorting date
df['year_quarter'] = df['year'].astype(str) + ' Q' + df['quarter'].astype(str)
quarter_month_map = {1: '01', 2: '04', 3: '07', 4: '10'}
df['quarter_start'] = pd.to_datetime(
    df['year'].astype(str) + '-' + df['quarter'].map(quarter_month_map) + '-01'
)

# --- GROUPING ---

grouped = df.groupby(
    ['year_quarter', 'quarter_start', 'internal_industry', 'internal_class_clean']
)['leasedSF'].sum().reset_index()

grouped['sector_quality'] = grouped['internal_industry'] + ' - ' + grouped['internal_class_clean']

# --- COMPLETE ALL COMBOS ---

all_quarters = grouped[['year_quarter', 'quarter_start']].drop_duplicates().sort_values('quarter_start')
all_combos = pd.MultiIndex.from_product(
    [all_quarters['year_quarter'], grouped['sector_quality'].unique()],
    names=['year_quarter', 'sector_quality']
)

pivot_df = grouped.pivot_table(
    index='year_quarter',
    columns='sector_quality',
    values='leasedSF',
    aggfunc='sum'
).reindex(all_combos).fillna(0).reset_index()

pivot_df = pd.merge(pivot_df, all_quarters, on='year_quarter', how='left').sort_values('quarter_start')
pivot_df.set_index('year_quarter', inplace=True)
pivot_df.drop(columns='quarter_start', inplace=True)

# --- SHOW ENTIRE DATAFRAME ---
pd.set_option('display.max_rows', None)
print("\n📄 Full Grouped and Completed DataFrame (All Quarters + Sector-Class Combos):")
print(pivot_df)

# --- PLOT ---

plt.figure(figsize=(18, 9))
pivot_df.plot(marker='o', figsize=(18, 9))
plt.title('Leased Square Footage by Quarter and Year (All Quarters, Filtered Sectors)')
plt.xlabel('Year-Quarter')
plt.ylabel('Leased Square Footage')
plt.xticks(rotation=90)
plt.grid(True)
plt.legend(title='Sector - Class', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
import pandas as pd
import matplotlib.pyplot as plt

# Load data
file_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(file_path)

# --- CLEANING ---

# Drop rows missing key fields
df = df.dropna(subset=['leasedSF', 'quarter', 'year', 'internal_industry', 'internal_class'])

# Convert quarter like "Q1" → 1
df['quarter'] = df['quarter'].astype(str).str.extract(r'Q([1-4])')[0].astype(int)

# Standardize numeric types
df['year'] = df['year'].astype(int)
df['leasedSF'] = df['leasedSF'].astype(int)

# Normalize internal_class to A, O, Other
df['internal_class'] = df['internal_class'].astype(str)
df['internal_class_clean'] = df['internal_class'].where(df['internal_class'].isin(['A', 'O']), 'Other')

# Convert internal_industry to string
df['internal_industry'] = df['internal_industry'].astype(str)

# Optional: Filter top industries
selected_industries = [
    'Financial Services and Insurance',
    'Technology, Advertising, Media, and Information',
    'Legal Services'
]
df = df[df['internal_industry'].isin(selected_industries)]

# --- LABELING ---

# Create quarter label and sorting date
df['year_quarter'] = df['year'].astype(str) + ' Q' + df['quarter'].astype(str)
quarter_month_map = {1: '01', 2: '04', 3: '07', 4: '10'}
df['quarter_start'] = pd.to_datetime(
    df['year'].astype(str) + '-' + df['quarter'].map(quarter_month_map) + '-01'
)

# --- GROUPING ---

grouped = df.groupby(
    ['year_quarter', 'quarter_start', 'internal_industry', 'internal_class_clean']
)['leasedSF'].sum().reset_index()

grouped['sector_quality'] = grouped['internal_industry'] + ' - ' + grouped['internal_class_clean']

# --- COMPLETE ALL COMBOS ---

all_quarters = grouped[['year_quarter', 'quarter_start']].drop_duplicates().sort_values('quarter_start')
all_combos = pd.MultiIndex.from_product(
    [all_quarters['year_quarter'], grouped['sector_quality'].unique()],
    names=['year_quarter', 'sector_quality']
)

pivot_df = grouped.pivot_table(
    index='year_quarter',
    columns='sector_quality',
    values='leasedSF',
    aggfunc='sum'
).reindex(all_combos).fillna(0).reset_index()

pivot_df = pd.merge(pivot_df, all_quarters, on='year_quarter', how='left').sort_values('quarter_start')
pivot_df.set_index('year_quarter', inplace=True)
pivot_df.drop(columns='quarter_start', inplace=True)

# --- SHOW ENTIRE DATAFRAME ---
pd.set_option('display.max_rows', None)
print("\n📄 Full Grouped and Completed DataFrame (All Quarters + Sector-Class Combos):")
print(pivot_df)

# --- PLOT ---

plt.figure(figsize=(18, 9))
pivot_df.plot(marker='o', figsize=(18, 9))
plt.title('Leased Square Footage by Quarter and Year (All Quarters, Filtered Sectors)')
plt.xlabel('Year-Quarter')
plt.ylabel('Leased Square Footage')
plt.xticks(rotation=90)
plt.grid(True)
plt.legend(title='Sector - Class', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
