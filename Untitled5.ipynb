```python
import requests
import time
import pandas as pd
from datetime import datetime, timedelta

def get_historical_crypto_data(crypto_id, date, api_key):
    if not crypto_id:
        return {"error": "Invalid crypto ID"}
    
    date_str = date.strftime('%d-%m-%Y')  # Format date for API request
    url = f"https://api.coingecko.com/api/v3/coins/{crypto_id}/history?date={date_str}"
    headers = {"x-cg-api-key": api_key}
    
    while True:
        response = requests.get(url, headers=headers)
        if response.status_code == 200:
            data = response.json()
            market_data = data.get("market_data", {})
            return {
                "price": market_data.get("current_price", {}).get("usd", "N/A"),
                "market_cap": market_data.get("market_cap", {}).get("usd", "N/A"),
                "total_volume": market_data.get("total_volume", {}).get("usd", "N/A")
            }
        elif response.status_code == 429:
            print("Rate limit reached. Waiting 60 seconds before retrying...")
            time.sleep(60)  # Wait longer to respect rate limit
        else:
            return {"error": f"API request failed with status code {response.status_code}"}

def process_csv(file_path, output_path, api_key):
    df = pd.read_csv(file_path)
    results = []
    request_count = 0
    
    for index, row in df.iterrows():
        if row['wasRekt'] != 1:
            continue  # Only process rows where wasRekt = 1
        
        if request_count >= 30:
            print("Rate limit reached, waiting 60 seconds...")
            time.sleep(60)
            request_count = 0
        
        crypto_id = row['id']
        date_str = str(row['Date'])
        
        try:
            target_date = datetime.strptime(date_str, '%m.%d.%Y') - timedelta(days=1)  # One day before
            crypto_data = get_historical_crypto_data(crypto_id, target_date, api_key)
            row["price"] = crypto_data.get("price")
            row["market_cap"] = crypto_data.get("market_cap")
            row["volume"] = crypto_data.get("total_volume")
            results.append(row)
            request_count += 1
        except Exception as e:
            print(f"Error processing {crypto_id} on {date_str}: {str(e)}")
    
    result_df = pd.DataFrame(results)
    result_df.to_csv(output_path, index=False)
    print(f"Updated CSV saved to {output_path}")
    return result_df

# Example Usage
if __name__ == "__main__":
    API_KEY = "CG-7nM4znZ3HbrPjpBWWq8HpU1u"  # Replace with actual API key
    FILE_PATH = "C:/Users/pcagm/Downloads/merged_backup_3_3_0052.csv"
    OUTPUT_PATH = "C:/Users/pcagm/Downloads/rekt_filled.csv"
    final_data = process_csv(FILE_PATH, OUTPUT_PATH, API_KEY)
    print(final_data)
```
