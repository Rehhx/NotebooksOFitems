```python
# orb_15m_retest_all_parquet_BALANCED.py
# ORB 15m Retest with "balanced" quality mode to increase trades while avoiding overfit.
# Keeps the core idea simple: break → retest → enter. Fewer hard filters, wider gates.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path
from math import sqrt, isfinite

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
SAVE_DIR    = Path(".")
TZ_LOCAL    = "America/New_York"

# Session times
REG_START = "09:30"
OR_END    = "09:45"
REG_END   = "16:00"

# Trade timing guardrails (kept strict; robust, not curve-fitty)
TRADE_START = "09:46"
TRADE_END   = "15:50"
MAX_RETEST_MIN = 240
RETEST_OVERSHOOT_FRAC = 0.60   # allow deeper kisses of the level to get filled

# Quality mode: "strict" | "balanced" | "loose"
QUALITY_MODE = "balanced"

# Quality knobs (relaxed vs prior)
LOOKBACK_DAYS_OR = 60
MIN_OR_PCTILE = 5.0
MAX_OR_PCTILE = 95.0

REQUIRE_VOL_SPIKE = False     # OFF by default (too selective intraday)
VOL_SPIKE_MULT = 1.10

REQUIRE_VWAP_ALIGN = True     # keep it, but make it softer (either breakout OR entry)
USE_VWAP_SLOPE = False        # slope was too restrictive for many symbols
VWAP_SLOPE_WINDOW = 5

# Exits
EXIT_POLICY = "scaleout_be"   # keeps it simple and improves Win% a bit
RR_TARGET = 2.0
SCALEOUT_R1 = 1.0
ATR_LEN = 20
TRAIL_K_ATR = 2.0
STOP_FIRST_IF_BOTH_HIT = True

# Annualization
TRADING_DAYS_PER_YEAR = 252

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

def hm(ts): return ts.strftime("%H:%M")

# ========= INDICATORS =========
def add_intraday_vwap(day: pd.DataFrame) -> pd.DataFrame:
    pv = day["Close"] * day["Volume"].fillna(0)
    cv = pv.cumsum()
    vv = day["Volume"].fillna(0).cumsum()
    vwap = np.where(vv > 0, cv / vv, np.nan)
    day["VWAP"] = vwap
    day["VWAP_slope"] = pd.Series(vwap).diff(VWAP_SLOPE_WINDOW)
    return day

def add_atr_intraday(day: pd.DataFrame, prev_close: float) -> pd.DataFrame:
    close_shift = day["Close"].shift(1, fill_value=prev_close)
    tr = pd.concat([
        (day["High"] - day["Low"]).abs(),
        (day["High"] - close_shift).abs(),
        (day["Low"] - close_shift).abs()
    ], axis=1).max(axis=1)
    day["TR"] = tr
    day["ATR"] = tr.rolling(ATR_LEN, min_periods=1).mean()
    return day

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def simulate_exit_path(after_entry_df: pd.DataFrame,
                       side: str, entry: float, stop_init: float,
                       r_target_px: float, or_width: float,
                       policy: str,
                       trail_k_atr: float,
                       stop_first: bool) -> tuple:
    curr_stop = stop_init
    tp_px = r_target_px
    hit_scaleout = False
    pnl_accum = 0.0

    for idx, row in after_entry_df.iterrows():
        px = float(row["Close"])
        hh = float(row["High"])
        ll = float(row["Low"])
        atr = float(row.get("ATR", np.nan))

        # Scale-out + BE policy
        if policy in ("scaleout_be", "scaleout_be_trail"):
            # +1R level
            tp1 = entry + (or_width if side == "long" else -or_width)
            # check if +1R touched this bar
            touched_1R = (hh >= tp1) if side == "long" else (ll <= tp1)
            if touched_1R and not hit_scaleout:
                pnl_accum += 0.5 * ((tp1 - entry) if side == "long" else (entry - tp1))
                hit_scaleout = True
                curr_stop = entry  # move to BE

        # Optional chandelier trail after BE (disabled in default policy)
        if policy == "scaleout_be_trail" and hit_scaleout and np.isfinite(atr):
            if side == "long":
                chand = hh - trail_k_atr * atr
                curr_stop = max(curr_stop, chand, entry)
            else:
                chand = ll + trail_k_atr * atr
                curr_stop = min(curr_stop, chand, entry)

        # Check stop/tp hits
        hit_stop = (ll <= curr_stop) if side == "long" else (hh >= curr_stop)
        hit_tp   = (hh >= tp_px) if side == "long" else (ll <= tp_px)

        if hit_stop and hit_tp:
            if stop_first:
                pnl_accum += (0.5 if hit_scaleout else 1.0) * ((curr_stop - entry) if side == "long" else (entry - curr_stop))
                return "STOP", curr_stop, idx, pnl_accum
            else:
                pnl_accum += (0.5 if hit_scaleout else 1.0) * ((tp_px - entry) if side == "long" else (entry - tp_px))
                return "TP", tp_px, idx, pnl_accum

        if hit_tp:
            pnl_accum += (0.5 if hit_scaleout else 1.0) * ((tp_px - entry) if side == "long" else (entry - tp_px))
            return "TP", tp_px, idx, pnl_accum

        if hit_stop:
            pnl_accum += (0.5 if hit_scaleout else 1.0) * ((curr_stop - entry) if side == "long" else (entry - curr_stop))
            return "STOP", curr_stop, idx, pnl_accum

    # EOD
    last_idx = after_entry_df.index[-1]
    last_px  = float(after_entry_df.loc[last_idx, "Close"])
    pnl_accum += (0.5 if hit_scaleout else 1.0) * ((last_px - entry) if side == "long" else (entry - last_px))
    return "EOD", last_px, last_idx, pnl_accum

def calc_basic_metrics(trades_df):
    pnl = trades_df["PnL_$"].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= QUANT-STANDARD (daily PnL) =========
def _max_drawdown_from_series(equity: pd.Series) -> float:
    roll_max = equity.cummax()
    drawdown = equity - roll_max
    return float(-drawdown.min()) if len(equity) else 0.0

def _sortino_ratio(daily_pnl: pd.Series) -> float:
    if daily_pnl.empty: return 0.0
    mean = daily_pnl.mean()
    dstd = daily_pnl[daily_pnl < 0].std(ddof=1)
    if dstd == 0 or np.isnan(dstd): return (np.inf if mean > 0 else 0.0)
    return float((mean / dstd) * sqrt(TRADING_DAYS_PER_YEAR))

def _sharpe_ratio(daily_pnl: pd.Series) -> float:
    if daily_pnl.empty: return 0.0
    mu = daily_pnl.mean()
    sd = daily_pnl.std(ddof=1)
    if sd == 0 or np.isnan(sd): return (np.inf if mu > 0 else 0.0)
    return float((mu / sd) * sqrt(TRADING_DAYS_PER_YEAR))

def _t_stat_of_mean(daily_pnl: pd.Series) -> float:
    n = len(daily_pnl)
    if n < 2: return 0.0
    mu = daily_pnl.mean()
    sd = daily_pnl.std(ddof=1)
    if sd == 0 or np.isnan(sd): return (np.inf if mu > 0 else 0.0)
    return float(mu / (sd / sqrt(n)))

def _cvar(daily_pnl: pd.Series, alpha: float = 0.05) -> float:
    if daily_pnl.empty: return 0.0
    q = daily_pnl.quantile(alpha)
    tail = daily_pnl[daily_pnl <= q]
    return float(tail.mean()) if not tail.empty else 0.0

def quant_indicators(trades_df: pd.DataFrame):
    if trades_df.empty:
        return {k: 0.0 for k in [
            "Sharpe_daily","Sortino_daily","MaxDD_$","Calmar","t_stat_daily_mean",
            "CVaR_5%","CVaR_1%","Skew_daily","ExKurt_daily"
        ]}
    daily = trades_df.groupby("Date", as_index=True)["PnL_$"].sum().sort_index()
    sharpe = _sharpe_ratio(daily)
    sortino = _sortino_ratio(daily)
    equity = daily.cumsum()
    maxdd = _max_drawdown_from_series(equity)
    ann_pnl = float(daily.mean() * TRADING_DAYS_PER_YEAR)
    calmar = (ann_pnl / maxdd) if maxdd > 0 else np.inf
    t_stat = _t_stat_of_mean(daily)
    cvar5 = _cvar(daily, 0.05)
    cvar1 = _cvar(daily, 0.01)
    skew = float(daily.skew()) if len(daily) > 2 else 0.0
    exkurt = float(daily.kurt()) if len(daily) > 3 else 0.0
    return {
        "Sharpe_daily": float(sharpe if isfinite(sharpe) else 0.0),
        "Sortino_daily": float(sortino if isfinite(sortino) else 0.0),
        "MaxDD_$": float(maxdd),
        "Calmar": float(calmar if isfinite(calmar) else 0.0),
        "t_stat_daily_mean": float(t_stat if isfinite(t_stat) else 0.0),
        "CVaR_5%": float(cvar5),
        "CVaR_1%": float(cvar1),
        "Skew_daily": float(skew),
        "ExKurt_daily": float(exkurt),
    }

# ========= CORE BACKTEST =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close", "Volume"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")

    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Session slice
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()
    if df.empty:
        return pd.DataFrame()

    # Precompute day groups
    day_groups = list(df.groupby("date"))
    or_width_hist = []
    or_vol_hist   = []

    records = []

    for d, day in day_groups:
        if day.empty: continue

        # OR window
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any(): continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if not np.isfinite(or_width) or or_width <= 0: continue

        or_vol_today = float(day.loc[or_mask, "Volume"].sum())

        # Lookback stats (prior days only)
        prior_widths = np.array(or_width_hist[-LOOKBACK_DAYS_OR:]) if or_width_hist else np.array([])
        prior_vols   = np.array(or_vol_hist[-LOOKBACK_DAYS_OR:])   if or_vol_hist else np.array([])

        # Percentiles (enabled only with enough history)
        or_ok, p_lo, p_hi = True, np.nan, np.nan
        if prior_widths.size >= 30:  # need enough to be meaningful but not tight
            p_lo = float(np.percentile(prior_widths, MIN_OR_PCTILE))
            p_hi = float(np.percentile(prior_widths, MAX_OR_PCTILE))
            or_ok = (or_width >= p_lo) and (or_width <= p_hi)

        vol_ok, med_or_vol = True, np.nan
        if REQUIRE_VOL_SPIKE and prior_vols.size >= 30:
            med_or_vol = float(np.median(prior_vols))
            vol_ok = (or_vol_today >= VOL_SPIKE_MULT * med_or_vol)

        # Stash today's stats for *future* days
        or_width_hist.append(or_width)
        or_vol_hist.append(or_vol_today)

        # Indicators
        pre_or = day.loc[day["t"] < REG_START]
        prev_close = float(pre_or["Close"].iloc[-1]) if not pre_or.empty else float(day["Close"].iloc[0])
        day = add_intraday_vwap(day)
        day = add_atr_intraday(day, prev_close)

        after_or = day.loc[day["t"] >= OR_END].copy()
        if after_or.empty: continue

        # First *close* breakout
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None: continue

        # VWAP checks (softened)
        def vwap_pass(idx):
            row = day.loc[idx]
            px = float(row["Close"])
            v  = float(row["VWAP"])
            if not USE_VWAP_SLOPE:
                return (px >= v) if side == "long" else (px <= v)
            else:
                vs = float(row["VWAP_slope"])
                return ((px >= v) and (vs >= 0)) if side == "long" else ((px <= v) and (vs <= 0))

        vwap_ok_break = (vwap_pass(brk_idx) if REQUIRE_VWAP_ALIGN else True)

        # Find retest
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty: continue

        level = orh if side == "long" else orl
        retest_idx = first_retest_touch(after_breakout_close, level)
        if retest_idx is None: continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry_hm = hm(entry_time)

        # Guardrails
        time_ok = (entry_hm >= TRADE_START) and (entry_hm <= TRADE_END)
        minutes_since_breakout = (entry_time - day.loc[brk_idx, "EventAt"]).total_seconds() / 60.0
        delay_ok = (minutes_since_breakout <= MAX_RETEST_MIN)

        entry_bar = day.loc[retest_idx]
        overshoot_ok = True
        if side == "long":
            overshoot_ok = (level - float(entry_bar["Low"])) <= (RETEST_OVERSHOOT_FRAC * or_width)
        else:
            overshoot_ok = (float(entry_bar["High"]) - level) <= (RETEST_OVERSHOOT_FRAC * or_width)

        vwap_ok_entry = (vwap_pass(retest_idx) if REQUIRE_VWAP_ALIGN else True)

        # ====== QUALITY GATE (non-overfit, higher throughput) ======
        # Always require robust stuff:
        robust_ok = time_ok and delay_ok and overshoot_ok
        if not robust_ok:
            records.append({
                "Date": d, "Symbol": symbol, "Side": side, "Outcome": "FILTERED",
                "ORH": orh, "ORL": orl, "OR_Width": or_width, "OR_pLo": p_lo, "OR_pHi": p_hi,
                "OR_Vol_Today": or_vol_today, "OR_Vol_Med": med_or_vol,
                "VWAP_OK_brk": vwap_ok_break, "VWAP_OK_entry": vwap_ok_entry,
                "Time_OK": time_ok, "Delay_OK": delay_ok, "Overshoot_OK": overshoot_ok,
                "PnL_$": 0.0
            })
            continue

        if QUALITY_MODE == "strict":
            quality_pass = (or_ok and vol_ok and vwap_ok_break and vwap_ok_entry)
        elif QUALITY_MODE == "loose":
            # pass if ANY of the soft signals is true
            soft_any = (or_ok or vwap_ok_break or vwap_ok_entry)
            quality_pass = soft_any
        else:  # "balanced"
            # pass if OR-width is reasonable OR (VWAP ok at breakout OR entry)
            soft_ok = (or_ok or vwap_ok_break or vwap_ok_entry)
            # ignoring vol_ok by default; turn it back on if you want it stricter
            quality_pass = soft_ok

        if not quality_pass:
            records.append({
                "Date": d, "Symbol": symbol, "Side": side, "Outcome": "FILTERED",
                "ORH": orh, "ORL": orl, "OR_Width": or_width, "OR_pLo": p_lo, "OR_pHi": p_hi,
                "OR_Vol_Today": or_vol_today, "OR_Vol_Med": med_or_vol,
                "VWAP_OK_brk": vwap_ok_break, "VWAP_OK_entry": vwap_ok_entry,
                "Time_OK": time_ok, "Delay_OK": delay_ok, "Overshoot_OK": overshoot_ok,
                "PnL_$": 0.0
            })
            continue

        # Construct trade
        entry = level
        stop  = orl if side == "long" else orh
        target_px = entry + (RR_TARGET * or_width if side == "long" else -RR_TARGET * or_width)
        after_entry = day.loc[day.index > retest_idx].copy()
        if after_entry.empty:
            records.append({
                "Date": d, "Symbol": symbol, "Side": side, "Outcome": "EOD",
                "ORH": orh, "ORL": orl, "OR_Width": or_width,
                "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target": target_px,
                "ExitPx": entry, "ExitTime": entry_time, "PnL_$": 0.0
            })
            continue

        # Add ATR for possible future use; not needed for scaleout_be
        outcome, exit_px, exit_idx, pnl = simulate_exit_path(
            after_entry, side, entry, stop, target_px, or_width,
            EXIT_POLICY, TRAIL_K_ATR, STOP_FIRST_IF_BOTH_HIT
        )

        records.append({
            "Date": d, "Symbol": symbol, "Side": side, "Outcome": outcome,
            "ORH": orh, "ORL": orl, "OR_Width": or_width,
            "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target": target_px,
            "ExitPx": exit_px, "ExitTime": day.loc[exit_idx, "EventAt"] if exit_idx in day.index else None,
            "PnL_$": float(pnl)
        })

    return pd.DataFrame(records)

# ========= RUN ALL FILES =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return
    SAVE_DIR.mkdir(parents=True, exist_ok=True)

    all_trades, metrics_rows = [], []

    for f in files:
        symbol = f.stem.upper()
        try:
            trades = backtest_one_symbol(f, symbol)
        except Exception as e:
            print(f"[SKIP] {symbol}: {e}")
            continue

        # Executed trades only (exclude FILTERED diagnostics)
        exec_trades = trades[~trades["Outcome"].eq("FILTERED")].copy()
        if exec_trades.empty:
            print(f"=== ORB 15m Retest — {symbol} (BALANCED) ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            print("— Risk & Quality (daily) — Sharpe: 0.00 | Sortino: 0.00 | MaxDD: $0.00 | Calmar: 0.00 | t-stat: 0.00 | CVaR5%: $0.00 | CVaR1%: $0.00 | Skew: 0.00 | ExKurt: 0.00")
            continue

        exec_trades["Date"] = pd.to_datetime(exec_trades["Date"])

        # Save both
        (SAVE_DIR / f"orb_trades_{symbol}_BAL_full.csv").write_text(trades.to_csv(index=False))
        save_path = SAVE_DIR / f"orb_trades_{symbol}_BAL_exec.csv"
        exec_trades.to_csv(save_path, index=False)

        # Metrics
        t, pnl, wr, pf = calc_basic_metrics(exec_trades)
        print(f"=== ORB 15m Retest — {symbol} (BALANCED) ===")
        print(f"Trades: {t} | Total PnL: ${pnl:,.2f} | Win%: {wr:.2f}% | Profit Factor: {pf:.3f}")
        print(f"Saved trades (exec) → {save_path.resolve()}")

        qi = quant_indicators(exec_trades)
        print(
            "— Risk & Quality (daily) — "
            f"Sharpe: {qi['Sharpe_daily']:.3f} | Sortino: {qi['Sortino_daily']:.3f} | "
            f"MaxDD: ${qi['MaxDD_$']:,.2f} | Calmar: {qi['Calmar']:.3f} | "
            f"t-stat: {qi['t_stat_daily_mean']:.3f} | "
            f"CVaR5%: ${qi['CVaR_5%']:,.2f} | CVaR1%: ${qi['CVaR_1%']:,.2f} | "
            f"Skew: {qi['Skew_daily']:.3f} | ExKurt: {qi['ExKurt_daily']:.3f}"
        )

        all_trades.append(exec_trades)
        metrics_rows.append({"Symbol": symbol, "Trades": t, "TotalPnL_$": pnl, "WinRate_%": wr, "ProfitFactor": pf, **qi})

    if not all_trades:
        print("\nNo executed trades generated for any symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    t, pnl, wr, pf = calc_basic_metrics(overall)
    save_all = SAVE_DIR / "orb_trades_ALL_BAL_exec.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== ORB 15m Retest — OVERALL (BALANCED) ===")
    print(f"Trades: {t} | Total PnL: ${pnl:,.2f} | Win%: {wr:.2f}% | Profit Factor: {pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

    oq = quant_indicators(overall)
    print(
        "— Risk & Quality (daily) — "
        f"Sharpe: {oq['Sharpe_daily']:.3f} | Sortino: {oq['Sortino_daily']:.3f} | "
        f"MaxDD: ${oq['MaxDD_$']:,.2f} | Calmar: {oq['Calmar']:.3f} | "
        f"t-stat: {oq['t_stat_daily_mean']:.3f} | "
        f"CVaR5%: ${oq['CVaR_5%']:,.2f} | CVaR1%: ${oq['CVaR_1%']:,.2f} | "
        f"Skew: {oq['Skew_daily']:.3f} | ExKurt: {oq['ExKurt_daily']:.3f}"
    )

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_all_parquet_BALANCED_plus_NEAR_and_FIXEDRR.py
# Goals:
#   • More trades w/o overfit: allow NEAR-retest fills (within δ of the level)
#   • More PnL: default EXIT_POLICY = "fixed_rr" (keep trend-day winners)
#   • Keep robust guardrails (time window, retest delay, overshoot)

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path
from math import sqrt, isfinite

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
SAVE_DIR    = Path(".")
TZ_LOCAL    = "America/New_York"

# Session times
REG_START = "09:30"
OR_END    = "09:45"
REG_END   = "16:00"

# Guardrails (robust, not curve-fit)
TRADE_START = "09:46"
TRADE_END   = "15:55"
MAX_RETEST_MIN = 999                 # allow the whole session (max throughput)
RETEST_OVERSHOOT_FRAC = 0.60         # allow deeper “kiss” to get filled

# NEAR-retest (new): accept retests that come within δ * OR_width of the level
RETEST_NEAR_FRAC = 0.10              # 10% of OR width
RETEST_MODE = "touch_or_near"        # "touch_only" | "touch_or_near"

# Optional: if no retest within deadline, enter on breakout close to boost trades
CONTINUATION_FALLBACK = False        # set True to add more trades (momentum style)

# Quality mode
QUALITY_MODE = "balanced"            # "strict" | "balanced" | "loose"

# OR width distribution (generic)
LOOKBACK_DAYS_OR = 60
MIN_OR_PCTILE = 5.0
MAX_OR_PCTILE = 95.0

# VWAP (kept soft; still generic)
REQUIRE_VWAP_ALIGN = True            # require px vs VWAP at breakout OR entry
USE_VWAP_SLOPE = False               # slope is often too restrictive intraday
VWAP_SLOPE_WINDOW = 5

# Volume spike filter (off by default to avoid undertrading)
REQUIRE_VOL_SPIKE = False
VOL_SPIKE_MULT = 1.10

# Exits (default = fixed 2R to restore PnL on trend names like AMZN)
EXIT_POLICY = "fixed_rr"             # "fixed_rr" | "scaleout_be" | "scaleout_be_trail"
RR_TARGET = 2.0
SCALEOUT_R1 = 1.0
SCALEOUT_FRACTION = 0.33             # if you use scaleout, keep more runner than 50/50
ATR_LEN = 20
TRAIL_K_ATR = 2.0
STOP_FIRST_IF_BOTH_HIT = True

# Annualization
TRADING_DAYS_PER_YEAR = 252

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

def hm(ts): return ts.strftime("%H:%M")

# ========= INDICATORS =========
def add_intraday_vwap(day: pd.DataFrame) -> pd.DataFrame:
    pv = day["Close"] * day["Volume"].fillna(0)
    cv = pv.cumsum()
    vv = day["Volume"].fillna(0).cumsum()
    vwap = np.where(vv > 0, cv / vv, np.nan)
    day["VWAP"] = vwap
    day["VWAP_slope"] = pd.Series(vwap).diff(VWAP_SLOPE_WINDOW)
    return day

def add_atr_intraday(day: pd.DataFrame, prev_close: float) -> pd.DataFrame:
    close_shift = day["Close"].shift(1, fill_value=prev_close)
    tr = pd.concat([
        (day["High"] - day["Low"]).abs(),
        (day["High"] - close_shift).abs(),
        (day["Low"] - close_shift).abs()
    ], axis=1).max(axis=1)
    day["TR"] = tr
    day["ATR"] = tr.rolling(ATR_LEN, min_periods=1).mean()
    return day

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch_or_near(df_after_breakout_close, level, or_width, near_frac):
    """
    Return first index that either touches the level or comes within 'near_frac * or_width'.
    Distance to the [Low,High] interval is 0 if it spans the level; else min(|level-Low|, |level-High|).
    """
    lo = df_after_breakout_close["Low"].values
    hi = df_after_breakout_close["High"].values
    spans = (lo <= level) & (hi >= level)
    dist  = np.where(spans, 0.0, np.minimum(np.abs(lo - level), np.abs(hi - level)))
    near  = dist <= (near_frac * or_width)
    if not near.any():
        return None, False  # idx, touched?
    idx = df_after_breakout_close.index[np.argmax(near)]
    touched = spans[np.argmax(near)]
    return idx, bool(touched)

def simulate_exit_path(after_entry_df: pd.DataFrame,
                       side: str, entry: float, stop_init: float,
                       r_target_px: float, risk_R: float,
                       policy: str,
                       trail_k_atr: float,
                       scaleout_fraction: float,
                       stop_first: bool) -> tuple:
    """
    Returns outcome, exit_px, exit_idx, pnl_dollars.
    risk_R is the absolute $ distance from entry to stop (>=0).
    """
    curr_stop = stop_init
    tp_px = r_target_px
    hit_scaleout = False
    pnl_accum = 0.0

    for idx, row in after_entry_df.iterrows():
        px = float(row["Close"])
        hh = float(row["High"])
        ll = float(row["Low"])
        atr = float(row.get("ATR", np.nan))

        if policy in ("scaleout_be", "scaleout_be_trail"):
            tp1 = entry + (risk_R if side == "long" else -risk_R)
            touched_1R = (hh >= tp1) if side == "long" else (ll <= tp1)
            if touched_1R and not hit_scaleout:
                pnl_accum += scaleout_fraction * ((tp1 - entry) if side == "long" else (entry - tp1))
                hit_scaleout = True
                curr_stop = entry  # move to breakeven

        if policy == "scaleout_be_trail" and hit_scaleout and np.isfinite(atr):
            if side == "long":
                chand = hh - trail_k_atr * atr
                curr_stop = max(curr_stop, chand, entry)
            else:
                chand = ll + trail_k_atr * atr
                curr_stop = min(curr_stop, chand, entry)

        hit_stop = (ll <= curr_stop) if side == "long" else (hh >= curr_stop)
        hit_tp   = (hh >= tp_px) if side == "long" else (ll <= tp_px)

        if hit_stop and hit_tp:
            if stop_first:
                pnl_accum += (1.0 - (scaleout_fraction if hit_scaleout else 0.0)) * ((curr_stop - entry) if side == "long" else (entry - curr_stop))
                return "STOP", curr_stop, idx, pnl_accum
            else:
                pnl_accum += (1.0 - (scaleout_fraction if hit_scaleout else 0.0)) * ((tp_px - entry) if side == "long" else (entry - tp_px))
                return "TP", tp_px, idx, pnl_accum

        if hit_tp:
            pnl_accum += (1.0 - (scaleout_fraction if hit_scaleout else 0.0)) * ((tp_px - entry) if side == "long" else (entry - tp_px))
            return "TP", tp_px, idx, pnl_accum

        if hit_stop:
            pnl_accum += (1.0 - (scaleout_fraction if hit_scaleout else 0.0)) * ((curr_stop - entry) if side == "long" else (entry - curr_stop))
            return "STOP", curr_stop, idx, pnl_accum

    last_idx = after_entry_df.index[-1]
    last_px  = float(after_entry_df.loc[last_idx, "Close"])
    pnl_accum += (1.0 - (scaleout_fraction if hit_scaleout else 0.0)) * ((last_px - entry) if side == "long" else (entry - last_px))
    return "EOD", last_px, last_idx, pnl_accum

def calc_basic_metrics(trades_df):
    pnl = trades_df["PnL_$"].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= QUANT-STANDARD (daily PnL) =========
def _max_drawdown_from_series(equity: pd.Series) -> float:
    roll_max = equity.cummax()
    drawdown = equity - roll_max
    return float(-drawdown.min()) if len(equity) else 0.0

def _sortino_ratio(daily_pnl: pd.Series) -> float:
    if daily_pnl.empty: return 0.0
    mean = daily_pnl.mean()
    dstd = daily_pnl[daily_pnl < 0].std(ddof=1)
    if dstd == 0 or np.isnan(dstd): return (np.inf if mean > 0 else 0.0)
    return float((mean / dstd) * sqrt(TRADING_DAYS_PER_YEAR))

def _sharpe_ratio(daily_pnl: pd.Series) -> float:
    if daily_pnl.empty: return 0.0
    mu = daily_pnl.mean()
    sd = daily_pnl.std(ddof=1)
    if sd == 0 or np.isnan(sd): return (np.inf if mu > 0 else 0.0)
    return float((mu / sd) * sqrt(TRADING_DAYS_PER_YEAR))

def _t_stat_of_mean(daily_pnl: pd.Series) -> float:
    n = len(daily_pnl)
    if n < 2: return 0.0
    mu = daily_pnl.mean()
    sd = daily_pnl.std(ddof=1)
    if sd == 0 or np.isnan(sd): return (np.inf if mu > 0 else 0.0)
    return float(mu / (sd / sqrt(n)))

def _cvar(daily_pnl: pd.Series, alpha: float = 0.05) -> float:
    if daily_pnl.empty: return 0.0
    q = daily_pnl.quantile(alpha)
    tail = daily_pnl[daily_pnl <= q]
    return float(tail.mean()) if not tail.empty else 0.0

def quant_indicators(trades_df: pd.DataFrame):
    if trades_df.empty:
        return {k: 0.0 for k in [
            "Sharpe_daily","Sortino_daily","MaxDD_$","Calmar","t_stat_daily_mean",
            "CVaR_5%","CVaR_1%","Skew_daily","ExKurt_daily"
        ]}
    daily = trades_df.groupby("Date", as_index=True)["PnL_$"].sum().sort_index()
    sharpe = _sharpe_ratio(daily)
    sortino = _sortino_ratio(daily)
    equity = daily.cumsum()
    maxdd = _max_drawdown_from_series(equity)
    ann_pnl = float(daily.mean() * TRADING_DAYS_PER_YEAR)
    calmar = (ann_pnl / maxdd) if maxdd > 0 else np.inf
    t_stat = _t_stat_of_mean(daily)
    cvar5 = _cvar(daily, 0.05)
    cvar1 = _cvar(daily, 0.01)
    skew = float(daily.skew()) if len(daily) > 2 else 0.0
    exkurt = float(daily.kurt()) if len(daily) > 3 else 0.0
    return {
        "Sharpe_daily": float(sharpe if isfinite(sharpe) else 0.0),
        "Sortino_daily": float(sortino if isfinite(sortino) else 0.0),
        "MaxDD_$": float(maxdd),
        "Calmar": float(calmar if isfinite(calmar) else 0.0),
        "t_stat_daily_mean": float(t_stat if isfinite(t_stat) else 0.0),
        "CVaR_5%": float(cvar5),
        "CVaR_1%": float(cvar1),
        "Skew_daily": float(skew),
        "ExKurt_daily": float(exkurt),
    }

# ========= CORE BACKTEST =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close", "Volume"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")

    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()
    if df.empty:
        return pd.DataFrame()

    day_groups = list(df.groupby("date"))
    or_width_hist, or_vol_hist = [], []

    records = []

    for d, day in day_groups:
        if day.empty: continue

        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any(): continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if not np.isfinite(or_width) or or_width <= 0: continue

        or_vol_today = float(day.loc[or_mask, "Volume"].sum())

        # Prior stats for simple percentile gating
        prior_widths = np.array(or_width_hist[-LOOKBACK_DAYS_OR:]) if or_width_hist else np.array([])
        prior_vols   = np.array(or_vol_hist[-LOOKBACK_DAYS_OR:])   if or_vol_hist else np.array([])

        or_ok, p_lo, p_hi = True, np.nan, np.nan
        if prior_widths.size >= 30:
            p_lo = float(np.percentile(prior_widths, MIN_OR_PCTILE))
            p_hi = float(np.percentile(prior_widths, MAX_OR_PCTILE))
            or_ok = (or_width >= p_lo) and (or_width <= p_hi)

        vol_ok, med_or_vol = True, np.nan
        if REQUIRE_VOL_SPIKE and prior_vols.size >= 30:
            med_or_vol = float(np.median(prior_vols))
            vol_ok = (or_vol_today >= VOL_SPIKE_MULT * med_or_vol)

        # stash for future days
        or_width_hist.append(or_width)
        or_vol_hist.append(or_vol_today)

        # Indicators
        pre_or = day.loc[day["t"] < REG_START]
        prev_close = float(pre_or["Close"].iloc[-1]) if not pre_or.empty else float(day["Close"].iloc[0])
        day = add_intraday_vwap(day)
        day = add_atr_intraday(day, prev_close)

        after_or = day.loc[day["t"] >= OR_END].copy()
        if after_or.empty: continue

        # Breakout close
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # Soft VWAP checks
        def vwap_pass(idx):
            row = day.loc[idx]
            px = float(row["Close"])
            v  = float(row["VWAP"])
            if not USE_VWAP_SLOPE:
                return (px >= v) if side == "long" else (px <= v)
            else:
                vs = float(row["VWAP_slope"])
                return ((px >= v) and (vs >= 0)) if side == "long" else ((px <= v) and (vs <= 0))

        vwap_ok_break = (vwap_pass(brk_idx) if REQUIRE_VWAP_ALIGN else True)

        # Retest (touch or near)
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        level = orh if side == "long" else orl

        if RETEST_MODE == "touch_only":
            # legacy touch
            touched = (after_breakout_close["Low"] <= level) & (after_breakout_close["High"] >= level)
            retest_idx = after_breakout_close.index[touched].min() if touched.any() else None
            touched_flag = True if retest_idx is not None else False
        else:
            res = first_retest_touch_or_near(after_breakout_close, level, or_width, RETEST_NEAR_FRAC)
            retest_idx, touched_flag = res if res is not None else (None, False)

        # Optional continuation fallback if no retest at all
        if retest_idx is None and CONTINUATION_FALLBACK:
            retest_idx = brk_idx
            touched_flag = False  # we’ll “enter” on the breakout close below

        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry_hm = hm(entry_time)

        time_ok = (entry_hm >= TRADE_START) and (entry_hm <= TRADE_END)
        minutes_since_breakout = (entry_time - day.loc[brk_idx, "EventAt"]).total_seconds() / 60.0
        delay_ok = (minutes_since_breakout <= MAX_RETEST_MIN)

        entry_bar = day.loc[retest_idx]
        # Overshoot tolerance: only count depth below/above, ignore negative depth
        if side == "long":
            depth = max(level - float(entry_bar["Low"]), 0.0)
        else:
            depth = max(float(entry_bar["High"]) - level, 0.0)
        overshoot_ok = (depth <= (RETEST_OVERSHOOT_FRAC * or_width))

        vwap_ok_entry = (vwap_pass(retest_idx) if REQUIRE_VWAP_ALIGN else True)

        # Robust gate
        if not (time_ok and delay_ok and overshoot_ok):
            records.append({
                "Date": d, "Symbol": symbol, "Side": side, "Outcome": "FILTERED",
                "PnL_$": 0.0, "Reason": "robust_gate_fail"
            })
            continue

        # Soft gate per QUALITY_MODE
        if QUALITY_MODE == "strict":
            soft_ok = (or_ok and vwap_ok_break and vwap_ok_entry and (vol_ok))
        elif QUALITY_MODE == "loose":
            soft_ok = (or_ok or vwap_ok_break or vwap_ok_entry)
        else:
            soft_ok = (or_ok or vwap_ok_break or vwap_ok_entry)

        if not soft_ok:
            records.append({
                "Date": d, "Symbol": symbol, "Side": side, "Outcome": "FILTERED",
                "PnL_$": 0.0, "Reason": "soft_gate_fail"
            })
            continue

        # ===== Entry price =====
        # If the bar actually touched the level, assume limit at the level fills there.
        # If it only came NEAR (no touch), enter at the bar's CLOSE to stay realistic.
        if touched_flag:
            entry = level
        else:
            entry = float(entry_bar["Close"])

        stop  = orl if side == "long" else orh
        # per-trade R in dollars from actual entry
        risk_R = abs(entry - stop)
        if risk_R == 0 or not np.isfinite(risk_R):
            continue

        target_px = entry + (RR_TARGET * risk_R if side == "long" else -RR_TARGET * risk_R)

        after_entry = day.loc[day.index > retest_idx].copy()
        if after_entry.empty:
            records.append({
                "Date": d, "Symbol": symbol, "Side": side, "Outcome": "EOD",
                "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target": target_px,
                "ExitPx": entry, "ExitTime": entry_time, "PnL_$": 0.0
            })
            continue

        outcome, exit_px, exit_idx, pnl = simulate_exit_path(
            after_entry, side, entry, stop, target_px, risk_R,
            EXIT_POLICY, TRAIL_K_ATR, SCALEOUT_FRACTION, STOP_FIRST_IF_BOTH_HIT
        )

        records.append({
            "Date": d, "Symbol": symbol, "Side": side, "Outcome": outcome,
            "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target": target_px,
            "ExitPx": exit_px, "ExitTime": day.loc[exit_idx, "EventAt"] if exit_idx in day.index else None,
            "PnL_$": float(pnl)
        })

    return pd.DataFrame(records)

# ========= RUN ALL FILES =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return
    SAVE_DIR.mkdir(parents=True, exist_ok=True)

    all_trades, metrics_rows = [], []

    for f in files:
        symbol = f.stem.upper()
        try:
            trades = backtest_one_symbol(f, symbol)
        except Exception as e:
            print(f"[SKIP] {symbol}: {e}")
            continue

        exec_trades = trades[~trades["Outcome"].eq("FILTERED")].copy()
        if exec_trades.empty:
            print(f"=== ORB 15m Retest — {symbol} (NEAR+FIXEDRR) ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            print("— Risk & Quality (daily) — Sharpe: 0.00 | Sortino: 0.00 | MaxDD: $0.00 | Calmar: 0.00 | t-stat: 0.00 | CVaR5%: $0.00 | CVaR1%: $0.00 | Skew: 0.00 | ExKurt: 0.00")
            continue

        exec_trades["Date"] = pd.to_datetime(exec_trades["Date"])

        full_path = SAVE_DIR / f"orb_trades_{symbol}_NEAR_FIXEDRR_full.csv"
        trades.to_csv(full_path, index=False)
        save_path = SAVE_DIR / f"orb_trades_{symbol}_NEAR_FIXEDRR_exec.csv"
        exec_trades.to_csv(save_path, index=False)

        t, pnl, wr, pf = calc_basic_metrics(exec_trades)
        print(f"=== ORB 15m Retest — {symbol} (NEAR+FIXEDRR) ===")
        print(f"Trades: {t} | Total PnL: ${pnl:,.2f} | Win%: {wr:.2f}% | Profit Factor: {pf:.3f}")
        print(f"Saved trades (exec) → {save_path.resolve()}")

        qi = quant_indicators(exec_trades)
        print(
            "— Risk & Quality (daily) — "
            f"Sharpe: {qi['Sharpe_daily']:.3f} | Sortino: {qi['Sortino_daily']:.3f} | "
            f"MaxDD: ${qi['MaxDD_$']:,.2f} | Calmar: {qi['Calmar']:.3f} | "
            f"t-stat: {qi['t_stat_daily_mean']:.3f} | "
            f"CVaR5%: ${qi['CVaR_5%']:,.2f} | CVaR1%: ${qi['CVaR_1%']:,.2f} | "
            f"Skew: {qi['Skew_daily']:.3f} | ExKurt: {qi['ExKurt_daily']:.3f}"
        )

        all_trades.append(exec_trades)
        metrics_rows.append({"Symbol": symbol, "Trades": t, "TotalPnL_$": pnl, "WinRate_%": wr, "ProfitFactor": pf, **qi})

    if not all_trades:
        print("\nNo executed trades generated for any symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    t, pnl, wr, pf = calc_basic_metrics(overall)
    save_all = SAVE_DIR / "orb_trades_ALL_NEAR_FIXEDRR_exec.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== ORB 15m Retest — OVERALL (NEAR+FIXEDRR) ===")
    print(f"Trades: {t} | Total PnL: ${pnl:,.2f} | Win%: {wr:.2f}% | Profit Factor: {pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

    oq = quant_indicators(overall)
    print(
        "— Risk & Quality (daily) — "
        f"Sharpe: {oq['Sharpe_daily']:.3f} | Sortino: {oq['Sortino_daily']:.3f} | "
        f"MaxDD: ${oq['MaxDD_$']:,.2f} | Calmar: {oq['Calmar']:.3f} | "
        f"t-stat: {oq['t_stat_daily_mean']:.3f} | "
        f"CVaR5%: ${oq['CVaR_5%']:,.2f} | CVaR1%: ${oq['CVaR_1%']:,.2f} | "
        f"Skew: {oq['Skew_daily']:.3f} | ExKurt: {oq['ExKurt_daily']:.3f}"
    )

if __name__ == "__main__":
    main()
```

```python
# multi_intraday_bundle.py
# Runs multiple intraday strategies on minute OHLCV parquet files:
#   1) ORB Retest (touch/near), optional continuation
#   2) ORB Continuation (breakout close)
#   3) VWAP Pullback Trend
#   4) Previous-Day High/Low Retest
#
# Includes:
#   - $ risk per trade sizing
#   - Slippage + fees
#   - One-open-position-per-strategy-per-symbol-per-day
#   - Portfolio rollup & quant-style daily metrics (NET of costs)

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path
from math import sqrt, isfinite

# =================== USER SETTINGS ===================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
SAVE_DIR    = Path(".")

TZ_LOCAL = "America/New_York"
REG_START = "09:30"
OR_END    = "09:45"
REG_END   = "16:00"

# Portfolio / sizing / costs
RISK_PER_TRADE_USD   = 100.0     # fixed $ risk per trade
FEE_PER_SHARE        = 0.003     # per-share fee (commissions + ECN etc.)
FIXED_FEE_PER_TRADE  = 0.00      # flat fee per open/close (applied once per side below)
SLIPPAGE_PER_SHARE   = 0.01      # $ per share each side (entry and exit)

# Reporting
TRADING_DAYS_PER_YEAR = 252

# Enable/disable strategies
ENABLED_STRATEGIES = {
    "ORB_RETEST": True,
    "ORB_CONTINUATION": True,
    "VWAP_PULLBACK": True,
    "PDH_PDL_RETEST": True,
}

# ==== ORB shared knobs ====
TRADE_START = "09:46"
TRADE_END   = "15:50"
MAX_RETEST_MIN = 240
RETEST_OVERSHOOT_FRAC = 0.60
RETEST_NEAR_FRAC = 0.10   # 10% of OR width = "near" band
CONTINUATION_FALLBACK = False  # inside retest: if no retest, can flip to breakout close
RR_TARGET = 2.0

# ==== VWAP Pullback knobs ====
ATR_LEN = 20
VWAP_BAND_ATR = 0.5  # consider "near VWAP" if within 0.5*ATR of VWAP
VWAP_TREND_MIN_SLOPE = 0.0  # slope >= 0 uptrend, <= 0 downtrend
VWAP_TRADE_START = "10:00"   # avoid opening noise
VWAP_TRADE_END   = "15:45"
VWAP_R_MULT_TP   = 2.0       # take-profit 2R default
VWAP_STOP_ATR    = 1.0       # initial stop 1*ATR from entry (vs last swing)

# ==== PDH/PDL knobs ====
PD_TRADE_START = "09:46"
PD_TRADE_END   = "15:45"
PD_MAX_WAIT_MIN = 240
PD_RETEST_NEAR_FRAC = 0.10
PD_R_MULT_TP   = 2.0

# =====================================================

def to_ny(ts):
    ts = pd.to_datetime(ts, errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

def add_time_cols(df):
    df["date"] = df["EventAt"].dt.date
    df["hm"]   = df["EventAt"].dt.strftime("%H:%M")
    return df

def session_slice(df):
    return df[(df["hm"] >= REG_START) & (df["hm"] <= REG_END)].copy()

def add_vwap(day: pd.DataFrame) -> pd.DataFrame:
    pv = day["Close"] * day["Volume"].fillna(0)
    cv = pv.cumsum()
    vv = day["Volume"].fillna(0).cumsum()
    vwap = np.where(vv > 0, cv / vv, np.nan)
    day["VWAP"] = vwap
    # simple slope (n-bar diff)
    day["VWAP_Slope"] = pd.Series(vwap).diff(5)
    return day

def add_atr(day: pd.DataFrame, prev_close: float) -> pd.DataFrame:
    cs = day["Close"].shift(1, fill_value=prev_close)
    tr = pd.concat([
        (day["High"] - day["Low"]).abs(),
        (day["High"] - cs).abs(),
        (day["Low"] - cs).abs()
    ], axis=1).max(axis=1)
    day["TR"] = tr
    day["ATR"] = tr.rolling(ATR_LEN, min_periods=1).mean()
    return day

def shares_for_risk(entry, stop, risk_usd):
    risk_per_share = abs(entry - stop)
    if not np.isfinite(risk_per_share) or risk_per_share <= 0:
        return 0
    return int(np.floor(risk_usd / risk_per_share))

def apply_slippage(side, px, which):
    """
    which: 'entry' or 'exit'
    For longs: entry pays +slip; exit receives -slip (worse).
    For shorts: entry receives -slip; exit pays +slip (worse).
    """
    s = SLIPPAGE_PER_SHARE
    if side == "long":
        return px + s if which == "entry" else px - s
    else:
        return px - s if which == "entry" else px + s

def fees_cost(shares):
    # round-trip: we charge both at entry and exit separately in simulate
    return shares * FEE_PER_SHARE + FIXED_FEE_PER_TRADE

def _max_dd(equity):
    rm = equity.cummax()
    dd = equity - rm
    return float(-dd.min()) if len(equity) else 0.0

def _sharpe(daily):
    if daily.empty: return 0.0
    mu, sd = daily.mean(), daily.std(ddof=1)
    if sd == 0 or np.isnan(sd): return (np.inf if mu > 0 else 0.0)
    return float((mu/sd) * sqrt(TRADING_DAYS_PER_YEAR))

def _sortino(daily):
    if daily.empty: return 0.0
    mu = daily.mean()
    dstd = daily[daily < 0].std(ddof=1)
    if dstd == 0 or np.isnan(dstd): return (np.inf if mu > 0 else 0.0)
    return float((mu/dstd) * sqrt(TRADING_DAYS_PER_YEAR))

def _tstat(daily):
    n = len(daily)
    if n < 2: return 0.0
    mu, sd = daily.mean(), daily.std(ddof=1)
    if sd == 0 or np.isnan(sd): return (np.inf if mu>0 else 0.0)
    return float(mu / (sd/np.sqrt(n)))

# ---------- Strategy 1: ORB Retest (touch or near) ----------
def strat_orb_retest(day, orh, orl, symbol):
    out = []
    # after OR window
    after_or = day[day["hm"] >= OR_END].copy()
    if after_or.empty: return out

    # breakout by close
    above = after_or["Close"] > orh
    below = after_or["Close"] < orl
    i_long  = after_or.index[above].min() if above.any() else None
    i_short = after_or.index[below].min() if below.any() else None
    if i_long is None and i_short is None: return out
    if i_long is not None and (i_short is None or i_long <= i_short):
        side, brk_idx, level = "long", i_long, orh
    else:
        side, brk_idx, level = "short", i_short, orl

    after_brk = after_or.loc[after_or.index > brk_idx]
    if after_brk.empty:
        # optional continuation fallback
        if CONTINUATION_FALLBACK:
            return _enter_orb(day, brk_idx, side, level, symbol, retest=False)
        return out

    # first touch/near
    lo = after_brk["Low"].values
    hi = after_brk["High"].values
    or_width = orh - orl
    spans = (lo <= level) & (hi >= level)
    dist  = np.where(spans, 0.0, np.minimum(np.abs(lo - level), np.abs(hi - level)))
    near  = dist <= (RETEST_NEAR_FRAC * abs(or_width))
    if near.any():
        first_idx = after_brk.index[np.argmax(near)]
        return _enter_orb(day, first_idx, side, level, symbol, retest=True)
    else:
        if CONTINUATION_FALLBACK:
            return _enter_orb(day, brk_idx, side, level, symbol, retest=False)
        return out

def _enter_orb(day, idx, side, level, symbol, retest=True):
    out=[]
    entry_bar = day.loc[idx]
    etime = entry_bar["EventAt"]
    hm = entry_bar["hm"]
    if not (hm >= TRADE_START and hm <= TRADE_END): return out

    # compute OR values from day's first 15 minutes
    or_mask = (day["hm"] >= REG_START) & (day["hm"] < OR_END)
    orh = float(day.loc[or_mask, "High"].max())
    orl = float(day.loc[or_mask, "Low"].min())
    or_width = orh - orl

    if retest:
        depth = (level - float(entry_bar["Low"])) if side=="long" else (float(entry_bar["High"]) - level)
        depth = max(depth, 0.0)
        if depth > (RETEST_OVERSHOOT_FRAC * abs(or_width)): return out
        touched = (float(entry_bar["Low"]) <= level <= float(entry_bar["High"]))
        entry_px = level if touched else float(entry_bar["Close"])
    else:
        entry_px = float(entry_bar["Close"])

    stop = orl if side=="long" else orh
    risk_per_share = abs(entry_px - stop)
    shares = shares_for_risk(entry_px, stop, RISK_PER_TRADE_USD)
    if shares <= 0: return out

    # slippage & fees at entry
    fill_entry = apply_slippage(side, entry_px, "entry")
    fees_entry = fees_cost(shares)

    target = fill_entry + (RR_TARGET * risk_per_share if side=="long" else -RR_TARGET * risk_per_share)

    # walk forward
    after = day.loc[day.index > idx].copy()
    if after.empty:
        net = -fees_entry  # opened but no exit? assume flat at entry EOD (simplified)
        out.append(_rec(symbol, day, side, "ORB_RETEST", etime, fill_entry, stop, target,
                        etime, fill_entry, "EOD", shares, -fees_entry, 0.0, net))
        return out

    outcome, exit_px = _simulate_path(after, side, fill_entry, stop, target)
    # slippage & fees at exit
    exit_fill = apply_slippage(side, exit_px, "exit")
    fees_exit = fees_cost(shares)

    gross = (exit_fill - fill_entry) * shares if side=="long" else (fill_entry - exit_fill) * shares
    costs = fees_entry + fees_exit
    net   = gross - costs

    out.append(_rec(symbol, day, side, "ORB_RETEST", etime, fill_entry, stop, target,
                    after.iloc[-1]["EventAt"] if outcome=="EOD" else after.loc[after.index[0],"EventAt"],
                    exit_fill, outcome, shares, gross, costs, net))
    return out

# ---------- Strategy 2: ORB Continuation (explicit) ----------
def strat_orb_continuation(day, orh, orl, symbol):
    out=[]
    after_or = day[day["hm"] >= OR_END].copy()
    if after_or.empty: return out
    above = after_or["Close"] > orh
    below = after_or["Close"] < orl
    i_long  = after_or.index[above].min() if above.any() else None
    i_short = after_or.index[below].min() if below.any() else None
    if i_long is None and i_short is None: return out
    if i_long is not None and (i_short is None or i_long <= i_short):
        side, brk_idx = "long", i_long
        stop = orl
    else:
        side, brk_idx = "short", i_short
        stop = orh

    etime = day.loc[brk_idx, "EventAt"]
    hm = day.loc[brk_idx, "hm"]
    if not (hm >= TRADE_START and hm <= TRADE_END): return out

    entry_px = float(day.loc[brk_idx, "Close"])
    shares = shares_for_risk(entry_px, stop, RISK_PER_TRADE_USD)
    if shares <= 0: return out

    entry_fill = apply_slippage(side, entry_px, "entry")
    fees_entry = fees_cost(shares)
    risk_per_share = abs(entry_fill - stop)
    target = entry_fill + (RR_TARGET * risk_per_share if side=="long" else -RR_TARGET * risk_per_share)

    after = day.loc[day.index > brk_idx].copy()
    if after.empty:
        net = -fees_entry
        out.append(_rec(symbol, day, side, "ORB_CONT", etime, entry_fill, stop, target,
                        etime, entry_fill, "EOD", shares, -fees_entry, 0.0, net))
        return out

    outcome, exit_px = _simulate_path(after, side, entry_fill, stop, target)
    exit_fill = apply_slippage(side, exit_px, "exit")
    fees_exit = fees_cost(shares)
    gross = (exit_fill - entry_fill) * shares if side=="long" else (entry_fill - exit_fill) * shares
    net = gross - (fees_entry + fees_exit)
    out.append(_rec(symbol, day, side, "ORB_CONT", etime, entry_fill, stop, target,
                    after.iloc[-1]["EventAt"] if outcome=="EOD" else after.loc[after.index[0],"EventAt"],
                    exit_fill, outcome, shares, gross, fees_entry+fees_exit, net))
    return out

# ---------- Strategy 3: VWAP Pullback Trend ----------
def strat_vwap_pullback(day, symbol):
    out=[]
    # Only trade between VWAP_TRADE_START and VWAP_TRADE_END
    intr = day[(day["hm"] >= VWAP_TRADE_START) & (day["hm"] <= VWAP_TRADE_END)].copy()
    if intr.empty: return out

    # Trend up/down & pullback to VWAP band; enter next bar
    for idx in intr.index[:-1]:
        row = intr.loc[idx]
        px, vw, vs, atr = float(row["Close"]), float(row["VWAP"]), float(row["VWAP_Slope"]), float(row["ATR"])
        if not np.isfinite(vw) or not np.isfinite(atr): continue
        band = VWAP_BAND_ATR * atr

        # Long setup
        if (px >= vw) and (vs >= VWAP_TREND_MIN_SLOPE) and (abs(px - vw) <= band):
            side = "long"
            entry_px = float(intr.loc[idx+1, "Close"])  # next bar execution
            stop = vw - VWAP_STOP_ATR * atr
            shares = shares_for_risk(entry_px, stop, RISK_PER_TRADE_USD)
            if shares > 0:
                entry_fill = apply_slippage(side, entry_px, "entry")
                fees_entry = fees_cost(shares)
                risk_ps = abs(entry_fill - stop)
                target = entry_fill + VWAP_R_MULT_TP * risk_ps
                after = day.loc[day.index > (idx+1)].copy()
                if not after.empty:
                    outcome, exit_px = _simulate_path(after, side, entry_fill, stop, target)
                    exit_fill = apply_slippage(side, exit_px, "exit")
                    fees_exit = fees_cost(shares)
                    gross = (exit_fill - entry_fill) * shares
                    net = gross - (fees_entry + fees_exit)
                    out.append(_rec(symbol, day, side, "VWAP_PULLBACK", row["EventAt"], entry_fill, stop, target,
                                    after.iloc[-1]["EventAt"] if outcome=="EOD" else after.loc[after.index[0],"EventAt"],
                                    exit_fill, outcome, shares, gross, fees_entry+fees_exit, net))

        # Short setup
        if (px <= vw) and (vs <= -VWAP_TREND_MIN_SLOPE) and (abs(px - vw) <= band):
            side = "short"
            entry_px = float(intr.loc[idx+1, "Close"])
            stop = vw + VWAP_STOP_ATR * atr
            shares = shares_for_risk(entry_px, stop, RISK_PER_TRADE_USD)
            if shares > 0:
                entry_fill = apply_slippage(side, entry_px, "entry")
                fees_entry = fees_cost(shares)
                risk_ps = abs(entry_fill - stop)
                target = entry_fill - VWAP_R_MULT_TP * risk_ps
                after = day.loc[day.index > (idx+1)].copy()
                if not after.empty:
                    outcome, exit_px = _simulate_path(after, side, entry_fill, stop, target)
                    exit_fill = apply_slippage(side, exit_px, "exit")
                    fees_exit = fees_cost(shares)
                    gross = (entry_fill - exit_fill) * shares
                    net = gross - (fees_entry + fees_exit)
                    out.append(_rec(symbol, day, side, "VWAP_PULLBACK", row["EventAt"], entry_fill, stop, target,
                                    after.iloc[-1]["EventAt"] if outcome=="EOD" else after.loc[after.index[0],"EventAt"],
                                    exit_fill, outcome, shares, gross, fees_entry+fees_exit, net))
    return out

# ---------- Strategy 4: Previous-Day High/Low Retest ----------
def strat_pdh_pdl(day, prev_high, prev_low, symbol):
    out=[]
    intr = day[(day["hm"] >= PD_TRADE_START) & (day["hm"] <= PD_TRADE_END)].copy()
    if intr.empty: return out

    # Breakout close beyond PDH/PDL
    above = intr["Close"] > prev_high
    below = intr["Close"] < prev_low
    i_long  = intr.index[above].min() if above.any() else None
    i_short = intr.index[below].min() if below.any() else None

    entries = []
    if i_long is not None:  entries.append(("long", i_long, prev_high))
    if i_short is not None: entries.append(("short", i_short, prev_low))

    for side, brk_idx, level in entries:
        after = intr.loc[intr.index > brk_idx]
        if after.empty: continue

        lo = after["Low"].values
        hi = after["High"].values
        # near band based on today's ATR at breakout bar
        atr_b = float(intr.loc[brk_idx, "ATR"])
        near_band = PD_RETEST_NEAR_FRAC * (atr_b if np.isfinite(atr_b) else 1.0)
        spans = (lo <= level) & (hi >= level)
        dist  = np.where(spans, 0.0, np.minimum(np.abs(lo - level), np.abs(hi - level)))
        near  = dist <= near_band
        if not near.any(): continue
        idx = after.index[np.argmax(near)]
        entry_bar = day.loc[idx]
        entry_px = level if (float(entry_bar["Low"]) <= level <= float(entry_bar["High"])) else float(entry_bar["Close"])

        stop = prev_low if side=="long" else prev_high
        shares = shares_for_risk(entry_px, stop, RISK_PER_TRADE_USD)
        if shares <= 0: continue

        entry_fill = apply_slippage(side, entry_px, "entry")
        fees_entry = fees_cost(shares)
        risk_ps = abs(entry_fill - stop)
        target = entry_fill + (PD_R_MULT_TP * risk_ps if side=="long" else -PD_R_MULT_TP * risk_ps)

        after2 = day.loc[day.index > idx].copy()
        if after2.empty: continue
        outcome, exit_px = _simulate_path(after2, side, entry_fill, stop, target)
        exit_fill = apply_slippage(side, exit_px, "exit")
        fees_exit = fees_cost(shares)
        gross = (exit_fill - entry_fill) * shares if side=="long" else (entry_fill - exit_fill) * shares
        net = gross - (fees_entry + fees_exit)
        out.append(_rec(symbol, day, side, "PDH_PDL_RETEST", entry_bar["EventAt"], entry_fill, stop, target,
                        after2.iloc[-1]["EventAt"] if outcome=="EOD" else after2.loc[after2.index[0],"EventAt"],
                        exit_fill, outcome, shares, gross, fees_entry+fees_exit, net))
    return out

# ---------- Path Simulator ----------
def _simulate_path(fwd_df, side, entry_fill, stop, target):
    for _, r in fwd_df.iterrows():
        lo, hi = float(r["Low"]), float(r["High"])
        hit_stop = (lo <= stop) if side=="long" else (hi >= stop)
        hit_tp   = (hi >= target) if side=="long" else (lo <= target)
        if hit_stop and hit_tp:
            # conservative: stop-first sequencing
            return "STOP", stop
        if hit_tp:   return "TP", target
        if hit_stop: return "STOP", stop
    # end of day exit at last close
    return "EOD", float(fwd_df.iloc[-1]["Close"])

# ---------- Trade record ----------
def _rec(symbol, day, side, strategy, etime, entry, stop, target, xtime, exit_px, outcome, shares, gross, costs, net):
    d = pd.to_datetime(etime).date()
    return {
        "Date": d, "Symbol": symbol, "Strategy": strategy, "Side": side,
        "EntryTime": etime, "EntryPx": float(entry), "StopPx": float(stop), "TargetPx": float(target),
        "ExitTime": xtime, "ExitPx": float(exit_px), "Outcome": outcome,
        "Shares": int(shares), "GrossPnL_$": float(gross), "Costs_$": float(costs), "NetPnL_$": float(net)
    }

# ---------- Per-symbol engine ----------
def run_symbol(fpath: Path):
    symbol = fpath.stem.upper()
    df = pd.read_parquet(fpath)
    for c in ["EventAt","Open","High","Low","Close","Volume"]:
        if c not in df.columns:
            print(f"[SKIP] {symbol}: missing {c}")
            return symbol, pd.DataFrame()

    df["EventAt"] = to_ny(df["EventAt"])
    df = df.sort_values("EventAt").reset_index(drop=True)
    df = add_time_cols(df)
    df = session_slice(df)
    if df.empty:
        return symbol, pd.DataFrame()

    trades = []
    # Build previous day H/L map
    by_day = list(df.groupby("date"))

    prev_day_high = None
    prev_day_low  = None

    for d, day in by_day:
        if day.empty: 
            continue

        # opening range
        or_mask = (day["hm"] >= REG_START) & (day["hm"] < OR_END)
        if not or_mask.any():
            # update prev day H/L after finishing day
            prev_day_high = float(day["High"].max())
            prev_day_low  = float(day["Low"].min())
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())

        # Indicators
        pre_or = day[day["hm"] < REG_START]
        prev_close = float(pre_or["Close"].iloc[-1]) if not pre_or.empty else float(day["Close"].iloc[0])
        day = add_vwap(day)
        day = add_atr(day, prev_close)

        # 1) ORB_RETEST
        if ENABLED_STRATEGIES.get("ORB_RETEST", False):
            trades += strat_orb_retest(day, orh, orl, symbol)

        # 2) ORB_CONTINUATION
        if ENABLED_STRATEGIES.get("ORB_CONTINUATION", False):
            trades += strat_orb_continuation(day, orh, orl, symbol)

        # 3) VWAP_PULLBACK
        if ENABLED_STRATEGIES.get("VWAP_PULLBACK", False):
            trades += strat_vwap_pullback(day, symbol)

        # 4) PDH/PDL_RETEST  (needs yesterday's levels)
        if ENABLED_STRATEGIES.get("PDH_PDL_RETEST", False) and prev_day_high is not None and prev_day_low is not None:
            trades += strat_pdh_pdl(day, prev_day_high, prev_day_low, symbol)

        # finally update yesterday H/L for next day
        prev_day_high = float(day["High"].max())
        prev_day_low  = float(day["Low"].min())

    if not trades:
        return symbol, pd.DataFrame()

    tdf = pd.DataFrame(trades)
    # keep one open per strategy/symbol/day by dropping duplicates on EntryTime (rare collisions)
    tdf = tdf.sort_values(["Date","EntryTime"]).drop_duplicates(subset=["Date","Symbol","Strategy","EntryTime"], keep="first")
    return symbol, tdf

# ---------- Metrics ----------
def summarize(trades):
    if trades.empty:
        return {"Trades":0, "NetPnL_$":0.0, "WinRate_%":0.0, "ProfitFactor":0.0,
                "Sharpe":0.0,"Sortino":0.0,"MaxDD_$":0.0,"Calmar":0.0,"t_stat":0.0,
                "CVaR5%":0.0,"CVaR1%":0.0,"Skew":0.0,"ExKurt":0.0}

    wins = trades["NetPnL_$"] > 0
    losses = trades["NetPnL_$"] < 0
    gross_pos = trades.loc[wins, "NetPnL_$"].sum()
    gross_neg = -trades.loc[losses, "NetPnL_$"].sum()
    pf = (gross_pos / gross_neg) if gross_neg > 0 else np.inf
    wr = 100.0 * wins.mean()

    daily = trades.groupby("Date")["NetPnL_$"].sum().sort_index()
    sharpe = _sharpe(daily); sortino = _sortino(daily)
    equity = daily.cumsum()
    maxdd = _max_dd(equity)
    ann_pnl = float(daily.mean() * TRADING_DAYS_PER_YEAR)
    calmar = (ann_pnl / maxdd) if maxdd > 0 else np.inf
    tstat = _tstat(daily)
    cvar5 = float(daily[daily <= daily.quantile(0.05)].mean()) if not daily.empty else 0.0
    cvar1 = float(daily[daily <= daily.quantile(0.01)].mean()) if not daily.empty else 0.0
    skew = float(daily.skew()) if len(daily) > 2 else 0.0
    exkurt = float(daily.kurt()) if len(daily) > 3 else 0.0

    return {
        "Trades": int(len(trades)),
        "NetPnL_$": float(trades["NetPnL_$"].sum()),
        "WinRate_%": float(wr),
        "ProfitFactor": float(pf if isfinite(pf) else 0.0),
        "Sharpe": float(sharpe if isfinite(sharpe) else 0.0),
        "Sortino": float(sortino if isfinite(sortino) else 0.0),
        "MaxDD_$": float(maxdd),
        "Calmar": float(calmar if isfinite(calmar) else 0.0),
        "t_stat": float(tstat if isfinite(tstat) else 0.0),
        "CVaR5%": float(cvar5 if np.isfinite(cvar5) else 0.0),
        "CVaR1%": float(cvar1 if np.isfinite(cvar1) else 0.0),
        "Skew": float(skew if np.isfinite(skew) else 0.0),
        "ExKurt": float(exkurt if np.isfinite(exkurt) else 0.0),
    }

# ---------- Runner ----------
def main():
    SAVE_DIR.mkdir(parents=True, exist_ok=True)
    files = sorted(Path(PARQUET_DIR).glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {PARQUET_DIR}")
        return

    all_trades = []
    for f in files:
        symbol, tdf = run_symbol(f)
        if tdf.empty:
            print(f"=== {symbol} — no trades ===")
            continue
        # save per symbol trades
        outp = SAVE_DIR / f"multi_trades_{symbol}.csv"
        tdf.to_csv(outp, index=False)
        s = summarize(tdf)
        print(f"=== {symbol} — Multi-Strategy ===")
        print(f"Trades: {s['Trades']} | NetPnL: ${s['NetPnL_$']:,.2f} | Win%: {s['WinRate_%']:.2f}% | PF: {s['ProfitFactor']:.3f}")
        print(f"— Daily — Sharpe: {s['Sharpe']:.3f} | Sortino: {s['Sortino']:.3f} | MaxDD: ${s['MaxDD_$']:,.2f} | Calmar: {s['Calmar']:.3f} | t-stat: {s['t_stat']:.3f}\n")
        all_trades.append(tdf)

    if not all_trades:
        print("\nNo trades generated.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    overall_path = SAVE_DIR / "multi_trades_ALL.csv"
    overall.to_csv(overall_path, index=False)

    s = summarize(overall)
    print(f"=== OVERALL — Multi-Strategy ===")
    print(f"Trades: {s['Trades']} | NetPnL: ${s['NetPnL_$']:,.2f} | Win%: {s['WinRate_%']:.2f}% | PF: {s['ProfitFactor']:.3f}")
    print(f"— Daily — Sharpe: {s['Sharpe']:.3f} | Sortino: {s['Sortino']:.3f} | MaxDD: ${s['MaxDD_$']:,.2f} | Calmar: {s['Calmar']:.3f} | t-stat: {s['t_stat']:.3f} | "
          f"CVaR5%: ${s['CVaR5%']:,.2f} | CVaR1%: ${s['CVaR1%']:,.2f} | Skew: {s['Skew']:.3f} | ExKurt: {s['ExKurt']:.3f}")
    print(f"Saved trades → {overall_path.resolve()}")

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_all_parquet_RR_1_to_2_DIAG_RISK.py
# ORB 15m Retest (RR 1:2) across ALL parquet files with diagnostics + risk-normalized sizing.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from math import floor
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"  # your folder
TZ_LOCAL = "America/New_York"

REG_START = "09:30"   # session start
OR_END    = "09:45"   # end of first 15-min bar (exclusive)
REG_END   = "16:00"   # session end

STOP_FIRST_IF_BOTH_HIT = True
SAVE_DIR = Path(".")

# Risk-normalized sizing (per-trade $ risk ~ opening range width per share)
RISK_PER_TRADE_USD = 100.0   # e.g., $100 risk budget per trade

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def simulate_path(after_entry_df, side, entry, stop, target, stop_first=True):
    for idx, row in after_entry_df.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop and hit_tp:
            return ("STOP", stop, idx) if stop_first else ("TP", target, idx)
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    last_idx = after_entry_df.index[-1]
    return "EOD", float(after_entry_df.loc[last_idx, "Close"]), last_idx

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

def calc_metrics(trades_df, pnl_col):
    pnl = trades_df[pnl_col].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= CORE BACKTEST + DIAGNOSTIC FIELDS =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Keep only regular session minutes
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    records = []

    for d, day in df.groupby("date"):
        # Opening range: first 15-minute bar [09:30, 09:45)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # Day-level
        day_open = float(day.iloc[0]["Open"])
        day_close = float(day.iloc[-1]["Close"])
        day_dir = 1 if (day_close > day_open) else (-1 if day_close < day_open else 0)
        or_width_bps = (or_width / day_open) * 10_000.0  # basis points

        # After OR window (from 09:45 to close)
        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        # 1) First confirmed breakout by CLOSE
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # 2) After breakout close, retest entry at the broken level
        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        # 3) First bar that *touches* the level → entry
        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop       = orl if side == "long" else orh
        target     = entry + ( 2 * or_width if side == "long" else -2 * or_width)

        # Retest latency & trend agreement
        breakout_time = day.loc[brk_idx, "EventAt"]
        latency_min = (entry_time - breakout_time).total_seconds() / 60.0
        trade_dir = 1 if side == "long" else -1
        trend_agree = 1 if (trade_dir == day_dir and day_dir != 0) else 0

        # Risk-normalized shares (floor to an integer; at least 1 share)
        shares_norm = int(max(1, floor(RISK_PER_TRADE_USD / or_width))) if or_width > 0 else 1

        # Forward simulate after the entry bar
        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            pnl_1 = 0.0
            pnl_norm = pnl_1 * shares_norm
            records.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": or_width, "OR_Width_bps": or_width_bps,
                "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
                "Outcome_1to2": "EOD", "ExitPx_1to2": entry, "ExitTime_1to2": entry_time,
                "PnL_1to2": pnl_1, "Shares_Norm": shares_norm, "PnL_Norm": pnl_norm,
                "RetestLatency_min": latency_min, "TrendAgree": trend_agree, "DayDir": day_dir,
            })
            continue

        outcome, exit_px, exit_idx = simulate_path(
            after_entry, side, entry, stop, target, stop_first=STOP_FIRST_IF_BOTH_HIT
        )
        pnl_1 = pnl_one_share(side, entry, exit_px)
        pnl_norm = pnl_1 * shares_norm

        records.append({
            "Date": d, "Symbol": symbol, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": or_width, "OR_Width_bps": or_width_bps,
            "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
            "Outcome_1to2": outcome, "ExitPx_1to2": exit_px,
            "ExitTime_1to2": day.loc[exit_idx, "EventAt"] if exit_idx in day.index else None,
            "PnL_1to2": pnl_1, "Shares_Norm": shares_norm, "PnL_Norm": pnl_norm,
            "RetestLatency_min": latency_min, "TrendAgree": trend_agree, "DayDir": day_dir,
        })

    return pd.DataFrame(records)

# ========= RUN ALL FILES + DIAGNOSTICS =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    all_trades = []

    for f in files:
        symbol = f.stem.upper()
        try:
            trades = backtest_one_symbol(f, symbol)
        except Exception as e:
            print(f"[SKIP] {symbol}: {e}")
            continue

        if trades.empty:
            print(f"=== ORB 15m Retest — {symbol} (1 share, RR 1:2) ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            continue

        all_trades.append(trades)

        # Per-symbol metrics (both modes)
        t1, p1, w1, pf1 = calc_metrics(trades, "PnL_1to2")
        t2, p2, w2, pf2 = calc_metrics(trades, "PnL_Norm")

        save_path = SAVE_DIR / f"orb_trades_{symbol}_RR1to2.csv"
        trades.to_csv(save_path, index=False)

        print(f"=== ORB 15m Retest — {symbol} ===")
        print(f"[1 share]      Trades: {t1} | Total PnL: ${p1:,.2f} | Win%: {w1:.2f}% | PF: {pf1:.3f}")
        print(f"[Risk-normal]  Trades: {t2} | Total PnL: ${p2:,.2f} | Win%: {w2:.2f}% | PF: {pf2:.3f}")
        print(f"Saved trades → {save_path.resolve()}")

    if not all_trades:
        print("\nNo trades generated for any symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)

    # ---------- DIAGNOSTICS ----------
    trades = overall.copy()
    trades["Win_1"] = (trades["PnL_1to2"] > 0).astype(int)
    trades["Win_Norm"] = (trades["PnL_Norm"] > 0).astype(int)

    def _pf(s):
        pos = s[s > 0].sum()
        neg = -s[s < 0].sum()
        return (pos / neg) if neg > 0 else np.inf

    per_sym = (
        trades.groupby("Symbol")
        .agg(
            Trades=("PnL_1to2", "size"),
            PnL_sum_1=("PnL_1to2", "sum"),
            PnL_sum_Norm=("PnL_Norm", "sum"),
            WinRate_pct_1=("Win_1", "mean"),
            WinRate_pct_Norm=("Win_Norm", "mean"),
            PF_1=("PnL_1to2", _pf),
            PF_Norm=("PnL_Norm", _pf),
            OR_usd=("OR_Width", "mean"),
            OR_bps=("OR_Width_bps", "median"),
            Latency_min=("RetestLatency_min", "median"),
            TrendAgree_pct=("TrendAgree", "mean"),
            TP_rate=("Outcome_1to2", lambda s: np.mean(np.array(s)=="TP")),
            STOP_rate=("Outcome_1to2", lambda s: np.mean(np.array(s)=="STOP")),
            EOD_rate=("Outcome_1to2", lambda s: np.mean(np.array(s)=="EOD")),
        )
        .reset_index()
    )
    per_sym["WinRate_pct_1"] *= 100.0
    per_sym["WinRate_pct_Norm"] *= 100.0
    per_sym["TrendAgree_pct"] *= 100.0

    # Cross-sectional correlations per symbol (robust Spearman)
    print("\n--- Cross-sectional Spearman correlations (per symbol) ---")
    for target in ["PnL_sum_1", "PnL_sum_Norm", "WinRate_pct_1", "WinRate_pct_Norm", "PF_1", "PF_Norm"]:
        for feat in ["OR_usd", "OR_bps", "Latency_min", "TrendAgree_pct"]:
            try:
                rho = per_sym[[feat, target]].corr(method="spearman").iloc[0,1]
                print(f"Spearman({feat} , {target}) = {rho:+.3f}")
            except Exception:
                print(f"Spearman({feat} , {target}) = n/a")

    # Per-trade correlations (pooled)
    print("\n--- Per-trade Spearman correlations (pooled) ---")
    pooled = trades[["PnL_1to2","PnL_Norm","Win_1","Win_Norm","OR_Width","OR_Width_bps","RetestLatency_min","TrendAgree"]].dropna()
    for y in ["Win_1","Win_Norm","PnL_1to2","PnL_Norm"]:
        for x in ["OR_Width","OR_Width_bps","RetestLatency_min","TrendAgree"]:
            try:
                rho = pooled[[x, y]].corr(method="spearman").iloc[0,1]
                print(f"Spearman({x} , {y}) = {rho:+.3f}")
            except Exception:
                print(f"Spearman({x} , {y}) = n/a")

    # Optional: OLS (skip if statsmodels not installed)
    try:
        import statsmodels.api as sm
        X = pooled[["OR_Width_bps","RetestLatency_min","TrendAgree"]].copy()
        X = (X - X.mean()) / X.std(ddof=0)
        X = sm.add_constant(X)
        for yname in ["PnL_1to2","PnL_Norm"]:
            y = pooled[yname].values
            res = sm.OLS(y, X).fit()
            print(f"\n--- OLS: {yname} ~ OR_Width_bps + RetestLatency_min + TrendAgree ---")
            print(res.summary().as_text())
    except Exception as e:
        print("\n(OLS summary skipped:", e, ")")

    # Rank symbols: 1-share and risk-normal
    ranked_1 = per_sym.sort_values(
        ["PF_1","WinRate_pct_1","TrendAgree_pct","OR_bps"],
        ascending=[False, False, False, False]
    ).head(10)
    ranked_N = per_sym.sort_values(
        ["PF_Norm","WinRate_pct_Norm","TrendAgree_pct","OR_bps"],
        ascending=[False, False, False, False]
    ).head(10)

    print("\nTop 10 symbols (1 share) by PF/Win/Trend/OR_bps:")
    print(ranked_1[["Symbol","Trades","PnL_sum_1","PF_1","WinRate_pct_1","OR_bps","TrendAgree_pct","Latency_min","TP_rate","STOP_rate"]].to_string(index=False))

    print("\nTop 10 symbols (Risk-normalized) by PF/Win/Trend/OR_bps:")
    print(ranked_N[["Symbol","Trades","PnL_sum_Norm","PF_Norm","WinRate_pct_Norm","OR_bps","TrendAgree_pct","Latency_min","TP_rate","STOP_rate"]].to_string(index=False))

    # Save everything
    save_all = SAVE_DIR / "orb_trades_ALL_RR1to2.csv"
    trades.to_csv(save_all, index=False)
    print(f"\n=== ORB 15m Retest — OVERALL ===")
    t1, p1, w1, pf1 = calc_metrics(trades, "PnL_1to2")
    t2, p2, w2, pf2 = calc_metrics(trades, "PnL_Norm")
    print(f"[1 share]     Trades: {t1} | Total PnL: ${p1:,.2f} | Win%: {w1:.2f}% | PF: {pf1:.3f}")
    print(f"[Risk-normal] Trades: {t2} | Total PnL: ${p2:,.2f} | Win%: {w2:.2f}% | PF: {pf2:.3f}")
    print(f"Saved all-trades CSV → {save_all.resolve()}")

    diag_path = SAVE_DIR / "orb_diagnostics_per_symbol.csv"
    per_sym.to_csv(diag_path, index=False)
    print(f"Saved per-symbol diagnostics → {diag_path.resolve()}")

if __name__ == "__main__":
    main()
```
