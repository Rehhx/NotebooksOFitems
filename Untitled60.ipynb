```python
# ============================================
# Divergence + Level-to-Level Futures Backtest
# QuestDB (Postgres wire) • READ-ONLY
# ============================================

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
from dataclasses import dataclass
from typing import List, Optional, Tuple, Dict

import numpy as np
import pandas as pd
from sqlalchemy import create_engine, text

# =============================
# DB SETTINGS (QuestDB)
# =============================
ENGINE_URL = "postgresql+psycopg2://admin:quest@24.128.195.68:8812/qdb"
TABLE      = "futures_candles"
SAVE_DIR   = Path("."); SAVE_DIR.mkdir(parents=True, exist_ok=True)

# Build engine in AUTOCOMMIT (avoids SAVEPOINT/rollback issues)
engine = create_engine(
    ENGINE_URL,
    isolation_level="AUTOCOMMIT",
    pool_pre_ping=True,
    future=True,
)

# =============================
# RUN SETTINGS
# =============================
INTERVAL       = "1m"             # '1m' rows from your table
SYMBOLS        = ["NQ"]           # exact symbols OR set SYMBOL_PREFIX below
SYMBOL_PREFIX: Optional[str] = None   # e.g. "NQ" to grab all NQ* contracts

START_DATE: Optional[str] = None  # e.g. "2024-10-01"
END_DATE:   Optional[str] = None  # e.g. "2024-12-01"

# =============================
# STRATEGY PARAMETERS
# =============================
TZ_LOCAL = "America/New_York"
REG_START = "09:30"   # RTH session start
REG_END   = "16:00"   # RTH session end

RSI_LEN = 14
ATR_LEN = 14
SWING_K = 5                   # swing window (2k+1 centered)

ATR_MULT_STOP   = 1.0         # stop distance = 1.0 * ATR
R_MULT_TARGET   = 1.5         # target = 1.5R
CONFIRM_ENTRY   = True        # require next bar break of signal bar

USE_LEVELS               = True
LEVEL_PROX_THRESHOLD_ATR = 0.5  # must be within X*ATR of a level
USE_DAILY_OPEN_AS_LEVEL  = True # include session open as a level

# Instrument specs (per symbol). Default to NQ if missing.
SYMBOL_SPECS = {
    "NQ": {"POINT_VALUE": 20.0, "TICK_SIZE": 0.25},  # E-mini NQ
    "MNQ": {"POINT_VALUE": 2.0, "TICK_SIZE": 0.25},  # Micro NQ
    "ES": {"POINT_VALUE": 50.0, "TICK_SIZE": 0.25},  # E-mini S&P
    "MES": {"POINT_VALUE": 5.0, "TICK_SIZE": 0.25},  # Micro ES
    "RTY": {"POINT_VALUE": 50.0, "TICK_SIZE": 0.10}, # E-mini Russell
    "M2K": {"POINT_VALUE": 5.0, "TICK_SIZE": 0.10},
    "YM": {"POINT_VALUE": 5.0,  "TICK_SIZE": 1.0},   # Dow (pts=$5)
    "MYM": {"POINT_VALUE": 0.5, "TICK_SIZE": 1.0},
}

# =============================
# Helpers
# =============================
def _spec_for(symbol: str) -> Tuple[float, float]:
    """Return (POINT_VALUE, TICK_SIZE) with a sensible default."""
    for k, v in SYMBOL_SPECS.items():
        if symbol.startswith(k):
            return v["POINT_VALUE"], v["TICK_SIZE"]
    # default to NQ if unknown
    return 20.0, 0.25

def _literal_in_list(values: List[str]) -> str:
    return ",".join("'" + str(v).replace("'", "''") + "'" for v in values)

def to_ny(ts: pd.Series) -> pd.Series:
    utc = pd.to_datetime(ts, utc=True)  # your DB uses "timestamp without tz" but is UTC; force-interpret as UTC
    return utc.dt.tz_convert(TZ_LOCAL)

def rsi(close: pd.Series, length: int = 14) -> pd.Series:
    delta = close.diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    roll_up = up.ewm(alpha=1/length, adjust=False).mean()
    roll_down = down.ewm(alpha=1/length, adjust=False).mean()
    rs = roll_up / roll_down.replace(0, np.nan)
    return 100 - (100 / (1 + rs))

def atr(df: pd.DataFrame, length: int = 14) -> pd.Series:
    high, low, close = df["high"], df["low"], df["close"]
    prev_close = close.shift(1)
    tr = pd.concat([
        (high - low),
        (high - prev_close).abs(),
        (low - prev_close).abs()
    ], axis=1).max(axis=1)
    return tr.ewm(alpha=1/length, adjust=False).mean()

def is_swing_high(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).max())

def is_swing_low(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).min())

def daily_levels(day_df: pd.DataFrame, prev_day_df: Optional[pd.DataFrame]) -> List[float]:
    levels = []
    if prev_day_df is not None and not prev_day_df.empty:
        levels += [prev_day_df["high"].max(), prev_day_df["low"].min()]
    if USE_DAILY_OPEN_AS_LEVEL and not day_df.empty:
        levels.append(float(day_df.iloc[0]["open"]))
    return sorted(list({round(x, 4) for x in levels}))

def nearest_level(price: float, levels: List[float]) -> Tuple[float, float]:
    if not levels:
        return np.nan, np.inf
    d = [abs(price - lv) for lv in levels]
    i = int(np.argmin(d))
    return levels[i], d[i]

@dataclass
class Trade:
    symbol: str
    entry_time: pd.Timestamp
    side: str            # "long" / "short"
    entry: float
    stop: float
    target: float
    exit_time: pd.Timestamp
    exit_price: float
    reason: str          # "target" / "stop" / "eod"
    r_mult: float
    points: float
    ticks: int
    pnl_usd: float
    day: str

# =============================
# Data Loader (read-only, QuestDB)
# =============================
def load_data(
    table: str,
    interval: str,
    symbols: Optional[List[str]] = None,
    symbol_prefix: Optional[str] = None,
    start: Optional[str] = None,
    end: Optional[str] = None,
) -> pd.DataFrame:

    where = ["interval = :interval"]
    params = {"interval": interval}

    if symbols and len(symbols) > 0 and symbol_prefix is None:
        where.append(f"symbol IN ({_literal_in_list(symbols)})")
    elif symbol_prefix:
        where.append("symbol LIKE :prefix")
        params["prefix"] = f"{symbol_prefix}%"

    if start:
        where.append("event_at >= TIMESTAMP :start")
        params["start"] = start
    if end:
        where.append("event_at < TIMESTAMP :end")
        params["end"] = end

    sql = text(f"""
        SELECT symbol, interval, event_at, open, high, low, close, volume
        FROM {table}
        WHERE {' AND '.join(where)}
        ORDER BY symbol, event_at
    """)

    with engine.connect() as conn:
        df = pd.read_sql_query(sql, conn, params=params)

    if df.empty:
        return df

    df["ts_ny"]  = to_ny(df["event_at"])
    df["date_ny"] = df["ts_ny"].dt.date
    return df

# =============================
# Signals & Backtest
# =============================
def build_signals(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["rsi"] = rsi(df["close"], RSI_LEN)
    df["atr"] = atr(df, ATR_LEN)

    df["sh"] = is_swing_high(df["high"], SWING_K).fillna(False)
    df["sl"] = is_swing_low(df["low"],   SWING_K).fillna(False)
    df["rsi_sh"] = is_swing_high(df["rsi"], SWING_K).fillna(False)
    df["rsi_sl"] = is_swing_low(df["rsi"],  SWING_K).fillna(False)

    # prior swing values to compare for divergence
    df["last_price_sh"] = np.where(df["sh"], df["high"], np.nan).ffill()
    df["last_price_sl"] = np.where(df["sl"], df["low"],  np.nan).ffill()
    df["last_rsi_sh"]   = np.where(df["rsi_sh"], df["rsi"], np.nan).ffill()
    df["last_rsi_sl"]   = np.where(df["rsi_sl"], df["rsi"], np.nan).ffill()

    # Bearish divergence: price higher high, RSI lower high
    df["bear_div"] = (
        df["sh"] & df["rsi_sh"] &
        (df["high"] > df["last_price_sh"].shift(1)) &
        (df["rsi"]  < df["last_rsi_sh"].shift(1))
    )

    # Bullish divergence: price lower low, RSI higher low
    df["bull_div"] = (
        df["sl"] & df["rsi_sl"] &
        (df["low"] < df["last_price_sl"].shift(1)) &
        (df["rsi"] > df["last_rsi_sl"].shift(1))
    )
    return df

def _finalize_trade(symbol: str, pos: Dict, exit_time: pd.Timestamp, exit_price: float,
                    reason: str, r_mult: float, pnl_points: float,
                    point_value: float, tick_size: float) -> Trade:
    ticks = int(round(pnl_points / tick_size))
    pnl_usd = pnl_points * point_value
    return Trade(
        symbol=symbol,
        entry_time=pos["entry_time"], side=pos["side"],
        entry=pos["entry"], stop=pos["stop"], target=pos["target"],
        exit_time=exit_time, exit_price=exit_price,
        reason=reason, r_mult=r_mult, points=pnl_points,
        ticks=ticks, pnl_usd=pnl_usd, day=str(pos["entry_time"].date())
    )

def simulate_intraday(symbol: str, df: pd.DataFrame) -> List[Trade]:
    trades: List[Trade] = []
    df = build_signals(df)

    POINT_VALUE, TICK_SIZE = _spec_for(symbol)

    for day, g in df.groupby("date_ny"):
        # RTH window only
        day_slice = g[(g["ts_ny"].dt.strftime("%H:%M") >= REG_START) &
                      (g["ts_ny"].dt.strftime("%H:%M") <= REG_END)].copy()
        if day_slice.empty:
            continue

        # Previous day for levels
        prev_df = df[df["date_ny"] == (pd.to_datetime(day) - pd.Timedelta(days=1)).date()]
        levels = daily_levels(day_slice, prev_df) if USE_LEVELS else []

        active_pos = None

        for i in range(1, len(day_slice)):
            prev = day_slice.iloc[i-1]
            row  = day_slice.iloc[i]

            # manage open position first
            if active_pos:
                if active_pos["side"] == "long":
                    if row["low"] <= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        pnl_points = exit_px - active_pos["entry"]
                        r = pnl_points / (active_pos["entry"] - active_pos["stop"])
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "stop", r, pnl_points, POINT_VALUE, TICK_SIZE))
                        active_pos = None
                    elif row["high"] >= active_pos["target"]:
                        exit_px = active_pos["target"]
                        pnl_points = exit_px - active_pos["entry"]
                        r = R_MULT_TARGET
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "target", r, pnl_points, POINT_VALUE, TICK_SIZE))
                        active_pos = None
                else:  # short
                    if row["high"] >= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        pnl_points = active_pos["entry"] - exit_px
                        r = pnl_points / (active_pos["stop"] - active_pos["entry"])
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "stop", r, pnl_points, POINT_VALUE, TICK_SIZE))
                        active_pos = None
                    elif row["low"] <= active_pos["target"]:
                        exit_px = active_pos["target"]
                        pnl_points = active_pos["entry"] - exit_px
                        r = R_MULT_TARGET
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "target", r, pnl_points, POINT_VALUE, TICK_SIZE))
                        active_pos = None

            # if flat, check new signals
            if active_pos is None:
                atr_now = max(1e-6, row["atr"])
                near_level = True
                if USE_LEVELS and levels:
                    _, dist = nearest_level(row["close"], levels)
                    near_level = (dist <= LEVEL_PROX_THRESHOLD_ATR * atr_now)

                # bearish divergence -> short
                if prev["bear_div"] and near_level:
                    signal_hi = prev["high"]
                    entry = min(row["open"], row["low"]) if CONFIRM_ENTRY else row["low"]
                    entry = min(entry, signal_hi - TICK_SIZE)  # 1 tick below signal bar
                    stop  = entry + ATR_MULT_STOP * atr_now
                    target = entry - R_MULT_TARGET * (stop - entry)
                    active_pos = {"side": "short", "entry_time": row["ts_ny"], "entry": entry,
                                  "stop": stop, "target": target, "reason": "bear_div"}

                # bullish divergence -> long
                elif prev["bull_div"] and near_level:
                    signal_lo = prev["low"]
                    entry = max(row["open"], row["high"]) if CONFIRM_ENTRY else row["high"]
                    entry = max(entry, signal_lo + TICK_SIZE)  # 1 tick above signal bar
                    stop  = entry - ATR_MULT_STOP * atr_now
                    target = entry + R_MULT_TARGET * (entry - stop)
                    active_pos = {"side": "long", "entry_time": row["ts_ny"], "entry": entry,
                                  "stop": stop, "target": target, "reason": "bull_div"}

        # flatten at end of session
        if active_pos is not None:
            last = day_slice.iloc[-1]
            exit_px = last["close"]
            if active_pos["side"] == "long":
                pnl_points = exit_px - active_pos["entry"]
                r = pnl_points / (active_pos["entry"] - active_pos["stop"])
            else:
                pnl_points = active_pos["entry"] - exit_px
                r = pnl_points / (active_pos["stop"] - active_pos["entry"])
            trades.append(_finalize_trade(symbol, active_pos, last["ts_ny"], exit_px, "eod", r, pnl_points, POINT_VALUE, TICK_SIZE))
            active_pos = None

    return trades

# =============================
# RUN
# =============================
def main():
    raw = load_data(
        table=TABLE,
        interval=INTERVAL,
        symbols=SYMBOLS,
        symbol_prefix=SYMBOL_PREFIX,
        start=START_DATE,
        end=END_DATE,
    )
    if raw.empty:
        print("No data returned. Check TABLE/SYMBOLS/DATE/INTERVAL.")
        return

    results: List[Trade] = []
    for sym, sdf in raw.groupby("symbol"):
        trades = simulate_intraday(sym, sdf.reset_index(drop=True))
        results.extend(trades)

    if not results:
        print("No trades generated with current params.")
        return

    tl = pd.DataFrame([t.__dict__ for t in results]).sort_values(["day", "entry_time"])
    # per-symbol instrument spec for proper tick calc was already applied in Trade creation
    tl["cum_pnl_usd"] = tl.groupby("symbol")["pnl_usd"].cumsum()

    # Summary
    wins = (tl["pnl_usd"] > 0).sum()
    losses = (tl["pnl_usd"] <= 0).sum()
    total = len(tl)
    winrate = wins / total if total else 0.0
    total_pnl = tl["pnl_usd"].sum()
    dd = (tl["cum_pnl_usd"].groupby(tl["symbol"]).apply(lambda s: float((s.cummax()-s).max())).sum())

    print("\n=== TRADE SUMMARY (All Symbols) ===")
    print(f"Trades: {total} | Wins: {wins} | Losses: {losses} | Winrate: {winrate:.1%}")
    print(f"Total PnL: ${total_pnl:,.2f}")
    print(f"Approx Max Drawdown (sum of per-symbol): ${dd:,.2f}")

    # Save trade log
    out_csv = SAVE_DIR / "divergence_levels_backtest_trades.csv"
    tl.to_csv(out_csv, index=False)
    print(f"\nSaved trade log → {out_csv.resolve()}")

if __name__ == "__main__":
    main()
```

```python
import math
from dataclasses import dataclass
from pathlib import Path
from typing import List, Optional, Tuple, Dict

import numpy as np
import pandas as pd
from sqlalchemy import create_engine, text

# =============================
# DB & RUN SETTINGS (EDIT ME)
# =============================

ENGINE_URL = "postgresql+psycopg2://admin:quest@24.128.195.68:8812/qdb"
TABLE      = "futures_candles"     # columns: symbol, interval, event_at, open, high, low, close, volume

INTERVAL   = "1m"
SYMBOLS    = ["NQ"]                 # or [] and use SYMBOL_PREFIX
SYMBOL_PREFIX: Optional[str] = None # e.g., "NQ" to match NQ* contracts

# Optional UTC window
START_DATE: Optional[str] = None    # "2024-10-01"
END_DATE:   Optional[str] = None    # "2024-12-01"

# Outputs
SAVE_DIR = Path("."); SAVE_DIR.mkdir(parents=True, exist_ok=True)

# =============================
# MARKET & STRATEGY PARAMETERS
# =============================

TZ_LOCAL = "America/New_York"
REG_START = "09:30"   # RTH window
REG_END   = "16:00"

RSI_LEN = 14
SWING_K = 5
ATR_LEN = 14

ATR_MULT_STOP = 1.0
R_MULT_TARGET = 1.5
CONFIRM_ENTRY = True

USE_LEVELS               = True
LEVEL_PROX_THRESHOLD_ATR = 0.5
USE_DAILY_OPEN_AS_LEVEL  = True

# Default instrument specs (used as fallback)
DEFAULT_POINT_VALUE = 20.0
DEFAULT_TICK_SIZE   = 0.25

# =============================
# ENGINE (QuestDB-friendly)
# =============================

engine = create_engine(
    ENGINE_URL,
    isolation_level="AUTOCOMMIT",   # critical for QuestDB PG wire
    pool_pre_ping=True,
    future=True,
)

# =============================
# Helpers
# =============================

def _literal_in_list(values: List[str]) -> str:
    """Safely quote list for SQL IN ('a','b',...)."""
    return ",".join("'" + str(v).replace("'", "''") + "'" for v in values)

def to_ny(ts: pd.Series) -> pd.Series:
    # Treat incoming timestamps as UTC; convert to NY tz
    return pd.to_datetime(ts, utc=True).dt.tz_convert(TZ_LOCAL)

def rsi(close: pd.Series, length: int = 14) -> pd.Series:
    delta = close.diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    roll_up = up.ewm(alpha=1/length, adjust=False).mean()
    roll_down = down.ewm(alpha=1/length, adjust=False).mean()
    rs = roll_up / roll_down.replace(0, np.nan)
    out = 100 - (100 / (1 + rs))
    return out.fillna(method="bfill")

def atr(df: pd.DataFrame, length: int = 14) -> pd.Series:
    high, low, close = df["high"], df["low"], df["close"]
    prev_close = close.shift(1)
    tr = pd.concat([
        (high - low),
        (high - prev_close).abs(),
        (low - prev_close).abs()
    ], axis=1).max(axis=1)
    return tr.ewm(alpha=1/length, adjust=False).mean()

def is_swing_high(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).max())

def is_swing_low(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).min())

def nearest_level(price: float, levels: List[float]) -> Tuple[float, float]:
    if not levels:
        return (np.nan, np.inf)
    d = [abs(price - lv) for lv in levels]
    i = int(np.argmin(d))
    return (levels[i], d[i])

def _spec_for(symbol: str) -> Tuple[float, float]:
    """Return (POINT_VALUE, TICK_SIZE) for the given symbol."""
    s = (symbol or "").upper()
    if s.startswith("MNQ"):   return 2.0, 0.25            # Micro NQ
    if s.startswith("NQ"):    return 20.0, 0.25           # E-mini NQ
    if s.startswith("MES"):   return 5.0, 0.25            # Micro ES
    if s.startswith("ES"):    return 50.0, 0.25           # E-mini ES
    if s.startswith("M2K"):   return 5.0, 0.10            # Micro RTY
    if s.startswith("RTY"):   return 50.0, 0.10           # E-mini RTY
    if s.startswith("MYM"):   return 0.5, 1.0             # Micro YM ($0.5/point; 1 point tick)
    if s.startswith("YM"):    return 5.0, 1.0             # E-mini YM
    return DEFAULT_POINT_VALUE, DEFAULT_TICK_SIZE

@dataclass
class Trade:
    symbol: str
    entry_time: pd.Timestamp
    side: str
    entry: float
    stop: float
    target: float
    exit_time: pd.Timestamp
    exit_price: float
    reason: str
    r_mult: float
    points: float
    ticks: int
    pnl_usd: float
    day: str

# =============================
# Data (READ-ONLY)
# =============================

def load_data(
    symbols: List[str],
    interval: str,
    start: Optional[str] = None,
    end: Optional[str] = None,
    symbol_prefix: Optional[str] = None,
) -> pd.DataFrame:

    where = ["interval = :interval"]
    params = {"interval": interval}

    if symbols and len(symbols) > 0 and symbol_prefix is None:
        where.append(f"symbol IN ({_literal_in_list(symbols)})")
    elif symbol_prefix:
        where.append("symbol LIKE :prefix")
        params["prefix"] = f"{symbol_prefix}%"

    if start:
        where.append("event_at >= TIMESTAMP :start")
        params["start"] = start
    if end:
        where.append("event_at < TIMESTAMP :end")
        params["end"] = end

    sql = text(f"""
        SELECT symbol, interval, event_at, open, high, low, close, volume
        FROM {TABLE}
        WHERE {' AND '.join(where)}
        ORDER BY symbol, event_at
    """)

    with engine.connect() as conn:
        df = pd.read_sql_query(sql, conn, params=params)

    if df.empty:
        return df

    df["ts_ny"] = to_ny(df["event_at"])
    df["date_ny"] = df["ts_ny"].dt.date
    return df

# =============================
# Signals (Divergence + Levels)
# =============================

def build_signals(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    # Indicators
    df["rsi"] = rsi(df["close"], RSI_LEN)
    df["atr"] = atr(df, ATR_LEN)

    # Swing points (centered window)
    df["sh"]      = is_swing_high(df["high"], SWING_K).fillna(False)  # price swing high
    df["sl"]      = is_swing_low(df["low"],  SWING_K).fillna(False)   # price swing low
    df["rsi_sh"]  = is_swing_high(df["rsi"],  SWING_K).fillna(False)  # rsi swing high
    df["rsi_sl"]  = is_swing_low(df["rsi"],   SWING_K).fillna(False)  # rsi swing low

    # Capture swing values as Series (not numpy arrays)
    df["price_sh"]   = df["high"].where(df["sh"])
    df["price_sl"]   = df["low"].where(df["sl"])
    df["rsi_sh_val"] = df["rsi"].where(df["rsi_sh"])
    df["rsi_sl_val"] = df["rsi"].where(df["rsi_sl"])

    # Previous swing values (current swing compared to previous swing)
    df["prev_price_sh"] = df["price_sh"].ffill().shift(1)
    df["prev_price_sl"] = df["price_sl"].ffill().shift(1)
    df["prev_rsi_sh"]   = df["rsi_sh_val"].ffill().shift(1)
    df["prev_rsi_sl"]   = df["rsi_sl_val"].ffill().shift(1)

    # Divergence rules
    df["bear_div"] = (
        df["sh"] & df["rsi_sh"] &
        (df["high"] > df["prev_price_sh"]) &
        (df["rsi"]  < df["prev_rsi_sh"])
    ).fillna(False)

    df["bull_div"] = (
        df["sl"] & df["rsi_sl"] &
        (df["low"] < df["prev_price_sl"]) &
        (df["rsi"] > df["prev_rsi_sl"])
    ).fillna(False)

    return df

def daily_levels(day_df: pd.DataFrame, prev_day_df: Optional[pd.DataFrame]) -> List[float]:
    levels = []
    if prev_day_df is not None and not prev_day_df.empty:
        levels += [prev_day_df["high"].max(), prev_day_df["low"].min()]
    if USE_DAILY_OPEN_AS_LEVEL and not day_df.empty:
        levels.append(float(day_df.iloc[0]["open"]))
    return sorted(list({round(lv, 4) for lv in levels}))

# =============================
# Backtest
# =============================

def simulate_intraday(symbol: str, df: pd.DataFrame) -> List[Trade]:
    trades: List[Trade] = []
    df = build_signals(df)

    POINT_VALUE, TICK_SIZE = _spec_for(symbol)

    for day, g in df.groupby("date_ny"):
        day_slice = g[(g["ts_ny"].dt.strftime("%H:%M") >= REG_START) &
                      (g["ts_ny"].dt.strftime("%H:%M") <= REG_END)].copy()
        if day_slice.empty:
            continue

        prev_df = df[df["date_ny"] == (pd.to_datetime(day) - pd.Timedelta(days=1)).date()]
        levels = daily_levels(day_slice, prev_df)
        active_pos = None

        for i in range(1, len(day_slice)):
            prev = day_slice.iloc[i-1]
            row  = day_slice.iloc[i]

            # manage open position
            if active_pos:
                if active_pos["side"] == "long":
                    if row["low"] <= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        pnl_points = exit_px - active_pos["entry"]
                        r = pnl_points / (active_pos["entry"] - active_pos["stop"])
                        trades.append(_finalize(symbol, active_pos, row["ts_ny"], exit_px, "stop",
                                               r, pnl_points, POINT_VALUE, TICK_SIZE))
                        active_pos = None
                    elif row["high"] >= active_pos["target"]:
                        exit_px = active_pos["target"]
                        pnl_points = exit_px - active_pos["entry"]
                        trades.append(_finalize(symbol, active_pos, row["ts_ny"], exit_px, "target",
                                               R_MULT_TARGET, pnl_points, POINT_VALUE, TICK_SIZE))
                        active_pos = None
                else:  # short
                    if row["high"] >= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        pnl_points = active_pos["entry"] - exit_px
                        r = pnl_points / (active_pos["stop"] - active_pos["entry"])
                        trades.append(_finalize(symbol, active_pos, row["ts_ny"], exit_px, "stop",
                                               r, pnl_points, POINT_VALUE, TICK_SIZE))
                        active_pos = None
                    elif row["low"] <= active_pos["target"]:
                        exit_px = active_pos["target"]
                        pnl_points = active_pos["entry"] - exit_px
                        trades.append(_finalize(symbol, active_pos, row["ts_ny"], exit_px, "target",
                                               R_MULT_TARGET, pnl_points, POINT_VALUE, TICK_SIZE))
                        active_pos = None

            # if flat, look for signals
            if active_pos is None:
                atr_now = max(1e-6, row["atr"])
                near_level = True
                if USE_LEVELS and levels:
                    _, dist = nearest_level(row["close"], levels)
                    near_level = (dist <= LEVEL_PROX_THRESHOLD_ATR * atr_now)

                # bearish divergence → short
                if prev["bear_div"] and near_level:
                    signal_hi = prev["high"]
                    entry = min(row["open"], row["low"]) if CONFIRM_ENTRY else row["low"]
                    entry = min(entry, signal_hi - TICK_SIZE)  # below signal bar
                    stop  = entry + ATR_MULT_STOP * atr_now
                    target = entry - R_MULT_TARGET * (stop - entry)
                    active_pos = {"side": "short", "entry_time": row["ts_ny"], "entry": entry,
                                  "stop": stop, "target": target, "reason": "bear_div"}

                # bullish divergence → long
                elif prev["bull_div"] and near_level:
                    signal_lo = prev["low"]
                    entry = max(row["open"], row["high"]) if CONFIRM_ENTRY else row["high"]
                    entry = max(entry, signal_lo + TICK_SIZE)  # above signal bar
                    stop  = entry - ATR_MULT_STOP * atr_now
                    target = entry + R_MULT_TARGET * (entry - stop)
                    active_pos = {"side": "long", "entry_time": row["ts_ny"], "entry": entry,
                                  "stop": stop, "target": target, "reason": "bull_div"}

        # flatten at session end
        if active_pos is not None:
            last = day_slice.iloc[-1]
            exit_px = last["close"]
            if active_pos["side"] == "long":
                pnl_points = exit_px - active_pos["entry"]
                r = pnl_points / (active_pos["entry"] - active_pos["stop"])
            else:
                pnl_points = active_pos["entry"] - exit_px
                r = pnl_points / (active_pos["stop"] - active_pos["entry"])
            trades.append(_finalize(symbol, active_pos, last["ts_ny"], exit_px, "eod",
                                   r, pnl_points, POINT_VALUE, TICK_SIZE))
            active_pos = None

    return trades

def _finalize(symbol: str, pos: Dict, exit_time: pd.Timestamp, exit_price: float,
              reason: str, r_mult: float, pnl_points: float,
              POINT_VALUE: float, TICK_SIZE: float) -> Trade:
    ticks = int(round(pnl_points / TICK_SIZE))
    pnl_usd = pnl_points * POINT_VALUE
    return Trade(
        symbol=symbol,
        entry_time=pos["entry_time"], side=pos["side"],
        entry=pos["entry"], stop=pos["stop"], target=pos["target"],
        exit_time=exit_time, exit_price=exit_price,
        reason=reason, r_mult=r_mult,
        points=pnl_points, ticks=ticks, pnl_usd=pnl_usd,
        day=str(pos["entry_time"].date())
    )

# =============================
# Run
# =============================

def main():
    raw = load_data(SYMBOLS, INTERVAL, START_DATE, END_DATE, SYMBOL_PREFIX)
    if raw.empty:
        print("No data returned. Check TABLE/SYMBOLS/DATE/INTERVAL window.")
        return

    results: List[Trade] = []
    for sym, sdf in raw.groupby("symbol"):
        trades = simulate_intraday(sym, sdf.reset_index(drop=True))
        results.extend(trades)

    if not results:
        print("No trades generated with current params.")
        return

    tl = pd.DataFrame([t.__dict__ for t in results]).sort_values(["day", "entry_time"])
    tl["cum_pnl_usd"] = tl["pnl_usd"].cumsum()

    wins = (tl["pnl_usd"] > 0).sum()
    losses = (tl["pnl_usd"] <= 0).sum()
    total = len(tl)
    winrate = wins / total if total else 0.0

    print("\n=== TRADE SUMMARY ===")
    print(f"Trades: {total} | Wins: {wins} | Losses: {losses} | Winrate: {winrate:.1%}")
    print(f"Total PnL: ${tl['pnl_usd'].sum():.2f} | Max DD: ${float((tl['cum_pnl_usd'].cummax()-tl['cum_pnl_usd']).max()):.2f}")

    out_csv = SAVE_DIR / "divergence_levels_backtest_trades.csv"
    tl.to_csv(out_csv, index=False)
    print(f"\nSaved trade log → {out_csv.resolve()}")

if __name__ == "__main__":
    main()
```
