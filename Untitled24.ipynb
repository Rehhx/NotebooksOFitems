```python
import requests
import pandas as pd
import time

# Your SportsDataIO API key
API_KEY = "b08e876c6d53418dbc224b988eb82293"

# URLs
GAMES_URL = f"https://api.sportsdata.io/v3/cfb/scores/json/Games/2024?key={API_KEY}"
PROPS_URL_TEMPLATE = "https://api.sportsdata.io/v3/cfb/odds/json/BettingPlayerPropsByGameID/{gameId}?key={api_key}"

# STEP 1: Get all 2024 GameIDs
print("Fetching all 2024 CFB games...")
response = requests.get(GAMES_URL)
if response.status_code != 200:
    print(f"Error fetching game list: {response.status_code}")
    exit()

games = response.json()
game_ids = [game["GameID"] for game in games]

# Optional: save games list
df_games = pd.DataFrame(games)
df_games.to_csv("cfb_2024_game_ids.csv", index=False)
print(f"Retrieved {len(game_ids)} GameIDs")

# STEP 2: Loop through GameIDs to get player props
all_props = []

print("Fetching player props for each game...")
for i, game_id in enumerate(game_ids):
    url = PROPS_URL_TEMPLATE.format(gameId=game_id, api_key=API_KEY)
    res = requests.get(url)

    if res.status_code == 200:
        try:
            props = res.json()
            for prop in props:
                for outcome in prop.get("PlayerProps", []):
                    all_props.append({
                        "GameID": prop.get("GameID"),
                        "PlayerName": outcome.get("PlayerName"),
                        "Team": outcome.get("Team"),
                        "StatType": outcome.get("StatType"),
                        "Line": outcome.get("Value"),
                        "OverOdds": outcome.get("OverOdds"),
                        "UnderOdds": outcome.get("UnderOdds"),
                        "Sportsbook": outcome.get("SportsBook")
                    })
        except Exception as e:
            print(f"Error parsing props for GameID {game_id}: {e}")
    else:
        print(f"Failed to get props for GameID {game_id}: {res.status_code}")

    # Avoid rate limits
    time.sleep(0.5)

# STEP 3: Save to CSV
df_props = pd.DataFrame(all_props)
df_props.to_csv("cfb_player_props_2024.csv", index=False)
print("✅ Saved all player props to cfb_player_props_2024.csv")
```
