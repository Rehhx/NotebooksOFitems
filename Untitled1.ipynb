```python
!pip install snscrape
!pip install --upgrade certifi
import ssl
import certifi

ssl_context = ssl.create_default_context(cafile=certifi.where())

print(f"SSL Certificate Path: {certifi.where()}")
print(f"SSL Context: {ssl_context}")


import snscrape.modules.twitter as sntwitter
import pandas as pd

# Define search query for "$HAWK"
query = "$HAWK since:2024-01-01 until:2024-02-28 lang:en"
max_tweets = 10  # Number of tweets to scrape

# List to store tweet data
tweets = []

# Scrape tweets
for i, tweet in enumerate(sntwitter.TwitterSearchScraper(query).get_items()):
    if i >= max_tweets:
        break
    tweets.append([
        tweet.date, tweet.user.username, tweet.content, 
        tweet.likeCount, tweet.retweetCount
    ])

# Convert to DataFrame
if tweets:  # Check if tweets were collected
    df = pd.DataFrame(tweets, columns=["Date", "Username", "Tweet", "Likes", "Retweets"])
    
    # Ensure proper data types
    df["Likes"] = df["Likes"].astype(int)
    df["Retweets"] = df["Retweets"].astype(int)

    # Count total tweets
    total_tweets = df.shape[0]

    # Sum up likes and impressions (likes + retweets)
    total_likes = df["Likes"].sum()
    total_impressions = df["Likes"].sum() + df["Retweets"].sum()

    # Display results
    print(f"Total Tweets: {total_tweets}")
    print(f"Total Likes: {total_likes}")
    print(f"Total Impressions (Likes + Retweets): {total_impressions}")

    # Save to CSV (optional)
    df.to_csv("hawk_tweets_analysis.csv", index=False)

else:
    print("No tweets found for the given query.")





```
