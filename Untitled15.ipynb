```python
import pandas as pd

# Load the CSV file
file_path = r'C:\\Users\\pcagm\\Downloads\\time_series_data_checkpoint_cleaned.csv'
data = pd.read_csv(file_path)

# Clean whitespace from the 'symbol' column
data['symbol'] = data['symbol'].str.strip()

# Prompt user for token symbol
symbol = input("Enter the token symbol to extract data: ").strip()

# Filter the data by token symbol (case-insensitive and trimmed)
filtered_data = data[data['symbol'].str.upper() == symbol.upper()]

# Check if data exists for the given symbol
if filtered_data.empty:
    print(f"No data found for token symbol: {symbol}")
else:
    print(f"Data for {symbol}:")
    print(filtered_data.to_string(index=False))
```

```python
import pandas as pd

# Define the file path
file_path = r'C:\\Users\\pcagm\\Downloads\\time_series_data_checkpoint_cleaned.csv'

# Initialize an empty DataFrame to store extracted tokens
extracted_tokens = pd.DataFrame()

# Continuously prompt user for token symbols until they enter 'quit'
while True:
    symbol = input("Enter the token symbol to extract data (or type 'quit' to exit): ").strip()

    if symbol.lower() == 'quit':
        print("Exiting...")
        break

    # Load the CSV file in chunks to handle large files
    chunksize = 10000
    found = False

    for chunk in pd.read_csv(file_path, chunksize=chunksize):
        chunk['symbol'] = chunk['symbol'].str.strip()
        filtered_data = chunk[chunk['symbol'].str.upper() == symbol.upper()]

        if not filtered_data.empty:
            print(f"Data for {symbol}:")
            print(filtered_data.to_string(index=False))

            extracted_tokens = pd.concat([extracted_tokens, filtered_data], ignore_index=True)
            found = True

    if not found:
        print(f"No data found for token symbol: {symbol}")

# Optional: Save extracted tokens to a CSV file
output_file = 'extracted_tokens.csv'
extracted_tokens.to_csv(output_file, index=False)
print(f"Extracted tokens saved to {output_file}")
```
