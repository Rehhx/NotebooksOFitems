```python
import pandas as pd
import time
from selenium import webdriver
from selenium.webdriver.common.by import By

# Load the CSV file
file_path = r"C:\Users\pcagm\Downloads\split_file_6.csv"
df = pd.read_csv(file_path)

# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors='coerce')

# Filter token IDs and dates where wasRekt = 1 and volume is empty
rekt_data = df[(df["wasRekt"] == 1) & (df["volume"].isna())][["id", "Date"]].dropna()

# Initialize Selenium WebDriver
options = webdriver.ChromeOptions()
driver = webdriver.Chrome(options=options)

historical_data = []

for index, row in rekt_data.iterrows():
    try:
        token_id = row["id"]
        date_str = row["Date"]
        date_obj = pd.to_datetime(date_str, errors='coerce')
        if pd.isna(date_obj):
            print(f"Skipping {token_id} due to invalid date format")
            continue

        max_years_back = 7 * 365  # Maximum search range in days (7 years)
        max_years_forward = (pd.Timestamp.today() - date_obj).days  # Days until current date
        
        # Backward search intervals
        backward_intervals = [90, 180, 365, 730, 1095]  # 3 months, 6 months, 1 year, 2 years, 3 years
        while backward_intervals[-1] < max_years_back:
            backward_intervals.append(backward_intervals[-1] + 90)  # 3-month intervals after 3 years

        # Forward search intervals
        forward_intervals = [1, 3, 9, 27, 60, 120, 180]  # 1 day, 3 days, 9 days, 27 days, 2 months, 4 months, 6 months
        while forward_intervals[-1] < max_years_forward:
            forward_intervals.append(forward_intervals[-1] + 90)  # 3-month intervals after 6 months

        data_found = False

        # Backward search
        for days_back in backward_intervals:
            if data_found:
                break
            start_date = (date_obj - pd.Timedelta(days=days_back)).strftime("%Y-%m-%d")
            end_date = (date_obj - pd.Timedelta(days=days_back - 1)).strftime("%Y-%m-%d")
            url = f"https://www.coingecko.com/en/coins/{token_id}/historical_data?start={start_date}&end={end_date}"
            driver.get(url)
            print(f"Opened: {url}")
            time.sleep(5)

            try:
                rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")
                if rows:
                    row = rows[0]  # Only grab the first available row
                    cols = row.find_elements(By.TAG_NAME, "td")
                    if len(cols) >= 4:
                        date = cols[0].text
                        market_cap = cols[1].text
                        volume = cols[2].text
                        open_price = cols[3].text
                        close_price = cols[4].text if len(cols) > 4 else "N/A"
                        historical_data.append([token_id, date, market_cap, volume, open_price, close_price])
                        print(f"{token_id} - {date}: Market Cap={market_cap}, Volume={volume}, Open={open_price}, Close={close_price}")
                        data_found = True
            except:
                print(f"No historical data found for {token_id} on {start_date}")

        # Forward search if no data found
        if not data_found:
            for days_forward in forward_intervals:
                start_date = (date_obj + pd.Timedelta(days=days_forward)).strftime("%Y-%m-%d")
                end_date = (date_obj + pd.Timedelta(days=days_forward + 1)).strftime("%Y-%m-%d")
                url = f"https://www.coingecko.com/en/coins/{token_id}/historical_data?start={start_date}&end={end_date}"
                driver.get(url)
                print(f"Opened: {url}")
                time.sleep(5)

                try:
                    rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")
                    if rows:
                        row = rows[0]
                        cols = row.find_elements(By.TAG_NAME, "td")
                        if len(cols) >= 4:
                            date = cols[0].text
                            market_cap = cols[1].text
                            volume = cols[2].text
                            open_price = cols[3].text
                            close_price = cols[4].text if len(cols) > 4 else "N/A"
                            historical_data.append([token_id, date, market_cap, volume, open_price, close_price])
                            print(f"{token_id} - {date}: Market Cap={market_cap}, Volume={volume}, Open={open_price}, Close={close_price}")
                            data_found = True
                            break
                except:
                    print(f"No historical data found for {token_id} on {start_date}")
    
    except Exception as e:
        print(f"Error fetching historical data for {token_id}: {e}")

# Convert to DataFrame and save
historical_df = pd.DataFrame(historical_data, columns=["id", "Date", "market_cap", "volume", "open", "close"])
historical_df.to_csv("historical_data_split_file_6.csv", index=False)
print("Historical data saved to historical_data_split_file_6.csv")

driver.quit()
```
