```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import time

# Initialize WebDriver
driver = webdriver.Chrome()

# Open the De.Fi Rekt Database webpage
driver.get("https://de.fi/rekt-database")

# Wait for the page to load fully
WebDriverWait(driver, 15).until(EC.presence_of_element_located((By.CLASS_NAME, "mono-selector-header")))

# Click the dropdown menu to open options
dropdown = driver.find_element(By.CLASS_NAME, "mono-selector-header")
dropdown.click()

# Wait for dropdown items to appear
WebDriverWait(driver, 10).until(EC.presence_of_all_elements_located((By.CLASS_NAME, "item")))

# Find and click the "Show by 200" option
options = driver.find_elements(By.CLASS_NAME, "item")
for option in options:
    if "Show by 200" in option.text:
        option.click()
        break

# Wait for the table to refresh and fully load
time.sleep(5)
WebDriverWait(driver, 15).until(EC.presence_of_element_located((By.CLASS_NAME, "scam-database-body")))

# Scroll down to ensure all rows are loaded
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(5)

# Extract all table rows
rows = driver.find_elements(By.CLASS_NAME, "grid-row-template.minimized")

# Lists to store extracted data
tokens, chains, categories, issue_types, funds_lost, dates = [], [], [], [], [], []

# Debugging: Print number of rows found
print(f"Rows found: {len(rows)}")

# Loop through each row and extract data
for row in rows:
    try:
        token = row.find_element(By.CSS_SELECTOR, ".name").text
        chain = row.find_element(By.CSS_SELECTOR, ".column.tokens.with-extra-info.network").text
        category = row.find_element(By.CSS_SELECTOR, ".column.column-rekt-function").text
        issue_type = row.find_element(By.CSS_SELECTOR, ".column.with-extra-info.column-rekt-function").text
        fund_lost = row.find_element(By.CSS_SELECTOR, ".funds-lost").text
        date = row.find_element(By.CSS_SELECTOR, ".column.date").text
        
        tokens.append(token)
        chains.append(chain)
        categories.append(category)
        issue_types.append(issue_type)
        funds_lost.append(fund_lost)
        dates.append(date)

    except Exception as e:
        print(f"Error extracting row: {e}")

# Close the browser
driver.quit()

# Create a DataFrame
df = pd.DataFrame({
    "Token Name": tokens,
    "Chain": chains,
    "Category": categories,
    "Type of Issue": issue_types,
    "Funds Lost": funds_lost,
    "Date": dates
})

# Save to CSV
df.to_csv("rekt_database.csv", index=False)

# Display the DataFrame
print(df)
```

```python
import os
import time
import random
import pandas as pd
from datetime import datetime, timedelta
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains

# File paths
REKT_DATABASE_PATH = r"C:\Users\pcagm\Downloads\rekt_database_with_ticker.xls"
DOWNLOAD_FOLDER = r"C:\Users\pcagm\Downloads"
FINAL_CSV_PATH = os.path.join(DOWNLOAD_FOLDER, "extracted_market_data.csv")

# Setup Selenium
chrome_options = webdriver.ChromeOptions()
chrome_options.add_experimental_option("prefs", {
    "download.default_directory": DOWNLOAD_FOLDER,
    "download.prompt_for_download": False,
    "download.directory_upgrade": True,
    "safebrowsing.enabled": True
})

# Initialize WebDriver
driver = webdriver.Chrome(options=chrome_options)
driver.get("https://coinmarketcap.com")
wait = WebDriverWait(driver, 15)
actions = ActionChains(driver)

# Function to find elements robustly
def find_element_stale_proof(by, value, retries=3, wait_time=10):
    for attempt in range(retries):
        try:
            return WebDriverWait(driver, wait_time).until(EC.presence_of_element_located((by, value)))
        except Exception as e:
            print(f"Retrying finding element: {value} (Attempt {attempt+1}) - Error: {e}")
            time.sleep(2)
    raise Exception(f"Failed to find element after {retries} attempts: {value}")

# Load rekt database XLS file
rekt_data = pd.read_excel(REKT_DATABASE_PATH)

# Filter valid ticker symbols
valid_tickers = rekt_data[(rekt_data["Ticker Symbol"].notna()) & (rekt_data["Ticker Symbol"] != "N/A")]

# Convert the "Date" column to datetime format
valid_tickers["Date"] = pd.to_datetime(valid_tickers["Date"], errors='coerce')

# Create a DataFrame to store results
extracted_data = pd.DataFrame(columns=["Date", "MarketCap", "Volume"])

# Iterate through each ticker
for index, row in valid_tickers.iterrows():
    ticker = row["Ticker Symbol"]
    funds_lost_date = row["Date"]
    prev_date = funds_lost_date - timedelta(days=1)  # One day before funds lost

    print(f"\n🔍 Searching for {ticker} and capturing data for {prev_date.strftime('%Y-%m-%d')}")

    # Step 1: Search for the ticker
    try:
        search_icon = find_element_stale_proof(By.CLASS_NAME, "sc-f48f9b47-1")
        driver.execute_script("arguments[0].click();", search_icon)
        time.sleep(2)

        search_input = find_element_stale_proof(By.CLASS_NAME, "search-input.desktop-input")
        search_input.send_keys(ticker)
        time.sleep(random.randint(5, 7))
        search_input.send_keys(Keys.ENTER)
        time.sleep(5)
    except Exception as e:
        print(f"Error searching for {ticker}: {e}")
        continue

    # Step 2: Click "All" button
    try:
        all_button = find_element_stale_proof(By.XPATH, "//h5[contains(text(),'All')]")
        driver.execute_script("arguments[0].click();", all_button)
        time.sleep(2)
    except Exception as e:
        print(f"Error clicking All button for {ticker}: {e}")
        continue

    # Step 3: Hover over the three dots (`...`) button
    try:
        three_dots_button = find_element_stale_proof(By.XPATH, "//div[contains(@class, 'range-tab-custom-btn')]")
        actions.move_to_element(three_dots_button).perform()
        time.sleep(2)

        # Wait for dropdown to appear
        dropdown_container = find_element_stale_proof(By.CLASS_NAME, "dropdown-container")
        actions.move_to_element(dropdown_container).perform()
        time.sleep(2)
    except Exception as e:
        print(f"Error hovering over three dots button for {ticker}: {e}")
        continue

    # Step 4: Click "Download price history (.csv)"
    try:
        csv_download_button = find_element_stale_proof(By.XPATH, "//button[contains(., 'Download price history (.csv)')]")
        actions.move_to_element(csv_download_button).click().perform()
        print(f"✅ CSV file download started for {ticker}!")
        time.sleep(10)  # Wait for download to complete
    except Exception as e:
        print(f"Error clicking CSV download button for {ticker}: {e}")
        continue

    # Step 5: Locate the downloaded CSV and extract data
    try:
        # Find latest downloaded CSV file
        downloaded_files = sorted(os.listdir(DOWNLOAD_FOLDER), key=lambda x: os.path.getmtime(os.path.join(DOWNLOAD_FOLDER, x)), reverse=True)
        csv_file_path = os.path.join(DOWNLOAD_FOLDER, downloaded_files[0])
        
        # Load CSV
        price_data = pd.read_csv(csv_file_path)

        # Convert timestamp to datetime
        price_data["timestamp"] = pd.to_datetime(price_data["timestamp"], errors='coerce')

        # Filter for the row before the "Funds Lost" date
        filtered_data = price_data[price_data["timestamp"] < prev_date]
        latest_entry = filtered_data.iloc[-1] if not filtered_data.empty else None

        if latest_entry is not None:
            market_cap = latest_entry["marketCap"]
            volume = latest_entry["volume"]
            print(f"📊 Extracted for {ticker} on {latest_entry['timestamp']} -> Market Cap: {market_cap}, Volume: {volume}")

            # Add data to DataFrame
            extracted_data.loc[ticker] = [latest_entry["timestamp"], market_cap, volume]
        else:
            print(f"⚠️ No historical data found for {ticker} before {prev_date.strftime('%Y-%m-%d')}")

    except Exception as e:
        print(f"Error processing CSV for {ticker}: {e}")
        continue

# Close the browser
driver.quit()

# Save extracted data to CSV
extracted_data.to_csv(FINAL_CSV_PATH, index=True)

print(f"\n🎉 Process completed! Extracted data saved as '{FINAL_CSV_PATH}'")
```
