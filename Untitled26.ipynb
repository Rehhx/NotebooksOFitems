```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import pandas as pd
import time

def scrape_247_transfers():
    url = "https://247sports.com/Season/2024-Football/TransferPortal/"

    # Setup headless Chrome
    options = Options()
    options.add_argument("--headless=new")
    driver = webdriver.Chrome(service=Service(), options=options)

    try:
        driver.get(url)
        time.sleep(5)  # wait for full table render

        players = []
        rows = driver.find_elements(By.CSS_SELECTOR, "div.TransferPortalPlayer")

        for row in rows:
            try:
                name = row.find_element(By.CSS_SELECTOR, "a.name-link").text
                pos = row.find_element(By.CSS_SELECTOR, "div.position").text
                from_school = row.find_element(By.CSS_SELECTOR, "div.from span.school").text
                to_school = row.find_element(By.CSS_SELECTOR, "div.to span.school").text if "to" in row.text else "Uncommitted"
                status = row.find_element(By.CSS_SELECTOR, "div.status").text
                date = row.find_element(By.CSS_SELECTOR, "div.entry-date").text

                players.append({
                    "Player": name,
                    "Position": pos,
                    "From School": from_school,
                    "To School": to_school,
                    "Transfer Date": date,
                    "Status": status
                })
            except Exception as e:
                print(f"Skipping row: {e}")

        df = pd.DataFrame(players)
        df.to_csv("cfb_transfer_tracker_2024.csv", index=False)
        print("✅ Saved: cfb_transfer_tracker_2024.csv")
        return df

    finally:
        driver.quit()

if __name__ == "__main__":
    df = scrape_247_transfers()
    print(df.head())
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time

def visit_each_acc_transfer_team():
    url = "https://247sports.com/college/transfer-portal/season/2025-football/transferteamrankings/?Conference=ACC"

    # Set up headless Chrome
    options = Options()
    options.add_argument("--headless=new")
    driver = webdriver.Chrome(service=Service(), options=options)

    try:
        driver.get(url)
        time.sleep(5)  # Wait for page to render

        # Find all team links in the ACC rankings list
        team_links = driver.find_elements(By.CSS_SELECTOR, "div.team a.rankings-page__name-link")
        print(f"🔗 Found {len(team_links)} team links")

        for i, link in enumerate(team_links):
            team_name = link.text.strip()
            team_url = link.get_attribute("href")
            print(f"🔄 Visiting [{i+1}/{len(team_links)}] {team_name} → {team_url}")

            # Open team transfer portal page
            driver.execute_script("window.open(arguments[0]);", team_url)
            driver.switch_to.window(driver.window_handles[-1])

            time.sleep(2)  # simulate page view / give time to load

            # Close the tab and return to main page
            driver.close()
            driver.switch_to.window(driver.window_handles[0])

        print("✅ All team pages visited.")

    finally:
        driver.quit()

if __name__ == "__main__":
    visit_each_acc_transfer_team()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time
import pandas as pd

def scrape_acc_transfer_players():
    start_url = "https://247sports.com/college/transfer-portal/season/2025-football/transferteamrankings/?Conference=ACC"

    # Setup Chrome
    options = Options()
    options.add_argument("--headless=new")
    driver = webdriver.Chrome(service=Service(), options=options)

    all_players = []

    try:
        driver.get(start_url)
        time.sleep(5)

        # Get all team URLs from ACC rankings
        team_links = driver.find_elements(By.CSS_SELECTOR, "div.team a.rankings-page__name-link")

        for i, team in enumerate(team_links):
            team_name = team.text.strip()
            team_url = team.get_attribute("href")
            print(f"\n🔄 Visiting [{i+1}/{len(team_links)}] {team_name} → {team_url}")

            driver.execute_script("window.open(arguments[0]);", team_url)
            driver.switch_to.window(driver.window_handles[-1])
            time.sleep(4)

            try:
                player_elements = driver.find_elements(By.CSS_SELECTOR, "li.transfer-player")

                for el in player_elements:
                    try:
                        name_el = el.find_element(By.CSS_SELECTOR, "h3 a")
                        name = name_el.text.strip()
                        profile_url = name_el.get_attribute("href").strip()

                        pos = el.find_element(By.CSS_SELECTOR, "div.position").text.strip()
                        rating = el.find_element(By.CSS_SELECTOR, "div.rating").text.strip()
                        status = el.find_element(By.CSS_SELECTOR, "div.status").text.strip()

                        # HT/WT might be missing on some entries
                        try:
                            htwt = el.find_element(By.CSS_SELECTOR, "div.htwt").text.strip()
                        except:
                            htwt = ""

                        # From → To team (via logo alt text)
                        logos = el.find_elements(By.CSS_SELECTOR, "div.transfer-teams img")
                        from_team = logos[0].get_attribute("alt") if len(logos) > 0 else ""
                        to_team = logos[1].get_attribute("alt") if len(logos) > 1 else team_name

                        all_players.append({
                            "Player": name,
                            "Profile URL": profile_url,
                            "Rating": rating,
                            "Position": pos,
                            "HT/WT": htwt,
                            "Status": status,
                            "From School": from_team,
                            "To School": to_team,
                            "Team Page": team_url
                        })

                    except Exception as e:
                        print(f"⚠️ Skipping player row: {e}")

            except Exception as e:
                print(f"⚠️ Failed to process players for {team_name}: {e}")

            driver.close()
            driver.switch_to.window(driver.window_handles[0])

    finally:
        driver.quit()

    df = pd.DataFrame(all_players)
    df.to_csv("acc_2025_incoming_transfers_full.csv", index=False)
    print("\n✅ Full ACC transfer data saved to acc_2025_incoming_transfers_full.csv")
    return df

if __name__ == "__main__":
    df = scrape_acc_transfer_players()
    print(df.head())
```
