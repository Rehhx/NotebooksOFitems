```python
# orb_15m_retest_MAG7_scaled_1R_2R_3R.py
# ORB 15m Retest entry, 3-target scale-out:
#  - TP@1R -> move stop to BE
#  - TP@2R -> move stop to +1R
#  - TP@3R -> move stop to +2R (flat)
#
# Runs only the Magnificent 7 from PARQUET_DIR and prints per-symbol + overall metrics.
# Saves per-symbol CSVs and an ALL CSV.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
TZ_LOCAL    = "America/New_York"

REG_START = "09:30"   # session start
OR_END    = "09:45"   # end of first 15-min bar (exclusive)
REG_END   = "16:00"   # session end

STOP_FIRST_IF_BOTH_HIT = True
SAVE_DIR = Path(".")  # where to write CSVs

# Only run the Magnificent 7
MAG7 = {"AAPL", "MSFT", "AMZN", "GOOGL", "META", "NVDA", "TSLA"}

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    """
    Robustly parse EventAt:
    - If tz-naive: assume UTC then convert to New York.
    - If tz-aware: convert to New York.
    """
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    """Return ('long'/'short', idx) for the first bar *closing* outside the OR."""
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    """First bar that *touches* the level after the breakout close."""
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def pnl(side, entry, exit_px, qty=1.0):
    per_share = (exit_px - entry) if side == "long" else (entry - exit_px)
    return qty * per_share

def simulate_scaled(after_entry_df, side, entry, stop0, R, stop_first=True):
    """
    Simulate 3-unit scale-out with staged stops:
      - Hit T1=1R: realize 1 unit at T1, move stop -> BE
      - Hit T2=2R: realize 1 unit at T2, move stop -> entry+/-1R
      - Hit T3=3R: realize last unit at T3 (flat)
    If a bar hits both stop and a target, use `stop_first` to resolve ordering.

    Returns:
      dict with:
        'Outcome': 'ALL_TP' | 'STOP' | 'EOD_PARTIAL' | 'EOD'
        'Realized_PnL': float  (sum over realized units)
        'ExitTime': last event time index (pandas index) where flat or EOD
        'Hits': dict(flags for T1/T2/T3)
        'FinalStop': last stop level used (for diagnostics)
    """
    # Targets for long/short
    sgn = +1 if side == "long" else -1
    T1 = entry + sgn * R
    T2 = entry + sgn * 2*R
    T3 = entry + sgn * 3*R

    # Active state
    stop = stop0
    units_remaining = 3
    realized = 0.0
    hits = {"Hit1R": False, "Hit2R": False, "Hit3R": False}
    last_idx = None

    for idx, row in after_entry_df.iterrows():
        lo, hi, close = float(row["Low"]), float(row["High"]), float(row["Close"])
        last_idx = idx

        # At most 3 fills can occur on a single bar (rare but possible if wide range)
        while units_remaining > 0:
            crossed_stop   = (lo <= stop) if side == "long" else (hi >= stop)
            next_target_px = T1 if not hits["Hit1R"] else (T2 if not hits["Hit2R"] else (T3 if not hits["Hit3R"] else None))
            if next_target_px is None:
                # No more targets; we're only managing trailing stop on last unit(s)
                # If stop is not hit on this bar, break to next bar
                if crossed_stop:
                    # Stop out remaining units
                    realized += pnl(side, entry, stop, qty=units_remaining)
                    units_remaining = 0
                    return {
                        "Outcome": "STOP",
                        "Realized_PnL": realized,
                        "ExitTime": idx,
                        "Hits": hits,
                        "FinalStop": stop,
                    }
                else:
                    break  # proceed to next bar

            crossed_target = (hi >= next_target_px) if side == "long" else (lo <= next_target_px)

            if crossed_stop and crossed_target:
                # Ambiguous same-bar touch: choose ordering by policy
                if stop_first:
                    realized += pnl(side, entry, stop, qty=units_remaining)
                    units_remaining = 0
                    return {
                        "Outcome": "STOP",
                        "Realized_PnL": realized,
                        "ExitTime": idx,
                        "Hits": hits,
                        "FinalStop": stop,
                    }
                else:
                    # Fill the target first (1 unit), update stop, and loop again to see if further levels also hit
                    realized += pnl(side, entry, next_target_px, qty=1.0)
                    if not hits["Hit1R"]:
                        hits["Hit1R"] = True
                        # move stop to BE
                        stop = entry
                    elif not hits["Hit2R"]:
                        hits["Hit2R"] = True
                        # move stop to +1R
                        stop = entry + sgn * R
                    else:
                        hits["Hit3R"] = True
                        units_remaining -= 1
                        # last unit taken; if units_remaining becomes 0 -> done
                        if units_remaining == 0:
                            return {
                                "Outcome": "ALL_TP",
                                "Realized_PnL": realized,
                                "ExitTime": idx,
                                "Hits": hits,
                                "FinalStop": stop,
                            }
                        # if somehow more units remain (shouldn't), continue
                        continue

                    # we consumed 1 unit at target
                    units_remaining -= 1
                    # continue the while loop to see if the same bar also hits stop or next target

            elif crossed_target:
                # Hit the next target cleanly
                realized += pnl(side, entry, next_target_px, qty=1.0)
                if not hits["Hit1R"]:
                    hits["Hit1R"] = True
                    stop = entry  # move to BE
                elif not hits["Hit2R"]:
                    hits["Hit2R"] = True
                    stop = entry + sgn * R  # move to +1R
                else:
                    hits["Hit3R"] = True
                    units_remaining -= 1
                    if units_remaining == 0:
                        return {
                            "Outcome": "ALL_TP",
                            "Realized_PnL": realized,
                            "ExitTime": idx,
                            "Hits": hits,
                            "FinalStop": stop,
                        }
                    # continue to see if more happens in same bar
                    continue

                units_remaining -= 1
                # keep looping to allow multiple targets in same bar

            elif crossed_stop:
                # Stopped on remaining units
                realized += pnl(side, entry, stop, qty=units_remaining)
                units_remaining = 0
                return {
                    "Outcome": "STOP",
                    "Realized_PnL": realized,
                    "ExitTime": idx,
                    "Hits": hits,
                    "FinalStop": stop,
                }
            else:
                # nothing hit on this bar; move to next bar
                break

    # If we get here, session ended with units remaining: mark-to-market at last close
    if units_remaining > 0:
        # Realize remaining at last close
        last_close = float(after_entry_df.loc[last_idx, "Close"])
        realized += pnl(side, entry, last_close, qty=units_remaining)
        return {
            "Outcome": "EOD_PARTIAL" if any(hits.values()) else "EOD",
            "Realized_PnL": realized,
            "ExitTime": last_idx,
            "Hits": hits,
            "FinalStop": stop,
        }

    # Flat already (shouldn’t reach)
    return {
        "Outcome": "ALL_TP",
        "Realized_PnL": realized,
        "ExitTime": last_idx,
        "Hits": hits,
        "FinalStop": stop,
    }

def calc_metrics(trades_df, pnl_col):
    pnl = trades_df[pnl_col].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= CORE BACKTEST =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Keep only regular session minutes
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    records = []

    for d, day in df.groupby("date"):
        # Opening range: first 15-minute bar [09:30, 09:45)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # After OR window (from 09:45 to close)
        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        # 1) Wait for first confirmed breakout by CLOSE
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # 2) After that breakout close, place retest entry at the broken level
        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        # 3) Wait for the first bar that *touches* the broken level → entry
        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop0      = orl if side == "long" else orh
        R          = (orh - orl)

        # Forward simulate after the entry bar with scaled exits
        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            # No bars after entry; flat at entry (no PnL)
            records.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": R,
                "EntryTime": entry_time, "Entry": entry, "Stop0": stop0,
                "Outcome_scaled": "EOD",
                "PnL_scaled_$": 0.0,
                "Hit1R": False, "Hit2R": False, "Hit3R": False,
                "ExitTime_scaled": entry_time,
            })
            continue

        sim = simulate_scaled(
            after_entry_df=after_entry,
            side=side,
            entry=entry,
            stop0=stop0,
            R=R,
            stop_first=STOP_FIRST_IF_BOTH_HIT
        )

        records.append({
            "Date": d, "Symbol": symbol, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": R,
            "EntryTime": entry_time, "Entry": entry, "Stop0": stop0,
            "Outcome_scaled": sim["Outcome"],
            "PnL_scaled_$": sim["Realized_PnL"],
            "Hit1R": bool(sim["Hits"]["Hit1R"]),
            "Hit2R": bool(sim["Hits"]["Hit2R"]),
            "Hit3R": bool(sim["Hits"]["Hit3R"]),
            "ExitTime_scaled": day.loc[sim["ExitTime"], "EventAt"] if sim["ExitTime"] in day.index else None,
            "FinalStop": sim["FinalStop"],
        })

    return pd.DataFrame(records)

# ========= RUN MAG 7 =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    # filter to MAG7 by stem
    wanted = {f.stem.upper(): f for f in files if f.stem.upper() in MAG7}
    if not wanted:
        print("No MAG7 parquet files found. Expected one or more of:", ", ".join(sorted(MAG7)))
        return

    all_trades = []
    for sym in sorted(wanted.keys()):
        f = wanted[sym]
        try:
            trades = backtest_one_symbol(f, sym)
        except Exception as e:
            print(f"[SKIP] {sym}: {e}")
            continue

        if trades.empty:
            print(f"=== ORB 15m Retest (Scaled Exits) — {sym} ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            continue

        all_trades.append(trades)

        # Per-symbol metrics + print
        N, total_pnl, winrate, pf = calc_metrics(trades, pnl_col="PnL_scaled_$")

        save_path = SAVE_DIR / f"orb_trades_{sym}_scaled_1R_2R_3R.csv"
        trades.to_csv(save_path, index=False)

        print(f"=== ORB 15m Retest (Scaled Exits) — {sym} ===")
        print(f"Trades: {N} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {pf:.3f}")
        print(f"Saved trades → {save_path.resolve()}")

    # ===== OVERALL roll-up =====
    if not all_trades:
        print("\nNo trades generated for MAG7 symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    N, total_pnl, winrate, pf = calc_metrics(overall, pnl_col="PnL_scaled_$")

    save_all = SAVE_DIR / "orb_trades_MAG7_scaled_1R_2R_3R_ALL.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== ORB 15m Retest (Scaled Exits) — OVERALL MAG7 ===")
    print(f"Trades: {N} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_MAG7_scaled_100_shares.py
# ORB 15m Retest entry, scaled exits for a 100-share position:
#  - Take 33 shares @ 1R -> move stop to BE
#  - Take 33 shares @ 2R -> move stop to +1R
#  - Take 34 shares @ 3R -> move stop to +2R (flat)
#
# Runs only the Magnificent 7 from PARQUET_DIR and prints per-symbol + overall metrics.
# Saves per-symbol CSVs and an ALL CSV.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
TZ_LOCAL    = "America/New_York"

REG_START = "09:30"   # session start
OR_END    = "09:45"   # end of first 15-min bar (exclusive)
REG_END   = "16:00"   # session end

STOP_FIRST_IF_BOTH_HIT = True
SAVE_DIR = Path(".")

# Only run the Magnificent 7
MAG7 = {"AAPL", "MSFT", "AMZN", "GOOGL", "META", "NVDA", "TSLA"}

# === Position sizing (exact integer shares) ===
UNITS_SHARES = 100
TP_SPLITS = [33, 33, 34]  # at 1R, 2R, 3R respectively; must sum to UNITS_SHARES

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def pnl(side, entry, exit_px, qty=1.0):
    per_share = (exit_px - entry) if side == "long" else (entry - exit_px)
    return qty * per_share

def simulate_scaled_shares(after_entry_df, side, entry, stop0, R, tp_splits, stop_first=True):
    """
    100-share style (integer) scale-out with stop ratchets:
      - Take tp_splits[0] @ 1R -> stop -> BE
      - Take tp_splits[1] @ 2R -> stop -> entry +/- 1R
      - Take tp_splits[2] @ 3R -> stop -> entry +/- 2R (flat)
    If both stop and target are hit in the same bar, ordering is chosen via stop_first.

    Returns dict:
      Outcome, Realized_PnL, ExitTime, Hits{Hit1R,Hit2R,Hit3R}, FinalStop
    """
    assert sum(tp_splits) > 0, "TP_SPLITS must sum to > 0"
    sgn = +1 if side == "long" else -1
    T1 = entry + sgn * R
    T2 = entry + sgn * 2*R
    T3 = entry + sgn * 3*R

    stop = stop0
    units_remaining = int(sum(tp_splits))
    realized = 0.0
    hits = {"Hit1R": False, "Hit2R": False, "Hit3R": False}
    last_idx = None

    # Copy splits so we can decrement as we fill
    splits = tp_splits.copy()

    for idx, row in after_entry_df.iterrows():
        lo, hi, close = float(row["Low"]), float(row["High"]), float(row["Close"])
        last_idx = idx

        # Loop in-case multiple events happen on same bar
        while units_remaining > 0:
            crossed_stop = (lo <= stop) if side == "long" else (hi >= stop)
            # Determine next target and shares to sell
            if not hits["Hit1R"]:
                next_target_px = T1
                next_qty = splits[0]
            elif not hits["Hit2R"]:
                next_target_px = T2
                next_qty = splits[1]
            elif not hits["Hit3R"]:
                next_target_px = T3
                next_qty = splits[2]
            else:
                next_target_px = None
                next_qty = 0

            if next_target_px is None:
                # No more targets; just manage stop for any remaining shares (should be 0)
                if crossed_stop:
                    realized += pnl(side, entry, stop, qty=units_remaining)
                    units_remaining = 0
                    return {"Outcome":"STOP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
                break

            crossed_target = (hi >= next_target_px) if side == "long" else (lo <= next_target_px)

            if crossed_stop and crossed_target:
                if stop_first:
                    realized += pnl(side, entry, stop, qty=units_remaining)
                    units_remaining = 0
                    return {"Outcome":"STOP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
                # else: process target first and loop again
                realized += pnl(side, entry, next_target_px, qty=next_qty)
                units_remaining -= next_qty
                if not hits["Hit1R"]:
                    hits["Hit1R"] = True
                    splits[0] = 0
                    stop = entry
                elif not hits["Hit2R"]:
                    hits["Hit2R"] = True
                    splits[1] = 0
                    stop = entry + sgn * R
                else:
                    hits["Hit3R"] = True
                    splits[2] = 0
                    if units_remaining == 0:
                        return {"Outcome":"ALL_TP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
                continue

            if crossed_target:
                realized += pnl(side, entry, next_target_px, qty=next_qty)
                units_remaining -= next_qty
                if not hits["Hit1R"]:
                    hits["Hit1R"] = True
                    splits[0] = 0
                    stop = entry  # move to BE
                elif not hits["Hit2R"]:
                    hits["Hit2R"] = True
                    splits[1] = 0
                    stop = entry + sgn * R  # move to +1R
                else:
                    hits["Hit3R"] = True
                    splits[2] = 0
                    if units_remaining == 0:
                        return {"Outcome":"ALL_TP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
                continue

            if crossed_stop:
                realized += pnl(side, entry, stop, qty=units_remaining)
                units_remaining = 0
                return {"Outcome":"STOP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}

            # neither hit → move to next bar
            break

    # Session end with leftover shares → mark to market at last close
    if units_remaining > 0:
        last_close = float(after_entry_df.loc[last_idx, "Close"])
        realized += pnl(side, entry, last_close, qty=units_remaining)
        return {"Outcome":"EOD_PARTIAL" if any(hits.values()) else "EOD",
                "Realized_PnL":realized,"ExitTime":last_idx,"Hits":hits,"FinalStop":stop}

    return {"Outcome":"ALL_TP","Realized_PnL":realized,"ExitTime":last_idx,"Hits":hits,"FinalStop":stop}

def calc_metrics(trades_df, pnl_col):
    pnl = trades_df[pnl_col].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= CORE BACKTEST =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Keep only regular session minutes
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    records = []

    for d, day in df.groupby("date"):
        # Opening range: first 15-minute bar [09:30, 09:45)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        R   = orh - orl
        if R <= 0:
            continue

        # After OR window (from 09:45 to close)
        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        # 1) First confirmed breakout by CLOSE
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # 2) Place retest entry at broken level
        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        # 3) First bar that touches the broken level → entry
        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop0      = orl if side == "long" else orh

        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            records.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": R,
                "EntryTime": entry_time, "Entry": entry, "Stop0": stop0,
                "Outcome_scaled": "EOD",
                "PnL_scaled_$": 0.0,
                "Hit1R": False, "Hit2R": False, "Hit3R": False,
                "ExitTime_scaled": entry_time,
                "Shares": UNITS_SHARES
            })
            continue

        sim = simulate_scaled_shares(
            after_entry_df=after_entry,
            side=side,
            entry=entry,
            stop0=stop0,
            R=R,
            tp_splits=TP_SPLITS,
            stop_first=STOP_FIRST_IF_BOTH_HIT
        )

        records.append({
            "Date": d, "Symbol": symbol, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": R,
            "EntryTime": entry_time, "Entry": entry, "Stop0": stop0,
            "Outcome_scaled": sim["Outcome"],
            "PnL_scaled_$": sim["Realized_PnL"],
            "Hit1R": bool(sim["Hits"]["Hit1R"]),
            "Hit2R": bool(sim["Hits"]["Hit2R"]),
            "Hit3R": bool(sim["Hits"]["Hit3R"]),
            "ExitTime_scaled": day.loc[sim["ExitTime"], "EventAt"] if sim["ExitTime"] in day.index else None,
            "FinalStop": sim["FinalStop"],
            "Shares": UNITS_SHARES
        })

    return pd.DataFrame(records)

# ========= RUN MAG 7 =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    # filter to MAG7 by stem
    wanted = {f.stem.upper(): f for f in files if f.stem.upper() in MAG7}
    if not wanted:
        print("No MAG7 parquet files found. Expected one or more of:", ", ".join(sorted(MAG7)))
        return

    all_trades = []
    for sym in sorted(wanted.keys()):
        f = wanted[sym]
        try:
            trades = backtest_one_symbol(f, sym)
        except Exception as e:
            print(f"[SKIP] {sym}: {e}")
            continue

        if trades.empty:
            print(f"=== ORB 15m Retest (100 sh, Scaled Exits) — {sym} ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            continue

        all_trades.append(trades)

        N, total_pnl, winrate, pf = calc_metrics(trades, pnl_col="PnL_scaled_$")

        save_path = SAVE_DIR / f"orb_trades_{sym}_scaled_100sh.csv"
        trades.to_csv(save_path, index=False)

        print(f"=== ORB 15m Retest (100 sh, Scaled Exits) — {sym} ===")
        print(f"Trades: {N} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {pf:.3f}")
        print(f"Saved trades → {save_path.resolve()}")

    if not all_trades:
        print("\nNo trades generated for MAG7 symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    N, total_pnl, winrate, pf = calc_metrics(overall, pnl_col="PnL_scaled_$")

    save_all = SAVE_DIR / "orb_trades_MAG7_scaled_100sh_ALL.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== ORB 15m Retest (100 sh, Scaled Exits) — OVERALL MAG7 ===")
    print(f"Trades: {N} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_MAG7_scaled_1R_2R_3R.py
# ORB 15m Retest entry, 3-target scale-out:
#  - TP@1R -> move stop to BE
#  - TP@2R -> move stop to +1R
#  - TP@3R -> move stop to +2R (flat)
#
# Runs only the Magnificent 7 from PARQUET_DIR and prints per-symbol + overall metrics.
# Saves per-symbol CSVs and an ALL CSV.
# Charts:
#  (1) overall equity curve (cumulative PnL)
#  (2) yearly total PnL
#  (3) yearly PnL by ticker (CSV + grouped bars + stacked bars)

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# NEW
import matplotlib
matplotlib.use("Agg")  # headless-safe
import matplotlib.pyplot as plt

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
TZ_LOCAL    = "America/New_York"

REG_START = "09:30"   # session start
OR_END    = "09:45"   # end of first 15-min bar (exclusive)
REG_END   = "16:00"   # session end

STOP_FIRST_IF_BOTH_HIT = True
SAVE_DIR = Path(".")  # where to write CSVs/PNGs

# Only run the Magnificent 7
MAG7 = {"AAPL", "MSFT", "AMZN", "GOOGL", "META", "NVDA", "TSLA"}

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def pnl(side, entry, exit_px, qty=1.0):
    per_share = (exit_px - entry) if side == "long" else (entry - exit_px)
    return qty * per_share

def simulate_scaled(after_entry_df, side, entry, stop0, R, stop_first=True):
    sgn = +1 if side == "long" else -1
    T1 = entry + sgn * R
    T2 = entry + sgn * 2*R
    T3 = entry + sgn * 3*R

    stop = stop0
    units_remaining = 3
    realized = 0.0
    hits = {"Hit1R": False, "Hit2R": False, "Hit3R": False}
    last_idx = None

    for idx, row in after_entry_df.iterrows():
        lo, hi, close = float(row["Low"]), float(row["High"]), float(row["Close"])
        last_idx = idx

        while units_remaining > 0:
            crossed_stop   = (lo <= stop) if side == "long" else (hi >= stop)
            next_target_px = T1 if not hits["Hit1R"] else (T2 if not hits["Hit2R"] else (T3 if not hits["Hit3R"] else None))
            if next_target_px is None:
                if crossed_stop:
                    realized += pnl(side, entry, stop, qty=units_remaining)
                    units_remaining = 0
                    return {"Outcome":"STOP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
                else:
                    break

            crossed_target = (hi >= next_target_px) if side == "long" else (lo <= next_target_px)

            if crossed_stop and crossed_target:
                if stop_first:
                    realized += pnl(side, entry, stop, qty=units_remaining)
                    units_remaining = 0
                    return {"Outcome":"STOP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
                else:
                    realized += pnl(side, entry, next_target_px, qty=1.0)
                    if not hits["Hit1R"]:
                        hits["Hit1R"] = True; stop = entry
                    elif not hits["Hit2R"]:
                        hits["Hit2R"] = True; stop = entry + sgn * R
                    else:
                        hits["Hit3R"] = True
                        units_remaining -= 1
                        if units_remaining == 0:
                            return {"Outcome":"ALL_TP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
                        continue
                    units_remaining -= 1

            elif crossed_target:
                realized += pnl(side, entry, next_target_px, qty=1.0)
                if not hits["Hit1R"]:
                    hits["Hit1R"] = True; stop = entry
                elif not hits["Hit2R"]:
                    hits["Hit2R"] = True; stop = entry + sgn * R
                else:
                    hits["Hit3R"] = True
                    units_remaining -= 1
                    if units_remaining == 0:
                        return {"Outcome":"ALL_TP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
                    continue
                units_remaining -= 1

            elif crossed_stop:
                realized += pnl(side, entry, stop, qty=units_remaining)
                units_remaining = 0
                return {"Outcome":"STOP","Realized_PnL":realized,"ExitTime":idx,"Hits":hits,"FinalStop":stop}
            else:
                break

    if units_remaining > 0:
        last_close = float(after_entry_df.loc[last_idx, "Close"])
        realized += pnl(side, entry, last_close, qty=units_remaining)
        return {"Outcome":"EOD_PARTIAL" if any(hits.values()) else "EOD",
                "Realized_PnL":realized,"ExitTime":last_idx,"Hits":hits,"FinalStop":stop}

    return {"Outcome":"ALL_TP","Realized_PnL":realized,"ExitTime":last_idx,"Hits":hits,"FinalStop":stop}

def calc_metrics(trades_df, pnl_col):
    pnl = trades_df[pnl_col].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= CORE BACKTEST =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Keep only regular session minutes
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    records = []

    for d, day in df.groupby("date"):
        # Opening range: first 15-minute bar [09:30, 09:45)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # After OR window (from 09:45 to close)
        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        # 1) Wait for first confirmed breakout by CLOSE
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # 2) After that breakout close, place retest entry at the broken level
        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:   # <-- FIXED: was .empty()
            continue

        # 3) Wait for the first bar that *touches* the broken level → entry
        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop0      = orl if side == "long" else orh
        R          = (orh - orl)

        # Forward simulate after the entry bar with scaled exits
        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            # No bars after entry; flat at entry (no PnL)
            records.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": R,
                "EntryTime": entry_time, "Entry": entry, "Stop0": stop0,
                "Outcome_scaled": "EOD",
                "PnL_scaled_$": 0.0,
                "Hit1R": False, "Hit2R": False, "Hit3R": False,
                "ExitTime_scaled": entry_time,
            })
            continue

        sim = simulate_scaled(
            after_entry_df=after_entry,
            side=side,
            entry=entry,
            stop0=stop0,
            R=R,
            stop_first=STOP_FIRST_IF_BOTH_HIT
        )

        records.append({
            "Date": d, "Symbol": symbol, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": R,
            "EntryTime": entry_time, "Entry": entry, "Stop0": stop0,
            "Outcome_scaled": sim["Outcome"],
            "PnL_scaled_$": sim["Realized_PnL"],
            "Hit1R": bool(sim["Hits"]["Hit1R"]),
            "Hit2R": bool(sim["Hits"]["Hit2R"]),
            "Hit3R": bool(sim["Hits"]["Hit3R"]),
            "ExitTime_scaled": day.loc[sim["ExitTime"], "EventAt"] if sim["ExitTime"] in day.index else None,
            "FinalStop": sim["FinalStop"],
        })

    return pd.DataFrame(records)

# ========= CHART HELPERS =========
def _ensure_datetime(series):
    s = pd.to_datetime(series, errors="coerce")
    if getattr(s.dt, "tz", None) is None:
        s = s.dt.tz_localize(TZ_LOCAL)
    return s

def make_pnl_charts(overall_trades: pd.DataFrame, save_dir: Path):
    if overall_trades.empty:
        print("No trades for plotting.")
        return

    df = overall_trades.copy()
    ts = df["ExitTime_scaled"].copy()
    if ts.isna().all():
        ts = df["EntryTime"]
    ts = _ensure_datetime(ts)
    df["When"] = ts

    df = df.sort_values("When")
    df["CumPnL_$"] = df["PnL_scaled_$"].fillna(0.0).cumsum()

    # (1) Overall equity curve
    fig1, ax1 = plt.subplots(figsize=(12, 6))
    ax1.plot(df["When"], df["CumPnL_$"])
    ax1.set_title("MAG7 ORB 15m Retest — Overall Equity Curve (Cumulative PnL)")
    ax1.set_xlabel("Date"); ax1.set_ylabel("Cumulative PnL ($)")
    ax1.grid(True, alpha=0.3)
    fig1.tight_layout()
    out1 = save_dir / "equity_MAG7_overall.png"
    fig1.savefig(out1, dpi=144); plt.close(fig1)
    print(f"Saved chart → {out1.resolve()}")

    # (2) Yearly total PnL (all tickers together)
    df["Year"] = df["When"].dt.year
    yearly_total = df.groupby("Year", dropna=True)["PnL_scaled_$"].sum().reset_index()

    fig2, ax2 = plt.subplots(figsize=(10, 5))
    ax2.bar(yearly_total["Year"].astype(str), yearly_total["PnL_scaled_$"])
    ax2.set_title("MAG7 ORB 15m Retest — Yearly Total PnL (All Tickers)")
    ax2.set_xlabel("Year"); ax2.set_ylabel("PnL ($)")
    ax2.grid(True, axis="y", alpha=0.3)
    fig2.tight_layout()
    out2 = save_dir / "yearly_pnl_MAG7.png"
    fig2.savefig(out2, dpi=144); plt.close(fig2)
    print(f"Saved chart → {out2.resolve()}")

    # (3) Yearly PnL by ticker (CSV + grouped & stacked bars)
    by_ticker = df.groupby(["Year", "Symbol"], dropna=True)["PnL_scaled_$"].sum().reset_index()
    pivot = by_ticker.pivot(index="Year", columns="Symbol", values="PnL_scaled_$").fillna(0.0).sort_index()
    # Save CSV
    csv_path = save_dir / "yearly_pnl_by_ticker_MAG7.csv"
    pivot.to_csv(csv_path)
    print(f"Saved CSV  → {csv_path.resolve()}")

    # Grouped bar chart
    years = pivot.index.astype(int).tolist()
    symbols = list(pivot.columns)
    values = pivot.values

    x = np.arange(len(years))
    width = 0.8 / max(1, len(symbols))

    fig3, ax3 = plt.subplots(figsize=(max(10, 1.2*len(years)), 6))
    for i, sym in enumerate(symbols):
        ax3.bar(x + i*width, pivot[sym].values, width, label=sym)
    ax3.set_xticks(x + (len(symbols)-1)*width/2)
    ax3.set_xticklabels([str(y) for y in years], rotation=0)
    ax3.set_title("MAG7 ORB 15m Retest — Yearly PnL by Ticker (Grouped)")
    ax3.set_xlabel("Year"); ax3.set_ylabel("PnL ($)")
    ax3.grid(True, axis="y", alpha=0.3)
    ax3.legend(ncols=min(4, len(symbols)))
    fig3.tight_layout()
    out3 = save_dir / "yearly_pnl_by_ticker_grouped.png"
    fig3.savefig(out3, dpi=144); plt.close(fig3)
    print(f"Saved chart → {out3.resolve()}")

    # Stacked bar chart
    fig4, ax4 = plt.subplots(figsize=(max(10, 1.2*len(years)), 6))
    bottoms = np.zeros(len(years))
    for sym in symbols:
        ax4.bar([str(y) for y in years], pivot[sym].values, bottom=bottoms, label=sym)
        bottoms += pivot[sym].values
    ax4.set_title("MAG7 ORB 15m Retest — Yearly PnL by Ticker (Stacked)")
    ax4.set_xlabel("Year"); ax4.set_ylabel("PnL ($)")
    ax4.grid(True, axis="y", alpha=0.3)
    ax4.legend(ncols=min(4, len(symbols)))
    fig4.tight_layout()
    out4 = save_dir / "yearly_pnl_by_ticker_stacked.png"
    fig4.savefig(out4, dpi=144); plt.close(fig4)
    print(f"Saved chart → {out4.resolve()}")

# ========= RUN MAG 7 =========
def main():
    SAVE_DIR.mkdir(parents=True, exist_ok=True)

    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    # filter to MAG7 by stem
    wanted = {f.stem.upper(): f for f in files if f.stem.upper() in MAG7}
    if not wanted:
        print("No MAG7 parquet files found. Expected one or more of:", ", ".join(sorted(MAG7)))
        return

    all_trades = []
    for sym in sorted(wanted.keys()):
        f = wanted[sym]
        try:
            trades = backtest_one_symbol(f, sym)
        except Exception as e:
            print(f"[SKIP] {sym}: {e}")
            continue

        if trades.empty:
            print(f"=== ORB 15m Retest (Scaled Exits) — {sym} ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            continue

        # Per-symbol metrics + save CSV
        N, total_pnl, winrate, pf = calc_metrics(trades, pnl_col="PnL_scaled_$")
        save_path = SAVE_DIR / f"orb_trades_{sym}_scaled_1R_2R_3R.csv"
        trades.to_csv(save_path, index=False)

        print(f"=== ORB 15m Retest (Scaled Exits) — {sym} ===")
        print(f"Trades: {N} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {pf:.3f}")
        print(f"Saved trades → {save_path.resolve()}")

        all_trades.append(trades)

    if not all_trades:
        print("\nNo trades generated for MAG7 symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    N, total_pnl, winrate, pf = calc_metrics(overall, pnl_col="PnL_scaled_$")

    save_all = SAVE_DIR / "orb_trades_MAG7_scaled_1R_2R_3R_ALL.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== ORB 15m Retest (Scaled Exits) — OVERALL MAG7 ===")
    print(f"Trades: {N} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

    # Charts
    make_pnl_charts(overall, SAVE_DIR)

if __name__ == "__main__":
    main()
```
