```python
import pandas as pd

# File paths
leases_path = r"C:\Users\pcagm\Downloads\Leases.csv"
unemp_path = r"C:\Users\pcagm\Downloads\Unemployment.csv"

# Load data
leases_df = pd.read_csv(leases_path)
unemp_df = pd.read_csv(unemp_path)

# Strip whitespace and convert to string for consistent merging
for col in ['state', 'year', 'quarter']:
    leases_df[col] = leases_df[col].astype(str).str.strip()
    unemp_df[col] = unemp_df[col].astype(str).str.strip()

# Merge on state, year, quarter using left join to preserve leases data
merged_df = pd.merge(
    leases_df,
    unemp_df[['state', 'year', 'quarter', 'unemployment_rate']],
    on=['state', 'year', 'quarter'],
    how='left'
)

# Confirm row counts match
print(f"Original Lease Rows: {len(leases_df)}")
print(f"Merged Rows: {len(merged_df)}")

# Optional: Fill missing unemployment rates with mean
merged_df['unemployment_rate'] = merged_df['unemployment_rate'].fillna(merged_df['unemployment_rate'].mean())

# Display first few rows to verify
print(merged_df[['state', 'year', 'quarter', 'unemployment_rate']].head(10))

# Save to CSV if needed
# merged_df.to_csv(r"C:\Users\pcagm\Downloads\Merged_Leases_with_Unemployment.csv", index=False)
```
