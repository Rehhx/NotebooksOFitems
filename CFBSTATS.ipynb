```python
import pandas as pd
import requests

# URLs of the TeamRankings pages
urls = {
    'Opponent Rushing Attempts per Game': 'https://www.teamrankings.com/college-football/stat/opponent-rushing-attempts-per-game',
    'Opponent Yards per Rush Attempt': 'https://www.teamrankings.com/college-football/stat/opponent-yards-per-rush-attempt',
    'Opponent Rushing Yards per Game': 'https://www.teamrankings.com/college-football/stat/opponent-rushing-yards-per-game',
    'Opponent Rushing Play Percentage': 'https://www.teamrankings.com/college-football/stat/opponent-rushing-play-pct',
    'Opponent Rushing Yards Percentage': 'https://www.teamrankings.com/college-football/stat/opponent-rushing-yards-pct'
}

# List of Kansas State's opponents
opponents = [
    'UT Martin',
    'Tulane',
    'Arizona',
    'BYU',
    'Oklahoma St',
    'Colorado',
    'West Virginia',
    'Kansas',
    'Houston'
]

# Dictionary to store each DataFrame by statistic
dataframes = {}

# Function to fetch and parse the table from a given URL
def fetch_table(url):
    response = requests.get(url)
    tables = pd.read_html(response.text)
    df = tables[0]
    # Print the columns for debugging
    print(f"\nFetched columns from {url}: {df.columns}")
    # Standardize column names
    df.columns = df.columns.str.strip()
    return df

# Fetch data for each statistic and filter for opponents
for stat_name, url in urls.items():
    # Fetch the table from the URL
    df = fetch_table(url)
    
    # Check if 'Team' column is present
    if 'Team' not in df.columns:
        print(f"'Team' column not found in the table from {url}")
        continue
    
    # Standardize the team names
    df['Team'] = df['Team'].str.strip()
    
    # Filter the data for Kansas State's opponents
    opponents_df = df[df['Team'].isin(opponents)]
    
    # Store the filtered DataFrame in the dictionary
    dataframes[stat_name] = opponents_df

# Display each DataFrame
for stat_name, df in dataframes.items():
    print(f"\n{stat_name}:\n{df}\n")
```

```python
import pandas as pd
import requests

# URLs to fetch player stats from
urls = {
    "Rushing Plays per Rush": "https://www.teamrankings.com/college-football/player-stat/rushing-plays?rate=per-rush",
    "Rushing Net Yards per Rush": "https://www.teamrankings.com/college-football/player-stat/rushing-net-yards?rate=per-rush",
    "Rushing Yards per Attempt per Rush": "https://www.teamrankings.com/college-football/player-stat/rushing-yards-per-attempt?rate=per-rush",
    "Rushing Touchdowns per Rush": "https://www.teamrankings.com/college-football/player-stat/rushing-touchdowns?rate=per-rush",
    "Rushing Touchdowns per Game": "https://www.teamrankings.com/college-football/player-stat/rushing-touchdowns?rate=per-game",
    "Rushing Plays per Game": "https://www.teamrankings.com/college-football/player-stat/rushing-plays?rate=per-game",
    "Rushing Net Yards per Game": "https://www.teamrankings.com/college-football/player-stat/rushing-net-yards?rate=per-game",
    "Rushing Yards per Attempt per Game": "https://www.teamrankings.com/college-football/player-stat/rushing-yards-per-attempt?rate=per-game"
}

def fetch_player_stats(url):
    """Fetch player stats from a given URL."""
    try:
        response = requests.get(url)
        response.raise_for_status()
        tables = pd.read_html(response.text)
        df = tables[0]
        return df
    except Exception as e:
        print(f"Error fetching data from {url}: {e}")
        return None

# Dictionary to store dataframes for each stat category
dataframes = {}

# Fetch data from each URL and store in the dictionary
for stat_name, url in urls.items():
    df = fetch_player_stats(url)
    if df is not None:
        # Rename the 'Value' column to the specific stat name for easier merging later
        df = df.rename(columns={"Value": stat_name})
        dataframes[stat_name] = df

# Merge all dataframes on 'Player' and 'Team'
merged_df = None

for stat_name, df in dataframes.items():
    if merged_df is None:
        merged_df = df
    else:
        # Use suffixes to avoid column conflicts
        merged_df = pd.merge(merged_df, df, on=["Player", "Team"], how="outer", suffixes=('', f'_{stat_name}'))

# Standardize team names by converting to lowercase and stripping whitespace
merged_df['Team'] = merged_df['Team'].str.lower().str.strip()

# List of teams that Kansas State has played against (normalized)
kansas_opponents = [
    "tulane", "ut martin", "arizona", "byu", "oklahoma state", 
    "colorado", "west virginia", "kansas", "houston"
]
kansas_opponents = [team.lower() for team in kansas_opponents]

# Filter the merged dataframe to include only players from teams that played against Kansas State
filtered_df = merged_df[merged_df['Team'].isin(kansas_opponents)]

# Display the filtered dataframe
print("\nFiltered Dataframe for players who played against Kansas State:")
print(filtered_df.head())

# Save to CSV for further analysis if needed
filtered_df.to_csv("kansas_opponents_player_stats.csv", index=False)
```
