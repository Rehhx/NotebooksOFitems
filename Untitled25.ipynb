```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

URL = "https://www.sports-reference.com/cfb/years/2024-rushing.html"

def scrape_rushing_stats(url):
    response = requests.get(url)
    response.raise_for_status()  # raise error if site is unreachable
    soup = BeautifulSoup(response.content, 'lxml')

    table = soup.find("table", {"id": "rushing"})
    headers = [th.get_text() for th in table.find("thead").find_all("th")][1:]  # skip Rk column

    data = []
    for row in table.find("tbody").find_all("tr"):
        if "class" in row.attrs and "thead" in row["class"]:
            continue  # skip subheaders

        cells = row.find_all("td")
        if not cells:
            continue  # skip empty rows

        row_data = {}
        for i, cell in enumerate(cells):
            text = cell.get_text(strip=True)
            row_data[headers[i]] = text

            # Add player profile link
            if headers[i] == "Player":
                link = cell.find("a")
                row_data["PlayerURL"] = (
                    "https://www.sports-reference.com" + link['href'] if link else None
                )

        data.append(row_data)

    return pd.DataFrame(data)

# Run the function
df = scrape_rushing_stats(URL)

# Preview top 5 rows
print(df.head())

# Save to CSV
df.to_csv("cfb_2024_rushing_leaders.csv", index=False)
print("Saved to cfb_2024_rushing_leaders.csv")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

def scrape_2024_rushing():
    url = "https://www.sports-reference.com/cfb/years/2024-rushing.html"
    resp = requests.get(url)
    resp.raise_for_status()
    soup = BeautifulSoup(resp.text, "html.parser")

    # The main table has id="rushing"
    table = soup.find("table", id="rushing")
    if not table:
        raise ValueError("Could not find rushing stats table on page.")

    # Extract headers
    headers = [th.get_text(strip=True) for th in table.find("thead").find_all("th")]
    # First header might be a rank or blank; adjust as needed
    # Example headers: ['Rk', 'Player', 'School', 'G', 'Att', 'Yds', 'Avg', 'TD', 'Yds/G']
    headers = headers[1:]  # Drop the first if it's a serial column

    rows = []
    for row in table.find("tbody").find_all("tr"):
        # Some rows may be separators or blanks
        if row.get("class") and "thead" in row.get("class"):
            continue

        cols = [td.get_text(strip=True) for td in row.find_all("td")]
        if not cols:
            continue
        # Drop first column if 'Rk' numbering included in <th>
        # If the first <td> is 'Rk', keep it; otherwise drop any empty column
        if len(cols) == len(headers) + 1:
            cols = cols[1:]

        rows.append(cols)

    df = pd.DataFrame(rows, columns=headers)
    # Convert numeric columns to appropriate dtype
    for col in ["G", "Att", "Yds", "Avg", "TD", "Yds/G"]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col].str.replace(',', ''), errors='coerce')

    return df

if __name__ == "__main__":
    df = scrape_2024_rushing()
    print(df.head())
    # Optionally, save to CSV
    df.to_csv("2024_cfb_rushing.csv", index=False)
```
