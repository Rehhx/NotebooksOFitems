```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# URLs of the webpages containing the data
team_url = "https://cfbstats.com/2024/leader/25354/team/defense/split01/category01/sort01.html"
player_url = "https://cfbstats.com/2024/leader/25354/player/split01/category01/sort01.html"

def fetch_table_data(url):
    # Send a GET request to fetch the webpage content
    response = requests.get(url)
    response.raise_for_status()  # Ensure the request was successful

    # Parse the webpage content
    soup = BeautifulSoup(response.content, "html.parser")

    # Find the table in the webpage (adjust class or id if necessary)
    table = soup.find("table")

    # Read the table into a pandas DataFrame
    if table:
        return pd.read_html(str(table))[0]
    else:
        print(f"No table found at {url}")
        return pd.DataFrame()

def fetch_player_game_logs(player_base_url):
    # Fetch player data page content
    response = requests.get(player_base_url)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, "html.parser")

    # Extract links to player pages from the table
    table = soup.find("table")
    player_df = pd.read_html(str(table))[0]
    player_links = [a['href'] for a in table.find_all('a', href=True)]

    all_game_logs = []
    for link in player_links:
        player_page_url = f"https://cfbstats.com{link}"
        response = requests.get(player_page_url)
        response.raise_for_status()
        player_soup = BeautifulSoup(response.content, "html.parser")

        # Find and click the "Rushing" -> "Game Log" link
        rushing_section = player_soup.find('a', href=True, text="Game Log")
        if rushing_section:
            game_log_url = f"https://cfbstats.com{rushing_section['href']}"
            print(f"Fetching game log from {game_log_url}")
            game_log_df = fetch_table_data(game_log_url)
            game_log_df['Player'] = player_page_url.split('/')[-2]  # Add player identifier
            all_game_logs.append(game_log_df)
        else:
            print(f"No game log found for {player_page_url}")

    # Concatenate all game logs
    if all_game_logs:
        return pd.concat(all_game_logs, ignore_index=True)
    else:
        return pd.DataFrame()

# Fetch team data
team_data = fetch_table_data(team_url)
print("Team Data:")
print(team_data)

# Save team data to a CSV file
team_data.to_csv("team_rushing_defense_stats.csv", index=False)

# Fetch player data
player_data = fetch_table_data(player_url)
print("Player Data:")
print(player_data)

# Save player data to a CSV file
player_data.to_csv("player_rushing_stats.csv", index=False)

# Fetch player game logs
game_logs = fetch_player_game_logs(player_url)
print("Player Game Logs:")
print(game_logs)

# Save game logs to a CSV file
game_logs.to_csv("player_game_logs.csv", index=False)
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re

# List of player game log URLs
player_game_logs = [
    "https://cfbstats.com/2024/player/28/1116606/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/700/1113531/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/327/1122985/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/51/2000838/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/768/2001366/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/311/2000741/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/768/1175674/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/768/1114151/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/77/2003175/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/51/2000771/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/311/2002332/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/698/2000294/rushing/gamelog.html",
    "https://cfbstats.com/2024/player/288/2000116/rushing/gamelog.html"
]

# List of team rushing defense URLs
team_defense_logs = [
    "https://cfbstats.com/2024/team/327/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/28/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/128/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/732/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/288/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/77/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/768/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/700/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/157/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/328/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/51/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/140/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/698/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/311/rushing/defense/gamelog.html",
    "https://cfbstats.com/2024/team/521/rushing/defense/gamelog.html"
]

def fetch_table_data(url):
    """Fetch and parse an HTML table from a URL into a DataFrame."""
    response = requests.get(url)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, "html.parser")
    table = soup.find("table")
    if table:
        return pd.read_html(str(table))[0]
    else:
        print(f"No table found at {url}")
        return pd.DataFrame()

def get_all_player_logs(player_urls):
    """Fetch all player game logs and combine into a DataFrame."""
    all_logs = []
    for url in player_urls:
        print(f"Fetching player data from {url}")
        player_log = fetch_table_data(url)
        if not player_log.empty:
            player_log['Player'] = url.split('/')[-3]  # Extract player ID as identifier
            all_logs.append(player_log)
    return pd.concat(all_logs, ignore_index=True) if all_logs else pd.DataFrame()

def get_all_team_logs(team_urls):
    """Fetch all team defensive logs and combine into a DataFrame."""
    all_logs = []
    for url in team_urls:
        print(f"Fetching team defense data from {url}")
        team_log = fetch_table_data(url)
        if not team_log.empty:
            team_log['Team'] = url.split('/')[-3]  # Extract team ID as identifier
            all_logs.append(team_log)
    return pd.concat(all_logs, ignore_index=True) if all_logs else pd.DataFrame()

def convert_to_numeric(df, columns):
    """Convert specified DataFrame columns to numeric, handling errors gracefully."""
    for column in columns:
        df[column] = pd.to_numeric(df[column], errors='coerce')
    return df

def match_and_calculate_all(player_df, team_df):
    """Match player game logs against team defense logs for all teams and calculate differences."""
    results = []
    for _, player_row in player_df.iterrows():
        opponent_team = player_row['Opponent']
        team_matches = team_df[team_df['Opponent'].str.contains(re.escape(opponent_team), case=False, na=False)]
        if not team_matches.empty:
            for _, team_row in team_matches.iterrows():
                # Extract necessary stats
                try:
                    player_row = convert_to_numeric(player_row, ['Att', 'Yards', 'Avg.'])
                    team_row = convert_to_numeric(team_row, ['Att', 'Yards'])
                    player_att = player_row['Att']
                    player_yards = player_row['Yards']
                    overall_avg_att = team_row['Att']
                    overall_avg_yards = team_row['Yards']
                    
                    # Calculate differences
                    diff_att = player_att - overall_avg_att
                    diff_yards = player_yards - overall_avg_yards

                    # Append to results
                    results.append({
                        'Player': player_row['Player'],
                        'Opponent': opponent_team,
                        'Game_Att': player_att,
                        'Game_Yards': player_yards,
                        'Diff_Att': diff_att,
                        'Diff_Yards': diff_yards
                    })
                except (ValueError, TypeError):
                    continue
    return pd.DataFrame(results)

def remove_nan_rows(df):
    """Remove rows with NaN values."""
    return df.dropna()

# Fetch player game logs
player_game_logs_df = get_all_player_logs(player_game_logs)
player_game_logs_df = remove_nan_rows(player_game_logs_df)
print("Player Game Logs:")
print(player_game_logs_df)

# Fetch team defense game logs
team_defense_logs_df = get_all_team_logs(team_defense_logs)
team_defense_logs_df = remove_nan_rows(team_defense_logs_df)
print("Team Defense Logs:")
print(team_defense_logs_df)

# Match and calculate differences for all opponents
if not player_game_logs_df.empty and not team_defense_logs_df.empty:
    final_results = match_and_calculate_all(player_game_logs_df, team_defense_logs_df)
    print("Player Stats Matched Against All Opponents:")
    print(final_results)
    final_results.to_csv("player_vs_all_opponents.csv", index=False)
else:
    print("No data available for matching.")
```
