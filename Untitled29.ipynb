```python



# =========================
# Munger Indicator (S&P 500)
# Python 3.9 compatible
# =========================

# 0) Install (first time only):
# !pip install -U yfinance pandas numpy requests

import time
from typing import Optional, Tuple, List
import requests
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt

# -----------------------------
# CONFIG
# -----------------------------
WIKI_URL = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

# Touch sensitivity: None = exact touch (Low <= SMA <= High)
# e.g., set to 0.5 to count within ±0.5% of SMA_200W
TOUCH_BAND_PCT: Optional[float] = None

# “On the SMA” classification for ranking view:
# If |Pct_from_SMA| <= ON_BAND_PCT → Position = "On"
ON_BAND_PCT = 0.10  # 0.10% default (adjust to taste)

# Limit number of plots to avoid spamming
PLOT_MAX = 12

# -----------------------------
# 1) Pull S&P 500 tickers
# -----------------------------
def sp500_from_wikipedia() -> pd.DataFrame:
    html = requests.get(WIKI_URL, headers={"User-Agent":"Mozilla/5.0"}, timeout=30).text
    table = pd.read_html(html)[0]  # first table = constituents
    df = table.rename(columns={"Symbol":"SymbolRaw"}).copy()
    df["Symbol"] = df["SymbolRaw"].astype(str).str.strip()
    # Wikipedia uses "." (e.g., BRK.B); yfinance expects "-"
    df["Ticker"] = df["Symbol"].str.replace(r"\.", "-", regex=True)
    keep = ["Ticker","Symbol","Security","GICS Sector","GICS Sub-Industry","Headquarters Location"]
    return df[keep]

sp500_df = sp500_from_wikipedia()
sp500_tickers = sp500_df["Ticker"].dropna().unique().tolist()
print(f"S&P 500 tickers loaded: {len(sp500_tickers)}")

# -----------------------------
# 2) Price helpers
# -----------------------------
def download_weekly_prices(tickers: List[str]) -> pd.DataFrame:
    """Bulk weekly OHLCV; returns long DF: Ticker, Date, Open, High, Low, Close, Adj Close, Volume."""
    if not tickers:
        return pd.DataFrame()
    raw = yf.download(
        tickers, period="max", interval="1wk",
        auto_adjust=False, group_by="ticker", progress=False
    )
    frames = []
    if isinstance(raw.columns, pd.MultiIndex):
        lvl0 = set(raw.columns.get_level_values(0))
        for t in tickers:
            if t in lvl0:
                df_t = raw[t].dropna(how="all").copy()
                if df_t.empty:
                    continue
                df_t["Ticker"] = t
                frames.append(df_t.reset_index().rename(columns={"index":"Date"}))
    else:  # single ticker
        df_t = raw.dropna(how="all").copy()
        df_t["Ticker"] = tickers[0]
        frames.append(df_t.reset_index().rename(columns={"index":"Date"}))
    out = pd.concat(frames, ignore_index=True)
    out.columns = [c if isinstance(c, str) else c[0] for c in out.columns]
    return out[["Ticker","Date","Open","High","Low","Close","Adj Close","Volume"]]

def add_200w_sma(df: pd.DataFrame) -> pd.DataFrame:
    df = df.sort_values(["Ticker","Date"]).copy()
    df["SMA_200W"] = (
        df.groupby("Ticker")["Adj Close"]
          .transform(lambda s: s.rolling(200, min_periods=200).mean())
    )
    return df

def touched_200w_row(high: float, low: float, sma: float, band_pct: Optional[float]=None) -> bool:
    if np.isnan(sma):
        return False
    if band_pct is None:                     # exact touch
        return low <= sma <= high
    band = sma * (band_pct / 100.0)          # near-touch band
    return not (high < sma - band or low > sma + band)

# -----------------------------
# 3) Profitability helper
# -----------------------------
def is_profitable_ttm(ticker: str) -> Tuple[bool, Optional[float]]:
    """Profitability = TTM Net Income > 0 (fallback to last annual if quarterly missing)."""
    tk = yf.Ticker(ticker)
    ttm = None

    # Try quarterly (TTM = sum last 4)
    try:
        q = tk.quarterly_income_stmt
        if q is not None and not q.empty:
            q = q.copy()
            q.index = q.index.map(str.lower)
            key = "net income" if "net income" in q.index else ("netincome" if "netincome" in q.index else None)
            if key:
                vals = q.loc[key].dropna().astype(float)
                if len(vals) >= 4:
                    ttm = float(vals.iloc[:4].sum())
    except Exception:
        pass

    # Fallback: annual last value
    if ttm is None:
        try:
            a = tk.income_stmt
            if a is not None and not a.empty:
                a = a.copy()
                a.index = a.index.map(str.lower)
                key = "net income" if "net income" in a.index else ("netincome" if "netincome" in a.index else None)
                if key:
                    vals = a.loc[key].dropna().astype(float)
                    if len(vals) >= 1:
                        ttm = float(vals.iloc[0])
        except Exception:
            pass

    return (ttm is not None and ttm > 0, ttm)

# -----------------------------
# 4) Run scan (bulk prices → SMA → latest bar → touch → profitability)
# -----------------------------
print("Downloading weekly prices (this can take a bit for ~500 tickers)...")
prices_w = download_weekly_prices(sp500_tickers)
prices_w = add_200w_sma(prices_w)

# Latest weekly bar per ticker
latest = (
    prices_w.sort_values(["Ticker","Date"])
            .groupby("Ticker", as_index=False)
            .tail(1)
)

# Touch test on latest bar
latest["Touched200W"] = latest.apply(
    lambda r: touched_200w_row(float(r["High"]), float(r["Low"]), float(r["SMA_200W"]), band_pct=TOUCH_BAND_PCT),
    axis=1
)

# Profitability check only for those that touched
rows = []
candidates = latest[latest["Touched200W"]].copy()

for _, r in candidates.iterrows():
    t = r["Ticker"]
    try:
        profitable, ttm = is_profitable_ttm(t)
    except Exception:
        profitable, ttm = (False, None)
    if profitable:
        rows.append({
            "Ticker": t,
            "Week_End": pd.to_datetime(r["Date"]).date(),
            "Low": float(r["Low"]),
            "High": float(r["High"]),
            "Adj_Close": float(r["Adj Close"]),
            "SMA_200W": float(r["SMA_200W"]),
            "TTM_NetIncome": None if ttm is None else float(ttm)
        })
    # (optional) tiny pause to be polite to Yahoo; uncomment if you hit rate limits
    # time.sleep(0.02)

df_hits = pd.DataFrame(rows).sort_values("Ticker").reset_index(drop=True)

# -----------------------------
# 5) Distance-to-SMA columns, Position, Ranking
# -----------------------------
df_hits["Diff_from_SMA"] = df_hits["Adj_Close"] - df_hits["SMA_200W"]
df_hits["Pct_from_SMA"]  = 100.0 * df_hits["Diff_from_SMA"] / df_hits["SMA_200W"]

def _classify(pct: float) -> str:
    return "On" if abs(pct) <= ON_BAND_PCT else ("Above" if pct > 0 else "Below")

df_hits["Position"] = df_hits["Pct_from_SMA"].apply(_classify)
df_hits["AbsPct_from_SMA"] = df_hits["Pct_from_SMA"].abs()

ranked = (
    df_hits.sort_values(
        by=["Position", "AbsPct_from_SMA"],
        ascending=[True, True],
        key=lambda col: col.map({"On":0,"Above":1,"Below":2}) if col.name=="Position" else col
    ).reset_index(drop=True)
)

# -----------------------------
# 6) Display & save
# -----------------------------
print(f"\nTotal hits (profitable + touch 200W): {len(df_hits)}")
display(ranked)

ranked.to_csv("munger_sp500_hits_ranked.csv", index=False)
df_hits.to_csv("munger_sp500_hits.csv", index=False)
print("Saved CSVs: munger_sp500_hits_ranked.csv  |  munger_sp500_hits.csv")

# -----------------------------
# 7) Plots (Adj Close vs 200W SMA) for first few ranked names
# -----------------------------
def plot_ticker_200w_from_prices(df_prices: pd.DataFrame, ticker: str):
    w = df_prices[df_prices["Ticker"] == ticker].sort_values("Date").copy()
    if w.empty:
        return
    w.set_index("Date", inplace=True)
    ax = w["Adj Close"].plot(figsize=(9,4), label="Adj Close")
    w["SMA_200W"].plot(ax=ax, label="SMA 200W")
    plt.title(f"{ticker} — Adj Close vs 200W SMA")
    plt.legend(); plt.show()

print("\nPlotting first few ranked names...")
for t in ranked["Ticker"].tolist()[:PLOT_MAX]:
    plot_ticker_200w_from_prices(prices_w, t)
```

```python
# orb_15min_retest_yf60d_fixed.py
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import requests
import yfinance as yf
from datetime import timedelta
from typing import Optional, List, Tuple

# =============================
# USER PARAMETERS
# =============================
# Data granularity
INTERVAL = "15m"         # true ORB timeframe; valid: "1m","2m","5m","15m","30m"
PERIOD = "60d"           # Yahoo limit for intraday < 60m
USE_PERIOD = True        # must be True for 1–30m to avoid YFPricesMissingError

# If you switch to 60m later, you can use start/end again:
START_DATE = "2025-07-01"   # unused when USE_PERIOD=True and INTERVAL < 60m
END_DATE   = "2025-09-03"   # unused when USE_PERIOD=True and INTERVAL < 60m

# Universe / session settings
DEFAULT_TICKERS = ["SPY"]
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic
OR_WINDOW_MIN = 15              # opening range = first 15-min bar (already native when INTERVAL="15m")
RETEST_CONFIRM_CLOSE = True     # retest bar must close back in breakout direction
MAX_RETEST_MIN = 120            # give up if no retest within X minutes

# Risk / exits
R_MULTIPLES = [1.0, 2.0]        # partials at 1R and 2R
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0              # 1bp per fill
FEES_PER_TRADE = 0.00

# Plotting (optional)
PLOT_MAX_DAYS_PER_TICKER = 3    # set to 0 to disable plots

# =============================
# S&P 500 scraping (your snippet)
# =============================
WIKI_URL = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

def sp500_from_wikipedia() -> pd.DataFrame:
    html = requests.get(WIKI_URL, headers={"User-Agent":"Mozilla/5.0"}, timeout=30).text
    table = pd.read_html(html)[0]  # first table = constituents
    df = table.rename(columns={"Symbol":"SymbolRaw"}).copy()
    df["Symbol"] = df["SymbolRaw"].astype(str).str.strip()
    # Wikipedia uses "." (e.g., BRK.B); yfinance expects "-"
    df["Ticker"] = df["Symbol"].str.replace(r"\.", "-", regex=True)
    keep = ["Ticker","Symbol","Security","GICS Sector","GICS Sub-Industry","Headquarters Location"]
    return df[keep]

# =============================
# Data helpers
# =============================
def fetch_intraday(
    ticker: str,
    start: str,
    end: str,
    interval: str = "15m",
    tz: str = "America/New_York",
    period: str = "60d",
    use_period: bool = True
) -> pd.DataFrame:
    """
    Yahoo/yfinance intraday limits:
      - 1m/2m/5m/15m/30m: ~60 days max => must use 'period'
      - 60m: ~2 years allowed => start/end OK
    This function auto-selects the safe path, flattens columns, and fixes tz handling.
    """
    intraday_small = interval in {"1m","2m","5m","15m","30m"}

    try:
        if intraday_small and use_period:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column"
            )
        else:
            df = yf.download(
                ticker, start=start, end=end, interval=interval,
                auto_adjust=False, progress=False, group_by="column"
            )
    except Exception as e:
        print(f"[{ticker}] download error: {e}")
        return pd.DataFrame()

    if df is None or df.empty:
        print(f"[{ticker}] no data returned (interval={interval}, period={period}, start={start}, end={end})")
        return pd.DataFrame()

    # --- FLATTEN COLUMNS IF NEEDED ---
    if isinstance(df.columns, pd.MultiIndex):
        # Use top-level names only: ("Open","SPY") -> "Open"
        df.columns = [c[0] if isinstance(c, tuple) and len(c) > 0 else c for c in df.columns]

    # --- TZ handling ---
    if df.index.tz is None:
        df = df.tz_localize("UTC")
    df = df.tz_convert(tz)

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    # Clean, add ticker
    df = df.drop(columns=[c for c in df.columns if str(c).lower().startswith("adj")], errors="ignore")
    df["Ticker"] = ticker
    df = df.sort_index()
    return df

def resample_to_15m(df: pd.DataFrame) -> pd.DataFrame:
    """Not used if INTERVAL already 15m; included for completeness."""
    if df.empty:
        return df
    o = df["Open"].resample("15T").first()
    h = df["High"].resample("15T").max()
    l = df["Low"].resample("15T").min()
    c = df["Close"].resample("15T").last()
    v = df["Volume"].resample("15T").sum()
    out = pd.concat({"Open":o,"High":h,"Low":l,"Close":c,"Volume":v}, axis=1).dropna(how="any")
    out["Ticker"] = df["Ticker"].iloc[0]
    return out

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    """One session per calendar trading day in local time."""
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB logic
# =============================
def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    """
    Mark ORH/ORL from the first 15-min candle of each session (single bar in 15m data).
    """
    df = df_15.copy()

    # ensure ORH/ORL columns exist (float)
    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan

    df["Session"] = sessionize(df)

    # first 15m bar per day
    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)

    # forward-fill ORH/ORL across the session
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    """Worsen the fill by bps: side='buy' or 'sell'."""
    factor = 1 + (bps/10000.0)
    if side == "buy":
        return price * factor
    return price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = 120,
    r_targets: List[float] = [1.0, 2.0],
    dollars: float = 10_000,
    slippage_bps: float = 1.0,
    fees: float = 0.0,
    retest_confirm_close: bool = True
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    For each session:
      1) Confirm breakout CLOSE beyond ORH/ORL after the first 15m bar.
      2) Wait for retest of breakout level (bar range touches ORH/ORL).
         Optional: require retest bar to CLOSE back in breakout direction.
      3) Stop = opposite side of opening range.
      4) Targets = list of R-multiples.
      5) One trade per session (first valid setup).
    Returns: (trade_log, equity_curve_by_session)
    """
    df = df_15.copy()
    df["Session"] = sessionize(df)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        # OR from the first 15m bar
        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]

        # after the first bar, find first breakout close
        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        breakout_row = None
        direction = None
        if not long_break.empty and not short_break.empty:
            if long_break.index[0] < short_break.index[0]:
                breakout_row = long_break.iloc[0]; direction = "long"
            else:
                breakout_row = short_break.iloc[0]; direction = "short"
        elif not long_break.empty:
            breakout_row = long_break.iloc[0]; direction = "long"
        elif not short_break.empty:
            breakout_row = short_break.iloc[0]; direction = "short"
        else:
            continue  # no breakout that day

        # Retest search window
        btime = breakout_row.name
        after_break = sdf[sdf.index > btime].copy()
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = after_break[after_break.index <= cutoff]
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            valid = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not valid:
                continue

        # Entry at level (slippage applied)
        entry = _apply_slippage(level, slippage_bps, "buy" if direction=="long" else "sell")
        stop = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1

        # Targets (R)
        t_prices = [(entry + r*risk_per_share) if direction=="long" else (entry - r*risk_per_share) for r in r_targets]

        # Walk forward for exits
        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for stamp, row in sdf_run.iterrows():
            high, low = row["High"], row["Low"]
            # check stop first (worst-case intrabar)
            if low <= stop <= high:
                exit_px = _apply_slippage(stop, slippage_bps, "sell" if direction=="long" else "buy")
                exits.append(("stop", stamp, exit_px))
                break

            # check targets (in order)
            hit_any = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    side = "sell" if direction=="long" else "buy"
                    exits.append((f"tp{i+1}", stamp, _apply_slippage(tp, slippage_bps, side)))
                    hit_any = True
            if hit_any:
                break

        if not exits:
            # flat at session last close (conservative)
            last_row = sdf_run.iloc[-1]
            flat_px = _apply_slippage(last_row["Close"], slippage_bps, "sell" if direction=="long" else "buy")
            exits.append(("eod", sdf_run.index[-1], flat_px))

        # P&L with equal partials per exit
        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = 0.0
        for tag, tstamp, px in exits:
            cash_pnl += (px - entry) * side_mult * qty_each
        cash_pnl -= FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh,
            "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": entry,
            "Stop": stop,
            "Targets": t_prices,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Plotting (optional)
# =============================
def plot_example_days(df_15: pd.DataFrame, trade_log: pd.DataFrame, max_days: int = 3, title_prefix: str = ""):
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        return
    if max_days <= 0 or trade_log.empty:
        return

    sessions = trade_log["Session"].unique()[:max_days]
    for ses in sessions:
        sdf = df_15[df_15.index.date == pd.to_datetime(ses).date()]
        if sdf.empty:
            continue
        tr = trade_log[trade_log["Session"] == ses].iloc[0]

        plt.figure(figsize=(11,5))
        plt.plot(sdf.index, sdf["Close"], label="Close")
        plt.axhline(tr["ORH"], linestyle="--", label="ORH")
        plt.axhline(tr["ORL"], linestyle="--", label="ORL")
        plt.scatter([tr["EntryTime"]], [tr["Entry"]], marker="^" if tr["Direction"]=="long" else "v", s=80, label="Entry")
        for tag, tstamp, px in tr["Exits"]:
            plt.scatter([tstamp], [px], marker="x", s=80, label=f"Exit {tag}")
        plt.title(f"{title_prefix}{tr['Ticker']} {pd.to_datetime(ses).date()} ORB Retest (15m)")
        plt.legend()
        plt.tight_layout()
        plt.show()

# =============================
# Runner
# =============================
def run_for_tickers(tickers: List[str]):
    all_trades, all_equity = [], []
    for tk in tickers:
        print(f"== {tk} ==")
        raw = fetch_intraday(
            tk,
            START_DATE,
            END_DATE,
            interval=INTERVAL,
            tz=TZ,
            period=PERIOD,
            use_period=USE_PERIOD
        )
        if raw.empty:
            print("No data.")
            continue

        # Already 15m bars because INTERVAL="15m"
        df15 = raw.copy()

        # Safety: regular session only (already filtered)
        df15 = df15.between_time(REG_SESSION_START, REG_SESSION_END)

        df15 = compute_opening_range(df15)

        tlog, eq = backtest_orb_retest(
            df15,
            max_retest_min=MAX_RETEST_MIN,
            r_targets=R_MULTIPLES,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE
        )
        if not tlog.empty:
            all_trades.append(tlog.assign(Ticker=tk))
            all_equity.append(eq.assign(Ticker=tk))

        plot_example_days(df15, tlog, PLOT_MAX_DAYS_PER_TICKER, title_prefix="")

    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    if not trades.empty:
        to_save = trades.copy()
        to_save["Targets"] = to_save["Targets"].apply(lambda xs: ";".join([f"{p:.4f}" for p in xs]))
        to_save["Exits"] = to_save["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.4f}" for t in xs]))
        to_save.sort_values(["Ticker","Session"], inplace=True)
        trades_csv = "orb_trades.csv"
        to_save.to_csv(trades_csv, index=False)
        print(f"Saved trade log: {trades_csv}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity_csv = "orb_equity_curve.csv"
        equity.to_csv(equity_csv, index=False)
        print(f"Saved equity curve: {equity_csv}")

    if not trades.empty:
        summary = (trades.groupby("Ticker")["PnL_$"]
                   .agg(["count","sum","mean","std"])
                   .rename(columns={"count":"Trades","sum":"PnL_$","mean":"AvgPnL_$","std":"StdPnL_$"}))
        print("\n=== Summary by Ticker ===")
        print(summary)
    else:
        print("No trades generated with current settings.")

if __name__ == "__main__":
    # Example 1: quick run on SPY (last ~60 days, true 15m)
    run_for_tickers(DEFAULT_TICKERS)

    # Example 2: S&P 500 universe (heavy; consider batching)
    # sp500_df = sp500_from_wikipedia()
    # tickers = sp500_df["Ticker"].dropna().unique().tolist()
    # run_for_tickers(tickers)
```

```python
# orb_15min_retest_sp500.py
import warnings
warnings.filterwarnings("ignore")

import time
from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
import requests
import yfinance as yf

# =============================
# USER PARAMETERS
# =============================
# Data (Yahoo intraday <60m => last ~60d only)
INTERVAL = "15m"
PERIOD = "60d"
USE_PERIOD = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120  # minutes after breakout to wait for retest

# Risk / exits
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Batch / persistence
PLOT_MAX_DAYS_PER_TICKER = 0     # disable plots for bulk run
SLEEP_BETWEEN_TICKERS = 0.5      # polite throttle to Yahoo
AUTOSAVE_EVERY = 25              # write partial CSVs every N tickers

# Output files
TRADES_CSV_ALL = "orb_trades_sp500.csv"
EQUITY_CSV_ALL = "orb_equity_sp500.csv"

# =============================
# S&P 500 scraping
# =============================
WIKI_URL = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

def sp500_from_wikipedia() -> pd.DataFrame:
    html = requests.get(WIKI_URL, headers={"User-Agent":"Mozilla/5.0"}, timeout=30).text
    table = pd.read_html(html)[0]
    df = table.rename(columns={"Symbol":"SymbolRaw"}).copy()
    df["Symbol"] = df["SymbolRaw"].astype(str).str.strip()
    df["Ticker"] = df["Symbol"].str.replace(r"\.", "-", regex=True)  # BRK.B -> BRK-B
    keep = ["Ticker","Symbol","Security","GICS Sector","GICS Sub-Industry","Headquarters Location"]
    return df[keep]

# =============================
# Data helpers
# =============================
def fetch_intraday(
    ticker: str,
    interval: str = "15m",
    tz: str = "America/New_York",
    period: str = "60d",
    use_period: bool = True
) -> pd.DataFrame:
    """
    Safe downloader:
      - 1m/2m/5m/15m/30m -> use 'period' (<= ~60d)
      - returns tz-aware index in tz
      - flattens columns to avoid MultiIndex issues
    """
    intraday_small = interval in {"1m","2m","5m","15m","30m"}
    try:
        if intraday_small and use_period:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
        else:
            # not used here, but kept for completeness
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
    except Exception as e:
        print(f"[{ticker}] download error: {e}")
        return pd.DataFrame()

    if df is None or df.empty:
        print(f"[{ticker}] no data (interval={interval}, period={period})")
        return pd.DataFrame()

    # Flatten columns if needed
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) and len(c) else c for c in df.columns]

    # TZ handling
    if df.index.tz is None:
        df = df.tz_localize("UTC")
    df = df.tz_convert(tz)

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    # Cleanup
    df = df.drop(columns=[c for c in df.columns if str(c).lower().startswith("adj")], errors="ignore")
    df["Ticker"] = ticker
    df = df.sort_index()
    return df

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB logic
# =============================
def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()

    # Ensure columns exist
    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan

    df["Session"] = sessionize(df)

    # First 15m bar per day
    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)

    # Forward-fill within session
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = 120,
    r_targets: List[float] = [1.0, 2.0],
    dollars: float = 10_000,
    slippage_bps: float = 1.0,
    fees: float = 0.0,
    retest_confirm_close: bool = True
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout CLOSE after the opening bar
        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            # whichever happened earlier
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction=="long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        # Retest within window
        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction=="long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction=="long" else "sell")
        stop = orl if direction=="long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction=="long" else -1

        # Targets
        t_prices = [(entry + r*risk_per_share) if direction=="long" else (entry - r*risk_per_share) for r in r_targets]

        # Forward simulate exits
        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = row["High"], row["Low"]
            # Stop first
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction=="long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction=="long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            # Flat on last bar close
            last = sdf_run.iloc[-1]
            px = _apply_slippage(last["Close"], slippage_bps, "sell" if direction=="long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        # P&L with equal partials per exit
        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = 0.0
        for tag, ts, px in exits:
            cash_pnl += (px - entry) * side_mult * qty_each
        cash_pnl -= FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": t_prices,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Optional plotting (kept off by default here)
# =============================
def plot_example_days(df_15: pd.DataFrame, trade_log: pd.DataFrame, max_days: int = 0, title_prefix: str = ""):
    if max_days <= 0 or trade_log.empty:
        return
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        return
    sessions = trade_log["Session"].unique()[:max_days]
    for ses in sessions:
        sdf = df_15[df_15.index.date == pd.to_datetime(ses).date()]
        if sdf.empty:
            continue
        tr = trade_log[trade_log["Session"] == ses].iloc[0]
        plt.figure(figsize=(11,5))
        plt.plot(sdf.index, sdf["Close"], label="Close")
        plt.axhline(tr["ORH"], linestyle="--", label="ORH")
        plt.axhline(tr["ORL"], linestyle="--", label="ORL")
        plt.scatter([tr["EntryTime"]], [tr["Entry"]], marker="^" if tr["Direction"]=="long" else "v", s=80, label="Entry")
        for tag, tstamp, px in tr["Exits"]:
            plt.scatter([tstamp], [px], marker="x", s=80, label=f"Exit {tag}")
        plt.title(f"{title_prefix}{tr['Ticker']} {pd.to_datetime(ses).date()} ORB Retest (15m)")
        plt.legend(); plt.tight_layout(); plt.show()

# =============================
# Runner (S&P 500)
# =============================
def run_sp500():
    sp = sp500_from_wikipedia()
    tickers = sp["Ticker"].dropna().unique().tolist()
    print(f"Tickers loaded: {len(tickers)}")

    all_trades = []
    all_equity = []
    processed = 0

    for tk in tickers:
        print(f"\n== {tk} ==")
        raw = fetch_intraday(tk, interval=INTERVAL, tz=TZ, period=PERIOD, use_period=USE_PERIOD)
        if raw.empty:
            print("No data.")
            time.sleep(SLEEP_BETWEEN_TICKERS)
            continue

        df15 = raw.copy()
        df15 = df15.between_time(REG_SESSION_START, REG_SESSION_END)
        df15 = compute_opening_range(df15)

        tlog, eq = backtest_orb_retest(
            df15,
            max_retest_min=MAX_RETEST_MIN,
            r_targets=R_MULTIPLES,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE
        )

        if not tlog.empty:
            # Stringify lists for CSV
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"] = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy()
            eq_out["Ticker"] = tk
            all_equity.append(eq_out)

        plot_example_days(df15, tlog, PLOT_MAX_DAYS_PER_TICKER, title_prefix="")

        processed += 1
        # Autosave every N tickers to avoid losing progress
        if processed % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

        time.sleep(SLEEP_BETWEEN_TICKERS)

    # Final save
    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    if not trades.empty:
        summary = (trades.groupby("Ticker")["PnL_$"]
                   .agg(Trades="count", PnL_$="sum", AvgPnL_$="mean", StdPnL_$="std")).reset_index()
        print("\n=== Summary sample ===")
        print(summary.head(15).to_string(index=False))
    else:
        print("No trades generated with current settings.")

if __name__ == "__main__":
    run_sp500()
```

```python
# orb_15min_retest_sp500.py
import warnings
warnings.filterwarnings("ignore")

import time
from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
import requests
import yfinance as yf

# =============================
# USER PARAMETERS
# =============================
# Data (Yahoo intraday <60m => last ~60d only)
INTERVAL = "15m"
PERIOD = "60d"
USE_PERIOD = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120  # minutes after breakout to wait for retest

# Risk / exits
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Batch / persistence
PLOT_MAX_DAYS_PER_TICKER = 0     # disable plots for bulk run
SLEEP_BETWEEN_TICKERS = 0.5      # polite throttle to Yahoo
AUTOSAVE_EVERY = 25              # write partial CSVs every N tickers

# Output files
TRADES_CSV_ALL = "orb_trades_sp500.csv"
EQUITY_CSV_ALL = "orb_equity_sp500.csv"

# =============================
# S&P 500 scraping
# =============================
WIKI_URL = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

def sp500_from_wikipedia() -> pd.DataFrame:
    html = requests.get(WIKI_URL, headers={"User-Agent":"Mozilla/5.0"}, timeout=30).text
    table = pd.read_html(html)[0]
    df = table.rename(columns={"Symbol":"SymbolRaw"}).copy()
    df["Symbol"] = df["SymbolRaw"].astype(str).str.strip()
    df["Ticker"] = df["Symbol"].str.replace(r"\.", "-", regex=True)  # BRK.B -> BRK-B
    keep = ["Ticker","Symbol","Security","GICS Sector","GICS Sub-Industry","Headquarters Location"]
    return df[keep]

# =============================
# Data helpers
# =============================
def fetch_intraday(
    ticker: str,
    interval: str = "15m",
    tz: str = "America/New_York",
    period: str = "60d",
    use_period: bool = True
) -> pd.DataFrame:
    """
    Safe downloader:
      - 1m/2m/5m/15m/30m -> use 'period' (<= ~60d)
      - returns tz-aware index in tz
      - flattens columns to avoid MultiIndex issues
    """
    intraday_small = interval in {"1m","2m","5m","15m","30m"}
    try:
        if intraday_small and use_period:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
        else:
            # not used here, but kept for completeness
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
    except Exception as e:
        print(f"[{ticker}] download error: {e}")
        return pd.DataFrame()

    if df is None or df.empty:
        print(f"[{ticker}] no data (interval={interval}, period={period})")
        return pd.DataFrame()

    # Flatten columns if needed
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) and len(c) else c for c in df.columns]

    # TZ handling
    if df.index.tz is None:
        df = df.tz_localize("UTC")
    df = df.tz_convert(tz)

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    # Cleanup
    df = df.drop(columns=[c for c in df.columns if str(c).lower().startswith("adj")], errors="ignore")
    df["Ticker"] = ticker
    df = df.sort_index()
    return df

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB logic
# =============================
def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()

    # Ensure columns exist
    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan

    df["Session"] = sessionize(df)

    # First 15m bar per day
    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)

    # Forward-fill within session
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = 120,
    r_targets: List[float] = [1.0, 2.0],
    dollars: float = 10_000,
    slippage_bps: float = 1.0,
    fees: float = 0.0,
    retest_confirm_close: bool = True
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout CLOSE after the opening bar
        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            # whichever happened earlier
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction=="long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        # Retest within window
        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction=="long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction=="long" else "sell")
        stop = orl if direction=="long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction=="long" else -1

        # Targets
        t_prices = [(entry + r*risk_per_share) if direction=="long" else (entry - r*risk_per_share) for r in r_targets]

        # Forward simulate exits
        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = row["High"], row["Low"]
            # Stop first
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction=="long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction=="long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            # Flat on last bar close
            last = sdf_run.iloc[-1]
            px = _apply_slippage(last["Close"], slippage_bps, "sell" if direction=="long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        # P&L with equal partials per exit
        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = 0.0
        for tag, ts, px in exits:
            cash_pnl += (px - entry) * side_mult * qty_each
        cash_pnl -= FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": t_prices,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Optional plotting (kept off by default here)
# =============================
def plot_example_days(df_15: pd.DataFrame, trade_log: pd.DataFrame, max_days: int = 0, title_prefix: str = ""):
    if max_days <= 0 or trade_log.empty:
        return
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        return
    sessions = trade_log["Session"].unique()[:max_days]
    for ses in sessions:
        sdf = df_15[df_15.index.date == pd.to_datetime(ses).date()]
        if sdf.empty:
            continue
        tr = trade_log[trade_log["Session"] == ses].iloc[0]
        plt.figure(figsize=(11,5))
        plt.plot(sdf.index, sdf["Close"], label="Close")
        plt.axhline(tr["ORH"], linestyle="--", label="ORH")
        plt.axhline(tr["ORL"], linestyle="--", label="ORL")
        plt.scatter([tr["EntryTime"]], [tr["Entry"]], marker="^" if tr["Direction"]=="long" else "v", s=80, label="Entry")
        for tag, tstamp, px in tr["Exits"]:
            plt.scatter([tstamp], [px], marker="x", s=80, label=f"Exit {tag}")
        plt.title(f"{title_prefix}{tr['Ticker']} {pd.to_datetime(ses).date()} ORB Retest (15m)")
        plt.legend(); plt.tight_layout(); plt.show()

# =============================
# Runner (S&P 500)
# =============================
def run_sp500():
    sp = sp500_from_wikipedia()
    tickers = sp["Ticker"].dropna().unique().tolist()
    print(f"Tickers loaded: {len(tickers)}")

    all_trades = []
    all_equity = []
    processed = 0

    for tk in tickers:
        print(f"\n== {tk} ==")
        raw = fetch_intraday(tk, interval=INTERVAL, tz=TZ, period=PERIOD, use_period=USE_PERIOD)
        if raw.empty:
            print("No data.")
            time.sleep(SLEEP_BETWEEN_TICKERS)
            continue

        df15 = raw.copy()
        df15 = df15.between_time(REG_SESSION_START, REG_SESSION_END)
        df15 = compute_opening_range(df15)

        tlog, eq = backtest_orb_retest(
            df15,
            max_retest_min=MAX_RETEST_MIN,
            r_targets=R_MULTIPLES,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE
        )

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"] = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy()
            eq_out["Ticker"] = tk
            all_equity.append(eq_out)

        plot_example_days(df15, tlog, PLOT_MAX_DAYS_PER_TICKER, title_prefix="")

        processed += 1
        # Autosave every N tickers to avoid losing progress
        if processed % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

        time.sleep(SLEEP_BETWEEN_TICKERS)

    # Final save
    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    # --- Fixed summary (no invalid kwarg names) ---
    if not trades.empty:
        summary = (
            trades.groupby("Ticker")["PnL_$"]
                  .agg(['count', 'sum', 'mean', 'std'])
                  .rename(columns={
                      'count': 'Trades',
                      'sum':   'PnL_sum',
                      'mean':  'AvgPnL',
                      'std':   'StdPnL'
                  })
                  .reset_index()
        )
        print("\n=== Summary by Ticker (first 15) ===")
        print(summary.head(500).to_string(index=False))
        summary.to_csv("orb_summary_sp500.csv", index=False)
    else:
        print("No trades generated with current settings.")

if __name__ == "__main__":
    run_sp500()
```

```python
# orb_15min_retest_sp500.py
import warnings
warnings.filterwarnings("ignore")

import time
from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
import requests
import yfinance as yf

# =============================
# USER PARAMETERS
# =============================
# Data (Yahoo intraday <60m => last ~60d only)
INTERVAL = "15m"
PERIOD = "60d"
USE_PERIOD = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120  # minutes after breakout to wait for retest

# Risk / exits
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Batch / persistence
PLOT_MAX_DAYS_PER_TICKER = 0     # disable plots for bulk run
SLEEP_BETWEEN_TICKERS = 0.5      # polite throttle to Yahoo
AUTOSAVE_EVERY = 25              # write partial CSVs every N tickers

# Output files
TRADES_CSV_ALL = "orb_trades_sp500.csv"
EQUITY_CSV_ALL = "orb_equity_sp500.csv"

# =============================
# S&P 500 scraping
# =============================
WIKI_URL = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

def sp500_from_wikipedia() -> pd.DataFrame:
    html = requests.get(WIKI_URL, headers={"User-Agent":"Mozilla/5.0"}, timeout=30).text
    table = pd.read_html(html)[0]
    df = table.rename(columns={"Symbol":"SymbolRaw"}).copy()
    df["Symbol"] = df["SymbolRaw"].astype(str).str.strip()
    df["Ticker"] = df["Symbol"].str.replace(r"\.", "-", regex=True)  # BRK.B -> BRK-B
    keep = ["Ticker","Symbol","Security","GICS Sector","GICS Sub-Industry","Headquarters Location"]
    return df[keep]

# =============================
# Data helpers
# =============================
def fetch_intraday(
    ticker: str,
    interval: str = "15m",
    tz: str = "America/New_York",
    period: str = "60d",
    use_period: bool = True
) -> pd.DataFrame:
    """
    Safe downloader:
      - 1m/2m/5m/15m/30m -> use 'period' (<= ~60d)
      - returns tz-aware index in tz
      - flattens columns to avoid MultiIndex issues
    """
    intraday_small = interval in {"1m","2m","5m","15m","30m"}
    try:
        if intraday_small and use_period:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
        else:
            # not used here, but kept for completeness
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
    except Exception as e:
        print(f"[{ticker}] download error: {e}")
        return pd.DataFrame()

    if df is None or df.empty:
        print(f"[{ticker}] no data (interval={interval}, period={period})")
        return pd.DataFrame()

    # Flatten columns if needed
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) and len(c) else c for c in df.columns]

    # TZ handling
    if df.index.tz is None:
        df = df.tz_localize("UTC")
    df = df.tz_convert(tz)

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    # Cleanup
    df = df.drop(columns=[c for c in df.columns if str(c).lower().startswith("adj")], errors="ignore")
    df["Ticker"] = ticker
    df = df.sort_index()
    return df

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB logic
# =============================
def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()

    # Ensure columns exist
    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan

    df["Session"] = sessionize(df)

    # First 15m bar per day
    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)

    # Forward-fill within session
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = 120,
    r_targets: List[float] = [1.0, 2.0],
    dollars: float = 10_000,
    slippage_bps: float = 1.0,
    fees: float = 0.0,
    retest_confirm_close: bool = True
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        # First breakout CLOSE after the opening bar
        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            # whichever happened earlier
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction=="long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        # Retest within window
        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction=="long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction=="long" else "sell")
        stop = orl if direction=="long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction=="long" else -1

        # Targets
        t_prices = [(entry + r*risk_per_share) if direction=="long" else (entry - r*risk_per_share) for r in r_targets]

        # Forward simulate exits
        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = row["High"], row["Low"]
            # Stop first
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction=="long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction=="long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            # Flat on last bar close
            last = sdf_run.iloc[-1]
            px = _apply_slippage(last["Close"], slippage_bps, "sell" if direction=="long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        # P&L with equal partials per exit
        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = 0.0
        for tag, ts, px in exits:
            cash_pnl += (px - entry) * side_mult * qty_each
        cash_pnl -= FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": t_prices,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Optional plotting (kept off by default here)
# =============================
def plot_example_days(df_15: pd.DataFrame, trade_log: pd.DataFrame, max_days: int = 0, title_prefix: str = ""):
    if max_days <= 0 or trade_log.empty:
        return
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        return
    sessions = trade_log["Session"].unique()[:max_days]
    for ses in sessions:
        sdf = df_15[df_15.index.date == pd.to_datetime(ses).date()]
        if sdf.empty:
            continue
        tr = trade_log[trade_log["Session"] == ses].iloc[0]
        plt.figure(figsize=(11,5))
        plt.plot(sdf.index, sdf["Close"], label="Close")
        plt.axhline(tr["ORH"], linestyle="--", label="ORH")
        plt.axhline(tr["ORL"], linestyle="--", label="ORL")
        plt.scatter([tr["EntryTime"]], [tr["Entry"]], marker="^" if tr["Direction"]=="long" else "v", s=80, label="Entry")
        for tag, tstamp, px in tr["Exits"]:
            plt.scatter([tstamp], [px], marker="x", s=80, label=f"Exit {tag}")
        plt.title(f"{title_prefix}{tr['Ticker']} {pd.to_datetime(ses).date()} ORB Retest (15m)")
        plt.legend(); plt.tight_layout(); plt.show()

# =============================
# Runner (S&P 500)
# =============================
def run_sp500():
    sp = sp500_from_wikipedia()
    tickers = sp["Ticker"].dropna().unique().tolist()
    print(f"Tickers loaded: {len(tickers)}")

    all_trades = []
    all_equity = []
    processed = 0

    for tk in tickers:
        print(f"\n== {tk} ==")
        raw = fetch_intraday(tk, interval=INTERVAL, tz=TZ, period=PERIOD, use_period=USE_PERIOD)
        if raw.empty:
            print("No data.")
            time.sleep(SLEEP_BETWEEN_TICKERS)
            continue

        df15 = raw.copy()
        df15 = df15.between_time(REG_SESSION_START, REG_SESSION_END)
        df15 = compute_opening_range(df15)

        tlog, eq = backtest_orb_retest(
            df15,
            max_retest_min=MAX_RETEST_MIN,
            r_targets=R_MULTIPLES,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE
        )

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"] = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy()
            eq_out["Ticker"] = tk
            all_equity.append(eq_out)

        plot_example_days(df15, tlog, PLOT_MAX_DAYS_PER_TICKER, title_prefix="")

        processed += 1
        # Autosave every N tickers to avoid losing progress
        if processed % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

        time.sleep(SLEEP_BETWEEN_TICKERS)

    # Final save
    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    # --- Summary & totals (adds Total PnL and Win %) ---
    if not trades.empty:
        # Per-ticker summary (includes win rate)
        per_ticker = (
            trades.assign(Win=trades["PnL_$"] > 0)
                  .groupby("Ticker")
                  .agg(
                      Trades=("PnL_$", "count"),
                      PnL_sum=("PnL_$", "sum"),
                      AvgPnL=("PnL_$", "mean"),
                      StdPnL=("PnL_$", "std"),
                      WinRate=("Win", "mean"),
                      AvgR=("R_multiple", "mean"),
                      MedianR=("R_multiple", "median")
                  )
                  .reset_index()
        )
        per_ticker["WinRate"] = (per_ticker["WinRate"] * 100).round(2)  # %

        # Save per-ticker table
        per_ticker.to_csv("orb_summary_sp500.csv", index=False)

        # Overall totals
        total_trades = len(trades)
        total_pnl = trades["PnL_$"].sum()
        win_rate = (trades["PnL_$"] > 0).mean() * 100.0
        avg_r = trades["R_multiple"].mean()
        median_r = trades["R_multiple"].median()

        print("\n=== Overall Totals ===")
        print(f"Total trades: {total_trades}")
        print(f"Total PnL ($): {total_pnl:,.2f}")
        print(f"Win rate: {win_rate:.2f}%")
        print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")

        print("\n=== Per-Ticker Summary (first 15) ===")
        print(per_ticker.head(15).to_string(index=False))
    else:
        print("No trades generated with current settings.")

if __name__ == "__main__":
    run_sp500()
```

```python
# orb_15min_retest_sp500.py
import warnings
warnings.filterwarnings("ignore")

import time
from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
import requests
import yfinance as yf

# =============================
# USER PARAMETERS
# =============================
# Data (Yahoo intraday <60m => last ~60d only)
INTERVAL = "15m"
PERIOD = "60d"
USE_PERIOD = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120  # minutes after breakout to wait for retest

# Risk / exits
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Batch / persistence
PLOT_MAX_DAYS_PER_TICKER = 0     # disable plots for bulk run
SLEEP_BETWEEN_TICKERS = 0.5      # polite throttle to Yahoo
AUTOSAVE_EVERY = 25              # write partial CSVs every N tickers

# Output files
TRADES_CSV_ALL = "orb_trades_sp500.csv"
EQUITY_CSV_ALL = "orb_equity_sp500.csv"

# =============================
# S&P 500 scraping
# =============================
WIKI_URL = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

def sp500_from_wikipedia() -> pd.DataFrame:
    html = requests.get(WIKI_URL, headers={"User-Agent":"Mozilla/5.0"}, timeout=30).text
    table = pd.read_html(html)[0]
    df = table.rename(columns={"Symbol":"SymbolRaw"}).copy()
    df["Symbol"] = df["SymbolRaw"].astype(str).str.strip()
    df["Ticker"] = df["Symbol"].str.replace(r"\.", "-", regex=True)  # BRK.B -> BRK-B
    keep = ["Ticker","Symbol","Security","GICS Sector","GICS Sub-Industry","Headquarters Location"]
    return df[keep]

# =============================
# Data helpers
# =============================
def fetch_intraday(
    ticker: str,
    interval: str = "15m",
    tz: str = "America/New_York",
    period: str = "60d",
    use_period: bool = True
) -> pd.DataFrame:
    """
    Safe downloader:
      - 1m/2m/5m/15m/30m -> use 'period' (<= ~60d)
      - returns tz-aware index in tz
      - flattens columns to avoid MultiIndex issues
    """
    intraday_small = interval in {"1m","2m","5m","15m","30m"}
    try:
        if intraday_small and use_period:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
        else:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
    except Exception as e:
        print(f"[{ticker}] download error: {e}")
        return pd.DataFrame()

    if df is None or df.empty:
        print(f"[{ticker}] no data (interval={interval}, period={period})")
        return pd.DataFrame()

    # Flatten columns if needed
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) and len(c) else c for c in df.columns]

    # TZ handling
    if df.index.tz is None:
        df = df.tz_localize("UTC")
    df = df.tz_convert(tz)

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    # Cleanup
    df = df.drop(columns=[c for c in df.columns if str(c).lower().startswith("adj")], errors="ignore")
    df["Ticker"] = ticker
    df = df.sort_index()
    return df

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB logic
# =============================
def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()

    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan

    df["Session"] = sessionize(df)

    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)

    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = 120,
    r_targets: List[float] = [1.0, 2.0],
    dollars: float = 10_000,
    slippage_bps: float = 1.0,
    fees: float = 0.0,
    retest_confirm_close: bool = True
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction=="long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction=="long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction=="long" else "sell")
        stop = orl if direction=="long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction=="long" else -1

        t_prices = [(entry + r*risk_per_share) if direction=="long" else (entry - r*risk_per_share) for r in r_targets]

        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = row["High"], row["Low"]
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction=="long" else "buy")
                exits.append(("stop", ts, px))
                break
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction=="long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(last["Close"], slippage_bps, "sell" if direction=="long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = 0.0
        for tag, ts, px in exits:
            cash_pnl += (px - entry) * side_mult * qty_each
        cash_pnl -= FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": t_prices,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Optional plotting (kept off by default here)
# =============================
def plot_example_days(df_15: pd.DataFrame, trade_log: pd.DataFrame, max_days: int = 0, title_prefix: str = ""):
    if max_days <= 0 or trade_log.empty:
        return
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        return
    sessions = trade_log["Session"].unique()[:max_days]
    for ses in sessions:
        sdf = df_15[df_15.index.date == pd.to_datetime(ses).date()]
        if sdf.empty:
            continue
        tr = trade_log[trade_log["Session"] == ses].iloc[0]
        plt.figure(figsize=(11,5))
        plt.plot(sdf.index, sdf["Close"], label="Close")
        plt.axhline(tr["ORH"], linestyle="--", label="ORH")
        plt.axhline(tr["ORL"], linestyle="--", label="ORL")
        plt.scatter([tr["EntryTime"]], [tr["Entry"]], marker="^" if tr["Direction"]=="long" else "v", s=80, label="Entry")
        for tag, tstamp, px in tr["Exits"]:
            plt.scatter([tstamp], [px], marker="x", s=80, label=f"Exit {tag}")
        plt.title(f"{title_prefix}{tr['Ticker']} {pd.to_datetime(ses).date()} ORB Retest (15m)")
        plt.legend(); plt.tight_layout(); plt.show()

# =============================
# Runner (S&P 500)
# =============================
def run_sp500():
    sp = sp500_from_wikipedia()
    tickers = sp["Ticker"].dropna().unique().tolist()
    print(f"Tickers loaded: {len(tickers)}")

    all_trades = []
    all_equity = []
    processed = 0

    for tk in tickers:
        print(f"\n== {tk} ==")
        raw = fetch_intraday(tk, interval=INTERVAL, tz=TZ, period=PERIOD, use_period=USE_PERIOD)
        if raw.empty:
            print("No data.")
            time.sleep(SLEEP_BETWEEN_TICKERS)
            continue

        df15 = raw.copy()
        df15 = df15.between_time(REG_SESSION_START, REG_SESSION_END)
        df15 = compute_opening_range(df15)

        tlog, eq = backtest_orb_retest(
            df15,
            max_retest_min=MAX_RETEST_MIN,
            r_targets=R_MULTIPLES,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE
        )

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"] = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy()
            eq_out["Ticker"] = tk
            all_equity.append(eq_out)

        # plotting disabled by default for speed
        processed += 1
        if processed % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")
        time.sleep(SLEEP_BETWEEN_TICKERS)

    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    # ========= NEW: full per-ticker table + AvgWin/AvgLoss, and overall PF =========
    if not trades.empty:
        g = trades.groupby("Ticker")

        # Per-ticker metrics
        per_ticker = pd.DataFrame({
            "Trades":   g["PnL_$"].count(),
            "PnL_sum":  g["PnL_$"].sum(),
            "AvgPnL":   g["PnL_$"].mean(),
            "StdPnL":   g["PnL_$"].std(),
            "Wins":     g["PnL_$"].apply(lambda s: (s > 0).sum()),
            "Losses":   g["PnL_$"].apply(lambda s: (s < 0).sum()),
            "WinRate":  g["PnL_$"].apply(lambda s: (s > 0).mean()),
            "AvgWin":   g["PnL_$"].apply(lambda s: s[s > 0].mean() if (s > 0).any() else np.nan),
            "AvgLoss":  g["PnL_$"].apply(lambda s: s[s < 0].mean() if (s < 0).any() else np.nan),
            "AvgR":     g["R_multiple"].mean(),
            "MedianR":  g["R_multiple"].median(),
            "PF":       g["PnL_$"].apply(lambda s: (s[s > 0].sum() / abs(s[s < 0].sum()))
                                         if abs(s[s < 0].sum()) > 0 else np.nan)
        }).reset_index()

        per_ticker["WinRate"] = (per_ticker["WinRate"] * 100).round(2)
        # Optional: show loss magnitude as positive number alongside AvgLoss
        per_ticker["LossMag"] = per_ticker["AvgLoss"].abs()

        # Save the full table
        per_ticker.to_csv("orb_summary_sp500.csv", index=False)

        # Overall totals
        total_trades = len(trades)
        total_pnl    = trades["PnL_$"].sum()
        win_rate     = (trades["PnL_$"] > 0).mean() * 100.0
        avg_r        = trades["R_multiple"].mean()
        median_r     = trades["R_multiple"].median()
        avg_win_all  = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
        avg_loss_all = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()  # negative
        pf_all = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())

        print("\n=== Overall Totals ===")
        print(f"Total trades: {total_trades}")
        print(f"Total PnL ($): {total_pnl:,.2f}")
        print(f"Win rate: {win_rate:.2f}%")
        print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
        print(f"Avg Win ($): {0.0 if pd.isna(avg_win_all) else avg_win_all:,.2f}")
        print(f"Avg Loss ($): {0.0 if pd.isna(avg_loss_all) else avg_loss_all:,.2f}")
        print(f"Profit Factor: {pf_all:.3f}")

        # Show ALL ~500 tickers in the console (sorted by ticker)
        print("\n=== Per-Ticker Summary (ALL) ===")
        print(per_ticker.sort_values("Ticker")
                        .to_string(index=False,
                                   float_format=lambda x: f"{x:,.4f}"))
    else:
        print("No trades generated with current settings.")

if __name__ == "__main__":
    run_sp500()
```

```python
# orb_15min_retest_sp500.py
# Runs 15-minute ORB (retest) across all S&P 500 tickers for the last ~60 days (Yahoo limit).
# Saves CSV + Excel + XML outputs and prints a full per-ticker summary (all ~500 rows).

import warnings
warnings.filterwarnings("ignore")

import time
from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
import requests
import yfinance as yf

# =============================
# USER PARAMETERS
# =============================
# Data (Yahoo intraday <60m => last ~60d only)
INTERVAL = "15m"
PERIOD = "60d"
USE_PERIOD = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120  # minutes after breakout to wait for retest

# Risk / exits
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Batch / persistence
PLOT_MAX_DAYS_PER_TICKER = 0     # disable plots for bulk run
SLEEP_BETWEEN_TICKERS = 0.5      # polite throttle to Yahoo
AUTOSAVE_EVERY = 25              # write partial CSVs every N tickers

# Output files
TRADES_CSV_ALL = "orb_trades_sp500.csv"
EQUITY_CSV_ALL = "orb_equity_sp500.csv"
SUMMARY_CSV    = "orb_summary_sp500.csv"

# NEW: Excel & XML outputs
EXCEL_OUT   = "orb_results_sp500.xlsx"   # workbook with Summary, Trades, Equity
XML_TRADES  = "orb_trades_sp500.xml"
XML_EQUITY  = "orb_equity_sp500.xml"
XML_SUMMARY = "orb_summary_sp500.xml"

# =============================
# S&P 500 scraping
# =============================
WIKI_URL = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

def sp500_from_wikipedia() -> pd.DataFrame:
    html = requests.get(WIKI_URL, headers={"User-Agent":"Mozilla/5.0"}, timeout=30).text
    table = pd.read_html(html)[0]
    df = table.rename(columns={"Symbol":"SymbolRaw"}).copy()
    df["Symbol"] = df["SymbolRaw"].astype(str).str.strip()
    df["Ticker"] = df["Symbol"].str.replace(r"\.", "-", regex=True)  # BRK.B -> BRK-B
    keep = ["Ticker","Symbol","Security","GICS Sector","GICS Sub-Industry","Headquarters Location"]
    return df[keep]

# =============================
# Data helpers
# =============================
def fetch_intraday(
    ticker: str,
    interval: str = "15m",
    tz: str = "America/New_York",
    period: str = "60d",
    use_period: bool = True
) -> pd.DataFrame:
    """
    Safe downloader:
      - 1m/2m/5m/15m/30m -> use 'period' (<= ~60d)
      - returns tz-aware index in tz
      - flattens columns to avoid MultiIndex issues
    """
    intraday_small = interval in {"1m","2m","5m","15m","30m"}
    try:
        if intraday_small and use_period:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
        else:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
    except Exception as e:
        print(f"[{ticker}] download error: {e}")
        return pd.DataFrame()

    if df is None or df.empty:
        print(f"[{ticker}] no data (interval={interval}, period={period})")
        return pd.DataFrame()

    # Flatten columns if needed
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) and len(c) else c for c in df.columns]

    # TZ handling
    if df.index.tz is None:
        df = df.tz_localize("UTC")
    df = df.tz_convert(tz)

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    # Cleanup
    df = df.drop(columns=[c for c in df.columns if str(c).lower().startswith("adj")], errors="ignore")
    df["Ticker"] = ticker
    df = df.sort_index()
    return df

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB logic
# =============================
def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()

    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan

    df["Session"] = sessionize(df)

    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)

    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = 120,
    r_targets: List[float] = [1.0, 2.0],
    dollars: float = 10_000,
    slippage_bps: float = 1.0,
    fees: float = 0.0,
    retest_confirm_close: bool = True
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction=="long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction=="long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction=="long" else "sell")
        stop = orl if direction=="long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction=="long" else -1

        t_prices = [(entry + r*risk_per_share) if direction=="long" else (entry - r*risk_per_share) for r in r_targets]

        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = row["High"], row["Low"]
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction=="long" else "buy")
                exits.append(("stop", ts, px))
                break
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction=="long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(last["Close"], slippage_bps, "sell" if direction=="long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = 0.0
        for tag, ts, px in exits:
            cash_pnl += (px - entry) * side_mult * qty_each
        cash_pnl -= FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": t_prices,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Optional plotting (off by default)
# =============================
def plot_example_days(df_15: pd.DataFrame, trade_log: pd.DataFrame, max_days: int = 0, title_prefix: str = ""):
    if max_days <= 0 or trade_log.empty:
        return
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        return
    sessions = trade_log["Session"].unique()[:max_days]
    for ses in sessions:
        sdf = df_15[df_15.index.date == pd.to_datetime(ses).date()]
        if sdf.empty:
            continue
        tr = trade_log[trade_log["Session"] == ses].iloc[0]
        plt.figure(figsize=(11,5))
        plt.plot(sdf.index, sdf["Close"], label="Close")
        plt.axhline(tr["ORH"], linestyle="--", label="ORH")
        plt.axhline(tr["ORL"], linestyle="--", label="ORL")
        plt.scatter([tr["EntryTime"]], [tr["Entry"]], marker="^" if tr["Direction"]=="long" else "v", s=80, label="Entry")
        for tag, tstamp, px in tr["Exits"]:
            plt.scatter([tstamp], [px], marker="x", s=80, label=f"Exit {tag}")
        plt.title(f"{title_prefix}{tr['Ticker']} {pd.to_datetime(ses).date()} ORB Retest (15m)")
        plt.legend(); plt.tight_layout(); plt.show()

# =============================
# XML writer helper
# =============================
def df_to_xml(df: pd.DataFrame, root_tag: str, row_tag: str, file_path: str) -> None:
    """Write a DataFrame to a simple XML file."""
    import xml.etree.ElementTree as ET

    def _to_text(v):
        if pd.isna(v):
            return ""
        if isinstance(v, pd.Timestamp):
            return v.isoformat()
        return str(v)

    root = ET.Element(root_tag)
    for _, row in df.iterrows():
        item = ET.SubElement(root, row_tag)
        for col, val in row.items():
            tag = str(col).replace(" ", "_")
            ET.SubElement(item, tag).text = _to_text(val)

    tree = ET.ElementTree(root)
    try:
        ET.indent(tree, space="  ", level=0)  # pretty print (py3.9+)
    except Exception:
        pass
    tree.write(file_path, encoding="utf-8", xml_declaration=True)

# =============================
# Runner (S&P 500)
# =============================
def run_sp500():
    sp = sp500_from_wikipedia()
    tickers = sp["Ticker"].dropna().unique().tolist()
    print(f"Tickers loaded: {len(tickers)}")

    all_trades = []
    all_equity = []
    processed = 0

    for tk in tickers:
        print(f"\n== {tk} ==")
        raw = fetch_intraday(tk, interval=INTERVAL, tz=TZ, period=PERIOD, use_period=USE_PERIOD)
        if raw.empty:
            print("No data.")
            time.sleep(SLEEP_BETWEEN_TICKERS)
            continue

        df15 = raw.copy()
        df15 = df15.between_time(REG_SESSION_START, REG_SESSION_END)
        df15 = compute_opening_range(df15)

        tlog, eq = backtest_orb_retest(
            df15,
            max_retest_min=MAX_RETEST_MIN,
            r_targets=R_MULTIPLES,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE
        )

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"] = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy()
            eq_out["Ticker"] = tk
            all_equity.append(eq_out)

        # autosave every N tickers
        processed += 1
        if processed % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")
        time.sleep(SLEEP_BETWEEN_TICKERS)

    # Final tables
    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    # ========= Summary (prints all tickers) + Excel & XML exports =========
    if not trades.empty:
        g = trades.groupby("Ticker")

        # Per-ticker metrics (incl. AvgWin/AvgLoss, PF)
        per_ticker = pd.DataFrame({
            "Trades":   g["PnL_$"].count(),
            "PnL_sum":  g["PnL_$"].sum(),
            "AvgPnL":   g["PnL_$"].mean(),
            "StdPnL":   g["PnL_$"].std(),
            "Wins":     g["PnL_$"].apply(lambda s: (s > 0).sum()),
            "Losses":   g["PnL_$"].apply(lambda s: (s < 0).sum()),
            "WinRate":  g["PnL_$"].apply(lambda s: (s > 0).mean()),
            "AvgWin":   g["PnL_$"].apply(lambda s: s[s > 0].mean() if (s > 0).any() else np.nan),
            "AvgLoss":  g["PnL_$"].apply(lambda s: s[s < 0].mean() if (s < 0).any() else np.nan),
            "AvgR":     g["R_multiple"].mean(),
            "MedianR":  g["R_multiple"].median(),
            "PF":       g["PnL_$"].apply(lambda s: (s[s > 0].sum() / abs(s[s < 0].sum()))
                                         if abs(s[s < 0].sum()) > 0 else np.nan)
        }).reset_index()
        per_ticker["WinRate"] = (per_ticker["WinRate"] * 100).round(2)
        per_ticker_sorted = per_ticker.sort_values("Ticker")

        # Save the per-ticker CSV
        per_ticker_sorted.to_csv(SUMMARY_CSV, index=False)
        print(f"Saved summary: {SUMMARY_CSV}")

        # Overall totals
        total_trades = len(trades)
        total_pnl    = trades["PnL_$"].sum()
        win_rate     = (trades["PnL_$"] > 0).mean() * 100.0
        avg_r        = trades["R_multiple"].mean()
        median_r     = trades["R_multiple"].median()
        avg_win_all  = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
        avg_loss_all = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()  # negative
        pf_all = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())

        print("\n=== Overall Totals ===")
        print(f"Total trades: {total_trades}")
        print(f"Total PnL ($): {total_pnl:,.2f}")
        print(f"Win rate: {win_rate:.2f}%")
        print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
        print(f"Avg Win ($): {0.0 if pd.isna(avg_win_all) else avg_win_all:,.2f}")
        print(f"Avg Loss ($): {0.0 if pd.isna(avg_loss_all) else avg_loss_all:,.2f}")
        print(f"Profit Factor: {pf_all:.3f}")

        # Show ALL ~500 tickers in console
        print("\n=== Per-Ticker Summary (ALL) ===")
        print(per_ticker_sorted.to_string(index=False, float_format=lambda x: f"{x:,.4f}"))

        # Excel workbook with 3 sheets
        # (engine=openpyxl; install via: pip install openpyxl)
        with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
            per_ticker_sorted.to_excel(writer, sheet_name="Summary", index=False)
            trades.to_excel(writer,      sheet_name="Trades",  index=False)
            equity.to_excel(writer,      sheet_name="Equity",  index=False)
        print(f"Wrote Excel workbook: {EXCEL_OUT}")

        # XML exports
        df_to_xml(per_ticker_sorted, "PerTickerSummary", "TickerRow", XML_SUMMARY)
        df_to_xml(trades,            "Trades",           "Trade",     XML_TRADES)
        df_to_xml(equity,            "EquityCurve",      "Row",       XML_EQUITY)
        print(f"Wrote XML files: {XML_SUMMARY}, {XML_TRADES}, {XML_EQUITY}")

    else:
        print("No trades generated with current settings.")

if __name__ == "__main__":
    run_sp500()
```

```python
# orb_15min_retest_sp500.py
# Runs 15-minute ORB (retest) across all S&P 500 tickers for the last ~60 days (Yahoo limit).
# Saves CSV + Excel + XML outputs and prints a full per-ticker summary (all ~500 rows).

import warnings
warnings.filterwarnings("ignore")

import time
from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd
import requests
import yfinance as yf
from pandas.api.types import is_datetime64tz_dtype

# =============================
# USER PARAMETERS
# =============================
# Data (Yahoo intraday <60m => last ~60d only)
INTERVAL = "15m"
PERIOD = "60d"
USE_PERIOD = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120  # minutes after breakout to wait for retest

# Risk / exits
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Batch / persistence
PLOT_MAX_DAYS_PER_TICKER = 0     # disable plots for bulk run
SLEEP_BETWEEN_TICKERS = 0.5      # polite throttle to Yahoo
AUTOSAVE_EVERY = 25              # write partial CSVs every N tickers

# Output files
TRADES_CSV_ALL = "orb_trades_sp500.csv"
EQUITY_CSV_ALL = "orb_equity_sp500.csv"
SUMMARY_CSV    = "orb_summary_sp500.csv"

# Excel & XML outputs
EXCEL_OUT   = "orb_results_sp500.xlsx"   # workbook with Summary, Trades, Equity
XML_TRADES  = "orb_trades_sp500.xml"
XML_EQUITY  = "orb_equity_sp500.xml"
XML_SUMMARY = "orb_summary_sp500.xml"

# =============================
# S&P 500 scraping
# =============================
WIKI_URL = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

def sp500_from_wikipedia() -> pd.DataFrame:
    html = requests.get(WIKI_URL, headers={"User-Agent":"Mozilla/5.0"}, timeout=30).text
    table = pd.read_html(html)[0]
    df = table.rename(columns={"Symbol":"SymbolRaw"}).copy()
    df["Symbol"] = df["SymbolRaw"].astype(str).str.strip()
    df["Ticker"] = df["Symbol"].str.replace(r"\.", "-", regex=True)  # BRK.B -> BRK-B
    keep = ["Ticker","Symbol","Security","GICS Sector","GICS Sub-Industry","Headquarters Location"]
    return df[keep]

# =============================
# Data helpers
# =============================
def fetch_intraday(
    ticker: str,
    interval: str = "15m",
    tz: str = "America/New_York",
    period: str = "60d",
    use_period: bool = True
) -> pd.DataFrame:
    """
    Safe downloader:
      - 1m/2m/5m/15m/30m -> use 'period' (<= ~60d)
      - returns tz-aware index in tz
      - flattens columns to avoid MultiIndex issues
    """
    intraday_small = interval in {"1m","2m","5m","15m","30m"}
    try:
        if intraday_small and use_period:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
        else:
            df = yf.download(
                ticker, period=period, interval=interval,
                auto_adjust=False, progress=False, group_by="column", threads=True
            )
    except Exception as e:
        print(f"[{ticker}] download error: {e}")
        return pd.DataFrame()

    if df is None or df.empty:
        print(f"[{ticker}] no data (interval={interval}, period={period})")
        return pd.DataFrame()

    # Flatten columns if needed
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [c[0] if isinstance(c, tuple) and len(c) else c for c in df.columns]

    # TZ handling
    if df.index.tz is None:
        df = df.tz_localize("UTC")
    df = df.tz_convert(tz)

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    # Cleanup
    df = df.drop(columns=[c for c in df.columns if str(c).lower().startswith("adj")], errors="ignore")
    df["Ticker"] = ticker
    df = df.sort_index()
    return df

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB logic
# =============================
def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()

    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan

    df["Session"] = sessionize(df)

    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)

    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = 120,
    r_targets: List[float] = [1.0, 2.0],
    dollars: float = 10_000,
    slippage_bps: float = 1.0,
    fees: float = 0.0,
    retest_confirm_close: bool = True
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction=="long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction=="long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction=="long" else "sell")
        stop = orl if direction=="long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction=="long" else -1

        t_prices = [(entry + r*risk_per_share) if direction=="long" else (entry - r*risk_per_share) for r in r_targets]

        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = row["High"], row["Low"]
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction=="long" else "buy")
                exits.append(("stop", ts, px))
                break
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction=="long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(last["Close"], slippage_bps, "sell" if direction=="long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = 0.0
        for tag, ts, px in exits:
            cash_pnl += (px - entry) * side_mult * qty_each
        cash_pnl -= FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": t_prices,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Optional plotting (off by default)
# =============================
def plot_example_days(df_15: pd.DataFrame, trade_log: pd.DataFrame, max_days: int = 0, title_prefix: str = ""):
    if max_days <= 0 or trade_log.empty:
        return
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        return
    sessions = trade_log["Session"].unique()[:max_days]
    for ses in sessions:
        sdf = df_15[df_15.index.date == pd.to_datetime(ses).date()]
        if sdf.empty:
            continue
        tr = trade_log[trade_log["Session"] == ses].iloc[0]
        plt.figure(figsize=(11,5))
        plt.plot(sdf.index, sdf["Close"], label="Close")
        plt.axhline(tr["ORH"], linestyle="--", label="ORH")
        plt.axhline(tr["ORL"], linestyle="--", label="ORL")
        plt.scatter([tr["EntryTime"]], [tr["Entry"]], marker="^" if tr["Direction"]=="long" else "v", s=80, label="Entry")
        for tag, tstamp, px in tr["Exits"]:
            plt.scatter([tstamp], [px], marker="x", s=80, label=f"Exit {tag}")
        plt.title(f"{title_prefix}{tr['Ticker']} {pd.to_datetime(ses).date()} ORB Retest (15m)")
        plt.legend(); plt.tight_layout(); plt.show()

# =============================
# XML writer helper
# =============================
def df_to_xml(df: pd.DataFrame, root_tag: str, row_tag: str, file_path: str) -> None:
    """Write a DataFrame to a simple XML file."""
    import xml.etree.ElementTree as ET

    def _to_text(v):
        if pd.isna(v):
            return ""
        if isinstance(v, pd.Timestamp):
            return v.isoformat()
        return str(v)

    root = ET.Element(root_tag)
    for _, row in df.iterrows():
        item = ET.SubElement(root, row_tag)
        for col, val in row.items():
            tag = str(col).replace(" ", "_")
            ET.SubElement(item, tag).text = _to_text(val)

    tree = ET.ElementTree(root)
    try:
        ET.indent(tree, space="  ", level=0)  # pretty print (py3.9+)
    except Exception:
        pass
    tree.write(file_path, encoding="utf-8", xml_declaration=True)

# =============================
# Excel safety helper (remove tz-aware datetimes)
# =============================
def make_excel_compatible(df: pd.DataFrame, local_tz: str = TZ) -> pd.DataFrame:
    """Return a copy with tz-aware datetimes converted to naive (in local_tz)."""
    out = df.copy()
    for col in out.columns:
        if is_datetime64tz_dtype(out[col]):
            # convert to desired timezone, then drop tz info
            out[col] = out[col].dt.tz_convert(local_tz).dt.tz_localize(None)
        elif out[col].dtype == "object":
            # if any tz-aware timestamps are hiding in object columns, stringify them
            out[col] = out[col].apply(
                lambda v: v.tz_convert(local_tz).tz_localize(None)
                if isinstance(v, pd.Timestamp) and v.tzinfo is not None
                else v
            )
    return out

# =============================
# Runner (S&P 500)
# =============================
def run_sp500():
    sp = sp500_from_wikipedia()
    tickers = sp["Ticker"].dropna().unique().tolist()
    print(f"Tickers loaded: {len(tickers)}")

    all_trades = []
    all_equity = []
    processed = 0

    for tk in tickers:
        print(f"\n== {tk} ==")
        raw = fetch_intraday(tk, interval=INTERVAL, tz=TZ, period=PERIOD, use_period=USE_PERIOD)
        if raw.empty:
            print("No data.")
            time.sleep(SLEEP_BETWEEN_TICKERS)
            continue

        df15 = raw.copy()
        df15 = df15.between_time(REG_SESSION_START, REG_SESSION_END)
        df15 = compute_opening_range(df15)

        tlog, eq = backtest_orb_retest(
            df15,
            max_retest_min=MAX_RETEST_MIN,
            r_targets=R_MULTIPLES,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE
        )

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"] = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy()
            eq_out["Ticker"] = tk
            all_equity.append(eq_out)

        # autosave every N tickers
        processed += 1
        if processed % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")
        time.sleep(SLEEP_BETWEEN_TICKERS)

    # Final tables
    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    # ========= Summary (prints all tickers) + Excel & XML exports =========
    if not trades.empty:
        g = trades.groupby("Ticker")

        # Per-ticker metrics (incl. AvgWin/AvgLoss, PF)
        per_ticker = pd.DataFrame({
            "Trades":   g["PnL_$"].count(),
            "PnL_sum":  g["PnL_$"].sum(),
            "AvgPnL":   g["PnL_$"].mean(),
            "StdPnL":   g["PnL_$"].std(),
            "Wins":     g["PnL_$"].apply(lambda s: (s > 0).sum()),
            "Losses":   g["PnL_$"].apply(lambda s: (s < 0).sum()),
            "WinRate":  g["PnL_$"].apply(lambda s: (s > 0).mean()),
            "AvgWin":   g["PnL_$"].apply(lambda s: s[s > 0].mean() if (s > 0).any() else np.nan),
            "AvgLoss":  g["PnL_$"].apply(lambda s: s[s < 0].mean() if (s < 0).any() else np.nan),
            "AvgR":     g["R_multiple"].mean(),
            "MedianR":  g["R_multiple"].median(),
            "PF":       g["PnL_$"].apply(lambda s: (s[s > 0].sum() / abs(s[s < 0].sum()))
                                         if abs(s[s < 0].sum()) > 0 else np.nan)
        }).reset_index()
        per_ticker["WinRate"] = (per_ticker["WinRate"] * 100).round(2)
        per_ticker_sorted = per_ticker.sort_values("Ticker")

        # Save the per-ticker CSV
        per_ticker_sorted.to_csv(SUMMARY_CSV, index=False)
        print(f"Saved summary: {SUMMARY_CSV}")

        # Overall totals
        total_trades = len(trades)
        total_pnl    = trades["PnL_$"].sum()
        win_rate     = (trades["PnL_$"] > 0).mean() * 100.0
        avg_r        = trades["R_multiple"].mean()
        median_r     = trades["R_multiple"].median()
        avg_win_all  = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
        avg_loss_all = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()  # negative
        pf_all = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())

        print("\n=== Overall Totals ===")
        print(f"Total trades: {total_trades}")
        print(f"Total PnL ($): {total_pnl:,.2f}")
        print(f"Win rate: {win_rate:.2f}%")
        print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
        print(f"Avg Win ($): {0.0 if pd.isna(avg_win_all) else avg_win_all:,.2f}")
        print(f"Avg Loss ($): {0.0 if pd.isna(avg_loss_all) else avg_loss_all:,.2f}")
        print(f"Profit Factor: {pf_all:.3f}")

        # Show ALL ~500 tickers in console
        print("\n=== Per-Ticker Summary (ALL) ===")
        print(per_ticker_sorted.to_string(index=False, float_format=lambda x: f"{x:,.4f}"))

        # -------- Excel workbook (timezone-safe) --------
        # Convert any tz-aware datetime columns to naive before writing
        trades_x  = make_excel_compatible(trades, TZ)
        equity_x  = make_excel_compatible(equity, TZ)
        summary_x = make_excel_compatible(per_ticker_sorted, TZ)

        # (engine=openpyxl; install via: pip install openpyxl)
        with pd.ExcelWriter(EXCEL_OUT, engine="openpyxl") as writer:
            summary_x.to_excel(writer, sheet_name="Summary", index=False)
            trades_x.to_excel(writer,  sheet_name="Trades",  index=False)
            equity_x.to_excel(writer,  sheet_name="Equity",  index=False)
        print(f"Wrote Excel workbook: {EXCEL_OUT}")

        # -------- XML exports --------
        df_to_xml(per_ticker_sorted, "PerTickerSummary", "TickerRow", XML_SUMMARY)
        df_to_xml(trades,            "Trades",           "Trade",     XML_TRADES)
        df_to_xml(equity,            "EquityCurve",      "Row",       XML_EQUITY)
        print(f"Wrote XML files: {XML_SUMMARY}, {XML_TRADES}, {XML_EQUITY}")

    else:
        print("No trades generated with current settings.")

if __name__ == "__main__":
    run_sp500()
```

```python
import pandas as pd
import numpy as np

# === CONFIG ===
PARQUET_PATH = r"C:\Users\pcagm\OneDrive\Desktop\downloads\AAPL.parquet"

# If your file already has a PnL column, set it here (e.g., "PnL_$") and the script will use it.
PNL_COL = None              # e.g., "PnL_$" or leave as None to auto/compute
USE_PCT_RETURNS = False     # if no PnL col: False => dollar change (Close.diff), True => % return (Close.pct_change())

# Candidate columns (auto-detect if not set)
TICKER_COL_CANDIDATES = ["Ticker", "Symbol", "ticker", "symbol"]
DATE_COL_CANDIDATES   = ["Date", "date", "Timestamp", "timestamp", "ExitDate", "TradeDate", "EventAt", "Datetime", "datetime"]
PRICE_COL_CANDIDATES  = ["Close", "Adj Close", "close", "adj_close"]
R_COL_CANDIDATES      = ["R", "r", "R_multiple", "Rmultiple"]  # optional

def _find_col(df, candidates, required=True):
    for c in candidates:
        if c in df.columns:
            return c
    if required:
        raise ValueError(f"Could not find any of {candidates} in columns: {list(df.columns)}")
    return None

def _summarize_group(g, pnl_col, r_col=None):
    pnl = pd.to_numeric(g[pnl_col], errors="coerce").dropna()

    trades = len(pnl)
    wins_mask = pnl > 0
    losses_mask = pnl < 0
    wins = int(wins_mask.sum())
    losses = int(losses_mask.sum())

    pos_sum = pnl[wins_mask].sum()
    neg_sum = pnl[losses_mask].sum()  # negative or 0

    pf = np.nan
    if neg_sum < 0:
        pf = pos_sum / abs(neg_sum)
    elif pos_sum > 0 and neg_sum == 0:
        pf = np.inf  # no losses

    out = {
        "Trades": trades,
        "PnL_Sum": pnl.sum(),
        "AvgPnL": pnl.mean(),
        "StdPnL": pnl.std(ddof=1) if trades > 1 else 0.0,
        "Wins": wins,
        "Losses": losses,
        "WinRate": (wins / trades * 100.0) if trades else np.nan,
        "AvgWin": pnl[wins_mask].mean() if wins else np.nan,
        "AvgLoss": pnl[losses_mask].mean() if losses else np.nan,
        "PF": pf,
    }

    # R-multiple (optional)
    if r_col and r_col in g.columns:
        rvals = pd.to_numeric(g[r_col], errors="coerce").dropna()
        out["AvgR"] = rvals.mean() if len(rvals) else np.nan
        out["MedianR"] = rvals.median() if len(rvals) else np.nan
    else:
        out["AvgR"] = np.nan
        out["MedianR"] = np.nan

    return pd.Series(out)

# ===== Load =====
df = pd.read_parquet(PARQUET_PATH, engine="pyarrow")  # or engine="fastparquet"

# Auto-detect key columns
TICKER_COL = _find_col(df, TICKER_COL_CANDIDATES)
DATE_COL   = _find_col(df, DATE_COL_CANDIDATES)
if PNL_COL is None:
    PRICE_COL = _find_col(df, PRICE_COL_CANDIDATES)

# Ensure datetime (make tz-naive for grouping)
dt = pd.to_datetime(df[DATE_COL], errors="coerce", utc=True)
df[DATE_COL] = dt.dt.tz_localize(None)

# Sort for diff/pct_change to make sense
df = df.sort_values([TICKER_COL, DATE_COL])

# If no explicit PnL column, compute proxy from price moves
if PNL_COL is None:
    if USE_PCT_RETURNS:
        df["__PnL__"] = df.groupby(TICKER_COL)[PRICE_COL].pct_change()
    else:
        df["__PnL__"] = df.groupby(TICKER_COL)[PRICE_COL].diff()
    PNL_COL = "__PnL__"

# Optional R column (many market datasets don’t have this)
R_COL = None
for c in R_COL_CANDIDATES:
    if c in df.columns:
        R_COL = c
        break

# ===== Overall per-ticker =====
overall = (
    df.groupby(TICKER_COL, dropna=False)
      .apply(lambda g: _summarize_group(g, pnl_col=PNL_COL, r_col=R_COL))
      .reset_index()
)

# ===== Per-ticker, per-month =====
df["Month"] = df[DATE_COL].dt.to_period("M").dt.to_timestamp("MS")
monthly = (
    df.groupby([TICKER_COL, "Month"], dropna=False)
      .apply(lambda g: _summarize_group(g, pnl_col=PNL_COL, r_col=R_COL))
      .reset_index()
)

# Pretty rounding for display
to_round = ["PnL_Sum","AvgPnL","StdPnL","WinRate","AvgWin","AvgLoss","AvgR","MedianR","PF"]
for tbl in (overall, monthly):
    for c in to_round:
        if c in tbl.columns:
            tbl[c] = tbl[c].astype(float).round(4)

print("=== Overall per Ticker ===")
print(overall.to_string(index=False))

print("\n=== Per Ticker, Per Month ===")
print(monthly.sort_values(["Month", TICKER_COL]).tail(30).to_string(index=False))

# === Optionally save ===
# overall.to_excel(r"C:\Users\pcagm\OneDrive\Desktop\downloads\summary_overall.xlsx", index=False)
# monthly.to_excel(r"C:\Users\pcagm\OneDrive\Desktop\downloads\summary_monthly.xlsx", index=False)
```
