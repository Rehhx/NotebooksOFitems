```python
import pandas as pd
import requests

# Base stat URLs (without or with ?date=...) — we'll add the date dynamically
stats = {
    'Opponent Pass Attempts per Game': 'https://www.teamrankings.com/college-football/stat/opponent-pass-attempts-per-game',
    'Opponent Completions per Game': 'https://www.teamrankings.com/college-football/stat/opponent-completions-per-game',
    'Opponent Incompletions per Game': 'https://www.teamrankings.com/college-football/stat/opponent-incompletions-per-game',
    'Opponent Completion Pct': 'https://www.teamrankings.com/college-football/stat/opponent-completion-pct',
    'Opponent Passing Yards per Game': 'https://www.teamrankings.com/college-football/stat/opponent-passing-yards-per-game',
    'Opponent Passing First Downs per Game': 'https://www.teamrankings.com/college-football/stat/opponent-passing-first-downs-per-game',
    'Opponent Avg Team Passer Rating': 'https://www.teamrankings.com/college-football/stat/opponent-average-team-passer-rating',
    'Sack Pct': 'https://www.teamrankings.com/college-football/stat/sack-pct',
    'Opponent Passing Play Pct': 'https://www.teamrankings.com/college-football/stat/opponent-passing-play-pct',
    'Opponent Passing Yards Pct': 'https://www.teamrankings.com/college-football/stat/opponent-passing-yards-pct',
    'Sacks per Game': 'https://www.teamrankings.com/college-football/stat/sacks-per-game',
    'Opponent Yards per Pass Attempt': 'https://www.teamrankings.com/college-football/stat/opponent-yards-per-pass-attempt',
    'Opponent Yards per Completion': 'https://www.teamrankings.com/college-football/stat/opponent-yards-per-completion'
}

# Years: 2025 to 2013
years = range(2025, 2012, -1)

# Master dictionary: {stat_label: {year: DataFrame}}
dfs = {}

for stat_label, base_url in stats.items():
    print(f"\n📊 Scraping: {stat_label}")
    stat_dfs = {}

    for year in years:
        dated_url = f"{base_url}?date={year}-01-21"
        print(f"  🔄 Year {year} → {dated_url}")
        
        try:
            tables = pd.read_html(dated_url)
            if tables:
                df = tables[0]
                stat_dfs[year] = df

                # Save to file
                safe_name = stat_label.lower().replace(" ", "_").replace("%", "pct")
                filename = f"{safe_name}_{year}.csv"
                df.to_csv(filename, index=False)
                print(f"    ✅ Saved {filename}")
            else:
                print(f"    ⚠️ No table found for {year}")
        except Exception as e:
            print(f"    ❌ Error in {year}: {e}")

    dfs[stat_label] = stat_dfs

# Example access: dfs['Sack Pct'][2020]
```
