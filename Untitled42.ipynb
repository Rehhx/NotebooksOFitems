```python
# orb_15m_retest_spy.py
# 15m ORB breakout with entry on retest AFTER a breakout-close beyond ORH/ORL.

import os
import sys
import math
from dataclasses import dataclass
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.patches import Rectangle

# =============================
# USER SETTINGS
# =============================
PARQUET = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet\SPY.parquet"
TZ = "America/New_York"
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

REG_START = "09:30"
REG_END   = "16:00"

# Retest behavior (your spec: entry on TOUCH after a breakout close)
MAX_RETEST_MIN: Optional[int] = 120     # None => wait until EOD
RETEST_NEEDS_CLOSE_THROUGH = False      # keep False to enter on touch

# Risk / exits
R_MULTIPLES: List[float] = [1.0, 2.0]   # 1R, 2R
POSITION_DOLLARS = 10_000.0
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Output
TRADES_CSV = "trades_spy_orb.csv"
EQUITY_CSV = "equity_spy_orb.csv"
CHART_DIR  = "charts"

# =============================
# Utilities
# =============================
STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                      rename[lower[k]] = std
        elif k.capitalize() in df.columns:  rename[k.capitalize()] = std
        elif k.upper() in df.columns:       rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _infer_median_interval_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    d = (idx[1:] - idx[:-1]).total_seconds() / 60.0
    return float(np.median(d)) if len(d) else np.nan

def _resample_to_15m_if_needed(df: pd.DataFrame) -> pd.DataFrame:
    """
    If bars finer than 15m, resample to right-labeled, right-closed 15T.
    If already ~15m, keep.
    If coarser, return empty.
    """
    m = _infer_median_interval_minutes(df.index)
    if np.isnan(m): return pd.DataFrame()
    if m < 15 - 1e-6:
        agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
        if "Volume" in df.columns: agg["Volume"] = "sum"
        out = (df.sort_index()
                 .resample("15T", label="right", closed="right")
                 .agg(agg)
                 .dropna(subset=["Open","High","Low","Close"]))
        return out
    elif abs(m - 15) <= 1e-6:
        return df.sort_index()
    else:
        return pd.DataFrame()

def load_spy_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()

    # TZ handling
    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df = df.tz_localize("UTC").tz_convert(tz)
        else:
            df = df.tz_localize(tz)
    else:
        df = df.tz_convert(tz)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    df = _resample_to_15m_if_needed(df)
    if df.empty:
        raise ValueError("Input is coarser than 15m or empty after normalization.")

    # Regular trading hours
    df = df.between_time(REG_START, REG_END).sort_index()
    return df

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB helpers
# =============================
def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    """
    ORH/ORL come from the COMPLETED first 15m RTH bar (right-labeled ~09:45).
    """
    df = df_15.copy()
    df["Session"] = sessionize(df)
    df["ORH"] = np.nan
    df["ORL"] = np.nan

    for ses, g in df.groupby("Session"):
        g = g.sort_index()
        if g.empty:
            continue
        first_ts = g.index[0]
        df.loc[first_ts, "ORH"] = float(g["High"].iloc[0])
        df.loc[first_ts, "ORL"] = float(g["Low"].iloc[0])
        # forward-fill within session
        df.loc[g.index, ["ORH","ORL"]] = df.loc[g.index, ["ORH","ORL"]].ffill()

    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

@dataclass
class Trade:
    Session: pd.Timestamp
    Direction: str
    BreakoutTime: pd.Timestamp
    EntryTime: pd.Timestamp
    Entry: float
    Stop: float
    ORH: float
    ORL: float
    Targets: List[float]
    Exits: List[Tuple[str, pd.Timestamp, float]]
    Qty: float
    PnL_USD: float
    R_multiple: float

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: Optional[int] = MAX_RETEST_MIN,
    r_targets: Optional[List[float]] = None,
    dollars: float = POSITION_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_needs_close: bool = RETEST_NEEDS_CLOSE_THROUGH
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    • ORH/ORL from first 15m bar only.
    • Find first later bar (strictly after first bar) whose Close > ORH or Close < ORL (breakout).
    • AFTER that breakout bar CLOSES, wait for first retest TOUCH of the level and enter.
      (If retest_needs_close=True, require the retest bar to CLOSE back across the level.)
    • Stop = opposite OR; 1R/2R targets; stop has priority.
    """
    if r_targets is None:
        r_targets = [1.0, 2.0]

    df = df_15.copy()
    df["Session"] = sessionize(df)
    trades: List[Trade] = []

    for ses, sdf in df.groupby("Session"):
        sdf = sdf.sort_index()
        if len(sdf) < 3:
            continue

        first_ts = sdf.index[0]
        orh = float(sdf.loc[first_ts, "ORH"])
        orl = float(sdf.loc[first_ts, "ORL"])

        # breakout search strictly AFTER first bar close
        scan = sdf[sdf.index > first_ts]
        long_break  = scan[scan["Close"] > orh].head(1)
        short_break = scan[scan["Close"] < orl].head(1)
        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_time = long_break.index[0] if direction == "long" else short_break.index[0]
        elif not long_break.empty:
            direction = "long";  breakout_time = long_break.index[0]
        else:
            direction = "short"; breakout_time = short_break.index[0]

        # retest strictly AFTER the breakout close
        if max_retest_min is None:
            after = sdf[sdf.index > breakout_time]
        else:
            cutoff = breakout_time + pd.Timedelta(minutes=int(max_retest_min))
            after = sdf[(sdf.index > breakout_time) & (sdf.index <= cutoff)]
        if after.empty:
            continue

        level = orh if direction == "long" else orl

        # Find first TOUCH of level
        touch = after[(after["Low"] <= level) & (after["High"] >= level)].head(1)
        if touch.empty:
            continue

        # Optional retest close back across level
        if retest_needs_close:
            row = touch.iloc[0]
            ok = (row["Close"] >= level) if direction == "long" else (row["Close"] <= level)
            if not ok:
                continue

        retest_time = touch.index[0]
        entry = _apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop  = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1
        targets = [(entry + r*risk_per_share) if direction == "long" else (entry - r*risk_per_share)
                   for r in r_targets]

        # simulate exits from entry bar forward
        sdf_run = sdf[sdf.index >= retest_time]
        exits: List[Tuple[str, pd.Timestamp, float]] = []
        for ts, row in sdf_run.iterrows():
            high, low = float(row["High"]), float(row["Low"])
            # stop first
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px))
                break
            # then targets (first one that hits)
            hit = False
            for i, tp in enumerate(targets):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
                    break
            if hit:
                break

        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(float(last["Close"]), slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _, _, px in exits) - fees
        r_mult = cash_pnl / (risk_per_share * qty)

        trades.append(Trade(
            Session=ses, Direction=direction, BreakoutTime=breakout_time,
            EntryTime=retest_time, Entry=float(entry), Stop=float(stop),
            ORH=orh, ORL=orl, Targets=[float(x) for x in targets],
            Exits=[(t, ts, float(px)) for (t, ts, px) in exits],
            Qty=float(qty), PnL_USD=float(cash_pnl), R_multiple=float(r_mult)
        ))

    # build dataframes
    if not trades:
        return pd.DataFrame(), pd.DataFrame(columns=["Session","CumPnL_USD"])

    tlog = pd.DataFrame([t.__dict__ for t in trades]).sort_values("Session")
    eq = (tlog.groupby("Session")["PnL_USD"].sum()
          .sort_index().cumsum().reset_index().rename(columns={"PnL_USD": "CumPnL_USD"}))
    return tlog, eq

# =============================
# Plotting
# =============================
def _draw_candles(ax, df_win: pd.DataFrame, width_minutes=8):
    """Simple OHLC candles (white up, black down) — robust across Matplotlib versions."""
    width_days = width_minutes / (24.0 * 60.0)
    for ts, row in df_win.iterrows():
        o, h, l, c = row["Open"], row["High"], row["Low"], row["Close"]
        color = "white" if c >= o else "black"
        # wick
        ax.vlines(ts, l, h, linewidth=1, color="black")
        # body
        y = min(o, c)
        height = max(abs(c - o), 1e-6)
        x_left = mdates.date2num(ts) - width_days/2.0
        rect = Rectangle((x_left, y), width_days, height, edgecolor="black",
                         facecolor=color, linewidth=1.0)
        ax.add_patch(rect)
    ax.xaxis_date()

def _format_time(ts: pd.Timestamp) -> str:
    return ts.tz_convert(TZ).strftime("%H:%M")

def plot_trade(df_15: pd.DataFrame, trade_row: pd.Series, outdir: str, dpi: int = 140):
    ses = trade_row["Session"]
    direction = trade_row["Direction"]

    sdf = df_15[sessionize(df_15) == pd.to_datetime(ses)]
    if sdf.empty:
        return

    df_win = sdf.copy()

    fig, ax = plt.subplots(figsize=(12, 5), dpi=dpi)
    _draw_candles(ax, df_win)

    # OR lines (constant)
    orh = float(trade_row["ORH"]); orl = float(trade_row["ORL"])
    ax.axhline(orh, ls="--", lw=1.4, color="tab:blue")
    ax.axhline(orl, ls="--", lw=1.4, color="tab:blue")

    # Entry/Stop
    entry = float(trade_row["Entry"]); stop = float(trade_row["Stop"])
    ax.axhline(entry, ls="-", lw=1.2, color="tab:green")
    ax.axhline(stop,  ls="-", lw=1.2, color="tab:red")

    etime = trade_row["EntryTime"]
    btime = trade_row["BreakoutTime"]
    ax.axvline(btime, ls=":", lw=1.0, color="tab:orange")
    ax.axvline(etime, ls=":", lw=1.0, color="tab:green")

    ax.text(df_win.index[-1], df_win["Close"].min(),
            f"PnL: ${float(trade_row['PnL_USD']):,.2f}", ha="right", va="bottom",
            bbox=dict(fc="white", ec="0.7"))

    ax.text(etime, entry, f"Entry: {entry:.2f}\n{_format_time(etime)}",
            ha="left", va="bottom", fontsize=9,
            bbox=dict(fc="white", ec="0.6"))

    ax.set_title(f"SPY {pd.to_datetime(ses).date()}  ({direction})")
    ax.grid(True, alpha=0.25)
    fig.autofmt_xdate()

    os.makedirs(outdir, exist_ok=True)
    fname = os.path.join(outdir, f"UpdatedCharts_SPY_{pd.to_datetime(ses).date()}_{direction}.png")
    plt.savefig(fname, bbox_inches="tight")
    plt.close()
    print(f"Wrote {fname}")

# =============================
# Runner
# =============================
def main():
    # Friendly to notebooks: ignore unknown args (e.g., -f)
    import argparse
    ap = argparse.ArgumentParser()
    ap.add_argument("--parquet", default=PARQUET)
    ap.add_argument("--charts",  default=CHART_DIR)
    ap.add_argument("--dpi",     type=int, default=140)
    ap.add_argument("--limit",   type=int, default=0, help="max charts to render (0 = all)")
    ap.add_argument("--start",   type=str, default=None)
    ap.add_argument("--end",     type=str, default=None)
    args, _ = ap.parse_known_args()

    print(f"Loading: {args.parquet}")
    df15 = load_spy_15m(args.parquet, tz=TZ)
    # optional session range
    if args.start:
        df15 = df15[df15.index.date >= pd.to_datetime(args.start).date()]
    if args.end:
        df15 = df15[df15.index.date <= pd.to_datetime(args.end).date()]

    df15 = compute_opening_range(df15)

    print("[RTH] {}  {} → {}  (first 15m is the bar labeled ~09:45)".format(
        df15.index[0].date(), REG_START, REG_END))

    print("Running ORB retest backtest…")
    tlog, eq = backtest_orb_retest(
        df15,
        max_retest_min=MAX_RETEST_MIN,
        r_targets=R_MULTIPLES,
        dollars=POSITION_DOLLARS,
        slippage_bps=SLIPPAGE_BPS,
        fees=FEES_PER_TRADE,
        retest_needs_close=RETEST_NEEDS_CLOSE_THROUGH
    )

    if not tlog.empty:
        tlog.to_csv(TRADES_CSV, index=False); print(f"Saved trades: {TRADES_CSV}")
    if not eq.empty:
        eq.to_csv(EQUITY_CSV, index=False);   print(f"Saved equity: {EQUITY_CSV}")

    # Overall totals
    if tlog is None or tlog.empty:
        print("No trades.")
        return

    total_trades = len(tlog)
    total_pnl    = tlog["PnL_USD"].sum()
    win_rate     = (tlog["PnL_USD"] > 0).mean() * 100.0
    avg_r        = tlog["R_multiple"].mean()
    median_r     = tlog["R_multiple"].median()

    print(f"\n=== Overall Totals (SPY) ===")
    print(f"Total trades: {total_trades}")
    print(f"Total PnL ($): {total_pnl:,.2f}")
    print(f"Win rate: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")

    # Charts
    outdir = args.charts
    os.makedirs(outdir, exist_ok=True)
    n = 0
    for _, row in tlog.iterrows():
        plot_trade(df15, row, outdir, dpi=args.dpi)
        n += 1
        if args.limit and n >= args.limit:
            break

if __name__ == "__main__":
    main()
```

```python
# validate_spy_daily_prices.py
import warnings
warnings.filterwarnings("ignore")

import os, glob
from typing import Tuple
import numpy as np
import pandas as pd

# =============================
# CONFIG
# =============================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
DAILY_CSV   = "spy_daily_prices.csv"              # created by your ORB script
OUT_VALIDATION_CSV = "spy_daily_price_validation.csv"
OUT_PARQUET_DAILY  = "spy_daily_prices_from_parquet.csv"
OUT_DAILY_CLOSE    = "spy_daily_close.csv"        # simple “price of the day” = Close

ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# Tolerances for price comparisons
ABS_TOL = 0.01      # $0.01 absolute
REL_TOL = 1e-4      # 1 basis point relative (0.01%) ~ tighter than ABS_TOL usually

# =============================
# Helpers
# =============================
STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                      rename[lower[k]] = std
        elif k.capitalize() in df.columns:  rename[k.capitalize()] = std
        elif k.upper() in df.columns:       rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _infer_median_interval_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    d = (idx[1:] - idx[:-1]).total_seconds() / 60.0
    return float(np.median(d)) if len(d) else np.nan

def _resample_to_15m_if_needed(df: pd.DataFrame) -> pd.DataFrame:
    """
    If bars are finer than 15m, resample to exact 15m OHLC(V)
    using right-closed/right-labeled bins so first bar ends 09:45.
    If already ~15m, keep as-is; if coarser, return empty.
    """
    m = _infer_median_interval_minutes(df.index)
    if np.isnan(m): return pd.DataFrame()
    if m < 15 - 1e-6:
        agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
        if "Volume" in df.columns: agg["Volume"] = "sum"
        out = (df.sort_index()
                 .resample("15T", label="right", closed="right")
                 .agg(agg)
                 .dropna(subset=["Open","High","Low","Close"]))
        return out
    elif abs(m - 15) <= 1e-6:
        return df.sort_index()
    else:
        # Coarser than 15m → we can't reconstruct 15m; still allow daily aggregation directly
        return df.sort_index()

def _find_spy_file(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    if not files:
        return ""
    candidates = [f for f in files if "spy" in os.path.basename(f).lower()]
    if candidates:
        # prefer a clean name if present
        exact = [f for f in candidates if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(candidates)[0]
    strict = os.path.join(folder, "SPY.parquet")
    return strict if os.path.exists(strict) else ""

def load_spy_15m(folder: str) -> pd.DataFrame:
    fpath = _find_spy_file(folder)
    if not fpath:
        raise FileNotFoundError(f"SPY parquet not found in: {folder}")

    df = pd.read_parquet(fpath, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()

    # TZ handling
    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df = df.tz_localize("UTC").tz_convert(TZ)
        else:
            df = df.tz_localize(TZ)
    else:
        df = df.tz_convert(TZ)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(fpath)} missing required column: {need}")

    # Resample if needed, then restrict to RTH
    df15 = _resample_to_15m_if_needed(df)
    if df15.empty:
        # If coarser than 15m, attempt RTH filter on original df
        base = df
    else:
        base = df15

    base = base.between_time(REG_SESSION_START, REG_SESSION_END).sort_index()
    return base

def compute_daily_from_bars(df_rth: pd.DataFrame) -> pd.DataFrame:
    g = df_rth.copy()
    g["Session"] = pd.to_datetime(g.index.date)
    agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
    if "Volume" in g.columns:
        agg["Volume"] = "sum"
    daily = g.groupby("Session").agg(agg).reset_index()
    return daily

def _rel_diff(a: float, b: float) -> float:
    denom = max(abs(a), abs(b), 1e-12)
    return (a - b) / denom

def compare_daily(parquet_daily: pd.DataFrame, csv_daily: pd.DataFrame) -> pd.DataFrame:
    left = parquet_daily.copy()
    right = csv_daily.copy()

    # Normalize dtypes
    left["Session"]  = pd.to_datetime(left["Session"]).dt.date
    right["Session"] = pd.to_datetime(right["Session"]).dt.date

    cols = ["Open","High","Low","Close"] + (["Volume"] if "Volume" in left.columns and "Volume" in right.columns else [])

    m = left.merge(right[["Session"] + cols], on="Session", suffixes=("_parquet","_csv"), how="inner")

    # Compute diffs
    for c in ["Open","High","Low","Close"]:
        m[f"{c}_diff_abs"] = (m[f"{c}_parquet"] - m[f"{c}_csv"]).abs()
        m[f"{c}_diff_rel"] = m.apply(lambda r: abs(_rel_diff(r[f"{c}_parquet"], r[f"{c}_csv"])), axis=1)
        m[f"{c}_pass"] = (m[f"{c}_diff_abs"] <= ABS_TOL) | (m[f"{c}_diff_rel"] <= REL_TOL)

    if "Volume" in cols:
        m["Volume_diff"] = (m["Volume_parquet"] - m["Volume_csv"]).abs()
        # Volume tolerance: allow small rounding/aggregation differences
        m["Volume_pass"] = m["Volume_diff"] <= 1_000  # adjust if needed

    # Overall pass flag
    pass_cols = [c for c in m.columns if c.endswith("_pass")]
    m["ALL_PASS"] = m[pass_cols].all(axis=1) if pass_cols else True
    return m.sort_values("Session")

def main():
    print("Loading SPY bars from parquet…")
    df_rth = load_spy_15m(PARQUET_DIR)
    print(f"Bars loaded: {len(df_rth):,} | tz={df_rth.index.tz} | median interval minutes≈{_infer_median_interval_minutes(df_rth.index):.2f}")

    print("Aggregating to daily OHLC…")
    daily_from_parquet = compute_daily_from_bars(df_rth)
    daily_from_parquet.to_csv(OUT_PARQUET_DAILY, index=False)
    print(f"Wrote {OUT_PARQUET_DAILY} with {len(daily_from_parquet):,} rows.")

    # Also write a simple daily close (“price of the day”)
    daily_close = daily_from_parquet[["Session","Close"]].rename(columns={"Close":"CostOfDay_Close"})
    daily_close.to_csv(OUT_DAILY_CLOSE, index=False)
    print(f"Wrote {OUT_DAILY_CLOSE} (Close as price-of-day).")

    if not os.path.exists(DAILY_CSV):
        print(f"Reference CSV {DAILY_CSV} not found. Skipping comparison.")
        return

    print(f"Reading reference daily CSV: {DAILY_CSV}")
    ref = pd.read_csv(DAILY_CSV)

    print("Comparing parquet-derived daily OHLC vs CSV…")
    validation = compare_daily(daily_from_parquet, ref)
    validation.to_csv(OUT_VALIDATION_CSV, index=False)
    n = len(validation)
    n_ok = int(validation["ALL_PASS"].sum())
    n_bad = n - n_ok
    print(f"Wrote {OUT_VALIDATION_CSV}. Rows: {n} | PASS: {n_ok} | FAIL: {n_bad}")

    if n_bad > 0:
        # Show a few failing examples
        fails = validation[~validation["ALL_PASS"]].head(10)
        print("\nExample mismatches (top 10):")
        with pd.option_context("display.max_columns", None, "display.width", 140):
            print(fails)

    # Extra sanity: within each day, check Open/Close match first/last 15m bars
    print("\nSanity-checking Open/Close vs first/last 15m bar of each session…")
    g = df_rth.copy()
    g["Session"] = pd.to_datetime(g.index.date)
    firsts = g.groupby("Session").head(1).set_index("Session")[["Open"]].rename(columns={"Open":"FirstBarOpen"})
    lasts  = g.groupby("Session").tail(1).set_index("Session")[["Close"]].rename(columns={"Close":"LastBarClose"})
    z = daily_from_parquet.set_index("Session").join(firsts).join(lasts).reset_index()
    z["Open_match"]  = (z["Open"]  - z["FirstBarOpen"]).abs() <= ABS_TOL
    z["Close_match"] = (z["Close"] - z["LastBarClose"]).abs() <= ABS_TOL
    open_ok  = int(z["Open_match"].sum())
    close_ok = int(z["Close_match"].sum())
    print(f"Open matches first 15m: {open_ok}/{len(z)} | Close matches last 15m: {close_ok}/{len(z)}")
    if open_ok < len(z) or close_ok < len(z):
        bad = z[(~z["Open_match"]) | (~z["Close_match"])].head(10)
        print("\nOpen/Close mismatch examples (top 10):")
        with pd.option_context("display.max_columns", None, "display.width", 140):
            print(bad)

if __name__ == "__main__":
    main()
```

```python
# plot_orb_trades_spy.py
import warnings
warnings.filterwarnings("ignore")

import os, glob
from datetime import timedelta
from typing import List, Tuple
import ast

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.patches import Rectangle

# =============================
# CONFIG
# =============================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
OUT_DIR     = "charts"                         # where PNGs go
TRADES_CSV  = "orb_trades_sp500.csv"          # optional: if present, we’ll use it
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.0

# =============================
# Utilities shared with your ORB script
# =============================
STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                      rename[lower[k]] = std
        elif k.capitalize() in df.columns:  rename[k.capitalize()] = std
        elif k.upper() in df.columns:       rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _infer_median_interval_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    d = (idx[1:] - idx[:-1]).total_seconds() / 60.0
    return float(np.median(d)) if len(d) else np.nan

def _resample_to_15m_if_needed(df: pd.DataFrame) -> pd.DataFrame:
    m = _infer_median_interval_minutes(df.index)
    if np.isnan(m): return pd.DataFrame()
    if m < 15 - 1e-6:
        agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
        if "Volume" in df.columns: agg["Volume"] = "sum"
        out = (df.sort_index()
                 .resample("15T", label="right", closed="right")
                 .agg(agg)
                 .dropna(subset=["Open","High","Low","Close"]))
        return out
    elif abs(m - 15) <= 1e-6:
        return df.sort_index()
    else:
        return pd.DataFrame()

def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()
    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df = df.tz_localize("UTC").tz_convert(tz)
        else:
            df = df.tz_localize(tz)
    else:
        df = df.tz_convert(tz)
    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")
    df = _resample_to_15m_if_needed(df)
    if df.empty:
        return df
    return df.between_time(REG_SESSION_START, REG_SESSION_END).sort_index()

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()
    for col in ["ORH","ORL"]:
        if col not in df.columns: df[col] = np.nan
    df["Session"] = sessionize(df)
    first_idx = df.groupby("Session").head(1).index
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    trades = []
    for ses, sdf in df.groupby("Session"):
        if len(sdf) < 3: continue
        orh = float(sdf["ORH"].iloc[0]); orl = float(sdf["ORL"].iloc[0])
        if pd.isna(orh) or pd.isna(orl): continue
        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)
        if long_break.empty and short_break.empty: continue
        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction == "long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty: continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty: continue
        retest_bar = touch.iloc[0]; retest_time = retest_bar.name
        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok: continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop  = orl if direction == "long" else orh
        risk  = abs(entry - stop)
        if risk <= 1e-12: continue
        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1
        t_prices = [(entry + r*risk) if direction == "long" else (entry - r*risk) for r in r_targets]

        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            hi, lo = float(row["High"]), float(row["Low"])
            if lo <= stop <= hi:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px)); break
            hit = False
            for i, tp in enumerate(t_prices):
                if lo <= tp <= hi:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px)); hit = True
            if hit: break
        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(float(last["Close"]), slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        n_parts = len(exits); qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _,_,px in exits) - FEES_PER_TRADE
        trades.append({
            "Ticker": "SPY",
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": float(entry), "Stop": float(stop),
            "Targets": [float(x) for x in t_prices],
            "Exits": [(str(t[0]), t[1], float(t[2])) for t in exits],
            "Qty": float(qty),
            "PnL_$": float(cash_pnl),
        })
    tlog = pd.DataFrame(trades)
    eq = (tlog.groupby("Session")["PnL_$"].sum()
          .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"})) if not tlog.empty else pd.DataFrame()
    return tlog, eq

def _find_spy_file(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    candidates = [f for f in files if "spy" in os.path.basename(f).lower()]
    if candidates:
        exact = [f for f in candidates if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(candidates)[0]
    strict = os.path.join(folder, "SPY.parquet")
    return strict if os.path.exists(strict) else ""

# =============================
# Plotting
# =============================
def _parse_exits(exits_field):
    """
    exits_field is either a list of (tag, ts, px) or a CSV string of 'tag|ts|px;...'
    Return (first_tag, first_ts, first_px) or (None, None, None)
    """
    if isinstance(exits_field, list):
        tag, ts, px = exits_field[0]
        return tag, pd.to_datetime(ts), float(px)
    if isinstance(exits_field, str):
        try:
            # CSV form from saved file
            first = exits_field.split(";")[0]
            tag, ts, px = first.split("|")
            return tag, pd.to_datetime(ts), float(px)
        except Exception:
            pass
    # Sometimes pandas may have repr of list-as-string; try ast.literal_eval
    try:
        v = ast.literal_eval(exits_field)
        if isinstance(v, list) and v:
            tag, ts, px = v[0]
            return tag, pd.to_datetime(ts), float(px)
    except Exception:
        return None, None, None
    return None, None, None

def _draw_candles(ax, df_day: pd.DataFrame, width_minutes=10):
    """
    Minimal candlestick without specifying colors.
    """
    if df_day.empty: return
    width = (width_minutes / (24*60))  # in matplotlib date units
    times = mdates.date2num(df_day.index.to_pydatetime())

    for i, (t, row) in enumerate(df_day.iterrows()):
        o, h, l, c = float(row["Open"]), float(row["High"]), float(row["Low"]), float(row["Close"])
        # wick
        ax.plot([times[i], times[i]], [l, h], linewidth=1.0)
        # body (filled if close >= open; outlined if close < open)
        y = min(o, c); height = abs(c - o)
        if height == 0:
            height = 1e-6
        rect = Rectangle((times[i] - width/2.0, y), width, height, fill=False)  # outlined body
        ax.add_patch(rect)

def plot_trade(ax, df_day: pd.DataFrame, trade_row: pd.Series, title_prefix="SPY"):
    """
    Draw one trade on a given day's 15m candles.
    """
    _draw_candles(ax, df_day)
    ax.set_title(f"{title_prefix}  {pd.to_datetime(trade_row['Session']).date()}")

    # X axis formatting
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))
    ax.grid(True, linestyle=":", linewidth=0.6)

    # OR lines
    orh = float(trade_row["ORH"]); orl = float(trade_row["ORL"])
    t0, t1 = df_day.index.min(), df_day.index.max()
    ax.hlines(orh, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0, label="ORH")
    ax.hlines(orl, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0, label="ORL")

    # Entry + vertical marker
    entry_px  = float(trade_row["Entry"])
    entry_ts  = pd.to_datetime(trade_row["EntryTime"])
    ax.hlines(entry_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="solid", linewidth=1.2)
    ax.vlines(mdates.date2num(entry_ts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(), linestyles="dotted", linewidth=0.9)
    ax.annotate(f"Entry: {entry_px:.2f}\n{entry_ts.tz_convert(TZ).strftime('%H:%M') if entry_ts.tzinfo else entry_ts.strftime('%H:%M')}",
                xy=(mdates.date2num(entry_ts), entry_px),
                xytext=(10, 10), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    # Stop
    stop_px = float(trade_row["Stop"])
    ax.hlines(stop_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashdot", linewidth=1.0)
    ax.annotate(f"Stop: {stop_px:.2f}",
                xy=(mdates.date2num(t0), stop_px), xytext=(5, -12),
                textcoords="offset points", fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    # Exit (first realized exit)
    tag, ex_ts, ex_px = _parse_exits(trade_row["Exits"])
    if tag is not None:
        ax.scatter([mdates.date2num(ex_ts)], [ex_px], s=25)
        ax.annotate(f"Exit: {ex_px:.2f} ({tag})",
                    xy=(mdates.date2num(ex_ts), ex_px),
                    xytext=(10, -15), textcoords="offset points",
                    fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    # PnL label
    pnl = float(trade_row["PnL_$"])
    ax.text(0.99, 0.02, f"PnL: ${pnl:,.2f}", transform=ax.transAxes,
            ha="right", va="bottom", fontsize=9, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

def load_spy_and_trades(parquet_dir: str, trades_csv: str):
    # Try to reuse CSV of trades; if missing, recompute
    tlog = None
    if os.path.exists(trades_csv):
        tlog = pd.read_csv(trades_csv)
        # Try to bring Exits back to parseable form if they look like list strings
        if "EntryTime" in tlog.columns:
            tlog["EntryTime"] = pd.to_datetime(tlog["EntryTime"])
        else:
            # Older files may lack; we’ll skip those rows
            tlog["EntryTime"] = pd.NaT
    # Load SPY 15m dataframe
    spy_path = _find_spy_file(parquet_dir)
    if not spy_path:
        raise FileNotFoundError(f"SPY parquet not found in {parquet_dir}")
    df15 = load_parquet_15m(spy_path, tz=TZ)
    df15["Ticker"] = "SPY"
    df15 = compute_opening_range(df15)

    if tlog is None or tlog.empty:
        print("Trades CSV not found or empty; running backtest to create trade log in-memory.")
        tlog, _ = backtest_orb_retest(df15)
    return df15, tlog

def plot_all_spy_trades(parquet_dir: str = PARQUET_DIR, trades_csv: str = TRADES_CSV,
                        out_dir: str = OUT_DIR, limit: int = 50):
    os.makedirs(out_dir, exist_ok=True)
    df15, tlog = load_spy_and_trades(parquet_dir, trades_csv)
    if tlog is None or tlog.empty:
        print("No trades to plot.")
        return

    # Only SPY rows and ensure proper sorting
    spy_trades = tlog[tlog["Ticker"] == "SPY"].copy()
    spy_trades.sort_values(["Session", "EntryTime"], inplace=True)

    count = 0
    for _, tr in spy_trades.iterrows():
        ses = pd.to_datetime(tr["Session"]).date()
        df_day = df15[df15["Session"] == pd.to_datetime(ses)].copy()
        if df_day.empty:
            continue

        fig, ax = plt.subplots(figsize=(12, 6))
        plot_trade(ax, df_day, tr, title_prefix="SPY")
        fig.autofmt_xdate()
        fig.tight_layout()

        direction = str(tr.get("Direction", ""))
        fname = f"SPY_{ses}_{direction}.png".replace(" ", "_")
        out_path = os.path.join(out_dir, fname)
        fig.savefig(out_path, dpi=160)
        plt.close(fig)

        print(f"Wrote {out_path}")
        count += 1
        if limit and count >= limit:
            break

if __name__ == "__main__":
    # Change 'limit=None' to export all trades
    plot_all_spy_trades(limit=20)
```

```python
# plot_orb_trades_spy.py  (Jupyter-safe, Py 3.8+)
import warnings
warnings.filterwarnings("ignore")

import os, glob, argparse, ast, sys
from datetime import timedelta
from typing import Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.patches import Rectangle

# =============================
# CONFIG (defaults)
# =============================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
TRADES_CSV  = "orb_trades_sp500.csv"          # if present, we’ll reuse
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.0

# =============================
# Backtest helpers (same logic as your ORB script)
# =============================
STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                      rename[lower[k]] = std
        elif k.capitalize() in df.columns:  rename[k.capitalize()] = std
        elif k.upper() in df.columns:       rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _infer_median_interval_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    d = (idx[1:] - idx[:-1]).total_seconds() / 60.0
    return float(np.median(d)) if len(d) else np.nan

def _resample_to_15m_if_needed(df: pd.DataFrame) -> pd.DataFrame:
    m = _infer_median_interval_minutes(df.index)
    if np.isnan(m): return pd.DataFrame()
    if m < 15 - 1e-6:
        agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
        if "Volume" in df.columns: agg["Volume"] = "sum"
        out = (df.sort_index()
                 .resample("15T", label="right", closed="right")
                 .agg(agg)
                 .dropna(subset=["Open","High","Low","Close"]))
        return out
    elif abs(m - 15) <= 1e-6:
        return df.sort_index()
    else:
        return pd.DataFrame()

def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()
    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df = df.tz_localize("UTC").tz_convert(tz)
        else:
            df = df.tz_localize(tz)
    else:
        df = df.tz_convert(tz)
    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")
    df = _resample_to_15m_if_needed(df)
    if df.empty:
        return df
    return df.between_time(REG_SESSION_START, REG_SESSION_END).sort_index()

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()
    for col in ["ORH","ORL"]:
        if col not in df.columns: df[col] = np.nan
    df["Session"] = sessionize(df)
    first_idx = df.groupby("Session").head(1).index
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE
):
    df = df_15.copy()
    df["Session"] = sessionize(df)
    trades = []
    for ses, sdf in df.groupby("Session"):
        if len(sdf) < 3: continue
        orh = float(sdf["ORH"].iloc[0]); orl = float(sdf["ORL"].iloc[0])
        if pd.isna(orh) or pd.isna(orl): continue
        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)
        if long_break.empty and short_break.empty: continue
        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction == "long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty: continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty: continue
        retest_bar = touch.iloc[0]; retest_time = retest_bar.name
        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok: continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop  = orl if direction == "long" else orh
        risk  = abs(entry - stop)
        if risk <= 1e-12: continue
        qty = POSITION_SIZE_DOLLARS / entry
        side_mult = 1 if direction == "long" else -1
        t_prices = [(entry + r*risk) if direction == "long" else (entry - r*risk) for r in r_targets]

        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            hi, lo = float(row["High"]), float(row["Low"])
            if lo <= stop <= hi:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px)); break
            hit = False
            for i, tp in enumerate(t_prices):
                if lo <= tp <= hi:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px)); hit = True
            if hit: break
        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(float(last["Close"]), slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        n_parts = len(exits); qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _,_,px in exits) - FEES_PER_TRADE
        trades.append({
            "Ticker": "SPY",
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": float(entry), "Stop": float(stop),
            "Targets": [float(x) for x in t_prices],
            "Exits": [(str(t[0]), t[1], float(t[2])) for t in exits],
            "Qty": float(qty),
            "PnL_$": float(cash_pnl),
        })
    tlog = pd.DataFrame(trades)
    eq = (tlog.groupby("Session")["PnL_$"].sum()
          .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"})) if not tlog.empty else pd.DataFrame()
    return tlog, eq

def _find_spy_file(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    candidates = [f for f in files if "spy" in os.path.basename(f).lower()]
    if candidates:
        exact = [f for f in candidates if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(candidates)[0]
    strict = os.path.join(folder, "SPY.parquet")
    return strict if os.path.exists(strict) else ""

# =============================
# Plotting
# =============================
def _parse_exits(exits_field):
    if isinstance(exits_field, list) and exits_field:
        tag, ts, px = exits_field[0]
        return tag, pd.to_datetime(ts), float(px)
    if isinstance(exits_field, str):
        try:
            first = exits_field.split(";")[0]
            tag, ts, px = first.split("|")
            return tag, pd.to_datetime(ts), float(px)
        except Exception:
            pass
    try:
        v = ast.literal_eval(exits_field)
        if isinstance(v, list) and v:
            tag, ts, px = v[0]
            return tag, pd.to_datetime(ts), float(px)
    except Exception:
        return None, None, None
    return None, None, None

def _draw_candles(ax, df_day: pd.DataFrame, width_minutes=10):
    width = (width_minutes / (24*60))
    times = mdates.date2num(df_day.index.to_pydatetime())
    for i, (t, row) in enumerate(df_day.iterrows()):
        o, h, l, c = float(row["Open"]), float(row["High"]), float(row["Low"]), float(row["Close"])
        ax.plot([times[i], times[i]], [l, h], linewidth=1.0)  # wick
        y = min(o, c); height = abs(c - o) or 1e-6
        rect = Rectangle((times[i] - width/2.0, y), width, height, fill=False)
        ax.add_patch(rect)

def plot_trade(ax, df_day: pd.DataFrame, trade_row: pd.Series, title_prefix="SPY"):
    _draw_candles(ax, df_day)
    ax.set_title(f"{title_prefix}  {pd.to_datetime(trade_row['Session']).date()}")
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))
    ax.grid(True, linestyle=":", linewidth=0.6)

    orh = float(trade_row["ORH"]); orl = float(trade_row["ORL"])
    t0, t1 = df_day.index.min(), df_day.index.max()
    ax.hlines(orh, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)
    ax.hlines(orl, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)

    entry_px  = float(trade_row["Entry"])
    entry_ts  = pd.to_datetime(trade_row["EntryTime"])
    ax.hlines(entry_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="solid", linewidth=1.2)
    ax.vlines(mdates.date2num(entry_ts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(),
              linestyles="dotted", linewidth=0.9)
    lbl_time = entry_ts.tz_convert(TZ).strftime('%H:%M') if entry_ts.tzinfo else entry_ts.strftime('%H:%M')
    ax.annotate(f"Entry: {entry_px:.2f}\n{lbl_time}",
                xy=(mdates.date2num(entry_ts), entry_px),
                xytext=(10, 10), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    stop_px = float(trade_row["Stop"])
    ax.hlines(stop_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashdot", linewidth=1.0)
    ax.annotate(f"Stop: {stop_px:.2f}", xy=(mdates.date2num(t0), stop_px),
                xytext=(5, -12), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    tag, ex_ts, ex_px = _parse_exits(trade_row["Exits"])
    if tag is not None:
        ax.scatter([mdates.date2num(ex_ts)], [ex_px], s=25)
        ax.annotate(f"Exit: {ex_px:.2f} ({tag})",
                    xy=(mdates.date2num(ex_ts), ex_px),
                    xytext=(10, -15), textcoords="offset points",
                    fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    pnl = float(trade_row["PnL_$"])
    ax.text(0.99, 0.02, f"PnL: ${pnl:,.2f}", transform=ax.transAxes,
            ha="right", va="bottom", fontsize=9, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

def _find_spy_file(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    candidates = [f for f in files if "spy" in os.path.basename(f).lower()]
    if candidates:
        exact = [f for f in candidates if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(candidates)[0]
    strict = os.path.join(folder, "SPY.parquet")
    return strict if os.path.exists(strict) else ""

def _load_spy_df(parquet_dir: str) -> pd.DataFrame:
    spy_path = _find_spy_file(parquet_dir)
    if not spy_path:
        raise FileNotFoundError(f"SPY parquet not found in {parquet_dir}")
    df15 = load_parquet_15m(spy_path, tz=TZ)
    df15["Ticker"] = "SPY"
    return compute_opening_range(df15)

def _load_trades(df15: pd.DataFrame, trades_csv: str) -> pd.DataFrame:
    if os.path.exists(trades_csv):
        tlog = pd.read_csv(trades_csv)
        if "EntryTime" in tlog.columns:
            tlog["EntryTime"] = pd.to_datetime(tlog["EntryTime"])
        else:
            tlog["EntryTime"] = pd.NaT
        return tlog[tlog["Ticker"] == "SPY"].copy()
    tlog, _ = backtest_orb_retest(df15)
    return tlog

def plot_all_spy_trades(outdir: str, dpi: int, limit: Optional[int],
                        start: Optional[str], end: Optional[str],
                        skip_existing: bool):
    os.makedirs(outdir, exist_ok=True)
    df15 = _load_spy_df(PARQUET_DIR)
    tlog = _load_trades(df15, TRADES_CSV)
    if tlog is None or tlog.empty:
        print("No trades to plot."); return

    tlog["Session"] = pd.to_datetime(tlog["Session"])
    if start: tlog = tlog[tlog["Session"] >= pd.to_datetime(start)]
    if end:   tlog = tlog[tlog["Session"] <= pd.to_datetime(end)]

    tlog.sort_values(["Session", "EntryTime"], inplace=True)
    total = len(tlog); printed = 0

    for _, tr in tlog.iterrows():
        ses = tr["Session"].date()
        direction = str(tr.get("Direction", ""))
        fname = f"SPY_{ses}_{direction}.png".replace(" ", "_")
        out_path = os.path.join(outdir, fname)
        if skip_existing and os.path.exists(out_path):
            printed += 1
            continue

        df_day = df15[df15["Session"] == pd.to_datetime(ses)].copy()
        if df_day.empty:
            continue

        fig, ax = plt.subplots(figsize=(12, 6))
        plot_trade(ax, df_day, tr, title_prefix="SPY")
        fig.autofmt_xdate()
        fig.tight_layout()
        fig.savefig(out_path, dpi=dpi)
        plt.close(fig)
        print(f"Wrote {out_path}")

        printed += 1
        if (limit is not None) and (printed >= limit):
            break

    print(f"Done. Rendered {printed}/{total} charts.")

# ---------- JUPYTER-SAFE ARGPARSE ----------
def parse_args():
    p = argparse.ArgumentParser(description="Plot ORB retest entry charts for SPY trades.")
    p.add_argument("--outdir", default="charts", help="Output folder for PNGs.")
    p.add_argument("--dpi", type=int, default=160, help="Image DPI.")
    p.add_argument("--limit", type=int, default=-1,
                   help="Max charts to render. Use -1 for all (or --all).")
    p.add_argument("--all", action="store_true",
                   help="Render all charts (equivalent to --limit -1).")
    p.add_argument("--start", type=str, default=None,
                   help="Start date YYYY-MM-DD (optional).")
    p.add_argument("--end", type=str, default=None,
                   help="End date YYYY-MM-DD (optional).")
    p.add_argument("--skip-existing", action="store_true",
                   help="Skip charts that already exist on disk.")
    # IMPORTANT: Jupyter passes an extra "-f <json>" arg; ignore unknowns.
    args, _unknown = p.parse_known_args()
    return args

if __name__ == "__main__":
    args = parse_args()
    use_all = args.all or (args.limit == -1)
    limit = None if use_all else args.limit
    plot_all_spy_trades(
        outdir=args.outdir,
        dpi=args.dpi,
        limit=limit,
        start=args.start,
        end=args.end,
        skip_existing=args.skip_existing
    )
```

```python
# inspect_spy_parquet.py  — open a parquet, print what's inside, dump to CSV (no strategy)
import warnings
warnings.filterwarnings("ignore")

import os, glob, argparse
from typing import Optional
import numpy as np
import pandas as pd

# -------- config defaults --------
DEFAULT_FOLDER = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
TZ = "America/New_York"              # only used if you ask to convert

CANDIDATE_TIME_COLS = [
    "EventAt","Datetime","datetime","Timestamp","timestamp",
    "DateTime","DATE","Date","date","Time","time"
]

def _find_spy_parquet(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    if not files:
        raise FileNotFoundError(f"No parquet files in {folder}")
    cands = [f for f in files if "spy" in os.path.basename(f).lower()]
    if cands:
        exact = [f for f in cands if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(cands)[0]
    # fallback: first parquet
    return sorted(files)[0]

def _pick_time_col(df: pd.DataFrame) -> Optional[str]:
    if isinstance(df.index, pd.DatetimeIndex):
        return None  # already datetime index
    for c in CANDIDATE_TIME_COLS:
        if c in df.columns:
            return c
    return None

def _median_interval_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    d = (idx[1:] - idx[:-1]).total_seconds() / 60.0
    return float(np.median(d))

def inspect_and_export(path: str,
                       out_csv_raw: str,
                       parse_ts: bool,
                       out_csv_with_ts: Optional[str],
                       convert_to_tz: Optional[str]):
    print(f"\nReading parquet: {path}")
    df_raw = pd.read_parquet(path, engine="pyarrow")
    print(f"Rows: {len(df_raw):,}  | Cols: {len(df_raw.columns)}")
    print("Columns:", list(df_raw.columns))

    # Print dtypes + null counts
    dtypes = df_raw.dtypes.astype(str)
    nulls = df_raw.isna().sum()
    print("\n--- dtypes & nulls ---")
    for c in df_raw.columns:
        print(f"{c:>20}  {dtypes[c]:>12}  nulls={int(nulls[c]):,}")

    # Time diagnostics (without mutating df_raw)
    idx_info = "None"
    median_min = np.nan
    if isinstance(df_raw.index, pd.DatetimeIndex):
        idx_info = f"DatetimeIndex tz={df_raw.index.tz}"
        median_min = _median_interval_minutes(df_raw.index)
    else:
        tcol = _pick_time_col(df_raw)
        if tcol:
            try:
                ts = pd.to_datetime(df_raw[tcol], errors="coerce")
                tz = ts.dt.tz
                idx_info = f"Column '{tcol}' tz={tz}"
                if ts.notna().any():
                    s = ts.dropna().sort_values()
                    if len(s) >= 2:
                        median_min = _median_interval_minutes(pd.DatetimeIndex(s))
            except Exception as e:
                idx_info = f"Found '{tcol}', but parse error: {e}"
        else:
            idx_info = "No clear datetime column found"

    print("\n--- time diagnostics ---")
    print("Time source:", idx_info)
    if isinstance(df_raw.index, pd.DatetimeIndex) and len(df_raw.index) > 0:
        print(f"Index range: {df_raw.index.min()}  →  {df_raw.index.max()}")
    else:
        c = _pick_time_col(df_raw)
        if c:
            try:
                ts = pd.to_datetime(df_raw[c], errors="coerce")
                if ts.notna().any():
                    print(f"{c} range: {ts.min()}  →  {ts.max()}")
            except Exception:
                pass
    print(f"Median interval (minutes): {median_min if not np.isnan(median_min) else 'n/a'}")

    # Preview a few rows exactly as stored
    print("\n--- head(5) raw ---")
    with pd.option_context("display.width", 140, "display.max_columns", None):
        print(df_raw.head(5))
    print("\n--- tail(5) raw ---")
    with pd.option_context("display.width", 140, "display.max_columns", None):
        print(df_raw.tail(5))

    # Write raw CSV EXACTLY AS READ
    df_raw.to_csv(out_csv_raw, index=True)
    print(f"\nWrote RAW CSV (no changes): {out_csv_raw}")

    if parse_ts and out_csv_with_ts:
        df2 = df_raw.copy()
        # If already DatetimeIndex, just copy to a column so it's visible in CSV
        if isinstance(df2.index, pd.DatetimeIndex):
            df2.insert(0, "IndexTimestamp", df2.index)
            ts = df2["IndexTimestamp"]
        else:
            time_col = _pick_time_col(df2)
            if not time_col:
                print("No timestamp column detected; skipping parsed CSV.")
                return
            ts = pd.to_datetime(df2[time_col], errors="coerce")

        # Optional TZ conversion (only affects the parsed column we add)
        if convert_to_tz:
            try:
                if ts.dt.tz is None:
                    # treat naive as UTC, then convert
                    ts = ts.dt.tz_localize("UTC").dt.tz_convert(convert_to_tz)
                else:
                    ts = ts.dt.tz_convert(convert_to_tz)
            except Exception as e:
                print(f"[WARN] TZ conversion failed ({e}); keeping parsed timestamps as-is.")

        df2.insert(1, "ParsedTimestamp", ts)
        df2.to_csv(out_csv_with_ts, index=True)
        print(f"Wrote CSV with parsed timestamp column: {out_csv_with_ts}")

def main():
    ap = argparse.ArgumentParser(description="Inspect a SPY parquet and export to CSV (no strategy).")
    ap.add_argument("--file", type=str, default=None, help="Path to a specific parquet file.")
    ap.add_argument("--folder", type=str, default=DEFAULT_FOLDER, help="Folder to search (if --file not given).")
    ap.add_argument("--out", type=str, default=None, help="Base name for CSV outputs (without extension).")
    ap.add_argument("--parse-ts", action="store_true", help="Also write a CSV adding a parsed timestamp column.")
    ap.add_argument("--tz", type=str, default=None, help="Convert parsed timestamp to this timezone (e.g., 'America/New_York').")
    # Jupyter-friendly (ignores the kernel's -f arg)
    args, _ = ap.parse_known_args()

    path = args.file if args.file else _find_spy_parquet(args.folder)
    base = args.out if args.out else os.path.splitext(os.path.basename(path))[0]
    out_csv_raw = f"{base}__raw.csv"
    out_csv_with_ts = f"{base}__with_ts.csv" if args.parse_ts else None

    inspect_and_export(
        path=path,
        out_csv_raw=out_csv_raw,
        parse_ts=args.parse_ts,
        out_csv_with_ts=out_csv_with_ts,
        convert_to_tz=args.tz
    )

if __name__ == "__main__":
    main()
```

```python
# --------------------------------------------------------------------
# Helpers for robust timezone normalization (paste above load_parquet_15m)
# --------------------------------------------------------------------
def _is_rth_wallclock(idx: pd.DatetimeIndex) -> bool:
    """
    Heuristic: if timestamps are labeled UTC but wall-clock hours mostly look
    like 09:xx–16:xx, they’re probably New York times mislabeled as UTC.
    """
    if not isinstance(idx, pd.DatetimeIndex) or len(idx) == 0:
        return False
    hrs = pd.Series(idx.hour)
    return ((hrs >= 9) & (hrs <= 16)).mean() > 0.80  # lenient

def _reinterpret_as_tz(idx: pd.DatetimeIndex, tz: str) -> pd.DatetimeIndex:
    """Keep the wall-clock (e.g., 09:45 stays 09:45), only change the tz label."""
    if idx.tz is None:
        return idx.tz_localize(tz)
    # drop current tz info (to naive) then localize to target tz
    return idx.tz_convert(None).tz_localize(tz)

def _fix_timezone(idx: pd.DatetimeIndex,
                  assume_naive_are_utc: bool,
                  target_tz: str) -> pd.DatetimeIndex:
    """
    Normalize timestamps to target_tz.

    - If NAIVE:
        * assume_naive_are_utc=True  -> localize UTC, then convert to target_tz (clock shifts)
        * assume_naive_are_utc=False -> localize target_tz (clock does NOT shift)

    - If tz-AWARE:
        * If labeled UTC AND wall-clock looks like RTH (9–16), reinterpret as target_tz (no shift)
        * Else convert to target_tz normally (clock shifts)
    """
    if not isinstance(idx, pd.DatetimeIndex):
        raise TypeError("Expect DatetimeIndex")

    if idx.tz is None:
        return (idx.tz_localize("UTC").tz_convert(target_tz)
                if assume_naive_are_utc else
                idx.tz_localize(target_tz))

    # tz-aware
    if str(idx.tz) in ("UTC", "UTC+00:00", "tzutc()", "pytz.UTC"):
        return _reinterpret_as_tz(idx, target_tz) if _is_rth_wallclock(idx) else idx.tz_convert(target_tz)
    return idx.tz_convert(target_tz)

def _debug_session_bounds(df15: pd.DataFrame, n_days: int = 1):
    """
    Quick sanity print: show first N sessions' start/end and first bar label.
    Helps confirm first 15m ends at 09:45 and we’re within 09:30–16:00.
    """
    if df15.empty or "Session" not in df15.columns:
        return
    shown = 0
    for ses, g in df15.groupby("Session"):
        print(f"[RTH] {ses.date()}  {g.index.min().time()} → {g.index.max().time()}  "
              f"(first15 ends {g.index[0].time()})")
        shown += 1
        if shown >= n_days:
            break

# --------------------------------------------------------------------
# REPLACE your current load_parquet_15m() with this version
# --------------------------------------------------------------------
def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()

    # --- Robust timezone normalization (handles "mislabeled UTC" wall-clock) ---
    df.index = _fix_timezone(df.index, ASSUME_NAIVE_TIMESTAMPS_ARE_UTC, tz)

    # Standardize OHLCV names
    df = _standardize_ohlcv_columns(df)
    for need in ["Open", "High", "Low", "Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    # Resample to exact 15m if needed (right-closed, right-labeled -> first bar is 09:45)
    df = _resample_to_15m_if_needed(df)
    if df.empty:
        return df

    # Restrict to regular trading hours (RTH)
    df = df.between_time(REG_SESSION_START, REG_SESSION_END).sort_index()

    # Add Session (date) for downstream logic
    df["Session"] = pd.to_datetime(df.index.date)

    # One-line sanity print so you can SEE it’s inside 09:30–16:00 & first15=09:45
    _debug_session_bounds(df, n_days=1)

    return df
```

```python
# orb_retest_spy.py
# SPY 15m ORB breakout with entry on retest — robust parquet loader + charting

import warnings
warnings.filterwarnings("ignore")

import os, glob, argparse, ast, time
from datetime import timedelta
from typing import Optional, List, Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.dates import MinuteLocator, DateFormatter
from matplotlib.patches import Rectangle

# =============================
# USER SETTINGS (edit as needed)
# =============================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# Strategy
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN       = 120
R_MULTIPLES          = [1.0, 2.0]
POSITION_SIZE_DOLLARS= 10_000
SLIPPAGE_BPS         = 1.0
FEES_PER_TRADE       = 0.00

# IO
TRADES_CSV  = "trades_spy_orb.csv"
EQUITY_CSV  = "equity_spy_orb.csv"
CHARTS_DIR  = "charts"

# =============================
# Column normalize
# =============================
STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                      rename[lower[k]] = std
        elif k.capitalize() in df.columns:  rename[k.capitalize()] = std
        elif k.upper() in df.columns:       rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _median_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    return float(np.median((idx[1:] - idx[:-1]).total_seconds()/60.0))

# =============================
# Timezone handling (robust)
# =============================
def _is_rth_wallclock(idx: pd.DatetimeIndex) -> bool:
    if not isinstance(idx, pd.DatetimeIndex) or len(idx) == 0:
        return False
    hrs = pd.Series(idx.hour)
    return ((hrs >= 9) & (hrs <= 16)).mean() > 0.80

def _reinterpret_as_tz(idx: pd.DatetimeIndex, tz: str) -> pd.DatetimeIndex:
    if idx.tz is None:
        return idx.tz_localize(tz)
    return idx.tz_convert(None).tz_localize(tz)

def _fix_timezone(idx: pd.DatetimeIndex,
                  assume_naive_are_utc: bool,
                  target_tz: str) -> pd.DatetimeIndex:
    if idx.tz is None:
        return (idx.tz_localize("UTC").tz_convert(target_tz)
                if assume_naive_are_utc else
                idx.tz_localize(target_tz))
    if str(idx.tz) in ("UTC", "UTC+00:00", "tzutc()", "pytz.UTC"):
        return _reinterpret_as_tz(idx, target_tz) if _is_rth_wallclock(idx) else idx.tz_convert(target_tz)
    return idx.tz_convert(target_tz)

def _debug_session_bounds(df15: pd.DataFrame, n_days: int = 1):
    if df15.empty or "Session" not in df15.columns:
        return
    shown = 0
    for ses, g in df15.groupby("Session"):
        print(f"[RTH] {ses.date()}  {g.index.min().time()} → {g.index.max().time()}  "
              f"(first15 ends {g.index[0].time()})")
        shown += 1
        if shown >= n_days:
            break

# =============================
# Resample
# =============================
def _resample_to_15m_if_needed(df: pd.DataFrame) -> pd.DataFrame:
    m = _median_minutes(df.index)
    if np.isnan(m): return pd.DataFrame()
    if m < 15 - 1e-6:
        agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
        if "Volume" in df.columns: agg["Volume"] = "sum"
        out = (df.sort_index()
                 .resample("15T", label="right", closed="right")
                 .agg(agg)
                 .dropna(subset=["Open","High","Low","Close"]))
        return out
    elif abs(m - 15) <= 1e-6:
        return df.sort_index()
    else:
        return pd.DataFrame()

# =============================
# Loader
# =============================
def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()
    df.index = _fix_timezone(df.index, ASSUME_NAIVE_TIMESTAMPS_ARE_UTC, tz)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    df = _resample_to_15m_if_needed(df)
    if df.empty:
        return df

    df = df.between_time(REG_SESSION_START, REG_SESSION_END).sort_index()
    df["Session"] = pd.to_datetime(df.index.date)

    _debug_session_bounds(df, n_days=1)
    return df

def _find_spy_file(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    if not files:
        return ""
    cands = [f for f in files if "spy" in os.path.basename(f).lower()]
    if cands:
        exact = [f for f in cands if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(cands)[0]
    # fallback: first parquet
    return sorted(files)[0]

# =============================
# ORB logic
# =============================
def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()
    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan
    df["Session"] = sessionize(df)
    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    trades = []
    for ses, sdf in df.groupby("Session"):
        if len(sdf) < 3: continue
        orh = float(sdf["ORH"].iloc[0]); orl = float(sdf["ORL"].iloc[0])
        if pd.isna(orh) or pd.isna(orl): continue

        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)
        if long_break.empty and short_break.empty: continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction == "long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty: continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty: continue

        retest_bar = touch.iloc[0]; retest_time = retest_bar.name
        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok: continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop  = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12: continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1
        t_prices = [(entry + r*risk_per_share) if direction == "long" else (entry - r*risk_per_share) for r in r_targets]

        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            hi, lo = float(row["High"]), float(row["Low"])
            if lo <= stop <= hi:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px)); break
            hit = False
            for i, tp in enumerate(t_prices):
                if lo <= tp <= hi:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px)); hit = True
            if hit: break

        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(float(last["Close"]), slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        n_parts = len(exits); qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _,_,px in exits) - FEES_PER_TRADE

        trades.append({
            "Ticker": "SPY",
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": float(entry), "Stop": float(stop),
            "Targets": [float(x) for x in t_prices],
            "Exits": [(str(t[0]), t[1], float(t[2])) for t in exits],
            "Qty": float(qty),
            "PnL_$": float(cash_pnl),
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Plotting
# =============================
def _parse_exits(exits_field):
    if isinstance(exits_field, list) and exits_field:
        tag, ts, px = exits_field[0]
        return tag, pd.to_datetime(ts), float(px)
    if isinstance(exits_field, str):
        try:
            first = exits_field.split(";")[0]
            tag, ts, px = first.split("|")
            return tag, pd.to_datetime(ts), float(px)
        except Exception:
            pass
    try:
        v = ast.literal_eval(exits_field)
        if isinstance(v, list) and v:
            tag, ts, px = v[0]
            return tag, pd.to_datetime(ts), float(px)
    except Exception:
        return None, None, None
    return None, None, None

def _draw_candles(ax, df_day: pd.DataFrame, width_minutes=10):
    width = (width_minutes / (24*60))
    times = mdates.date2num(df_day.index.to_pydatetime())
    for i, (t, row) in enumerate(df_day.iterrows()):
        o, h, l, c = float(row["Open"]), float(row["High"]), float(row["Low"]), float(row["Close"])
        # wick
        ax.plot([times[i], times[i]], [l, h], linewidth=1.0)
        # body (outline only; no explicit colors)
        y = min(o, c); height = abs(c - o) or 1e-6
        rect = Rectangle((times[i] - width/2.0, y), width, height, fill=False)
        ax.add_patch(rect)

def plot_trade(ax, df_day: pd.DataFrame, trade_row: pd.Series, title_prefix="SPY"):
    _draw_candles(ax, df_day)
    ax.set_title(f"{title_prefix}  {pd.to_datetime(trade_row['Session']).date()}")

    # 15-minute ticks + HH:MM labels
    ax.set_xlim(df_day.index.min(), df_day.index.max())
    ax.xaxis.set_major_locator(MinuteLocator(byminute=[0, 15, 30, 45]))
    try:
        ax.xaxis.set_major_formatter(DateFormatter("%H:%M", tz=df_day.index.tz))
    except TypeError:
        ax.xaxis.set_major_formatter(DateFormatter("%H:%M"))
    ax.grid(True, linestyle=":", linewidth=0.6)

    # OR lines
    orh = float(trade_row["ORH"]); orl = float(trade_row["ORL"])
    t0, t1 = df_day.index.min(), df_day.index.max()
    ax.hlines(orh, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)
    ax.hlines(orl, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)

    # Entry (line + vertical marker + time label)
    entry_px = float(trade_row["Entry"])
    entry_ts = pd.to_datetime(trade_row["EntryTime"])
    ax.hlines(entry_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="solid", linewidth=1.2)
    ax.vlines(mdates.date2num(entry_ts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(),
              linestyles="dotted", linewidth=0.9)
    lbl_time = entry_ts.tz_convert(TZ).strftime('%H:%M') if entry_ts.tzinfo else entry_ts.strftime('%H:%M')
    ax.annotate(f"Entry: {entry_px:.2f}\n{lbl_time}",
                xy=(mdates.date2num(entry_ts), entry_px),
                xytext=(10, 10), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    # Stop
    stop_px = float(trade_row["Stop"])
    ax.hlines(stop_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashdot", linewidth=1.0)
    ax.annotate(f"Stop: {stop_px:.2f}", xy=(mdates.date2num(t0), stop_px),
                xytext=(5, -12), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    # First realized exit (marker + label)
    tag, ex_ts, ex_px = _parse_exits(trade_row["Exits"])
    if tag is not None:
        ax.scatter([mdates.date2num(ex_ts)], [ex_px], s=25)
        ax.annotate(f"Exit: {ex_px:.2f} ({tag})",
                    xy=(mdates.date2num(ex_ts), ex_px),
                    xytext=(10, -15), textcoords="offset points",
                    fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    # PnL label
    pnl = float(trade_row["PnL_$"])
    ax.text(0.99, 0.02, f"PnL: ${pnl:,.2f}", transform=ax.transAxes,
            ha="right", va="bottom", fontsize=9, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

# =============================
# Runner + CLI
# =============================
def parse_args():
    p = argparse.ArgumentParser(description="SPY 15m ORB (retest) from parquet + charts")
    p.add_argument("--parquet-dir", type=str, default=PARQUET_DIR, help="Folder with SPY parquet.")
    p.add_argument("--outdir",      type=str, default=CHARTS_DIR,   help="Charts output directory.")
    p.add_argument("--dpi",         type=int, default=170,          help="Chart DPI.")
    p.add_argument("--limit",       type=int, default=-1,           help="Max charts (use -1 for all).")
    p.add_argument("--start",       type=str, default=None,         help="Start date YYYY-MM-DD (optional).")
    p.add_argument("--end",         type=str, default=None,         help="End date YYYY-MM-DD (optional).")
    p.add_argument("--date",        type=str, default=None,         help="Plot only this YYYY-MM-DD session.")
    p.add_argument("--skip-existing", action="store_true",          help="Skip PNGs that already exist.")
    # ignore unknown (Jupyter's -f)
    args, _ = p.parse_known_args()
    return args

def main():
    args = parse_args()
    os.makedirs(args.outdir, exist_ok=True)

    spy_path = _find_spy_file(args.parquet_dir)
    if not spy_path:
        raise FileNotFoundError(f"No parquet in {args.parquet_dir}")

    print(f"Loading: {spy_path}")
    df15 = load_parquet_15m(spy_path, tz=TZ)
    if df15.empty:
        print("No usable 15m data after normalization/resample/session filter.")
        return
    df15["Ticker"] = "SPY"
    df15 = compute_opening_range(df15)

    print("Running ORB retest backtest…")
    trades, equity = backtest_orb_retest(df15)

    if trades.empty:
        print("No trades generated with current settings.")
        return

    trades.to_csv(TRADES_CSV, index=False)
    print(f"Saved trades: {TRADES_CSV}")

    if not equity.empty:
        equity.to_csv(EQUITY_CSV, index=False)
        print(f"Saved equity: {EQUITY_CSV}")

    # Render charts
    tlog = trades.copy()
    tlog["Session"] = pd.to_datetime(tlog["Session"])
    if args.date:
        day = pd.to_datetime(args.date).normalize()
        tlog = tlog[tlog["Session"] == day]
    else:
        if args.start: tlog = tlog[tlog["Session"] >= pd.to_datetime(args.start)]
        if args.end:   tlog = tlog[tlog["Session"] <= pd.to_datetime(args.end)]
    tlog.sort_values(["Session","EntryTime"], inplace=True)

    printed, total = 0, len(tlog)
    for _, tr in tlog.iterrows():
        ses = tr["Session"].date()
        direction = str(tr.get("Direction",""))
        fname = f"SPY_{ses}_{direction}.png"
        out_path = os.path.join(args.outdir, fname)
        if args.skip_existing and os.path.exists(out_path):
            printed += 1
            continue

        df_day = df15[df15["Session"] == pd.to_datetime(ses)].copy()
        if df_day.empty:
            continue

        fig, ax = plt.subplots(figsize=(12, 6))
        plot_trade(ax, df_day, tr, title_prefix="SPY")
        fig.autofmt_xdate()
        fig.tight_layout()
        fig.savefig(out_path, dpi=args.dpi)
        plt.close(fig)
        print(f"Wrote {out_path}")
        printed += 1
        if (args.limit != -1) and (printed >= args.limit):
            break

    print(f"Done. Rendered {printed}/{total} charts.")

if __name__ == "__main__":
    main()
```

```python
# orb_retest_spy.py
# SPY 15m ORB breakout with entry on retest — parquet loader, backtest, charts

import warnings
warnings.filterwarnings("ignore")

import os, glob, argparse, ast
from datetime import timedelta
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.dates import MinuteLocator, DateFormatter
from matplotlib.patches import Rectangle

# =============================
# USER SETTINGS
# =============================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# Strategy — EXACT rule requested:
# Breakout: bar CLOSE beyond ORH/ORL
# Entry:    first retest TOUCH of that level (no close confirmation)
# Timeout:  None (wait the rest of the session)
MAX_RETEST_MIN: Optional[int] = None
R_MULTIPLES           = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS          = 1.0
FEES_PER_TRADE        = 0.00

# Output
TRADES_CSV = "trades_spy_orb.csv"
EQUITY_CSV = "equity_spy_orb.csv"
CHARTS_DIR = "charts"

# =============================
# Column normalize
# =============================
STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                      rename[lower[k]] = std
        elif k.capitalize() in df.columns:  rename[k.capitalize()] = std
        elif k.upper() in df.columns:       rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _median_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    return float(np.median((idx[1:] - idx[:-1]).total_seconds()/60.0))

# =============================
# Timezone handling
# =============================
def _is_rth_wallclock(idx: pd.DatetimeIndex) -> bool:
    if not isinstance(idx, pd.DatetimeIndex) or len(idx) == 0:
        return False
    hrs = pd.Series(idx.hour)
    return ((hrs >= 9) & (hrs <= 16)).mean() > 0.80

def _reinterpret_as_tz(idx: pd.DatetimeIndex, tz: str) -> pd.DatetimeIndex:
    if idx.tz is None:
        return idx.tz_localize(tz)
    return idx.tz_convert(None).tz_localize(tz)

def _fix_timezone(idx: pd.DatetimeIndex,
                  assume_naive_are_utc: bool,
                  target_tz: str) -> pd.DatetimeIndex:
    if idx.tz is None:
        return (idx.tz_localize("UTC").tz_convert(target_tz)
                if assume_naive_are_utc else
                idx.tz_localize(target_tz))
    if str(idx.tz) in ("UTC", "UTC+00:00", "tzutc()", "pytz.UTC"):
        return _reinterpret_as_tz(idx, target_tz) if _is_rth_wallclock(idx) else idx.tz_convert(target_tz)
    return idx.tz_convert(target_tz)

# =============================
# Resample — force right-labeled 15m (first label 09:45)
# =============================
def _resample_to_15m_if_needed(df: pd.DataFrame) -> pd.DataFrame:
    """
    Normalize to *right-labeled, right-closed* 15-minute bars so the first
    regular-session bar is labeled 09:45 (covering 09:30–09:45).
    """
    m = _median_minutes(df.index)
    if np.isnan(m):
        return pd.DataFrame()

    if m < 15 - 1e-6:
        # Finer data → aggregate to true 15m, right-labeled
        agg = {"Open":"first", "High":"max", "Low":"min", "Close":"last"}
        if "Volume" in df.columns:
            agg["Volume"] = "sum"
        out = (df.sort_index()
                 .resample("15T", label="right", closed="right")
                 .agg(agg)
                 .dropna(subset=["Open","High","Low","Close"]))
        return out

    elif abs(m - 15) <= 1e-6:
        dfx = df.sort_index().copy()
        # Detect left-labeled 15m by sampling many sessions (first RTH stamp per day)
        probe = dfx.between_time("09:00", "10:00")
        if not probe.empty:
            firsts = probe.groupby(probe.index.date).head(1).index
            if len(firsts) >= 3 and (pd.Series(firsts).map(lambda t: t.minute == 30).mean() > 0.5):
                dfx.index = dfx.index + pd.Timedelta(minutes=15)
        return dfx

    else:
        # Coarser than 15m not supported
        return pd.DataFrame()

# =============================
# Loader
# =============================
def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()
    df.index = _fix_timezone(df.index, ASSUME_NAIVE_TIMESTAMPS_ARE_UTC, tz)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    df = _resample_to_15m_if_needed(df)
    if df.empty:
        return df

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END).sort_index()
    df["Session"] = pd.to_datetime(df.index.date)

    # quick sanity print for first RTH session window
    ses0 = df["Session"].iloc[0]
    g = df[df["Session"] == ses0]
    print(f"[RTH] {ses0.date()}  {g.index.min().time()} → {g.index.max().time()}  (first 15m label {g.index[0].time()})")
    return df

def _find_spy_file(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    if not files:
        return ""
    cands = [f for f in files if "spy" in os.path.basename(f).lower()]
    if cands:
        exact = [f for f in cands if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(cands)[0]
    return sorted(files)[0]

# =============================
# ORB: opening range from first 15m bar
# =============================
def compute_opening_range(df_15: pd.DataFrame, debug_first_n: int = 3) -> pd.DataFrame:
    """
    Opening Range = High/Low of the *first* 15m RTH bar (should be labeled 09:45).
    Levels are constant for the rest of the session.
    """
    df = df_15.copy()
    df["Session"] = pd.to_datetime(df.index.date)

    df["ORH"] = np.nan
    df["ORL"] = np.nan

    shown = 0
    for ses, g in df.groupby("Session"):
        g = g.sort_index()
        first_ts = g.index[0]
        orh = float(g["High"].iloc[0])
        orl = float(g["Low"].iloc[0])

        df.loc[g.index[0], "ORH"] = orh
        df.loc[g.index[0], "ORL"] = orl
        df.loc[g.index, ["ORH","ORL"]] = df.loc[g.index, ["ORH","ORL"]].ffill()

        if shown < debug_first_n:
            print(f"[OR] {ses.date()}  first bar label={first_ts.time()}  ORH={orh:.4f}  ORL={orl:.4f}")
            shown += 1

    return df

def assert_first_bar_is_0945(df15: pd.DataFrame, max_warn: int = 5):
    warns = 0
    for ses, g in df15.groupby(pd.to_datetime(df15.index.date)):
        t = g.index[0]
        if (t.hour, t.minute) != (9, 45):
            print(f"[WARN] {ses.date()} first label is {t.time()} (expected 09:45). Using that bar for OR anyway.")
            warns += 1
            if warns >= max_warn:
                break

# =============================
# Backtest
# =============================
def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: Optional[int] = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Breakout = first bar *close* beyond ORH/ORL.
    Entry    = first *touch* retest of the level (no close confirmation).
    Timeout  = None (valid until EOD), unless max_retest_min is set.
    Stop     = opposite OR. Targets at R multiples; stop has priority.
    """
    df = df_15.copy()
    df["Session"] = pd.to_datetime(df.index.date)
    trades = []

    for ses, sdf in df.groupby("Session"):
        if len(sdf) < 3:
            continue

        orh = float(sdf["ORH"].iloc[0]); orl = float(sdf["ORL"].iloc[0])
        if pd.isna(orh) or pd.isna(orl):
            continue

        # --- Breakout: bar CLOSE beyond OR ---
        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction == "long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        breakout_time = breakout_row.name

        # --- Retest window: whole session (or until cutoff) ---
        if max_retest_min is None:
            after_break = sdf[sdf.index > breakout_time].copy()
        else:
            cutoff = breakout_time + timedelta(minutes=int(max_retest_min))
            after_break = sdf[(sdf.index > breakout_time) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl

        # --- Retest: enter on TOUCH ---
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar  = touch.iloc[0]
        retest_time = retest_bar.name

        # --- Entry/stop/targets ---
        entry = _apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop  = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1
        t_prices = [(entry + r*risk_per_share) if direction == "long" else (entry - r*risk_per_share) for r in r_targets]

        # --- Forward simulate exits ---
        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = float(row["High"]), float(row["Low"])
            # Stop first
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets (first hit wins)
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(float(last["Close"]), slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        # Equal partials among realized exits
        n_parts = len(exits); qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _,_,px in exits) - FEES_PER_TRADE
        r_multiple = cash_pnl / (risk_per_share * qty)

        trades.append({
            "Ticker": "SPY",
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "BreakoutTime": breakout_time,
            "EntryTime": retest_time,
            "Entry": float(entry), "Stop": float(stop),
            "Targets": [float(x) for x in t_prices],
            "Exits": [(str(t[0]), t[1], float(t[2])) for t in exits],
            "Qty": float(qty),
            "PnL_$": float(cash_pnl),
            "R_multiple": float(r_multiple),
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Plotting
# =============================
def _parse_exits(exits_field):
    if isinstance(exits_field, list) and exits_field:
        tag, ts, px = exits_field[0]
        return tag, pd.to_datetime(ts), float(px)
    if isinstance(exits_field, str):
        try:
            first = exits_field.split(";")[0]
            tag, ts, px = first.split("|")
            return tag, pd.to_datetime(ts), float(px)
        except Exception:
            pass
    try:
        v = ast.literal_eval(exits_field)
        if isinstance(v, list) and v:
            tag, ts, px = v[0]
            return tag, pd.to_datetime(ts), float(px)
    except Exception:
        return None, None, None
    return None, None, None

def _draw_candles(ax, df_day: pd.DataFrame, width_minutes=10):
    width = (width_minutes / (24*60))
    times = mdates.date2num(df_day.index.to_pydatetime())
    for i, (_, row) in enumerate(df_day.iterrows()):
        o, h, l, c = float(row["Open"]), float(row["High"]), float(row["Low"]), float(row["Close"])
        ax.plot([times[i], times[i]], [l, h], linewidth=1.0)
        y = min(o, c); height = abs(c - o) or 1e-6
        rect = Rectangle((times[i] - width/2.0, y), width, height, fill=False)
        ax.add_patch(rect)

def plot_trade(ax, df_day: pd.DataFrame, trade_row: pd.Series, title_prefix="SPY"):
    _draw_candles(ax, df_day)
    ax.set_title(f"{title_prefix}  {pd.to_datetime(trade_row['Session']).date()}")

    ax.set_xlim(df_day.index.min(), df_day.index.max())
    ax.xaxis.set_major_locator(MinuteLocator(byminute=[0, 15, 30, 45]))
    try:
        ax.xaxis.set_major_formatter(DateFormatter("%H:%M", tz=df_day.index.tz))
    except TypeError:
        ax.xaxis.set_major_formatter(DateFormatter("%H:%M"))
    ax.grid(True, linestyle=":", linewidth=0.6)

    orh = float(trade_row["ORH"]); orl = float(trade_row["ORL"])
    t0, t1 = df_day.index.min(), df_day.index.max()
    ax.hlines(orh, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)
    ax.hlines(orl, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)

    # breakout marker
    if "BreakoutTime" in trade_row and pd.notna(trade_row["BreakoutTime"]):
        bts = pd.to_datetime(trade_row["BreakoutTime"])
        ax.vlines(mdates.date2num(bts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(),
                  linestyles="dashed", linewidth=0.8)
        ax.annotate("Breakout", xy=(mdates.date2num(bts), df_day["High"].max()),
                    xytext=(5, -15), textcoords="offset points", fontsize=8,
                    bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    # entry marker
    entry_px = float(trade_row["Entry"])
    entry_ts = pd.to_datetime(trade_row["EntryTime"])
    ax.hlines(entry_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="solid", linewidth=1.2)
    ax.vlines(mdates.date2num(entry_ts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(),
              linestyles="dotted", linewidth=0.9)
    lbl_time = entry_ts.tz_convert(TZ).strftime('%H:%M') if entry_ts.tzinfo else entry_ts.strftime('%H:%M')
    ax.annotate(f"Entry: {entry_px:.2f}\n{lbl_time}",
                xy=(mdates.date2num(entry_ts), entry_px),
                xytext=(10, 10), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    # stop
    stop_px = float(trade_row["Stop"])
    ax.hlines(stop_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashdot", linewidth=1.0)
    ax.annotate(f"Stop: {stop_px:.2f}", xy=(mdates.date2num(t0), stop_px),
                xytext=(5, -12), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    # first realized exit
    tag, ex_ts, ex_px = _parse_exits(trade_row["Exits"])
    if tag is not None:
        ax.scatter([mdates.date2num(ex_ts)], [ex_px], s=25)
        ax.annotate(f"Exit: {ex_px:.2f} ({tag})",
                    xy=(mdates.date2num(ex_ts), ex_px),
                    xytext=(10, -15), textcoords="offset points",
                    fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    pnl = float(trade_row["PnL_$"])
    ax.text(0.99, 0.02, f"PnL: ${pnl:,.2f}", transform=ax.transAxes,
            ha="right", va="bottom", fontsize=9, bbox=dict(boxstyle="round", fc="w", alpha=0.4))
    ax.text(0.01, 0.98, "UpdatedCharts", transform=ax.transAxes,
            ha="left", va="top", fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.3))

# =============================
# Runner + CLI
# =============================
def parse_args():
    p = argparse.ArgumentParser(description="SPY 15m ORB (retest) from parquet + charts")
    p.add_argument("--parquet-dir", type=str, default=PARQUET_DIR, help="Folder with SPY parquet.")
    p.add_argument("--outdir",      type=str, default=CHARTS_DIR,   help="Charts output directory.")
    p.add_argument("--dpi",         type=int, default=170,          help="Chart DPI.")
    p.add_argument("--limit",       type=int, default=-1,           help="Max charts (use -1 for all).")
    p.add_argument("--start",       type=str, default=None,         help="Start date YYYY-MM-DD (optional).")
    p.add_argument("--end",         type=str, default=None,         help="End date YYYY-MM-DD (optional).")
    p.add_argument("--date",        type=str, default=None,         help="Plot only this YYYY-MM-DD session.")
    args, _ = p.parse_known_args()  # ignore Jupyter's -f
    return args

def main():
    args = parse_args()
    os.makedirs(args.outdir, exist_ok=True)

    spy_path = _find_spy_file(args.parquet_dir)
    if not spy_path:
        raise FileNotFoundError(f"No parquet in {args.parquet_dir}")

    print(f"Loading: {spy_path}")
    df15 = load_parquet_15m(spy_path, tz=TZ)
    if df15.empty:
        print("No usable 15m data after normalization/resample/session filter.")
        return

    # Assert first bar is 09:45 (warn if not), then compute OR
    assert_first_bar_is_0945(df15)
    df15["Ticker"] = "SPY"
    df15 = compute_opening_range(df15)

    print("Running ORB retest backtest…")
    trades, equity = backtest_orb_retest(df15)

    if trades.empty:
        print("No trades generated with current settings.")
        return

    trades.to_csv(TRADES_CSV, index=False)
    print(f"Saved trades: {TRADES_CSV}")

    if not equity.empty:
        equity.to_csv(EQUITY_CSV, index=False)
        print(f"Saved equity: {EQUITY_CSV}")

    # --- Overall Totals (SPY) ---
    total_trades = len(trades)
    total_pnl    = trades["PnL_$"].sum()
    win_rate     = (trades["PnL_$"] > 0).mean() * 100.0
    avg_r        = trades["R_multiple"].mean()
    median_r     = trades["R_multiple"].median()
    print("\n=== Overall Totals (SPY) ===")
    print(f"Total trades: {total_trades}")
    print(f"Total PnL ($): {total_pnl:,.2f}")
    print(f"Win rate: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")

    # --- Render charts (ALWAYS regenerate; add UpdatedCharts to filename) ---
    tlog = trades.copy()
    tlog["Session"] = pd.to_datetime(tlog["Session"])
    if args.date:
        day = pd.to_datetime(args.date).normalize()
        tlog = tlog[tlog["Session"] == day]
    else:
        if args.start: tlog = tlog[tlog["Session"] >= pd.to_datetime(args.start)]
        if args.end:   tlog = tlog[tlog["Session"] <= pd.to_datetime(args.end)]
    tlog.sort_values(["Session","EntryTime"], inplace=True)

    printed, total = 0, len(tlog)
    for _, tr in tlog.iterrows():
        ses = tr["Session"].date()
        direction = str(tr.get("Direction",""))
        fname = f"SPY_UpdatedCharts_{ses}_{direction}.png"
        out_path = os.path.join(args.outdir, fname)

        df_day = df15[df15["Session"] == pd.to_datetime(ses)].copy()
        if df_day.empty:
            continue

        fig, ax = plt.subplots(figsize=(12, 6))
        plot_trade(ax, df_day, tr, title_prefix="SPY")
        fig.autofmt_xdate()
        fig.tight_layout()
        fig.savefig(out_path, dpi=args.dpi)
        plt.close(fig)
        print(f"Wrote {out_path}")
        printed += 1
        if (args.limit != -1) and (printed >= args.limit):
            break

    print(f"Done. Rendered {printed}/{total} charts.")

if __name__ == "__main__":
    main()
```

```python
# orb_retest_spy.py
# SPY 15m ORB breakout with entry on retest — robust parquet loader + charting
import warnings
warnings.filterwarnings("ignore")

import os, glob, argparse, ast
from datetime import timedelta
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.dates import MinuteLocator, DateFormatter
from matplotlib.patches import Rectangle

# =============================
# USER SETTINGS
# =============================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# Strategy
# --- Set to match the "dark chart" behavior ---
BREAKOUT_RULE = "highlow"      # "close" (bar close beyond OR) or "highlow" (intrabar High/Low beyond OR)
RETEST_RULE   = "touch"        # "close" (retest bar must close back through) or "touch" (first touch is enough)
MAX_RETEST_MIN: Optional[int] = None   # None = wait to EOD; or set minutes (e.g., 120)

R_MULTIPLES          = [1.0, 2.0]
POSITION_SIZE_DOLLARS= 10_000
SLIPPAGE_BPS         = 1.0
FEES_PER_TRADE       = 0.00

# IO
TRADES_CSV  = "trades_spy_orb.csv"
EQUITY_CSV  = "equity_spy_orb.csv"
CHARTS_DIR  = "charts"

# =============================
# Column normalize
# =============================
STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                      rename[lower[k]] = std
        elif k.capitalize() in df.columns:  rename[k.capitalize()] = std
        elif k.upper() in df.columns:       rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _median_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    return float(np.median((idx[1:] - idx[:-1]).total_seconds()/60.0))

# =============================
# Timezone handling (robust)
# =============================
def _is_rth_wallclock(idx: pd.DatetimeIndex) -> bool:
    if not isinstance(idx, pd.DatetimeIndex) or len(idx) == 0:
        return False
    hrs = pd.Series(idx.hour)
    return ((hrs >= 9) & (hrs <= 16)).mean() > 0.80

def _reinterpret_as_tz(idx: pd.DatetimeIndex, tz: str) -> pd.DatetimeIndex:
    if idx.tz is None:
        return idx.tz_localize(tz)
    return idx.tz_convert(None).tz_localize(tz)

def _fix_timezone(idx: pd.DatetimeIndex,
                  assume_naive_are_utc: bool,
                  target_tz: str) -> pd.DatetimeIndex:
    if idx.tz is None:
        return (idx.tz_localize("UTC").tz_convert(target_tz)
                if assume_naive_are_utc else
                idx.tz_localize(target_tz))
    if str(idx.tz) in ("UTC", "UTC+00:00", "tzutc()", "pytz.UTC"):
        return _reinterpret_as_tz(idx, target_tz) if _is_rth_wallclock(idx) else idx.tz_convert(target_tz)
    return idx.tz_convert(target_tz)

# =============================
# Resample — force right-labeled 15m (first label 09:45)
# =============================
def _resample_to_15m_if_needed(df: pd.DataFrame) -> pd.DataFrame:
    """
    Normalize bars to *right-labeled, right-closed* 15-minute intervals
    so the first regular-session bar is labeled 09:45.
    """
    m = _median_minutes(df.index)
    if np.isnan(m):
        return pd.DataFrame()

    if m < 15 - 1e-6:
        agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
        if "Volume" in df.columns:
            agg["Volume"] = "sum"
        out = (df.sort_index()
                 .resample("15T", label="right", closed="right")
                 .agg(agg)
                 .dropna(subset=["Open","High","Low","Close"]))
        return out

    elif abs(m - 15) <= 1e-6:
        dfx = df.sort_index().copy()
        probe = dfx.between_time("09:00", "10:00")
        if not probe.empty:
            first_stamp = probe.index[0]
            if first_stamp.minute == 30:
                # left-labeled → shift forward to make right-labeled
                dfx.index = dfx.index + pd.Timedelta(minutes=15)
        return dfx

    else:
        return pd.DataFrame()

# =============================
# Loader
# =============================
def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()
    df.index = _fix_timezone(df.index, ASSUME_NAIVE_TIMESTAMPS_ARE_UTC, tz)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    df = _resample_to_15m_if_needed(df)
    if df.empty:
        return df

    df = df.between_time(REG_SESSION_START, REG_SESSION_END).sort_index()
    df["Session"] = pd.to_datetime(df.index.date)

    # quick sanity print for first day
    ses0 = df["Session"].iloc[0]
    g = df[df["Session"] == ses0]
    print(f"[RTH] {ses0.date()}  {g.index.min().time()} → {g.index.max().time()}  (first 15m label {g.index[0].time()})")
    return df

def _find_spy_file(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    if not files:
        return ""
    cands = [f for f in files if "spy" in os.path.basename(f).lower()]
    if cands:
        exact = [f for f in cands if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(cands)[0]
    return sorted(files)[0]

# =============================
# ORB logic
# =============================
def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()
    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan
    df["Session"] = sessionize(df)
    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: Optional[int] = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    breakout_rule: str = BREAKOUT_RULE,
    retest_rule: str = RETEST_RULE
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    trades = []

    for ses, sdf in df.groupby("Session"):
        if len(sdf) < 3:
            continue

        orh = float(sdf["ORH"].iloc[0]); orl = float(sdf["ORL"].iloc[0])
        if pd.isna(orh) or pd.isna(orl):
            continue

        # --- breakout selection (first after opening bar) ---
        sdf_after = sdf.iloc[1:].copy()
        if breakout_rule.lower() == "highlow":
            long_break  = sdf_after[sdf_after["High"] > orh].head(1)
            short_break = sdf_after[sdf_after["Low"]  < orl].head(1)
        else:  # "close"
            long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
            short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction == "long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        breakout_time = breakout_row.name

        # --- retest search window ---
        if max_retest_min is None:
            after_break = sdf[sdf.index > breakout_time].copy()
        else:
            cutoff = breakout_time + timedelta(minutes=int(max_retest_min))
            after_break = sdf[(sdf.index > breakout_time) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl

        # --- retest detection ---
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar  = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_rule.lower() == "close":
            ok = (retest_bar["Close"] >= level) if direction == "long" else (retest_bar["Close"] <= level)
            if not ok:
                continue
        # "touch" requires no extra confirmation

        # --- entry/stop/targets ---
        entry = _apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop  = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1
        t_prices = [(entry + r*risk_per_share) if direction == "long" else (entry - r*risk_per_share) for r in r_targets]

        # --- forward simulate exits ---
        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = float(row["High"]), float(row["Low"])
            # Stop first
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets (first hit wins)
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            # EOD flatten on last bar close
            last = sdf_run.iloc[-1]
            px = _apply_slippage(float(last["Close"]), slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        # Equal partials among realized exits
        n_parts = len(exits); qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _,_,px in exits) - FEES_PER_TRADE
        r_multiple = cash_pnl / (risk_per_share * qty)

        trades.append({
            "Ticker": "SPY",
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "BreakoutTime": breakout_time,
            "EntryTime": retest_time,
            "Entry": float(entry), "Stop": float(stop),
            "Targets": [float(x) for x in t_prices],
            "Exits": [(str(t[0]), t[1], float(t[2])) for t in exits],
            "Qty": float(qty),
            "PnL_$": float(cash_pnl),
            "R_multiple": float(r_multiple),
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Plotting
# =============================
def _parse_exits(exits_field):
    if isinstance(exits_field, list) and exits_field:
        tag, ts, px = exits_field[0]
        return tag, pd.to_datetime(ts), float(px)
    if isinstance(exits_field, str):
        try:
            first = exits_field.split(";")[0]
            tag, ts, px = first.split("|")
            return tag, pd.to_datetime(ts), float(px)
        except Exception:
            pass
    try:
        v = ast.literal_eval(exits_field)
        if isinstance(v, list) and v:
            tag, ts, px = v[0]
            return tag, pd.to_datetime(ts), float(px)
    except Exception:
        return None, None, None
    return None, None, None

def _draw_candles(ax, df_day: pd.DataFrame, width_minutes=10):
    width = (width_minutes / (24*60))
    times = mdates.date2num(df_day.index.to_pydatetime())
    for i, (t, row) in enumerate(df_day.iterrows()):
        o, h, l, c = float(row["Open"]), float(row["High"]), float(row["Low"]), float(row["Close"])
        ax.plot([times[i], times[i]], [l, h], linewidth=1.0)
        y = min(o, c); height = abs(c - o) or 1e-6
        rect = Rectangle((times[i] - width/2.0, y), width, height, fill=False)
        ax.add_patch(rect)

def plot_trade(ax, df_day: pd.DataFrame, trade_row: pd.Series, title_prefix="SPY"):
    _draw_candles(ax, df_day)
    ax.set_title(f"{title_prefix}  {pd.to_datetime(trade_row['Session']).date()}")

    ax.set_xlim(df_day.index.min(), df_day.index.max())
    ax.xaxis.set_major_locator(MinuteLocator(byminute=[0, 15, 30, 45]))
    try:
        ax.xaxis.set_major_formatter(DateFormatter("%H:%M", tz=df_day.index.tz))
    except TypeError:
        ax.xaxis.set_major_formatter(DateFormatter("%H:%M"))
    ax.grid(True, linestyle=":", linewidth=0.6)

    orh = float(trade_row["ORH"]); orl = float(trade_row["ORL"])
    t0, t1 = df_day.index.min(), df_day.index.max()
    ax.hlines(orh, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)
    ax.hlines(orl, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)

    # breakout marker
    if "BreakoutTime" in trade_row and pd.notna(trade_row["BreakoutTime"]):
        bts = pd.to_datetime(trade_row["BreakoutTime"])
        ax.vlines(mdates.date2num(bts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(),
                  linestyles="dashed", linewidth=0.8)
        ax.annotate("Breakout", xy=(mdates.date2num(bts), df_day["High"].max()),
                    xytext=(5, -15), textcoords="offset points", fontsize=8,
                    bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    # entry marker
    entry_px = float(trade_row["Entry"])
    entry_ts = pd.to_datetime(trade_row["EntryTime"])
    ax.hlines(entry_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="solid", linewidth=1.2)
    ax.vlines(mdates.date2num(entry_ts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(),
              linestyles="dotted", linewidth=0.9)
    lbl_time = entry_ts.tz_convert(TZ).strftime('%H:%M') if entry_ts.tzinfo else entry_ts.strftime('%H:%M')
    ax.annotate(f"Entry: {entry_px:.2f}\n{lbl_time}",
                xy=(mdates.date2num(entry_ts), entry_px),
                xytext=(10, 10), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    # stop
    stop_px = float(trade_row["Stop"])
    ax.hlines(stop_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashdot", linewidth=1.0)
    ax.annotate(f"Stop: {stop_px:.2f}", xy=(mdates.date2num(t0), stop_px),
                xytext=(5, -12), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    # first realized exit
    tag, ex_ts, ex_px = _parse_exits(trade_row["Exits"])
    if tag is not None:
        ax.scatter([mdates.date2num(ex_ts)], [ex_px], s=25)
        ax.annotate(f"Exit: {ex_px:.2f} ({tag})",
                    xy=(mdates.date2num(ex_ts), ex_px),
                    xytext=(10, -15), textcoords="offset points",
                    fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    pnl = float(trade_row["PnL_$"])
    ax.text(0.99, 0.02, f"PnL: ${pnl:,.2f}", transform=ax.transAxes,
            ha="right", va="bottom", fontsize=9, bbox=dict(boxstyle="round", fc="w", alpha=0.4))
    ax.text(0.01, 0.98, "UpdatedCharts", transform=ax.transAxes,
            ha="left", va="top", fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.3))

# =============================
# Runner + CLI
# =============================
def parse_args():
    p = argparse.ArgumentParser(description="SPY 15m ORB (retest) from parquet + charts")
    p.add_argument("--parquet-dir", type=str, default=PARQUET_DIR, help="Folder with SPY parquet.")
    p.add_argument("--outdir",      type=str, default=CHARTS_DIR,   help="Charts output directory.")
    p.add_argument("--dpi",         type=int, default=170,          help="Chart DPI.")
    p.add_argument("--limit",       type=int, default=-1,           help="Max charts (use -1 for all).")
    p.add_argument("--start",       type=str, default=None,         help="Start date YYYY-MM-DD (optional).")
    p.add_argument("--end",         type=str, default=None,         help="End date YYYY-MM-DD (optional).")
    p.add_argument("--date",        type=str, default=None,         help="Plot only this YYYY-MM-DD session.")
    args, _ = p.parse_known_args()  # ignore Jupyter's -f
    return args

def main():
    args = parse_args()
    os.makedirs(args.outdir, exist_ok=True)

    spy_path = _find_spy_file(args.parquet_dir)
    if not spy_path:
        raise FileNotFoundError(f"No parquet in {args.parquet_dir}")

    print(f"Loading: {spy_path}")
    df15 = load_parquet_15m(spy_path, tz=TZ)
    if df15.empty:
        print("No usable 15m data after normalization/resample/session filter.")
        return
    df15["Ticker"] = "SPY"
    df15 = compute_opening_range(df15)

    print("Running ORB retest backtest…")
    trades, equity = backtest_orb_retest(df15)

    if trades.empty:
        print("No trades generated with current settings.")
        return

    trades.to_csv(TRADES_CSV, index=False)
    print(f"Saved trades: {TRADES_CSV}")

    if not equity.empty:
        equity.to_csv(EQUITY_CSV, index=False)
        print(f"Saved equity: {EQUITY_CSV}")

    # --- Overall Totals (SPY) block ---
    total_trades = len(trades)
    total_pnl    = trades["PnL_$"].sum()
    win_rate     = (trades["PnL_$"] > 0).mean() * 100.0
    avg_r        = trades["R_multiple"].mean()
    median_r     = trades["R_multiple"].median()
    print("\n=== Overall Totals (SPY) ===")
    print(f"Total trades: {total_trades}")
    print(f"Total PnL ($): {total_pnl:,.2f}")
    print(f"Win rate: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")

    # --- Render charts (ALWAYS regenerate) ---
    tlog = trades.copy()
    tlog["Session"] = pd.to_datetime(tlog["Session"])
    if args.date:
        day = pd.to_datetime(args.date).normalize()
        tlog = tlog[tlog["Session"] == day]
    else:
        if args.start: tlog = tlog[tlog["Session"] >= pd.to_datetime(args.start)]
        if args.end:   tlog = tlog[tlog["Session"] <= pd.to_datetime(args.end)]
    tlog.sort_values(["Session","EntryTime"], inplace=True)

    printed, total = 0, len(tlog)
    for _, tr in tlog.iterrows():
        ses = tr["Session"].date()
        direction = str(tr.get("Direction",""))
        fname = f"SPY_UpdatedCharts_{ses}_{direction}.png"
        out_path = os.path.join(args.outdir, fname)

        df_day = df15[df15["Session"] == pd.to_datetime(ses)].copy()
        if df_day.empty:
            continue

        fig, ax = plt.subplots(figsize=(12, 6))
        plot_trade(ax, df_day, tr, title_prefix="SPY")
        fig.autofmt_xdate()
        fig.tight_layout()
        fig.savefig(out_path, dpi=args.dpi)
        plt.close(fig)
        print(f"Wrote {out_path}")
        printed += 1
        if (args.limit != -1) and (printed >= args.limit):
            break

    print(f"Done. Rendered {printed}/{total} charts.")

if __name__ == "__main__":
    main()
```

```python
# orb_retest_spy.py
# SPY 15m ORB breakout with entry on retest — parquet loader, backtest, charts

import warnings
warnings.filterwarnings("ignore")

import os, glob, argparse, ast
from datetime import timedelta
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.dates import MinuteLocator, DateFormatter
from matplotlib.patches import Rectangle

# =============================
# USER SETTINGS
# =============================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# Strategy — EXACT rule requested:
# Breakout: bar CLOSE beyond ORH/ORL
# Entry:    first retest TOUCH of that level (no close confirmation)
# Timeout:  None (wait the rest of the session)
BREAKOUT_RULE = "close"            # don't change
RETEST_RULE   = "touch"            # don't change
MAX_RETEST_MIN: Optional[int] = None

R_MULTIPLES           = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS          = 1.0
FEES_PER_TRADE        = 0.00

# Output
TRADES_CSV = "trades_spy_orb.csv"
EQUITY_CSV = "equity_spy_orb.csv"
CHARTS_DIR = "charts"

# =============================
# Column normalize
# =============================
STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                      rename[lower[k]] = std
        elif k.capitalize() in df.columns:  rename[k.capitalize()] = std
        elif k.upper() in df.columns:       rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _median_minutes(idx: pd.DatetimeIndex) -> float:
    if len(idx) < 2: return np.nan
    return float(np.median((idx[1:] - idx[:-1]).total_seconds()/60.0))

# =============================
# Timezone handling
# =============================
def _is_rth_wallclock(idx: pd.DatetimeIndex) -> bool:
    if not isinstance(idx, pd.DatetimeIndex) or len(idx) == 0:
        return False
    hrs = pd.Series(idx.hour)
    return ((hrs >= 9) & (hrs <= 16)).mean() > 0.80

def _reinterpret_as_tz(idx: pd.DatetimeIndex, tz: str) -> pd.DatetimeIndex:
    if idx.tz is None:
        return idx.tz_localize(tz)
    return idx.tz_convert(None).tz_localize(tz)

def _fix_timezone(idx: pd.DatetimeIndex,
                  assume_naive_are_utc: bool,
                  target_tz: str) -> pd.DatetimeIndex:
    if idx.tz is None:
        return (idx.tz_localize("UTC").tz_convert(target_tz)
                if assume_naive_are_utc else
                idx.tz_localize(target_tz))
    if str(idx.tz) in ("UTC", "UTC+00:00", "tzutc()", "pytz.UTC"):
        return _reinterpret_as_tz(idx, target_tz) if _is_rth_wallclock(idx) else idx.tz_convert(target_tz)
    return idx.tz_convert(target_tz)

# =============================
# Resample — force right-labeled 15m (first label 09:45)
# =============================
def _resample_to_15m_if_needed(df: pd.DataFrame) -> pd.DataFrame:
    """
    Normalize to right-labeled, right-closed 15m bars so the first RTH label is 09:45.
    """
    m = _median_minutes(df.index)
    if np.isnan(m):
        return pd.DataFrame()

    if m < 15 - 1e-6:
        agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
        if "Volume" in df.columns:
            agg["Volume"] = "sum"
        out = (df.sort_index()
                 .resample("15T", label="right", closed="right")
                 .agg(agg)
                 .dropna(subset=["Open","High","Low","Close"]))
        return out

    elif abs(m - 15) <= 1e-6:
        dfx = df.sort_index().copy()
        probe = dfx.between_time("09:00", "10:00")
        if not probe.empty:
            first_stamp = probe.index[0]
            if first_stamp.minute == 30:
                dfx.index = dfx.index + pd.Timedelta(minutes=15)
        return dfx

    else:
        return pd.DataFrame()

# =============================
# Loader
# =============================
def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()
    df.index = _fix_timezone(df.index, ASSUME_NAIVE_TIMESTAMPS_ARE_UTC, tz)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    df = _resample_to_15m_if_needed(df)
    if df.empty:
        return df

    df = df.between_time(REG_SESSION_START, REG_SESSION_END).sort_index()
    df["Session"] = pd.to_datetime(df.index.date)

    # quick sanity print for first RTH session window
    ses0 = df["Session"].iloc[0]
    g = df[df["Session"] == ses0]
    print(f"[RTH] {ses0.date()}  {g.index.min().time()} → {g.index.max().time()}  (first 15m label {g.index[0].time()})")
    return df

def _find_spy_file(folder: str) -> str:
    files = glob.glob(os.path.join(folder, "*.parquet"))
    if not files:
        return ""
    cands = [f for f in files if "spy" in os.path.basename(f).lower()]
    if cands:
        exact = [f for f in cands if os.path.basename(f).lower() in ("spy.parquet","spy_15m.parquet")]
        return exact[0] if exact else sorted(cands)[0]
    return sorted(files)[0]

# =============================
# ORB logic
# =============================
def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()
    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan
    df["Session"] = sessionize(df)
    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: Optional[int] = MAX_RETEST_MIN,
    r_targets: List[float] = R_MULTIPLES,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    breakout_rule: str = BREAKOUT_RULE,
    retest_rule: str = RETEST_RULE
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    trades = []

    for ses, sdf in df.groupby("Session"):
        if len(sdf) < 3:
            continue

        orh = float(sdf["ORH"].iloc[0]); orl = float(sdf["ORL"].iloc[0])
        if pd.isna(orh) or pd.isna(orl):
            continue

        # --- Breakout: bar CLOSE beyond ORH/ORL (per your rule) ---
        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction == "long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        breakout_time = breakout_row.name

        # --- Retest window: whole session (no timeout) unless max_retest_min set ---
        if max_retest_min is None:
            after_break = sdf[sdf.index > breakout_time].copy()
        else:
            cutoff = breakout_time + timedelta(minutes=int(max_retest_min))
            after_break = sdf[(sdf.index > breakout_time) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl

        # --- Retest: enter on TOUCH (no close confirmation) ---
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar  = touch.iloc[0]
        retest_time = retest_bar.name

        # --- Entry/stop/targets ---
        entry = _apply_slippage(level, slippage_bps, "buy" if direction == "long" else "sell")
        stop  = orl if direction == "long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction == "long" else -1
        t_prices = [(entry + r*risk_per_share) if direction == "long" else (entry - r*risk_per_share) for r in r_targets]

        # --- Forward simulate exits ---
        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = float(row["High"]), float(row["Low"])
            # Stop first
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction == "long" else "buy")
                exits.append(("stop", ts, px))
                break
            # Targets (first hit wins)
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction == "long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            # EOD flatten on last bar close
            last = sdf_run.iloc[-1]
            px = _apply_slippage(float(last["Close"]), slippage_bps, "sell" if direction == "long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        # Equal partials among realized exits
        n_parts = len(exits); qty_each = qty / n_parts
        cash_pnl = sum((px - entry) * side_mult * qty_each for _,_,px in exits) - FEES_PER_TRADE
        r_multiple = cash_pnl / (risk_per_share * qty)

        trades.append({
            "Ticker": "SPY",
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "BreakoutTime": breakout_time,
            "EntryTime": retest_time,
            "Entry": float(entry), "Stop": float(stop),
            "Targets": [float(x) for x in t_prices],
            "Exits": [(str(t[0]), t[1], float(t[2])) for t in exits],
            "Qty": float(qty),
            "PnL_$": float(cash_pnl),
            "R_multiple": float(r_multiple),
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Plotting
# =============================
def _parse_exits(exits_field):
    if isinstance(exits_field, list) and exits_field:
        tag, ts, px = exits_field[0]
        return tag, pd.to_datetime(ts), float(px)
    if isinstance(exits_field, str):
        try:
            first = exits_field.split(";")[0]
            tag, ts, px = first.split("|")
            return tag, pd.to_datetime(ts), float(px)
        except Exception:
            pass
    try:
        v = ast.literal_eval(exits_field)
        if isinstance(v, list) and v:
            tag, ts, px = v[0]
            return tag, pd.to_datetime(ts), float(px)
    except Exception:
        return None, None, None
    return None, None, None

def _draw_candles(ax, df_day: pd.DataFrame, width_minutes=10):
    width = (width_minutes / (24*60))
    times = mdates.date2num(df_day.index.to_pydatetime())
    for i, (_, row) in enumerate(df_day.iterrows()):
        o, h, l, c = float(row["Open"]), float(row["High"]), float(row["Low"]), float(row["Close"])
        ax.plot([times[i], times[i]], [l, h], linewidth=1.0)
        y = min(o, c); height = abs(c - o) or 1e-6
        rect = Rectangle((times[i] - width/2.0, y), width, height, fill=False)
        ax.add_patch(rect)

def plot_trade(ax, df_day: pd.DataFrame, trade_row: pd.Series, title_prefix="SPY"):
    _draw_candles(ax, df_day)
    ax.set_title(f"{title_prefix}  {pd.to_datetime(trade_row['Session']).date()}")

    ax.set_xlim(df_day.index.min(), df_day.index.max())
    ax.xaxis.set_major_locator(MinuteLocator(byminute=[0, 15, 30, 45]))
    try:
        ax.xaxis.set_major_formatter(DateFormatter("%H:%M", tz=df_day.index.tz))
    except TypeError:
        ax.xaxis.set_major_formatter(DateFormatter("%H:%M"))
    ax.grid(True, linestyle=":", linewidth=0.6)

    orh = float(trade_row["ORH"]); orl = float(trade_row["ORL"])
    t0, t1 = df_day.index.min(), df_day.index.max()
    ax.hlines(orh, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)
    ax.hlines(orl, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashed", linewidth=1.0)

    # breakout marker
    if "BreakoutTime" in trade_row and pd.notna(trade_row["BreakoutTime"]):
        bts = pd.to_datetime(trade_row["BreakoutTime"])
        ax.vlines(mdates.date2num(bts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(),
                  linestyles="dashed", linewidth=0.8)
        ax.annotate("Breakout", xy=(mdates.date2num(bts), df_day["High"].max()),
                    xytext=(5, -15), textcoords="offset points", fontsize=8,
                    bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    # entry marker
    entry_px = float(trade_row["Entry"])
    entry_ts = pd.to_datetime(trade_row["EntryTime"])
    ax.hlines(entry_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="solid", linewidth=1.2)
    ax.vlines(mdates.date2num(entry_ts), ymin=df_day["Low"].min(), ymax=df_day["High"].max(),
              linestyles="dotted", linewidth=0.9)
    lbl_time = entry_ts.tz_convert(TZ).strftime('%H:%M') if entry_ts.tzinfo else entry_ts.strftime('%H:%M')
    ax.annotate(f"Entry: {entry_px:.2f}\n{lbl_time}",
                xy=(mdates.date2num(entry_ts), entry_px),
                xytext=(10, 10), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    # stop
    stop_px = float(trade_row["Stop"])
    ax.hlines(stop_px, mdates.date2num(t0), mdates.date2num(t1), linestyles="dashdot", linewidth=1.0)
    ax.annotate(f"Stop: {stop_px:.2f}", xy=(mdates.date2num(t0), stop_px),
                xytext=(5, -12), textcoords="offset points",
                fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.4))

    # first realized exit
    tag, ex_ts, ex_px = _parse_exits(trade_row["Exits"])
    if tag is not None:
        ax.scatter([mdates.date2num(ex_ts)], [ex_px], s=25)
        ax.annotate(f"Exit: {ex_px:.2f} ({tag})",
                    xy=(mdates.date2num(ex_ts), ex_px),
                    xytext=(10, -15), textcoords="offset points",
                    fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.5))

    pnl = float(trade_row["PnL_$"])
    ax.text(0.99, 0.02, f"PnL: ${pnl:,.2f}", transform=ax.transAxes,
            ha="right", va="bottom", fontsize=9, bbox=dict(boxstyle="round", fc="w", alpha=0.4))
    ax.text(0.01, 0.98, "UpdatedCharts", transform=ax.transAxes,
            ha="left", va="top", fontsize=8, bbox=dict(boxstyle="round", fc="w", alpha=0.3))

# =============================
# Runner + CLI
# =============================
def parse_args():
    p = argparse.ArgumentParser(description="SPY 15m ORB (retest) from parquet + charts")
    p.add_argument("--parquet-dir", type=str, default=PARQUET_DIR, help="Folder with SPY parquet.")
    p.add_argument("--outdir",      type=str, default=CHARTS_DIR,   help="Charts output directory.")
    p.add_argument("--dpi",         type=int, default=170,          help="Chart DPI.")
    p.add_argument("--limit",       type=int, default=-1,           help="Max charts (use -1 for all).")
    p.add_argument("--start",       type=str, default=None,         help="Start date YYYY-MM-DD (optional).")
    p.add_argument("--end",         type=str, default=None,         help="End date YYYY-MM-DD (optional).")
    p.add_argument("--date",        type=str, default=None,         help="Plot only this YYYY-MM-DD session.")
    args, _ = p.parse_known_args()  # ignore Jupyter's -f
    return args

def main():
    args = parse_args()
    os.makedirs(args.outdir, exist_ok=True)

    spy_path = _find_spy_file(args.parquet_dir)
    if not spy_path:
        raise FileNotFoundError(f"No parquet in {args.parquet_dir}")

    print(f"Loading: {spy_path}")
    df15 = load_parquet_15m(spy_path, tz=TZ)
    if df15.empty:
        print("No usable 15m data after normalization/resample/session filter.")
        return
    df15["Ticker"] = "SPY"
    df15 = compute_opening_range(df15)

    print("Running ORB retest backtest…")
    trades, equity = backtest_orb_retest(df15)

    if trades.empty:
        print("No trades generated with current settings.")
        return

    trades.to_csv(TRADES_CSV, index=False)
    print(f"Saved trades: {TRADES_CSV}")

    if not equity.empty:
        equity.to_csv(EQUITY_CSV, index=False)
        print(f"Saved equity: {EQUITY_CSV}")

    # --- Overall Totals (SPY) ---
    total_trades = len(trades)
    total_pnl    = trades["PnL_$"].sum()
    win_rate     = (trades["PnL_$"] > 0).mean() * 100.0
    avg_r        = trades["R_multiple"].mean()
    median_r     = trades["R_multiple"].median()
    print("\n=== Overall Totals (SPY) ===")
    print(f"Total trades: {total_trades}")
    print(f"Total PnL ($): {total_pnl:,.2f}")
    print(f"Win rate: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")

    # --- Render charts (ALWAYS regenerate; add UpdatedCharts to filename) ---
    tlog = trades.copy()
    tlog["Session"] = pd.to_datetime(tlog["Session"])
    if args.date:
        day = pd.to_datetime(args.date).normalize()
        tlog = tlog[tlog["Session"] == day]
    else:
        if args.start: tlog = tlog[tlog["Session"] >= pd.to_datetime(args.start)]
        if args.end:   tlog = tlog[tlog["Session"] <= pd.to_datetime(args.end)]
    tlog.sort_values(["Session","EntryTime"], inplace=True)

    printed, total = 0, len(tlog)
    for _, tr in tlog.iterrows():
        ses = tr["Session"].date()
        direction = str(tr.get("Direction",""))
        fname = f"SPY_UpdatedCharts_{ses}_{direction}.png"
        out_path = os.path.join(args.outdir, fname)

        df_day = df15[df15["Session"] == pd.to_datetime(ses)].copy()
        if df_day.empty:
            continue

        fig, ax = plt.subplots(figsize=(12, 6))
        plot_trade(ax, df_day, tr, title_prefix="SPY")
        fig.autofmt_xdate()
        fig.tight_layout()
        fig.savefig(out_path, dpi=args.dpi)
        plt.close(fig)
        print(f"Wrote {out_path}")
        printed += 1
        if (args.limit != -1) and (printed >= args.limit):
            break

    print(f"Done. Rendered {printed}/{total} charts.")

if __name__ == "__main__":
    main()
```
