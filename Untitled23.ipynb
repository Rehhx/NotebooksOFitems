```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
leases_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(leases_path)

# Clean & filter
df = df.dropna(subset=['leasedSF', 'overall_rent', 'availability_proportion', 'market', 'year', 'quarter'])
df = df[df['leasedSF'] > 0]

# Compute metrics
df['year_quarter'] = df['year'].astype(str) + " Q" + df['quarter'].astype(str)
df['rent_per_sf'] = df['overall_rent'] / df['leasedSF']

# Focus on main cities
cities = ['Houston', 'Atlanta', 'San Francisco', 'Manhattan']
filtered_df = df[df['market'].isin(cities)]

# Group by city and quarter
grouped = filtered_df.groupby(['market', 'year_quarter']).agg({
    'rent_per_sf': 'mean',
    'availability_proportion': 'mean'
}).reset_index()

# Pivot for plotting
pivot_rent = grouped.pivot(index='year_quarter', columns='market', values='rent_per_sf')
pivot_avail = grouped.pivot(index='year_quarter', columns='market', values='availability_proportion')

# Sort by date
pivot_rent = pivot_rent.sort_index()
pivot_avail = pivot_avail.sort_index()

# Plot
fig, ax1 = plt.subplots(figsize=(16, 7))

# Left y-axis: Rent
colors = {
    'Houston': 'blue',
    'Atlanta': 'orange',
    'San Francisco': 'red',
    'Manhattan': 'brown'
}
for city in cities:
    ax1.plot(pivot_rent.index, pivot_rent[city], label=f"{city} Rent", color=colors[city])
ax1.set_ylabel("Average Rent ($/sqft)", fontsize=12)
ax1.set_xlabel("Time", fontsize=12)
ax1.tick_params(axis='x', rotation=45)

# Right y-axis: Availability
ax2 = ax1.twinx()
for city in cities:
    ax2.plot(pivot_avail.index, pivot_avail[city], label=f"{city} Availability", color=colors[city], linestyle='dashed')
ax2.set_ylabel("Average Availability Proportion", fontsize=12)

# Combine legends
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines + lines2, labels + labels2, loc='upper left', fontsize=10)

# Title
plt.title("Average Rent (Solid) & Availability (Dotted) Over Time by Market", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the lease data
leases_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(leases_path)

# Clean and filter
df = df.dropna(subset=['leasedSF', 'overall_rent', 'availability_proportion', 'market', 'state', 'city', 'year', 'quarter'])
df = df[df['leasedSF'] > 0]

# Add computed columns
df['rent_per_sf'] = df['overall_rent'] / df['leasedSF']
df['occupancy_rate'] = 1 - df['availability_proportion']
df['year_quarter'] = df['year'].astype(str) + " Q" + df['quarter'].astype(str)

# Define state groupings
state_groups = {
    'TX': 'Texas',
    'GA': 'Georgia',
    'CA': 'California',
    'NY': 'New York'
}

# Filter only for those 4 states
state_df = df[df['state'].isin(state_groups.keys())].copy()

# Replace state code with full name
state_df['state_label'] = state_df['state'].map(state_groups)

# Group by state and year_quarter to get average metrics
state_summary = state_df.groupby(['state_label', 'year_quarter']).agg({
    'rent_per_sf': 'mean',
    'availability_proportion': 'mean',
    'leasedSF': 'mean'
}).reset_index()

# Sort by year_quarter for plotting
state_summary = state_summary.sort_values(by='year_quarter')

# Plot 1: Average Rent per SF by State
plt.figure(figsize=(14, 6))
for state in state_summary['state_label'].unique():
    data = state_summary[state_summary['state_label'] == state]
    plt.plot(data['year_quarter'], data['rent_per_sf'], label=state)
plt.title("Average Rent ($/SF) Over Time by State")
plt.xlabel("Year - Quarter")
plt.ylabel("Rent per SF")
plt.xticks(rotation=45)
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()

# Plot 2: Availability Proportion by State
plt.figure(figsize=(14, 6))
for state in state_summary['state_label'].unique():
    data = state_summary[state_summary['state_label'] == state]
    plt.plot(data['year_quarter'], data['availability_proportion'], label=state, linestyle='dashed')
plt.title("Availability Proportion Over Time by State")
plt.xlabel("Year - Quarter")
plt.ylabel("Availability Proportion")
plt.xticks(rotation=45)
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()

# Plot 3: Average Leased SF by State
plt.figure(figsize=(14, 6))
for state in state_summary['state_label'].unique():
    data = state_summary[state_summary['state_label'] == state]
    plt.plot(data['year_quarter'], data['leasedSF'], label=state)
plt.title("Average Leased SF Over Time by State")
plt.xlabel("Year - Quarter")
plt.ylabel("Avg Leased SF")
plt.xticks(rotation=45)
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```
