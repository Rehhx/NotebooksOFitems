```python
import pandas as pd

# Load both datasets
leases_path = r"C:\Users\pcagm\Downloads\Leases.csv"
occupancy_path = r"C:\Users\pcagm\Downloads\Major Market Occupancy Data-revised.csv"

df_leases = pd.read_csv(leases_path)
df_occupancy = pd.read_csv(occupancy_path)

# 🧼 Clean and align 'quarter' format (e.g., Q1 → 1)
df_leases['quarter'] = df_leases['quarter'].astype(str).str.extract(r'Q?([1-4])')[0].astype(int)
df_occupancy['quarter'] = df_occupancy['quarter'].astype(str).str.extract(r'Q?([1-4])')[0].astype(int)

# Ensure 'year' is integer
df_leases['year'] = df_leases['year'].astype(int)
df_occupancy['year'] = df_occupancy['year'].astype(int)

# Clean whitespace and formatting in 'market'
df_leases['market'] = df_leases['market'].astype(str).str.strip().str.lower()
df_occupancy['market'] = df_occupancy['market'].astype(str).str.strip().str.lower()

# 🔗 Merge on year, quarter, and market
merged_df = pd.merge(
    df_leases,
    df_occupancy,
    on=['year', 'quarter', 'market'],
    how='inner'  # Change to 'left' if you want to keep all lease rows
)

# 🖨️ Preview the merged data
pd.set_option('display.max_rows', None)
print("\n✅ Merged DataFrame (Top 10 rows):")
print(merged_df.head(10))
print(f"\n✅ Total merged rows: {len(merged_df)}")

# 💾 Optional: Save to CSV
output_path = r"C:\Users\pcagm\Downloads\Merged_Leases_Occupancy.csv"
merged_df.to_csv(output_path, index=False)
print(f"\n📁 Merged data exported to: {output_path}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Group by year, quarter, market, and internal_industry (sector)
grouped_occupancy = merged_df.groupby(
    ['year', 'quarter', 'market', 'internal_industry']
)['avg_occupancy_proportion'].mean().reset_index()

# Combine year and quarter into one label for time series
grouped_occupancy['year_quarter'] = grouped_occupancy['year'].astype(str) + ' Q' + grouped_occupancy['quarter'].astype(str)

# Sort for chronological plotting
grouped_occupancy = grouped_occupancy.sort_values(['year', 'quarter'])

# Pivot for multi-line plotting: one line per (market + sector)
grouped_occupancy['market_sector'] = grouped_occupancy['market'].str.title() + ' - ' + grouped_occupancy['internal_industry']
pivot_occ = grouped_occupancy.pivot(index='year_quarter', columns='market_sector', values='avg_occupancy_proportion')

# Plot
plt.figure(figsize=(20, 10))
pivot_occ.plot(marker='o', figsize=(20, 10))
plt.title('Average Occupancy Proportion by Market and Sector (Quarterly)')
plt.xlabel('Year-Quarter')
plt.ylabel('Avg Occupancy Proportion')
plt.xticks(rotation=90)
plt.grid(True)
plt.legend(title='Market - Sector', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Filter merged_df to selected sectors only
selected_sectors = [
    'Financial Services and Insurance',
    'Technology, Advertising, Media, and Information',
    'Legal Services'
]
filtered_df = merged_df[merged_df['internal_industry'].isin(selected_sectors)]

# Group: Average occupancy by year, quarter, market, and sector
grouped = filtered_df.groupby(
    ['year', 'quarter', 'market', 'internal_industry']
)['avg_occupancy_proportion'].mean().reset_index()

# Create a combined year_quarter label
grouped['year_quarter'] = grouped['year'].astype(str) + ' Q' + grouped['quarter'].astype(str)

# Sort for time series plotting
grouped = grouped.sort_values(['year', 'quarter'])

# 🖨️ Print clean summary table
pd.set_option('display.max_rows', None)
print("\n📄 Avg Occupancy Proportion by Market and Sector:")
print(grouped[['year_quarter', 'market', 'internal_industry', 'avg_occupancy_proportion']])

# ✅ Plot one chart per market, line per sector
markets = grouped['market'].unique()
num_markets = len(markets)

fig, axes = plt.subplots(num_markets, 1, figsize=(20, 6 * num_markets), sharex=True)

if num_markets == 1:
    axes = [axes]

for i, market in enumerate(markets):
    ax = axes[i]
    df_market = grouped[grouped['market'] == market]

    for sector in selected_sectors:
        df_sector = df_market[df_market['internal_industry'] == sector]
        if not df_sector.empty:
            ax.plot(df_sector['year_quarter'], df_sector['avg_occupancy_proportion'], marker='o', label=sector)

    ax.set_title(f"{market.title()} - Avg Occupancy by Sector")
    ax.set_xlabel('Year-Quarter')
    ax.set_ylabel('Avg Occupancy Proportion')
    ax.tick_params(axis='x', rotation=90)
    ax.grid(True)
    ax.legend(title='Sector', bbox_to_anchor=(1.02, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
