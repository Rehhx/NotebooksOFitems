```python
import snscrape.modules.twitter as sntwitter
import pandas as pd
from textblob import TextBlob
import matplotlib.pyplot as plt

# Define search parameters
query = "Bitcoin OR BTC -filter:retweets lang:en"
max_tweets = 200  # Number of tweets to fetch

# Scrape tweets
tweets = []
for i, tweet in enumerate(sntwitter.TwitterSearchScraper(query).get_items()):
    if i >= max_tweets:
        break
    tweets.append([tweet.date, tweet.content])

# Convert to DataFrame
df = pd.DataFrame(tweets, columns=['Date', 'Tweet'])

# Function to get sentiment polarity
def get_sentiment(text):
    return TextBlob(text).sentiment.polarity

# Apply sentiment analysis
df['Sentiment'] = df['Tweet'].apply(get_sentiment)

# Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date']).dt.date

# Aggregate daily sentiment
sentiment_trend = df.groupby('Date')['Sentiment'].mean().reset_index()

# Plot sentiment over time
plt.figure(figsize=(10,5))
plt.plot(sentiment_trend['Date'], sentiment_trend['Sentiment'], marker='o', linestyle='-')
plt.axhline(0, color='red', linestyle='dashed', alpha=0.5)
plt.xlabel("Date")
plt.ylabel("Average Sentiment")
plt.title("Bitcoin Twitter Sentiment Trend")
plt.xticks(rotation=45)
plt.show()
```
