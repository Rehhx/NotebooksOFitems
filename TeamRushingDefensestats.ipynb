```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd

def scrape_opponent_rush_yards_selenium():
    url = "https://www.teamrankings.com/college-football/stat/opponent-rushing-yards-per-game"

    # Setup ChromeDriver (headless for speed)
    options = Options()
    options.add_argument("--headless=new")
    options.add_argument("--disable-gpu")
    options.add_argument("--no-sandbox")

    driver = webdriver.Chrome(service=Service(), options=options)

    try:
        driver.get(url)

        # Wait until table with correct ID is visible
        WebDriverWait(driver, 15).until(
            EC.presence_of_element_located((By.ID, "DataTables_Table_0"))
        )

        # Now fetch rows within the tbody
        table = driver.find_element(By.ID, "DataTables_Table_0")
        rows = table.find_elements(By.CSS_SELECTOR, "tbody tr")

        data = []
        for row in rows:
            cells = row.find_elements(By.TAG_NAME, "td")
            if len(cells) == 8:
                row_data = [cell.text.strip() for cell in cells]
                data.append(row_data)

        # Define headers
        columns = ["Rank", "Team", "Opp Rush Yds/G", "Games", "Last 3", "Last 1", "Home", "Away"]
        df = pd.DataFrame(data, columns=columns)

        # Convert numeric fields
        for col in columns[2:]:
            df[col] = pd.to_numeric(df[col].str.replace(',', '').str.replace('–', ''), errors='coerce')

        return df

    finally:
        driver.quit()

if __name__ == "__main__":
    df = scrape_opponent_rush_yards_selenium()
    print(df.head())
    df.to_csv("opponent_rushing_yards_per_game.csv", index=False)
```

```python
import pandas as pd
import requests

# URLs of the TeamRankings stats
urls = {
    'Opponent Rushing Attempts per Game': 'https://www.teamrankings.com/college-football/stat/opponent-rushing-attempts-per-game',
    'Opponent Yards per Rush Attempt': 'https://www.teamrankings.com/college-football/stat/opponent-yards-per-rush-attempt',
    'Opponent Rushing Yards per Game': 'https://www.teamrankings.com/college-football/stat/opponent-rushing-yards-per-game',
    'Opponent Rushing Play Percentage': 'https://www.teamrankings.com/college-football/stat/opponent-rushing-play-pct',
    'Opponent Rushing Yards Percentage': 'https://www.teamrankings.com/college-football/stat/opponent-rushing-yards-pct'
}

# Fetch and save each table
for label, url in urls.items():
    print(f"Scraping: {label}")
    response = requests.get(url)
    tables = pd.read_html(response.text)
    
    if tables:
        df = tables[0]
        filename = label.lower().replace(" ", "_").replace("%", "pct") + ".csv"
        df.to_csv(filename, index=False)
        print(f"✅ Saved: {filename}")
    else:
        print(f"⚠️ No table found for: {label}")
```

```python
import pandas as pd
import requests

# Base URL pattern with a placeholder for year
base_url = "https://www.teamrankings.com/college-football/stat/opponent-rushing-yards-per-game?date={year}-01-21"

# Range of years to scrape (2025 to 2013)
years = range(2025, 2012, -1)

all_years_data = []

for year in years:
    url = base_url.format(year=year)
    print(f"Scraping: {url}")
    try:
        tables = pd.read_html(url)
        if tables:
            df = tables[0]
            df["Year"] = year  # Add year as a column
            all_years_data.append(df)
            print(f"✅ Scraped {year} with {len(df)} rows")
        else:
            print(f"⚠️ No table found for {year}")
    except Exception as e:
        print(f"❌ Error scraping {year}: {e}")

# Combine all years into one DataFrame
if all_years_data:
    combined_df = pd.concat(all_years_data, ignore_index=True)
    combined_df.to_csv("opponent_rushing_yards_per_game_2013_2025.csv", index=False)
    print("✅ Saved combined CSV: opponent_rushing_yards_per_game_2013_2025.csv")
else:
    print("❌ No data collected.")
```
