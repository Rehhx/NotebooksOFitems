```python
import requests
import json
import os
import time
import pandas as pd
from bs4 import BeautifulSoup

API_KEY = "FTYWE23FK7GGUNFQPMTB2KAQSCZWDPTV7P"

OUTPUT_DIR = "./scraped_data"
os.makedirs(OUTPUT_DIR, exist_ok=True)

CALLS_PER_SECOND = 5
DELAY = 1 / CALLS_PER_SECOND

ETHERSCAN_TOKENS_URL = "https://etherscan.io/tokens?sort=percent_change_24h&order={order}"

def scrape_top_movers(order="desc"):
    """Scrapes the top 100 tokens sorted by percentage change from Etherscan and stores in a DataFrame."""
    url = ETHERSCAN_TOKENS_URL.format(order=order)
    response = requests.get(url)
    
    if response.status_code != 200:
        print("Error fetching token data from Etherscan.")
        return pd.DataFrame()
    
    soup = BeautifulSoup(response.text, 'html.parser')
    token_rows = soup.select("table tbody tr")
    
    data = []
    for row in token_rows[:100]:
        columns = row.find_all("td")
        token_name = columns[1].text.strip()
        contract_link = columns[1].find("a")["href"] if columns[1].find("a") else ""
        contract_address = contract_link.split("/")[-1] if contract_link else ""
        percent_change = columns[3].text.strip()
        
        data.append([token_name, contract_address, percent_change])
    
    df = pd.DataFrame(data, columns=["Token Name", "Contract Address", "Percent Change"])
    return df

def get_token_holders(contract_address):
    """Fetches token holders using known wallet balances and stores in a DataFrame."""
    known_wallets = [
        "0x742d35Cc6634C0532925a3b844Bc454e4438f44e",
        "0x564286362092D8e7936F0549571a803B203aAceD"
    ]
    data = []
    for wallet in known_wallets:
        url = f"https://api.etherscan.io/api?module=account&action=tokenbalance&contractaddress={contract_address}&address={wallet}&tag=latest&apikey={API_KEY}"
        response = requests.get(url)
        time.sleep(DELAY)
        result = response.json()
        
        if result.get('status') == '1':
            data.append([wallet, result.get('result')])
        else:
            print(f"Error fetching token balance for {wallet}:", result.get('message', 'Unknown error'))
    
    df = pd.DataFrame(data, columns=["Wallet Address", "Balance"])
    return df

def get_wallet_transactions(wallet_address):
    """Fetches the major transactions of a wallet address and stores in a DataFrame."""
    url = f"https://api.etherscan.io/api?module=account&action=txlist&address={wallet_address}&startblock=0&endblock=99999999&sort=desc&apikey={API_KEY}"
    response = requests.get(url)
    time.sleep(DELAY)
    data = response.json()
    
    if data.get('status') == '1':
        df = pd.DataFrame(data['result'])
        return df
    else:
        print("Error fetching transactions:", data.get('message', 'Unknown error'))
        return pd.DataFrame()

def save_dataframe(df, filename):
    """Saves a Pandas DataFrame to CSV and prints it."""
    if not df.empty:
        filepath = os.path.join(OUTPUT_DIR, filename)
        df.to_csv(filepath, index=False)
        print(f"Data saved to {filepath}")
        print(df.head())
    else:
        print(f"No data available for {filename}.")

if __name__ == "__main__":
    print("Fetching top 100 positive movers...")
    positive_movers_df = scrape_top_movers(order="desc")
    save_dataframe(positive_movers_df, "top_positive_movers.csv")
    
    print("Fetching top 100 negative movers...")
    negative_movers_df = scrape_top_movers(order="asc")
    save_dataframe(negative_movers_df, "top_negative_movers.csv")
    
    combined_movers = pd.concat([positive_movers_df, negative_movers_df])
    
    for _, token in combined_movers.iterrows():
        contract_address = token["Contract Address"]
        token_name = token["Token Name"]
        print(f"Processing {token_name} ({contract_address})...")
        
        top_holders_df = get_token_holders(contract_address)
        save_dataframe(top_holders_df, f"{token_name}_holders.csv")
        
        for _, holder in top_holders_df.iterrows():
            wallet_address = holder["Wallet Address"]
            transactions_df = get_wallet_transactions(wallet_address)
            save_dataframe(transactions_df, f"{wallet_address}_transactions.csv")
    
    print("Scraping completed. CSV files saved in ./scraped_data/")
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import re
from IPython.display import display  # For better DataFrame visualization in Jupyter Notebook

# Define the URL to scrape
url = "https://etherscan.io/tokens?sort=percent_change_24h&order=desc"

# Headers to mimic a real browser visit
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
}

# Make the request
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# Extract token data
tokens = []
percent_changes = []
market_caps = []
holders = []

# Locate table rows containing token information
rows = soup.find_all("tr")

for row in rows[1:]:  # Skip header row
    cols = row.find_all("td")
    if len(cols) < 7:
        continue  # Skip rows without sufficient columns

    token_name = cols[1].text.strip()

    # Extract and clean percentage change
    percent_change_raw = cols[3].text.strip()
    percent_change_cleaned = re.findall(r"-?\d+\.\d+|-?\d+", percent_change_raw)  # Extract numeric values
    percent_change_value = float(percent_change_cleaned[0]) if percent_change_cleaned else 0

    # Extract and clean market cap
    market_cap_raw = cols[5].text.strip()
    market_cap_cleaned = re.findall(r"\d+\.\d+|\d+", market_cap_raw.replace(",", "").replace("$", ""))
    market_cap_value = float(market_cap_cleaned[0]) if market_cap_cleaned else 0

    # Extract and clean holders count
    holder_count_raw = cols[6].text.strip()
    holder_count_cleaned = re.findall(r"\d+", holder_count_raw.replace(",", ""))
    holder_count_value = int(holder_count_cleaned[0]) if holder_count_cleaned else 0

    # Append cleaned values
    tokens.append(token_name)
    percent_changes.append(percent_change_value)
    market_caps.append(market_cap_value)
    holders.append(holder_count_value)

# Create a DataFrame
df = pd.DataFrame({
    "Token": tokens,
    "% Change": percent_changes,
    "Market Cap": market_caps,
    "Holders": holders
})

# Set index as Token name
df.set_index("Token", inplace=True)

# Sort in ascending order
df_sorted = df.sort_values(by="% Change", ascending=True)

# Display the sorted DataFrame in Jupyter Notebook
display(df_sorted)
```
