```python
from concurrent.futures import ThreadPoolExecutor, as_completed
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut
import time
import random

# Initialize geolocator
geolocator = Nominatim(user_agent="lease_locator", timeout=10)

# Retry + Exponential Backoff
def geocode_address(row, retries=3, base_delay=1):
    full_addr = row['full_address']
    idx = row.name
    for attempt in range(retries):
        try:
            print(f"[{idx}] ⏳ Geocoding: {full_addr}")
            location = geolocator.geocode(full_addr)
            if location:
                print(f"[{idx}] ✅ Got ({location.latitude}, {location.longitude})")
                return pd.Series([location.latitude, location.longitude])
            else:
                print(f"[{idx}] ❌ Not found")
                return pd.Series([None, None])
        except GeocoderTimedOut:
            delay = base_delay * (2 ** attempt) + random.uniform(0, 0.5)
            print(f"[{idx}] ⏳ Timed out. Retrying in {round(delay, 2)}s...")
            time.sleep(delay)
        except Exception as e:
            print(f"[{idx}] ❌ Error: {e}")
            return pd.Series([None, None])
    return pd.Series([None, None])

# === Load & prepare data ===
file_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(file_path)
df['zip_str'] = df['zip'].fillna(0).astype(int).astype(str).str.zfill(5)
df['full_address'] = df['address'].astype(str) + ', ' + df['city'].astype(str) + ', ' + df['state'].astype(str) + ' ' + df['zip_str']

# === Threaded + rate-respecting execution ===
start_time = time.time()
results = []
max_workers = 3  # Small thread pool to prevent overloading (recommended max 1–3)

print(f"🚀 Starting threaded geocoding with {max_workers} workers...")

def wrapped_geocode(row):
    result = geocode_address(row)
    time.sleep(1.1)  # throttle each request to ~1 request/second
    return (row.name, *result)

with ThreadPoolExecutor(max_workers=max_workers) as executor:
    future_to_index = {executor.submit(wrapped_geocode, row): idx for idx, row in df.iterrows()}

    for future in as_completed(future_to_index):
        try:
            idx, lat, lon = future.result()
        except Exception:
            idx, lat, lon = future_to_index[future], None, None
        results.append((idx, lat, lon))

# === Assign results back to DataFrame ===
df[['latitude', 'longitude']] = pd.DataFrame.from_dict(
    {idx: [lat, lon] for idx, lat, lon in results},
    orient='index'
)

print(f"✅ Geocoding completed in {round(time.time() - start_time, 2)} seconds")
```

```python
import pandas as pd
import time
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut

# === Initialize geolocator ===
geolocator = Nominatim(user_agent="lease_locator", timeout=10)

def geocode_address(row, retries=3, base_delay=1):
    full_addr = row['full_address']
    idx = row.name
    for attempt in range(retries):
        try:
            print(f"[{idx}] ⏳ Geocoding: {full_addr}")
            location = geolocator.geocode(full_addr)
            if location:
                print(f"[{idx}] ✅ Got ({location.latitude}, {location.longitude})")
                return pd.Series([location.latitude, location.longitude])
            else:
                print(f"[{idx}] ❌ Not found")
                return pd.Series([None, None])
        except GeocoderTimedOut:
            delay = base_delay * (2 ** attempt)
            print(f"[{idx}] ⏳ Timeout, retry in {delay}s")
            time.sleep(delay)
        except Exception as e:
            print(f"[{idx}] ❌ Error: {e}")
            return pd.Series([None, None])
    return pd.Series([None, None])

# === Load and prepare data ===
file_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(file_path)

df['zip_str'] = df['zip'].fillna(0).astype(int).astype(str).str.zfill(5)
df['full_address'] = df['address'].astype(str) + ', ' + df['city'].astype(str) + ', ' + df['state'].astype(str) + ' ' + df['zip_str']

# === Filter cities with >=100 entries ===
city_counts = df['city'].value_counts()
eligible_cities = city_counts[city_counts >= 100].index

# === Select 100 entries per qualifying city ===
df_filtered = df[df['city'].isin(eligible_cities)].groupby('city').head(100).copy()
df_filtered['latitude'] = None
df_filtered['longitude'] = None

# === Start geocoding ===
print(f"🚀 Starting geocoding for {len(eligible_cities)} cities (100 rows each)...")

start_time = time.time()
for idx, row in df_filtered.iterrows():
    df_filtered.loc[idx, ['latitude', 'longitude']] = geocode_address(row)
    time.sleep(1.3)  # Respectful delay for Nominatim

print(f"✅ Geocoding completed in {round(time.time() - start_time, 2)} seconds")

# Optional: Save to CSV
# df_filtered.to_csv("Geocoded_Leases_100_Per_Eligible_City.csv", index=False)
```
