```python
# orb_15m_retest_from_filelist.py
import warnings
warnings.filterwarnings("ignore")

import os, time
from datetime import timedelta
from typing import List, Tuple

import numpy as np
import pandas as pd

# =============================
# EXACT FILE LIST (edit here)
# =============================
FILE_PATHS = [
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\SPY.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\TEM.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\TSLA.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\TSM.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\AAPL.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\AMD.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\AMZN.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\CAT.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\CHWY.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\CVNA.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\GLD.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\GOOGL.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\GS.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\JPM.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\MSFT.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\NET.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\NFLX.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\NVDA.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\QQQ.parquet",
    r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\SE.parquet",
]

# =============================
# USER PARAMETERS
# =============================
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# Input data handling
RESAMPLE_TO_15M = True

# ORB logic
RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120  # minutes after breakout to wait for retest

# Risk / exits
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00

# Autosave cadence
AUTOSAVE_EVERY = 10

# Output files
TRADES_CSV_ALL = "orb_trades_from_parquet_list.csv"
EQUITY_CSV_ALL = "orb_equity_from_parquet_list.csv"
SUMMARY_CSV    = "orb_summary_from_parquet_list.csv"

# =============================
# Loading helpers
# =============================

def infer_ticker_from_filename(path: str) -> str:
    base = os.path.basename(path)
    name, _ = os.path.splitext(base)
    return name.upper()

def _normalize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    cols = {c.lower(): c for c in df.columns}
    mapping = {}
    for want, cand in {
        "Open":   ["open","o"],
        "High":   ["high","h"],
        "Low":    ["low","l"],
        "Close":  ["close","c"],
        "Volume": ["volume","vol","v"],
    }.items():
        for c in cand:
            if c in cols:
                mapping[cols[c]] = want
                break
    if mapping:
        df = df.rename(columns=mapping)
    for c in list(df.columns):
        if str(c).lower().startswith("adj"):
            df = df.drop(columns=[c], errors="ignore")
    return df

def _ensure_datetime_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    candidate_cols = [c for c in df.columns if str(c).lower() in {"datetime","date","timestamp","time"}]
    if not candidate_cols:
        try:
            df.index = pd.to_datetime(df.index, utc=False, errors="raise")
            return df
        except Exception:
            pass
        raise ValueError("No datetime index/column (expected Datetime/Date/Timestamp/Time).")
    dt_col = candidate_cols[0]
    df[dt_col] = pd.to_datetime(df[dt_col], utc=False, errors="coerce")
    df = df.set_index(dt_col, drop=True)
    return df

def _localize_and_convert_tz(df: pd.DataFrame, target_tz: str) -> pd.DataFrame:
    idx = df.index
    if idx.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df.index = idx.tz_localize("UTC").tz_convert(target_tz)
        else:
            df.index = idx.tz_localize(target_tz)
    else:
        df.index = idx.tz_convert(target_tz)
    return df

def _resample_15m(df: pd.DataFrame) -> pd.DataFrame:
    need = {"Open","High","Low","Close"}
    missing = need - set(df.columns)
    if missing:
        raise ValueError(f"Cannot resample: missing columns {missing}")
    agg = {"Open":"first", "High":"max", "Low":"min", "Close":"last"}
    if "Volume" in df.columns:
        agg["Volume"] = "sum"
    out = (df.sort_index()
             .resample("15T", label="right", closed="right")
             .agg(agg)
             .dropna(subset=["Open","High","Low","Close"], how="any"))
    return out

def load_ticker_from_parquet(path: str, tz: str) -> pd.DataFrame:
    if not os.path.exists(path):
        print(f"[MISSING] {path}")
        return pd.DataFrame()
    try:
        raw = pd.read_parquet(path, engine="pyarrow")
    except Exception as e:
        print(f"[READ ERR] {os.path.basename(path)}: {e}")
        return pd.DataFrame()
    if raw is None or raw.empty:
        print(f"[EMPTY] {os.path.basename(path)}")
        return pd.DataFrame()

    df = raw.copy()
    df = _normalize_ohlcv_columns(df)
    df = _ensure_datetime_index(df)
    df = _localize_and_convert_tz(df, tz)
    df = df[~df.index.duplicated(keep="last")].sort_index()

    if RESAMPLE_TO_15M:
        try:
            df = _resample_15m(df)
        except Exception as e:
            print(f"[RESAMPLE ERR] {os.path.basename(path)}: {e}")
            return pd.DataFrame()

    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            print(f"[COL MISSING] {os.path.basename(path)} missing {need}")
            return pd.DataFrame()

    df["Ticker"] = infer_ticker_from_filename(path)
    return df

def sessionize(df_15: pd.DataFrame) -> pd.Series:
    return pd.to_datetime(df_15.index.date)

# =============================
# ORB logic
# =============================

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = df_15.copy()
    for col in ["ORH", "ORL"]:
        if col not in df.columns:
            df[col] = np.nan
    df["Session"] = sessionize(df)
    first_bar_idx = df.groupby("Session").head(1).index
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float)
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float)
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def backtest_orb_retest(
    df_15: pd.DataFrame,
    max_retest_min: int = 120,
    r_targets: List[float] = [1.0, 2.0],
    dollars: float = 10_000,
    slippage_bps: float = 1.0,
    fees: float = 0.0,
    retest_confirm_close: bool = True
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    df = df_15.copy()
    df["Session"] = sessionize(df)
    sessions = df["Session"].unique()

    trades = []
    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh = sdf["ORH"].iloc[0]
        orl = sdf["ORL"].iloc[0]
        if pd.isna(orh) or pd.isna(orl):
            continue

        sdf_after = sdf.iloc[1:].copy()
        long_break  = sdf_after[sdf_after["Close"] > orh].head(1)
        short_break = sdf_after[sdf_after["Close"] < orl].head(1)

        if long_break.empty and short_break.empty:
            continue

        if not long_break.empty and not short_break.empty:
            direction = "long" if long_break.index[0] < short_break.index[0] else "short"
            breakout_row = long_break.iloc[0] if direction=="long" else short_break.iloc[0]
        elif not long_break.empty:
            direction = "long"; breakout_row = long_break.iloc[0]
        else:
            direction = "short"; breakout_row = short_break.iloc[0]

        btime = breakout_row.name
        cutoff = btime + timedelta(minutes=max_retest_min)
        after_break = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
        if after_break.empty:
            continue

        level = orh if direction == "long" else orl
        touch = after_break[(after_break["Low"] <= level) & (after_break["High"] >= level)].head(1)
        if touch.empty:
            continue

        retest_bar = touch.iloc[0]
        retest_time = retest_bar.name

        if retest_confirm_close:
            ok = (retest_bar["Close"] >= level) if direction=="long" else (retest_bar["Close"] <= level)
            if not ok:
                continue

        entry = _apply_slippage(level, slippage_bps, "buy" if direction=="long" else "sell")
        stop = orl if direction=="long" else orh
        risk_per_share = abs(entry - stop)
        if risk_per_share <= 1e-12:
            continue

        qty = dollars / entry
        side_mult = 1 if direction=="long" else -1

        t_prices = [(entry + r*risk_per_share) if direction=="long" else (entry - r*risk_per_share) for r in r_targets]

        sdf_run = sdf[sdf.index >= retest_time].copy()
        exits = []
        for ts, row in sdf_run.iterrows():
            high, low = row["High"], row["Low"]
            if low <= stop <= high:
                px = _apply_slippage(stop, slippage_bps, "sell" if direction=="long" else "buy")
                exits.append(("stop", ts, px))
                break
            hit = False
            for i, tp in enumerate(t_prices):
                if low <= tp <= high:
                    px = _apply_slippage(tp, slippage_bps, "sell" if direction=="long" else "buy")
                    exits.append((f"tp{i+1}", ts, px))
                    hit = True
            if hit:
                break

        if not exits:
            last = sdf_run.iloc[-1]
            px = _apply_slippage(last["Close"], slippage_bps, "sell" if direction=="long" else "buy")
            exits.append(("eod", sdf_run.index[-1], px))

        n_parts = len(exits)
        qty_each = qty / n_parts
        cash_pnl = 0.0
        for tag, ts, px in exits:
            cash_pnl += (px - entry) * side_mult * qty_each
        cash_pnl -= FEES_PER_TRADE

        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": direction,
            "EntryTime": retest_time,
            "Entry": entry, "Stop": stop,
            "Targets": t_prices,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": cash_pnl,
            "R_multiple": cash_pnl / (risk_per_share * qty)
        })

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Runner
# =============================

def run_from_list():
    files = FILE_PATHS
    print(f"Using {len(files)} parquet files from explicit list.")
    all_trades, all_equity = [], []
    processed = 0

    for i, path in enumerate(files, start=1):
        ticker = infer_ticker_from_filename(path)
        print(f"\n== {ticker} ({i}/{len(files)}) ==\n{path}")

        df = load_ticker_from_parquet(path, tz=TZ)
        if df.empty:
            print("No usable data after normalization/resample/session filter.")
            continue

        df = df.between_time(REG_SESSION_START, REG_SESSION_END)
        df = compute_opening_range(df)

        tlog, eq = backtest_orb_retest(
            df,
            max_retest_min=MAX_RETEST_MIN,
            r_targets=R_MULTIPLES,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE
        )

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"] = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs]))
            all_trades.append(tlog_out)

        if not eq.empty:
            eq_out = eq.copy()
            eq_out["Ticker"] = ticker
            all_equity.append(eq_out)

        processed += 1
        if processed % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    if not trades.empty:
        g = trades.groupby("Ticker")
        per_ticker = pd.DataFrame({
            "Trades":   g["PnL_$"].count(),
            "PnL_sum":  g["PnL_$"].sum(),
            "AvgPnL":   g["PnL_$"].mean(),
            "StdPnL":   g["PnL_$"].std(),
            "Wins":     g["PnL_$"].apply(lambda s: (s > 0).sum()),
            "Losses":   g["PnL_$"].apply(lambda s: (s < 0).sum()),
            "WinRate":  g["PnL_$"].apply(lambda s: (s > 0).mean()),
            "AvgWin":   g["PnL_$"].apply(lambda s: s[s > 0].mean() if (s > 0).any() else np.nan),
            "AvgLoss":  g["PnL_$"].apply(lambda s: s[s < 0].mean() if (s < 0).any() else np.nan),
            "AvgR":     g["R_multiple"].mean(),
            "MedianR":  g["R_multiple"].median(),
            "PF":       g["PnL_$"].apply(lambda s: (s[s > 0].sum() / abs(s[s < 0].sum()))
                                         if abs(s[s < 0].sum()) > 0 else np.nan)
        }).reset_index()

        per_ticker["WinRate"] = (per_ticker["WinRate"] * 100).round(2)
        per_ticker["LossMag"] = per_ticker["AvgLoss"].abs()
        per_ticker.to_csv(SUMMARY_CSV, index=False)

        total_trades = len(trades)
        total_pnl    = trades["PnL_$"].sum()
        win_rate     = (trades["PnL_$"] > 0).mean() * 100.0
        avg_r        = trades["R_multiple"].mean()
        median_r     = trades["R_multiple"].median()
        avg_win_all  = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
        avg_loss_all = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()
        pf_all = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())

        print("\n=== Overall Totals ===")
        print(f"Total trades: {total_trades}")
        print(f"Total PnL ($): {total_pnl:,.2f}")
        print(f"Win rate: {win_rate:.2f}%")
        print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
        print(f"Avg Win ($): {0.0 if pd.isna(avg_win_all) else avg_win_all:,.2f}")
        print(f"Avg Loss ($): {0.0 if pd.isna(avg_loss_all) else avg_loss_all:,.2f}")
        print(f"Profit Factor: {pf_all:.3f}")

        print("\n=== Per-Ticker Summary (ALL) ===")
        print(per_ticker.sort_values("Ticker")
                        .to_string(index=False,
                                   float_format=lambda x: f"{x:,.4f}"))
    else:
        print("No trades generated with current settings.")

if __name__ == "__main__":
    run_from_list()
```

```python
# orb_15m_retest_from_folder.py
import warnings
warnings.filterwarnings("ignore")

import os, glob
from datetime import timedelta
from typing import List, Tuple
import numpy as np
import pandas as pd

# =============================
# USER PARAMETERS
# =============================
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"

ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

RESAMPLE_TO_15M = True

RETEST_CONFIRM_CLOSE = True
MAX_RETEST_MIN = 120
R_MULTIPLES = [1.0, 2.0]
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00
AUTOSAVE_EVERY = 10

TRADES_CSV_ALL = "orb_trades_from_folder.csv"
EQUITY_CSV_ALL = "orb_equity_from_folder.csv"
SUMMARY_CSV    = "orb_summary_from_folder.csv"

# =============================
# Helpers
# =============================
def list_parquet_files(folder: str):
    return sorted(glob.glob(os.path.join(folder, "*.parquet")))

def infer_ticker(path: str) -> str:
    return os.path.splitext(os.path.basename(path))[0].upper()

def normalize(df: pd.DataFrame) -> pd.DataFrame:
    cols = {c.lower(): c for c in df.columns}
    mapping = {}
    for want, cand in {
        "Open": ["open","o"],
        "High": ["high","h"],
        "Low":  ["low","l"],
        "Close":["close","c"],
        "Volume":["volume","vol","v"],
    }.items():
        for c in cand:
            if c in cols:
                mapping[cols[c]] = want
                break
    df = df.rename(columns=mapping)
    # Drop Adj* columns if present
    for c in list(df.columns):
        if str(c).lower().startswith("adj"):
            df = df.drop(columns=[c], errors="ignore")
    # Enforce float where appropriate
    for c in ["Open","High","Low","Close","Volume"]:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")
    return df

def ensure_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    candidates = [c for c in df.columns if str(c).lower() in {"datetime","date","timestamp","time"}]
    if not candidates:
        df.index = pd.to_datetime(df.index, errors="coerce", utc=False)
    else:
        dt = candidates[0]
        df[dt] = pd.to_datetime(df[dt], errors="coerce", utc=False)
        df = df.set_index(dt, drop=True)
    return df

def localize(df: pd.DataFrame, tz: str) -> pd.DataFrame:
    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df.index = df.index.tz_localize("UTC").tz_convert(tz)
        else:
            df.index = df.index.tz_localize(tz)
    else:
        df.index = df.index.tz_convert(tz)
    return df

def resample_15m(df: pd.DataFrame) -> pd.DataFrame:
    # Require OHLC present
    need = {"Open","High","Low","Close"}
    missing = need - set(df.columns)
    if missing:
        raise ValueError(f"Cannot resample: missing {missing}")
    agg = {"Open":"first","High":"max","Low":"min","Close":"last"}
    if "Volume" in df.columns:
        agg["Volume"] = "sum"
    out = (df.sort_index()
             .resample("15T", label="right", closed="right")
             .agg(agg)
             .dropna(subset=["Open","High","Low","Close"], how="any"))
    return out

def load_file(path: str) -> pd.DataFrame:
    try:
        raw = pd.read_parquet(path, engine="pyarrow")
    except Exception as e:
        print(f"[READ ERR] {path}: {e}")
        return pd.DataFrame()

    if raw is None or raw.empty:
        print(f"[EMPTY] {path}")
        return pd.DataFrame()

    df = normalize(raw)
    # Basic column presence check before proceeding further
    if not {"Open","High","Low","Close"}.intersection(set(df.columns)):
        print(f"[COL ERR] {path}: missing OHLC columns")
        return pd.DataFrame()

    df = ensure_dt_index(df)
    # Drop rows with NaT index
    df = df[~df.index.isna()]
    if df.empty:
        print(f"[DT ERR] {path}: no valid timestamps")
        return pd.DataFrame()

    df = localize(df, TZ)
    # Ensure unique, sorted timestamps
    df = df[~df.index.duplicated(keep="last")].sort_index()

    if RESAMPLE_TO_15M:
        try:
            df = resample_15m(df)
        except Exception as e:
            print(f"[RESAMPLE ERR] {os.path.basename(path)}: {e}")
            return pd.DataFrame()

    # Regular session only
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)

    # Final OHLC check
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            print(f"[COL MISSING] {os.path.basename(path)} missing {need}")
            return pd.DataFrame()

    df["Ticker"] = infer_ticker(path)
    return df

# =============================
# ORB logic
# =============================
def sessionize(df): 
    return pd.to_datetime(df.index.date)

def compute_opening_range(df):
    df = df.copy()
    df["Session"] = sessionize(df)
    # Opening range = first 15m bar high/low of the session
    first_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_idx, "ORH"] = df.loc[first_idx, "High"].astype(float)
    df.loc[first_idx, "ORL"] = df.loc[first_idx, "Low"].astype(float)
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return df

def _apply_slip(px, bps, side):
    factor = 1 + bps/10000.0
    return px * factor if side == "buy" else px / factor

def backtest_orb(df):
    df = df.copy()
    df["Session"] = sessionize(df)
    trades = []
    for ses in df["Session"].unique():
        sdf = df[df["Session"] == ses]
        if len(sdf) < 3:
            continue

        orh = float(sdf["ORH"].iloc[0])
        orl = float(sdf["ORL"].iloc[0])

        # first breakout after the opening bar
        sdf_after = sdf.iloc[1:]
        long = sdf_after[sdf_after["Close"] > orh].head(1)
        short = sdf_after[sdf_after["Close"] < orl].head(1)
        if long.empty and short.empty:
            continue
        if not long.empty and not short.empty:
            dir_ = "long" if long.index[0] < short.index[0] else "short"
            br = long.iloc[0] if dir_ == "long" else short.iloc[0]
        elif not long.empty:
            dir_ = "long"; br = long.iloc[0]
        else:
            dir_ = "short"; br = short.iloc[0]

        btime = br.name
        cutoff = btime + timedelta(minutes=MAX_RETEST_MIN)
        after = sdf[(sdf.index > btime) & (sdf.index <= cutoff)]
        if after.empty:
            continue

        level = orh if dir_ == "long" else orl
        touch = after[(after["Low"] <= level) & (after["High"] >= level)].head(1)
        if touch.empty:
            continue

        bar = touch.iloc[0]
        rtime = bar.name
        if RETEST_CONFIRM_CLOSE:
            if dir_ == "long" and bar["Close"] < level:
                continue
            if dir_ == "short" and bar["Close"] > level:
                continue

        entry = _apply_slip(level, SLIPPAGE_BPS, "buy" if dir_ == "long" else "sell")
        stop  = orl if dir_ == "long" else orh
        risk  = abs(entry - stop)
        if risk <= 1e-12:
            continue

        qty  = POSITION_SIZE_DOLLARS / entry
        side = 1 if dir_ == "long" else -1
        tps  = [entry + r*risk if dir_ == "long" else entry - r*risk for r in R_MULTIPLES]

        run = sdf[sdf.index >= rtime]
        exits = []
        for ts, row in run.iterrows():
            low, high = row["Low"], row["High"]
            # stop first
            if low <= stop <= high:
                exits.append(("stop", ts, _apply_slip(stop, SLIPPAGE_BPS, "sell" if dir_=="long" else "buy")))
                break
            # then first target hit
            hit = False
            for i, tp in enumerate(tps):
                if low <= tp <= high:
                    exits.append((f"tp{i+1}", ts, _apply_slip(tp, SLIPPAGE_BPS, "sell" if dir_=="long" else "buy")))
                    hit = True
            if hit:
                break

        if not exits:
            last = run.iloc[-1]
            exits.append(("eod", last.name, _apply_slip(last["Close"], SLIPPAGE_BPS, "sell" if dir_=="long" else "buy")))

        pnl = sum((px - entry) * side * (qty/len(exits)) for _,_,px in exits) - FEES_PER_TRADE
        trades.append({
            "Ticker": sdf["Ticker"].iloc[0],
            "Session": ses,
            "ORH": orh, "ORL": orl,
            "Direction": dir_,
            "EntryTime": rtime,
            "Entry": entry, "Stop": stop,
            "Targets": tps,
            "Exits": exits,
            "Qty": qty,
            "PnL_$": pnl,
            "R_multiple": pnl / (risk * qty)
        })

    tlog = pd.DataFrame(trades)
    if tlog.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return tlog, eq

    eq = (tlog.groupby("Session")["PnL_$"].sum()
            .sort_index().cumsum().reset_index()
            .rename(columns={"PnL_$":"CumPnL_$"}))
    return tlog, eq

# =============================
# Runner
# =============================
def run_folder():
    files = list_parquet_files(PARQUET_DIR)
    print(f"Found {len(files)} parquet files.")
    all_trades, all_eq = [], []

    for i, path in enumerate(files, 1):
        tk = infer_ticker(path)
        print(f"\n== {tk} ({i}/{len(files)}) ==\n{path}")
        df = load_file(path)
        if df.empty:
            print("No usable data after normalization/resample/session filter.")
            continue

        df = compute_opening_range(df)
        tlog, eq = backtest_orb(df)

        if not tlog.empty:
            out = tlog.copy()
            out["Targets"] = out["Targets"].apply(lambda xs: ";".join(f"{p:.6f}" for p in xs))
            out["Exits"]   = out["Exits"].apply(lambda xs: ";".join(f"{t[0]}|{t[1]}|{t[2]:.6f}" for t in xs))
            all_trades.append(out)

        if not eq.empty:
            eq2 = eq.copy()
            eq2["Ticker"] = tk
            all_eq.append(eq2)

        # Autosave
        if i % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_eq:
                pd.concat(all_eq, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

    # Final outputs
    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_eq, ignore_index=True) if all_eq else pd.DataFrame()

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    # ===== Per-ticker summary + overall totals =====
    if not trades.empty:
        g = trades.groupby("Ticker")
        per_ticker = pd.DataFrame({
            "Trades":   g["PnL_$"].count(),
            "PnL_sum":  g["PnL_$"].sum(),
            "AvgPnL":   g["PnL_$"].mean(),
            "StdPnL":   g["PnL_$"].std(),
            "Wins":     g["PnL_$"].apply(lambda s: (s > 0).sum()),
            "Losses":   g["PnL_$"].apply(lambda s: (s < 0).sum()),
            "WinRate":  g["PnL_$"].apply(lambda s: (s > 0).mean()),
            "AvgWin":   g["PnL_$"].apply(lambda s: s[s > 0].mean() if (s > 0).any() else np.nan),
            "AvgLoss":  g["PnL_$"].apply(lambda s: s[s < 0].mean() if (s < 0).any() else np.nan),
            "AvgR":     g["R_multiple"].mean(),
            "MedianR":  g["R_multiple"].median(),
            "PF":       g["PnL_$"].apply(lambda s: (s[s > 0].sum() / abs(s[s < 0].sum()))
                                         if abs(s[s < 0].sum()) > 0 else np.nan),
        }).reset_index()

        per_ticker["WinRate"] = (per_ticker["WinRate"] * 100).round(2)
        per_ticker["LossMag"] = per_ticker["AvgLoss"].abs()
        per_ticker.to_csv(SUMMARY_CSV, index=False)
        print(f"Saved summary: {SUMMARY_CSV}")

        total_trades = len(trades)
        total_pnl    = trades["PnL_$"].sum()
        win_rate     = (trades["PnL_$"] > 0).mean() * 100.0
        avg_r        = trades["R_multiple"].mean()
        median_r     = trades["R_multiple"].median()
        avg_win_all  = trades.loc[trades["PnL_$"] > 0, "PnL_$"].mean()
        avg_loss_all = trades.loc[trades["PnL_$"] < 0, "PnL_$"].mean()
        pf_all = trades.loc[trades["PnL_$"] > 0, "PnL_$"].sum() / abs(trades.loc[trades["PnL_$"] < 0, "PnL_$"].sum())

        print("\n=== Overall Totals ===")
        print(f"Total trades: {total_trades}")
        print(f"Total PnL ($): {total_pnl:,.2f}")
        print(f"Win rate: {win_rate:.2f}%")
        print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
        print(f"Avg Win ($): {0.0 if pd.isna(avg_win_all) else avg_win_all:,.2f}")
        print(f"Avg Loss ($): {0.0 if pd.isna(avg_loss_all) else avg_loss_all:,.2f}")
        print(f"Profit Factor: {pf_all:.3f}")

        print("\n=== Per-Ticker Summary (ALL) ===")
        print(per_ticker.sort_values("Ticker")
                        .to_string(index=False,
                                   float_format=lambda x: f"{x:,.4f}"))
    else:
        print("No trades generated with current settings.")

if __name__=="__main__":
    run_folder()
```
