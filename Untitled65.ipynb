```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
SolarAPP+ Availability Scraper
--------------------------------
Scrapes the SolarAPP+ availability page and produces a CSV tracking
County/City (AHJ), State, and whether PV and PV+ST have checkmarks.

Default behavior:
  - Filters to California (State == "CA")
  - Requires both PV and PV+ST to be checked
  - Outputs columns: AHJ, State, PV, PV+ST

Usage:
    python solarapp_scraper.py
    python solarapp_scraper.py --out solarapp_ca_tracking.csv --debug

Optional flags:
    --all-states      Do NOT filter to CA only (include all states)
    --allow-missing   Do NOT require checks; include rows even if PV or PV+ST are missing
"""
import re
import sys
import argparse
from typing import List, Dict, Optional

import requests
from bs4 import BeautifulSoup
import pandas as pd

URL = "https://help.gosolarapp.org/article/108-where-is-solarapp-available"

# Strings that may indicate a checkmark. Covers text and common attributes.
CHECK_STRINGS = {"√", "✓", "✔", "check", "checked", "true", "yes"}

def is_checked(cell) -> bool:
    """
    Return True if a table cell appears to contain a check mark.
    Handles text checkmarks and possible <img alt="check"> icons.
    """
    if cell is None:
        return False
    txt = cell.get_text(strip=True).lower()
    if any(s in txt for s in {c.lower() for c in CHECK_STRINGS}):
        return True
    for img in cell.find_all("img"):
        alt = (img.get("alt") or "").lower()
        title = (img.get("title") or "").lower()
        if any(s in alt for s in CHECK_STRINGS) or any(s in title for s in CHECK_STRINGS):
            return True
    aria = (cell.get("aria-label") or "").lower()
    if any(s in aria for s in CHECK_STRINGS):
        return True
    return False

def normalize_text(cell) -> str:
    return cell.get_text(" ", strip=True) if cell else ""

def extract_state(ahj_text: str) -> Optional[str]:
    m = re.search(r",\s*([A-Z]{2})\s*$", ahj_text)
    return m.group(1) if m else None

def scrape(debug: bool=False) -> pd.DataFrame:
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/120.0.0.0 Safari/537.36"
    }
    resp = requests.get(URL, headers=headers, timeout=30)
    resp.raise_for_status()
    soup = BeautifulSoup(resp.text, "lxml")

    # Find tables (target often has class 'hs-table')
    tables = soup.select("table.hs-table, table")
    if not tables:
        raise RuntimeError("Could not find any <table> elements on the page. The layout may have changed.")
    rows: List[Dict] = []
    for table in tables:
        for tr in table.find_all("tr"):
            tds = tr.find_all(["td", "th"])
            if len(tds) < 6:
                continue  # skip header or malformed rows
            # Expected columns: AHJ | Local License | PV | PV+ST | MPU | MBD
            ahj_cell, lic_cell, pv_cell, pvst_cell, mpu_cell, mbd_cell = tds[:6]

            ahj_text = normalize_text(ahj_cell)
            # Skip header-like rows
            header_like = {"ahj", "local", "license", "pv+st", "mpu", "mbd"}
            if any(h in ahj_text.lower() for h in header_like):
                continue

            state = extract_state(ahj_text) or ""
            record = {
                "AHJ": ahj_text,
                "State": state,
                "PV": "✓" if is_checked(pv_cell) else "",
                "PV+ST": "✓" if is_checked(pvst_cell) else "",
            }
            if "," in record["AHJ"]:
                rows.append(record)

    if not rows:
        raise RuntimeError("Parsed 0 rows. The page structure may have changed.")

    df = pd.DataFrame(rows)

    if debug:
        print(f"Total rows parsed: {len(df)}")
        print(df.head(10).to_string(index=False))

    return df

def main():
    parser = argparse.ArgumentParser(description="Scrape SolarAPP+ availability table and output AHJ/State with PV & PV+ST checkmarks.")
    parser.add_argument("--out", default="solarapp_ca_tracking.csv", help="Output CSV filename")
    parser.add_argument("--debug", action="store_true", help="Print debug info")
    parser.add_argument("--all-states", action="store_true", help="Do not filter to California only")
    parser.add_argument("--allow-missing", action="store_true", help="Include rows even if PV or PV+ST are not checked")
    args = parser.parse_args()

    try:
        df = scrape(debug=args.debug)
    except Exception as e:
        print(f"[ERROR] {e}")
        sys.exit(1)

    # Apply filters
    if not args.all_states:
        df = df[df["State"] == "CA"]

    def col_is_check(s: pd.Series) -> pd.Series:
        return s.fillna("").astype(str).str.contains(r"[✓✔√]", regex=True)

    if not args.allow_missing:
        df = df[col_is_check(df["PV"]) & col_is_check(df["PV+ST"])]

    # Keep only requested columns
    df = df[["AHJ", "State", "PV", "PV+ST"]].reset_index(drop=True)

    # Save
    df.to_csv(args.out, index=False, encoding="utf-8")
    print(f"Saved {len(df)} rows to {args.out}")
    if len(df):
        print(df.head(min(20, len(df))).to_string(index=False))

if __name__ == "__main__":
    main()
```
