```python
# === pip (optional, once) ===
# import sys
# !{sys.executable} -m pip install "psycopg[binary]" pandas numpy pyarrow

import warnings
warnings.filterwarnings("ignore")

import math
from pathlib import Path
from typing import List, Optional, Tuple

import numpy as np
import pandas as pd
import psycopg

# =============================
# DB & RUN SETTINGS (EDIT ME)
# =============================

CONN_STR   = "host=24.128.195.68 port=8812 dbname=qdb user=admin password=quest"
TABLE      = "equities_candles"
INTERVAL   = "1m"                 # rows to read from DB (e.g., '1m', '5m', etc.)
BATCH_SIZE = 20                   # process tickers in groups of 20
MAX_SYMBOLS: Optional[int] = None # None = all symbols; or set an int to cap

# Optional date window to limit rows pulled from DB (UTC timestamps in QuestDB)
# e.g. START_DATE = "2016-01-01"; END_DATE = "2025-09-01"
START_DATE: Optional[str] = None
END_DATE:   Optional[str] = None

# Save CSVs to here
SAVE_DIR = Path(".")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# =============================
# STRATEGY PARAMETERS (same as before)
# =============================

TZ_LOCAL = "America/New_York"

REG_START = "09:30"   # session start
OR_END    = "09:45"   # end of first 15-min bar (exclusive)
REG_END   = "16:00"   # session end

STOP_FIRST_IF_BOTH_HIT = True  # tie-breaker if stop & target are inside the same bar

# =============================
# Helpers
# =============================

def to_ny(df: pd.DataFrame, col: str = "EventAt") -> pd.Series:
    """
    Robustly parse timestamps from DB:
    - If tz-naive: assume UTC → convert to New York
    - If tz-aware: convert to New York
    """
    ts = pd.to_datetime(df[col], errors="coerce", utc=True)  # DB should be UTC; force-interpret as UTC
    return ts.dt.tz_convert(TZ_LOCAL)

def first_breakout_close(df_after_or: pd.DataFrame, orh: float, orl: float) -> Tuple[Optional[str], Optional[int]]:
    """Return ('long'/'short', idx) for first bar *closing* outside the OR."""
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close: pd.DataFrame, level: float) -> Optional[int]:
    """First bar that *touches* the level after the breakout close."""
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return int(df_after_breakout_close.index[touched].min())

# ---------- NEW: ladder simulator ----------
def simulate_ladder(
    after_entry_df: pd.DataFrame,
    side: str,
    entry: float,
    stop0: float,
    R: float,
    stop_first: bool = True
) -> Tuple[str, float, Optional[int], float, dict]:
    """
    Scale-out ladder (3 equal units):
      - TP @ 1R -> move stop to BE (entry)
      - TP @ 2R -> move stop to +1R (entry ± R)
      - TP @ 3R -> flat
    Returns:
      outcome_label, last_exit_px, last_exit_idx, realized_pnl_total, details(dict)
    """
    sgn = 1 if side == "long" else -1
    T1 = entry + sgn * R
    T2 = entry + sgn * (2*R)
    T3 = entry + sgn * (3*R)

    cur_stop = stop0
    units_left = 3
    realized_pnl = 0.0

    hits = {
        "Hit_T1": False,
        "Hit_T2": False,
        "Hit_T3": False,
        "Stop_Level_On_Exit": None
    }

    def unit_pnl(px_exit: float) -> float:
        return (px_exit - entry) if side == "long" else (entry - px_exit)

    last_idx = None
    last_px  = None

    for idx, row in after_entry_df.iterrows():
        lo, hi, close = float(row["Low"]), float(row["High"]), float(row["Close"])

        progressed = True
        while progressed and units_left > 0:
            progressed = False

            next_target = None
            if not hits["Hit_T1"]:
                next_target = T1
            elif not hits["Hit_T2"]:
                next_target = T2
            elif not hits["Hit_T3"]:
                next_target = T3

            stop_hit = (lo <= cur_stop <= hi)  # for either side, stop level between low/high means touched
            targ_hit = False
            if next_target is not None:
                targ_hit = (hi >= next_target) if side == "long" else (lo <= next_target)

            if stop_hit and targ_hit:
                if stop_first:
                    realized_pnl += units_left * unit_pnl(cur_stop)
                    hits["Stop_Level_On_Exit"] = cur_stop
                    return "STOP_TIE", cur_stop, idx, realized_pnl, hits
                else:
                    # Fill target first (1 unit), then loop (might still stop after the stop moves)
                    realized_pnl += 1 * unit_pnl(next_target)
                    units_left -= 1
                    progressed = True
                    if next_target == T1:
                        hits["Hit_T1"] = True
                        cur_stop = entry               # move to BE
                    elif next_target == T2:
                        hits["Hit_T2"] = True
                        cur_stop = entry + sgn * R     # move to +1R
                    elif next_target == T3:
                        hits["Hit_T3"] = True
                        return "TP3_FULL", next_target, idx, realized_pnl, hits
                    continue  # keep checking within the same bar

            if stop_hit:
                realized_pnl += units_left * unit_pnl(cur_stop)
                hits["Stop_Level_On_Exit"] = cur_stop
                tag = ("STOP_BE" if cur_stop == entry else
                       "STOP_+1R" if cur_stop == entry + sgn*R else
                       "STOP_INIT")
                return tag, cur_stop, idx, realized_pnl, hits

            if targ_hit and next_target is not None:
                realized_pnl += 1 * unit_pnl(next_target)
                units_left -= 1
                progressed = True
                if next_target == T1:
                    hits["Hit_T1"] = True
                    cur_stop = entry               # to BE
                elif next_target == T2:
                    hits["Hit_T2"] = True
                    cur_stop = entry + sgn * R     # to +1R
                elif next_target == T3:
                    hits["Hit_T3"] = True
                    return "TP3_FULL", next_target, idx, realized_pnl, hits

        last_idx = idx
        last_px = close

    # End of session: flat remaining at last close
    if last_idx is None:
        return "EOD_NO_BARS", entry, None, realized_pnl, hits

    if units_left > 0 and last_px is not None:
        realized_pnl += units_left * unit_pnl(last_px)
        return "EOD", last_px, last_idx, realized_pnl, hits

    return "EOD", last_px if last_px is not None else entry, last_idx, realized_pnl, hits

def pnl_one_share(side: str, entry: float, exit_px: float) -> float:
    return (exit_px - entry) if side == "long" else (entry - exit_px)

def calc_metrics(trades_df: pd.DataFrame) -> Tuple[int, float, float, float]:
    # Use ladder PnL
    pnl = trades_df["PnL_Ladder"].fillna(0.0)
    wins = int((pnl > 0).sum())
    losses = int((pnl < 0).sum())
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = float(pnl[pnl > 0].sum())
    gross_neg = float(-pnl[pnl < 0].sum())
    profit_factor = (gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, float(profit_factor)

# =============================
# DB loaders
# =============================

def fetch_all_symbols(conn) -> List[str]:
    q = f"""
        SELECT DISTINCT symbol
        FROM {TABLE}
        WHERE "interval" = %s
        ORDER BY symbol
    """
    df = pd.read_sql(q, conn, params=(INTERVAL,))
    syms = df["symbol"].dropna().astype(str).tolist()
    if MAX_SYMBOLS is not None:
        syms = syms[:MAX_SYMBOLS]
    return syms

def fetch_symbol_data(conn, symbol: str) -> pd.DataFrame:
    """
    Pull one symbol’s rows from QuestDB for the chosen INTERVAL and optional date window.
    """
    base = f"""
        SELECT event_at, open, high, low, close, volume
        FROM {TABLE}
        WHERE symbol = %s
          AND "interval" = %s
    """
    params: List[object] = [symbol, INTERVAL]

    if START_DATE and END_DATE:
        base += " AND event_at >= %s AND event_at < %s"
        params.extend([START_DATE, END_DATE])
    elif START_DATE:
        base += " AND event_at >= %s"
        params.append(START_DATE)
    elif END_DATE:
        base += " AND event_at < %s"
        params.append(END_DATE)

    base += " ORDER BY event_at"

    df = pd.read_sql(base, conn, params=tuple(params))

    if df.empty:
        return df

    # Normalize column names to match backtest logic
    df = df.rename(columns={
        "event_at": "EventAt",
        "open": "Open",
        "high": "High",
        "low": "Low",
        "close": "Close",
        "volume": "Volume",
    })

    # Convert to NY time and add handy columns
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")

    # keep only regular session
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    return df.loc[session].copy()

# =============================
# Core backtest on a DataFrame (now with ladder exits)
# =============================

def backtest_df(df: pd.DataFrame, symbol: str) -> pd.DataFrame:
    if df.empty:
        return pd.DataFrame()

    for c in ["EventAt", "Open", "High", "Low", "Close", "date", "t"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: missing required column {c}")

    records = []

    for d, day in df.groupby("date"):
        # Opening range: first 15-minute bar [09:30, 09:45)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # After OR (09:45+)
        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        # 1) first confirmed breakout by CLOSE
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # 2) retest entry at broken level
        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop0      = orl if side == "long" else orh
        R          = or_width  # since entry at ORH/ORL and stop at opposite bound

        # Targets for record
        T1 = entry + ( R if side == "long" else -R)
        T2 = entry + ( 2*R if side == "long" else -2*R)
        T3 = entry + ( 3*R if side == "long" else -3*R)

        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            # no bars after entry → flat at entry (zero)
            records.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": or_width,
                "EntryTime": entry_time, "Entry": entry,
                "Stop0": stop0, "Target1": T1, "Target2": T2, "Target3": T3,
                "Outcome_Ladder": "EOD_NO_BARS",
                "ExitPx_Ladder": entry,
                "ExitTime_Ladder": entry_time,
                "PnL_Ladder": 0.0,
                "Hit_T1": False, "Hit_T2": False, "Hit_T3": False, "Stop_Level_On_Exit": None
            })
            continue

        outcome, exit_px, exit_idx, realized_pnl, hits = simulate_ladder(
            after_entry, side, entry, stop0, R, stop_first=STOP_FIRST_IF_BOTH_HIT
        )

        records.append({
            "Date": d, "Symbol": symbol, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": or_width,
            "EntryTime": entry_time, "Entry": entry,
            "Stop0": stop0, "Target1": T1, "Target2": T2, "Target3": T3,
            "Outcome_Ladder": outcome, "ExitPx_Ladder": float(exit_px) if exit_px is not None else None,
            "ExitTime_Ladder": day.loc[exit_idx, "EventAt"] if (exit_idx is not None and exit_idx in day.index) else None,
            "PnL_Ladder": float(realized_pnl),
            "Hit_T1": bool(hits.get("Hit_T1", False)),
            "Hit_T2": bool(hits.get("Hit_T2", False)),
            "Hit_T3": bool(hits.get("Hit_T3", False)),
            "Stop_Level_On_Exit": hits.get("Stop_Level_On_Exit", None),
        })

    return pd.DataFrame(records)

# =============================
# Batch runner
# =============================

def chunked(seq: List[str], n: int) -> List[List[str]]:
    return [seq[i:i+n] for i in range(0, len(seq), n)]

def main():
    try:
        conn = psycopg.connect(CONN_STR)
    except Exception as e:
        print(f"DB connection failed: {e}")
        return

    with conn:
        symbols = fetch_all_symbols(conn)
        if not symbols:
            print(f"No symbols found for interval='{INTERVAL}'.")
            return

        batches = chunked(symbols, BATCH_SIZE)
        all_trades = []

        print(f"Found {len(symbols)} symbols for interval '{INTERVAL}'. Processing in {len(batches)} batch(es) of {BATCH_SIZE}...")

        for bi, batch in enumerate(batches, start=1):
            print(f"\n=== Batch {bi}/{len(batches)} ({len(batch)} symbols) ===")

            for symbol in batch:
                try:
                    df = fetch_symbol_data(conn, symbol)
                    if df.empty:
                        print(f"=== ORB 15m Retest — {symbol} (ladder 1R/2R/3R) ===")
                        print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
                        continue

                    trades = backtest_df(df, symbol)

                    if trades.empty:
                        print(f"=== ORB 15m Retest — {symbol} (ladder 1R/2R/3R) ===")
                        print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
                        continue

                    all_trades.append(trades)

                    # Per-symbol metrics + print in required format
                    total_trades, total_pnl, winrate, profit_factor = calc_metrics(trades)

                    save_path = SAVE_DIR / f"orb_trades_{symbol}_ladder_1R_2R_3R.csv"
                    trades.to_csv(save_path, index=False)

                    print(f"=== ORB 15m Retest — {symbol} (ladder 1R/2R/3R) ===")
                    print(f"Trades: {total_trades} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {profit_factor:.3f}")
                    print(f"Saved trades → {save_path.resolve()}")

                except Exception as e:
                    print(f"[SKIP] {symbol}: {e}")

        # ===== OVERALL roll-up =====
        if not all_trades:
            print("\nNo trades generated for any symbols.")
            return

        overall = pd.concat(all_trades, ignore_index=True)
        o_trades, o_pnl, o_winrate, o_pf = calc_metrics(overall)

        save_all = SAVE_DIR / "orb_trades_ALL_ladder_1R_2R_3R.csv"
        overall.to_csv(save_all, index=False)

        print(f"\n=== ORB 15m Retest — OVERALL (ladder 1R/2R/3R) ===")
        print(f"Trades: {o_trades} | Total PnL: ${o_pnl:,.2f} | Win%: {o_winrate:.2f}% | Profit Factor: {o_pf:.3f}")
        print(f"Saved trades → {save_all.resolve()}")

if __name__ == "__main__":
    main()
```
