```python
import math
from dataclasses import dataclass
from pathlib import Path
from typing import List, Optional, Tuple, Dict, Set
from datetime import time as dtime

import numpy as np
import pandas as pd
import psycopg
import pytz

# =============================
# DB & RUN SETTINGS (EDIT ME)
# =============================

CONN_STR   = "host=24.128.195.68 port=8812 dbname=qdb user=admin password=quest"
FUTURES_TABLE = "futures_candles"
VOLATILITY_TABLE = "volatility_candles"
INTERVAL   = "1m"

# Futures and their volatility index pairs
FUTURES_VOL_MAP = {
    "ES": "VIX",    # S&P 500 E-mini -> VIX
    "NQ": "VXN",    # Nasdaq 100 E-mini -> VXN
    "YM": "VXD",    # Dow Jones E-mini -> VXD
    "RTY": "RVX"    # Russell 2000 E-mini -> RVX
}

# Optional UTC date window (ISO-8601)
START_DATE: Optional[str] = "2024-01-01"
END_DATE:   Optional[str] = None

# Output
SAVE_DIR = Path("./divergence_results")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# =============================
# MARKET & STRATEGY PARAMETERS
# =============================

TZ_LOCAL = "America/New_York"

# RTH session - day starts at 9:30 AM EST/EDT
SESSION_START = dtime(9, 30)
SESSION_END   = dtime(16, 0)

# Contract specifications
CONTRACT_SPECS = {
    "ES": {"point_value": 50.0, "tick_size": 0.25},
    "NQ": {"point_value": 20.0, "tick_size": 0.25},
    "YM": {"point_value": 5.0, "tick_size": 1.0},
    "RTY": {"point_value": 50.0, "tick_size": 0.1}
}

# =============================
# Helpers
# =============================

def to_ny(ts: pd.Series) -> pd.Series:
    """Convert UTC timestamps to New York timezone."""
    utc = pd.to_datetime(ts, utc=True)
    return utc.dt.tz_convert(TZ_LOCAL)

@dataclass
class Trade:
    symbol: str
    entry_time: pd.Timestamp
    side: str              # "long"/"short"
    entry: float
    stop: float
    divergence_symbol: str  # which symbol had the divergence
    exit_time: Optional[pd.Timestamp] = None
    exit_price: Optional[float] = None
    reason: Optional[str] = None  # "target" / "stop" / "eod"
    points: Optional[float] = None
    ticks: Optional[int] = None
    pnl_usd: Optional[float] = None

    def close_trade(self, exit_time: pd.Timestamp, exit_price: float, reason: str, tick_size: float, point_value: float):
        """Close the trade and calculate P&L."""
        self.exit_time = exit_time
        self.exit_price = exit_price
        self.reason = reason
        
        if self.side == "long":
            self.points = exit_price - self.entry
        else:  # short
            self.points = self.entry - exit_price
        
        self.ticks = int(round(self.points / tick_size))
        self.pnl_usd = self.points * point_value

# =============================
# Data Loading
# =============================

def load_futures_and_volatility(conn_str: str,
                                futures_table: str,
                                volatility_table: str,
                                futures_symbols: List[str],
                                vol_symbols: List[str],
                                interval: str,
                                start: Optional[str],
                                end: Optional[str]) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Load futures and volatility index data from separate QuestDB tables.
    """
    
    # Build WHERE clause for dates and interval
    def build_where(symbols: List[str]) -> str:
        where = [f"interval = '{interval}'"]
        if symbols:
            syms = ",".join("'" + s.replace("'", "''") + "'" for s in symbols)
            where.append(f"symbol IN ({syms})")
        if start:
            where.append(f"event_at >= TIMESTAMP '{start}'")
        if end:
            where.append(f"event_at < TIMESTAMP '{end}'")
        return " AND ".join(where)
    
    # Load futures
    futures_sql = f"""
        SELECT symbol, interval, event_at, open, high, low, close, volume
        FROM {futures_table}
        WHERE {build_where(futures_symbols)}
        ORDER BY symbol, event_at
    """
    
    # Load volatility indices
    vol_sql = f"""
        SELECT symbol, interval, event_at, open, high, low, close, volume
        FROM {volatility_table}
        WHERE {build_where(vol_symbols)}
        ORDER BY symbol, event_at
    """
    
    print(f"Loading futures from {futures_table}...")
    with psycopg.connect(conn_str) as conn:
        futures_df = pd.read_sql(futures_sql, conn)
    
    print(f"Loading volatility indices from {volatility_table}...")
    with psycopg.connect(conn_str) as conn:
        vol_df = pd.read_sql(vol_sql, conn)
    
    # Process timestamps
    if not futures_df.empty:
        futures_df["ts_ny"] = to_ny(futures_df["event_at"])
        futures_df["date_ny"] = futures_df["ts_ny"].dt.date
        futures_df["time_ny"] = futures_df["ts_ny"].dt.time
        print(f"Loaded {len(futures_df)} futures rows for {futures_df['symbol'].nunique()} symbols")
    
    if not vol_df.empty:
        vol_df["ts_ny"] = to_ny(vol_df["event_at"])
        vol_df["date_ny"] = vol_df["ts_ny"].dt.date
        vol_df["time_ny"] = vol_df["ts_ny"].dt.time
        print(f"Loaded {len(vol_df)} volatility rows for {vol_df['symbol'].nunique()} symbols")
    
    return futures_df, vol_df

# =============================
# Divergence Detection
# =============================

def detect_divergences(futures_df: pd.DataFrame, 
                       vol_df: pd.DataFrame,
                       future_symbol: str,
                       vol_symbol: str,
                       trend_direction: Optional[str] = None) -> pd.DataFrame:
    """
    Detect divergences between a future and its volatility index.
    
    Returns DataFrame with divergence signals including:
    - divergence_type: "high_div" or "low_div"
    - candle_color: "bull" or "bear"
    - signal: True if divergence + candle color match for entry
    """
    
    # Filter to session hours only (9:30 - 16:00)
    futures_df = futures_df[
        (futures_df["time_ny"] >= SESSION_START) & 
        (futures_df["time_ny"] <= SESSION_END)
    ].copy()
    
    vol_df = vol_df[
        (vol_df["time_ny"] >= SESSION_START) & 
        (vol_df["time_ny"] <= SESSION_END)
    ].copy()
    
    # Group by trading day
    results = []
    
    for date in sorted(futures_df["date_ny"].unique()):
        day_futures = futures_df[futures_df["date_ny"] == date].copy()
        day_vol = vol_df[vol_df["date_ny"] == date].copy()
        
        if day_futures.empty or day_vol.empty:
            continue
        
        # Calculate running highs and lows for the day
        day_futures["day_high"] = day_futures["high"].expanding().max()
        day_futures["day_low"] = day_futures["low"].expanding().min()
        day_futures["prev_day_high"] = day_futures["day_high"].shift(1)
        day_futures["prev_day_low"] = day_futures["day_low"].shift(1)
        
        day_vol["day_high"] = day_vol["high"].expanding().max()
        day_vol["day_low"] = day_vol["low"].expanding().min()
        day_vol["prev_day_high"] = day_vol["day_high"].shift(1)
        day_vol["prev_day_low"] = day_vol["day_low"].shift(1)
        
        # Detect new highs/lows
        day_futures["new_high"] = day_futures["high"] > day_futures["prev_day_high"]
        day_futures["new_low"] = day_futures["low"] < day_futures["prev_day_low"]
        
        # Merge with vol data by timestamp
        merged = pd.merge_asof(
            day_futures.sort_values("ts_ny"),
            day_vol[["ts_ny", "high", "low", "day_high", "prev_day_high", "day_low", "prev_day_low"]].sort_values("ts_ny"),
            on="ts_ny",
            direction="backward",
            suffixes=("", "_vol")
        )
        
        # Check if vol made new high/low
        merged["vol_new_high"] = merged["high_vol"] > merged["prev_day_high_vol"]
        merged["vol_new_low"] = merged["low_vol"] < merged["prev_day_low_vol"]
        
        # Detect divergences
        # High divergence: future new high BUT vol did NOT make new high
        merged["high_div"] = merged["new_high"] & ~merged["vol_new_high"]
        
        # Low divergence: future new low BUT vol did NOT make new low
        merged["low_div"] = merged["new_low"] & ~merged["vol_new_low"]
        
        # Determine candle color
        merged["candle_color"] = np.where(merged["close"] > merged["open"], "bull", "bear")
        
        # Entry signals based on divergence + candle color
        # Low div + green candle = potential long signal
        merged["low_div_signal"] = merged["low_div"] & (merged["candle_color"] == "bull")
        
        # High div + red candle = potential short signal
        merged["high_div_signal"] = merged["high_div"] & (merged["candle_color"] == "bear")
        
        results.append(merged)
    
    if not results:
        return pd.DataFrame()
    
    return pd.concat(results, ignore_index=True)

# =============================
# Trading Logic
# =============================

def simulate_trades(all_futures_data: Dict[str, pd.DataFrame],
                   divergence_signals: Dict[str, pd.DataFrame],
                   trend_filter: Optional[Dict[str, pd.DataFrame]] = None) -> List[Trade]:
    """
    Simulate trades based on divergence signals.
    
    When one future shows divergence, we can trade ANY of the 4 futures that also
    make new highs/lows on the same day.
    """
    
    trades = []
    
    # Process each day
    all_dates = set()
    for signals in divergence_signals.values():
        all_dates.update(signals["date_ny"].unique())
    
    for date in sorted(all_dates):
        # Track which futures made new highs/lows this day
        futures_new_highs = set()
        futures_new_lows = set()
        
        for symbol in all_futures_data.keys():
            day_data = all_futures_data[symbol][all_futures_data[symbol]["date_ny"] == date]
            if not day_data.empty:
                if day_data["new_high"].any():
                    futures_new_highs.add(symbol)
                if day_data["new_low"].any():
                    futures_new_lows.add(symbol)
        
        # Check for divergence signals
        for div_symbol, signals in divergence_signals.items():
            day_signals = signals[signals["date_ny"] == date]
            
            # Low divergence signals (potential longs)
            low_div_bars = day_signals[day_signals["low_div_signal"]]
            for _, signal_bar in low_div_bars.iterrows():
                # Trade all futures that made new lows
                for trade_symbol in futures_new_lows:
                    # Apply trend filter if provided
                    if trend_filter and trade_symbol in trend_filter:
                        trend_data = trend_filter[trade_symbol]
                        trend_at_signal = trend_data[
                            trend_data["ts_ny"] <= signal_bar["ts_ny"]
                        ].iloc[-1] if not trend_data.empty else None
                        
                        if trend_at_signal is not None and trend_at_signal.get("trend") != "up":
                            continue  # Skip if trend is not up
                    
                    # Entry at high of divergence candle
                    entry_price = signal_bar["high"]
                    stop_price = signal_bar["low"] - CONTRACT_SPECS[trade_symbol]["tick_size"]
                    
                    trade = Trade(
                        symbol=trade_symbol,
                        entry_time=signal_bar["ts_ny"],
                        side="long",
                        entry=entry_price,
                        stop=stop_price,
                        divergence_symbol=div_symbol
                    )
                    trades.append(trade)
            
            # High divergence signals (potential shorts)
            high_div_bars = day_signals[day_signals["high_div_signal"]]
            for _, signal_bar in high_div_bars.iterrows():
                # Trade all futures that made new highs
                for trade_symbol in futures_new_highs:
                    # Apply trend filter if provided
                    if trend_filter and trade_symbol in trend_filter:
                        trend_data = trend_filter[trade_symbol]
                        trend_at_signal = trend_data[
                            trend_data["ts_ny"] <= signal_bar["ts_ny"]
                        ].iloc[-1] if not trend_data.empty else None
                        
                        if trend_at_signal is not None and trend_at_signal.get("trend") != "down":
                            continue  # Skip if trend is not down
                    
                    # Entry at low of divergence candle
                    entry_price = signal_bar["low"]
                    stop_price = signal_bar["high"] + CONTRACT_SPECS[trade_symbol]["tick_size"]
                    
                    trade = Trade(
                        symbol=trade_symbol,
                        entry_time=signal_bar["ts_ny"],
                        side="short",
                        entry=entry_price,
                        stop=stop_price,
                        divergence_symbol=div_symbol
                    )
                    trades.append(trade)
    
    # Exit trades based on stop or end of day
    for trade in trades:
        if trade.exit_time is not None:
            continue  # Already closed
        
        symbol_data = all_futures_data[trade.symbol]
        after_entry = symbol_data[symbol_data["ts_ny"] > trade.entry_time]
        after_entry = after_entry[after_entry["date_ny"] == trade.entry_time.date()]
        
        specs = CONTRACT_SPECS[trade.symbol]
        
        for _, bar in after_entry.iterrows():
            # Check for stop hit
            if trade.side == "long":
                if bar["low"] <= trade.stop:
                    trade.close_trade(
                        bar["ts_ny"], trade.stop, "stop",
                        specs["tick_size"], specs["point_value"]
                    )
                    break
            else:  # short
                if bar["high"] >= trade.stop:
                    trade.close_trade(
                        bar["ts_ny"], trade.stop, "stop",
                        specs["tick_size"], specs["point_value"]
                    )
                    break
        
        # If still open, close at end of day
        if trade.exit_time is None and not after_entry.empty:
            last_bar = after_entry.iloc[-1]
            trade.close_trade(
                last_bar["ts_ny"], last_bar["close"], "eod",
                specs["tick_size"], specs["point_value"]
            )
    
    return trades

# =============================
# Main Execution
# =============================

def main():
    # Build lists of symbols
    futures_symbols = list(FUTURES_VOL_MAP.keys())
    vol_symbols = list(FUTURES_VOL_MAP.values())
    
    print(f"Futures symbols: {futures_symbols}")
    print(f"Volatility symbols: {vol_symbols}")
    
    # Load data from both tables
    futures_df, vol_df = load_futures_and_volatility(
        CONN_STR, 
        FUTURES_TABLE, 
        VOLATILITY_TABLE,
        futures_symbols,
        vol_symbols,
        INTERVAL, 
        START_DATE, 
        END_DATE
    )
    
    if futures_df.empty:
        print("No futures data loaded. Exiting.")
        return
    
    if vol_df.empty:
        print("No volatility data loaded. Exiting.")
        return
    
    # Check which symbols were actually found
    found_futures = futures_df["symbol"].unique().tolist()
    found_vol = vol_df["symbol"].unique().tolist()
    
    print(f"\nFound futures: {found_futures}")
    print(f"Found volatility indices: {found_vol}")
    
    missing_futures = set(futures_symbols) - set(found_futures)
    missing_vol = set(vol_symbols) - set(found_vol)
    
    if missing_futures:
        print(f"WARNING: Missing futures: {missing_futures}")
    if missing_vol:
        print(f"WARNING: Missing volatility indices: {missing_vol}")
    
    # Separate data by symbol
    futures_data = {}
    vol_data = {}
    
    for symbol in FUTURES_VOL_MAP.keys():
        if symbol in found_futures:
            futures_data[symbol] = futures_df[futures_df["symbol"] == symbol].copy()
        else:
            print(f"Skipping {symbol} - no data")
            
    for symbol in FUTURES_VOL_MAP.values():
        if symbol in found_vol:
            vol_data[symbol] = vol_df[vol_df["symbol"] == symbol].copy()
        else:
            print(f"Skipping {symbol} - no data")
    
    # Detect divergences for each future/vol pair
    print("\nDetecting divergences...")
    divergence_signals = {}
    
    for future_symbol, vol_symbol in FUTURES_VOL_MAP.items():
        if futures_data[future_symbol].empty or vol_data[vol_symbol].empty:
            print(f"Skipping {future_symbol}/{vol_symbol} - missing data")
            continue
        
        print(f"Processing {future_symbol} vs {vol_symbol}...")
        signals = detect_divergences(
            futures_data[future_symbol],
            vol_data[vol_symbol],
            future_symbol,
            vol_symbol
        )
        divergence_signals[future_symbol] = signals
        
        # Add tracking for new highs/lows
        futures_data[future_symbol] = signals
    
    # Simulate trades
    print("\nSimulating trades...")
    trades = simulate_trades(futures_data, divergence_signals)
    
    print(f"\nTotal trades: {len(trades)}")
    
    # Convert to DataFrame for analysis
    if trades:
        trades_df = pd.DataFrame([
            {
                "symbol": t.symbol,
                "entry_time": t.entry_time,
                "side": t.side,
                "entry": t.entry,
                "stop": t.stop,
                "exit_time": t.exit_time,
                "exit_price": t.exit_price,
                "reason": t.reason,
                "points": t.points,
                "ticks": t.ticks,
                "pnl_usd": t.pnl_usd,
                "divergence_symbol": t.divergence_symbol
            }
            for t in trades
        ])
        
        # Save results
        output_file = SAVE_DIR / f"divergence_trades_{pd.Timestamp.now().strftime('%Y%m%d_%H%M%S')}.csv"
        trades_df.to_csv(output_file, index=False)
        print(f"\nTrades saved to: {output_file}")
        
        # Print summary statistics
        print("\n" + "="*60)
        print("PERFORMANCE SUMMARY")
        print("="*60)
        
        for symbol in sorted(trades_df["symbol"].unique()):
            symbol_trades = trades_df[trades_df["symbol"] == symbol]
            wins = symbol_trades[symbol_trades["pnl_usd"] > 0]
            losses = symbol_trades[symbol_trades["pnl_usd"] <= 0]
            
            print(f"\n{symbol}:")
            print(f"  Total trades: {len(symbol_trades)}")
            print(f"  Wins: {len(wins)} ({len(wins)/len(symbol_trades)*100:.1f}%)")
            print(f"  Losses: {len(losses)} ({len(losses)/len(symbol_trades)*100:.1f}%)")
            print(f"  Total P&L: ${symbol_trades['pnl_usd'].sum():.2f}")
            print(f"  Avg P&L per trade: ${symbol_trades['pnl_usd'].mean():.2f}")
            if len(wins) > 0:
                print(f"  Avg win: ${wins['pnl_usd'].mean():.2f}")
            if len(losses) > 0:
                print(f"  Avg loss: ${losses['pnl_usd'].mean():.2f}")
        
        print(f"\n{'='*60}")
        print(f"OVERALL: ${trades_df['pnl_usd'].sum():.2f}")
        print(f"{'='*60}\n")
    else:
        print("No trades generated.")

if __name__ == "__main__":
    main()
```
