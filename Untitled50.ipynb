```python
# =======================
# ORB 15m Retest + ML Meta-Label Filter (LOCAL ONLY)
# - Reads from QuestDB/Postgres
# - Computes features at each ORB retest entry (no leakage)
# - Labels from realized outcome (PnL>0 => 1 else 0)
# - Trains LightGBM (fallback: Logistic) with walk-forward CV
# - Applies EV-based threshold and re-runs backtest with ML filter
# - Prints rule-only vs ML-filtered stats per symbol and overall
# =======================

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
from typing import List, Tuple, Optional, Dict
import numpy as np
import pandas as pd
import psycopg
from datetime import timedelta

# ===== Try LightGBM; fallback to Logistic if unavailable =====
USE_LGBM = True
try:
    from lightgbm import LGBMClassifier
except Exception:
    USE_LGBM = False
from sklearn.linear_model import LogisticRegression
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import roc_auc_score

# =============================
# CONFIG
# =============================
CONN_STR   = "host=24.128.195.68 port=8812 dbname=qdb user=admin password=quest"
TABLE      = "equities_candles"
INTERVAL   = "1m"
BATCH_SIZE = 20
MAX_SYMBOLS: Optional[int] = None   # None = all

START_DATE: Optional[str] = None    # e.g. "2018-01-01"
END_DATE:   Optional[str] = None

SAVE_DIR = Path(".")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

TZ_LOCAL = "America/New_York"
REG_START = "09:30"
OR_END    = "09:45"
REG_END   = "16:00"
STOP_FIRST_IF_BOTH_HIT = True

# =============================
# TIME / STRATEGY HELPERS
# =============================
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce", utc=True)
    return ts.dt.tz_convert(TZ_LOCAL)

def first_breakout_close(df_after_or, orh, orl):
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return int(df_after_breakout_close.index[touched].min())

def simulate_path(after_entry_df, side, entry, stop, target, stop_first=True):
    for idx, row in after_entry_df.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop and hit_tp:
            return (("STOP", stop, idx) if stop_first else ("TP", target, idx))
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    last_idx = int(after_entry_df.index[-1])
    return "EOD", float(after_entry_df.loc[last_idx, "Close"]), last_idx

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

def calc_metrics(trades_df):
    pnl = trades_df["PnL_1to2"].fillna(0.0)
    wins = int((pnl > 0).sum())
    losses = int((pnl < 0).sum())
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = float(pnl[pnl > 0].sum())
    gross_neg = float(-pnl[pnl < 0].sum())
    profit_factor = (gross_pos / gross_neg) if gross_neg > 0 else float("inf") if total_trades>0 else 0.0
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, float(profit_factor)

# =============================
# DB LOADERS
# =============================
def fetch_all_symbols(conn) -> List[str]:
    q = f"""SELECT DISTINCT symbol FROM {TABLE} WHERE interval=%s ORDER BY symbol"""
    df = pd.read_sql(q, conn, params=(INTERVAL,))
    syms = df["symbol"].dropna().astype(str).tolist()
    if MAX_SYMBOLS:
        syms = syms[:MAX_SYMBOLS]
    return syms

def fetch_symbol_data(conn, symbol: str) -> pd.DataFrame:
    base = f"""
        SELECT event_at, open, high, low, close, volume
        FROM {TABLE}
        WHERE symbol=%s AND interval=%s
    """
    params = [symbol, INTERVAL]
    if START_DATE and END_DATE:
        base += " AND event_at >= %s AND event_at < %s"
        params.extend([START_DATE, END_DATE])
    elif START_DATE:
        base += " AND event_at >= %s"
        params.append(START_DATE)
    elif END_DATE:
        base += " AND event_at < %s"
        params.append(END_DATE)
    base += " ORDER BY event_at"

    df = pd.read_sql(base, conn, params=tuple(params))
    if df.empty:
        return df
    df = df.rename(columns={"event_at":"EventAt","open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"})
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    return df.loc[session].copy()

# =============================
# FEATURE ENGINEERING (no leakage)
# =============================
def compute_features_for_entry(day_df: pd.DataFrame, retest_idx: int, side: str, orh: float, orl: float) -> Dict[str, float]:
    """
    Features that use ONLY info up to the entry bar (inclusive).
    """
    up = day_df.loc[:retest_idx]

    entry_price = orh if side == "long" else orl
    or_width = orh - orl
    prev_pos = up.index.get_loc(retest_idx)
    prev_idx = up.index[prev_pos - 1] if prev_pos > 0 else retest_idx
    prev = up.loc[prev_idx]

    # Intraday stats so far
    high_so_far = float(up["High"].max())
    low_so_far  = float(up["Low"].min())
    range_so_far = (high_so_far - low_so_far) / max(1e-9, entry_price)

    # Local vol proxy (last 30 bars)
    lastN = up.tail(30)
    hl_rng = lastN["High"] - lastN["Low"]
    hl_range_mean = float(hl_rng.mean()) if len(lastN) else 0.0
    hl_range_std  = float(hl_rng.std(ddof=1)) if len(lastN) > 1 else 0.0

    # Wick balance of previous bar
    prev_body_hi = max(prev["Open"], prev["Close"])
    prev_body_lo = min(prev["Open"], prev["Close"])
    upper_wick = max(0.0, prev["High"] - prev_body_hi)
    lower_wick = max(0.0, prev_body_lo - prev["Low"])
    wick_ratio = (upper_wick + 1e-9) / (lower_wick + 1e-9)

    # Minutes since REG_START on the SAME day, TZ-aware
    # Ensure Timedelta uses HH:MM:SS
    session_open = up.iloc[-1]["EventAt"].normalize() + pd.to_timedelta(REG_START + ":00")
    minutes_since_open = int((up.iloc[-1]["EventAt"] - session_open).total_seconds() // 60)

    feats = {
        "OR_Width": or_width,
        "OR_Width_Pct": or_width / max(1e-9, entry_price),
        "SideLong": 1 if side == "long" else 0,
        "RangeSoFar_Pct": range_so_far,
        "HLRange_mean": hl_range_mean,
        "HLRange_std": hl_range_std,
        "Prev_Wick_Ratio": wick_ratio,
        "BarVol_mean30": float(lastN["Volume"].mean()) if len(lastN) else 0.0,
        "BarVol_std30": float(lastN["Volume"].std(ddof=1)) if len(lastN) > 1 else 0.0,
        "MinutesSinceOpen": minutes_since_open,
    }
    return feats


# =============================
# Build Events (entries), Features, Labels
# =============================
def build_events_features_labels(df: pd.DataFrame, symbol: str) -> pd.DataFrame:
    """
    Returns one row per *candidate trade* with:
      - features (X)
      - realized outcome (PnL_1to2) and label (LabelWin: 1/0)
    """
    rows = []
    for d, day in df.groupby("date"):
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop       = orl if side == "long" else orh
        target     = entry + ( 2 * or_width if side == "long" else -2 * or_width)
        after_entry = day.loc[day.index > retest_idx]

        if after_entry.empty:
            outcome, exit_px = "EOD", entry
            exit_idx = retest_idx
        else:
            outcome, exit_px, exit_idx = simulate_path(after_entry, side, entry, stop, target, STOP_FIRST_IF_BOTH_HIT)

        pnl = pnl_one_share(side, entry, exit_px)
        feats = compute_features_for_entry(day, retest_idx, side, orh, orl)

        rows.append({
            "Date": d, "Symbol": symbol, "EntryTime": entry_time, "Side": side,
            "Entry": entry, "Stop": stop, "Target_1to2": target,
            "Outcome_1to2": outcome, "ExitPx_1to2": exit_px,
            "ExitTime_1to2": day.loc[exit_idx, "EventAt"] if exit_idx in day.index else None,
            "PnL_1to2": pnl,
            "LabelWin": 1 if pnl > 0 else 0,
            **feats
        })
    return pd.DataFrame(rows)

# =============================
# Walk-Forward CV & Training
# =============================
def make_walk_forward_splits(dates: np.ndarray, n_folds: int = 5):
    """
    dates: sorted unique date array for events.
    Returns list of (train_idx, valid_idx) over the *event rows* indices.
    """
    uniq = np.array(sorted(pd.unique(dates)))
    if len(uniq) < n_folds + 1:
        n_folds = max(2, min(3, len(uniq)//2))
    folds = []
    # Split the unique dates into sequential folds
    bins = np.array_split(uniq, n_folds)
    # For each fold k, train on all prior bins, validate on bin k
    cum = []
    for k, b in enumerate(bins):
        if k == 0:
            cum = list(b)
            continue
        train_dates = np.concatenate(bins[:k])
        valid_dates = b
        folds.append((train_dates, valid_dates))
    return folds

def train_meta_model(events_df: pd.DataFrame):
    if events_df.empty:
        raise ValueError("No events to train on.")

    feature_cols = [c for c in events_df.columns if c not in
                    ("Date","Symbol","EntryTime","Side","Entry","Stop","Target_1to2",
                     "Outcome_1to2","ExitPx_1to2","ExitTime_1to2","PnL_1to2","LabelWin")]

    X = events_df[feature_cols].astype(float).values
    y = events_df["LabelWin"].astype(int).values
    dates = events_df["Date"].values

    # Standardize a few skewed features lightly (avoid leakage by fitting per train fold)
    # We'll let tree models handle scaling; for Logistic fallback, CalibratedClassifier handles probs.

    folds = make_walk_forward_splits(dates, n_folds=5)
    if not folds:
        # Not enough dates; just fit on all and use simple calibration
        base = LGBMClassifier(n_estimators=300, max_depth=-1, learning_rate=0.05,
                              subsample=0.8, colsample_bytree=0.8, random_state=42) if USE_LGBM \
               else LogisticRegression(max_iter=2000, C=1.0)
        model = CalibratedClassifierCV(base, method="isotonic", cv=3)
        model.fit(X, y)
        # Use empirical gains/losses overall
        G = float(events_df.loc[events_df["LabelWin"]==1, "PnL_1to2"].mean()) if (events_df["LabelWin"]==1).any() else 1.0
        L = float(-events_df.loc[events_df["LabelWin"]==0, "PnL_1to2"].mean()) if (events_df["LabelWin"]==0).any() else 1.0
        p_star = (L) / (G + L) if (G + L) > 0 else 0.5
        return model, feature_cols, p_star

    # Walk-forward training with threshold selection by EV on validation
    best_pstar_list, aucs = [], []
    models = []
    for train_dates, valid_dates in folds:
        tr_idx = events_df["Date"].isin(train_dates)
        va_idx = events_df["Date"].isin(valid_dates)

        Xtr, ytr = events_df.loc[tr_idx, feature_cols].values, events_df.loc[tr_idx, "LabelWin"].values
        Xva, yva = events_df.loc[va_idx, feature_cols].values, events_df.loc[va_idx, "LabelWin"].values

        base = LGBMClassifier(n_estimators=400, learning_rate=0.05, subsample=0.8,
                              colsample_bytree=0.8, num_leaves=63, random_state=42) if USE_LGBM \
               else LogisticRegression(max_iter=3000, C=0.5)

        model = CalibratedClassifierCV(base, method="isotonic", cv=3)
        model.fit(Xtr, ytr)

        p = model.predict_proba(Xva)[:,1]
        try:
            aucs.append(roc_auc_score(yva, p))
        except Exception:
            pass

        # Empirical EV threshold from TRAIN fold
        G = float(events_df.loc[tr_idx & (events_df["LabelWin"]==1), "PnL_1to2"].mean()) if (events_df.loc[tr_idx,"LabelWin"]==1).any() else 1.0
        L = float(-events_df.loc[tr_idx & (events_df["LabelWin"]==0), "PnL_1to2"].mean()) if (events_df.loc[tr_idx,"LabelWin"]==0).any() else 1.0
        p_star = (L) / (G + L) if (G + L) > 0 else 0.5
        best_pstar_list.append(p_star)
        models.append(model)

    # Refit on ALL data with same hyperparams (common quant practice after CV)
    base_final = LGBMClassifier(n_estimators=400, learning_rate=0.05, subsample=0.8,
                                colsample_bytree=0.8, num_leaves=63, random_state=42) if USE_LGBM \
                 else LogisticRegression(max_iter=3000, C=0.5)
    model_final = CalibratedClassifierCV(base_final, method="isotonic", cv=3)
    model_final.fit(X, y)

    # Use median p* across folds for robustness
    p_star_final = float(np.median(best_pstar_list)) if best_pstar_list else 0.33
    if np.isnan(p_star_final): p_star_final = 0.33

    return model_final, feature_cols, p_star_final

# =============================
# Batch Runner with ML
# =============================
def chunked(seq: List[str], n: int) -> List[List[str]]:
    return [seq[i:i+n] for i in range(0, len(seq), n)]

def main():
    # ---- Connect and enumerate symbols ----
    try:
        conn = psycopg.connect(CONN_STR)
    except Exception as e:
        print(f"DB connection failed: {e}")
        return

    all_rule_trades = []
    per_symbol_rule_metrics = {}
    per_symbol_ml_metrics   = {}

    with conn:
        symbols = fetch_all_symbols(conn)
        if not symbols:
            print(f"No symbols found for interval='{INTERVAL}'.")
            return

        batches = chunked(symbols, BATCH_SIZE)
        print(f"Found {len(symbols)} symbols for '{INTERVAL}'. Running in {len(batches)} batch(es) of {BATCH_SIZE}.\n")

        # ===== 1) Build dataset across all symbols (features + labels) =====
        events_all = []

        for bi, batch in enumerate(batches, start=1):
            print(f"=== Batch {bi}/{len(batches)} ===")
            for symbol in batch:
                try:
                    df = fetch_symbol_data(conn, symbol)
                    if df.empty:
                        print(f"[{symbol}] no data.")
                        continue
                    ev = build_events_features_labels(df, symbol)
                    if ev.empty:
                        print(f"[{symbol}] no ORB entries.")
                        continue
                    events_all.append(ev)

                    # Save rule-only per-symbol CSV (same as before)
                    save_path = SAVE_DIR / f"orb_trades_{symbol}_one_share_RR1to2.csv"
                    ev.to_csv(save_path, index=False)

                    # rule-only metrics
                    t, pnl, wr, pf = calc_metrics(ev)
                    per_symbol_rule_metrics[symbol] = (t, pnl, wr, pf)
                    all_rule_trades.append(ev)

                    print(f"=== ORB 15m Retest — {symbol} (1 share, RR 1:2) ===")
                    print(f"Trades: {t} | Total PnL: ${pnl:,.2f} | Win%: {wr:.2f}% | Profit Factor: {pf:.3f}")
                    print(f"Saved trades → {save_path.resolve()}")
                except Exception as e:
                    print(f"[SKIP] {symbol}: {e}")

        if not events_all:
            print("\nNo events found across symbols.")
            return

        events_all = pd.concat(events_all, ignore_index=True)

        # ===== 2) Train ML meta-model locally =====
        print("\nTraining meta-label model (local only)...")
        model, feat_cols, p_star = train_meta_model(events_all)
        print(f"Model: {'LightGBM' if USE_LGBM else 'LogisticRegression'} (calibrated)")
        print(f"Features ({len(feat_cols)}): {feat_cols}")
        print(f"Probability threshold (EV-based): {p_star:.3f}")

        # ===== 3) Apply ML filter per symbol and print ML-filtered stats =====
        overall_ml = []
        for symbol, ev in events_all.groupby("Symbol"):
            Xs = ev[feat_cols].astype(float).values
            ps = model.predict_proba(Xs)[:,1]
            take = ps >= p_star
            ev_ml = ev.loc[take].copy()
            overall_ml.append(ev_ml)

            t, pnl, wr, pf = calc_metrics(ev_ml)
            per_symbol_ml_metrics[symbol] = (t, pnl, wr, pf)

            print(f"\n=== ORB 15m Retest + ML Filter — {symbol} ===")
            print(f"Kept trades: {t} | Total PnL: ${pnl:,.2f} | Win%: {wr:.2f}% | PF: {pf:.3f} | (kept {take.mean()*100:.1f}% of signals)")

        # ===== 4) Overall roll-ups =====
        rule_overall = pd.concat(all_rule_trades, ignore_index=True)
        ml_overall   = pd.concat(overall_ml, ignore_index=True)

        rt, rpnl, rwr, rpf = calc_metrics(rule_overall)
        mt, mpnl, mwr, mpf = calc_metrics(ml_overall)

        save_all_rule = SAVE_DIR / "orb_trades_ALL_rule_only.csv"
        save_all_ml   = SAVE_DIR / "orb_trades_ALL_ML_filtered.csv"
        rule_overall.to_csv(save_all_rule, index=False)
        ml_overall.to_csv(save_all_ml, index=False)

        print("\n=== ORB 15m Retest — OVERALL (Rule-only) ===")
        print(f"Trades: {rt} | Total PnL: ${rpnl:,.2f} | Win%: {rwr:.2f}% | PF: {rpf:.3f}")
        print(f"Saved trades → {save_all_rule.resolve()}")

        print("\n=== ORB 15m Retest + ML Filter — OVERALL ===")
        print(f"Trades: {mt} | Total PnL: ${mpnl:,.2f} | Win%: {mwr:.2f}% | PF: {mpf:.3f}")
        print(f"Saved trades → {save_all_ml.resolve()}")

        # ===== 5) Print a compact “model card” =====
        print("\n--- Model Card (local) ---")
        print(f"Classifier: {'LightGBM' if USE_LGBM else 'LogisticRegression'} + Isotonic calibration")
        try:
            # quick AUC on whole set (not a test metric; for sanity only)
            auc_all = roc_auc_score(rule_overall["LabelWin"].values,
                                    model.predict_proba(rule_overall[feat_cols].values)[:,1])
            print(f"AUC (in-sample sanity): {auc_all:.3f}")
        except Exception:
            pass
        print(f"Threshold p*: {p_star:.3f} (EV-based)")
        print(f"Feature count: {len(feat_cols)}")
        print("--------------------------")

if __name__ == "__main__":
    main()
```
