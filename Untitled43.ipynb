```python
# orb_15m_retest_spy_one_share_RR_1_to_2.py
# Implements EXACT spec:
#  - OR = first 15m candle (09:30–09:45 ET)
#  - Wait for first CLOSE outside OR (breakout confirmation)
#  - Then place entry at broken level and wait for a RETEST touch to fill
#  - 1 share, Stop at opposite OR side, Target = 2R (1:2)
#  - EOD flat if neither hit

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet\SPY.parquet"
SYMBOL   = "SPY"
TZ_LOCAL = "America/New_York"

REG_START = "09:30"   # session start
OR_END    = "09:45"   # end of first 15-min bar (exclusive)
REG_END   = "16:00"   # session end

STOP_FIRST_IF_BOTH_HIT = True
SAVE_TRADES_CSV = "orb_trades_SPY_one_share_RR1to2.csv"

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    """
    Robustly parse EventAt:
    - If tz-naive: assume UTC then convert to New York.
    - If tz-aware: convert to New York.
    """
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    """Return ('long'/'short', idx) for the first bar *closing* outside the OR."""
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    """First bar that *touches* the level after the breakout close."""
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def simulate_path(after_entry_df, side, entry, stop, target, stop_first=True):
    """
    Walk forward bar-by-bar after the entry.
    Returns outcome ('TP'/'STOP'/'EOD'), exit_price, exit_idx.
    """
    for idx, row in after_entry_df.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop and hit_tp:
            return ("STOP", stop, idx) if stop_first else ("TP", target, idx)
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    # neither hit by session end → exit at last close
    last_idx = after_entry_df.index[-1]
    return "EOD", float(after_entry_df.loc[last_idx, "Close"]), last_idx

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

# ========= MAIN =========
def main():
    # Load and normalize time
    df = pd.read_parquet(PARQUET)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"Parquet missing required column: {c}")
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Keep only regular session minutes
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    records = []

    for d, day in df.groupby("date"):
        # Opening range: first 15-minute bar [09:30, 09:45)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # After OR window (from 09:45 to close)
        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        # 1) Wait for first confirmed breakout by CLOSE
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # 2) After that breakout close, place retest entry at the broken level
        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        # 3) Wait for the first bar that *touches* the broken level → entry
        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop       = orl if side == "long" else orh
        target     = entry + ( 2 * or_width if side == "long" else -2 * or_width)

        # If the bar that retests moves through both stop/target intrabar,
        # simulate forward starting from the *next* bar after the retest
        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            # No bar after entry → flat at entry (no PnL)
            records.append({
                "Date": d, "Symbol": SYMBOL, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": or_width,
                "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
                "Outcome_1to2": "EOD", "ExitPx_1to2": entry, "ExitTime_1to2": entry_time,
                "PnL_1to2": 0.0,
            })
            continue

        outcome, exit_px, exit_idx = simulate_path(
            after_entry, side, entry, stop, target, stop_first=STOP_FIRST_IF_BOTH_HIT
        )
        pnl = pnl_one_share(side, entry, exit_px)

        records.append({
            "Date": d, "Symbol": SYMBOL, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": or_width,
            "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
            "Outcome_1to2": outcome, "ExitPx_1to2": exit_px,
            "ExitTime_1to2": day.loc[exit_idx, "EventAt"] if exit_idx in day.index else None,
            "PnL_1to2": pnl,
        })

    trades = pd.DataFrame(records)
    if trades.empty:
        print("No trades generated with current settings.")
        return

    # Daily green / red flag (did we profit with the 1:2 plan?)
    trades["Green_1to2"] = trades["PnL_1to2"] > 0

    # Metrics
    pnl = trades["PnL_1to2"].fillna(0.0)
    total_pnl = pnl.sum()
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    winrate = 100.0 * wins / max(1, wins + losses)
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = (gross_pos / gross_neg) if gross_neg > 0 else np.inf

    print(f"=== ORB 15m Retest — {SYMBOL} (1 share, RR 1:2) ===")
    print(f"Trades: {wins+losses} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {profit_factor:.3f}")
    print(f"Saved trades → {Path(SAVE_TRADES_CSV).resolve()}")

    trades.to_csv(SAVE_TRADES_CSV, index=False)

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_retest_RR_thirds_filters_robust.py
# 15-minute ORB breakout + RETEST entry, with your indicator/HTF filters and 1/3-1/3-1/3 exits.
# This version is HARDENED: never drops 'EventAt', and re-creates 'date' right before any operation using it.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET  = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet\SPY.parquet"
SYMBOL   = "SPY"
TZ_LOCAL = "America/New_York"

REG_START = "09:30"
OR_END    = "09:45"    # first 15-min candle is [09:30, 09:45)
REG_END   = "16:00"

# Lunch & cutoff
NO_TRADE_START = "12:00"
NO_TRADE_END   = "13:00"
LAST_ENTRY     = "15:00"
MAX_TRADES_PER_DAY = 2

# Filters toggles
USE_IBS                = True
USE_15M_EMA_TREND      = True
USE_15M_EMA_SLOPE      = True
USE_VWAP               = True
USE_RSI_14             = True
USE_ADX_14             = True
USE_OPPOSITE_WICK      = True
USE_BREAKOUT_CUSHION   = True
USE_DAILY_REGIME       = True
USE_DAILY_ATR_GATE     = True
USE_OPEN_VOL_QUANTILE  = True
USE_GAP_FILTER         = True
USE_SPY_CONFIRMATION   = True  # trading SPY: VWAP alignment is the same check

# Parameters
BREAKOUT_CUSHION_PCT = 0.0012      # 0.12%
EMA_SLOPE_MIN        = 0.0002      # 0.02% over last 5 bars (≈ 0.02%)
OPEN_VOL_LOOKBACK    = 60          # sessions
OPEN_VOL_Q           = 0.40
ATR_DAILY_MIN        = 0.15        # OR width >= 15% of ATR(14)
ATR_DAILY_MAX        = 1.20        # OR width <= 120% of ATR(14)
GAP_MIN              = 0.003       # 0.3%
GAP_MAX              = 0.04        # 4%
TRAIL_ATR_MULT       = 2.5         # trailing on last third after TP2
ATR_INTRADAY_LEN     = 14          # ATR length for intraday trailing

SAVE_TRADES_CSV = "orb_trades_SPY_one_share_FILTERS.csv"

# ========= HELPERS =========
def to_ny_series(ts):
    ts = pd.to_datetime(ts, errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC")
    return ts.dt.tz_convert(TZ_LOCAL)

def ensure_date_cols(df):
    """Guarantee EventAt (tz-aware NY) and date/t columns exist and are current."""
    if "EventAt" not in df.columns:
        # If it's index, pull it back; else fail loudly
        if isinstance(df.index, pd.DatetimeIndex):
            df = df.reset_index().rename(columns={"index": "EventAt"})
        else:
            raise KeyError("EventAt column is missing and index is not datetime.")
    df["EventAt"] = to_ny_series(df["EventAt"])
    df["date"] = df["EventAt"].dt.date
    df["t"] = df["EventAt"].dt.strftime("%H:%M")
    return df

def ema(s, n):
    return s.ewm(span=n, adjust=False).mean()

def rsi_14(close):
    delta = close.diff()
    up = delta.clip(lower=0.0)
    dn = -delta.clip(upper=0.0)
    roll_up = up.ewm(alpha=1/14, adjust=False).mean()
    roll_dn = dn.ewm(alpha=1/14, adjust=False).mean()
    rs = roll_up / roll_dn.replace(0, np.nan)
    return 100 - (100 / (1 + rs))

def adx_14(high, low, close, n=14):
    tr1 = (high - low)
    tr2 = (high - close.shift()).abs()
    tr3 = (low - close.shift()).abs()
    tr = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)

    up_move = high.diff().clip(lower=0.0)
    down_move = (-low.diff()).clip(lower=0.0)
    cond = up_move >= down_move
    dm_plus = up_move.where(cond, 0.0)
    dm_minus = down_move.where(~cond, 0.0)

    tr_n = tr.ewm(alpha=1/n, adjust=False).mean()
    dmp_n = dm_plus.ewm(alpha=1/n, adjust=False).mean()
    dmm_n = dm_minus.ewm(alpha=1/n, adjust=False).mean()

    dip = 100 * (dmp_n / tr_n.replace(0, np.nan))
    dim = 100 * (dmm_n / tr_n.replace(0, np.nan))
    dx = 100 * ((dip - dim).abs() / (dip + dim).replace(0, np.nan))
    return dx.ewm(alpha=1/n, adjust=False).mean()

def true_range(h,l,c_prev):
    return pd.DataFrame({
        "tr1": h - l,
        "tr2": (h - c_prev).abs(),
        "tr3": (l - c_prev).abs()
    }).max(axis=1)

def session_vwap_from_parts(high, low, close, vol, dates):
    tp = (high + low + close) / 3.0
    by = pd.Series(dates, index=close.index)
    cum_pv = (tp * vol).groupby(by).cumsum()
    cum_v  = vol.groupby(by).cumsum()
    return cum_pv / cum_v.replace(0, np.nan)

def slope_mask(series, lookback=5, min_frac=EMA_SLOPE_MIN, up=True):
    prev = series.shift(lookback)
    slope = (series - prev) / prev.replace(0, np.nan)
    return (slope >= min_frac) if up else (slope <= -min_frac)

def first_breakout_close(df_after_or, orh, orl, cushion_pct):
    above = df_after_or["Close"] > (orh * (1 + cushion_pct))
    below = df_after_or["Close"] < (orl * (1 - cushion_pct))
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    return None if not touched.any() else df_after_breakout_close.index[touched].min()

def opposite_wick_ok(row, side):
    rng = row["High"] - row["Low"]
    if rng <= 0: return False
    open_, close = row["Open"], row["Close"]
    if side == "long":
        upper_wick = row["High"] - max(open_, close)
        return (upper_wick / rng) <= 0.30
    else:
        lower_wick = min(open_, close) - row["Low"]
        return (lower_wick / rng) <= 0.30

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

# ========= MAIN =========
def main():
    # --- Load ---
    df = pd.read_parquet(PARQUET)
    need_cols = {"EventAt","Open","High","Low","Close","Volume"}
    missing = need_cols - set(df.columns)
    if missing:
        raise ValueError(f"Parquet missing required columns: {sorted(missing)}")

    df = ensure_date_cols(df)
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Regular session filter
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()
    df = ensure_date_cols(df)  # re-ensure after slicing

    # --- Build 15m bars per day ---
    def resample_15m_per_day(g):
        gi = g.set_index("EventAt")
        out = pd.DataFrame({
            "Open":  gi["Open"].resample("15T", label="right", closed="right").first(),
            "High":  gi["High"].resample("15T", label="right", closed="right").max(),
            "Low":   gi["Low"].resample("15T", label="right", closed="right").min(),
            "Close": gi["Close"].resample("15T", label="right", closed="right").last(),
            "Volume":gi["Volume"].resample("15T", label="right", closed="right").sum(),
        }).dropna(subset=["Open","High","Low","Close"])
        out = out.reset_index().rename(columns={"EventAt":"BarEnd"})
        out["date"] = g["date"].iloc[0]
        return out

    f15 = df.groupby("date", group_keys=False).apply(resample_15m_per_day)
    # 15m indicators
    f15["EMA20"] = ema(f15["Close"], 20)
    f15["EMA50"] = ema(f15["Close"], 50)
    # RSI/ADX per day (no lookahead across days)
    f15["RSI14"] = f15.groupby("date", group_keys=False).apply(lambda g: rsi_14(g["Close"]))
    f15["ADX14"] = f15.groupby("date", group_keys=False).apply(lambda g: adx_14(g["High"], g["Low"], g["Close"], 14))

    # Map 15m indicators to minute rows via merge_asof by 'date'
    map_cols = ["date","BarEnd","EMA20","EMA50","RSI14","ADX14"]
    f15_map = f15[map_cols].sort_values(["date","BarEnd"]).reset_index(drop=True)
    df = df.sort_values(["date","EventAt"]).reset_index(drop=True)
    df = pd.merge_asof(
        df, f15_map,
        left_on="EventAt", right_on="BarEnd",
        by="date", direction="backward", allow_exact_matches=True
    )
    # EMA slope masks on the mapped series
    df["EMA20_up"] = slope_mask(df["EMA20"], lookback=5, min_frac=EMA_SLOPE_MIN, up=True)
    df["EMA50_up"] = slope_mask(df["EMA50"], lookback=5, min_frac=EMA_SLOPE_MIN, up=True)
    df["EMA20_dn"] = slope_mask(df["EMA20"], lookback=5, min_frac=EMA_SLOPE_MIN, up=False)
    df["EMA50_dn"] = slope_mask(df["EMA50"], lookback=5, min_frac=EMA_SLOPE_MIN, up=False)

    # --- VWAP (session) & slope ---
    df["VWAP"] = session_vwap_from_parts(df["High"], df["Low"], df["Close"], df["Volume"], df["date"])
    df["VWAP_slope_up"] = df.groupby("date")["VWAP"].apply(lambda s: (s - s.shift(5)) > 0)
    df["VWAP_slope_dn"] = df.groupby("date")["VWAP"].apply(lambda s: (s - s.shift(5)) < 0)

    # --- Daily OHLC + regime (from intraday) ---
    daily = df.groupby("date").agg(
        Open=("Open","first"), High=("High","max"), Low=("Low","min"), Close=("Close","last")
    ).reset_index()
    daily["EMA20D"] = ema(daily["Close"], 20)
    daily["EMA50D"] = ema(daily["Close"], 50)
    tr_d = true_range(daily["High"], daily["Low"], daily["Close"].shift())
    daily["ATR14D"] = tr_d.ewm(alpha=1/14, adjust=False).mean()
    # gap %
    daily["PrevClose"] = daily["Close"].shift(1)
    daily["GapPct"] = (daily["Open"] - daily["PrevClose"]) / daily["PrevClose"]

    # Merge daily fields back
    df = df.merge(daily[["date","EMA20D","EMA50D","ATR14D","GapPct"]], on="date", how="left")

    # --- Opening 15m stats (IBS & volume) ---
    def opening_stats(g):
        m = (g["t"] >= REG_START) & (g["t"] < OR_END)
        if not m.any():
            return pd.Series({"ORH":np.nan,"ORL":np.nan,"ORClose":np.nan,"ORVol":np.nan})
        sub = g.loc[m]
        return pd.Series({
            "ORH": sub["High"].max(),
            "ORL": sub["Low"].min(),
            "ORClose": sub["Close"].iloc[-1],
            "ORVol": sub["Volume"].sum()
        })
    open15 = df.groupby("date").apply(opening_stats)
    open15["IBS"] = (open15["ORClose"] - open15["ORL"]) / (open15["ORH"] - open15["ORL"])
    open15["ORVol_q40"] = open15["ORVol"].rolling(OPEN_VOL_LOOKBACK, min_periods=10).quantile(OPEN_VOL_Q).shift(1)
    open15 = open15.reset_index()
    df = df.merge(open15[["date","ORH","ORL","IBS","ORVol","ORVol_q40"]], on="date", how="left")

    # --- Intraday ATR for trailing ---
    tr_i = true_range(df["High"], df["Low"], df["Close"].shift())
    df["ATRi14"] = tr_i.ewm(alpha=1/ATR_INTRADAY_LEN, adjust=False).mean()

    # ===== Sim loop per day =====
    records = []
    day_counts = {}

    # helpful index for lookups
    df = df.set_index("EventAt", drop=False)

    for d, day in df.groupby("date"):
        day_counts.setdefault(d, 0)

        # Opening range (first 15 min)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask,"High"].max())
        orl = float(day.loc[or_mask,"Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # Daily ATR gate
        atrd = float(day["ATR14D"].iloc[0]) if day["ATR14D"].notna().any() else np.nan
        if USE_DAILY_ATR_GATE and np.isfinite(atrd):
            if not (ATRDailyMin := atrd*ATR_DAILY_MIN) <= or_width <= (ATRDailyMax := atrd*ATR_DAILY_MAX):
                continue

        # Gap filter
        gap = float(day["GapPct"].iloc[0]) if day["GapPct"].notna().any() else 0.0
        if USE_GAP_FILTER and not (GAP_MIN <= abs(gap) <= GAP_MAX):
            continue

        # Opening volume quantile
        orvol = float(day["ORVol"].iloc[0]) if day["ORVol"].notna().any() else np.nan
        vol_q40 = float(day["ORVol_q40"].iloc[0]) if day["ORVol_q40"].notna().any() else np.nan
        if USE_OPEN_VOL_QUANTILE and np.isfinite(vol_q40) and orvol < vol_q40:
            continue

        # IBS bias
        ibs = float(day["IBS"].iloc[0]) if day["IBS"].notna().any() else np.nan

        # After OR window
        after_or = day.loc[day["t"] >= OR_END].copy()
        if after_or.empty:
            continue

        # Breakout close beyond cushion
        cushion = BREAKOUT_CUSHION_PCT if USE_BREAKOUT_CUSHION else 0.0
        side, brk_idx = first_breakout_close(after_or, orh, orl, cushion)
        if side is None:
            continue

        # Regime (daily EMAs)
        if USE_DAILY_REGIME:
            ema20d = float(day["EMA20D"].iloc[0]) if day["EMA20D"].notna().any() else np.nan
            ema50d = float(day["EMA50D"].iloc[0]) if day["EMA50D"].notna().any() else np.nan
            if side == "long" and not (ema20d > ema50d):  continue
            if side == "short" and not (ema20d < ema50d): continue

        # IBS side bias
        if USE_IBS:
            if side == "long" and not (ibs >= 0.55):  continue
            if side == "short" and not (ibs <= 0.45): continue

        # Indicator snapshot at breakout close
        brk_row = after_or.loc[brk_idx]

        if USE_15M_EMA_TREND:
            if side == "long" and not (brk_row["EMA20"] > brk_row["EMA50"]):  continue
            if side == "short" and not (brk_row["EMA20"] < brk_row["EMA50"]): continue

        if USE_15M_EMA_SLOPE:
            if side == "long" and not (bool(brk_row["EMA20_up"]) and bool(brk_row["EMA50_up"])):   continue
            if side == "short" and not (bool(brk_row["EMA20_dn"]) and bool(brk_row["EMA50_dn"])): continue

        if USE_RSI_14:
            if side == "long" and not (brk_row["RSI14"] >= 55): continue
            if side == "short" and not (brk_row["RSI14"] <= 45): continue

        if USE_ADX_14 and not (brk_row["ADX14"] >= 20):
            continue

        if USE_VWAP:
            if side == "long" and not (brk_row["Close"] > brk_row["VWAP"] and bool(brk_row["VWAP_slope_up"])): continue
            if side == "short" and not (brk_row["Close"] < brk_row["VWAP"] and bool(brk_row["VWAP_slope_dn"])): continue

        if USE_SPY_CONFIRMATION:
            pass  # same as VWAP checks for SPY here

        # Entry timing windows (avoid lunch chop & late entries)
        after_breakout_close = after_or.loc[after_or.index > brk_idx].copy()
        if after_breakout_close.empty:
            continue
        ok_time = ((after_breakout_close["t"] < NO_TRADE_START) | (after_breakout_close["t"] >= NO_TRADE_END)) \
                  & (after_breakout_close["t"] <= LAST_ENTRY)
        after_breakout_close = after_breakout_close.loc[ok_time]
        if after_breakout_close.empty:
            continue

        # Retest entry at broken level
        level = orh if side == "long" else orl
        retest_idx = first_retest_touch(after_breakout_close, level)
        if retest_idx is None:
            continue

        # Max trades/day
        if day_counts[d] >= MAX_TRADES_PER_DAY:
            continue

        entry_row = day.loc[day.index == retest_idx].iloc[0]
        if USE_OPPOSITE_WICK and not opposite_wick_ok(entry_row, side):
            continue

        entry_time = pd.Timestamp(entry_row["EventAt"])
        entry      = level
        stop       = orl if side == "long" else orh
        dist       = abs(entry - stop)
        if dist <= 0:
            continue

        # Targets
        tp1 = entry + ( dist if side=="long" else -dist)
        tp2 = entry + (2*dist if side=="long" else -2*dist)
        tp3 = entry + (3*dist if side=="long" else -3*dist)

        after_entry = day.loc[day.index > retest_idx].copy()
        if after_entry.empty:
            continue

        filled_tp1 = False
        filled_tp2 = False
        exit_px_final = None
        exit_time_final = None
        pnl_total = 0.0
        be_stop = stop
        trail_on = False
        trail_stop = None

        for idx, row in after_entry.iterrows():
            lo, hi, close = row["Low"], row["High"], row["Close"]
            atr_i = row["ATRi14"]

            active_stop = be_stop
            if trail_on and np.isfinite(atr_i):
                if side == "long":
                    trail_stop = max(trail_stop or -np.inf, close - TRAIL_ATR_MULT * atr_i)
                    active_stop = max(active_stop, trail_stop)
                else:
                    trail_stop = min(trail_stop or  np.inf, close + TRAIL_ATR_MULT * atr_i)
                    active_stop = min(active_stop, trail_stop)

            # conservative sequencing: stop first within bar
            hit_stop = (lo <= active_stop) if side=="long" else (hi >= active_stop)
            hit_tp1  = (hi >= tp1) if side=="long" else (lo <= tp1)
            hit_tp2  = (hi >= tp2) if side=="long" else (lo <= tp2)
            hit_tp3  = (hi >= tp3) if side=="long" else (lo <= tp3)

            if hit_stop and ((not filled_tp1 and hit_tp1) or (filled_tp1 and not filled_tp2 and hit_tp2) or (filled_tp2 and hit_tp3)):
                rem_thirds = (1/3.0) * ((0 if filled_tp1 else 1) + (0 if filled_tp2 else 1) + 1)
                pnl_total += rem_thirds * pnl_one_share(side, entry, active_stop)
                exit_px_final = active_stop
                exit_time_final = row["EventAt"]
                break

            if not filled_tp1 and hit_tp1:
                pnl_total += (1/3.0) * pnl_one_share(side, entry, tp1)
                filled_tp1 = True
                be_stop = entry  # move to break-even

            if filled_tp1 and not filled_tp2 and hit_tp2:
                pnl_total += (1/3.0) * pnl_one_share(side, entry, tp2)
                filled_tp2 = True
                trail_on = True  # enable trailing on last third

            if filled_tp1 and filled_tp2 and hit_tp3:
                pnl_total += (1/3.0) * pnl_one_share(side, entry, tp3)
                exit_px_final = tp3
                exit_time_final = row["EventAt"]
                break

            if trail_on:
                hit_trail = (lo <= active_stop) if side=="long" else (hi >= active_stop)
                if hit_trail:
                    pnl_total += (1/3.0) * pnl_one_share(side, entry, active_stop)
                    exit_px_final = active_stop
                    exit_time_final = row["EventAt"]
                    break

        if exit_px_final is None:
            # close remainder at EOD
            rem_thirds = (1/3.0) * ((0 if filled_tp1 else 1) + (0 if filled_tp2 else 1) + 1)
            eod_close = after_entry.iloc[-1]["Close"]
            pnl_total += rem_thirds * pnl_one_share(side, entry, eod_close)
            exit_px_final = eod_close
            exit_time_final = after_entry.iloc[-1]["EventAt"]

        records.append({
            "Date": d, "Symbol": SYMBOL, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": or_width,
            "EntryTime": entry_time, "Entry": entry, "Stop_init": stop,
            "TP1": tp1, "TP2": tp2, "TP3": tp3,
            "ExitPx_final": exit_px_final, "ExitTime_final": exit_time_final,
            "PnL_total": pnl_total,
            "Filled_TP1": filled_tp1, "Filled_TP2": filled_tp2,
            "Used_Trail": trail_on,
        })
        day_counts[d] += 1

    trades = pd.DataFrame(records)
    if trades.empty:
        print("No trades generated with current settings/filters.")
        return

    # Summary
    pnl = trades["PnL_total"].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    winrate = 100.0 * wins / max(1, wins + losses)
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = (gross_pos / gross_neg) if gross_neg > 0 else np.inf

    print(f"=== ORB 15m Retest + Filters — {SYMBOL} (1 share) ===")
    print(f"Trades: {wins+losses} | Total PnL: ${pnl.sum():,.2f} | Win%: {winrate:.2f}% | PF: {profit_factor:.3f}")
    print(f"Saved trades → {Path(SAVE_TRADES_CSV).resolve()}")

    trades.to_csv(SAVE_TRADES_CSV, index=False)

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_filters_quality.py
# 15m ORB breakout + RETEST entry with *quality-focused* filters & confirmations.
# Hardened pipeline: keeps EventAt/date intact; robust merges.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET  = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet\SPY.parquet"
SYMBOL   = "SPY"
TZ_LOCAL = "America/New_York"

REG_START = "09:30"
OR_END    = "09:45"    # first 15-min candle ends (exclusive)
REG_END   = "16:00"

# Lunch & cutoff (quality-biased)
NO_TRADE_START = "11:50"
NO_TRADE_END   = "13:10"
LAST_ENTRY     = "14:45"
MAX_TRADES_PER_DAY = 2

# ====== Filter toggles ======
USE_IBS                = True
USE_15M_EMA_TREND      = True
USE_15M_EMA_SLOPE      = True
USE_VWAP               = True
USE_RSI_14             = True
USE_ADX_14             = True
USE_OPPOSITE_WICK      = True
USE_BREAKOUT_CUSHION   = True
USE_DAILY_REGIME       = True
USE_DAILY_ATR_GATE     = True
USE_OPEN_VOL_QUANTILE  = True
USE_GAP_FILTER         = True
USE_SPY_CONFIRMATION   = True  # SPY VWAP alignment == our VWAP rule

# ====== Quality parameters ======
BREAKOUT_CUSHION_PCT = 0.0015   # 0.15% close beyond ORH/ORL
VWAP_SLOPE_LOOKBACK  = 5        # bars
EMA_SLOPE_MIN        = 0.0002   # 0.02% over 5 bars
OPEN_VOL_LOOKBACK    = 60       # sessions
OPEN_VOL_Q           = 0.45     # 45th percentile
ATR_DAILY_MIN        = 0.15     # OR width >= 15% of ATR(14)
ATR_DAILY_MAX        = 1.10     # OR width <= 110% of ATR(14)
GAP_MIN              = 0.002    # 0.2%
GAP_MAX              = 0.03     # 3%
OPPOSITE_WICK_MAX_FRAC = 0.30   # ≤30% of bar range
RSI_LONG_MIN         = 57
RSI_SHORT_MAX        = 43
ADX_MIN              = 22

# Extra confirmations to reduce whipsaws
RETEST_CLOSE_THROUGH          = True   # require close-back-through on retest bar
RETEST_CONFIRM_CUSHION_PCT    = 0.0003 # 0.03% beyond level in trade direction
ENGULFING_CONFIRM             = True   # 1-min engulfing pattern on entry bar

TRAIL_ATR_MULT       = 2.5      # trail last third after TP2
ATR_INTRADAY_LEN     = 14

SAVE_TRADES_CSV = "orb_trades_SPY_one_share_FILTERS.csv"

# ========= HELPERS =========
def to_ny_series(ts):
    ts = pd.to_datetime(ts, errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC")
    return ts.dt.tz_convert(TZ_LOCAL)

def ensure_date_cols(df):
    """Guarantee EventAt (tz-aware NY) and date/t columns exist and are current."""
    if "EventAt" not in df.columns:
        if isinstance(df.index, pd.DatetimeIndex):
            df = df.reset_index().rename(columns={"index": "EventAt"})
        else:
            raise KeyError("EventAt column is missing and index is not datetime.")
    df["EventAt"] = to_ny_series(df["EventAt"])
    df["date"] = df["EventAt"].dt.date
    df["t"] = df["EventAt"].dt.strftime("%H:%M")
    return df

def ema(s, n):
    return s.ewm(span=n, adjust=False).mean()

def rsi_14(close):
    delta = close.diff()
    up = delta.clip(lower=0.0)
    dn = -delta.clip(upper=0.0)
    roll_up = up.ewm(alpha=1/14, adjust=False).mean()
    roll_dn = dn.ewm(alpha=1/14, adjust=False).mean()
    rs = roll_up / roll_dn.replace(0, np.nan)
    return 100 - (100 / (1 + rs))

def adx_14(high, low, close, n=14):
    tr1 = (high - low)
    tr2 = (high - close.shift()).abs()
    tr3 = (low - close.shift()).abs()
    tr = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)

    up_move = high.diff().clip(lower=0.0)
    down_move = (-low.diff()).clip(lower=0.0)
    cond = up_move >= down_move
    dm_plus = up_move.where(cond, 0.0)
    dm_minus = down_move.where(~cond, 0.0)

    tr_n  = tr.ewm(alpha=1/n, adjust=False).mean()
    dmp_n = dm_plus.ewm(alpha=1/n, adjust=False).mean()
    dmm_n = dm_minus.ewm(alpha=1/n, adjust=False).mean()

    dip = 100 * (dmp_n / tr_n.replace(0, np.nan))
    dim = 100 * (dmm_n / tr_n.replace(0, np.nan))
    dx  = 100 * ((dip - dim).abs() / (dip + dim).replace(0, np.nan))
    return dx.ewm(alpha=1/n, adjust=False).mean()

def true_range(h,l,c_prev):
    return pd.DataFrame({
        "tr1": h - l,
        "tr2": (h - c_prev).abs(),
        "tr3": (l - c_prev).abs()
    }).max(axis=1)

def session_vwap_from_parts(high, low, close, vol, dates):
    tp = (high + low + close) / 3.0
    by = pd.Series(dates, index=close.index)
    cum_pv = (tp * vol).groupby(by).cumsum()
    cum_v  = vol.groupby(by).cumsum()
    return cum_pv / cum_v.replace(0, np.nan)

def slope_mask(series, lookback=5, min_frac=EMA_SLOPE_MIN, up=True):
    prev = series.shift(lookback)
    slope = (series - prev) / prev.replace(0, np.nan)
    return (slope >= min_frac) if up else (slope <= -min_frac)

def first_breakout_close(df_after_or, orh, orl, cushion_pct):
    above = df_after_or["Close"] > (orh * (1 + cushion_pct))
    below = df_after_or["Close"] < (orl * (1 - cushion_pct))
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    return None if not touched.any() else df_after_breakout_close.index[touched].min()

def opposite_wick_ok(row, side, max_frac=OPPOSITE_WICK_MAX_FRAC):
    rng = row["High"] - row["Low"]
    if rng <= 0: return False
    o, c = row["Open"], row["Close"]
    if side == "long":
        upper_wick = row["High"] - max(o, c)
        return (upper_wick / rng) <= max_frac
    else:
        lower_wick = min(o, c) - row["Low"]
        return (lower_wick / rng) <= max_frac

def engulfing_ok(prev_row, curr_row, side):
    if prev_row is None or curr_row is None:
        return False
    po, pc = prev_row["Open"], prev_row["Close"]
    co, cc = curr_row["Open"], curr_row["Close"]
    if side == "long":
        return (cc > po) and (co < pc)
    else:
        return (cc < po) and (co > pc)

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

# ========= MAIN =========
def main():
    # --- Load ---
    df = pd.read_parquet(PARQUET)
    need = {"EventAt","Open","High","Low","Close","Volume"}
    miss = need - set(df.columns)
    if miss:
        raise ValueError(f"Parquet missing required columns: {sorted(miss)}")

    df = ensure_date_cols(df)
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Regular session filter
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()
    df = ensure_date_cols(df)  # re-ensure after slicing

    # --- Build 15m bars per day ---
    def resample_15m_per_day(g):
        gi = g.set_index("EventAt")
        out = pd.DataFrame({
            "Open":  gi["Open"].resample("15T", label="right", closed="right").first(),
            "High":  gi["High"].resample("15T", label="right", closed="right").max(),
            "Low":   gi["Low"].resample("15T", label="right", closed="right").min(),
            "Close": gi["Close"].resample("15T", label="right", closed="right").last(),
            "Volume":gi["Volume"].resample("15T", label="right", closed="right").sum(),
        }).dropna(subset=["Open","High","Low","Close"])
        out = out.reset_index().rename(columns={"EventAt":"BarEnd"})
        out["date"] = g["date"].iloc[0]
        return out

    f15 = df.groupby("date", group_keys=False).apply(resample_15m_per_day)

    # 15m indicators
    f15["EMA20"] = ema(f15["Close"], 20)
    f15["EMA50"] = ema(f15["Close"], 50)
    f15["RSI14"] = f15.groupby("date", group_keys=False).apply(lambda g: rsi_14(g["Close"]))
    f15["ADX14"] = f15.groupby("date", group_keys=False).apply(lambda g: adx_14(g["High"], g["Low"], g["Close"], 14))

    # Map 15m indicators to minute rows via merge_asof by 'date'
    f15_map = f15[["date","BarEnd","EMA20","EMA50","RSI14","ADX14"]].sort_values(["date","BarEnd"]).reset_index(drop=True)
    df = df.sort_values(["date","EventAt"]).reset_index(drop=True)
    df = pd.merge_asof(
        df, f15_map,
        left_on="EventAt", right_on="BarEnd",
        by="date", direction="backward", allow_exact_matches=True
    )

    # EMA slope masks (quality)
    df["EMA20_up"] = slope_mask(df["EMA20"], lookback=5, min_frac=EMA_SLOPE_MIN, up=True)
    df["EMA50_up"] = slope_mask(df["EMA50"], lookback=5, min_frac=EMA_SLOPE_MIN, up=True)
    df["EMA20_dn"] = slope_mask(df["EMA20"], lookback=5, min_frac=EMA_SLOPE_MIN, up=False)
    df["EMA50_dn"] = slope_mask(df["EMA50"], lookback=5, min_frac=EMA_SLOPE_MIN, up=False)

    # --- VWAP (session) & slope (lookback=5) ---
    df["VWAP"] = session_vwap_from_parts(df["High"], df["Low"], df["Close"], df["Volume"], df["date"])
    df["VWAP_slope_up"] = df.groupby("date")["VWAP"].apply(lambda s: (s - s.shift(VWAP_SLOPE_LOOKBACK)) > 0)
    df["VWAP_slope_dn"] = df.groupby("date")["VWAP"].apply(lambda s: (s - s.shift(VWAP_SLOPE_LOOKBACK)) < 0)

    # --- Daily OHLC + regime (from intraday) ---
    daily = df.groupby("date").agg(
        Open=("Open","first"), High=("High","max"), Low=("Low","min"), Close=("Close","last")
    ).reset_index()
    daily["EMA20D"] = ema(daily["Close"], 20)
    daily["EMA50D"] = ema(daily["Close"], 50)
    tr_d = true_range(daily["High"], daily["Low"], daily["Close"].shift())
    daily["ATR14D"] = tr_d.ewm(alpha=1/14, adjust=False).mean()
    daily["PrevClose"] = daily["Close"].shift(1)
    daily["GapPct"] = (daily["Open"] - daily["PrevClose"]) / daily["PrevClose"]

    # Merge daily fields back
    df = df.merge(daily[["date","EMA20D","EMA50D","ATR14D","GapPct"]], on="date", how="left")

    # --- Opening 15m stats (IBS & volume) ---
    def opening_stats(g):
        m = (g["t"] >= REG_START) & (g["t"] < OR_END)
        if not m.any():
            return pd.Series({"ORH":np.nan,"ORL":np.nan,"ORClose":np.nan,"ORVol":np.nan})
        sub = g.loc[m]
        return pd.Series({
            "ORH": sub["High"].max(),
            "ORL": sub["Low"].min(),
            "ORClose": sub["Close"].iloc[-1],
            "ORVol": sub["Volume"].sum()
        })
    open15 = df.groupby("date").apply(opening_stats)
    open15["IBS"] = (open15["ORClose"] - open15["ORL"]) / (open15["ORH"] - open15["ORL"])
    open15["ORVol_q45"] = open15["ORVol"].rolling(OPEN_VOL_LOOKBACK, min_periods=10).quantile(OPEN_VOL_Q).shift(1)
    open15 = open15.reset_index()
    df = df.merge(open15[["date","ORH","ORL","IBS","ORVol","ORVol_q45"]], on="date", how="left")

    # --- Intraday ATR for trailing ---
    tr_i = true_range(df["High"], df["Low"], df["Close"].shift())
    df["ATRi14"] = tr_i.ewm(alpha=1/ATR_INTRADAY_LEN, adjust=False).mean()

    # ===== Sim loop per day =====
    records = []
    day_counts = {}

    # helpful index for lookups
    df = df.set_index("EventAt", drop=False)

    for d, day in df.groupby("date"):
        day_counts.setdefault(d, 0)

        # Opening range (first 15 min)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask,"High"].max())
        orl = float(day.loc[or_mask,"Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # Daily ATR gate
        atrd = float(day["ATR14D"].iloc[0]) if day["ATR14D"].notna().any() else np.nan
        if USE_DAILY_ATR_GATE and np.isfinite(atrd):
            if not (atrd*ATR_DAILY_MIN <= or_width <= atrd*ATR_DAILY_MAX):
                continue

        # Gap filter
        gap = float(day["GapPct"].iloc[0]) if day["GapPct"].notna().any() else 0.0
        if USE_GAP_FILTER and not (GAP_MIN <= abs(gap) <= GAP_MAX):
            continue

        # Opening volume quantile
        orvol   = float(day["ORVol"].iloc[0])   if day["ORVol"].notna().any() else np.nan
        vol_qth = float(day["ORVol_q45"].iloc[0]) if day["ORVol_q45"].notna().any() else np.nan
        if USE_OPEN_VOL_QUANTILE and np.isfinite(vol_qth) and orvol < vol_qth:
            continue

        # IBS bias
        ibs = float(day["IBS"].iloc[0]) if day["IBS"].notna().any() else np.nan

        # After OR window
        after_or = day.loc[day["t"] >= OR_END].copy()
        if after_or.empty:
            continue

        # Breakout close beyond cushion
        cushion = BREAKOUT_CUSHION_PCT if USE_BREAKOUT_CUSHION else 0.0
        side, brk_idx = first_breakout_close(after_or, orh, orl, cushion)
        if side is None:
            continue

        # Regime (daily EMAs)
        if USE_DAILY_REGIME:
            ema20d = float(day["EMA20D"].iloc[0]) if day["EMA20D"].notna().any() else np.nan
            ema50d = float(day["EMA50D"].iloc[0]) if day["EMA50D"].notna().any() else np.nan
            if side == "long" and not (ema20d > ema50d):  continue
            if side == "short" and not (ema20d < ema50d): continue

        # IBS side bias
        if USE_IBS:
            if side == "long" and not (ibs >= 0.55):  continue
            if side == "short" and not (ibs <= 0.45): continue

        # Indicator snapshot at breakout close
        brk_row = after_or.loc[brk_idx]

        if USE_15M_EMA_TREND:
            if side == "long" and not (brk_row["EMA20"] > brk_row["EMA50"]):  continue
            if side == "short" and not (brk_row["EMA20"] < brk_row["EMA50"]): continue

        if USE_15M_EMA_SLOPE:
            if side == "long" and not (bool(brk_row["EMA20_up"]) and bool(brk_row["EMA50_up"])):   continue
            if side == "short" and not (bool(brk_row["EMA20_dn"]) and bool(brk_row["EMA50_dn"])): continue

        if USE_RSI_14:
            if side == "long" and not (brk_row["RSI14"] >= RSI_LONG_MIN):  continue
            if side == "short" and not (brk_row["RSI14"] <= RSI_SHORT_MAX): continue

        if USE_ADX_14 and not (brk_row["ADX14"] >= ADX_MIN):
            continue

        if USE_VWAP:
            if side == "long" and not (brk_row["Close"] > brk_row["VWAP"] and bool(brk_row["VWAP_slope_up"])): continue
            if side == "short" and not (brk_row["Close"] < brk_row["VWAP"] and bool(brk_row["VWAP_slope_dn"])): continue

        if USE_SPY_CONFIRMATION:
            pass  # same as VWAP alignment for SPY here

        # Entry timing windows (avoid lunch chop & late entries)
        after_breakout_close = after_or.loc[after_or.index > brk_idx].copy()
        if after_breakout_close.empty:
            continue
        ok_time = ((after_breakout_close["t"] < NO_TRADE_START) | (after_breakout_close["t"] >= NO_TRADE_END)) \
                  & (after_breakout_close["t"] <= LAST_ENTRY)
        after_breakout_close = after_breakout_close.loc[ok_time]
        if after_breakout_close.empty:
            continue

        # Retest entry at the broken level
        level = orh if side == "long" else orl
        retest_idx = first_retest_touch(after_breakout_close, level)
        if retest_idx is None:
            continue

        # Max trades/day
        if day_counts[d] >= MAX_TRADES_PER_DAY:
            continue

        entry_row = day.loc[day.index == retest_idx].iloc[0]

        # Opposite-wick check
        if USE_OPPOSITE_WICK and not opposite_wick_ok(entry_row, side, OPPOSITE_WICK_MAX_FRAC):
            continue

        # Retest close-through confirmation
        if RETEST_CLOSE_THROUGH:
            if side == "long":
                needed = level * (1 + RETEST_CONFIRM_CUSHION_PCT)
                if entry_row["Close"] < needed:
                    continue
            else:
                needed = level * (1 - RETEST_CONFIRM_CUSHION_PCT)
                if entry_row["Close"] > needed:
                    continue

        # Engulfing confirmation (current vs previous 1-minute bar)
        if ENGULFING_CONFIRM:
            # find previous bar within this day
            loc = day.index.get_loc(retest_idx)
            prev_row = None if loc == 0 else day.iloc[loc - 1]
            if not engulfing_ok(prev_row, entry_row, side):
                continue

        # Final entry/stop/targets
        entry_time = pd.Timestamp(entry_row["EventAt"])
        entry      = level
        stop       = orl if side == "long" else orh
        dist       = abs(entry - stop)
        if dist <= 0:
            continue

        tp1 = entry + ( dist if side=="long" else -dist)
        tp2 = entry + (2*dist if side=="long" else -2*dist)
        tp3 = entry + (3*dist if side=="long" else -3*dist)

        after_entry = day.loc[day.index > retest_idx].copy()
        if after_entry.empty:
            continue

        filled_tp1 = False
        filled_tp2 = False
        exit_px_final = None
        exit_time_final = None
        pnl_total = 0.0
        be_stop = stop
        trail_on = False
        trail_stop = None

        for idx, row in after_entry.iterrows():
            lo, hi, close = row["Low"], row["High"], row["Close"]
            atr_i = row["ATRi14"]

            active_stop = be_stop
            if trail_on and np.isfinite(atr_i):
                if side == "long":
                    trail_stop = max(trail_stop or -np.inf, close - TRAIL_ATR_MULT * atr_i)
                    active_stop = max(active_stop, trail_stop)
                else:
                    trail_stop = min(trail_stop or  np.inf, close + TRAIL_ATR_MULT * atr_i)
                    active_stop = min(active_stop, trail_stop)

            hit_stop = (lo <= active_stop) if side=="long" else (hi >= active_stop)
            hit_tp1  = (hi >= tp1) if side=="long" else (lo <= tp1)
            hit_tp2  = (hi >= tp2) if side=="long" else (lo <= tp2)
            hit_tp3  = (hi >= tp3) if side=="long" else (lo <= tp3)

            if hit_stop and ((not filled_tp1 and hit_tp1) or (filled_tp1 and not filled_tp2 and hit_tp2) or (filled_tp2 and hit_tp3)):
                rem_thirds = (1/3.0) * ((0 if filled_tp1 else 1) + (0 if filled_tp2 else 1) + 1)
                pnl_total += rem_thirds * pnl_one_share(side, entry, active_stop)
                exit_px_final = active_stop
                exit_time_final = row["EventAt"]
                break

            if not filled_tp1 and hit_tp1:
                pnl_total += (1/3.0) * pnl_one_share(side, entry, tp1)
                filled_tp1 = True
                be_stop = entry  # BE after TP1

            if filled_tp1 and not filled_tp2 and hit_tp2:
                pnl_total += (1/3.0) * pnl_one_share(side, entry, tp2)
                filled_tp2 = True
                trail_on = True  # trail last third

            if filled_tp1 and filled_tp2 and hit_tp3:
                pnl_total += (1/3.0) * pnl_one_share(side, entry, tp3)
                exit_px_final = tp3
                exit_time_final = row["EventAt"]
                break

            if trail_on:
                hit_trail = (lo <= active_stop) if side=="long" else (hi >= active_stop)
                if hit_trail:
                    pnl_total += (1/3.0) * pnl_one_share(side, entry, active_stop)
                    exit_px_final = active_stop
                    exit_time_final = row["EventAt"]
                    break

        if exit_px_final is None:
            # close remainder at EOD
            rem_thirds = (1/3.0) * ((0 if filled_tp1 else 1) + (0 if filled_tp2 else 1) + 1)
            eod_close = after_entry.iloc[-1]["Close"]
            pnl_total += rem_thirds * pnl_one_share(side, entry, eod_close)
            exit_px_final = eod_close
            exit_time_final = after_entry.iloc[-1]["EventAt"]

        records.append({
            "Date": d, "Symbol": SYMBOL, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": or_width,
            "EntryTime": entry_time, "Entry": entry, "Stop_init": stop,
            "TP1": tp1, "TP2": tp2, "TP3": tp3,
            "ExitPx_final": exit_px_final, "ExitTime_final": exit_time_final,
            "PnL_total": pnl_total,
            "Filled_TP1": filled_tp1, "Filled_TP2": filled_tp2,
            "Used_Trail": trail_on,
        })
        day_counts[d] += 1

    trades = pd.DataFrame(records)
    if trades.empty:
        print("No trades generated with current settings/filters.")
        return

    # Summary
    pnl = trades["PnL_total"].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    winrate = 100.0 * wins / max(1, wins + losses)
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = (gross_pos / gross_neg) if gross_neg > 0 else np.inf

    print(f"=== ORB 15m Retest + Filters — {SYMBOL} (1 share) ===")
    print(f"Trades: {wins+losses} | Total PnL: ${pnl.sum():,.2f} | Win%: {winrate:.2f}% | PF: {profit_factor:.3f}")
    print(f"Saved trades → {Path(SAVE_TRADES_CSV).resolve()}")

    trades.to_csv(SAVE_TRADES_CSV, index=False)

if __name__ == "__main__":
    main()
```
