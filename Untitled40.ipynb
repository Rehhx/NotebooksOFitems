```python
# orb_15m_parquet_institutional.py
# ORB 15m Retest/Continuation on a folder of .parquet files (2016→today),
# with indicators, realistic fills (confirm close -> next open),
# Profit Factor, progress bars, and institutional-grade validation:
# - Purged K-Fold CV (time-aware)
# - Walk-Forward Validation (purge/embargo)
# - Nested WFO (Train→Calibration→Test) picking a single param
# - Final strictly-unseen Holdout with portfolio risk management

import warnings
warnings.filterwarnings("ignore")

import os, glob, time
from datetime import timedelta, time as dtime
from typing import List, Tuple, Optional, Dict

import numpy as np
import pandas as pd

# Optional progress bar
try:
    from tqdm import tqdm
    TQDM_OK = True
except Exception:
    TQDM_OK = False

# =============================
# USER PARAMETERS
# =============================

# Parquet directory (ALL files inside will be processed)
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"

# If timestamps in your parquet are NAIVE and represent UTC, keep True
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic (quality + enough trades)
MAX_RETEST_MIN = 120            # minutes after breakout to wait for retest
RETEST_CONFIRM_CLOSE = True     # retest bar must close back through the level
BREAKOUT_CUSHION_PCT = 0.0005   # 0.05% cushion beyond OR to confirm breakout/retest
MAX_OPPOSITE_WICK_FRAC = 0.40   # reject entry bar if opposite wick > 40% of bar range

# Multiple trades per day
MAX_TRADES_PER_SESSION = 2
ALLOW_BOTH_SIDES = True

# Scale-out in thirds
USE_SCALE_OUT = True
TARGETS_R = [1.0, 2.0, 2.5]
SCALE_SPLIT = (1/3, 1/3, 1/3)

# Breakeven policy
MOVE_STOP_TO_BE_AT_R = 1.0      # arm BE after TP1 (==1R)
BE_ON_NEXT_BAR = True           # move stop to BE only on the next bar

# Continuation (fallback if no retest)
USE_CONTINUATION = True

# Time windows
BREAKOUT_DEADLINE   = "11:00"   # ET deadline for breakout
RETEST_DEADLINE     = "12:00"   # ET deadline for retest (kept for reference)
CONTINUATION_CUTOFF = "12:00"   # ET cutoff for continuation entries

# --- Indicator filters (ON) ---
USE_VWAP_FILTER   = True
USE_TREND_FILTER  = True
EMA_FAST = 20
EMA_SLOW = 50

USE_RSI_FILTER = True
RSI_LEN = 14
RSI_THRESH_LONG = 50.0
RSI_THRESH_SHORT = 50.0

USE_ADX_FILTER = False
ADX_LEN = 14
ADX_MIN = 18.0

USE_DAILY_TREND_FILTER = True
DAILY_EMA_FAST = 20
DAILY_EMA_SLOW = 50
DAILY_ATR_LEN  = 14

# OR width vs daily ATR band
USE_OR_WIDTH_ATR_FILTER = True
OR_ATR_MIN = 0.10
OR_ATR_MAX = 2.00

# Opening 15m volume quantile
USE_OPENING_VOL_FILTER = True
OPENING_VOL_LOOKBACK = 60
OPENING_VOL_QUANTILE = 0.30

# Gap filters
USE_GAP_FILTER = True
GAP_MIN = 0.003                 # 0.3%
GAP_MAX = 0.04                  # 4.0%
ALIGN_WITH_GAP_DIR = False      # allow counter-gap trades

USE_DOW_FILTER = False
ALLOWED_DOW = {1, 2, 3}         # Tue–Thu if enabled

# Entry realism when using close confirmation
#   "next_open" (recommended) = fill at next bar open after confirming close
#   "confirm_close"           = fill at confirming bar close
ENTRY_ON_CONFIRM = "next_open"

# Risk / costs
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00
FEE_PER_FILL = True             # charge fee per exit leg (TP/stop)

# Batch / persistence
SLEEP_BETWEEN_TICKERS = 0.0
AUTOSAVE_EVERY = 25

# Output files
TRADES_CSV_ALL = "orb_trades.csv"
EQUITY_CSV_ALL = "orb_equity.csv"
SUMMARY_CSV    = "orb_summary.csv"

# =============================
# VALIDATION TOGGLES
# =============================

RUN_PURGED_KFOLD = True
KFOLD_SPLITS = 5
PURGE_DAYS   = 2
EMBARGO_DAYS = 2

RUN_WALK_FORWARD = True
WF_TRAIN_DAYS = 365 * 2
WF_TEST_DAYS  = 90

RUN_NESTED_WF = True   # Train→Calib→Test per fold; produces chosen params
NWF_TRAIN_DAYS  = 365 * 2
NWF_CALIB_DAYS  = 60
NWF_TEST_DAYS   = 90
NWF_PURGE_DAYS  = 2
NWF_EMBARGO_DAYS= 2
NWF_MIN_TRADES_CALIB = 150

RUN_FINAL_HOLDOUT_RISK = True
HOLDOUT_START_AFTER_WF = True   # set True to begin holdout the day AFTER last nested-WF test

RUN_SENSITIVITY = False
SENS_SLIPPAGE_LIST = [0.5, 1.0, 2.0]
SENS_FEE_LIST      = [0.00, 0.25]

# Parameter grid (small, safe to scale up later)
BUILD_PARAM_GRID = True
GRID_BREAKOUT_CUSHION = [0.0003, 0.0005, 0.0010]
GRID_WICK_FRAC        = [0.30, 0.40, 0.50]

# =============================
# Helpers
# =============================

def ensure_unique_columns(df: pd.DataFrame) -> pd.DataFrame:
    if getattr(df.columns, "duplicated", None) is not None and df.columns.duplicated().any():
        df = df.loc[:, ~df.columns.duplicated()].copy()
    return df

def sessionize(obj) -> pd.Series:
    idx = getattr(obj, "index", obj)
    return pd.to_datetime(pd.Index(idx).date)

def _to_clock(ts) -> dtime:
    return dtime(ts.hour, ts.minute)

def _clock_le(ts, hhmm: str) -> bool:
    h, m = map(int, hhmm.split(":"))
    return _to_clock(ts) <= dtime(h, m)

def wick_opposite_fraction(row: pd.Series, direction: str) -> float:
    h, l, o, c = float(row["High"]), float(row["Low"]), float(row.get("Open", np.nan)), float(row["Close"])
    rng = max(h - l, 1e-12)
    if direction == "long":
        opp = min(o, c) - l
    else:
        opp = h - max(o, c)
    return float(max(opp, 0.0) / rng)

def true_range(high, low, prev_close):
    return np.maximum.reduce([
        (high - low).values,
        np.abs(high - prev_close).values,
        np.abs(low - prev_close).values
    ])

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def _normalize_splits(num_targets: int, splits: tuple) -> List[float]:
    if num_targets <= 0:
        return [1.0]
    if not splits:
        return [1.0 / num_targets] * num_targets
    arr = list(splits[:num_targets])
    if len(arr) < num_targets:
        arr += [0.0] * (num_targets - len(arr))
    s = sum(arr)
    if s <= 0:
        return [1.0 / num_targets] * num_targets
    return [x / s for x in arr]

# Nicely formatted totals (with PF)
def print_overall_totals(trades: pd.DataFrame):
    print("\n=== Overall Totals ===")
    if trades is None or trades.empty:
        print("Total trades: 0")
        print("Total PnL ($): 0.00")
        print("Win rate: 0.00%")
        print("Avg R: 0.000 | Median R: 0.000")
        print("Profit Factor: 0.000")
        print("Avg Win ($): 0.00 | Avg Loss ($): 0.00")
        return

    pnl = pd.to_numeric(trades["PnL_$"], errors="coerce").fillna(0.0)
    wins_mask = pnl > 0
    losses_mask = pnl < 0

    total_trades = int(len(pnl))
    total_pnl    = float(pnl.sum())
    win_rate     = float(wins_mask.mean() * 100.0) if total_trades else 0.0
    avg_r        = float(pd.to_numeric(trades["R_multiple"], errors="coerce").mean())
    median_r     = float(pd.to_numeric(trades["R_multiple"], errors="coerce").median())

    gross_profit = float(pnl[wins_mask].sum())
    gross_loss   = float(-pnl[losses_mask].sum())
    pf = (gross_profit / gross_loss) if gross_loss > 0 else (float("inf") if gross_profit > 0 else 0.0)

    avg_win  = float(pnl[wins_mask].mean()) if wins_mask.any() else 0.0
    avg_loss = float(pnl[losses_mask].mean()) if losses_mask.any() else 0.0

    print(f"Total trades: {total_trades}")
    print(f"Total PnL ($): {total_pnl:,.2f}")
    print(f"Win rate: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Profit Factor: {pf:.3f}")
    print(f"Avg Win ($): {avg_win:,.2f} | Avg Loss ($): {avg_loss:,.2f}")

# =============================
# Indicators & features
# =============================

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    for c in ("ORH","ORL"):
        if (df.columns == c).sum() > 0:
            df = df.drop(columns=[c])
    for col in ("High","Low"):
        if col not in df.columns:
            raise ValueError(f"compute_opening_range: missing column {col}")
    if not df.index.is_monotonic_increasing:
        df = df.sort_index()
    df["Session"] = sessionize(df)
    first_bar_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float).values
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float).values
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return ensure_unique_columns(df)

def add_session_vwap(df_15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df["Session"] = sessionize(df)
    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0
    if "Volume" in df.columns:
        df["vwap_num"] = tp * df["Volume"]
        df["vwap_den"] = df["Volume"].replace(0, np.nan)
    else:
        df["vwap_num"] = tp
        df["vwap_den"] = 1.0
    df["VWAP"] = (df.groupby("Session")["vwap_num"].cumsum() /
                  df.groupby("Session")["vwap_den"].cumsum())
    return ensure_unique_columns(df.drop(columns=["vwap_num","vwap_den"]))

def add_ema_trend(df_15: pd.DataFrame, fast=EMA_FAST, slow=EMA_SLOW) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return ensure_unique_columns(df)

def rsi(series: pd.Series, length=14) -> pd.Series:
    delta = series.diff()
    up = delta.clip(lower=0)
    dn = -delta.clip(upper=0)
    avg_gain = up.ewm(alpha=1/length, adjust=False).mean()
    avg_loss = dn.ewm(alpha=1/length, adjust=False).mean()
    rs = avg_gain / (avg_loss.replace(0, np.nan))
    return 100 - (100 / (1 + rs))

def add_rsi(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df[f"RSI{length}"] = rsi(df["Close"], length)
    return ensure_unique_columns(df)

def add_adx(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    up_move = df["High"].diff()
    dn_move = -df["Low"].diff()
    plus_dm  = np.where((up_move > dn_move) & (up_move > 0), up_move, 0.0)
    minus_dm = np.where((dn_move > up_move) & (dn_move > 0), dn_move, 0.0)

    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    atr = tr.ewm(alpha=1/length, adjust=False).mean()

    plus_di = 100 * pd.Series(plus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    minus_di = 100 * pd.Series(minus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    dx = (abs(plus_di - minus_di) / (plus_di + minus_di).replace(0, np.nan)) * 100
    adx = dx.ewm(alpha=1/length, adjust=False).mean()
    df[f"ADX{length}"] = adx
    return ensure_unique_columns(df)

def add_atr_15m(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    df[f"ATR15_{length}"] = tr.ewm(alpha=1/length, adjust=False).mean()
    return ensure_unique_columns(df)

def build_daily_from_15m(df_15: pd.DataFrame) -> pd.DataFrame:
    d = ensure_unique_columns(df_15.copy())
    d["Session"] = sessionize(d)
    daily = d.groupby("Session").agg(
        Open=("Open","first"),
        High=("High","max"),
        Low=("Low","min"),
        Close=("Close","last"),
        Volume=("Volume","sum") if "Volume" in d.columns else ("Close","size"),
    )
    return daily

def add_daily_bias_and_atr(df_15: pd.DataFrame, ema_fast=20, ema_slow=50, atr_len=14) -> pd.DataFrame:
    dly = build_daily_from_15m(df_15)

    dly[f"EMA_D{ema_fast}"] = dly["Close"].ewm(span=ema_fast, adjust=False).mean()
    dly[f"EMA_D{ema_slow}"] = dly["Close"].ewm(span=ema_slow, adjust=False).mean()

    prev_close = dly["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (dly["High"] - dly["Low"]).values,
        np.abs(dly["High"] - prev_close).values,
        np.abs(dly["Low"]  - prev_close).values
    ]), index=dly.index)
    atr_col = f"ATR_D{atr_len}"
    dly[atr_col] = tr.ewm(alpha=1/atr_len, adjust=False).mean()

    if "Volume" in df_15.columns:
        tmp = df_15.copy()
        tmp["Session"] = sessionize(tmp)
        open15_vol = tmp.groupby("Session")["Volume"].first()
        dly["Open15_Vol"] = open15_vol
        dly["Open15_Vol_Thresh"] = (
            dly["Open15_Vol"]
            .rolling(window=OPENING_VOL_LOOKBACK, min_periods=10)
            .quantile(OPENING_VOL_QUANTILE)
        )

    feature_cols = [f"EMA_D{ema_fast}", f"EMA_D{ema_slow}", atr_col, "Open15_Vol", "Open15_Vol_Thresh"]
    feature_cols = [c for c in feature_cols if c in dly.columns]

    out = ensure_unique_columns(df_15.copy())
    out["Session"] = sessionize(out)
    out = out.merge(dly[feature_cols], left_on="Session", right_index=True, how="left")
    return ensure_unique_columns(out)

# =============================
# Backtest (per-ticker)
# =============================

def backtest_orb_retest(
    df_15: pd.DataFrame,
    cfg: Optional[Dict] = None
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    cfg can override any of the globals used below, e.g.:
      {"BREAKOUT_CUSHION_PCT":0.0005, "MAX_OPPOSITE_WICK_FRAC":0.4, ...}
    """
    C = lambda k, d=None: (cfg[k] if (cfg and k in cfg) else globals()[k]) if (d is None) else (cfg.get(k, d) if cfg else d)

    # Opening range + features
    df = compute_opening_range(df_15)
    if C("USE_VWAP_FILTER", True):  df = add_session_vwap(df)
    if C("USE_TREND_FILTER", True): df = add_ema_trend(df, C("EMA_FAST", EMA_FAST), C("EMA_SLOW", EMA_SLOW))
    df = add_daily_bias_and_atr(df, C("DAILY_EMA_FAST", DAILY_EMA_FAST), C("DAILY_EMA_SLOW", DAILY_EMA_SLOW), C("DAILY_ATR_LEN", DAILY_ATR_LEN))

    # Gap (first 15m vs previous day's close)
    df["Session"] = sessionize(df)
    first_idx = df.groupby("Session").head(1).index
    prev_close_series = df.groupby("Session")["Close"].last().shift(1)
    prev_close_map = df["Session"].map(prev_close_series)
    df.loc[first_idx, "GapPct"] = (df.loc[first_idx, "Open"] / prev_close_map.loc[first_idx] - 1.0)
    df["GapPct"] = df.groupby("Session")["GapPct"].ffill()

    # Intraday indicators
    if C("USE_RSI_FILTER", True):  df = add_rsi(df, C("RSI_LEN", RSI_LEN))
    if C("USE_ADX_FILTER", False): df = add_adx(df, C("ADX_LEN", ADX_LEN))
    df = ensure_unique_columns(df)

    sessions = df["Session"].unique()
    trades = []

    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh, orl = float(sdf["ORH"].iloc[0]), float(sdf["ORL"].iloc[0])
        if np.isnan(orh) or np.isnan(orl):
            continue

        # All breakout candidates after the OR bar (use cushion)
        after_open = sdf.iloc[1:].copy()
        bcp = C("BREAKOUT_CUSHION_PCT", BREAKOUT_CUSHION_PCT)
        long_breaks  = after_open[after_open["Close"] > orh * (1 + bcp)]
        short_breaks = after_open[after_open["Close"] < orl * (1 - bcp)]

        cands = []
        for idx, row in long_breaks.iterrows():
            cands.append(("long", idx, row))
        for idx, row in short_breaks.iterrows():
            cands.append(("short", idx, row))
        cands.sort(key=lambda x: x[1])  # by timestamp

        trades_this_session = 0
        took_long = False
        took_short = False

        for direction, btime, breakout_row in cands:
            if not _clock_le(breakout_row.name, C("BREAKOUT_DEADLINE", BREAKOUT_DEADLINE)):
                continue

            if not C("ALLOW_BOTH_SIDES", True):
                if took_long and direction == "short": break
                if took_short and direction == "long": break
            if direction == "long" and took_long:   continue
            if direction == "short" and took_short: continue

            if C("USE_DOW_FILTER", False) and btime.weekday() not in C("ALLOWED_DOW", ALLOWED_DOW):
                continue

            if C("USE_GAP_FILTER", True) and "GapPct" in sdf.columns:
                gap = sdf.loc[sdf.index.min(), "GapPct"]
                if pd.notna(gap) and not (C("GAP_MIN", GAP_MIN) <= abs(float(gap)) <= C("GAP_MAX", GAP_MAX)):
                    continue
                if C("ALIGN_WITH_GAP_DIR", False) and pd.notna(gap):
                    g = float(gap)
                    if (direction == "long" and g < 0) or (direction == "short" and g > 0):
                        continue

            cutoff = btime + timedelta(minutes=C("MAX_RETEST_MIN", MAX_RETEST_MIN))
            window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
            level = orh if direction == "long" else orl
            touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)

            retest_used = True
            if touch.empty:
                if C("USE_CONTINUATION", True) and _clock_le(breakout_row.name, C("CONTINUATION_CUTOFF", CONTINUATION_CUTOFF)):
                    retest_used = False
                    retest_time = breakout_row.name
                    retest_bar = breakout_row
                else:
                    continue
            else:
                retest_bar = touch.iloc[0]
                retest_time = retest_bar.name

            if retest_used and C("RETEST_CONFIRM_CLOSE", True):
                if direction == "long":
                    ok = (retest_bar["Close"] >= level * (1 + bcp))
                else:
                    ok = (retest_bar["Close"] <= level * (1 - bcp))
                if not ok:
                    continue

            if C("MAX_OPPOSITE_WICK_FRAC", MAX_OPPOSITE_WICK_FRAC) is not None:
                if wick_opposite_fraction(retest_bar, direction) > C("MAX_OPPOSITE_WICK_FRAC", MAX_OPPOSITE_WICK_FRAC):
                    continue

            # Indicator filters
            if C("USE_VWAP_FILTER", True) and "VWAP" in sdf.columns:
                v_prev = sdf.loc[:retest_time, "VWAP"].tail(3).dropna().values
                rising  = (len(v_prev) < 2) or np.all(np.diff(v_prev) >= 0)
                falling = (len(v_prev) < 2) or np.all(np.diff(v_prev) <= 0)
                if direction == "long":
                    if not (retest_bar["Close"] > retest_bar["VWAP"] and rising):
                        continue
                else:
                    if not (retest_bar["Close"] < retest_bar["VWAP"] and falling):
                        continue

            if C("USE_TREND_FILTER", True) and (f"EMA{C('EMA_FAST', EMA_FAST)}" in sdf.columns) and (f"EMA{C('EMA_SLOW', EMA_SLOW)}" in sdf.columns):
                emaf = retest_bar.get(f"EMA{C('EMA_FAST', EMA_FAST)}", np.nan); emas = retest_bar.get(f"EMA{C('EMA_SLOW', EMA_SLOW)}", np.nan)
                if np.isfinite(emaf) and np.isfinite(emas):
                    if direction == "long" and not (emaf > emas): continue
                    if direction == "short" and not (emaf < emas): continue

            if C("USE_DAILY_TREND_FILTER", True) and f"EMA_D{C('DAILY_EMA_FAST', DAILY_EMA_FAST)}" in sdf.columns and f"EMA_D{C('DAILY_EMA_SLOW', DAILY_EMA_SLOW)}" in sdf.columns:
                dfast = float(sdf[f"EMA_D{C('DAILY_EMA_FAST', DAILY_EMA_FAST)}"].iloc[0])
                dslow = float(sdf[f"EMA_D{C('DAILY_EMA_SLOW', DAILY_EMA_SLOW)}"].iloc[0])
                if direction == "long" and not (dfast > dslow): continue
                if direction == "short" and not (dfast < dslow): continue

            atr_col = f"ATR_D{C('DAILY_ATR_LEN', DAILY_ATR_LEN)}"
            if C("USE_OR_WIDTH_ATR_FILTER", True) and atr_col in sdf.columns:
                or_width = float(orh - orl)
                atr_d = float(sdf[atr_col].iloc[0])
                if atr_d <= 0: continue
                frac = or_width / atr_d
                if not (C("OR_ATR_MIN", OR_ATR_MIN) <= frac <= C("OR_ATR_MAX", OR_ATR_MAX)): continue

            if C("USE_RSI_FILTER", True) and f"RSI{C('RSI_LEN', RSI_LEN)}" in sdf.columns:
                rsi_val = float(retest_bar[f"RSI{C('RSI_LEN', RSI_LEN)}"])
                if direction == "long" and not (rsi_val >= C("RSI_THRESH_LONG", RSI_THRESH_LONG)): continue
                if direction == "short" and not (rsi_val <= 100 - C("RSI_THRESH_SHORT", RSI_THRESH_SHORT)): continue

            if C("USE_ADX_FILTER", False) and f"ADX{C('ADX_LEN', ADX_LEN)}" in sdf.columns:
                adx_val = float(retest_bar[f"ADX{C('ADX_LEN', ADX_LEN)}"])
                if not (adx_val >= C("ADX_MIN", ADX_MIN)): continue

            # Entry price (realistic)
            if retest_used and C("RETEST_CONFIRM_CLOSE", True):
                if C("ENTRY_ON_CONFIRM", ENTRY_ON_CONFIRM).lower() == "next_open":
                    next_bar = sdf[sdf.index > retest_time].head(1)
                    if next_bar.empty:
                        continue
                    entry_time = next_bar.index[0]
                    raw_entry = float(next_bar["Open"].iloc[0])
                else:
                    entry_time = retest_time
                    raw_entry = float(retest_bar["Close"])
            else:
                entry_time = retest_time
                raw_entry = float(level)

            entry = _apply_slippage(raw_entry, C("SLIPPAGE_BPS", SLIPPAGE_BPS), "buy" if direction == "long" else "sell")
            stop  = orl if direction == "long" else orh
            rps   = abs(entry - stop)
            if rps <= 1e-12:
                continue

            qty = C("POSITION_SIZE_DOLLARS", POSITION_SIZE_DOLLARS) / max(entry, 1e-12)
            side_mult = 1 if direction == "long" else -1

            targets_r = TARGETS_R if C("USE_SCALE_OUT", True) else [2.0]
            targets = [(entry + r * rps) if direction == "long" else (entry - r * rps) for r in targets_r]

            run = sdf[sdf.index >= entry_time].copy()
            qty_left = qty
            exits = []
            next_tp_idx = 0

            if C("USE_SCALE_OUT", True) and len(targets) >= 1:
                splits = _normalize_splits(len(targets), C("SCALE_SPLIT", SCALE_SPLIT))
                leg_sizes = [qty * s for s in splits]
            else:
                leg_sizes = [qty]

            be_level = entry
            be_pending = False
            be_set_time = None

            for ts, row in run.iterrows():
                hi, lo = float(row["High"]), float(row["Low"])

                if be_pending and (not C("BE_ON_NEXT_BAR", BE_ON_NEXT_BAR) or (be_set_time is not None and ts > be_set_time)):
                    stop = be_level
                    be_pending = False

                # Stop first
                if lo <= stop <= hi:
                    px = _apply_slippage(stop, C("SLIPPAGE_BPS", SLIPPAGE_BPS), "sell" if direction == "long" else "buy")
                    label = "breakeven" if abs(px - be_level) < 1e-10 else "stop"
                    exits.append((label, ts, px, qty_left))
                    qty_left = 0.0
                    break

                # Targets
                if next_tp_idx < len(targets):
                    tp = targets[next_tp_idx]
                    if lo <= tp <= hi:
                        px = _apply_slippage(tp, C("SLIPPAGE_BPS", SLIPPAGE_BPS), "sell" if direction == "long" else "buy")
                        fill_qty = leg_sizes[next_tp_idx] if next_tp_idx < len(leg_sizes) else qty_left
                        exits.append((f"tp{next_tp_idx+1}", ts, px, fill_qty))
                        qty_left -= fill_qty
                        if next_tp_idx == 0 and C("MOVE_STOP_TO_BE_AT_R", MOVE_STOP_TO_BE_AT_R) is not None:
                            be_pending = True
                            be_set_time = ts
                        next_tp_idx += 1
                        if qty_left <= 1e-9:
                            break

                if C("USE_CONTINUATION", True) and not _clock_le(ts, C("CONTINUATION_CUTOFF", CONTINUATION_CUTOFF)) and next_tp_idx == 0:
                    stop = be_level

            # EOD exit
            if qty_left > 1e-9:
                last = run.iloc[-1]
                px = _apply_slippage(float(last["Close"]), C("SLIPPAGE_BPS", SLIPPAGE_BPS), "sell" if direction == "long" else "buy")
                exits.append(("eod", run.index[-1], px, qty_left))
                qty_left = 0.0

            cash_pnl = sum((px - entry) * side_mult * q for (_, _, px, q) in exits)
            if C("FEE_PER_FILL", FEE_PER_FILL):
                cash_pnl -= C("FEES_PER_TRADE", FEES_PER_TRADE) * len(exits)
            else:
                cash_pnl -= C("FEES_PER_TRADE", FEES_PER_TRADE)

            trades.append({
                "Ticker": sdf["Ticker"].iloc[0] if "Ticker" in sdf.columns else "",
                "Session": ses,
                "Direction": direction,
                "ORH": orh, "ORL": orl,
                "EntryTime": entry_time,
                "Entry": entry, "Stop": stop,
                "Targets": targets,
                "Exits": [(lab, ts, px, q) for (lab, ts, px, q) in exits],
                "Qty": qty,
                "PnL_$": cash_pnl,
                "R_multiple": cash_pnl / (rps * max(qty, 1e-12))
            })

            if direction == "long":  took_long  = True
            if direction == "short": took_short = True
            trades_this_session += 1
            if trades_this_session >= C("MAX_TRADES_PER_SESSION", MAX_TRADES_PER_SESSION):
                break

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Parquet loading
# =============================

STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                rename[lower[k]] = std
        elif k.capitalize() in df.columns: rename[k.capitalize()] = std
        elif k.upper() in df.columns: rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _maybe_resample_to_15m(df: pd.DataFrame) -> pd.DataFrame:
    if len(df.index) < 3:
        return df
    deltas = (df.index[1:] - df.index[:-1]).asi8
    if len(deltas) == 0:
        return df
    med_ns = np.median(deltas)
    fifteen_ns = pd.Timedelta("15T").value
    if med_ns < fifteen_ns:
        o = df["Open"].resample("15T", label="left", closed="left").first()
        h = df["High"].resample("15T", label="left", closed="left").max()
        l = df["Low"].resample("15T", label="left", closed="left").min()
        c = df["Close"].resample("15T", label="left", closed="left").last()
        v = df["Volume"].resample("15T", label="left", closed="left").sum() if "Volume" in df.columns else None
        parts = {"Open":o,"High":h,"Low":l,"Close":c}
        if v is not None: parts["Volume"] = v
        out = pd.concat(parts, axis=1).dropna(subset=["Open","High","Low","Close"])
        return out
    return df

def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()

    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df = df.tz_localize("UTC").tz_convert(tz)
        else:
            df = df.tz_localize(tz)
    else:
        df = df.tz_convert(tz)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    df = _maybe_resample_to_15m(df)
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)
    return ensure_unique_columns(df.sort_index())

# =============================
# Runner (folder of parquet files)
# =============================

def run_folder(folder: str, cfg: Optional[Dict]=None) -> Tuple[pd.DataFrame, pd.DataFrame]:
    files = sorted(glob.glob(os.path.join(folder, "*.parquet")))
    print(f"Found {len(files)} parquet files.")
    all_trades, all_equity = [], []

    iterator = tqdm(files, desc="Backtesting files") if TQDM_OK else files
    for i, fpath in enumerate(iterator, 1):
        ticker = os.path.splitext(os.path.basename(fpath))[0].upper().replace("_","").replace("-","")
        msg = f"== {ticker} ({i}/{len(files)}) =="
        if not TQDM_OK:
            print("\n" + msg)
        try:
            df15 = load_parquet_15m(fpath)
            df15["Ticker"] = ticker
        except Exception as e:
            print(f"Skipping {ticker}: {e}")
            continue

        tlog, eq = backtest_orb_retest(df15, cfg=cfg)

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"]   = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}|{t[3]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy(); eq_out["Ticker"] = ticker
            all_equity.append(eq_out)

        if AUTOSAVE_EVERY and i % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

        if SLEEP_BETWEEN_TICKERS > 0:
            time.sleep(SLEEP_BETWEEN_TICKERS)

    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()
    return trades, equity

# =============================
# Validation utilities
# =============================

def metrics_from_trades(df: pd.DataFrame) -> dict:
    if df is None or df.empty:
        return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, PnL=0.0)
    pnl = pd.to_numeric(df["PnL_$"], errors="coerce").fillna(0.0)
    r   = pd.to_numeric(df["R_multiple"], errors="coerce").fillna(0.0)
    wins = pnl > 0
    gp = float(pnl[wins].sum())
    gl = float(-pnl[~wins & (pnl<0)].sum())
    pf = (gp / gl) if gl > 0 else (float("inf") if gp > 0 else 0.0)
    return dict(
        Trades = int(len(df)),
        WinRate = float(wins.mean() * 100.0) if len(df) else 0.0,
        PF = float(pf),
        AvgR = float(r.mean()),
        PnL = float(pnl.sum())
    )

def purged_kfold_cv(trades: pd.DataFrame, n_splits=5, purge_days=2, embargo_days=2):
    print("\n--- Purged K-Fold CV ---")
    if trades.empty:
        print("No trades.")
        return
    trades = trades.copy()
    trades["Session_dt"] = pd.to_datetime(trades["Session"])
    sessions = np.sort(trades["Session_dt"].unique())
    folds = np.array_split(sessions, n_splits)

    for i, test_sessions in enumerate(folds, 1):
        test_start = pd.Timestamp(test_sessions.min())
        test_end   = pd.Timestamp(test_sessions.max())

        purge_end = test_start - pd.Timedelta(days=purge_days)
        embargo_start = test_end + pd.Timedelta(days=1)
        embargo_end   = embargo_start + pd.Timedelta(days=embargo_days)

        train = trades[(trades["Session_dt"] <= purge_end) | (trades["Session_dt"] >= embargo_end)]
        test  = trades[(trades["Session_dt"] >= test_start) & (trades["Session_dt"] <= test_end)]

        m_tr = metrics_from_trades(train)
        m_te = metrics_from_trades(test)

        print(f"Fold {i:02d} Train [..→{pd.Timestamp(purge_end).date()}] | Trades: {m_tr['Trades']:,} | Win%: {m_tr['WinRate']:.2f}% | PF: {m_tr['PF']:.3f} | Avg R: {m_tr['AvgR']:.3f} | PnL: ${m_tr['PnL']:,.2f}")
        print(f"Fold {i:02d} Test  [{pd.Timestamp(test_start).date()}→{pd.Timestamp(test_end).date()}] | Trades: {m_te['Trades']:,} | Win%: {m_te['WinRate']:.2f}% | PF: {m_te['PF']:.3f} | Avg R: {m_te['AvgR']:.3f} | PnL: ${m_te['PnL']:,.2f}")

def walk_forward_validation(trades: pd.DataFrame, train_days=365*2, test_days=90, purge_days=2, embargo_days=2):
    print("\n--- Walk-Forward Validation ---")
    if trades.empty:
        print("No trades.")
        return
    df = trades.copy()
    df["Session_dt"] = pd.to_datetime(df["Session"])
    sessions_sorted = np.sort(df["Session_dt"].unique())
    t0, t1 = sessions_sorted.min(), sessions_sorted.max()

    fold = 1
    train_start = pd.Timestamp(t0)

    while True:
        train_end = train_start + pd.Timedelta(days=train_days) - pd.Timedelta(days=1)
        if train_end >= t1:
            break

        test_start = train_end + pd.Timedelta(days=1)
        test_end   = test_start + pd.Timedelta(days=test_days) - pd.Timedelta(days=1)
        if test_start > t1:
            break

        # purge from end of train
        train_cut = train_end - pd.Timedelta(days=purge_days)
        test_eval_start = test_start + pd.Timedelta(days=embargo_days)
        test_eval_end   = min(test_end, t1)

        trn = df[(df["Session_dt"] >= train_start) & (df["Session_dt"] <= train_cut)]
        tst = df[(df["Session_dt"] >= test_eval_start) & (df["Session_dt"] <= test_eval_end)]

        m_trn = metrics_from_trades(trn)
        m_tst = metrics_from_trades(tst)

        print(f"Fold {fold:02d} Train [{pd.Timestamp(train_start).date()}→{pd.Timestamp(train_cut).date()}] | Trades: {m_trn['Trades']:,} | Win%: {m_trn['WinRate']:.2f}% | PF: {m_trn['PF']:.3f} | Avg R: {m_trn['AvgR']:.3f} | PnL: ${m_trn['PnL']:,.2f}")
        print(f"Fold {fold:02d} Test  [{pd.Timestamp(test_eval_start).date()}→{pd.Timestamp(test_eval_end).date()}] | Trades: {m_tst['Trades']:,} | Win%: {m_tst['WinRate']:.2f}% | PF: {m_tst['PF']:.3f} | Avg R: {m_tst['AvgR']:.3f} | PnL: ${m_tst['PnL']:,.2f}")

        train_start = test_start
        fold += 1
        if train_start >= t1:
            break

# ---------- Nested WFO (Train→Calib→Test) + final holdout ----------
def _metrics_from_R(r: np.ndarray):
    if r.size == 0:
        return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, PnL=0.0)
    wr = 100.0 * (r > 0).mean()
    gp = float(r[r > 0].sum())
    gl = float(-r[r < 0].sum())
    pf = (gp / gl) if gl > 0 else (float("inf") if gp > 0 else 0.0)
    return dict(Trades=int(len(r)), WinRate=float(wr), PF=float(pf), AvgR=float(np.mean(r)), PnL=float(np.sum(r)))

def choose_final_param_from_wf(wf_records: list) -> Optional[str]:
    if not wf_records:
        return None
    df = pd.DataFrame(wf_records)
    counts = df["chosen_key"].value_counts()
    top_freq = counts.max()
    finalists = counts[counts == top_freq].index.tolist()
    if len(finalists) == 1:
        return finalists[0]
    med = (df[df["chosen_key"].isin(finalists)]
           .groupby("chosen_key")["test_AvgR"].median()
           .sort_values(ascending=False))
    return med.index[0]

def nested_walk_forward_with_calibration(
    grid_returns: Dict[str, pd.DataFrame],
    train_days: int = 365*2,
    calib_days: int = 60,
    test_days: int = 90,
    purge_days: int = 2,
    embargo_days: int = 2,
    min_trades_calib: int = 150,
) -> Tuple[list, Optional[pd.Timestamp]]:
    # Sessions from any grid member
    all_sess = np.unique(np.concatenate([
        df["Session_dt"].values for df in grid_returns.values() if not df.empty
    ])) if grid_returns else np.array([])
    if all_sess.size == 0:
        print("\n--- Nested Walk-Forward (Train→Calib→Test) ---\nNo sessions.")
        return [], None

    all_sess = pd.to_datetime(all_sess)
    all_sess.sort_values(inplace=True)
    t0, t1 = all_sess.min(), all_sess.max()
    print("\n--- Nested Walk-Forward (Train→Calib→Test) ---")

    def seg_metrics(grid, start, end, min_trades=0):
        rows = []
        for key, df in grid.items():
            if df.empty:
                rows.append({"key": key, "Trades": 0, "WinRate": 0.0, "PF": 0.0, "AvgR": 0.0, "PnL": 0.0})
                continue
            msk = (df["Session_dt"] >= start) & (df["Session_dt"] <= end)
            r = df.loc[msk, "R_multiple"].astype(float).values
            m = _metrics_from_R(r)
            m["key"] = key
            rows.append(m)
        out = pd.DataFrame(rows)
        out["eligible"] = out["Trades"] >= int(min_trades)
        return out

    def pick_on_calib(m_calib: pd.DataFrame) -> str:
        elig = m_calib[m_calib["eligible"]].copy()
        if elig.empty:
            elig = m_calib.copy()
        elig = elig.sort_values(["AvgR","PF","WinRate","Trades"], ascending=[False, False, False, False])
        return str(elig.iloc[0]["key"])

    fold = 1
    train_start = pd.Timestamp(t0)
    records = []

    while True:
        train_end = train_start + pd.Timedelta(days=train_days) - pd.Timedelta(days=1)
        if train_end >= t1:
            break

        calib_start = train_end - pd.Timedelta(days=calib_days) + pd.Timedelta(days=1)
        train_cut   = calib_start - pd.Timedelta(days=purge_days)

        test_start = train_end + pd.Timedelta(days=1)
        test_end   = test_start + pd.Timedelta(days=test_days) - pd.Timedelta(days=1)
        if test_start > t1:
            break
        test_eval_start = test_start + pd.Timedelta(days=embargo_days)
        test_eval_end   = min(test_end, t1)

        m_train = seg_metrics(grid_returns, train_start, train_cut, min_trades=0)
        m_calib = seg_metrics(grid_returns, calib_start, train_end, min_trades=min_trades_calib)

        chosen_key = pick_on_calib(m_calib)
        df_best = grid_returns.get(chosen_key, pd.DataFrame())
        if df_best.empty:
            test_m = dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, PnL=0.0)
        else:
            r = df_best[(df_best["Session_dt"] >= test_eval_start) & (df_best["Session_dt"] <= test_eval_end)]["R_multiple"].astype(float).values
            test_m = _metrics_from_R(r)

        print(f"Fold {fold:02d}")
        print(f"  Train [{pd.Timestamp(train_start).date()}→{pd.Timestamp(train_cut).date()}]  (purge {purge_days}d)")
        print(f"  Calib [{pd.Timestamp(calib_start).date()}→{pd.Timestamp(train_end).date()}]  pick={chosen_key}")
        print(f"  Test  [{pd.Timestamp(test_eval_start).date()}→{pd.Timestamp(test_eval_end).date()}] | Trades: {test_m['Trades']:,} | Win%: {test_m['WinRate']:.2f}% | PF: {test_m['PF']:.3f} | AvgR: {test_m['AvgR']:.3f} | SumR: {test_m['PnL']:.2f}")

        records.append({
            "fold": fold,
            "train_start": train_start, "train_end": train_end,
            "calib_start": calib_start, "calib_end": train_end,
            "test_start": test_start, "test_end": test_end,
            "test_eval_start": test_eval_start, "test_eval_end": test_eval_end,
            "chosen_key": chosen_key,
            "test_Trades": test_m["Trades"],
            "test_WinRate": test_m["WinRate"],
            "test_PF": test_m["PF"],
            "test_AvgR": test_m["AvgR"],
            "test_PnL_R": test_m["PnL"],
        })

        train_start = test_start
        fold += 1
        if train_start >= t1:
            break

    if records:
        oos_wr = np.mean([r["test_WinRate"] for r in records])
        oos_pf = np.mean([r["test_PF"] for r in records if np.isfinite(r["test_PF"])])
        print(f"Nested WF OOS Avg | Win%: {oos_wr:.2f}% | PF: {oos_pf:.3f}")

    last_eval_end = records[-1]["test_eval_end"] if records else None
    return records, last_eval_end

def simulate_holdout_with_risk(df_param: pd.DataFrame,
                               start, end,
                               starting_equity: float = 100_000.0,
                               risk_per_trade_pct: float = 0.005,
                               max_daily_loss_pct: float = 0.02,
                               daily_trade_cap: int = 20):
    if df_param is None or df_param.empty:
        print("\n--- Final Holdout (Risk) ---\nNo data for chosen parameter.")
        return {}

    df = df_param.copy()
    df["Session_dt"] = pd.to_datetime(df["Session_dt"])
    df = df[(df["Session_dt"] >= pd.Timestamp(start)) & (df["Session_dt"] <= pd.Timestamp(end))].copy()
    df = df.sort_values("Session_dt")
    if df.empty:
        print("\n--- Final Holdout (Risk) ---\nNo trades in holdout window.")
        return {}

    equity = starting_equity
    eq_curve = []
    day = None
    day_start_equity = equity
    day_pnl = 0.0
    trades_today = 0

    executed_R = []
    executed_usd = []

    for _, row in df.iterrows():
        d = pd.Timestamp(row["Session_dt"]).date()
        r = float(row["R_multiple"])

        if (day is None) or (d != day):
            day = d
            day_start_equity = equity
            day_pnl = 0.0
            trades_today = 0

        if trades_today >= int(daily_trade_cap):
            continue
        if day_pnl <= -max_daily_loss_pct * day_start_equity:
            continue

        risk_unit = risk_per_trade_pct * equity
        pnl_usd = r * risk_unit

        equity += pnl_usd
        day_pnl += pnl_usd
        trades_today += 1

        executed_R.append(r)
        executed_usd.append(pnl_usd)
        eq_curve.append(equity)

    R_arr = np.array(executed_R, dtype=float)
    usd_arr = np.array(executed_usd, dtype=float)
    mR = _metrics_from_R(R_arr)

    if usd_arr.size:
        gp = float(usd_arr[usd_arr > 0].sum())
        gl = float(-usd_arr[usd_arr < 0].sum())
        pf_usd = (gp / gl) if gl > 0 else (float("inf") if gp > 0 else 0.0)
    else:
        pf_usd = 0.0

    if eq_curve:
        eq = np.array(eq_curve)
        hwm = np.maximum.accumulate(eq)
        mdd = float(np.max(hwm - eq))
    else:
        mdd = 0.0

    print("\n--- Final Holdout (Risk-Managed) ---")
    print(f"Holdout window: {pd.Timestamp(start).date()} → {pd.Timestamp(end).date()}")
    print(f"Executed trades: {len(executed_R)}")
    print(f"Win%: {mR['WinRate']:.2f}% | PF(R): {mR['PF']:.3f} | AvgR: {mR['AvgR']:.3f} | SumR: {mR['PnL']:.2f}")
    print(f"Total PnL $: {float(usd_arr.sum()):,.2f} | PF($): {pf_usd:.3f} | Max DD $: {mdd:,.2f} | End Equity: {equity:,.2f}")

    return {
        "Trades": int(len(executed_R)),
        "WinRate_%": float(mR["WinRate"]),
        "PF_R": float(mR["PF"]),
        "AvgR": float(mR["AvgR"]),
        "SumR": float(mR["PnL"]),
        "TotPnL_$": float(usd_arr.sum()) if usd_arr.size else 0.0,
        "PF_$": float(pf_usd),
        "MaxDD_$": float(mdd),
        "EndEquity_$": float(equity),
    }

# =============================
# Parameter grid (build grid_returns)
# =============================

def build_grid_returns(folder: str) -> Dict[str, pd.DataFrame]:
    """
    Re-runs the backtest over a small parameter grid and returns:
      key -> DataFrame[Session_dt, R_multiple]
    Keys look like: "cushion=0.0005|wick=0.40"
    """
    grid = []
    for c in GRID_BREAKOUT_CUSHION:
        for w in GRID_WICK_FRAC:
            grid.append({"BREAKOUT_CUSHION_PCT": c, "MAX_OPPOSITE_WICK_FRAC": w})
    grid_keys = [f"cushion={g['BREAKOUT_CUSHION_PCT']:.5f}|wick={g['MAX_OPPOSITE_WICK_FRAC']:.2f}" for g in grid]

    # Progress bar for grid
    iterator = tqdm(list(zip(grid_keys, grid)), desc="Param grid") if TQDM_OK else list(zip(grid_keys, grid))
    grid_returns = {}

    for key, cfg in iterator:
        trades, _ = run_folder(folder, cfg=cfg)
        if trades.empty:
            grid_returns[key] = pd.DataFrame(columns=["Session_dt","R_multiple"])
            continue
        tmp = trades[["Session","R_multiple"]].copy()
        tmp["Session_dt"] = pd.to_datetime(tmp["Session"])
        tmp = tmp[["Session_dt","R_multiple"]].sort_values("Session_dt").reset_index(drop=True)
        grid_returns[key] = tmp

    return grid_returns

# =============================
# Sensitivity (optional)
# =============================

def sensitivity_slippage_fee(backtest_func, folder: str, slippages: List[float], fees: List[float]):
    rows = []
    for bps in slippages:
        for fee in fees:
            cfg = {"SLIPPAGE_BPS": bps, "FEES_PER_TRADE": fee}
            trades, _ = run_folder(folder, cfg=cfg)
            m = metrics_from_trades(trades)
            rows.append({"slippage_bps": bps, "fee_$": fee, **m})
    out = pd.DataFrame(rows).sort_values(["slippage_bps","fee_$"]).reset_index(drop=True)
    out.to_csv("orb_sensitivity_grid.csv", index=False)
    print("\n=== Slippage/Fee Sensitivity Grid ===")
    print(out.to_string(index=False))
    return out

# =============================
# Main
# =============================

def main():
    # Base run (current default config)
    trades, equity = run_folder(PARQUET_DIR)
    print_overall_totals(trades)

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    if not trades.empty:
        summary = (
            trades.groupby("Ticker")["PnL_$"]
                  .agg(['count', 'sum', 'mean', 'std'])
                  .rename(columns={'count':'Trades','sum':'PnL_sum','mean':'AvgPnL','std':'StdPnL'})
                  .reset_index()
                  .sort_values("PnL_sum", ascending=False)
        )
        wr = trades.groupby("Ticker")["PnL_$"].apply(lambda s: (pd.to_numeric(s, errors="coerce") > 0).mean()*100.0).rename("WinRate_%")
        pos = trades.groupby("Ticker")["PnL_$"].apply(lambda s: pd.to_numeric(s, errors="coerce").clip(lower=0).sum()).rename("GrossProfit")
        neg = trades.groupby("Ticker")["PnL_$"].apply(lambda s: -pd.to_numeric(s, errors="coerce").clip(upper=0).sum()).rename("GrossLoss")
        pf = (pos / neg.replace(0, np.nan)).rename("PF")
        summary = summary.merge(wr, on="Ticker", how="left").merge(pf, on="Ticker", how="left")
        print("\n=== Summary by Ticker (top 50) ===")
        print(summary.head(50).to_string(index=False))
        summary.to_csv(SUMMARY_CSV, index=False)
    else:
        print("No trades generated with current settings.")

    # ===== Validations =====
    if RUN_PURGED_KFOLD and not trades.empty:
        purged_kfold_cv(trades, n_splits=KFOLD_SPLITS, purge_days=PURGE_DAYS, embargo_days=EMBARGO_DAYS)

    if RUN_WALK_FORWARD and not trades.empty:
        walk_forward_validation(trades, train_days=WF_TRAIN_DAYS, test_days=WF_TEST_DAYS,
                                purge_days=PURGE_DAYS, embargo_days=EMBARGO_DAYS)

    # --- Build grid_returns for nested WFO ---
    grid_returns = {}
    if RUN_NESTED_WF and BUILD_PARAM_GRID:
        grid_returns = build_grid_returns(PARQUET_DIR)

    if RUN_NESTED_WF and grid_returns:
        wf_records, last_wf_end = nested_walk_forward_with_calibration(
            grid_returns,
            train_days=NWF_TRAIN_DAYS,
            calib_days=NWF_CALIB_DAYS,
            test_days=NWF_TEST_DAYS,
            purge_days=NWF_PURGE_DAYS,
            embargo_days=NWF_EMBARGO_DAYS,
            min_trades_calib=NWF_MIN_TRADES_CALIB
        )
        final_key = choose_final_param_from_wf(wf_records)
        print(f"\nChosen FINAL parameter key from WFO: {final_key}")

        # Strictly unseen holdout AFTER WFO
        if RUN_FINAL_HOLDOUT_RISK and final_key is not None:
            all_sessions = pd.to_datetime(
                np.unique(np.concatenate([df["Session_dt"].values for df in grid_returns.values() if not df.empty]))
            )
            all_sessions.sort_values(inplace=True)
            if HOLDOUT_START_AFTER_WF and last_wf_end is not None:
                holdout_start = pd.Timestamp(last_wf_end) + pd.Timedelta(days=1)
            else:
                holdout_start = all_sessions.min()
            holdout_end = all_sessions.max()

            simulate_holdout_with_risk(
                grid_returns.get(final_key, pd.DataFrame()),
                start=holdout_start,
                end=holdout_end,
                starting_equity=100_000.0,
                risk_per_trade_pct=0.005,
                max_daily_loss_pct=0.02,
                daily_trade_cap=20
            )

    if RUN_SENSITIVITY:
        sensitivity_slippage_fee(backtest_orb_retest, PARQUET_DIR, SENS_SLIPPAGE_LIST, SENS_FEE_LIST)

# ===== Entry point =====
if __name__ == "__main__":
    main()
```

```python
# orb_15m_parquet_institutional.py
# ORB 15m Retest/Continuation on a folder of .parquet files (2016→today),
# with indicators, realistic fills (confirm close -> next open),
# Profit Factor, progress bars, and institutional-grade validation:
# - Purged K-Fold CV (time-aware)
# - Walk-Forward Validation (purge/embargo)
# - Nested WFO (Train→Calibration→Test) picking a single param
# - Final strictly-unseen Holdout with portfolio risk management

import warnings
warnings.filterwarnings("ignore")

import os, glob, time
from datetime import timedelta, time as dtime
from typing import List, Tuple, Optional, Dict

import numpy as np
import pandas as pd

# Optional progress bar
try:
    from tqdm import tqdm
    TQDM_OK = True
except Exception:
    TQDM_OK = False

# =============================
# USER PARAMETERS
# =============================

# Parquet directory (ALL files inside will be processed)
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"

# If timestamps in your parquet are NAIVE and represent UTC, keep True
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic (quality + enough trades)
MAX_RETEST_MIN = 120            # minutes after breakout to wait for retest
RETEST_CONFIRM_CLOSE = True     # retest bar must close back through the level
BREAKOUT_CUSHION_PCT = 0.0005   # 0.05% cushion beyond OR to confirm breakout/retest
MAX_OPPOSITE_WICK_FRAC = 0.40   # reject entry bar if opposite wick > 40% of bar range

# Multiple trades per day
MAX_TRADES_PER_SESSION = 2
ALLOW_BOTH_SIDES = True

# Scale-out in thirds
USE_SCALE_OUT = True
TARGETS_R = [1.0, 2.0, 2.5]
SCALE_SPLIT = (1/3, 1/3, 1/3)

# Breakeven policy
MOVE_STOP_TO_BE_AT_R = 1.0      # arm BE after TP1 (==1R)
BE_ON_NEXT_BAR = True           # move stop to BE only on the next bar

# Continuation (fallback if no retest)
USE_CONTINUATION = True

# Time windows
BREAKOUT_DEADLINE   = "11:00"   # ET deadline for breakout
RETEST_DEADLINE     = "12:00"   # ET deadline for retest (kept for reference)
CONTINUATION_CUTOFF = "12:00"   # ET cutoff for continuation entries

# --- Indicator filters (ON) ---
USE_VWAP_FILTER   = True
USE_TREND_FILTER  = True
EMA_FAST = 20
EMA_SLOW = 50

USE_RSI_FILTER = True
RSI_LEN = 14
RSI_THRESH_LONG = 50.0
RSI_THRESH_SHORT = 50.0

USE_ADX_FILTER = False
ADX_LEN = 14
ADX_MIN = 18.0

USE_DAILY_TREND_FILTER = True
DAILY_EMA_FAST = 20
DAILY_EMA_SLOW = 50
DAILY_ATR_LEN  = 14

# OR width vs daily ATR band
USE_OR_WIDTH_ATR_FILTER = True
OR_ATR_MIN = 0.10
OR_ATR_MAX = 2.00

# Opening 15m volume quantile
USE_OPENING_VOL_FILTER = True
OPENING_VOL_LOOKBACK = 60
OPENING_VOL_QUANTILE = 0.30

# Gap filters
USE_GAP_FILTER = True
GAP_MIN = 0.003                 # 0.3%
GAP_MAX = 0.04                  # 4.0%
ALIGN_WITH_GAP_DIR = False      # allow counter-gap trades

USE_DOW_FILTER = False
ALLOWED_DOW = {1, 2, 3}         # Tue–Thu if enabled

# Entry realism when using close confirmation
#   "next_open" (recommended) = fill at next bar open after confirming close
#   "confirm_close"           = fill at confirming bar close
ENTRY_ON_CONFIRM = "next_open"

# Risk / costs
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00
FEE_PER_FILL = True             # charge fee per exit leg (TP/stop)

# Batch / persistence
SLEEP_BETWEEN_TICKERS = 0.0
AUTOSAVE_EVERY = 25

# Output files
TRADES_CSV_ALL = "orb_trades.csv"
EQUITY_CSV_ALL = "orb_equity.csv"
SUMMARY_CSV    = "orb_summary.csv"

# =============================
# VALIDATION TOGGLES
# =============================

RUN_PURGED_KFOLD = True
KFOLD_SPLITS = 5
PURGE_DAYS   = 2
EMBARGO_DAYS = 2

RUN_WALK_FORWARD = True
WF_TRAIN_DAYS = 365 * 2
WF_TEST_DAYS  = 90

RUN_NESTED_WF = True   # Train→Calib→Test per fold; produces chosen params
NWF_TRAIN_DAYS  = 365 * 2
NWF_CALIB_DAYS  = 60
NWF_TEST_DAYS   = 90
NWF_PURGE_DAYS  = 2
NWF_EMBARGO_DAYS= 2
NWF_MIN_TRADES_CALIB = 150

RUN_FINAL_HOLDOUT_RISK = True
HOLDOUT_START_AFTER_WF = True   # set True to begin holdout the day AFTER last nested-WF test

RUN_SENSITIVITY = False
SENS_SLIPPAGE_LIST = [0.5, 1.0, 2.0]
SENS_FEE_LIST      = [0.00, 0.25]

# Parameter grid (small, safe to scale up later)
BUILD_PARAM_GRID = True
GRID_BREAKOUT_CUSHION = [0.0003, 0.0005, 0.0010]
GRID_WICK_FRAC        = [0.30, 0.40, 0.50]

# =============================
# Helpers
# =============================

def ensure_unique_columns(df: pd.DataFrame) -> pd.DataFrame:
    if getattr(df.columns, "duplicated", None) is not None and df.columns.duplicated().any():
        df = df.loc[:, ~df.columns.duplicated()].copy()
    return df

def sessionize(obj) -> pd.Series:
    idx = getattr(obj, "index", obj)
    return pd.to_datetime(pd.Index(idx).date)

def _to_clock(ts) -> dtime:
    return dtime(ts.hour, ts.minute)

def _clock_le(ts, hhmm: str) -> bool:
    h, m = map(int, hhmm.split(":"))
    return _to_clock(ts) <= dtime(h, m)

def wick_opposite_fraction(row: pd.Series, direction: str) -> float:
    h, l, o, c = float(row["High"]), float(row["Low"]), float(row.get("Open", np.nan)), float(row["Close"])
    rng = max(h - l, 1e-12)
    if direction == "long":
        opp = min(o, c) - l
    else:
        opp = h - max(o, c)
    return float(max(opp, 0.0) / rng)

def true_range(high, low, prev_close):
    return np.maximum.reduce([
        (high - low).values,
        np.abs(high - prev_close).values,
        np.abs(low - prev_close).values
    ])

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def _normalize_splits(num_targets: int, splits: tuple) -> List[float]:
    if num_targets <= 0:
        return [1.0]
    if not splits:
        return [1.0 / num_targets] * num_targets
    arr = list(splits[:num_targets])
    if len(arr) < num_targets:
        arr += [0.0] * (num_targets - len(arr))
    s = sum(arr)
    if s <= 0:
        return [1.0 / num_targets] * num_targets
    return [x / s for x in arr]

# Nicely formatted totals (with PF)
def print_overall_totals(trades: pd.DataFrame):
    print("\n=== Overall Totals ===")
    if trades is None or trades.empty:
        print("Total trades: 0")
        print("Total PnL ($): 0.00")
        print("Win rate: 0.00%")
        print("Avg R: 0.000 | Median R: 0.000")
        print("Profit Factor: 0.000")
        print("Avg Win ($): 0.00 | Avg Loss ($): 0.00")
        return

    pnl = pd.to_numeric(trades["PnL_$"], errors="coerce").fillna(0.0)
    wins_mask = pnl > 0
    losses_mask = pnl < 0

    total_trades = int(len(pnl))
    total_pnl    = float(pnl.sum())
    win_rate     = float(wins_mask.mean() * 100.0) if total_trades else 0.0
    avg_r        = float(pd.to_numeric(trades["R_multiple"], errors="coerce").mean())
    median_r     = float(pd.to_numeric(trades["R_multiple"], errors="coerce").median())

    gross_profit = float(pnl[wins_mask].sum())
    gross_loss   = float(-pnl[losses_mask].sum())
    pf = (gross_profit / gross_loss) if gross_loss > 0 else (float("inf") if gross_profit > 0 else 0.0)

    avg_win  = float(pnl[wins_mask].mean()) if wins_mask.any() else 0.0
    avg_loss = float(pnl[losses_mask].mean()) if losses_mask.any() else 0.0

    print(f"Total trades: {total_trades}")
    print(f"Total PnL ($): {total_pnl:,.2f}")
    print(f"Win rate: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Profit Factor: {pf:.3f}")
    print(f"Avg Win ($): {avg_win:,.2f} | Avg Loss ($): {avg_loss:,.2f}")

# =============================
# Indicators & features
# =============================

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    for c in ("ORH","ORL"):
        if (df.columns == c).sum() > 0:
            df = df.drop(columns=[c])
    for col in ("High","Low"):
        if col not in df.columns:
            raise ValueError(f"compute_opening_range: missing column {col}")
    if not df.index.is_monotonic_increasing:
        df = df.sort_index()
    df["Session"] = sessionize(df)
    first_bar_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float).values
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float).values
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return ensure_unique_columns(df)

def add_session_vwap(df_15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df["Session"] = sessionize(df)
    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0
    if "Volume" in df.columns:
        df["vwap_num"] = tp * df["Volume"]
        df["vwap_den"] = df["Volume"].replace(0, np.nan)
    else:
        df["vwap_num"] = tp
        df["vwap_den"] = 1.0
    df["VWAP"] = (df.groupby("Session")["vwap_num"].cumsum() /
                  df.groupby("Session")["vwap_den"].cumsum())
    return ensure_unique_columns(df.drop(columns=["vwap_num","vwap_den"]))

def add_ema_trend(df_15: pd.DataFrame, fast=EMA_FAST, slow=EMA_SLOW) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return ensure_unique_columns(df)

def rsi(series: pd.Series, length=14) -> pd.Series:
    delta = series.diff()
    up = delta.clip(lower=0)
    dn = -delta.clip(upper=0)
    avg_gain = up.ewm(alpha=1/length, adjust=False).mean()
    avg_loss = dn.ewm(alpha=1/length, adjust=False).mean()
    rs = avg_gain / (avg_loss.replace(0, np.nan))
    return 100 - (100 / (1 + rs))

def add_rsi(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df[f"RSI{length}"] = rsi(df["Close"], length)
    return ensure_unique_columns(df)

def add_adx(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    up_move = df["High"].diff()
    dn_move = -df["Low"].diff()
    plus_dm  = np.where((up_move > dn_move) & (up_move > 0), up_move, 0.0)
    minus_dm = np.where((dn_move > up_move) & (dn_move > 0), dn_move, 0.0)

    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    atr = tr.ewm(alpha=1/length, adjust=False).mean()

    plus_di = 100 * pd.Series(plus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    minus_di = 100 * pd.Series(minus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    dx = (abs(plus_di - minus_di) / (plus_di + minus_di).replace(0, np.nan)) * 100
    adx = dx.ewm(alpha=1/length, adjust=False).mean()
    df[f"ADX{length}"] = adx
    return ensure_unique_columns(df)

def add_atr_15m(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    prev_close = df["Close"].shift(1)
    tr = pd.Series(true_range(df["High"], df["Low"], prev_close), index=df.index)
    df[f"ATR15_{length}"] = tr.ewm(alpha=1/length, adjust=False).mean()
    return ensure_unique_columns(df)

def build_daily_from_15m(df_15: pd.DataFrame) -> pd.DataFrame:
    d = ensure_unique_columns(df_15.copy())
    d["Session"] = sessionize(d)
    daily = d.groupby("Session").agg(
        Open=("Open","first"),
        High=("High","max"),
        Low=("Low","min"),
        Close=("Close","last"),
        Volume=("Volume","sum") if "Volume" in d.columns else ("Close","size"),
    )
    return daily

def add_daily_bias_and_atr(df_15: pd.DataFrame, ema_fast=20, ema_slow=50, atr_len=14) -> pd.DataFrame:
    dly = build_daily_from_15m(df_15)

    dly[f"EMA_D{ema_fast}"] = dly["Close"].ewm(span=ema_fast, adjust=False).mean()
    dly[f"EMA_D{ema_slow}"] = dly["Close"].ewm(span=ema_slow, adjust=False).mean()

    prev_close = dly["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (dly["High"] - dly["Low"]).values,
        np.abs(dly["High"] - prev_close).values,
        np.abs(dly["Low"]  - prev_close).values
    ]), index=dly.index)
    atr_col = f"ATR_D{atr_len}"
    dly[atr_col] = tr.ewm(alpha=1/atr_len, adjust=False).mean()

    if "Volume" in df_15.columns:
        tmp = df_15.copy()
        tmp["Session"] = sessionize(tmp)
        open15_vol = tmp.groupby("Session")["Volume"].first()
        dly["Open15_Vol"] = open15_vol
        dly["Open15_Vol_Thresh"] = (
            dly["Open15_Vol"]
            .rolling(window=OPENING_VOL_LOOKBACK, min_periods=10)
            .quantile(OPENING_VOL_QUANTILE)
        )

    feature_cols = [f"EMA_D{ema_fast}", f"EMA_D{ema_slow}", atr_col, "Open15_Vol", "Open15_Vol_Thresh"]
    feature_cols = [c for c in feature_cols if c in dly.columns]

    out = ensure_unique_columns(df_15.copy())
    out["Session"] = sessionize(out)
    out = out.merge(dly[feature_cols], left_on="Session", right_index=True, how="left")
    return ensure_unique_columns(out)

# =============================
# Backtest (per-ticker)
# =============================

def backtest_orb_retest(
    df_15: pd.DataFrame,
    cfg: Optional[Dict] = None
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    cfg can override any of the globals used below, e.g.:
      {"BREAKOUT_CUSHION_PCT":0.0005, "MAX_OPPOSITE_WICK_FRAC":0.4, ...}
    """
    C = lambda k, d=None: (cfg[k] if (cfg and k in cfg) else globals()[k]) if (d is None) else (cfg.get(k, d) if cfg else d)

    # Opening range + features
    df = compute_opening_range(df_15)
    if C("USE_VWAP_FILTER", True):  df = add_session_vwap(df)
    if C("USE_TREND_FILTER", True): df = add_ema_trend(df, C("EMA_FAST", EMA_FAST), C("EMA_SLOW", EMA_SLOW))
    df = add_daily_bias_and_atr(df, C("DAILY_EMA_FAST", DAILY_EMA_FAST), C("DAILY_EMA_SLOW", DAILY_EMA_SLOW), C("DAILY_ATR_LEN", DAILY_ATR_LEN))

    # Gap (first 15m vs previous day's close)
    df["Session"] = sessionize(df)
    first_idx = df.groupby("Session").head(1).index
    prev_close_series = df.groupby("Session")["Close"].last().shift(1)
    prev_close_map = df["Session"].map(prev_close_series)
    df.loc[first_idx, "GapPct"] = (df.loc[first_idx, "Open"] / prev_close_map.loc[first_idx] - 1.0)
    df["GapPct"] = df.groupby("Session")["GapPct"].ffill()

    # Intraday indicators
    if C("USE_RSI_FILTER", True):  df = add_rsi(df, C("RSI_LEN", RSI_LEN))
    if C("USE_ADX_FILTER", False): df = add_adx(df, C("ADX_LEN", ADX_LEN))
    df = ensure_unique_columns(df)

    sessions = df["Session"].unique()
    trades = []

    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3:
            continue

        orh, orl = float(sdf["ORH"].iloc[0]), float(sdf["ORL"].iloc[0])
        if np.isnan(orh) or np.isnan(orl):
            continue

        # All breakout candidates after the OR bar (use cushion)
        after_open = sdf.iloc[1:].copy()
        bcp = C("BREAKOUT_CUSHION_PCT", BREAKOUT_CUSHION_PCT)
        long_breaks  = after_open[after_open["Close"] > orh * (1 + bcp)]
        short_breaks = after_open[after_open["Close"] < orl * (1 - bcp)]

        cands = []
        for idx, row in long_breaks.iterrows():
            cands.append(("long", idx, row))
        for idx, row in short_breaks.iterrows():
            cands.append(("short", idx, row))
        cands.sort(key=lambda x: x[1])  # by timestamp

        trades_this_session = 0
        took_long = False
        took_short = False

        for direction, btime, breakout_row in cands:
            if not _clock_le(breakout_row.name, C("BREAKOUT_DEADLINE", BREAKOUT_DEADLINE)):
                continue

            if not C("ALLOW_BOTH_SIDES", True):
                if took_long and direction == "short": break
                if took_short and direction == "long": break
            if direction == "long" and took_long:   continue
            if direction == "short" and took_short: continue

            if C("USE_DOW_FILTER", False) and btime.weekday() not in C("ALLOWED_DOW", ALLOWED_DOW):
                continue

            if C("USE_GAP_FILTER", True) and "GapPct" in sdf.columns:
                gap = sdf.loc[sdf.index.min(), "GapPct"]
                if pd.notna(gap) and not (C("GAP_MIN", GAP_MIN) <= abs(float(gap)) <= C("GAP_MAX", GAP_MAX)):
                    continue
                if C("ALIGN_WITH_GAP_DIR", False) and pd.notna(gap):
                    g = float(gap)
                    if (direction == "long" and g < 0) or (direction == "short" and g > 0):
                        continue

            cutoff = btime + timedelta(minutes=C("MAX_RETEST_MIN", MAX_RETEST_MIN))
            window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
            level = orh if direction == "long" else orl
            touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)

            retest_used = True
            if touch.empty:
                if C("USE_CONTINUATION", True) and _clock_le(breakout_row.name, C("CONTINUATION_CUTOFF", CONTINUATION_CUTOFF)):
                    retest_used = False
                    retest_time = breakout_row.name
                    retest_bar = breakout_row
                else:
                    continue
            else:
                retest_bar = touch.iloc[0]
                retest_time = retest_bar.name

            if retest_used and C("RETEST_CONFIRM_CLOSE", True):
                if direction == "long":
                    ok = (retest_bar["Close"] >= level * (1 + bcp))
                else:
                    ok = (retest_bar["Close"] <= level * (1 - bcp))
                if not ok:
                    continue

            if C("MAX_OPPOSITE_WICK_FRAC", MAX_OPPOSITE_WICK_FRAC) is not None:
                if wick_opposite_fraction(retest_bar, direction) > C("MAX_OPPOSITE_WICK_FRAC", MAX_OPPOSITE_WICK_FRAC):
                    continue

            # Indicator filters
            if C("USE_VWAP_FILTER", True) and "VWAP" in sdf.columns:
                v_prev = sdf.loc[:retest_time, "VWAP"].tail(3).dropna().values
                rising  = (len(v_prev) < 2) or np.all(np.diff(v_prev) >= 0)
                falling = (len(v_prev) < 2) or np.all(np.diff(v_prev) <= 0)
                if direction == "long":
                    if not (retest_bar["Close"] > retest_bar["VWAP"] and rising):
                        continue
                else:
                    if not (retest_bar["Close"] < retest_bar["VWAP"] and falling):
                        continue

            if C("USE_TREND_FILTER", True) and (f"EMA{C('EMA_FAST', EMA_FAST)}" in sdf.columns) and (f"EMA{C('EMA_SLOW', EMA_SLOW)}" in sdf.columns):
                emaf = retest_bar.get(f"EMA{C('EMA_FAST', EMA_FAST)}", np.nan); emas = retest_bar.get(f"EMA{C('EMA_SLOW', EMA_SLOW)}", np.nan)
                if np.isfinite(emaf) and np.isfinite(emas):
                    if direction == "long" and not (emaf > emas): continue
                    if direction == "short" and not (emaf < emas): continue

            if C("USE_DAILY_TREND_FILTER", True) and f"EMA_D{C('DAILY_EMA_FAST', DAILY_EMA_FAST)}" in sdf.columns and f"EMA_D{C('DAILY_EMA_SLOW', DAILY_EMA_SLOW)}" in sdf.columns:
                dfast = float(sdf[f"EMA_D{C('DAILY_EMA_FAST', DAILY_EMA_FAST)}"].iloc[0])
                dslow = float(sdf[f"EMA_D{C('DAILY_EMA_SLOW', DAILY_EMA_SLOW)}"].iloc[0])
                if direction == "long" and not (dfast > dslow): continue
                if direction == "short" and not (dfast < dslow): continue

            atr_col = f"ATR_D{C('DAILY_ATR_LEN', DAILY_ATR_LEN)}"
            if C("USE_OR_WIDTH_ATR_FILTER", True) and atr_col in sdf.columns:
                or_width = float(orh - orl)
                atr_d = float(sdf[atr_col].iloc[0])
                if atr_d <= 0: continue
                frac = or_width / atr_d
                if not (C("OR_ATR_MIN", OR_ATR_MIN) <= frac <= C("OR_ATR_MAX", OR_ATR_MAX)): continue

            if C("USE_RSI_FILTER", True) and f"RSI{C('RSI_LEN', RSI_LEN)}" in sdf.columns:
                rsi_val = float(retest_bar[f"RSI{C('RSI_LEN', RSI_LEN)}"])
                if direction == "long" and not (rsi_val >= C("RSI_THRESH_LONG", RSI_THRESH_LONG)): continue
                if direction == "short" and not (rsi_val <= 100 - C("RSI_THRESH_SHORT", RSI_THRESH_SHORT)): continue

            if C("USE_ADX_FILTER", False) and f"ADX{C('ADX_LEN', ADX_LEN)}" in sdf.columns:
                adx_val = float(retest_bar[f"ADX{C('ADX_LEN', ADX_LEN)}"])
                if not (adx_val >= C("ADX_MIN", ADX_MIN)): continue

            # Entry price (realistic)
            if retest_used and C("RETEST_CONFIRM_CLOSE", True):
                if C("ENTRY_ON_CONFIRM", ENTRY_ON_CONFIRM).lower() == "next_open":
                    next_bar = sdf[sdf.index > retest_time].head(1)
                    if next_bar.empty:
                        continue
                    entry_time = next_bar.index[0]
                    raw_entry = float(next_bar["Open"].iloc[0])
                else:
                    entry_time = retest_time
                    raw_entry = float(retest_bar["Close"])
            else:
                entry_time = retest_time
                raw_entry = float(level)

            entry = _apply_slippage(raw_entry, C("SLIPPAGE_BPS", SLIPPAGE_BPS), "buy" if direction == "long" else "sell")
            stop  = orl if direction == "long" else orh
            rps   = abs(entry - stop)
            if rps <= 1e-12:
                continue

            qty = C("POSITION_SIZE_DOLLARS", POSITION_SIZE_DOLLARS) / max(entry, 1e-12)
            side_mult = 1 if direction == "long" else -1

            targets_r = TARGETS_R if C("USE_SCALE_OUT", True) else [2.0]
            targets = [(entry + r * rps) if direction == "long" else (entry - r * rps) for r in targets_r]

            run = sdf[sdf.index >= entry_time].copy()
            qty_left = qty
            exits = []
            next_tp_idx = 0

            if C("USE_SCALE_OUT", True) and len(targets) >= 1:
                splits = _normalize_splits(len(targets), C("SCALE_SPLIT", SCALE_SPLIT))
                leg_sizes = [qty * s for s in splits]
            else:
                leg_sizes = [qty]

            be_level = entry
            be_pending = False
            be_set_time = None

            for ts, row in run.iterrows():
                hi, lo = float(row["High"]), float(row["Low"])

                if be_pending and (not C("BE_ON_NEXT_BAR", BE_ON_NEXT_BAR) or (be_set_time is not None and ts > be_set_time)):
                    stop = be_level
                    be_pending = False

                # Stop first
                if lo <= stop <= hi:
                    px = _apply_slippage(stop, C("SLIPPAGE_BPS", SLIPPAGE_BPS), "sell" if direction == "long" else "buy")
                    label = "breakeven" if abs(px - be_level) < 1e-10 else "stop"
                    exits.append((label, ts, px, qty_left))
                    qty_left = 0.0
                    break

                # Targets
                if next_tp_idx < len(targets):
                    tp = targets[next_tp_idx]
                    if lo <= tp <= hi:
                        px = _apply_slippage(tp, C("SLIPPAGE_BPS", SLIPPAGE_BPS), "sell" if direction == "long" else "buy")
                        fill_qty = leg_sizes[next_tp_idx] if next_tp_idx < len(leg_sizes) else qty_left
                        exits.append((f"tp{next_tp_idx+1}", ts, px, fill_qty))
                        qty_left -= fill_qty
                        if next_tp_idx == 0 and C("MOVE_STOP_TO_BE_AT_R", MOVE_STOP_TO_BE_AT_R) is not None:
                            be_pending = True
                            be_set_time = ts
                        next_tp_idx += 1
                        if qty_left <= 1e-9:
                            break

                if C("USE_CONTINUATION", True) and not _clock_le(ts, C("CONTINUATION_CUTOFF", CONTINUATION_CUTOFF)) and next_tp_idx == 0:
                    stop = be_level

            # EOD exit
            if qty_left > 1e-9:
                last = run.iloc[-1]
                px = _apply_slippage(float(last["Close"]), C("SLIPPAGE_BPS", SLIPPAGE_BPS), "sell" if direction == "long" else "buy")
                exits.append(("eod", run.index[-1], px, qty_left))
                qty_left = 0.0

            cash_pnl = sum((px - entry) * side_mult * q for (_, _, px, q) in exits)
            if C("FEE_PER_FILL", FEE_PER_FILL):
                cash_pnl -= C("FEES_PER_TRADE", FEES_PER_TRADE) * len(exits)
            else:
                cash_pnl -= C("FEES_PER_TRADE", FEES_PER_TRADE)

            trades.append({
                "Ticker": sdf["Ticker"].iloc[0] if "Ticker" in sdf.columns else "",
                "Session": ses,
                "Direction": direction,
                "ORH": orh, "ORL": orl,
                "EntryTime": entry_time,
                "Entry": entry, "Stop": stop,
                "Targets": targets,
                "Exits": [(lab, ts, px, q) for (lab, ts, px, q) in exits],
                "Qty": qty,
                "PnL_$": cash_pnl,
                "R_multiple": cash_pnl / (rps * max(qty, 1e-12))
            })

            if direction == "long":  took_long  = True
            if direction == "short": took_short = True
            trades_this_session += 1
            if trades_this_session >= C("MAX_TRADES_PER_SESSION", MAX_TRADES_PER_SESSION):
                break

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Parquet loading
# =============================

STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                rename[lower[k]] = std
        elif k.capitalize() in df.columns: rename[k.capitalize()] = std
        elif k.upper() in df.columns: rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _maybe_resample_to_15m(df: pd.DataFrame) -> pd.DataFrame:
    if len(df.index) < 3:
        return df
    deltas = (df.index[1:] - df.index[:-1]).asi8
    if len(deltas) == 0:
        return df
    med_ns = np.median(deltas)
    fifteen_ns = pd.Timedelta("15T").value
    if med_ns < fifteen_ns:
        o = df["Open"].resample("15T", label="left", closed="left").first()
        h = df["High"].resample("15T", label="left", closed="left").max()
        l = df["Low"].resample("15T", label="left", closed="left").min()
        c = df["Close"].resample("15T", label="left", closed="left").last()
        v = df["Volume"].resample("15T", label="left", closed="left").sum() if "Volume" in df.columns else None
        parts = {"Open":o,"High":h,"Low":l,"Close":c}
        if v is not None: parts["Volume"] = v
        out = pd.concat(parts, axis=1).dropna(subset=["Open","High","Low","Close"])
        return out
    return df

def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()

    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df = df.tz_localize("UTC").tz_convert(tz)
        else:
            df = df.tz_localize(tz)
    else:
        df = df.tz_convert(tz)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    df = _maybe_resample_to_15m(df)
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)
    return ensure_unique_columns(df.sort_index())

# =============================
# Runner (folder of parquet files)
# =============================

def run_folder(folder: str, cfg: Optional[Dict]=None) -> Tuple[pd.DataFrame, pd.DataFrame]:
    files = sorted(glob.glob(os.path.join(folder, "*.parquet")))
    print(f"Found {len(files)} parquet files.")
    all_trades, all_equity = [], []

    iterator = tqdm(files, desc="Backtesting files") if TQDM_OK else files
    for i, fpath in enumerate(iterator, 1):
        ticker = os.path.splitext(os.path.basename(fpath))[0].upper().replace("_","").replace("-","")
        msg = f"== {ticker} ({i}/{len(files)}) =="
        if not TQDM_OK:
            print("\n" + msg)
        try:
            df15 = load_parquet_15m(fpath)
            df15["Ticker"] = ticker
        except Exception as e:
            print(f"Skipping {ticker}: {e}")
            continue

        tlog, eq = backtest_orb_retest(df15, cfg=cfg)

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"]   = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}|{t[3]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy(); eq_out["Ticker"] = ticker
            all_equity.append(eq_out)

        if AUTOSAVE_EVERY and i % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

        if SLEEP_BETWEEN_TICKERS > 0:
            time.sleep(SLEEP_BETWEEN_TICKERS)

    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()
    return trades, equity

# =============================
# Validation utilities
# =============================

def metrics_from_trades(df: pd.DataFrame) -> dict:
    if df is None or df.empty:
        return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, PnL=0.0)
    pnl = pd.to_numeric(df["PnL_$"], errors="coerce").fillna(0.0)
    r   = pd.to_numeric(df["R_multiple"], errors="coerce").fillna(0.0)
    wins = pnl > 0
    gp = float(pnl[wins].sum())
    gl = float(-pnl[~wins & (pnl<0)].sum())
    pf = (gp / gl) if gl > 0 else (float("inf") if gp > 0 else 0.0)
    return dict(
        Trades = int(len(df)),
        WinRate = float(wins.mean() * 100.0) if len(df) else 0.0,
        PF = float(pf),
        AvgR = float(r.mean()),
        PnL = float(pnl.sum())
    )

def purged_kfold_cv(trades: pd.DataFrame, n_splits=5, purge_days=2, embargo_days=2):
    print("\n--- Purged K-Fold CV ---")
    if trades.empty:
        print("No trades.")
        return
    trades = trades.copy()
    trades["Session_dt"] = pd.to_datetime(trades["Session"])
    sessions = np.sort(trades["Session_dt"].unique())
    folds = np.array_split(sessions, n_splits)

    for i, test_sessions in enumerate(folds, 1):
        test_start = pd.Timestamp(test_sessions.min())
        test_end   = pd.Timestamp(test_sessions.max())

        purge_end = test_start - pd.Timedelta(days=purge_days)
        embargo_start = test_end + pd.Timedelta(days=1)
        embargo_end   = embargo_start + pd.Timedelta(days=embargo_days)

        train = trades[(trades["Session_dt"] <= purge_end) | (trades["Session_dt"] >= embargo_end)]
        test  = trades[(trades["Session_dt"] >= test_start) & (trades["Session_dt"] <= test_end)]

        m_tr = metrics_from_trades(train)
        m_te = metrics_from_trades(test)

        print(f"Fold {i:02d} Train [..→{pd.Timestamp(purge_end).date()}] | Trades: {m_tr['Trades']:,} | Win%: {m_tr['WinRate']:.2f}% | PF: {m_tr['PF']:.3f} | Avg R: {m_tr['AvgR']:.3f} | PnL: ${m_tr['PnL']:,.2f}")
        print(f"Fold {i:02d} Test  [{pd.Timestamp(test_start).date()}→{pd.Timestamp(test_end).date()}] | Trades: {m_te['Trades']:,} | Win%: {m_te['WinRate']:.2f}% | PF: {m_te['PF']:.3f} | Avg R: {m_te['AvgR']:.3f} | PnL: ${m_te['PnL']:,.2f}")

def walk_forward_validation(trades: pd.DataFrame, train_days=365*2, test_days=90, purge_days=2, embargo_days=2):
    print("\n--- Walk-Forward Validation ---")
    if trades.empty:
        print("No trades.")
        return
    df = trades.copy()
    df["Session_dt"] = pd.to_datetime(df["Session"])
    sessions_sorted = np.sort(df["Session_dt"].unique())
    t0, t1 = sessions_sorted.min(), sessions_sorted.max()

    fold = 1
    train_start = pd.Timestamp(t0)

    while True:
        train_end = train_start + pd.Timedelta(days=train_days) - pd.Timedelta(days=1)
        if train_end >= t1:
            break

        test_start = train_end + pd.Timedelta(days=1)
        test_end   = test_start + pd.Timedelta(days=test_days) - pd.Timedelta(days=1)
        if test_start > t1:
            break

        # purge from end of train
        train_cut = train_end - pd.Timedelta(days=purge_days)
        test_eval_start = test_start + pd.Timedelta(days=embargo_days)
        test_eval_end   = min(test_end, t1)

        trn = df[(df["Session_dt"] >= train_start) & (df["Session_dt"] <= train_cut)]
        tst = df[(df["Session_dt"] >= test_eval_start) & (df["Session_dt"] <= test_eval_end)]

        m_trn = metrics_from_trades(trn)
        m_tst = metrics_from_trades(tst)

        print(f"Fold {fold:02d} Train [{pd.Timestamp(train_start).date()}→{pd.Timestamp(train_cut).date()}] | Trades: {m_trn['Trades']:,} | Win%: {m_trn['WinRate']:.2f}% | PF: {m_trn['PF']:.3f} | Avg R: {m_trn['AvgR']:.3f} | PnL: ${m_trn['PnL']:,.2f}")
        print(f"Fold {fold:02d} Test  [{pd.Timestamp(test_eval_start).date()}→{pd.Timestamp(test_eval_end).date()}] | Trades: {m_tst['Trades']:,} | Win%: {m_tst['WinRate']:.2f}% | PF: {m_tst['PF']:.3f} | Avg R: {m_tst['AvgR']:.3f} | PnL: ${m_tst['PnL']:,.2f}")

        train_start = test_start
        fold += 1
        if train_start >= t1:
            break

# ---------- Nested WFO (Train→Calib→Test) + final holdout ----------
def _metrics_from_R(r: np.ndarray):
    if r.size == 0:
        return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, PnL=0.0)
    wr = 100.0 * (r > 0).mean()
    gp = float(r[r > 0].sum())
    gl = float(-r[r < 0].sum())
    pf = (gp / gl) if gl > 0 else (float("inf") if gp > 0 else 0.0)
    return dict(Trades=int(len(r)), WinRate=float(wr), PF=float(pf), AvgR=float(np.mean(r)), PnL=float(np.sum(r)))

def choose_final_param_from_wf(wf_records: list) -> Optional[str]:
    if not wf_records:
        return None
    df = pd.DataFrame(wf_records)
    counts = df["chosen_key"].value_counts()
    top_freq = counts.max()
    finalists = counts[counts == top_freq].index.tolist()
    if len(finalists) == 1:
        return finalists[0]
    med = (df[df["chosen_key"].isin(finalists)]
           .groupby("chosen_key")["test_AvgR"].median()
           .sort_values(ascending=False))
    return med.index[0]

def nested_walk_forward_with_calibration(
    grid_returns: Dict[str, pd.DataFrame],
    train_days: int = 365*2,
    calib_days: int = 60,
    test_days: int = 90,
    purge_days: int = 2,
    embargo_days: int = 2,
    min_trades_calib: int = 150,
) -> Tuple[list, Optional[pd.Timestamp]]:
    # Sessions from any grid member
    all_sess = np.unique(np.concatenate([
        df["Session_dt"].values for df in grid_returns.values() if not df.empty
    ])) if grid_returns else np.array([])
    if all_sess.size == 0:
        print("\n--- Nested Walk-Forward (Train→Calib→Test) ---\nNo sessions.")
        return [], None

    all_sess = pd.to_datetime(all_sess)
    all_sess.sort_values(inplace=True)
    t0, t1 = all_sess.min(), all_sess.max()
    print("\n--- Nested Walk-Forward (Train→Calib→Test) ---")

    def seg_metrics(grid, start, end, min_trades=0):
        rows = []
        for key, df in grid.items():
            if df.empty:
                rows.append({"key": key, "Trades": 0, "WinRate": 0.0, "PF": 0.0, "AvgR": 0.0, "PnL": 0.0})
                continue
            msk = (df["Session_dt"] >= start) & (df["Session_dt"] <= end)
            r = df.loc[msk, "R_multiple"].astype(float).values
            m = _metrics_from_R(r)
            m["key"] = key
            rows.append(m)
        out = pd.DataFrame(rows)
        out["eligible"] = out["Trades"] >= int(min_trades)
        return out

    def pick_on_calib(m_calib: pd.DataFrame) -> str:
        elig = m_calib[m_calib["eligible"]].copy()
        if elig.empty:
            elig = m_calib.copy()
        elig = elig.sort_values(["AvgR","PF","WinRate","Trades"], ascending=[False, False, False, False])
        return str(elig.iloc[0]["key"])

    fold = 1
    train_start = pd.Timestamp(t0)
    records = []

    while True:
        train_end = train_start + pd.Timedelta(days=train_days) - pd.Timedelta(days=1)
        if train_end >= t1:
            break

        calib_start = train_end - pd.Timedelta(days=calib_days) + pd.Timedelta(days=1)
        train_cut   = calib_start - pd.Timedelta(days=purge_days)

        test_start = train_end + pd.Timedelta(days=1)
        test_end   = test_start + pd.Timedelta(days=test_days) - pd.Timedelta(days=1)
        if test_start > t1:
            break
        test_eval_start = test_start + pd.Timedelta(days=embargo_days)
        test_eval_end   = min(test_end, t1)

        m_train = seg_metrics(grid_returns, train_start, train_cut, min_trades=0)
        m_calib = seg_metrics(grid_returns, calib_start, train_end, min_trades=min_trades_calib)

        chosen_key = pick_on_calib(m_calib)
        df_best = grid_returns.get(chosen_key, pd.DataFrame())
        if df_best.empty:
            test_m = dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, PnL=0.0)
        else:
            r = df_best[(df_best["Session_dt"] >= test_eval_start) & (df_best["Session_dt"] <= test_eval_end)]["R_multiple"].astype(float).values
            test_m = _metrics_from_R(r)

        print(f"Fold {fold:02d}")
        print(f"  Train [{pd.Timestamp(train_start).date()}→{pd.Timestamp(train_cut).date()}]  (purge {purge_days}d)")
        print(f"  Calib [{pd.Timestamp(calib_start).date()}→{pd.Timestamp(train_end).date()}]  pick={chosen_key}")
        print(f"  Test  [{pd.Timestamp(test_eval_start).date()}→{pd.Timestamp(test_eval_end).date()}] | Trades: {test_m['Trades']:,} | Win%: {test_m['WinRate']:.2f}% | PF: {test_m['PF']:.3f} | AvgR: {test_m['AvgR']:.3f} | SumR: {test_m['PnL']:.2f}")

        records.append({
            "fold": fold,
            "train_start": train_start, "train_end": train_end,
            "calib_start": calib_start, "calib_end": train_end,
            "test_start": test_start, "test_end": test_end,
            "test_eval_start": test_eval_start, "test_eval_end": test_eval_end,
            "chosen_key": chosen_key,
            "test_Trades": test_m["Trades"],
            "test_WinRate": test_m["WinRate"],
            "test_PF": test_m["PF"],
            "test_AvgR": test_m["AvgR"],
            "test_PnL_R": test_m["PnL"],
        })

        train_start = test_start
        fold += 1
        if train_start >= t1:
            break

    if records:
        oos_wr = np.mean([r["test_WinRate"] for r in records])
        oos_pf = np.mean([r["test_PF"] for r in records if np.isfinite(r["test_PF"])])
        print(f"Nested WF OOS Avg | Win%: {oos_wr:.2f}% | PF: {oos_pf:.3f}")

    last_eval_end = records[-1]["test_eval_end"] if records else None
    return records, last_eval_end

def simulate_holdout_with_risk(df_param: pd.DataFrame,
                               start, end,
                               starting_equity: float = 100_000.0,
                               risk_per_trade_pct: float = 0.005,
                               max_daily_loss_pct: float = 0.02,
                               daily_trade_cap: int = 20):
    if df_param is None or df_param.empty:
        print("\n--- Final Holdout (Risk) ---\nNo data for chosen parameter.")
        return {}

    df = df_param.copy()
    df["Session_dt"] = pd.to_datetime(df["Session_dt"])
    df = df[(df["Session_dt"] >= pd.Timestamp(start)) & (df["Session_dt"] <= pd.Timestamp(end))].copy()
    df = df.sort_values("Session_dt")
    if df.empty:
        print("\n--- Final Holdout (Risk) ---\nNo trades in holdout window.")
        return {}

    equity = starting_equity
    eq_curve = []
    day = None
    day_start_equity = equity
    day_pnl = 0.0
    trades_today = 0

    executed_R = []
    executed_usd = []

    for _, row in df.iterrows():
        d = pd.Timestamp(row["Session_dt"]).date()
        r = float(row["R_multiple"])

        if (day is None) or (d != day):
            day = d
            day_start_equity = equity
            day_pnl = 0.0
            trades_today = 0

        if trades_today >= int(daily_trade_cap):
            continue
        if day_pnl <= -max_daily_loss_pct * day_start_equity:
            continue

        risk_unit = risk_per_trade_pct * equity
        pnl_usd = r * risk_unit

        equity += pnl_usd
        day_pnl += pnl_usd
        trades_today += 1

        executed_R.append(r)
        executed_usd.append(pnl_usd)
        eq_curve.append(equity)

    R_arr = np.array(executed_R, dtype=float)
    usd_arr = np.array(executed_usd, dtype=float)
    mR = _metrics_from_R(R_arr)

    if usd_arr.size:
        gp = float(usd_arr[usd_arr > 0].sum())
        gl = float(-usd_arr[usd_arr < 0].sum())
        pf_usd = (gp / gl) if gl > 0 else (float("inf") if gp > 0 else 0.0)
    else:
        pf_usd = 0.0

    if eq_curve:
        eq = np.array(eq_curve)
        hwm = np.maximum.accumulate(eq)
        mdd = float(np.max(hwm - eq))
    else:
        mdd = 0.0

    print("\n--- Final Holdout (Risk-Managed) ---")
    print(f"Holdout window: {pd.Timestamp(start).date()} → {pd.Timestamp(end).date()}")
    print(f"Executed trades: {len(executed_R)}")
    print(f"Win%: {mR['WinRate']:.2f}% | PF(R): {mR['PF']:.3f} | AvgR: {mR['AvgR']:.3f} | SumR: {mR['PnL']:.2f}")
    print(f"Total PnL $: {float(usd_arr.sum()):,.2f} | PF($): {pf_usd:.3f} | Max DD $: {mdd:,.2f} | End Equity: {equity:,.2f}")

    return {
        "Trades": int(len(executed_R)),
        "WinRate_%": float(mR["WinRate"]),
        "PF_R": float(mR["PF"]),
        "AvgR": float(mR["AvgR"]),
        "SumR": float(mR["PnL"]),
        "TotPnL_$": float(usd_arr.sum()) if usd_arr.size else 0.0,
        "PF_$": float(pf_usd),
        "MaxDD_$": float(mdd),
        "EndEquity_$": float(equity),
    }

# =============================
# Parameter grid (build grid_returns)
# =============================

def build_grid_returns(folder: str) -> Dict[str, pd.DataFrame]:
    """
    Re-runs the backtest over a small parameter grid and returns:
      key -> DataFrame[Session_dt, R_multiple]
    Keys look like: "cushion=0.0005|wick=0.40"
    """
    grid = []
    for c in GRID_BREAKOUT_CUSHION:
        for w in GRID_WICK_FRAC:
            grid.append({"BREAKOUT_CUSHION_PCT": c, "MAX_OPPOSITE_WICK_FRAC": w})
    grid_keys = [f"cushion={g['BREAKOUT_CUSHION_PCT']:.5f}|wick={g['MAX_OPPOSITE_WICK_FRAC']:.2f}" for g in grid]

    # Progress bar for grid
    iterator = tqdm(list(zip(grid_keys, grid)), desc="Param grid") if TQDM_OK else list(zip(grid_keys, grid))
    grid_returns = {}

    for key, cfg in iterator:
        trades, _ = run_folder(folder, cfg=cfg)
        if trades.empty:
            grid_returns[key] = pd.DataFrame(columns=["Session_dt","R_multiple"])
            continue
        tmp = trades[["Session","R_multiple"]].copy()
        tmp["Session_dt"] = pd.to_datetime(tmp["Session"])
        tmp = tmp[["Session_dt","R_multiple"]].sort_values("Session_dt").reset_index(drop=True)
        grid_returns[key] = tmp

    return grid_returns

# =============================
# Sensitivity (optional)
# =============================

def sensitivity_slippage_fee(backtest_func, folder: str, slippages: List[float], fees: List[float]):
    rows = []
    for bps in slippages:
        for fee in fees:
            cfg = {"SLIPPAGE_BPS": bps, "FEES_PER_TRADE": fee}
            trades, _ = run_folder(folder, cfg=cfg)
            m = metrics_from_trades(trades)
            rows.append({"slippage_bps": bps, "fee_$": fee, **m})
    out = pd.DataFrame(rows).sort_values(["slippage_bps","fee_$"]).reset_index(drop=True)
    out.to_csv("orb_sensitivity_grid.csv", index=False)
    print("\n=== Slippage/Fee Sensitivity Grid ===")
    print(out.to_string(index=False))
    return out

# =============================
# Main
# =============================

def main():
    # Base run (current default config)
    trades, equity = run_folder(PARQUET_DIR)
    print_overall_totals(trades)

    if not trades.empty:
        trades.sort_values(["Ticker","Session"], inplace=True)
        trades.to_csv(TRADES_CSV_ALL, index=False)
        print(f"Saved trades: {TRADES_CSV_ALL}")

    if not equity.empty:
        equity.sort_values(["Ticker","Session"], inplace=True)
        equity.to_csv(EQUITY_CSV_ALL, index=False)
        print(f"Saved equity: {EQUITY_CSV_ALL}")

    if not trades.empty:
        summary = (
            trades.groupby("Ticker")["PnL_$"]
                  .agg(['count', 'sum', 'mean', 'std'])
                  .rename(columns={'count':'Trades','sum':'PnL_sum','mean':'AvgPnL','std':'StdPnL'})
                  .reset_index()
                  .sort_values("PnL_sum", ascending=False)
        )
        wr = trades.groupby("Ticker")["PnL_$"].apply(lambda s: (pd.to_numeric(s, errors="coerce") > 0).mean()*100.0).rename("WinRate_%")
        pos = trades.groupby("Ticker")["PnL_$"].apply(lambda s: pd.to_numeric(s, errors="coerce").clip(lower=0).sum()).rename("GrossProfit")
        neg = trades.groupby("Ticker")["PnL_$"].apply(lambda s: -pd.to_numeric(s, errors="coerce").clip(upper=0).sum()).rename("GrossLoss")
        pf = (pos / neg.replace(0, np.nan)).rename("PF")
        summary = summary.merge(wr, on="Ticker", how="left").merge(pf, on="Ticker", how="left")
        print("\n=== Summary by Ticker (top 50) ===")
        print(summary.head(50).to_string(index=False))
        summary.to_csv(SUMMARY_CSV, index=False)
    else:
        print("No trades generated with current settings.")

    # ===== Validations =====
    if RUN_PURGED_KFOLD and not trades.empty:
        purged_kfold_cv(trades, n_splits=KFOLD_SPLITS, purge_days=PURGE_DAYS, embargo_days=EMBARGO_DAYS)

    if RUN_WALK_FORWARD and not trades.empty:
        walk_forward_validation(trades, train_days=WF_TRAIN_DAYS, test_days=WF_TEST_DAYS,
                                purge_days=PURGE_DAYS, embargo_days=EMBARGO_DAYS)

    # --- Build grid_returns for nested WFO ---
    grid_returns = {}
    if RUN_NESTED_WF and BUILD_PARAM_GRID:
        grid_returns = build_grid_returns(PARQUET_DIR)

    if RUN_NESTED_WF and grid_returns:
        wf_records, last_wf_end = nested_walk_forward_with_calibration(
            grid_returns,
            train_days=NWF_TRAIN_DAYS,
            calib_days=NWF_CALIB_DAYS,
            test_days=NWF_TEST_DAYS,
            purge_days=NWF_PURGE_DAYS,
            embargo_days=NWF_EMBARGO_DAYS,
            min_trades_calib=NWF_MIN_TRADES_CALIB
        )
        final_key = choose_final_param_from_wf(wf_records)
        print(f"\nChosen FINAL parameter key from WFO: {final_key}")

        # Strictly unseen holdout AFTER WFO
        if RUN_FINAL_HOLDOUT_RISK and final_key is not None:
            all_sessions = pd.to_datetime(
                np.unique(np.concatenate([df["Session_dt"].values for df in grid_returns.values() if not df.empty]))
            )
            all_sessions.sort_values(inplace=True)
            if HOLDOUT_START_AFTER_WF and last_wf_end is not None:
                holdout_start = pd.Timestamp(last_wf_end) + pd.Timedelta(days=1)
            else:
                holdout_start = all_sessions.min()
            holdout_end = all_sessions.max()

            simulate_holdout_with_risk(
                grid_returns.get(final_key, pd.DataFrame()),
                start=holdout_start,
                end=holdout_end,
                starting_equity=100_000.0,
                risk_per_trade_pct=0.005,
                max_daily_loss_pct=0.02,
                daily_trade_cap=20
            )

    if RUN_SENSITIVITY:
        sensitivity_slippage_fee(backtest_orb_retest, PARQUET_DIR, SENS_SLIPPAGE_LIST, SENS_FEE_LIST)

# ===== Entry point =====
if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_parquet_industry_WF.py
# ORB 15m Retest/Continuation strategy on every .parquet in a folder.
# Includes indicators/filters, realistic entries, scale-out, PF, CV, WF, and Nested WFO (Train→Calibrate→Test).
# Robust date handling and progress bars. Fixed boolean-mask alignment in CV/WF.

import warnings
warnings.filterwarnings("ignore")

import os, glob, time, itertools
from datetime import timedelta, time as dtime
from typing import List, Tuple, Optional, Dict

import numpy as np
import pandas as pd

try:
    from tqdm import tqdm
except ImportError:
    tqdm = None  # fallback if tqdm isn't installed

# =============================
# USER PARAMETERS
# =============================

# Parquet directory (ALL files inside will be processed)
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"

# If timestamps in your parquet are NAIVE and represent UTC, keep True
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

# Session / timezone
TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic (quality + enough trades)
MAX_RETEST_MIN = 120            # minutes after breakout to wait for retest
RETEST_CONFIRM_CLOSE = True     # retest bar must close back through the level
BREAKOUT_CUSHION_PCT = 0.0005   # 0.05% cushion beyond OR to confirm breakout/retest
MAX_OPPOSITE_WICK_FRAC = 0.40   # reject entry bar if opposite wick > 40% of bar range

# Multiple trades per day
MAX_TRADES_PER_SESSION = 2
ALLOW_BOTH_SIDES = True

# Scale-out in thirds
USE_SCALE_OUT = True
TARGETS_R = [1.0, 2.0, 2.5]
SCALE_SPLIT = (1/3, 1/3, 1/3)

# Breakeven policy
MOVE_STOP_TO_BE_AT_R = 1.0      # arm BE after TP1 (==1R)
BE_ON_NEXT_BAR = True           # move stop to BE only on the next bar

# Continuation (fallback if no retest)
USE_CONTINUATION = True

# Time windows
BREAKOUT_DEADLINE   = "11:00"   # ET deadline for breakout
RETEST_DEADLINE     = "12:00"   # ET deadline for retest (kept for reference; not used separately)
CONTINUATION_CUTOFF = "12:00"   # ET cutoff for continuation entries

# --- Filters (kept & switchable) ---
# Intraday indicators
USE_VWAP_FILTER   = True
USE_TREND_FILTER  = True   # 15m EMA trend filter
EMA_FAST = 20
EMA_SLOW = 50

USE_RSI_FILTER = True
RSI_LEN = 14
RSI_THRESH_LONG = 50.0
RSI_THRESH_SHORT = 50.0

USE_ADX_FILTER = False
ADX_LEN = 14
ADX_MIN = 18.0

# Daily trend + ATR band for OR width
USE_DAILY_TREND_FILTER = True
DAILY_EMA_FAST = 20
DAILY_EMA_SLOW = 50
DAILY_ATR_LEN  = 14

USE_OR_WIDTH_ATR_FILTER = True
OR_ATR_MIN = 0.10
OR_ATR_MAX = 2.00

# Gap filter
USE_GAP_FILTER = True
GAP_MIN = 0.003                 # 0.3%
GAP_MAX = 0.04                  # 4.0%
ALIGN_WITH_GAP_DIR = False      # require trade direction to match gap direction

# Day-of-week filter (optional)
USE_DOW_FILTER = False
ALLOWED_DOW = {1, 2, 3}         # Tue–Thu if enabled

# Entry realism when using close confirmation
#   "next_open" (recommended) = fill at next bar open after confirming close
#   "confirm_close"           = fill at confirming bar close
ENTRY_ON_CONFIRM = "next_open"

# Risk / costs
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00
FEE_PER_FILL = True             # charge fee per exit leg (TP/stop)

# Batch / persistence
SLEEP_BETWEEN_TICKERS = 0.0
AUTOSAVE_EVERY = 25

# Output files
TRADES_CSV_ALL = "orb_trades_sp500.csv"
EQUITY_CSV_ALL = "orb_equity_sp500.csv"
SUMMARY_CSV    = "orb_summary_sp500.csv"

# =============================
# VALIDATION FLAGS
# =============================

RUN_BASELINE = True
RUN_PURGED_KFOLD = True
KFOLD_SPLITS = 5
PURGE_DAYS   = 2
EMBARGO_DAYS = 2

RUN_WALK_FORWARD = True
WF_TRAIN_DAYS = 365 * 2
WF_TEST_DAYS  = 180

RUN_NESTED_WF = True  # Train→Calibrate→Test with grid
# Nested WF window sizes
NWF_TRAIN_DAYS  = 365 * 2
NWF_CALIB_DAYS  = 120
NWF_TEST_DAYS   = 180
NWF_PURGE_DAYS  = 2
NWF_EMBARGO_DAYS= 2
NWF_MIN_TRADES_CALIB = 200

# Parameter grid (kept compact; expand as needed)
GRID_EMA_FAST = [12, 20]
GRID_EMA_SLOW = [26, 50]
GRID_BREAKOUT_CUSHION = [0.0005, 0.0010, 0.0015]

# =============================
# Helpers
# =============================

def ensure_unique_columns(df: pd.DataFrame) -> pd.DataFrame:
    if getattr(df.columns, "duplicated", None) is not None and df.columns.duplicated().any():
        df = df.loc[:, ~df.columns.duplicated()].copy()
    return df

def sessionize(obj) -> pd.Series:
    idx = getattr(obj, "index", obj)
    return pd.to_datetime(pd.Index(idx).date)

def session_series(df_or_series) -> pd.Series:
    """
    Return a tz-naive, normalized per-row session Series aligned to the original object.
    Use this to build boolean masks that match DataFrame length.
    """
    if isinstance(df_or_series, pd.DataFrame):
        s = df_or_series["Session"] if "Session" in df_or_series.columns else df_or_series.index
    else:
        s = df_or_series
    s = pd.to_datetime(s, errors="coerce")
    s = pd.Series(pd.DatetimeIndex(s).tz_localize(None).normalize(), index=getattr(df_or_series, "index", None))
    return s

def _to_clock(ts) -> dtime:
    ts = pd.Timestamp(ts)
    return dtime(ts.hour, ts.minute)

def _clock_le(ts, hhmm: str) -> bool:
    h, m = map(int, hhmm.split(":"))
    return _to_clock(ts) <= dtime(h, m)

def wick_opposite_fraction(row: pd.Series, direction: str) -> float:
    h, l, o, c = float(row["High"]), float(row["Low"]), float(row.get("Open", np.nan)), float(row["Close"])
    rng = max(h - l, 1e-12)
    if direction == "long":
        opp = min(o, c) - l
    else:
        opp = h - max(o, c)
    return float(max(opp, 0.0) / rng)

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def _normalize_splits(num_targets: int, splits: tuple) -> List[float]:
    if num_targets <= 0:
        return [1.0]
    if not splits:
        return [1.0 / num_targets] * num_targets
    arr = list(splits[:num_targets])
    if len(arr) < num_targets:
        arr += [0.0] * (num_targets - len(arr))
    s = sum(arr)
    if s <= 0:
        return [1.0 / num_targets] * num_targets
    return [x / s for x in arr]

def _to_dt_index(obj) -> pd.DatetimeIndex:
    """
    Coerce any array/Series/Index of datelike objects to a tz-naive
    normalized (date) DatetimeIndex, sorted and unique. Use only for ranges/boundaries.
    """
    if isinstance(obj, (pd.Series, pd.Index, np.ndarray, list, tuple)):
        di = pd.to_datetime(obj, errors="coerce")
    else:
        di = pd.to_datetime([obj], errors="coerce")
    di = pd.DatetimeIndex(di)
    di = di.dropna().tz_localize(None).normalize()
    di = di.sort_values().unique()
    return di

def _fmt(ts) -> str:
    return pd.Timestamp(ts).strftime("%Y-%m-%d")

# Overall Totals (adds Profit Factor & Avg Win/Loss)
def print_overall_totals(trades: pd.DataFrame):
    print("\n=== Overall Totals ===")
    if trades is None or trades.empty:
        print("Total trades: 0")
        print("Total PnL ($): 0.00")
        print("Win rate: 0.00%")
        print("Avg R: 0.000 | Median R: 0.000")
        print("Profit Factor: 0.000")
        print("Avg Win ($): 0.00 | Avg Loss ($): 0.00")
        return

    pnl = pd.to_numeric(trades["PnL_$"], errors="coerce").fillna(0.0)
    wins_mask = pnl > 0
    losses_mask = pnl < 0

    total_trades = int(len(pnl))
    total_pnl    = float(pnl.sum())
    win_rate     = float(wins_mask.mean() * 100.0) if total_trades else 0.0
    avg_r        = float(pd.to_numeric(trades["R_multiple"], errors="coerce").mean())
    median_r     = float(pd.to_numeric(trades["R_multiple"], errors="coerce").median())

    gross_profit = float(pnl[wins_mask].sum())
    gross_loss   = float(-pnl[losses_mask].sum())
    pf = (gross_profit / gross_loss) if gross_loss > 0 else (float("inf") if gross_profit > 0 else 0.0)

    avg_win  = float(pnl[wins_mask].mean()) if wins_mask.any() else 0.0
    avg_loss = float(pnl[losses_mask].mean()) if losses_mask.any() else 0.0

    print(f"Total trades: {total_trades:,}")
    print(f"Total PnL ($): {total_pnl:,.2f}")
    print(f"Win rate: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Profit Factor: {pf:.3f}")
    print(f"Avg Win ($): {avg_win:,.2f} | Avg Loss ($): {avg_loss:,.2f}")

# =============================
# Indicators & features
# =============================

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    for c in ("ORH","ORL"):
        if (df.columns == c).sum() > 0:
            df = df.drop(columns=[c])
    for col in ("High","Low"):
        if col not in df.columns:
            raise ValueError(f"compute_opening_range: missing column {col}")
    if not df.index.is_monotonic_increasing:
        df = df.sort_index()
    df["Session"] = sessionize(df)
    first_bar_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float).values
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float).values
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return ensure_unique_columns(df)

def add_session_vwap(df_15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df["Session"] = sessionize(df)
    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0
    if "Volume" in df.columns:
        df["vwap_num"] = tp * df["Volume"]
        df["vwap_den"] = df["Volume"].replace(0, np.nan)
    else:
        df["vwap_num"] = tp
        df["vwap_den"] = 1.0
    df["VWAP"] = (df.groupby("Session")["vwap_num"].cumsum() /
                  df.groupby("Session")["vwap_den"].cumsum())
    return ensure_unique_columns(df.drop(columns=["vwap_num","vwap_den"]))

def add_ema_trend(df_15: pd.DataFrame, fast=EMA_FAST, slow=EMA_SLOW) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return ensure_unique_columns(df)

def rsi(series: pd.Series, length=14) -> pd.Series:
    delta = series.diff()
    up = delta.clip(lower=0)
    dn = -delta.clip(upper=0)
    avg_gain = up.ewm(alpha=1/length, adjust=False).mean()
    avg_loss = dn.ewm(alpha=1/length, adjust=False).mean()
    rs = avg_gain / (avg_loss.replace(0, np.nan))
    return 100 - (100 / (1 + rs))

def add_rsi(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df[f"RSI{length}"] = rsi(df["Close"], length)
    return ensure_unique_columns(df)

def add_adx(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    up_move = df["High"].diff()
    dn_move = -df["Low"].diff()
    plus_dm  = np.where((up_move > dn_move) & (up_move > 0), up_move, 0.0)
    minus_dm = np.where((dn_move > up_move) & (dn_move > 0), dn_move, 0.0)
    prev_close = df["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (df["High"] - df["Low"]).values,
        np.abs(df["High"] - prev_close).values,
        np.abs(df["Low"]  - prev_close).values
    ]), index=df.index)
    atr = tr.ewm(alpha=1/length, adjust=False).mean()
    plus_di = 100 * pd.Series(plus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    minus_di = 100 * pd.Series(minus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    dx = (abs(plus_di - minus_di) / (plus_di + minus_di).replace(0, np.nan)) * 100
    adx = dx.ewm(alpha=1/length, adjust=False).mean()
    df[f"ADX{length}"] = adx
    return ensure_unique_columns(df)

def build_daily_from_15m(df_15: pd.DataFrame) -> pd.DataFrame:
    d = ensure_unique_columns(df_15.copy())
    d["Session"] = sessionize(d)
    daily = d.groupby("Session").agg(
        Open=("Open","first"),
        High=("High","max"),
        Low=("Low","min"),
        Close=("Close","last"),
        Volume=("Volume","sum") if "Volume" in d.columns else ("Close","size"),
    )
    return daily

def add_daily_bias_and_atr(df_15: pd.DataFrame, ema_fast=20, ema_slow=50, atr_len=14) -> pd.DataFrame:
    dly = build_daily_from_15m(df_15)
    dly[f"EMA_D{ema_fast}"] = dly["Close"].ewm(span=ema_fast, adjust=False).mean()
    dly[f"EMA_D{ema_slow}"] = dly["Close"].ewm(span=ema_slow, adjust=False).mean()
    prev_close = dly["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (dly["High"] - dly["Low"]).values,
        np.abs(dly["High"] - prev_close).values,
        np.abs(dly["Low"]  - prev_close).values
    ]), index=dly.index)
    atr_col = f"ATR_D{atr_len}"
    dly[atr_col] = tr.ewm(alpha=1/atr_len, adjust=False).mean()

    if "Volume" in df_15.columns:
        tmp = df_15.copy(); tmp["Session"] = sessionize(tmp)
        open15_vol = tmp.groupby("Session")["Volume"].first()
        dly["Open15_Vol"] = open15_vol
        dly["Open15_Vol_Thresh"] = (
            dly["Open15_Vol"].rolling(window=60, min_periods=10).quantile(0.30)
        )

    feature_cols = [f"EMA_D{ema_fast}", f"EMA_D{ema_slow}", atr_col, "Open15_Vol", "Open15_Vol_Thresh"]
    feature_cols = [c for c in feature_cols if c in dly.columns]

    out = ensure_unique_columns(df_15.copy())
    out["Session"] = sessionize(out)
    out = out.merge(dly[feature_cols], left_on="Session", right_index=True, how="left")
    return ensure_unique_columns(out)

# =============================
# Backtest (per-ticker)
# =============================

def backtest_orb_retest(
    df_15: pd.DataFrame,
    ema_fast:int = EMA_FAST,
    ema_slow:int = EMA_SLOW,
    breakout_cushion: float = BREAKOUT_CUSHION_PCT,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = TARGETS_R,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE
) -> Tuple[pd.DataFrame, pd.DataFrame]:

    # Opening range + features
    df = compute_opening_range(df_15)
    if USE_VWAP_FILTER:  df = add_session_vwap(df)
    if USE_TREND_FILTER: df = add_ema_trend(df, ema_fast, ema_slow)
    if USE_RSI_FILTER:   df = add_rsi(df, RSI_LEN)
    if USE_ADX_FILTER:   df = add_adx(df, ADX_LEN)
    df = add_daily_bias_and_atr(df, DAILY_EMA_FAST, DAILY_EMA_SLOW, DAILY_ATR_LEN)

    # Gap (first 15m vs previous day's close)
    df["Session"] = sessionize(df)
    first_idx = df.groupby("Session").head(1).index
    prev_close_series = df.groupby("Session")["Close"].last().shift(1)
    prev_close_map = df["Session"].map(prev_close_series)
    df.loc[first_idx, "GapPct"] = (df.loc[first_idx, "Open"] / prev_close_map.loc[first_idx] - 1.0)
    df["GapPct"] = df.groupby("Session")["GapPct"].ffill()

    sessions = df["Session"].unique()
    trades = []

    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3: continue

        orh, orl = float(sdf["ORH"].iloc[0]), float(sdf["ORL"].iloc[0])
        if np.isnan(orh) or np.isnan(orl): continue

        after_open = sdf.iloc[1:].copy()
        long_breaks  = after_open[after_open["Close"] > orh * (1 + breakout_cushion)]
        short_breaks = after_open[after_open["Close"] < orl * (1 - breakout_cushion)]

        cands = []
        for idx, row in long_breaks.iterrows():  cands.append(("long", idx, row))
        for idx, row in short_breaks.iterrows(): cands.append(("short", idx, row))
        cands.sort(key=lambda x: x[1])

        trades_this_session = 0
        took_long = False
        took_short = False

        for direction, btime, breakout_row in cands:
            if not _clock_le(breakout_row.name, BREAKOUT_DEADLINE): continue
            if not ALLOW_BOTH_SIDES:
                if took_long and direction == "short":  break
                if took_short and direction == "long":  break
            if direction == "long" and took_long:   continue
            if direction == "short" and took_short: continue

            # Filters
            if USE_DOW_FILTER and btime.weekday() not in ALLOWED_DOW: continue

            if USE_GAP_FILTER and "GapPct" in sdf.columns:
                gap = sdf.loc[sdf.index.min(), "GapPct"]
                if pd.notna(gap) and not (GAP_MIN <= abs(float(gap)) <= GAP_MAX):
                    continue
                if ALIGN_WITH_GAP_DIR and pd.notna(gap):
                    g = float(gap)
                    if (direction == "long" and g < 0) or (direction == "short" and g > 0):
                        continue

            if USE_VWAP_FILTER and "VWAP" in sdf.columns:
                rv = sdf.loc[:btime, "VWAP"].tail(3).dropna().values
                rising  = (len(rv) < 2) or np.all(np.diff(rv) >= 0)
                falling = (len(rv) < 2) or np.all(np.diff(rv) <= 0)
                if direction == "long":
                    if not (breakout_row["Close"] > breakout_row["VWAP"] and rising): continue
                else:
                    if not (breakout_row["Close"] < breakout_row["VWAP"] and falling): continue

            if USE_TREND_FILTER and (f"EMA{ema_fast}" in sdf.columns) and (f"EMA{ema_slow}" in sdf.columns):
                emaf = breakout_row.get(f"EMA{ema_fast}", np.nan); emas = breakout_row.get(f"EMA{ema_slow}", np.nan)
                if np.isfinite(emaf) and np.isfinite(emas):
                    if direction == "long" and not (emaf > emas): continue
                    if direction == "short" and not (emaf < emas): continue

            if USE_DAILY_TREND_FILTER and f"EMA_D{DAILY_EMA_FAST}" in sdf.columns and f"EMA_D{DAILY_EMA_SLOW}" in sdf.columns:
                dfast = float(sdf[f"EMA_D{DAILY_EMA_FAST}"].iloc[0])
                dslow = float(sdf[f"EMA_D{DAILY_EMA_SLOW}"].iloc[0])
                if direction == "long" and not (dfast > dslow): continue
                if direction == "short" and not (dfast < dslow): continue

            atr_col = f"ATR_D{DAILY_ATR_LEN}"
            if USE_OR_WIDTH_ATR_FILTER and atr_col in sdf.columns:
                or_width = float(orh - orl)
                atr_d = float(sdf[atr_col].iloc[0])
                if atr_d <= 0: continue
                frac = or_width / atr_d
                if not (OR_ATR_MIN <= frac <= OR_ATR_MAX): continue

            if USE_RSI_FILTER and f"RSI{RSI_LEN}" in sdf.columns:
                rsi_val = float(breakout_row[f"RSI{RSI_LEN}"])
                if direction == "long" and not (rsi_val >= RSI_THRESH_LONG): continue
                if direction == "short" and not (rsi_val <= 100 - RSI_THRESH_SHORT): continue

            if USE_ADX_FILTER and f"ADX{ADX_LEN}" in sdf.columns:
                adx_val = float(breakout_row[f"ADX{ADX_LEN}"])
                if not (adx_val >= ADX_MIN): continue

            # Retest window and level
            cutoff = btime + timedelta(minutes=max_retest_min)
            window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
            level = orh if direction == "long" else orl
            touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)

            retest_used = True
            if touch.empty:
                if USE_CONTINUATION and _clock_le(breakout_row.name, CONTINUATION_CUTOFF):
                    retest_used = False
                    retest_time = breakout_row.name
                    retest_bar = breakout_row
                else:
                    continue
            else:
                retest_bar = touch.iloc[0]
                retest_time = retest_bar.name

            # Confirm close through level if requested (with cushion)
            if retest_used and retest_confirm_close:
                if direction == "long":
                    ok = (retest_bar["Close"] >= level * (1 + breakout_cushion))
                else:
                    ok = (retest_bar["Close"] <= level * (1 - breakout_cushion))
                if not ok:
                    continue

            # Reject with big opposite wick
            if MAX_OPPOSITE_WICK_FRAC is not None:
                if wick_opposite_fraction(retest_bar, direction) > MAX_OPPOSITE_WICK_FRAC:
                    continue

            # ===== Entry price (realistic) =====
            if retest_used and retest_confirm_close:
                if ENTRY_ON_CONFIRM.lower() == "next_open":
                    next_bar = sdf[sdf.index > retest_time].head(1)
                    if next_bar.empty:
                        continue
                    entry_time = next_bar.index[0]
                    raw_entry = float(next_bar["Open"].iloc[0])
                elif ENTRY_ON_CONFIRM.lower() == "confirm_close":
                    entry_time = retest_time
                    raw_entry = float(retest_bar["Close"])
                else:
                    entry_time = retest_time
                    raw_entry = float(retest_bar["Close"])
            else:
                entry_time = retest_time
                raw_entry = float(level)

            entry = _apply_slippage(raw_entry, slippage_bps, "buy" if direction == "long" else "sell")
            stop  = orl if direction == "long" else orh
            rps   = abs(entry - stop)
            if rps <= 1e-12: continue

            qty = POSITION_SIZE_DOLLARS / max(entry, 1e-12)
            side_mult = 1 if direction == "long" else -1

            targets = [(entry + r * rps) if direction == "long" else (entry - r * rps) for r in TARGETS_R]

            # Walk forward bar-by-bar
            run = sdf[sdf.index >= entry_time].copy()
            qty_left = qty
            exits = []
            next_tp_idx = 0

            if USE_SCALE_OUT and len(targets) >= 1:
                splits = _normalize_splits(len(targets), SCALE_SPLIT)
                leg_sizes = [qty * s for s in splits]
            else:
                leg_sizes = [qty]

            be_level = entry
            be_pending = False
            be_set_time = None

            for ts, row in run.iterrows():
                hi, lo = float(row["High"]), float(row["Low"])

                # move to BE (armed after TP1), optionally on next bar
                if be_pending and (not BE_ON_NEXT_BAR or (be_set_time is not None and ts > be_set_time)):
                    stop = be_level
                    be_pending = False

                # Stop first
                if lo <= stop <= hi:
                    px = _apply_slippage(stop, SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                    label = "breakeven" if abs(px - be_level) < 1e-10 else "stop"
                    exits.append((label, ts, px, qty_left))
                    qty_left = 0.0
                    break

                # Targets (sequential)
                if next_tp_idx < len(targets):
                    tp = targets[next_tp_idx]
                    if lo <= tp <= hi:
                        px = _apply_slippage(tp, SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                        fill_qty = leg_sizes[next_tp_idx] if next_tp_idx < len(leg_sizes) else qty_left
                        exits.append((f"tp{next_tp_idx+1}", ts, px, fill_qty))
                        qty_left -= fill_qty
                        if next_tp_idx == 0 and MOVE_STOP_TO_BE_AT_R is not None:
                            be_pending = True
                            be_set_time = ts
                        next_tp_idx += 1
                        if qty_left <= 1e-9:
                            break

                # If no TP1 by continuation cutoff, tighten to BE
                if USE_CONTINUATION and not _clock_le(ts, CONTINUATION_CUTOFF) and next_tp_idx == 0:
                    stop = be_level

            # EOD exit for any leftover
            if qty_left > 1e-9:
                last = run.iloc[-1]
                px = _apply_slippage(float(last["Close"]), SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                exits.append(("eod", run.index[-1], px, qty_left))
                qty_left = 0.0

            cash_pnl = sum((px - entry) * side_mult * q for (_, _, px, q) in exits)
            if FEE_PER_FILL:
                cash_pnl -= FEES_PER_TRADE * len(exits)
            else:
                cash_pnl -= FEES_PER_TRADE

            trades.append({
                "Ticker": sdf["Ticker"].iloc[0] if "Ticker" in sdf.columns else "",
                "Session": ses,
                "Direction": direction,
                "ORH": orh, "ORL": orl,
                "EntryTime": entry_time,
                "Entry": entry, "Stop": stop,
                "Targets": targets,
                "Exits": [(lab, ts, px, q) for (lab, ts, px, q) in exits],
                "Qty": qty,
                "PnL_$": cash_pnl,
                "R_multiple": cash_pnl / (rps * max(qty, 1e-12))
            })

            if direction == "long":  took_long  = True
            if direction == "short": took_short = True
            trades_this_session += 1
            if trades_this_session >= MAX_TRADES_PER_SESSION:
                break

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Parquet loading
# =============================

STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                rename[lower[k]] = std
        elif k.capitalize() in df.columns: rename[k.capitalize()] = std
        elif k.upper() in df.columns: rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _maybe_resample_to_15m(df: pd.DataFrame) -> pd.DataFrame:
    if len(df.index) < 3:
        return df
    deltas = (df.index[1:] - df.index[:-1]).asi8
    if len(deltas) == 0:
        return df
    med_ns = np.median(deltas)
    fifteen_ns = pd.Timedelta("15T").value
    if med_ns < fifteen_ns:
        o = df["Open"].resample("15T", label="left", closed="left").first()
        h = df["High"].resample("15T", label="left", closed="left").max()
        l = df["Low"].resample("15T", label="left", closed="left").min()
        c = df["Close"].resample("15T", label="left", closed="left").last()
        v = df["Volume"].resample("15T", label="left", closed="left").sum() if "Volume" in df.columns else None
        parts = {"Open":o,"High":h,"Low":l,"Close":c}
        if v is not None: parts["Volume"] = v
        out = pd.concat(parts, axis=1).dropna(subset=["Open","High","Low","Close"])
        return out
    return df

def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()

    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df = df.tz_localize("UTC").tz_convert(tz)
        else:
            df = df.tz_localize(tz)
    else:
        df = df.tz_convert(tz)

    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")

    df = _maybe_resample_to_15m(df)
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)
    return ensure_unique_columns(df.sort_index())

# =============================
# Processing helpers
# =============================

def iter_parquet_files(folder: str):
    files = sorted(glob.glob(os.path.join(folder, "*.parquet")))
    return files

def run_folder(folder: str,
               ema_fast:int = EMA_FAST,
               ema_slow:int = EMA_SLOW,
               breakout_cushion: float = BREAKOUT_CUSHION_PCT) -> Tuple[pd.DataFrame, pd.DataFrame]:
    files = iter_parquet_files(folder)
    print(f"Found {len(files)} parquet files.")
    all_trades, all_equity = [], []

    iterator = files
    if tqdm is not None:
        iterator = tqdm(files, desc="Backtesting files", unit="file")

    for i, fpath in enumerate(iterator, 1):
        ticker = os.path.splitext(os.path.basename(fpath))[0].upper().replace("_","").replace("-","")
        try:
            df15 = load_parquet_15m(fpath)
            df15["Ticker"] = ticker
        except Exception as e:
            print(f"\nSkipping {ticker}: {e}")
            continue

        tlog, eq = backtest_orb_retest(
            df15,
            ema_fast=ema_fast,
            ema_slow=ema_slow,
            breakout_cushion=breakout_cushion,
            max_retest_min=MAX_RETEST_MIN,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=SLIPPAGE_BPS,
            fees=FEES_PER_TRADE,
            retest_confirm_close=RETEST_CONFIRM_CLOSE,
        )

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"]   = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{t[1]}|{t[2]:.6f}|{t[3]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy(); eq_out["Ticker"] = ticker
            all_equity.append(eq_out)

        if AUTOSAVE_EVERY and i % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

        if SLEEP_BETWEEN_TICKERS > 0:
            time.sleep(SLEEP_BETWEEN_TICKERS)

    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()
    return trades, equity

# =============================
# Validation: metrics and CV/WF
# =============================

def metrics(df: pd.DataFrame) -> Dict[str, float]:
    if df is None or df.empty:
        return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, SumR=0.0)
    r = pd.to_numeric(df.get("R_multiple", pd.Series(dtype=float)), errors="coerce").dropna()
    if r.empty:
        return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, SumR=0.0)
    wins = r[r > 0]; losses = r[r < 0]
    gp = wins.sum(); gl = -losses.sum()
    pf = (gp / gl) if gl > 1e-12 else (np.inf if gp > 0 else 0.0)
    return dict(
        Trades=int(r.size),
        WinRate=float((r > 0).mean() * 100.0),
        PF=float(pf),
        AvgR=float(r.mean()),
        SumR=float(r.sum())
    )

def _print_metrics(title: str, m: Dict[str, float]):
    print(f"{title} | Trades: {m['Trades']:,} | Win%: {m['WinRate']:.2f}% | PF: {m['PF']:.3f} | Avg R: {m['AvgR']:.3f} | SumR: {m['SumR']:,.2f}")

def purged_kfold_cv(trades: pd.DataFrame, n_splits=5, purge_days=2, embargo_days=2):
    print("\n--- Purged K-Fold CV ---")
    # Boundaries from unique sessions
    ses_unique = _to_dt_index(trades["Session"])
    if len(ses_unique) == 0:
        print("No sessions to split.")
        return
    smin, smax = ses_unique.min(), ses_unique.max()
    all_days = pd.date_range(smin, smax, freq="D")
    if n_splits <= 0 or len(all_days) < n_splits:
        print("Not enough days to split.")
        return

    # Per-row session series (aligned to trades)
    ses_row = session_series(trades)  # length == len(trades)

    fold_sizes = np.linspace(0, len(all_days), n_splits + 1, dtype=int)
    for i in range(n_splits):
        test_idx = all_days[fold_sizes[i]:fold_sizes[i+1]]
        if len(test_idx) == 0:
            continue
        test_start, test_end = test_idx.min(), test_idx.max()

        purge_start = max(all_days.min(), test_start - pd.Timedelta(days=purge_days))
        embargo_end = min(all_days.max(), test_end + pd.Timedelta(days=embargo_days))

        # Masks aligned to trades length
        train_mask = (ses_row < purge_start) | (ses_row > embargo_end)
        test_mask  = (ses_row >= test_start) & (ses_row <= test_end)

        trn = trades.loc[train_mask]
        tst = trades.loc[test_mask]

        _print_metrics(f"Fold {i+1:02d} Train [{_fmt(purge_start)} purge→]", metrics(trn))
        _print_metrics(f"Fold {i+1:02d} Test  [{_fmt(test_start)}→{_fmt(test_end)}]", metrics(tst))

def walk_forward_validation(trades: pd.DataFrame, train_days: int, test_days: int):
    print("\n--- Walk-Forward Validation ---")
    ses_unique = _to_dt_index(trades["Session"])
    if len(ses_unique) == 0:
        print("No sessions to split.")
        return
    t0, t1 = ses_unique.min(), ses_unique.max()

    # Per-row session series (aligned to trades)
    ses_row = session_series(trades)

    fold = 0
    train_start = t0
    while True:
        train_end = train_start + pd.Timedelta(days=train_days - 1)
        test_start = train_end + pd.Timedelta(days=1)
        test_end   = test_start + pd.Timedelta(days=test_days - 1)
        if test_end > t1:
            break
        fold += 1

        train_mask = (ses_row >= train_start) & (ses_row <= train_end)
        test_mask  = (ses_row >= test_start) & (ses_row <= test_end)

        trn = trades.loc[train_mask]
        tst = trades.loc[test_mask]

        _print_metrics(f"Fold {fold:02d} Train [{_fmt(train_start)}→{_fmt(train_end)}]", metrics(trn))
        _print_metrics(f"Fold {fold:02d} Test  [{_fmt(test_start)}→{_fmt(test_end)} ]", metrics(tst))

        train_start = test_end + pd.Timedelta(days=1)

# ---- Nested WFO (Train→Calibrate→Test) ----

def nested_walk_forward_with_calibration(
    grid_returns: dict,
    train_days: int,
    calib_days: int,
    test_days: int,
    purge_days: int = 2,
    embargo_days: int = 2,
    min_trades_calib: int = 100
):
    """
    grid_returns: dict[param_key] -> DataFrame with 'Session' (or index) + 'R_multiple'.
    Returns:
      wf_records: list of dicts (per fold) with train/calib/test metrics and chosen key
      last_wf_end: Timestamp end of the final test window (for downstream holdout)
    """
    # collect all sessions
    all_sessions = []
    for key, df in grid_returns.items():
        if df is None or len(df) == 0:
            continue
        if "Session" in df.columns:
            all_sessions.append(df["Session"])
        else:
            all_sessions.append(df.index)

    if not all_sessions:
        print("No sessions found in grid_returns.")
        return [], None

    all_sess = _to_dt_index(pd.concat([pd.to_datetime(s, errors="coerce") for s in all_sessions]))
    if len(all_sess) == 0:
        print("All sessions are NaT after coercion.")
        return [], None

    t0, t1 = all_sess.min(), all_sess.max()
    print("\n--- Nested Walk-Forward (Train→Calib→Test) ---")
    print(f"Global session range: {_fmt(t0)} → {_fmt(t1)}")

    wf_records = []
    last_wf_end = None

    def slice_sessions(df: pd.DataFrame, s0, s1):
        if df is None or df.empty:
            return df
        # per-row normalized series
        if "Session" in df.columns:
            ses = session_series(df)
        else:
            ses = pd.to_datetime(df.index, errors="coerce")
            ses = pd.Series(pd.DatetimeIndex(ses).tz_localize(None).normalize(), index=df.index)
        m = (ses >= s0) & (ses <= s1)
        return df.loc[m]

    def metrics_for_key(df_slice: pd.DataFrame) -> dict:
        if df_slice is None or df_slice.empty:
            return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, SumR=0.0)
        r = pd.to_numeric(df_slice.get("R_multiple", pd.Series(dtype=float)), errors="coerce").dropna()
        if r.empty:
            return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, SumR=0.0)
        wins = r[r > 0]; losses = r[r < 0]
        gp = wins.sum(); gl = -losses.sum()
        pf = (gp / gl) if gl > 1e-12 else (np.inf if gp > 0 else 0.0)
        return dict(
            Trades=int(r.size),
            WinRate=float((r > 0).mean() * 100.0),
            PF=float(pf),
            AvgR=float(r.mean()),
            SumR=float(r.sum())
        )

    fold = 0
    train_start = t0
    while True:
        train_end   = train_start + pd.Timedelta(days=train_days - 1)
        purge_end   = train_end + pd.Timedelta(days=purge_days)
        calib_end   = purge_end + pd.Timedelta(days=calib_days)
        embargo_end = calib_end + pd.Timedelta(days=embargo_days)
        test_end    = embargo_end + pd.Timedelta(days=test_days)

        if test_end > t1:
            break

        fold += 1
        print(f"\nFold {fold:02d}")
        print(f" Train:  {_fmt(train_start)} → {_fmt(train_end)}")
        print(f" Purge:  {_fmt(train_end + pd.Timedelta(days=1))} → {_fmt(purge_end)}")
        print(f" Calib:  {_fmt(purge_end + pd.Timedelta(days=1))} → {_fmt(calib_end)}")
        print(f" Embarg: {_fmt(calib_end + pd.Timedelta(days=1))} → {_fmt(embargo_end)}")
        print(f" Test:   {_fmt(embargo_end + pd.Timedelta(days=1))} → {_fmt(test_end)}")

        perkey_train, perkey_calib, perkey_test = {}, {}, {}
        for key, df in grid_returns.items():
            trn = slice_sessions(df, train_start, train_end)
            cal = slice_sessions(df, purge_end + pd.Timedelta(days=1), calib_end)
            tst = slice_sessions(df, embargo_end + pd.Timedelta(days=1), test_end)

            perkey_train[key] = metrics_for_key(trn)
            perkey_calib[key] = metrics_for_key(cal)
            perkey_test[key]  = metrics_for_key(tst)

        # Eligibility gate
        eligible = {k: v for k, v in perkey_calib.items() if v["Trades"] >= int(min_trades_calib)}
        if not eligible:
            eligible = {k: v for k, v in perkey_calib.items() if v["Trades"] > 0}
        if not eligible:
            print("  [warn] No eligible keys in calibration; skipping fold.")
            train_start = test_end + pd.Timedelta(days=1)
            continue

        # Rank by AvgR, PF, WinRate, Trades
        calib_sorted = sorted(
            eligible.items(),
            key=lambda kv: (kv[1]["AvgR"], kv[1]["PF"], kv[1]["WinRate"], kv[1]["Trades"]),
            reverse=True
        )
        chosen_key, chosen_calib = calib_sorted[0]
        chosen_test  = perkey_test[chosen_key]
        chosen_train = perkey_train[chosen_key]

        print(f"  Chosen key: {chosen_key}")
        _print_metrics("  Train ", chosen_train)
        _print_metrics("  Calib ", chosen_calib)
        _print_metrics("  Test  ", chosen_test)

        wf_records.append(dict(
            fold=fold,
            train_start=train_start, train_end=train_end,
            purge_end=purge_end, calib_end=calib_end,
            embargo_end=embargo_end, test_end=test_end,
            chosen_key=chosen_key,
            train_metrics=chosen_train,
            calib_metrics=chosen_calib,
            test_metrics=chosen_test
        ))
        last_wf_end = test_end

        # advance
        train_start = test_end + pd.Timedelta(days=1)

    if not wf_records:
        print("No folds produced.")
        return [], None

    test_wr = np.mean([rec["test_metrics"]["WinRate"] for rec in wf_records if rec["test_metrics"]["Trades"] > 0])
    test_pf = np.mean([rec["test_metrics"]["PF"] for rec in wf_records if rec["test_metrics"]["Trades"] > 0])
    print(f"\nWFO OOS (avg over folds) | Win%: {test_wr:.2f} | PF: {test_pf:.3f}")
    return wf_records, last_wf_end

def choose_final_param_from_wf(wf_records: List[dict]) -> Optional[str]:
    if not wf_records:
        return None
    # Majority vote by chosen_key across folds, tie-break by sum of test SumR
    keys = [rec["chosen_key"] for rec in wf_records]
    uniq = list(dict.fromkeys(keys))
    scores = {}
    for k in uniq:
        test_sumr = sum(rec["test_metrics"]["SumR"] for rec in wf_records if rec["chosen_key"] == k)
        votes = sum(1 for rec in wf_records if rec["chosen_key"] == k)
        scores[k] = (votes, test_sumr)
    best = sorted(scores.items(), key=lambda kv: (kv[1][0], kv[1][1]), reverse=True)[0][0]
    return best

# =============================
# Grid runner for nested WF
# =============================

def backtest_folder_for_params(folder: str, ema_fast:int, ema_slow:int, breakout_cushion: float) -> pd.DataFrame:
    trades, _ = run_folder(folder, ema_fast=ema_fast, ema_slow=ema_slow, breakout_cushion=breakout_cushion)
    return trades

def build_param_grid() -> List[tuple]:
    grid = list(itertools.product(GRID_EMA_FAST, GRID_EMA_SLOW, GRID_BREAKOUT_CUSHION))
    grid = [(f,s,c) for (f,s,c) in grid if f < s]
    return grid

def build_grid_returns(folder: str) -> Dict[str, pd.DataFrame]:
    grid = build_param_grid()
    print(f"Param grid size: {len(grid)}")
    iterator = grid
    if tqdm is not None:
        iterator = tqdm(grid, desc="Param grid", unit="combo")

    out = {}
    for ema_f, ema_s, cush in iterator:
        key = f"ema{ema_f}-{ema_s}_c{cush:.4f}"
        tr = backtest_folder_for_params(folder, ema_f, ema_s, cush)
        out[key] = tr[["Session","R_multiple"]].copy() if not tr.empty else pd.DataFrame(columns=["Session","R_multiple"])
    return out

# =============================
# Main
# =============================

def main():
    # ----- Baseline backtest -----
    if RUN_BASELINE:
        trades, equity = run_folder(PARQUET_DIR)
        print_overall_totals(trades)

        if not trades.empty:
            trades.sort_values(["Ticker","Session"], inplace=True)
            trades.to_csv(TRADES_CSV_ALL, index=False)
            print(f"Saved trades: {TRADES_CSV_ALL}")

        if not equity.empty:
            equity.sort_values(["Ticker","Session"], inplace=True)
            equity.to_csv(EQUITY_CSV_ALL, index=False)
            print(f"Saved equity: {EQUITY_CSV_ALL}")

        if not trades.empty:
            summary = (
                trades.groupby("Ticker")["PnL_$"]
                    .agg(['count', 'sum', 'mean', 'std'])
                    .rename(columns={'count':'Trades','sum':'PnL_sum','mean':'AvgPnL','std':'StdPnL'})
                    .reset_index()
                    .sort_values("PnL_sum", ascending=False)
            )
            wr = trades.groupby("Ticker")["PnL_$"].apply(lambda s: (pd.to_numeric(s, errors="coerce") > 0).mean()*100.0).rename("WinRate_%")
            pos = trades.groupby("Ticker")["PnL_$"].apply(lambda s: pd.to_numeric(s, errors="coerce").clip(lower=0).sum()).rename("GrossProfit")
            neg = trades.groupby("Ticker")["PnL_$"].apply(lambda s: -pd.to_numeric(s, errors="coerce").clip(upper=0).sum()).rename("GrossLoss")
            pf = (pos / neg.replace(0, np.nan)).rename("PF")
            summary = summary.merge(wr, on="Ticker", how="left").merge(pf, on="Ticker", how="left")
            print("\n=== Summary by Ticker (top 50) ===")
            print(summary.head(50).to_string(index=False))
            summary.to_csv(SUMMARY_CSV, index=False)
        else:
            print("No trades generated with current settings.")

    # ----- Purged K-Fold CV (fixed mask alignment) -----
    if RUN_PURGED_KFOLD and 'trades' in locals() and not trades.empty:
        purged_kfold_cv(trades, n_splits=KFOLD_SPLITS, purge_days=PURGE_DAYS, embargo_days=EMBARGO_DAYS)

    # ----- Walk-Forward Validation (fixed mask alignment) -----
    if RUN_WALK_FORWARD and 'trades' in locals() and not trades.empty:
        walk_forward_validation(trades, train_days=WF_TRAIN_DAYS, test_days=WF_TEST_DAYS)

    # ----- Nested Walk-Forward (Train→Calibrate→Test) -----
    if RUN_NESTED_WF:
        grid_returns = build_grid_returns(PARQUET_DIR)
        if grid_returns:
            wf_records, last_wf_end = nested_walk_forward_with_calibration(
                grid_returns,
                train_days=NWF_TRAIN_DAYS,
                calib_days=NWF_CALIB_DAYS,
                test_days=NWF_TEST_DAYS,
                purge_days=NWF_PURGE_DAYS,
                embargo_days=NWF_EMBARGO_DAYS,
                min_trades_calib=NWF_MIN_TRADES_CALIB
            )
            final_key = choose_final_param_from_wf(wf_records)
            print(f"\nChosen FINAL parameter key from WFO: {final_key}")
            # Optional final holdout: everything strictly AFTER last_wf_end using final_key
            if final_key is not None and final_key in grid_returns and last_wf_end is not None:
                df_all = grid_returns[final_key]
                ses_idx = session_series(df_all)
                holdout = df_all.loc[ses_idx > last_wf_end]
                print("\n--- Final Holdout (strictly after last WFO test window) ---")
                _print_metrics("Holdout", metrics(holdout))

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_parquet_industry_WF_PLUS.py
# ORB 15m Retest/Continuation strategy (multi-parquet).
# Includes: realistic entries, scale-out, filters; PF/Sharpe/Sortino/MaxDD/CVaR;
# purged K-Fold (session-based), Walk-Forward, Nested WFO; parameter heatmaps;
# slippage/fee surface; permutation (null) test; block-bootstrapped CIs;
# portfolio concurrency constraints; cleaner trade records.
#
# NOTE: This is a drop-in upgrade from your prior script.
# - Adds Stop_init/Stop_final, EntryType, ConfirmMode
# - Enforces optional RETEST_DEADLINE
# - Uses session-based K-Fold split (not dense calendar)
# - Adds a "fill_policy" for same-bar TP/stop sequencing (stop_first/tp_first/random)

import warnings
warnings.filterwarnings("ignore")

import os, glob, time, itertools, json, math, random
from datetime import timedelta, time as dtime
from typing import List, Tuple, Optional, Dict, Any

import numpy as np
import pandas as pd

try:
    from tqdm import tqdm
except ImportError:
    tqdm = None  # fallback if tqdm isn't installed

# =============================
# USER PARAMETERS
# =============================

PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
ASSUME_NAIVE_TIMESTAMPS_ARE_UTC = True

TZ = "America/New_York"
REG_SESSION_START = "09:30"
REG_SESSION_END   = "16:00"

# ORB logic
MAX_RETEST_MIN = 120
RETEST_CONFIRM_CLOSE = True
BREAKOUT_CUSHION_PCT = 0.0005
MAX_OPPOSITE_WICK_FRAC = 0.40
RETEST_DEADLINE = "12:00"  # NEW: optional cutoff for retest confirmation

MAX_TRADES_PER_SESSION = 2
ALLOW_BOTH_SIDES = True

# Scale-out
USE_SCALE_OUT = True
TARGETS_R = [1.0, 2.0, 2.5]
SCALE_SPLIT = (1/3, 1/3, 1/3)

# BE policy
MOVE_STOP_TO_BE_AT_R = 1.0
BE_ON_NEXT_BAR = True

# Continuation (fallback)
USE_CONTINUATION = True

# Time windows
BREAKOUT_DEADLINE   = "11:00"
CONTINUATION_CUTOFF = "12:00"

# Intraday indicators
USE_VWAP_FILTER   = True
USE_TREND_FILTER  = True
EMA_FAST = 20
EMA_SLOW = 50

USE_RSI_FILTER = True
RSI_LEN = 14
RSI_THRESH_LONG = 50.0
RSI_THRESH_SHORT = 50.0

USE_ADX_FILTER = False
ADX_LEN = 14
ADX_MIN = 18.0

# Daily trend + ATR band for OR width
USE_DAILY_TREND_FILTER = True
DAILY_EMA_FAST = 20
DAILY_EMA_SLOW = 50
DAILY_ATR_LEN  = 14

USE_OR_WIDTH_ATR_FILTER = True
OR_ATR_MIN = 0.10
OR_ATR_MAX = 2.00

# Gap filter
USE_GAP_FILTER = True
GAP_MIN = 0.003
GAP_MAX = 0.04
ALIGN_WITH_GAP_DIR = False

# DOW filter
USE_DOW_FILTER = False
ALLOWED_DOW = {1, 2, 3}  # Tue–Thu

# Entry realism
#   "next_open" (recommended) or "confirm_close"
ENTRY_ON_CONFIRM = "next_open"

# Execution realism: same-bar sequencing + partial-fills
FILL_POLICY = "stop_first"  # "stop_first" | "tp_first" | "random"
PARTIAL_TP1_IF_RANGE_FRAC = 0.10  # if bar range < 10% of risk, TP1 fills only 70%
PARTIAL_TP1_PCT = 0.70

# Risk / costs
POSITION_SIZE_DOLLARS = 10_000
SLIPPAGE_BPS = 1.0
FEES_PER_TRADE = 0.00
FEE_PER_FILL = True

# Batch / persistence
SLEEP_BETWEEN_TICKERS = 0.0
AUTOSAVE_EVERY = 25

# Output files
TRADES_CSV_ALL = "orb_trades_sp500.csv"
EQUITY_CSV_ALL = "orb_equity_sp500.csv"
SUMMARY_CSV    = "orb_summary_sp500.csv"
RESULTS_JSON   = "orb_results_suite.json"   # NEW: consolidated results

# =============================
# VALIDATION FLAGS
# =============================

RUN_BASELINE = True
RUN_PURGED_KFOLD = True
KFOLD_SPLITS = 5
PURGE_DAYS   = 2
EMBARGO_DAYS = 2

RUN_WALK_FORWARD = True
WF_TRAIN_DAYS = 365 * 2
WF_TEST_DAYS  = 180

RUN_NESTED_WF = True
# Nested WF window sizes
NWF_TRAIN_DAYS   = 365 * 2
NWF_CALIB_DAYS   = 120
NWF_TEST_DAYS    = 180
NWF_PURGE_DAYS   = 2
NWF_EMBARGO_DAYS = 2
NWF_MIN_TRADES_CALIB = 200

# Parameter grid
GRID_EMA_FAST = [12, 20]
GRID_EMA_SLOW = [26, 50]
GRID_BREAKOUT_CUSHION = [0.0005, 0.0010, 0.0015]

# Heatmap neighborhood (around chosen WFO key)
HEAT_EMA_FAST = [10, 12, 16, 20, 24]
HEAT_EMA_SLOW = [22, 26, 34, 50, 70]
HEAT_CUSHION  = [0.0003, 0.0005, 0.0010, 0.0015, 0.0020]

# Slippage/fee surface
SLIPPAGE_SWEEP_BPS = [1, 2, 3, 5, 8, 10]
FEE_SWEEP = [0.00, 0.0025, 0.0050]  # $ per fill or per trade depending on FEE_PER_FILL

# Permutation (null) test
PERMUTATIONS = 200
PERM_SEED = 1337

# Block bootstrap CIs
BOOTSTRAP_ITERS = 1000
BOOT_BLOCK_LEN  = 10  # sessions

# Portfolio constraints
REPLAY_MAX_POSITIONS = 10
REPLAY_PER_NAME_CAP  = 2

# =============================
# Helpers
# =============================

def ensure_unique_columns(df: pd.DataFrame) -> pd.DataFrame:
    if getattr(df.columns, "duplicated", None) is not None and df.columns.duplicated().any():
        df = df.loc[:, ~df.columns.duplicated()].copy()
    return df

def sessionize(obj) -> pd.Series:
    idx = getattr(obj, "index", obj)
    return pd.to_datetime(pd.Index(idx).date)

def session_series(df_or_series) -> pd.Series:
    if isinstance(df_or_series, pd.DataFrame):
        s = df_or_series["Session"] if "Session" in df_or_series.columns else df_or_series.index
    else:
        s = df_or_series
    s = pd.to_datetime(s, errors="coerce")
    s = pd.Series(pd.DatetimeIndex(s).tz_localize(None).normalize(), index=getattr(df_or_series, "index", None))
    return s

def _to_clock(ts) -> dtime:
    ts = pd.Timestamp(ts)
    return dtime(ts.hour, ts.minute)

def _clock_le(ts, hhmm: str) -> bool:
    h, m = map(int, hhmm.split(":"))
    return _to_clock(ts) <= dtime(h, m)

def wick_opposite_fraction(row: pd.Series, direction: str) -> float:
    h, l, o, c = float(row["High"]), float(row["Low"]), float(row.get("Open", np.nan)), float(row["Close"])
    rng = max(h - l, 1e-12)
    if direction == "long":
        opp = min(o, c) - l
    else:
        opp = h - max(o, c)
    return float(max(opp, 0.0) / rng)

def _apply_slippage(price: float, bps: float, side: str) -> float:
    factor = 1 + (bps/10000.0)
    return price * factor if side == "buy" else price / factor

def _normalize_splits(num_targets: int, splits: tuple) -> List[float]:
    if num_targets <= 0:
        return [1.0]
    if not splits:
        return [1.0 / num_targets] * num_targets
    arr = list(splits[:num_targets])
    if len(arr) < num_targets:
        arr += [0.0] * (num_targets - len(arr))
    s = sum(arr)
    if s <= 0:
        return [1.0 / num_targets] * num_targets
    return [x / s for x in arr]

def _to_dt_index(obj) -> pd.DatetimeIndex:
    if isinstance(obj, (pd.Series, pd.Index, np.ndarray, list, tuple)):
        di = pd.to_datetime(obj, errors="coerce")
    else:
        di = pd.to_datetime([obj], errors="coerce")
    di = pd.DatetimeIndex(di).dropna().tz_localize(None).normalize()
    di = di.sort_values().unique()
    return di

def _fmt(ts) -> str:
    return pd.Timestamp(ts).strftime("%Y-%m-%d")

def _rng_seed(seed: int = 42):
    np.random.seed(seed)
    random.seed(seed)

# Overall totals
def print_overall_totals(trades: pd.DataFrame):
    print("\n=== Overall Totals ===")
    if trades is None or trades.empty:
        print("Total trades: 0")
        print("Total PnL ($): 0.00")
        print("Win rate: 0.00%")
        print("Avg R: 0.000 | Median R: 0.000")
        print("Profit Factor: 0.000")
        print("Avg Win ($): 0.00 | Avg Loss ($): 0.00")
        return

    pnl = pd.to_numeric(trades["PnL_$"], errors="coerce").fillna(0.0)
    wins_mask = pnl > 0
    losses_mask = pnl < 0

    total_trades = int(len(pnl))
    total_pnl    = float(pnl.sum())
    win_rate     = float(wins_mask.mean() * 100.0) if total_trades else 0.0
    avg_r        = float(pd.to_numeric(trades["R_multiple"], errors="coerce").mean())
    median_r     = float(pd.to_numeric(trades["R_multiple"], errors="coerce").median())

    gross_profit = float(pnl[wins_mask].sum())
    gross_loss   = float(-pnl[losses_mask].sum())
    pf = (gross_profit / gross_loss) if gross_loss > 0 else (float("inf") if gross_profit > 0 else 0.0)

    avg_win  = float(pnl[wins_mask].mean()) if wins_mask.any() else 0.0
    avg_loss = float(pnl[losses_mask].mean()) if losses_mask.any() else 0.0

    print(f"Total trades: {total_trades:,}")
    print(f"Total PnL ($): {total_pnl:,.2f}")
    print(f"Win rate: {win_rate:.2f}%")
    print(f"Avg R: {avg_r:.3f} | Median R: {median_r:.3f}")
    print(f"Profit Factor: {pf:.3f}")
    print(f"Avg Win ($): {avg_win:,.2f} | Avg Loss ($): {avg_loss:,.2f}")

# =============================
# Indicators & features
# =============================

def compute_opening_range(df_15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    for c in ("ORH","ORL"):
        if (df.columns == c).sum() > 0:
            df = df.drop(columns=[c])
    for col in ("High","Low"):
        if col not in df.columns:
            raise ValueError(f"compute_opening_range: missing column {col}")
    if not df.index.is_monotonic_increasing:
        df = df.sort_index()
    df["Session"] = sessionize(df)
    first_bar_idx = df.groupby("Session").head(1).index
    df["ORH"] = np.nan
    df["ORL"] = np.nan
    df.loc[first_bar_idx, "ORH"] = df.loc[first_bar_idx, "High"].astype(float).values
    df.loc[first_bar_idx, "ORL"] = df.loc[first_bar_idx, "Low"].astype(float).values
    df[["ORH","ORL"]] = df.groupby("Session")[["ORH","ORL"]].ffill()
    return ensure_unique_columns(df)

def add_session_vwap(df_15: pd.DataFrame) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df["Session"] = sessionize(df)
    tp = (df["High"] + df["Low"] + df["Close"]) / 3.0
    if "Volume" in df.columns:
        df["vwap_num"] = tp * df["Volume"]
        df["vwap_den"] = df["Volume"].replace(0, np.nan)
    else:
        df["vwap_num"] = tp
        df["vwap_den"] = 1.0
    df["VWAP"] = (df.groupby("Session")["vwap_num"].cumsum() /
                  df.groupby("Session")["vwap_den"].cumsum())
    return ensure_unique_columns(df.drop(columns=["vwap_num","vwap_den"]))

def add_ema_trend(df_15: pd.DataFrame, fast=EMA_FAST, slow=EMA_SLOW) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df[f"EMA{fast}"] = df["Close"].ewm(span=fast, adjust=False).mean()
    df[f"EMA{slow}"] = df["Close"].ewm(span=slow, adjust=False).mean()
    return ensure_unique_columns(df)

def rsi(series: pd.Series, length=14) -> pd.Series:
    delta = series.diff()
    up = delta.clip(lower=0)
    dn = -delta.clip(upper=0)
    avg_gain = up.ewm(alpha=1/length, adjust=False).mean()
    avg_loss = dn.ewm(alpha=1/length, adjust=False).mean()
    rs = avg_gain / (avg_loss.replace(0, np.nan))
    return 100 - (100 / (1 + rs))

def add_rsi(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    df[f"RSI{length}"] = rsi(df["Close"], length)
    return ensure_unique_columns(df)

def add_adx(df_15: pd.DataFrame, length=14) -> pd.DataFrame:
    df = ensure_unique_columns(df_15.copy())
    up_move = df["High"].diff()
    dn_move = -df["Low"].diff()
    plus_dm  = np.where((up_move > dn_move) & (up_move > 0), up_move, 0.0)
    minus_dm = np.where((dn_move > up_move) & (dn_move > 0), dn_move, 0.0)
    prev_close = df["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (df["High"] - df["Low"]).values,
        np.abs(df["High"] - prev_close).values,
        np.abs(df["Low"]  - prev_close).values
    ]), index=df.index)
    atr = tr.ewm(alpha=1/length, adjust=False).mean()
    plus_di = 100 * pd.Series(plus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    minus_di = 100 * pd.Series(minus_dm, index=df.index).ewm(alpha=1/length, adjust=False).mean() / atr
    dx = (abs(plus_di - minus_di) / (plus_di + minus_di).replace(0, np.nan)) * 100
    adx = dx.ewm(alpha=1/length, adjust=False).mean()
    df[f"ADX{length}"] = adx
    return ensure_unique_columns(df)

def build_daily_from_15m(df_15: pd.DataFrame) -> pd.DataFrame:
    d = ensure_unique_columns(df_15.copy())
    d["Session"] = sessionize(d)
    daily = d.groupby("Session").agg(
        Open=("Open","first"),
        High=("High","max"),
        Low=("Low","min"),
        Close=("Close","last"),
        Volume=("Volume","sum") if "Volume" in d.columns else ("Close","size"),
    )
    return daily

def add_daily_bias_and_atr(df_15: pd.DataFrame, ema_fast=20, ema_slow=50, atr_len=14) -> pd.DataFrame:
    dly = build_daily_from_15m(df_15)
    dly[f"EMA_D{ema_fast}"] = dly["Close"].ewm(span=ema_fast, adjust=False).mean()
    dly[f"EMA_D{ema_slow}"] = dly["Close"].ewm(span=ema_slow, adjust=False).mean()
    prev_close = dly["Close"].shift(1)
    tr = pd.Series(np.maximum.reduce([
        (dly["High"] - dly["Low"]).values,
        np.abs(dly["High"] - prev_close).values,
        np.abs(dly["Low"]  - prev_close).values
    ]), index=dly.index)
    atr_col = f"ATR_D{atr_len}"
    dly[atr_col] = tr.ewm(alpha=1/atr_len, adjust=False).mean()

    if "Volume" in df_15.columns:
        tmp = df_15.copy(); tmp["Session"] = sessionize(tmp)
        open15_vol = tmp.groupby("Session")["Volume"].first()
        dly["Open15_Vol"] = open15_vol
        dly["Open15_Vol_Thresh"] = (
            dly["Open15_Vol"].rolling(window=60, min_periods=10).quantile(0.30)
        )

    feature_cols = [f"EMA_D{ema_fast}", f"EMA_D{ema_slow}", atr_col, "Open15_Vol", "Open15_Vol_Thresh"]
    feature_cols = [c for c in feature_cols if c in dly.columns]

    out = ensure_unique_columns(df_15.copy())
    out["Session"] = sessionize(out)
    out = out.merge(dly[feature_cols], left_on="Session", right_index=True, how="left")
    return ensure_unique_columns(out)

# =============================
# Backtest (per-ticker)
# =============================

def _tp_stop_order_this_bar(policy: str) -> List[str]:
    if policy == "tp_first":
        return ["tp", "stop"]
    if policy == "random":
        return ["tp","stop"] if random.random() < 0.5 else ["stop","tp"]
    return ["stop","tp"]

def backtest_orb_retest(
    df_15: pd.DataFrame,
    ema_fast:int = EMA_FAST,
    ema_slow:int = EMA_SLOW,
    breakout_cushion: float = BREAKOUT_CUSHION_PCT,
    max_retest_min: int = MAX_RETEST_MIN,
    r_targets: List[float] = TARGETS_R,
    dollars: float = POSITION_SIZE_DOLLARS,
    slippage_bps: float = SLIPPAGE_BPS,
    fees: float = FEES_PER_TRADE,
    retest_confirm_close: bool = RETEST_CONFIRM_CLOSE,
    fill_policy: str = FILL_POLICY
) -> Tuple[pd.DataFrame, pd.DataFrame]:

    df = compute_opening_range(df_15)
    if USE_VWAP_FILTER:  df = add_session_vwap(df)
    if USE_TREND_FILTER: df = add_ema_trend(df, ema_fast, ema_slow)
    if USE_RSI_FILTER:   df = add_rsi(df, RSI_LEN)
    if USE_ADX_FILTER:   df = add_adx(df, ADX_LEN)
    df = add_daily_bias_and_atr(df, DAILY_EMA_FAST, DAILY_EMA_SLOW, DAILY_ATR_LEN)

    df["Session"] = sessionize(df)
    first_idx = df.groupby("Session").head(1).index
    prev_close_series = df.groupby("Session")["Close"].last().shift(1)
    prev_close_map = df["Session"].map(prev_close_series)
    df.loc[first_idx, "GapPct"] = (df.loc[first_idx, "Open"] / prev_close_map.loc[first_idx] - 1.0)
    df["GapPct"] = df.groupby("Session")["GapPct"].ffill()

    sessions = df["Session"].unique()
    trades = []

    for ses in sessions:
        sdf = df[df["Session"] == ses].copy()
        if len(sdf) < 3: continue

        orh, orl = float(sdf["ORH"].iloc[0]), float(sdf["ORL"].iloc[0])
        if np.isnan(orh) or np.isnan(orl): continue

        after_open = sdf.iloc[1:].copy()
        long_breaks  = after_open[after_open["Close"] > orh * (1 + breakout_cushion)]
        short_breaks = after_open[after_open["Close"] < orl * (1 - breakout_cushion)]

        cands = []
        for idx, row in long_breaks.iterrows():  cands.append(("long", idx, row))
        for idx, row in short_breaks.iterrows(): cands.append(("short", idx, row))
        cands.sort(key=lambda x: x[1])

        trades_this_session = 0
        took_long = took_short = False

        for direction, btime, breakout_row in cands:
            if not _clock_le(breakout_row.name, BREAKOUT_DEADLINE): continue
            if not ALLOW_BOTH_SIDES:
                if took_long and direction == "short":  break
                if took_short and direction == "long":  break
            if direction == "long" and took_long:   continue
            if direction == "short" and took_short: continue

            if USE_DOW_FILTER and btime.weekday() not in ALLOWED_DOW: continue

            if USE_GAP_FILTER and "GapPct" in sdf.columns:
                gap = sdf.loc[sdf.index.min(), "GapPct"]
                if pd.notna(gap) and not (GAP_MIN <= abs(float(gap)) <= GAP_MAX):
                    continue
                if ALIGN_WITH_GAP_DIR and pd.notna(gap):
                    g = float(gap)
                    if (direction == "long" and g < 0) or (direction == "short" and g > 0):
                        continue

            if USE_VWAP_FILTER and "VWAP" in sdf.columns:
                rv = sdf.loc[:btime, "VWAP"].tail(3).dropna().values
                rising  = (len(rv) < 3) and (len(rv) >= 1) or (len(rv) >= 3 and np.all(np.diff(rv) >= 0))
                falling = (len(rv) < 3) and (len(rv) >= 1) or (len(rv) >= 3 and np.all(np.diff(rv) <= 0))
                if direction == "long":
                    if not (breakout_row.get("VWAP", np.nan) < breakout_row["Close"] and rising): continue
                else:
                    if not (breakout_row.get("VWAP", np.nan) > breakout_row["Close"] and falling): continue

            if USE_TREND_FILTER and (f"EMA{ema_fast}" in sdf.columns) and (f"EMA{ema_slow}" in sdf.columns):
                emaf = breakout_row.get(f"EMA{ema_fast}", np.nan); emas = breakout_row.get(f"EMA{ema_slow}", np.nan)
                if np.isfinite(emaf) and np.isfinite(emas):
                    if direction == "long" and not (emaf > emas): continue
                    if direction == "short" and not (emaf < emas): continue

            if USE_DAILY_TREND_FILTER and f"EMA_D{DAILY_EMA_FAST}" in sdf.columns and f"EMA_D{DAILY_EMA_SLOW}" in sdf.columns:
                dfast = float(sdf[f"EMA_D{DAILY_EMA_FAST}"].iloc[0])
                dslow = float(sdf[f"EMA_D{DAILY_EMA_SLOW}"].iloc[0])
                if direction == "long" and not (dfast > dslow): continue
                if direction == "short" and not (dfast < dslow): continue

            atr_col = f"ATR_D{DAILY_ATR_LEN}"
            if USE_OR_WIDTH_ATR_FILTER and atr_col in sdf.columns:
                or_width = float(orh - orl)
                atr_d = float(sdf[atr_col].iloc[0])
                if atr_d <= 0: continue
                frac = or_width / atr_d
                if not (OR_ATR_MIN <= frac <= OR_ATR_MAX): continue

            if USE_RSI_FILTER and f"RSI{RSI_LEN}" in sdf.columns:
                rsi_val = float(breakout_row[f"RSI{RSI_LEN}"])
                if not np.isfinite(rsi_val): continue
                if direction == "long" and not (rsi_val >= RSI_THRESH_LONG): continue
                if direction == "short" and not (rsi_val <= 100 - RSI_THRESH_SHORT): continue

            if USE_ADX_FILTER and f"ADX{ADX_LEN}" in sdf.columns:
                adx_val = float(breakout_row[f"ADX{ADX_LEN}"])
                if not np.isfinite(adx_val) or not (adx_val >= ADX_MIN): continue

            cutoff = btime + timedelta(minutes=max_retest_min)
            window = sdf[(sdf.index > btime) & (sdf.index <= cutoff)].copy()
            level = orh if direction == "long" else orl
            touch = window[(window["Low"] <= level) & (window["High"] >= level)].head(1)

            retest_used = True
            if touch.empty:
                if USE_CONTINUATION and _clock_le(breakout_row.name, CONTINUATION_CUTOFF):
                    retest_used = False
                    retest_time = breakout_row.name
                    retest_bar = breakout_row
                else:
                    continue
            else:
                retest_bar = touch.iloc[0]
                retest_time = retest_bar.name

            # Enforce optional retest deadline
            if retest_used and RETEST_DEADLINE and (not _clock_le(retest_time, RETEST_DEADLINE)):
                continue

            # Confirm close through level if requested
            if retest_used and retest_confirm_close:
                if direction == "long":
                    ok = (retest_bar["Close"] >= level * (1 + breakout_cushion))
                else:
                    ok = (retest_bar["Close"] <= level * (1 - breakout_cushion))
                if not ok:
                    continue

            # Reject with big opposite wick
            if MAX_OPPOSITE_WICK_FRAC is not None:
                if wick_opposite_fraction(retest_bar, direction) > MAX_OPPOSITE_WICK_FRAC:
                    continue

            # ===== Entry price (realistic) =====
            if retest_used and retest_confirm_close:
                if ENTRY_ON_CONFIRM.lower() == "next_open":
                    next_bar = sdf[sdf.index > retest_time].head(1)
                    if next_bar.empty:
                        continue
                    entry_time = next_bar.index[0]
                    raw_entry = float(next_bar["Open"].iloc[0])
                else:
                    entry_time = retest_time
                    raw_entry = float(retest_bar["Close"])
            else:
                entry_time = retest_time
                raw_entry = float(level)

            entry = _apply_slippage(raw_entry, slippage_bps, "buy" if direction == "long" else "sell")
            stop_init = orl if direction == "long" else orh          # initial risk stop (used for R)
            stop = stop_init                                          # may move to BE later
            rps  = abs(entry - stop_init)
            if rps <= 1e-12: continue

            qty = POSITION_SIZE_DOLLARS / max(entry, 1e-12)
            side_mult = 1 if direction == "long" else -1
            targets = [(entry + r * rps) if direction == "long" else (entry - r * rps) for r in TARGETS_R]

            run = sdf[sdf.index >= entry_time].copy()
            qty_left = qty
            exits = []
            next_tp_idx = 0
            splits = _normalize_splits(len(targets), SCALE_SPLIT) if (USE_SCALE_OUT and len(targets)>=1) else [1.0]
            leg_sizes = [qty * s for s in splits]

            be_level = entry
            be_pending = False
            be_set_time = None

            # Simulate forward
            for ts, row in run.iterrows():
                hi, lo = float(row["High"]), float(row["Low"])
                bar_range = hi - lo

                # move to BE if armed
                if be_pending and (not BE_ON_NEXT_BAR or (be_set_time is not None and ts > be_set_time)):
                    stop = be_level
                    be_pending = False

                # same-bar sequencing policy
                order = _tp_stop_order_this_bar(fill_policy)

                # Evaluate both target and stop hits; track sequential targets
                executed = False
                for first in order:
                    if first == "stop":
                        if lo <= stop <= hi:
                            px = _apply_slippage(stop, SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                            label = "breakeven" if abs(px - be_level) < 1e-10 else "stop"
                            exits.append((label, ts, px, qty_left))
                            qty_left = 0.0
                            executed = True
                            break
                    else:  # "tp"
                        if next_tp_idx < len(targets):
                            tp = targets[next_tp_idx]
                            if lo <= tp <= hi:
                                px = _apply_slippage(tp, SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                                fill_qty = leg_sizes[next_tp_idx] if next_tp_idx < len(leg_sizes) else qty_left

                                # Optional partial TP1 on very tight bar
                                if next_tp_idx == 0 and bar_range < PARTIAL_TP1_IF_RANGE_FRAC * rps:
                                    fill_qty = max(0.0, fill_qty * PARTIAL_TP1_PCT)

                                if fill_qty > 1e-9:
                                    exits.append((f"tp{next_tp_idx+1}", ts, px, fill_qty))
                                    qty_left -= fill_qty
                                if next_tp_idx == 0 and MOVE_STOP_TO_BE_AT_R is not None:
                                    be_pending = True
                                    be_set_time = ts
                                next_tp_idx += 1
                                if qty_left <= 1e-9:
                                    executed = True
                                    break
                if executed:
                    break

                # If no TP1 by continuation cutoff, tighten to BE
                if USE_CONTINUATION and not _clock_le(ts, CONTINUATION_CUTOFF) and next_tp_idx == 0:
                    stop = be_level

            # EOD exit
            if qty_left > 1e-9:
                last = run.iloc[-1]
                px = _apply_slippage(float(last["Close"]), SLIPPAGE_BPS, "sell" if direction == "long" else "buy")
                exits.append(("eod", run.index[-1], px, qty_left))
                qty_left = 0.0

            cash_pnl = sum((px - entry) * side_mult * q for (_, _, px, q) in exits)
            if FEE_PER_FILL:
                cash_pnl -= FEES_PER_TRADE * len(exits)
            else:
                cash_pnl -= FEES_PER_TRADE

            trades.append({
                "Ticker": sdf["Ticker"].iloc[0] if "Ticker" in sdf.columns else "",
                "Session": ses,
                "Direction": direction,
                "EntryType": "retest" if retest_used else "continuation",
                "ConfirmMode": ENTRY_ON_CONFIRM.lower(),
                "ORH": orh, "ORL": orl,
                "EntryTime": entry_time,
                "Entry": entry,
                "Stop_init": stop_init,           # NEW
                "Stop_final": stop,               # NEW (BE or last)
                "Targets": targets,
                "Exits": [(lab, ts, px, q) for (lab, ts, px, q) in exits],
                "Qty": qty,
                "PnL_$": cash_pnl,
                "R_multiple": cash_pnl / (rps * max(qty, 1e-12))
            })

            if direction == "long":  took_long  = True
            if direction == "short": took_short = True
            trades_this_session += 1
            if trades_this_session >= MAX_TRADES_PER_SESSION:
                break

    trade_log = pd.DataFrame(trades)
    if trade_log.empty:
        eq = pd.DataFrame(columns=["Session","CumPnL_$"])
        return trade_log, eq

    equity = (trade_log.groupby("Session")["PnL_$"].sum()
              .sort_index().cumsum().reset_index().rename(columns={"PnL_$":"CumPnL_$"}))
    return trade_log, equity

# =============================
# Parquet loading
# =============================

STD_MAP = {"open":"Open","high":"High","low":"Low","close":"Close","volume":"Volume"}

def _standardize_ohlcv_columns(df: pd.DataFrame) -> pd.DataFrame:
    rename = {}
    lower = {c.lower(): c for c in df.columns}
    for k, std in STD_MAP.items():
        if k in lower:                rename[lower[k]] = std
        elif k.capitalize() in df.columns: rename[k.capitalize()] = std
        elif k.upper() in df.columns: rename[k.upper()] = std
    out = df.rename(columns=rename)
    drop_cols = [c for c in out.columns if str(c).lower().startswith("adj")]
    return out.drop(columns=drop_cols, errors="ignore")

def _choose_dt_index(df: pd.DataFrame) -> pd.DataFrame:
    if isinstance(df.index, pd.DatetimeIndex):
        return df
    for cand in ["EventAt","Datetime","datetime","Timestamp","timestamp","Date","date","Time","time"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df = df.set_index(cand)
            break
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("No datetime index/column found.")
    return df

def _maybe_resample_to_15m(df: pd.DataFrame) -> pd.DataFrame:
    if len(df.index) < 3:
        return df
    deltas = (df.index[1:] - df.index[:-1]).asi8
    if len(deltas) == 0:
        return df
    med_ns = np.median(deltas)
    fifteen_ns = pd.Timedelta("15T").value
    if med_ns < fifteen_ns:
        o = df["Open"].resample("15T", label="left", closed="left").first()
        h = df["High"].resample("15T", label="left", closed="left").max()
        l = df["Low"].resample("15T", label="left", closed="left").min()
        c = df["Close"].resample("15T", label="left", closed="left").last()
        v = df["Volume"].resample("15T", label="left", closed="left").sum() if "Volume" in df.columns else None
        parts = {"Open":o,"High":h,"Low":l,"Close":c}
        if v is not None: parts["Volume"] = v
        out = pd.concat(parts, axis=1).dropna(subset=["Open","High","Low","Close"])
        return out
    return df

def load_parquet_15m(path: str, tz: str = TZ) -> pd.DataFrame:
    df = pd.read_parquet(path, engine="pyarrow")
    df = _choose_dt_index(df).sort_index()
    if df.index.tz is None:
        if ASSUME_NAIVE_TIMESTAMPS_ARE_UTC:
            df = df.tz_localize("UTC").tz_convert(tz)
        else:
            df = df.tz_localize(tz)
    else:
        df = df.tz_convert(tz)
    df = _standardize_ohlcv_columns(df)
    for need in ["Open","High","Low","Close"]:
        if need not in df.columns:
            raise ValueError(f"{os.path.basename(path)} missing required column: {need}")
    df = _maybe_resample_to_15m(df)
    df = df.between_time(REG_SESSION_START, REG_SESSION_END)
    return ensure_unique_columns(df.sort_index())

# =============================
# Processing helpers
# =============================

def iter_parquet_files(folder: str):
    return sorted(glob.glob(os.path.join(folder, "*.parquet")))

def run_folder(folder: str,
               ema_fast:int = EMA_FAST,
               ema_slow:int = EMA_SLOW,
               breakout_cushion: float = BREAKOUT_CUSHION_PCT,
               slippage_bps: float = SLIPPAGE_BPS,
               fees: float = FEES_PER_TRADE,
               fill_policy: str = FILL_POLICY) -> Tuple[pd.DataFrame, pd.DataFrame]:
    files = iter_parquet_files(folder)
    print(f"Found {len(files)} parquet files.")
    all_trades, all_equity = [], []
    iterator = tqdm(files, desc="Backtesting files", unit="file") if tqdm is not None else files

    for i, fpath in enumerate(iterator, 1):
        ticker = os.path.splitext(os.path.basename(fpath))[0].upper().replace("_","").replace("-","")
        try:
            df15 = load_parquet_15m(fpath)
            df15["Ticker"] = ticker
        except Exception as e:
            print(f"\nSkipping {ticker}: {e}")
            continue

        tlog, eq = backtest_orb_retest(
            df15,
            ema_fast=ema_fast,
            ema_slow=ema_slow,
            breakout_cushion=breakout_cushion,
            max_retest_min=MAX_RETEST_MIN,
            dollars=POSITION_SIZE_DOLLARS,
            slippage_bps=slippage_bps,
            fees=fees,
            retest_confirm_close=RETEST_CONFIRM_CLOSE,
            fill_policy=fill_policy
        )

        if not tlog.empty:
            tlog_out = tlog.copy()
            tlog_out["Targets"] = tlog_out["Targets"].apply(lambda xs: ";".join([f"{p:.6f}" for p in xs]))
            tlog_out["Exits"]   = tlog_out["Exits"].apply(lambda xs: ";".join([f"{t[0]}|{pd.Timestamp(t[1]).isoformat()}|{t[2]:.6f}|{t[3]:.6f}" for t in xs]))
            all_trades.append(tlog_out)
        if not eq.empty:
            eq_out = eq.copy(); eq_out["Ticker"] = ticker
            all_equity.append(eq_out)

        if AUTOSAVE_EVERY and i % AUTOSAVE_EVERY == 0:
            if all_trades:
                pd.concat(all_trades, ignore_index=True).to_csv(TRADES_CSV_ALL, index=False)
                print(f"[autosave] wrote {TRADES_CSV_ALL}")
            if all_equity:
                pd.concat(all_equity, ignore_index=True).to_csv(EQUITY_CSV_ALL, index=False)
                print(f"[autosave] wrote {EQUITY_CSV_ALL}")

        if SLEEP_BETWEEN_TICKERS > 0:
            time.sleep(SLEEP_BETWEEN_TICKERS)

    trades = pd.concat(all_trades, ignore_index=True) if all_trades else pd.DataFrame()
    equity = pd.concat(all_equity, ignore_index=True) if all_equity else pd.DataFrame()
    return trades, equity

# =============================
# Validation & advanced metrics
# =============================

def metrics_basic(df: pd.DataFrame) -> Dict[str, float]:
    if df is None or df.empty:
        return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, SumR=0.0)
    r = pd.to_numeric(df.get("R_multiple", pd.Series(dtype=float)), errors="coerce").dropna()
    if r.empty:
        return dict(Trades=0, WinRate=0.0, PF=0.0, AvgR=0.0, SumR=0.0)
    wins = r[r > 0]; losses = r[r < 0]
    gp = wins.sum(); gl = -losses.sum()
    pf = (gp / gl) if gl > 1e-12 else (np.inf if gp > 0 else 0.0)
    return dict(
        Trades=int(r.size),
        WinRate=float((r > 0).mean() * 100.0),
        PF=float(pf),
        AvgR=float(r.mean()),
        SumR=float(r.sum())
    )

def metrics_advanced(trades: pd.DataFrame) -> Dict[str, float]:
    if trades is None or trades.empty:
        return {"Sharpe":0.0, "Sortino":0.0, "MaxDD_$":0.0, "CVaR95_$":0.0, "Exposure_%":0.0}
    ses = trades.groupby("Session")["PnL_$"].sum().sort_index()
    if ses.empty:
        return {"Sharpe":0.0, "Sortino":0.0, "MaxDD_$":0.0, "CVaR95_$":0.0, "Exposure_%":0.0}
    rets = ses / max(POSITION_SIZE_DOLLARS, 1e-12)
    mu = rets.mean()
    sd = rets.std(ddof=1)
    neg = rets[rets<0]
    dd = ses.cumsum().cummax() - ses.cumsum()
    maxdd = float(dd.max()) if len(dd)>0 else 0.0
    # annualization by ~252 sessions ~ 252 trading days (approx)
    sharpe = float((np.sqrt(252) * mu / sd)) if sd > 1e-12 else 0.0
    downside_sd = neg.std(ddof=1)
    sortino = float((np.sqrt(252) * mu / downside_sd)) if downside_sd and downside_sd>1e-12 else 0.0
    # CVaR/ETL at 95% on $PnL per session
    losses = ses[ses < 0.0]
    if losses.empty:
        cvar95 = 0.0
    else:
        q = np.quantile(losses, 0.05)
        tail = losses[losses <= q]
        cvar95 = float(tail.mean()) if len(tail)>0 else float(q)
    # crude exposure: % sessions with any position assumed ~= sessions with trade
    exposure = 100.0 * (ses.ne(0).mean())
    return {"Sharpe":sharpe, "Sortino":sortino, "MaxDD_$":-maxdd, "CVaR95_$":cvar95, "Exposure_%":exposure}

def _print_metrics(title: str, m: Dict[str, float]):
    print(f"{title} | Trades: {m['Trades']:,} | Win%: {m['WinRate']:.2f}% | PF: {m['PF']:.3f} | Avg R: {m['AvgR']:.3f} | SumR: {m['SumR']:,.2f}")

def purged_kfold_cv(trades: pd.DataFrame, n_splits=5, purge_days=2, embargo_days=2) -> List[dict]:
    print("\n--- Purged K-Fold CV (session-based) ---")
    ses_unique = _to_dt_index(trades["Session"])
    if len(ses_unique) < n_splits:
        print("Not enough sessions to split.")
        return []
    ses_row = session_series(trades)

    edges = np.linspace(0, len(ses_unique), n_splits + 1, dtype=int)
    out = []
    for i in range(n_splits):
        test_ses = ses_unique[edges[i]:edges[i+1]]
        if len(test_ses) == 0: continue
        test_start, test_end = test_ses.min(), test_ses.max()
        purge_start = test_start - pd.Timedelta(days=purge_days)
        embargo_end = test_end   + pd.Timedelta(days=embargo_days)
        train_mask = (ses_row < purge_start) | (ses_row > embargo_end)
        test_mask  = (ses_row >= test_start) & (ses_row <= test_end)
        trn, tst = trades.loc[train_mask], trades.loc[test_mask]
        tr_m, te_m = metrics_basic(trn), metrics_basic(tst)
        _print_metrics(f"Fold {i+1:02d} Train [{_fmt(purge_start)} purge→]", tr_m)
        _print_metrics(f"Fold {i+1:02d} Test  [{_fmt(test_start)}→{_fmt(test_end)}]", te_m)
        out.append({"fold":i+1, "train":tr_m, "test":te_m,
                    "dates":{"test_start":_fmt(test_start),"test_end":_fmt(test_end)}})
    return out

def walk_forward_validation(trades: pd.DataFrame, train_days: int, test_days: int) -> List[dict]:
    print("\n--- Walk-Forward Validation ---")
    ses_unique = _to_dt_index(trades["Session"])
    if len(ses_unique) == 0:
        print("No sessions to split.")
        return []
    t0, t1 = ses_unique.min(), ses_unique.max()
    ses_row = session_series(trades)
    fold, train_start = 0, t0
    out = []
    while True:
        train_end = train_start + pd.Timedelta(days=train_days - 1)
        test_start = train_end + pd.Timedelta(days=1)
        test_end   = test_start + pd.Timedelta(days=test_days - 1)
        if test_end > t1: break
        fold += 1
        train_mask = (ses_row >= train_start) & (ses_row <= train_end)
        test_mask  = (ses_row >= test_start) & (ses_row <= test_end)
        trn, tst = trades.loc[train_mask], trades.loc[test_mask]
        tr_m, te_m = metrics_basic(trn), metrics_basic(tst)
        _print_metrics(f"Fold {fold:02d} Train [{_fmt(train_start)}→{_fmt(train_end)}]", tr_m)
        _print_metrics(f"Fold {fold:02d} Test  [{_fmt(test_start)}→{_fmt(test_end)} ]", te_m)
        out.append({"fold":fold, "train":tr_m, "test":te_m,
                    "dates":{"train_start":_fmt(train_start),"train_end":_fmt(train_end),
                             "test_start":_fmt(test_start),"test_end":_fmt(test_end)}})
        train_start = test_end + pd.Timedelta(days=1)
    return out

# ---- Nested WFO (Train→Calibrate→Test) ----

def nested_walk_forward_with_calibration(
    grid_returns: dict,
    train_days: int,
    calib_days: int,
    test_days: int,
    purge_days: int = 2,
    embargo_days: int = 2,
    min_trades_calib: int = 100
):
    all_sessions = []
    for key, df in grid_returns.items():
        if df is None or len(df) == 0:
            continue
        if "Session" in df.columns:
            all_sessions.append(df["Session"])
        else:
            all_sessions.append(df.index)
    if not all_sessions:
        print("No sessions found in grid_returns.")
        return [], None

    all_sess = _to_dt_index(pd.concat([pd.to_datetime(s, errors="coerce") for s in all_sessions]))
    if len(all_sess) == 0:
        print("All sessions are NaT after coercion.")
        return [], None

    t0, t1 = all_sess.min(), all_sess.max()
    print("\n--- Nested Walk-Forward (Train→Calib→Test) ---")
    print(f"Global session range: {_fmt(t0)} → {_fmt(t1)}")

    wf_records = []
    last_wf_end = None

    def slice_sessions(df: pd.DataFrame, s0, s1):
        if df is None or df.empty:
            return df
        if "Session" in df.columns:
            ses = session_series(df)
        else:
            ses = pd.to_datetime(df.index, errors="coerce")
            ses = pd.Series(pd.DatetimeIndex(ses).tz_localize(None).normalize(), index=df.index)
        m = (ses >= s0) & (ses <= s1)
        return df.loc[m]

    def metrics_for_key(df_slice: pd.DataFrame) -> dict:
        return metrics_basic(df_slice)

    fold = 0
    train_start = t0
    while True:
        train_end   = train_start + pd.Timedelta(days=train_days - 1)
        purge_end   = train_end + pd.Timedelta(days=purge_days)
        calib_end   = purge_end + pd.Timedelta(days=calib_days)
        embargo_end = calib_end + pd.Timedelta(days=embargo_days)
        test_end    = embargo_end + pd.Timedelta(days=test_days)
        if test_end > t1: break

        fold += 1
        print(f"\nFold {fold:02d}")
        print(f" Train:  {_fmt(train_start)} → {_fmt(train_end)}")
        print(f" Purge:  {_fmt(train_end + pd.Timedelta(days=1))} → {_fmt(purge_end)}")
        print(f" Calib:  {_fmt(purge_end + pd.Timedelta(days=1))} → {_fmt(calib_end)}")
        print(f" Embarg: {_fmt(calib_end + pd.Timedelta(days=1))} → {_fmt(embargo_end)}")
        print(f" Test:   {_fmt(embargo_end + pd.Timedelta(days=1))} → {_fmt(test_end)}")

        perkey_train, perkey_calib, perkey_test = {}, {}, {}
        for key, df in grid_returns.items():
            trn = slice_sessions(df, train_start, train_end)
            cal = slice_sessions(df, purge_end + pd.Timedelta(days=1), calib_end)
            tst = slice_sessions(df, embargo_end + pd.Timedelta(days=1), test_end)
            perkey_train[key] = metrics_for_key(trn)
            perkey_calib[key] = metrics_for_key(cal)
            perkey_test[key]  = metrics_for_key(tst)

        eligible = {k: v for k, v in perkey_calib.items() if v["Trades"] >= int(min_trades_calib)}
        if not eligible:
            eligible = {k: v for k, v in perkey_calib.items() if v["Trades"] > 0}
        if not eligible:
            print("  [warn] No eligible keys in calibration; skipping fold.")
            train_start = test_end + pd.Timedelta(days=1)
            continue

        calib_sorted = sorted(
            eligible.items(),
            key=lambda kv: (kv[1]["AvgR"], kv[1]["PF"], kv[1]["WinRate"], kv[1]["Trades"]),
            reverse=True
        )
        chosen_key, chosen_calib = calib_sorted[0]
        chosen_test  = perkey_test[chosen_key]
        chosen_train = perkey_train[chosen_key]

        print(f"  Chosen key: {chosen_key}")
        _print_metrics("  Train ", chosen_train)
        _print_metrics("  Calib ", chosen_calib)
        _print_metrics("  Test  ", chosen_test)

        wf_records.append(dict(
            fold=fold,
            train_start=train_start, train_end=train_end,
            purge_end=purge_end, calib_end=calib_end,
            embargo_end=embargo_end, test_end=test_end,
            chosen_key=chosen_key,
            train_metrics=chosen_train,
            calib_metrics=chosen_calib,
            test_metrics=chosen_test
        ))
        last_wf_end = test_end

        train_start = test_end + pd.Timedelta(days=1)

    if not wf_records:
        print("No folds produced.")
        return [], None

    test_wr = np.mean([rec["test_metrics"]["WinRate"] for rec in wf_records if rec["test_metrics"]["Trades"] > 0])
    test_pf = np.mean([rec["test_metrics"]["PF"] for rec in wf_records if rec["test_metrics"]["Trades"] > 0])
    print(f"\nWFO OOS (avg over folds) | Win%: {test_wr:.2f} | PF: {test_pf:.3f}")
    return wf_records, last_wf_end

def choose_final_param_from_wf(wf_records: List[dict]) -> Optional[str]:
    if not wf_records:
        return None
    keys = [rec["chosen_key"] for rec in wf_records]
    uniq = list(dict.fromkeys(keys))
    scores = {}
    for k in uniq:
        test_sumr = sum(rec["test_metrics"]["SumR"] for rec in wf_records if rec["chosen_key"] == k)
        votes = sum(1 for rec in wf_records if rec["chosen_key"] == k)
        scores[k] = (votes, test_sumr)
    best = sorted(scores.items(), key=lambda kv: (kv[1][0], kv[1][1]), reverse=True)[0][0]
    return best

def build_param_grid() -> List[tuple]:
    grid = list(itertools.product(GRID_EMA_FAST, GRID_EMA_SLOW, GRID_BREAKOUT_CUSHION))
    grid = [(f,s,c) for (f,s,c) in grid if f < s]
    return grid

def backtest_folder_for_params(folder: str, ema_fast:int, ema_slow:int, breakout_cushion: float,
                               slippage_bps: float = SLIPPAGE_BPS, fees: float = FEES_PER_TRADE) -> pd.DataFrame:
    trades, _ = run_folder(folder, ema_fast=ema_fast, ema_slow=ema_slow, breakout_cushion=breakout_cushion,
                           slippage_bps=slippage_bps, fees=fees)
    return trades

def build_grid_returns(folder: str) -> Dict[str, pd.DataFrame]:
    grid = build_param_grid()
    print(f"Param grid size: {len(grid)}")
    iterator = tqdm(grid, desc="Param grid", unit="combo") if tqdm is not None else grid
    out = {}
    for ema_f, ema_s, cush in iterator:
        key = f"ema{ema_f}-{ema_s}_c{cush:.4f}"
        tr = backtest_folder_for_params(folder, ema_f, ema_s, cush)
        out[key] = tr[["Session","R_multiple","PnL_$"]].copy() if not tr.empty else pd.DataFrame(columns=["Session","R_multiple","PnL_$"])
    return out

# =============================
# Extra tests (heatmap, slippage/fee, permutation, bootstrap, replay)
# =============================

def parameter_heatmap(folder: str, base_key: str) -> pd.DataFrame:
    # parse base_key like "ema20-26_c0.0015"
    try:
        ema_part, c_part = base_key.split("_c")
        f, s = ema_part.replace("ema","").split("-")
        base_f, base_s, base_c = int(f), int(s), float(c_part)
    except Exception:
        base_f, base_s, base_c = EMA_FAST, EMA_SLOW, BREAKOUT_CUSHION_PCT

    combos = [(ef, es, cu) for ef in HEAT_EMA_FAST for es in HEAT_EMA_SLOW for cu in HEAT_CUSHION if ef < es]
    rows = []
    iterator = tqdm(combos, desc="Heatmap combos", unit="combo") if tqdm is not None else combos
    for ef, es, cu in iterator:
        tr = backtest_folder_for_params(folder, ef, es, cu)
        mb = metrics_basic(tr)
        rows.append(dict(EMA_FAST=ef, EMA_SLOW=es, CUSH=cu, **mb))
    df = pd.DataFrame(rows)
    df.to_csv("orb_param_heatmap.csv", index=False)
    return df

def slippage_fee_surface(folder: str, ema_f: int, ema_s: int, cush: float) -> pd.DataFrame:
    rows = []
    iterator = itertools.product(SLIPPAGE_SWEEP_BPS, FEE_SWEEP)
    for slp, fee in iterator:
        tr = backtest_folder_for_params(folder, ema_f, ema_s, cush, slippage_bps=slp, fees=fee)
        mb = metrics_basic(tr); adv = metrics_advanced(tr)
        rows.append(dict(Slippage_bps=slp, Fee=fee, **mb, **adv))
    df = pd.DataFrame(rows).sort_values(["Slippage_bps","Fee"])
    df.to_csv("orb_slippage_fee_surface.csv", index=False)
    return df

def permutation_null_test(trades: pd.DataFrame, n_iter:int = PERMUTATIONS, seed:int = PERM_SEED) -> Dict[str, Any]:
    """Block by Session: shuffle sign of session PnL to build null PF distribution."""
    _rng_seed(seed)
    ses = trades.groupby("Session")["PnL_$"].sum().sort_index()
    if ses.empty:
        return {"PF_real":0.0,"PF_null_mean":0.0,"PF_null_95pct":0.0}
    real_gp = ses[ses>0].sum(); real_gl = -ses[ses<0].sum()
    real_pf = (real_gp / real_gl) if real_gl>1e-12 else (np.inf if real_gp>0 else 0.0)

    vals = ses.values.copy()
    null_pfs = []
    for _ in range(n_iter):
        signs = np.where(np.random.rand(len(vals))<0.5, 1.0, -1.0)
        perm = vals * signs
        gp = perm[perm>0].sum(); gl = -perm[perm<0].sum()
        pf = (gp / gl) if gl>1e-12 else (np.inf if gp>0 else 0.0)
        null_pfs.append(pf if np.isfinite(pf) else 1e9)
    null_pfs = np.array(null_pfs)
    out = {
        "PF_real": float(real_pf if np.isfinite(real_pf) else 1e9),
        "PF_null_mean": float(np.mean(null_pfs)),
        "PF_null_95pct": float(np.quantile(null_pfs, 0.95)),
    }
    with open("orb_permutation_test.json","w") as f:
        json.dump(out, f, indent=2)
    print("\n--- Permutation (null) test ---")
    print(f"Real PF: {out['PF_real']:.3f} | Null mean PF: {out['PF_null_mean']:.3f} | Null 95% PF: {out['PF_null_95pct']:.3f}")
    return out

def block_bootstrap_cis(trades: pd.DataFrame, iters:int=BOOTSTRAP_ITERS, block_len:int=BOOT_BLOCK_LEN, seed:int=123) -> Dict[str, Any]:
    """Moving-block bootstrap on session PnL to produce CIs for PF and Sharpe."""
    _rng_seed(seed)
    ses = trades.groupby("Session")["PnL_$"].sum().sort_index()
    if len(ses)==0:
        return {"PF_CI":[0,0], "Sharpe_CI":[0,0]}
    n = len(ses)
    idx = np.arange(n)
    vals = ses.values
    def _pf(x):
        gp = x[x>0].sum(); gl = -x[x<0].sum()
        return (gp/gl) if gl>1e-12 else (np.inf if gp>0 else 0.0)
    def _sh(x):
        r = x / max(POSITION_SIZE_DOLLARS, 1e-12)
        s = r.std(ddof=1)
        return float(np.sqrt(252) * r.mean() / s) if s>1e-12 else 0.0

    pfs, shs = [], []
    for _ in range(iters):
        # build boot sample by concatenating random blocks
        cur = []
        while len(cur) < n:
            start = np.random.randint(0, max(1, n - block_len + 1))
            cur.extend(vals[start:start+block_len])
        cur = np.array(cur[:n])
        pfv = _pf(cur)
        shv = _sh(cur)
        pfs.append(pfv if np.isfinite(pfv) else 1e9)
        shs.append(shv)
    pf_ci = [float(np.quantile(pfs, 0.025)), float(np.quantile(pfs, 0.975))]
    sh_ci = [float(np.quantile(shs, 0.025)), float(np.quantile(shs, 0.975))]
    out = {"PF_CI": pf_ci, "Sharpe_CI": sh_ci}
    with open("orb_bootstrap_cis.json","w") as f:
        json.dump(out, f, indent=2)
    print("\n--- Block-bootstrap CIs ---")
    print(f"PF 95% CI: [{pf_ci[0]:.3f}, {pf_ci[1]:.3f}] | Sharpe 95% CI: [{sh_ci[0]:.3f}, {sh_ci[1]:.3f}]")
    return out

def portfolio_replay_with_constraints(trades: pd.DataFrame,
                                      max_positions:int = REPLAY_MAX_POSITIONS,
                                      per_name_cap:int = REPLAY_PER_NAME_CAP) -> Dict[str, Any]:
    """
    Approximate portfolio concurrency constraints by allowing at most K open positions
    and at most M per name. Trades are 'open' at EntryTime and 'close' at last exit timestamp.
    We drop trades that would violate caps (FIFO by EntryTime).
    """
    if trades is None or trades.empty:
        print("\n--- Portfolio replay: no trades ---")
        return {"kept_trades":0, "dropped_trades":0, "metrics":metrics_basic(trades)}

    # parse last exit time & build intervals
    def last_exit_ts(x):
        xs = x if isinstance(x, list) else []
        if not xs: return pd.NaT
        return max(pd.Timestamp(t[1]) for t in xs)

    df = trades.copy()
    # Exits were stringified in CSV flow; ensure list if present
    if isinstance(df["Exits"].iloc[0], str):
        def _parse(s: str):
            out=[]
            for part in s.split(";"):
                if not part: continue
                lab, ts, px, q = part.split("|")
                out.append((lab, pd.Timestamp(ts), float(px), float(q)))
            return out
        df["Exits"] = df["Exits"].apply(_parse)

    df["EndTime"] = df["Exits"].apply(last_exit_ts)
    df = df.sort_values("EntryTime").reset_index(drop=True)

    open_positions: List[tuple] = []  # (ticker, endtime)
    held_per_name: Dict[str, int] = {}
    kept = []
    dropped = 0

    for _, tr in df.iterrows():
        ent = pd.Timestamp(tr["EntryTime"])
        # purge closed
        open_positions = [(tic, et) for (tic, et) in open_positions if pd.Timestamp(et) > ent]
        held_per_name = {}
        for tic, et in open_positions:
            held_per_name[tic] = held_per_name.get(tic, 0) + 1

        can_open = (len(open_positions) < max_positions) and (held_per_name.get(tr["Ticker"], 0) < per_name_cap)
        if can_open:
            kept.append(tr)
            open_positions.append((tr["Ticker"], tr["EndTime"]))
        else:
            dropped += 1

    kept_df = pd.DataFrame(kept) if kept else pd.DataFrame(columns=trades.columns)
    m = metrics_basic(kept_df)
    print("\n--- Portfolio replay (constraints) ---")
    print(f"Kept trades: {len(kept_df):,} | Dropped: {dropped:,}")
    _print_metrics("Replay metrics", m)
    kept_df.to_csv("orb_trades_replay_constrained.csv", index=False)
    return {"kept_trades": len(kept_df), "dropped_trades": dropped, "metrics": m}

# =============================
# Main
# =============================

def main():
    _rng_seed(42)

    results: Dict[str, Any] = {}

    # ----- Baseline -----
    if RUN_BASELINE:
        trades, equity = run_folder(PARQUET_DIR)
        print_overall_totals(trades)

        if not trades.empty:
            trades.sort_values(["Ticker","Session"], inplace=True)
            trades.to_csv(TRADES_CSV_ALL, index=False)
            print(f"Saved trades: {TRADES_CSV_ALL}")

        if not equity.empty:
            equity.sort_values(["Ticker","Session"], inplace=True)
            equity.to_csv(EQUITY_CSV_ALL, index=False)
            print(f"Saved equity: {EQUITY_CSV_ALL}")

        # Per-ticker summary with PF + Sharpe/MaxDD/ETL
        if not trades.empty:
            summary = (
                trades.groupby("Ticker")["PnL_$"]
                    .agg(['count', 'sum', 'mean', 'std'])
                    .rename(columns={'count':'Trades','sum':'PnL_sum','mean':'AvgPnL','std':'StdPnL'})
                    .reset_index()
                    .sort_values("PnL_sum", ascending=False)
            )
            wr = trades.groupby("Ticker")["PnL_$"].apply(lambda s: (pd.to_numeric(s, errors="coerce") > 0).mean()*100.0).rename("WinRate_%")
            pos = trades.groupby("Ticker")["PnL_$"].apply(lambda s: pd.to_numeric(s, errors="coerce").clip(lower=0).sum()).rename("GrossProfit")
            neg = trades.groupby("Ticker")["PnL_$"].apply(lambda s: -pd.to_numeric(s, errors="coerce").clip(upper=0).sum()).rename("GrossLoss")
            pf = (pos / neg.replace(0, np.nan)).rename("PF")
            summary = summary.merge(wr, on="Ticker", how="left").merge(pf, on="Ticker", how="left")
            summary.to_csv(SUMMARY_CSV, index=False)

        base_basic = metrics_basic(trades)
        base_adv   = metrics_advanced(trades)
        results["baseline"] = {"basic": base_basic, "advanced": base_adv}

    # ----- Purged K-Fold -----
    if RUN_PURGED_KFOLD and 'trades' in locals() and not trades.empty:
        kfold = purged_kfold_cv(trades, n_splits=KFOLD_SPLITS, purge_days=PURGE_DAYS, embargo_days=EMBARGO_DAYS)
        results["purged_kfold"] = kfold

    # ----- Walk-Forward -----
    if RUN_WALK_FORWARD and 'trades' in locals() and not trades.empty:
        wf = walk_forward_validation(trades, train_days=WF_TRAIN_DAYS, test_days=WF_TEST_DAYS)
        results["walk_forward"] = wf

    # ----- Nested WFO -----
    final_key = None
    last_wf_end = None
    if RUN_NESTED_WF:
        grid_returns = build_grid_returns(PARQUET_DIR)
        if grid_returns:
            wf_records, last_wf_end = nested_walk_forward_with_calibration(
                grid_returns,
                train_days=NWF_TRAIN_DAYS,
                calib_days=NWF_CALIB_DAYS,
                test_days=NWF_TEST_DAYS,
                purge_days=NWF_PURGE_DAYS,
                embargo_days=NWF_EMBARGO_DAYS,
                min_trades_calib=NWF_MIN_TRADES_CALIB
            )
            final_key = choose_final_param_from_wf(wf_records)
            results["nested_wfo"] = {"records": wf_records, "final_key": final_key}
            if final_key is not None and final_key in grid_returns and last_wf_end is not None:
                df_all = grid_returns[final_key]
                ses_idx = session_series(df_all)
                holdout = df_all.loc[ses_idx > last_wf_end]
                results["final_holdout"] = {"after": _fmt(last_wf_end), "metrics": metrics_basic(holdout)}

    # ----- Parameter heatmap around final key -----
    if final_key:
        heat = parameter_heatmap(PARQUET_DIR, final_key)
        results["param_heatmap_csv"] = "orb_param_heatmap.csv"

        # also store the top few combos in-memory
        top5 = heat.sort_values("PF", ascending=False).head(5).to_dict(orient="records")
        results["param_heatmap_top5"] = top5

        # for slippage/fee surface, pick parsed params from key
        ema_part, c_part = final_key.split("_c")
        ef, es = [int(x) for x in ema_part.replace("ema","").split("-")]
        cu = float(c_part)
        sf = slippage_fee_surface(PARQUET_DIR, ef, es, cu)
        results["slippage_fee_surface_csv"] = "orb_slippage_fee_surface.csv"

    # ----- Permutation (null) test -----
    if 'trades' in locals() and not trades.empty:
        perm = permutation_null_test(trades, n_iter=PERMUTATIONS, seed=PERM_SEED)
        results["permutation_test"] = perm

        cis = block_bootstrap_cis(trades, iters=BOOTSTRAP_ITERS, block_len=BOOT_BLOCK_LEN, seed=123)
        results["bootstrap_cis"] = cis

        # ----- Portfolio replay with constraints -----
        replay = portfolio_replay_with_constraints(trades, max_positions=REPLAY_MAX_POSITIONS, per_name_cap=REPLAY_PER_NAME_CAP)
        results["portfolio_replay"] = replay

    # Save consolidated results
    with open(RESULTS_JSON, "w") as f:
        json.dump(results, f, indent=2, default=str)
    print(f"\nSaved consolidated results to {RESULTS_JSON}")

if __name__ == "__main__":
    main()
```
