```python
import pandas as pd
import osmnx as ox
from geopy.geocoders import Nominatim
from shapely.geometry import Point
import time

# === Load your CSV ===
file_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(file_path)

# === Create full address column ===
df['full_address'] = df['address'].astype(str) + ', ' + df['city'].astype(str) + ', ' + df['state'].astype(str) + ' ' + df['zip'].astype(str)

# === Geocoder setup ===
geolocator = Nominatim(user_agent="lease_locator")

def geocode_address(addr):
    try:
        location = geolocator.geocode(addr, timeout=10)
        if location:
            return pd.Series([location.latitude, location.longitude])
    except Exception as e:
        return pd.Series([None, None])
    return pd.Series([None, None])

# === Geocode all addresses ===
print("📍 Geocoding lease addresses...")
df[['latitude', 'longitude']] = df['full_address'].apply(lambda x: geocode_address(x))
time.sleep(1)  # Avoid hitting API too fast

# === Define highway tags for OSM ===
tags_highway = {'highway': ['motorway', 'motorway_link', 'trunk', 'trunk_link']}

# === Find closest highway distance ===
def get_closest_highway_distance(lat, lon, radius=10000):
    try:
        point = Point(lon, lat)
        highways = ox.geometries_from_point((lat, lon), tags=tags_highway, dist=radius)
        if highways.empty:
            return None
        highway_geoms = highways.geometry.dropna()
        distances = highway_geoms.apply(lambda geom: geom.distance(point))
        return distances.min()
    except Exception as e:
        print(f"❌ Error @ ({lat},{lon}): {e}")
        return None

# === Apply distance function ===
print("📏 Calculating closest highway distance...")
df['closest_highway_m'] = df.apply(
    lambda row: get_closest_highway_distance(row['latitude'], row['longitude']) 
    if pd.notnull(row['latitude']) and pd.notnull(row['longitude']) else None,
    axis=1
)

# === Save result ===
output_path = r"C:\Users\pcagm\Downloads\Leases_with_Highway.csv"
df.to_csv(output_path, index=False)

print(f"✅ Saved with `closest_highway_m` column to:\n{output_path}")
```

```python
import pandas as pd
import osmnx as ox
from geopy.geocoders import Nominatim
from shapely.geometry import Point
import time

# === Load lease file ===
print("📂 Loading lease file...")
file_path = r"C:\Users\pcagm\Downloads\Leases.csv"
df = pd.read_csv(file_path)
print(f"✅ Loaded {len(df)} rows from {file_path}")

# === Format ZIP properly ===
print("📍 Fixing ZIP code formatting...")
df['zip_str'] = df['zip'].fillna(0).astype(int).astype(str).str.zfill(5)

# === Build full address string for geocoding ===
print("📍 Building full address strings...")
df['full_address'] = df['address'].astype(str) + ', ' + df['city'].astype(str) + ', ' + df['state'].astype(str) + ' ' + df['zip_str']

# === Setup geocoder ===
geolocator = Nominatim(user_agent="lease_locator")
print("🛰️ Geocoder initialized (Nominatim)")

# === Geocoding function ===
def geocode_address(addr, idx):
    try:
        print(f"🔎 [{idx}] Geocoding: {addr}")
        location = geolocator.geocode(addr, timeout=10)
        if location:
            print(f"   ✅ Coordinates: ({location.latitude}, {location.longitude})")
            return pd.Series([location.latitude, location.longitude])
        else:
            print(f"   ❌ Address not found.")
            return pd.Series([None, None])
    except Exception as e:
        print(f"   ❌ Error geocoding row {idx}: {e}")
        return pd.Series([None, None])

# === Wrapper to apply geocoding per row ===
def geocode_row(row):
    return geocode_address(row['full_address'], row.name)

print("🚀 Starting geocoding process...")
df[['latitude', 'longitude']] = df.apply(geocode_row, axis=1)

# === Define highway tags for OpenStreetMap ===
tags_highway = {'highway': ['motorway', 'motorway_link', 'trunk', 'trunk_link']}
print("🛣️ Highway tags defined")

# === Closest highway calculation ===
def get_closest_highway_distance(lat, lon, row_index, radius=10000):
    try:
        point = Point(lon, lat)
        print(f"🌍 [{row_index}] Searching highways within {radius}m of ({lat}, {lon})...")
        highways = ox.geometries_from_point((lat, lon), tags=tags_highway, dist=radius)
        if highways.empty:
            print("   ⚠️ No highways found.")
            return None
        highway_geoms = highways.geometry.dropna()
        distances = highway_geoms.apply(lambda geom: geom.distance(point))
        min_dist = distances.min()
        print(f"   ✅ Closest highway: {min_dist:.2f} meters")
        return min_dist
    except Exception as e:
        print(f"   ❌ Error @ row {row_index}: {e}")
        return None

# === Apply highway distance function ===
print("📏 Starting closest highway distance calculations...")
df['closest_highway_m'] = df.apply(
    lambda row: get_closest_highway_distance(row['latitude'], row['longitude'], row.name)
    if pd.notnull(row['latitude']) and pd.notnull(row['longitude']) else None,
    axis=1
)

# === Save result to CSV ===
output_path = r"C:\Users\pcagm\Downloads\Leases_with_Highway.csv"
df.to_csv(output_path, index=False)
print(f"\n💾 Done! Output saved with `closest_highway_m` column:\n{output_path}")
```
