```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# URL to scrape
url = "https://cfbstats.com/2024/player/29/1121315/rushing/gamelog.html"

def fetch_player_name(soup):
    """Extract the player's name from the page title or header."""
    # Find the player's name from a page heading or title if available
    title_tag = soup.find('h2')
    if title_tag:
        return title_tag.text.strip()
    
    # Fallback: Extract from the page title if the name is present there
    title = soup.title.text if soup.title else "Unknown Player"
    return title.split(" - ")[0] if " - " in title else title

def fetch_game_log(url):
    """Fetch the game log table and the player's name."""
    response = requests.get(url)
    if response.status_code != 200:
        print("Failed to retrieve the webpage")
        return None, None
    
    soup = BeautifulSoup(response.content, 'html.parser')
    player_name = fetch_player_name(soup)
    
    # Locate the game log table
    table = soup.find('table')
    if not table:
        print("Table not found.")
        return None, player_name
    
    # Extract headers
    headers = [header.text.strip() for header in table.find_all('th')]
    
    # Extract rows
    data = []
    rows = table.find_all('tr')[1:]
    for row in rows:
        cols = [col.text.strip() for col in row.find_all('td')]
        if cols:
            data.append(cols)
    
    # Convert to DataFrame
    df = pd.DataFrame(data, columns=headers)
    return df, player_name

def filter_kansas_state_games(df):
    """Filter rows where the opponent contains 'Kansas St.'."""
    return df[df['Opponent'].str.contains("Kansas St.", case=False, na=False)]

def main():
    # Fetch game log and player name
    game_log_df, player_name = fetch_game_log(url)
    
    if game_log_df is not None:
        print(f"Player Name: {player_name}")
        
        # Filter for games against Kansas State
        filtered_df = filter_kansas_state_games(game_log_df)
        
        # Print the filtered DataFrame
        print("Filtered Game Log for Kansas State:")
        print(filtered_df)
        
        # Save the data to CSV
        file_name = f"{player_name.replace(' ', '_')}_kansas_state_game_log.csv"
        filtered_df.to_csv(file_name, index=False)
        print(f"Data saved to {file_name}")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Step 1: Fetch the overall stats dataset
def fetch_overall_stats(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    
    # Find the table in the webpage
    table = soup.find('table')
    if not table:
        print("Table not found.")
        return None
    
    # Extract rows from the table
    rows = table.find_all('tr')
    
    # Extract headers
    headers = [header.text.strip() for header in rows[0].find_all('th')]
    
    # Extract data
    data = []
    for row in rows[1:]:
        cols = [col.text.strip() for col in row.find_all('td')]
        if cols:
            data.append(cols)
    
    # Convert to DataFrame
    df = pd.DataFrame(data, columns=headers)
    return df

# Step 2: Load the filtered game log dataset (already saved)
def load_filtered_game_log(filepath):
    return pd.read_csv(filepath)

# Step 3: Calculate percentage change
def calculate_percentage_change(overall_df, game_log_df):
    results = []
    
    # Iterate through each player in the game log dataset
    for _, game_row in game_log_df.iterrows():
        player_name = game_row['Player']
        team_name = game_row['Team']
        
        # Filter out rows with missing or invalid data
        if game_row['Att'] == '-' or game_row['Yards'] == '-' or game_row['Avg.'] == '-':
            continue
        
        # Find the player's overall stats
        overall_row = overall_df[overall_df['Name'].str.contains(player_name, case=False, na=False)]
        if overall_row.empty:
            continue
        
        # Extract relevant stats
        overall_att_g = float(overall_row['Att/G'].values[0])
        overall_yards_g = float(overall_row['Yards/G'].values[0])
        overall_avg = float(overall_row['Avg.'].values[0])
        
        # Extract game stats and convert to numeric
        try:
            game_att = int(game_row['Att'])
            game_yards = int(game_row['Yards'])
            game_avg = float(game_row['Avg.'])
        except ValueError:
            # Skip if conversion fails
            continue
        
        # Calculate percentage change
        change_att = ((game_att - overall_att_g) / overall_att_g) * 100 if overall_att_g != 0 else 0
        change_yards = ((game_yards - overall_yards_g) / overall_yards_g) * 100 if overall_yards_g != 0 else 0
        change_avg = ((game_avg - overall_avg) / overall_avg) * 100 if overall_avg != 0 else 0
        
        # Append results
        results.append({
            'Team': team_name,
            'Player': player_name,
            'Overall Att/G': overall_att_g,
            'Game Att': game_att,
            '% Change Att': change_att,
            'Overall Yards/G': overall_yards_g,
            'Game Yards': game_yards,
            '% Change Yards': change_yards,
            'Overall Avg.': overall_avg,
            'Game Avg.': game_avg,
            '% Change Avg': change_avg
        })
    
    # Convert results to DataFrame
    result_df = pd.DataFrame(results)
    return result_df

def main():
    # URL to fetch overall stats
    url = "https://cfbstats.com/2024/leader/25354/player/split01/category01/sort01.html"
    
    # Step 1: Fetch the overall stats dataset
    overall_df = fetch_overall_stats(url)
    if overall_df is None:
        return
    
    # Step 2: Load the filtered game log dataset
    game_log_df = load_filtered_game_log("combined_game_logs_vs_kansas_state.csv")
    
    # Step 3: Calculate percentage change
    comparison_df = calculate_percentage_change(overall_df, game_log_df)
    
    # Step 4: Display and save the results
    print("Comparison of Player Performance vs Kansas State:")
    print(comparison_df)
    
    # Save to CSV
    comparison_df.to_csv("player_performance_comparison.csv", index=False)
    print("Data saved to player_performance_comparison.csv")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Step 1: Fetch the overall stats dataset
def fetch_overall_stats(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    
    # Find the table in the webpage
    table = soup.find('table')
    if not table:
        print("Table not found.")
        return None
    
    # Extract rows from the table
    rows = table.find_all('tr')
    
    # Extract headers
    headers = [header.text.strip() for header in rows[0].find_all('th')]
    
    # Extract data
    data = []
    for row in rows[1:]:
        cols = [col.text.strip() for col in row.find_all('td')]
        if cols:
            data.append(cols)
    
    # Convert to DataFrame
    df = pd.DataFrame(data, columns=headers)
    return df

# Step 2: Load the filtered game log dataset (already saved)
def load_filtered_game_log(filepath):
    return pd.read_csv(filepath)

# Step 3: Calculate differences in performance
def calculate_performance_comparison(overall_df, game_log_df):
    results = []
    
    # Iterate through each player in the game log dataset
    for _, game_row in game_log_df.iterrows():
        player_name = game_row['Player']
        team_name = game_row['Team']
        
        # Filter out rows with missing or invalid data
        if game_row['Att'] == '-' or game_row['Yards'] == '-' or game_row['Avg.'] == '-':
            continue
        
        # Find the player's overall stats
        overall_row = overall_df[overall_df['Name'].str.contains(player_name, case=False, na=False)]
        if overall_row.empty:
            continue
        
        # Extract relevant stats
        overall_att_g = float(overall_row['Att/G'].values[0])
        overall_yards_g = float(overall_row['Yards/G'].values[0])
        overall_avg = float(overall_row['Avg.'].values[0])
        
        # Extract game stats and convert to numeric
        try:
            game_att = int(game_row['Att'])
            game_yards = int(game_row['Yards'])
            game_avg = float(game_row['Avg.'])
        except ValueError:
            continue
        
        # Calculate differences
        att_diff = game_att - overall_att_g
        yards_diff = game_yards - overall_yards_g
        avg_diff = game_avg - overall_avg
        
        # Append results
        results.append({
            'Team': team_name,
            'Player': player_name,
            'Overall Att/G': overall_att_g,
            'Game Att': game_att,
            'Att Diff': att_diff,
            'Overall Yards/G': overall_yards_g,
            'Game Yards': game_yards,
            'Yards Diff': yards_diff,
            'Overall Avg.': overall_avg,
            'Game Avg.': game_avg,
            'Avg Diff': avg_diff
        })
    
    # Convert results to DataFrame
    result_df = pd.DataFrame(results)
    return result_df

def main():
    # URL to fetch overall stats
    url = "https://cfbstats.com/2024/leader/25354/player/split01/category01/sort01.html"
    
    # Step 1: Fetch the overall stats dataset
    overall_df = fetch_overall_stats(url)
    if overall_df is None:
        return
    
    # Step 2: Load the filtered game log dataset
    game_log_df = load_filtered_game_log("combined_game_logs_vs_kansas_state.csv")
    
    # Step 3: Calculate performance comparison
    comparison_df = calculate_performance_comparison(overall_df, game_log_df)
    
    # Step 4: Display and save the results
    print("Comparison of Player Performance vs Kansas State:")
    print(comparison_df)
    
    # Save to CSV
    comparison_df.to_csv("player_performance_comparison_with_diffs.csv", index=False)
    print("Data saved to player_performance_comparison_with_diffs.csv")

if __name__ == "__main__":
    main()
```
