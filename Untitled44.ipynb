```python
# orb_15m_retest_all_parquet_RR_1_to_2.py
# Runs the same ORB 15m Retest (1 share, RR 1:2) across ALL parquet files in PARQUET_DIR.
# Prints a per-symbol line in your format and an OVERALL line; saves a CSV per symbol and an ALL CSV.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"  # <— your folder
TZ_LOCAL = "America/New_York"

REG_START = "09:30"   # session start
OR_END    = "09:45"   # end of first 15-min bar (exclusive)
REG_END   = "16:00"   # session end

STOP_FIRST_IF_BOTH_HIT = True
SAVE_DIR = Path(".")  # where to write CSVs (change if you prefer)

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    """
    Robustly parse EventAt:
    - If tz-naive: assume UTC then convert to New York.
    - If tz-aware: convert to New York.
    """
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    """Return ('long'/'short', idx) for the first bar *closing* outside the OR."""
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    """First bar that *touches* the level after the breakout close."""
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def simulate_path(after_entry_df, side, entry, stop, target, stop_first=True):
    """
    Walk forward bar-by-bar after the entry.
    Returns outcome ('TP'/'STOP'/'EOD'), exit_price, exit_idx.
    """
    for idx, row in after_entry_df.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop and hit_tp:
            return ("STOP", stop, idx) if stop_first else ("TP", target, idx)
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    # neither hit by session end → exit at last close
    last_idx = after_entry_df.index[-1]
    return "EOD", float(after_entry_df.loc[last_idx, "Close"]), last_idx

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

def calc_metrics(trades_df):
    pnl = trades_df["PnL_1to2"].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= CORE BACKTEST (unchanged logic) =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Keep only regular session minutes
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    records = []

    for d, day in df.groupby("date"):
        # Opening range: first 15-minute bar [09:30, 09:45)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # After OR window (from 09:45 to close)
        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        # 1) Wait for first confirmed breakout by CLOSE
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # 2) After that breakout close, place retest entry at the broken level
        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        # 3) Wait for the first bar that *touches* the broken level → entry
        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop       = orl if side == "long" else orh
        target     = entry + ( 2 * or_width if side == "long" else -2 * or_width)

        # Forward simulate after the entry bar
        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            records.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": or_width,
                "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
                "Outcome_1to2": "EOD", "ExitPx_1to2": entry, "ExitTime_1to2": entry_time,
                "PnL_1to2": 0.0,
            })
            continue

        outcome, exit_px, exit_idx = simulate_path(
            after_entry, side, entry, stop, target, stop_first=STOP_FIRST_IF_BOTH_HIT
        )
        pnl = pnl_one_share(side, entry, exit_px)

        records.append({
            "Date": d, "Symbol": symbol, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": or_width,
            "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
            "Outcome_1to2": outcome, "ExitPx_1to2": exit_px,
            "ExitTime_1to2": day.loc[exit_idx, "EventAt"] if exit_idx in day.index else None,
            "PnL_1to2": pnl,
        })

    return pd.DataFrame(records)

# ========= RUN ALL FILES =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    all_trades = []

    for f in files:
        symbol = f.stem.upper()  # e.g., SPY.parquet -> SPY
        try:
            trades = backtest_one_symbol(f, symbol)
        except Exception as e:
            print(f"[SKIP] {symbol}: {e}")
            continue

        if trades.empty:
            print(f"=== ORB 15m Retest — {symbol} (1 share, RR 1:2) ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            continue

        all_trades.append(trades)

        # Per-symbol metrics + print in required format
        total_trades, total_pnl, winrate, profit_factor = calc_metrics(trades)

        save_path = SAVE_DIR / f"orb_trades_{symbol}_one_share_RR1to2.csv"
        trades.to_csv(save_path, index=False)

        print(f"=== ORB 15m Retest — {symbol} (1 share, RR 1:2) ===")
        print(f"Trades: {total_trades} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {profit_factor:.3f}")
        print(f"Saved trades → {save_path.resolve()}")

    # ===== OVERALL roll-up =====
    if not all_trades:
        print("\nNo trades generated for any symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    o_trades, o_pnl, o_winrate, o_pf = calc_metrics(overall)

    save_all = SAVE_DIR / "orb_trades_ALL_one_share_RR1to2.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== ORB 15m Retest — OVERALL (1 share, RR 1:2) ===")
    print(f"Trades: {o_trades} | Total PnL: ${o_pnl:,.2f} | Win%: {o_winrate:.2f}% | Profit Factor: {o_pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_all_parquet_RR_1_to_2_quant.py
# Runs the same ORB 15m Retest (1 share, RR 1:2) across ALL parquet files in PARQUET_DIR.
# Prints your original per-symbol line + OVERALL line, and also prints a "quant-standard"
# risk block per symbol and overall:
#   Sharpe(daily), Sortino(daily), MaxDD, Calmar, t-stat(mean daily PnL),
#   CVaR@5% / @1% (daily), skew, excess kurtosis.
#
# NOTE: Strategy logic is unchanged.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path
from math import sqrt, isfinite

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"  # <— your folder
TZ_LOCAL = "America/New_York"

REG_START = "09:30"   # session start
OR_END    = "09:45"   # end of first 15-min bar (exclusive)
REG_END   = "16:00"   # session end

STOP_FIRST_IF_BOTH_HIT = True
SAVE_DIR = Path(".")  # where to write CSVs (change if you prefer)

TRADING_DAYS_PER_YEAR = 252  # for annualizations on *daily* PnL

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    """
    Robustly parse EventAt:
    - If tz-naive: assume UTC then convert to New York.
    - If tz-aware: convert to New York.
    """
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= STRATEGY HELPERS =========
def first_breakout_close(df_after_or, orh, orl):
    """Return ('long'/'short', idx) for the first bar *closing* outside the OR."""
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    """First bar that *touches* the level after the breakout close."""
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def simulate_path(after_entry_df, side, entry, stop, target, stop_first=True):
    """
    Walk forward bar-by-bar after the entry.
    Returns outcome ('TP'/'STOP'/'EOD'), exit_price, exit_idx.
    """
    for idx, row in after_entry_df.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop and hit_tp:
            return ("STOP", stop, idx) if stop_first else ("TP", target, idx)
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    # neither hit by session end → exit at last close
    last_idx = after_entry_df.index[-1]
    return "EOD", float(after_entry_df.loc[last_idx, "Close"]), last_idx

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

def calc_basic_metrics(trades_df):
    """Your original per-trade metrics."""
    pnl = trades_df["PnL_1to2"].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

# ========= QUANT-STANDARD INDICATORS (daily PnL) =========
def _max_drawdown_from_series(equity: pd.Series) -> float:
    """Max drawdown in *dollars* from an equity curve (running sum)."""
    roll_max = equity.cummax()
    drawdown = equity - roll_max
    return float(-drawdown.min()) if len(equity) else 0.0

def _sortino_ratio(daily_pnl: pd.Series) -> float:
    if daily_pnl.empty:
        return 0.0
    mean = daily_pnl.mean()
    downside = daily_pnl[daily_pnl < 0]
    dstd = downside.std(ddof=1)
    if dstd == 0 or np.isnan(dstd):
        return np.inf if mean > 0 else 0.0
    return float((mean / dstd) * sqrt(TRADING_DAYS_PER_YEAR))

def _sharpe_ratio(daily_pnl: pd.Series) -> float:
    if daily_pnl.empty:
        return 0.0
    mu = daily_pnl.mean()
    sd = daily_pnl.std(ddof=1)
    if sd == 0 or np.isnan(sd):
        return np.inf if mu > 0 else 0.0
    return float((mu / sd) * sqrt(TRADING_DAYS_PER_YEAR))

def _t_stat_of_mean(daily_pnl: pd.Series) -> float:
    n = len(daily_pnl)
    if n < 2:
        return 0.0
    mu = daily_pnl.mean()
    sd = daily_pnl.std(ddof=1)
    if sd == 0 or np.isnan(sd):
        return np.inf if mu > 0 else 0.0
    return float(mu / (sd / sqrt(n)))

def _cvar(daily_pnl: pd.Series, alpha: float = 0.05) -> float:
    """Expected shortfall (mean of worst alpha tail). Returns negative number if losses dominate."""
    if daily_pnl.empty:
        return 0.0
    q = daily_pnl.quantile(alpha)
    tail = daily_pnl[daily_pnl <= q]
    if tail.empty:
        return 0.0
    return float(tail.mean())

def quant_indicators(trades_df: pd.DataFrame):
    """
    Compute 'industry-standard' indicators on *daily aggregated PnL* from trades_df.
    Returns a dict with metrics.
    """
    if trades_df.empty:
        return {
            "Sharpe_daily": 0.0, "Sortino_daily": 0.0, "MaxDD_$": 0.0, "Calmar": 0.0,
            "t_stat_daily_mean": 0.0, "CVaR_5%": 0.0, "CVaR_1%": 0.0,
            "Skew_daily": 0.0, "ExKurt_daily": 0.0
        }

    # Aggregate to daily PnL in dollars
    daily = trades_df.groupby("Date", as_index=True)["PnL_1to2"].sum().sort_index()

    # Risk/quality metrics
    sharpe = _sharpe_ratio(daily)
    sortino = _sortino_ratio(daily)

    equity = daily.cumsum()
    maxdd = _max_drawdown_from_series(equity)

    # Annualized PnL (dollar drift) and Calmar
    ann_pnl = float(daily.mean() * TRADING_DAYS_PER_YEAR)
    calmar = (ann_pnl / maxdd) if maxdd > 0 else np.inf

    t_stat = _t_stat_of_mean(daily)
    cvar5 = _cvar(daily, 0.05)
    cvar1 = _cvar(daily, 0.01)

    skew = float(daily.skew()) if len(daily) > 2 else 0.0
    exkurt = float(daily.kurt()) if len(daily) > 3 else 0.0  # excess kurtosis (Fisher)

    return {
        "Sharpe_daily": float(sharpe if isfinite(sharpe) else 0.0),
        "Sortino_daily": float(sortino if isfinite(sortino) else 0.0),
        "MaxDD_$": float(maxdd),
        "Calmar": float(calmar if isfinite(calmar) else 0.0),
        "t_stat_daily_mean": float(t_stat if isfinite(t_stat) else 0.0),
        "CVaR_5%": float(cvar5),
        "CVaR_1%": float(cvar1),
        "Skew_daily": float(skew),
        "ExKurt_daily": float(exkurt),
    }

# ========= CORE BACKTEST (unchanged logic) =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    # Keep only regular session minutes
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    records = []

    for d, day in df.groupby("date"):
        # Opening range: first 15-minute bar [09:30, 09:45)
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        # After OR window (from 09:45 to close)
        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        # 1) Wait for first confirmed breakout by CLOSE
        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        # 2) After that breakout close, place retest entry at the broken level
        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        # 3) Wait for the first bar that *touches* the broken level → entry
        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop       = orl if side == "long" else orh
        target     = entry + ( 2 * or_width if side == "long" else -2 * or_width)

        # Forward simulate after the entry bar
        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            records.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": or_width,
                "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
                "Outcome_1to2": "EOD", "ExitPx_1to2": entry, "ExitTime_1to2": entry_time,
                "PnL_1to2": 0.0,
            })
            continue

        outcome, exit_px, exit_idx = simulate_path(
            after_entry, side, entry, stop, target, stop_first=STOP_FIRST_IF_BOTH_HIT
        )
        pnl = pnl_one_share(side, entry, exit_px)

        records.append({
            "Date": d, "Symbol": symbol, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": or_width,
            "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
            "Outcome_1to2": outcome, "ExitPx_1to2": exit_px,
            "ExitTime_1to2": day.loc[exit_idx, "EventAt"] if exit_idx in day.index else None,
            "PnL_1to2": pnl,
        })

    return pd.DataFrame(records)

# ========= RUN ALL FILES =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    SAVE_DIR.mkdir(parents=True, exist_ok=True)

    all_trades = []
    # optional: keep per-symbol quant summary for a roll-up csv
    metrics_rows = []

    for f in files:
        symbol = f.stem.upper()  # e.g., SPY.parquet -> SPY
        try:
            trades = backtest_one_symbol(f, symbol)
        except Exception as e:
            print(f"[SKIP] {symbol}: {e}")
            continue

        if trades.empty:
            print(f"=== ORB 15m Retest — {symbol} (1 share, RR 1:2) ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            print("— Risk & Quality (daily) — Sharpe: 0.00 | Sortino: 0.00 | MaxDD: $0.00 | Calmar: 0.00 | t-stat: 0.00 | CVaR5%: $0.00 | CVaR1%: $0.00 | Skew: 0.00 | ExKurt: 0.00")
            continue

        all_trades.append(trades)

        # Per-symbol metrics + print in required format
        total_trades, total_pnl, winrate, profit_factor = calc_basic_metrics(trades)

        save_path = SAVE_DIR / f"orb_trades_{symbol}_one_share_RR1to2.csv"
        trades.to_csv(save_path, index=False)

        print(f"=== ORB 15m Retest — {symbol} (1 share, RR 1:2) ===")
        print(f"Trades: {total_trades} | Total PnL: ${total_pnl:,.2f} | Win%: {winrate:.2f}% | Profit Factor: {profit_factor:.3f}")
        print(f"Saved trades → {save_path.resolve()}")

        # Quant-standard indicators (daily)
        qi = quant_indicators(trades)
        print(
            "— Risk & Quality (daily) — "
            f"Sharpe: {qi['Sharpe_daily']:.3f} | Sortino: {qi['Sortino_daily']:.3f} | "
            f"MaxDD: ${qi['MaxDD_$']:,.2f} | Calmar: {qi['Calmar']:.3f} | "
            f"t-stat: {qi['t_stat_daily_mean']:.3f} | "
            f"CVaR5%: ${qi['CVaR_5%']:,.2f} | CVaR1%: ${qi['CVaR_1%']:,.2f} | "
            f"Skew: {qi['Skew_daily']:.3f} | ExKurt: {qi['ExKurt_daily']:.3f}"
        )

        # for roll-up summary file
        metrics_rows.append({
            "Symbol": symbol,
            "Trades": total_trades,
            "TotalPnL_$": total_pnl,
            "WinRate_%": winrate,
            "ProfitFactor": profit_factor,
            **qi
        })

    # ===== OVERALL roll-up =====
    if not all_trades:
        print("\nNo trades generated for any symbols.")
        return

    overall = pd.concat(all_trades, ignore_index=True)
    o_trades, o_pnl, o_winrate, o_pf = calc_basic_metrics(overall)

    save_all = SAVE_DIR / "orb_trades_ALL_one_share_RR1to2.csv"
    overall.to_csv(save_all, index=False)

    print(f"\n=== ORB 15m Retest — OVERALL (1 share, RR 1:2) ===")
    print(f"Trades: {o_trades} | Total PnL: ${o_pnl:,.2f} | Win%: {o_winrate:.2f}% | Profit Factor: {o_pf:.3f}")
    print(f"Saved trades → {save_all.resolve()}")

    # Overall quant-standard block
    oq = quant_indicators(overall)
    print(
        "— Risk & Quality (daily) — "
        f"Sharpe: {oq['Sharpe_daily']:.3f} | Sortino: {oq['Sortino_daily']:.3f} | "
        f"MaxDD: ${oq['MaxDD_$']:,.2f} | Calmar: {oq['Calmar']:.3f} | "
        f"t-stat: {oq['t_stat_daily_mean']:.3f} | "
        f"CVaR5%: ${oq['CVaR_5%']:,.2f} | CVaR1%: ${oq['CVaR_1%']:,.2f} | "
        f"Skew: {oq['Skew_daily']:.3f} | ExKurt: {oq['ExKurt_daily']:.3f}"
    )

    # Optional: write a concise per-symbol summary with the quant indicators
    if metrics_rows:
        df_metrics = pd.DataFrame(metrics_rows)
        df_metrics.sort_values("TotalPnL_$", ascending=False, inplace=True)
        summary_path = SAVE_DIR / "orb_metrics_summary.csv"
        df_metrics.to_csv(summary_path, index=False)
        print(f"Saved metrics summary → {summary_path.resolve()}")

if __name__ == "__main__":
    main()
```

```python
# orb_15m_retest_all_parquet_RR_1_to_2_PROD.py
# Same ORB 15m Retest logic (1 share, RR 1:2) + production-grade portfolio controls:
#  - Per-symbol Regime Gate: trade only on days where prior rolling Sharpe > 0
#  - Cross-sectional Top-K selection by prior rolling Sharpe
#  - Optional volatility targeting (per-symbol, per-day) with caps
#
# Prints baseline metrics (ALL trades) AND "Production" metrics (GATED/Top-K/Vol-targeted).
# Strategy entry/exit rules are unchanged.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path
from math import sqrt, isfinite

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"  # your folder
TZ_LOCAL = "America/New_York"

REG_START = "09:30"
OR_END    = "09:45"
REG_END   = "16:00"

STOP_FIRST_IF_BOTH_HIT = True
SAVE_DIR = Path(".")

# ---- Production controls (tune lightly; defaults are conservative) ----
ROLL_DAYS          = 60     # lookback for rolling Sharpe on daily PnL
SHARPE_GATE        = 0.0    # require prior rolling Sharpe > this
TOP_K_PER_DAY      = 10     # trade only top-K symbols by prior rolling Sharpe each day
USE_VOL_TARGETING  = True   # per-symbol per-day size scaling by prior vol
TARGET_DOLLAR_VOL  = 300.0  # desired daily PnL volatility ($) per symbol (pre-cap)
SIZE_MULT_CAP      = 4.0    # cap multiplier to avoid extreme leverage
TRADING_DAYS_PER_YEAR = 252

# (Costs kept at zero by default so you can compare pure alpha; enable later if desired)
COMM_PER_SHARE = 0.00
SPREAD_BPS     = 0.0
ENTRY_CROSS    = False
EXIT_CROSS     = False

# ========= TIME HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

# ========= STRATEGY HELPERS (unchanged) =========
def first_breakout_close(df_after_or, orh, orl):
    above = df_after_or["Close"] > orh
    below = df_after_or["Close"] < orl
    i_long  = df_after_or.index[above].min()  if above.any() else None
    i_short = df_after_or.index[below].min()  if below.any() else None
    if i_long is None and i_short is None:
        return None, None
    if i_long is not None and (i_short is None or i_long <= i_short):
        return "long", i_long
    return "short", i_short

def first_retest_touch(df_after_breakout_close, level):
    touched = (df_after_breakout_close["Low"] <= level) & (df_after_breakout_close["High"] >= level)
    if not touched.any():
        return None
    return df_after_breakout_close.index[touched].min()

def simulate_path(after_entry_df, side, entry, stop, target, stop_first=True):
    for idx, row in after_entry_df.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop and hit_tp:
            return ("STOP", stop, idx) if stop_first else ("TP", target, idx)
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    last_idx = after_entry_df.index[-1]
    return "EOD", float(after_entry_df.loc[last_idx, "Close"]), last_idx

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

def apply_costs(entry, exit_px, shares=1):
    spread_cost = 0.0
    if ENTRY_CROSS:
        spread_cost += (SPREAD_BPS/1e4) * entry
    if EXIT_CROSS:
        spread_cost += (SPREAD_BPS/1e4) * exit_px
    fees = COMM_PER_SHARE * 2 * shares
    return spread_cost * shares + fees

# ========= METRICS =========
def calc_metrics(trades_df, col="PnL_1to2"):
    pnl = trades_df[col].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total_trades = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    profit_factor = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total_trades) if total_trades else 0.0
    return total_trades, total_pnl, winrate, profit_factor

def rolling_sharpe(series, lookback=ROLL_DAYS):
    mu = series.rolling(lookback, min_periods=lookback).mean()
    sd = series.rolling(lookback, min_periods=lookback).std(ddof=1)
    sharpe = (mu / sd) * np.sqrt(TRADING_DAYS_PER_YEAR)
    return sharpe.replace([np.inf, -np.inf], np.nan)

# ========= CORE BACKTEST (unchanged logic) =========
def backtest_one_symbol(parquet_path: Path, symbol: str) -> pd.DataFrame:
    df = pd.read_parquet(parquet_path)
    for c in ["EventAt", "Open", "High", "Low", "Close"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")
    df["EventAt"] = to_ny(df, "EventAt")
    df = df.sort_values("EventAt").reset_index(drop=True)

    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    records = []

    for d, day in df.groupby("date"):
        or_mask = (day["t"] >= REG_START) & (day["t"] < OR_END)
        if not or_mask.any():
            continue
        orh = float(day.loc[or_mask, "High"].max())
        orl = float(day.loc[or_mask, "Low"].min())
        or_width = orh - orl
        if or_width <= 0:
            continue

        after_or = day.loc[day["t"] >= OR_END]
        if after_or.empty:
            continue

        side, brk_idx = first_breakout_close(after_or, orh, orl)
        if side is None:
            continue

        broken_level = orh if side == "long" else orl
        after_breakout_close = after_or.loc[after_or.index > brk_idx]
        if after_breakout_close.empty:
            continue

        retest_idx = first_retest_touch(after_breakout_close, broken_level)
        if retest_idx is None:
            continue

        entry_time = day.loc[retest_idx, "EventAt"]
        entry      = broken_level
        stop       = orl if side == "long" else orh
        target     = entry + ( 2 * or_width if side == "long" else -2 * or_width)

        after_entry = day.loc[day.index > retest_idx]
        if after_entry.empty:
            pnl = 0.0
            costs = apply_costs(entry, entry, shares=1)
            pnl_net = pnl - costs
            records.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "ORH": orh, "ORL": orl, "OR_Width": or_width,
                "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
                "Outcome_1to2": "EOD", "ExitPx_1to2": entry, "ExitTime_1to2": entry_time,
                "PnL_1to2": pnl, "PnL_1to2_net": pnl_net
            })
            continue

        outcome, exit_px, exit_idx = simulate_path(
            after_entry, side, entry, stop, target, stop_first=STOP_FIRST_IF_BOTH_HIT
        )
        pnl = pnl_one_share(side, entry, exit_px)
        costs = apply_costs(entry, exit_px, shares=1)
        pnl_net = pnl - costs

        records.append({
            "Date": d, "Symbol": symbol, "Side": side,
            "ORH": orh, "ORL": orl, "OR_Width": or_width,
            "EntryTime": entry_time, "Entry": entry, "Stop": stop, "Target_1to2": target,
            "Outcome_1to2": outcome, "ExitPx_1to2": exit_px,
            "ExitTime_1to2": day.loc[exit_idx, "EventAt"] if exit_idx in day.index else None,
            "PnL_1to2": pnl, "PnL_1to2_net": pnl_net
        })

    return pd.DataFrame(records)

# ========= RUN ALL FILES WITH PRODUCTION CONTROLS =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    SAVE_DIR.mkdir(parents=True, exist_ok=True)

    # Containers for cross-sectional logic
    symbol_trades = {}       # symbol -> trades DataFrame
    symbol_daily  = {}       # symbol -> daily PnL (gross)
    symbol_rollS  = {}       # symbol -> rolling Sharpe series (shifted to be prior info)

    # 1) Run baseline backtest for each symbol
    for f in files:
        symbol = f.stem.upper()
        try:
            trades = backtest_one_symbol(f, symbol)
        except Exception as e:
            print(f"[SKIP] {symbol}: {e}")
            continue

        if trades.empty:
            print(f"=== ORB 15m Retest — {symbol} (1 share, RR 1:2) ===")
            print("Trades: 0 | Total PnL: $0.00 | Win%: 0.00% | Profit Factor: 0.000")
            continue

        # Save raw trades
        out_path = SAVE_DIR / f"orb_trades_{symbol}_one_share_RR1to2.csv"
        trades.to_csv(out_path, index=False)

        # Baseline metrics (gross)
        b_tr, b_pnl, b_wr, b_pf = calc_metrics(trades, col="PnL_1to2")
        print(f"=== ORB 15m Retest — {symbol} (1 share, RR 1:2) ===")
        print(f"Trades: {b_tr} | Total PnL: ${b_pnl:,.2f} | Win%: {b_wr:.2f}% | Profit Factor: {b_pf:.3f}")
        print(f"Saved trades → {out_path.resolve()}")

        # Build daily series & prior rolling Sharpe
        daily = trades.groupby("Date")["PnL_1to2"].sum().sort_index()
        rollS = rolling_sharpe(daily, lookback=ROLL_DAYS).shift(1)  # prior info only

        symbol_trades[symbol] = trades
        symbol_daily[symbol]  = daily
        symbol_rollS[symbol]  = rollS

    if not symbol_trades:
        print("\nNo trades generated for any symbols.")
        return

    # 2) Per-symbol Regime Gate (prior rolling Sharpe > SHARPE_GATE)
    gated_trades_by_symbol = {}
    for sym, trades in symbol_trades.items():
        rollS = symbol_rollS[sym]
        ok_days = rollS.index[rollS > SHARPE_GATE]
        gated = trades[trades["Date"].isin(ok_days)]
        gated_trades_by_symbol[sym] = gated

        # Per-symbol PROD metrics (gated only, still 1-share)
        if not gated.empty:
            g_tr, g_pnl, g_wr, g_pf = calc_metrics(gated, col="PnL_1to2")
            print(f"— Production (Regime-Gated) — {sym}: Trades: {g_tr} | Total PnL: ${g_pnl:,.2f} | Win%: {g_wr:.2f}% | PF: {g_pf:.3f}")

    # 3) Cross-sectional Top-K selection by prior rolling Sharpe (daily)
    # Build DataFrame of prior roll Sharpe per symbol
    all_dates = sorted(set().union(*[s.index for s in symbol_rollS.values()]))
    rollS_df = pd.DataFrame(index=pd.Index(all_dates, name="Date"))
    for sym, rs in symbol_rollS.items():
        rollS_df[sym] = rs

    # For each date, pick symbols with rollS > SHARPE_GATE and among top-K
    topk_allow = {}  # date -> set(symbols)
    for d, row in rollS_df.iterrows():
        valid = row[row > SHARPE_GATE].dropna()
        if valid.empty:
            topk_allow[d] = set()
            continue
        topk = valid.sort_values(ascending=False).head(TOP_K_PER_DAY).index.tolist()
        topk_allow[d] = set(topk)

    # 4) Build overall DataFrames
    all_trades_baseline = pd.concat(symbol_trades.values(), ignore_index=True)
    all_trades_gated    = pd.concat(gated_trades_by_symbol.values(), ignore_index=True)

    # Cross-sectional Top-K filter (on top of regime gate)
    mask_topk = all_trades_gated.apply(lambda r: r["Symbol"] in topk_allow.get(r["Date"], set()), axis=1)
    all_trades_prod = all_trades_gated[mask_topk].copy()

    # 5) Optional Volatility Targeting (per symbol/day, prior info)
    if USE_VOL_TARGETING and not all_trades_prod.empty:
        # compute prior rolling daily vol per symbol
        prior_vol = {}
        for sym, daily in symbol_daily.items():
            vol = daily.rolling(ROLL_DAYS, min_periods=ROLL_DAYS).std(ddof=1).shift(1)
            prior_vol[sym] = vol

        def size_mult(sym, d):
            vol = prior_vol.get(sym)
            if vol is None or d not in vol.index:
                return 1.0
            v = vol.loc[d]
            if pd.isna(v) or v <= 0:
                return 1.0
            m = min(TARGET_DOLLAR_VOL / v, SIZE_MULT_CAP)
            return float(max(m, 0.0))

        all_trades_prod["SizeMult"] = [
            size_mult(sym, d) for sym, d in zip(all_trades_prod["Symbol"], all_trades_prod["Date"])
        ]
        all_trades_prod["PnL_prod"] = all_trades_prod["PnL_1to2"] * all_trades_prod["SizeMult"]
    else:
        all_trades_prod["PnL_prod"] = all_trades_prod["PnL_1to2"]

    # ===== OVERALL REPORTING =====
    # Baseline (all trades)
    b_tr, b_pnl, b_wr, b_pf = calc_metrics(all_trades_baseline, col="PnL_1to2")
    print(f"\n=== ORB 15m Retest — OVERALL (Baseline; 1 share, RR 1:2) ===")
    print(f"Trades: {b_tr} | Total PnL: ${b_pnl:,.2f} | Win%: {b_wr:.2f}% | Profit Factor: {b_pf:.3f}")

    # Regime-Gated overall (still 1 share)
    if not all_trades_gated.empty:
        g_tr, g_pnl, g_wr, g_pf = calc_metrics(all_trades_gated, col="PnL_1to2")
        print(f"\n=== ORB 15m Retest — OVERALL (Regime-Gated) ===")
        print(f"Trades: {g_tr} | Total PnL: ${g_pnl:,.2f} | Win%: {g_wr:.2f}% | Profit Factor: {g_pf:.3f}")

    # Production (Gated + Top-K + Vol Targeting)
    if not all_trades_prod.empty:
        p_tr, p_pnl, p_wr, p_pf = calc_metrics(all_trades_prod, col="PnL_prod")
        save_all_prod = SAVE_DIR / "orb_trades_ALL_one_share_RR1to2_PROD.csv"
        all_trades_prod.to_csv(save_all_prod, index=False)

        print(f"\n=== ORB 15m Retest — OVERALL (PRODUCTION: Gated + Top-{TOP_K_PER_DAY}"
              f"{' + VolTarget' if USE_VOL_TARGETING else ''}) ===")
        print(f"Trades: {p_tr} | Total PnL: ${p_pnl:,.2f} | Win%: {p_wr:.2f}% | Profit Factor: {p_pf:.3f}")
        print(f"Saved trades → {save_all_prod.resolve()}")
    else:
        print("\nNo Production trades after gating/top-K settings (increase history or relax gates).")

if __name__ == "__main__":
    main()
```

```python
# vwap_pullback_trend_PROD.py  — fast, debugged, progress-printing
# VWAP Pullback Trend — pullbacks to session VWAP within established intraday trends.
# Production features: regime gate, Top-K selection, optional vol targeting, realistic costs.

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from pathlib import Path

# ========= USER SETTINGS =========
PARQUET_DIR = r"C:\Users\pcagm\OneDrive\Desktop\downloads\parquet\parquet"
SAVE_DIR    = Path(".")
TZ_LOCAL    = "America/New_York"

REG_START   = "09:30"
REG_END     = "16:00"
TRADE_START = "10:00"

# Optional throttles for faster testing in notebooks
MAX_FILES          = None   # e.g., 5 to process only 5 tickers
DATE_FROM          = None   # e.g., "2022-01-01"
DATE_TO            = None   # e.g., "2024-12-31"

# Entry/exit params
VWAP_TOL_BPS       = 5.0     # tolerance around VWAP in bps (5 = 0.05%)
ATR_MULT_STOP      = 1.5
RR_TARGET          = 2.0
MAX_TRADES_PER_DAY = 1

# Trend filters
EMA_FAST  = 50
EMA_SLOW  = 200
ADX_LEN   = 14
ADX_THRESH = 18.0

# Production controls
ROLL_DAYS            = 60
SHARPE_GATE          = 0.0
TOP_K_PER_DAY        = 10
USE_VOL_TARGETING    = True
TARGET_DOLLAR_VOL    = 300.0
SIZE_MULT_CAP        = 4.0
TRADING_DAYS_PER_YEAR = 252

# Costs
COMM_PER_SHARE = 0.003
SPREAD_BPS     = 1.0
ENTRY_CROSS    = False
EXIT_CROSS     = True

# ========= HELPERS =========
def to_ny(df, col="EventAt"):
    ts = pd.to_datetime(df[col], errors="coerce")
    if ts.dt.tz is None:
        ts = ts.dt.tz_localize("UTC").dt.tz_convert(TZ_LOCAL)
    else:
        ts = ts.dt.tz_convert(TZ_LOCAL)
    return ts

def ema(s, span):
    return s.ewm(span=span, adjust=False, min_periods=span).mean()

def true_range(h, l, c_prev):
    return pd.concat([(h - l).abs(), (h - c_prev).abs(), (l - c_prev).abs()], axis=1).max(axis=1)

def atr(df, n=14):
    c_prev = df["Close"].shift(1)
    tr = true_range(df["High"], df["Low"], c_prev)
    return tr.rolling(n, min_periods=n).mean()

def dx(df, n=14):
    up   = df["High"].diff()
    down = -df["Low"].diff()
    plus_dm  = np.where((up > down) & (up > 0),   up,   0.0)
    minus_dm = np.where((down > up) & (down > 0), down, 0.0)
    tr = true_range(df["High"], df["Low"], df["Close"].shift(1))
    atr_n = tr.rolling(n, min_periods=n).mean()

    plus_di  = 100 * pd.Series(plus_dm,  index=df.index).rolling(n, min_periods=n).mean() / atr_n
    minus_di = 100 * pd.Series(minus_dm, index=df.index).rolling(n, min_periods=n).mean() / atr_n
    denom = (plus_di + minus_di).replace(0, np.nan)
    return 100 * (plus_di - minus_di).abs() / denom

def adx(df, n=14):
    return dx(df, n).rolling(n, min_periods=n).mean()

def rsi(close, n=14):
    delta = close.diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    roll_up = up.rolling(n, min_periods=n).mean()
    roll_down = down.rolling(n, min_periods=n).mean()
    rs = roll_up / roll_down.replace(0, np.nan)
    return 100 - (100 / (1 + rs))

def session_vwap(df_day):
    # VWAP using typical price * volume; resets each day
    tp  = (df_day["High"] + df_day["Low"] + df_day["Close"]) / 3.0
    vol = pd.to_numeric(df_day["Volume"], errors="coerce").fillna(0.0).clip(lower=0.0)
    cum_pv = (tp * vol).cumsum()
    cum_v  = vol.cumsum().replace(0, np.nan)
    return cum_pv / cum_v

def apply_costs(entry, exit_px, shares=1):
    spread_cost = 0.0
    if ENTRY_CROSS:
        spread_cost += (SPREAD_BPS / 1e4) * entry
    if EXIT_CROSS:
        spread_cost += (SPREAD_BPS / 1e4) * exit_px
    fees = COMM_PER_SHARE * 2 * shares
    return spread_cost * shares + fees

def simulate_path(df_after_entry, side, entry, stop, target):
    if df_after_entry.empty:
        return "EOD", entry, None
    for idx, row in df_after_entry.iterrows():
        lo, hi = row["Low"], row["High"]
        hit_stop = (lo <= stop) if side == "long" else (hi >= stop)
        hit_tp   = (hi >= target) if side == "long" else (lo <= target)
        if hit_stop and hit_tp:
            return "STOP", stop, idx  # adverse selection
        if hit_stop:
            return "STOP", stop, idx
        if hit_tp:
            return "TP", target, idx
    last_idx = df_after_entry.index[-1]
    return "EOD", float(df_after_entry.loc[last_idx, "Close"]), last_idx

def pnl_one_share(side, entry, exit_px):
    return (exit_px - entry) if side == "long" else (entry - exit_px)

def calc_metrics(trades_df, col="PnL_net"):
    if trades_df.empty:
        return 0, 0.0, 0.0, 0.0
    pnl = trades_df[col].fillna(0.0)
    wins = (pnl > 0).sum()
    losses = (pnl < 0).sum()
    total = int(wins + losses)
    total_pnl = float(pnl.sum())
    gross_pos = pnl[pnl > 0].sum()
    gross_neg = -pnl[pnl < 0].sum()
    pf = float(gross_pos / gross_neg) if gross_neg > 0 else float("inf")
    winrate = (100.0 * wins / total) if total else 0.0
    return total, total_pnl, winrate, pf

def rolling_sharpe(daily_pnl, lookback=ROLL_DAYS):
    mu = daily_pnl.rolling(lookback, min_periods=lookback).mean()
    sd = daily_pnl.rolling(lookback, min_periods=lookback).std(ddof=1)
    sharpe = (mu / sd.replace(0, np.nan)) * np.sqrt(TRADING_DAYS_PER_YEAR)
    return sharpe.replace([np.inf, -np.inf], np.nan)

# ========= CORE STRATEGY =========
def backtest_symbol_vwap(parquet_path: Path, symbol: str) -> pd.DataFrame:
    print(f"[READ] {symbol} …", flush=True)
    df = pd.read_parquet(parquet_path, columns=["EventAt","Open","High","Low","Close","Volume"])
    for c in ["EventAt","Open","High","Low","Close","Volume"]:
        if c not in df.columns:
            raise ValueError(f"{symbol}: parquet missing required column: {c}")

    df["EventAt"] = to_ny(df, "EventAt")
    if DATE_FROM:
        df = df[df["EventAt"] >= pd.Timestamp(DATE_FROM, tz=TZ_LOCAL)]
    if DATE_TO:
        df = df[df["EventAt"] <= pd.Timestamp(DATE_TO, tz=TZ_LOCAL)]
    df = df.sort_values("EventAt").reset_index(drop=True)

    # keep regular session
    df["date"] = df["EventAt"].dt.date
    df["t"]    = df["EventAt"].dt.strftime("%H:%M")
    session = (df["t"] >= REG_START) & (df["t"] <= REG_END)
    df = df.loc[session].copy()

    if df.empty:
        print(f"[EMPTY] {symbol} after session filter.", flush=True)
        return pd.DataFrame()

    print(f"[INFO] {symbol}: rows={len(df):,}, days={df['date'].nunique():,}", flush=True)

    # Precompute indicators across full intraday series
    df["EMA_fast"] = ema(df["Close"], EMA_FAST)
    df["EMA_slow"] = ema(df["Close"], EMA_SLOW)
    df["ADX"]      = adx(df, ADX_LEN)
    df["ATR"]      = atr(df, 14)
    df["RSI3"]     = rsi(df["Close"], 3)

    rows = []

    # Group by day with progress every N days
    for di, (d, day) in enumerate(df.groupby("date", sort=True), 1):
        if day.empty:
            continue

        # session VWAP (reset each day)
        day = day.copy()
        day["VWAP"] = session_vwap(day)

        # trade window
        tradable = day[day["t"] >= TRADE_START].copy()
        if tradable.empty:
            continue

        trades_taken = 0

        # ===== Vectorized signal detection (faster) =====
        tol_arr = tradable["VWAP"] * (VWAP_TOL_BPS / 1e4)
        uptrend   = (df.loc[tradable.index, "EMA_fast"] > df.loc[tradable.index, "EMA_slow"]) & (df.loc[tradable.index, "ADX"] >= ADX_THRESH)
        downtrend = (df.loc[tradable.index, "EMA_fast"] < df.loc[tradable.index, "EMA_slow"]) & (df.loc[tradable.index, "ADX"] >= ADX_THRESH)

        long_sig  = uptrend & (tradable["Low"] <= tradable["VWAP"] - tol_arr) & (tradable["Close"] > tradable["VWAP"]) & (df.loc[tradable.index, "RSI3"] <= 35)
        short_sig = downtrend & (tradable["High"] >= tradable["VWAP"] + tol_arr) & (tradable["Close"] < tradable["VWAP"]) & (df.loc[tradable.index, "RSI3"] >= 65)

        signal_idx = tradable.index[np.where(long_sig | short_sig)[0]]

        for idx in signal_idx:
            if trades_taken >= MAX_TRADES_PER_DAY:
                break

            # signal state (use *signal bar* indicator/ATR; fill at future bar)
            side = "long" if long_sig.loc[idx] else "short"
            ema_fast = df.loc[idx, "EMA_fast"]; ema_slow = df.loc[idx, "EMA_slow"]
            adx_now  = df.loc[idx, "ADX"];       atr_now  = df.loc[idx, "ATR"]
            vwap_sig = day.loc[idx, "VWAP"]

            if pd.isna([ema_fast, ema_slow, adx_now, atr_now, vwap_sig]).any() or atr_now <= 0:
                continue

            after_signal = day.loc[day.index > idx]
            if after_signal.empty:
                break

            # fill when price touches VWAP (use entry bar's VWAP)
            touch = (after_signal["Low"] <= vwap_sig) & (after_signal["High"] >= vwap_sig)
            if not touch.any():
                continue
            entry_idx  = after_signal.index[touch].min()
            entry_time = day.loc[entry_idx, "EventAt"]
            entry      = float(day.loc[entry_idx, "VWAP"])

            # risk/target from signal ATR (prior-known)
            if side == "long":
                stop   = entry - ATR_MULT_STOP * float(atr_now)
                target = entry + RR_TARGET * (entry - stop)
            else:
                stop   = entry + ATR_MULT_STOP * float(atr_now)
                target = entry - RR_TARGET * (stop - entry)

            after_entry = day.loc[day.index > entry_idx]
            if after_entry.empty:
                continue

            outcome, exit_px, exit_idx = simulate_path(after_entry, side, entry, stop, target)
            gross = pnl_one_share(side, entry, exit_px)
            costs = apply_costs(entry, exit_px, shares=1)
            pnl_net = gross - costs

            rows.append({
                "Date": d, "Symbol": symbol, "Side": side,
                "EntryTime": entry_time, "Entry": entry,
                "Stop": stop, "Target": target,
                "ExitTime": day.loc[exit_idx, "EventAt"] if (exit_idx is not None and exit_idx in day.index) else None,
                "ExitPx": exit_px, "Outcome": outcome,
                "PnL_gross": gross, "PnL_net": pnl_net
            })

            trades_taken += 1

        if di % 25 == 0:  # heartbeat every 25 days
            print(f"[{symbol}] processed {di} days…", flush=True)

    return pd.DataFrame(rows)

# ========= DRIVER (gating, Top-K, vol targeting) =========
def main():
    parquet_dir = Path(PARQUET_DIR)
    files = sorted(parquet_dir.glob("*.parquet"))
    if MAX_FILES:
        files = files[:MAX_FILES]
    if not files:
        print(f"No parquet files found in {parquet_dir}")
        return

    SAVE_DIR.mkdir(parents=True, exist_ok=True)

    per_symbol_trades = {}
    per_symbol_daily  = {}
    per_symbol_rollS  = {}

    # 1) Build baseline trades per symbol
    for i, f in enumerate(files, 1):
        sym = f.stem.upper()
        print(f"\n[{i}/{len(files)}] Processing {sym} …", flush=True)
        try:
            trades = backtest_symbol_vwap(f, sym)
        except Exception as e:
            print(f"[SKIP] {sym}: {e}", flush=True)
            continue

        if trades.empty:
            print(f"=== VWAP Pullback Trend — {sym} ===")
            print("Trades: 0 | Total PnL (net): $0.00 | Win%: 0.00% | PF: 0.000")
            continue

        per_symbol_trades[sym] = trades
        out_path = SAVE_DIR / f"vwap_trades_{sym}.csv"
        trades.to_csv(out_path, index=False)

        tr, pnl, wr, pf = calc_metrics(trades, col="PnL_net")
        print(f"=== VWAP Pullback Trend — {sym} (Baseline, net) ===")
        print(f"Trades: {tr} | Total PnL: ${pnl:,.2f} | Win%: {wr:.2f}% | PF: {pf:.3f}")
        print(f"Saved trades → {out_path.resolve()}", flush=True)

        daily = trades.groupby("Date")["PnL_net"].sum().sort_index()
        rollS = rolling_sharpe(daily, lookback=ROLL_DAYS).shift(1)  # prior info only
        per_symbol_daily[sym] = daily
        per_symbol_rollS[sym] = rollS

    if not per_symbol_trades:
        print("\nNo trades generated for any symbols.")
        return

    # 2) Regime gate per symbol
    gated_list = []
    for sym, trades in per_symbol_trades.items():
        rollS = per_symbol_rollS[sym]
        ok_dates = rollS.index[rollS > SHARPE_GATE]
        gated = trades[trades["Date"].isin(ok_dates)]
        if not gated.empty:
            g_tr, g_pnl, g_wr, g_pf = calc_metrics(gated, col="PnL_net")
            print(f"— Production (Regime-Gated) — {sym}: "
                  f"Trades: {g_tr} | PnL: ${g_pnl:,.2f} | Win%: {g_wr:.2f}% | PF: {g_pf:.3f}", flush=True)
            gated_list.append(gated)

    if not gated_list:
        print("\nNo trades after regime gate; relax SHARPE_GATE or increase history.")
        return

    all_gated = pd.concat(gated_list, ignore_index=True)

    # 3) Top-K cross-sectional selection by prior rolling Sharpe (optional)
    if TOP_K_PER_DAY and TOP_K_PER_DAY > 0:
        all_dates = sorted(set().union(*[s.index for s in per_symbol_rollS.values()]))
        rs_df = pd.DataFrame(index=pd.Index(all_dates, name="Date"))
        for sym, rs in per_symbol_rollS.items():
            rs_df[sym] = rs

        allow = {}
        for d, row in rs_df.iterrows():
            valid = row[row > SHARPE_GATE].dropna()
            allow[d] = set(valid.sort_values(ascending=False).head(TOP_K_PER_DAY).index.tolist()) if not valid.empty else set()

        mask = all_gated.apply(lambda r: r["Symbol"] in allow.get(r["Date"], set()), axis=1)
        all_prod = all_gated[mask].copy()
    else:
        all_prod = all_gated.copy()

    if all_prod.empty:
        print("\nNo trades after Top-K selection; reduce TOP_K_PER_DAY.")
        return

    # 4) Vol targeting (optional)
    if USE_VOL_TARGETING:
        prior_vol = {}
        for sym, daily in per_symbol_daily.items():
            prior_vol[sym] = daily.rolling(ROLL_DAYS, min_periods=ROLL_DAYS).std(ddof=1).shift(1)

        def mult(sym, d):
            vol = prior_vol.get(sym)
            if vol is None or d not in vol.index:
                return 1.0
            v = vol.loc[d]
            if pd.isna(v) or v <= 0:
                return 1.0
            return float(np.clip(TARGET_DOLLAR_VOL / v, 0.0, SIZE_MULT_CAP))

        all_prod["SizeMult"] = [mult(s, d) for s, d in zip(all_prod["Symbol"], all_prod["Date"])]
        all_prod["PnL_prod"] = all_prod["PnL_net"] * all_prod["SizeMult"]
        metric_col = "PnL_prod"
    else:
        all_prod["PnL_prod"] = all_prod["PnL_net"]
        metric_col = "PnL_prod"

    # ===== OVERALL REPORT =====
    base_all = pd.concat(per_symbol_trades.values(), ignore_index=True)
    b_tr, b_pnl, b_wr, b_pf = calc_metrics(base_all, col="PnL_net")
    print(f"\n=== VWAP Pullback Trend — OVERALL (Baseline, net) ===")
    print(f"Trades: {b_tr} | Total PnL: ${b_pnl:,.2f} | Win%: {b_wr:.2f}% | PF: {b_pf:.3f}")

    g_tr, g_pnl, g_wr, g_pf = calc_metrics(all_gated, col="PnL_net")
    print(f"\n=== OVERALL (Regime-Gated, net) ===")
    print(f"Trades: {g_tr} | Total PnL: ${g_pnl:,.2f} | Win%: {g_wr:.2f}% | PF: {g_pf:.3f}")

    p_tr, p_pnl, p_wr, p_pf = calc_metrics(all_prod, col=metric_col)
    out_all = SAVE_DIR / "vwap_trades_ALL_PROD.csv"
    all_prod.to_csv(out_all, index=False)
    print(f"\n=== OVERALL (PRODUCTION: Gated + Top-{TOP_K_PER_DAY}"
          f"{' + VolTarget' if USE_VOL_TARGETING else ''}, net) ===")
    print(f"Trades: {p_tr} | Total PnL: ${p_pnl:,.2f} | Win%: {p_wr:.2f}% | PF: {p_pf:.3f}")
    print(f"Saved trades → {out_all.resolve()}", flush=True)

if __name__ == "__main__":
    main()
```
