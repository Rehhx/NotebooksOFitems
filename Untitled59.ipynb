```python
# === pip (once) ===
# import sys
# !{sys.executable} -m pip install "psycopg[binary]" pandas numpy pytz

import math
from dataclasses import dataclass
from pathlib import Path
from typing import List, Optional, Tuple, Dict

import numpy as np
import pandas as pd
import psycopg
import pytz

# =============================
# DB & RUN SETTINGS (EDIT ME)
# =============================

CONN_STR   = "host=24.128.195.68 port=8812 dbname=qdb user=admin password=quest"
TABLE      = "futures_candles"       # <— your table per screenshot
INTERVAL   = "1m"
SYMBOLS    = ["NQ"]                   # exact symbol(s) in your table, or use a prefix filter below
SYMBOL_PREFIX: Optional[str] = None   # e.g. "NQ" to grab all NQ variants; set to None to ignore

# Optional UTC date window (ISO-8601)
START_DATE: Optional[str] = None      # e.g. "2024-10-01"
END_DATE:   Optional[str] = None

# Output
SAVE_DIR = Path(".")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# =============================
# MARKET & STRATEGY PARAMETERS
# =============================

TZ_LOCAL = "America/New_York"

# RTH session for index futures (tweak if you use extended)
REG_START = "09:30"
REG_END   = "16:00"

# Indicators / swings
RSI_LEN   = 14
SWING_K   = 5               # swing high/low lookback-ahead window
ATR_LEN   = 14

# Execution / risk
ATR_MULT_STOP   = 1.0       # stop distance = 1 * ATR
R_MULT_TARGET   = 1.5       # target = 1.5R
CONFIRM_ENTRY   = True      # break of signal bar in divergence direction

# Level-to-level
USE_LEVELS                = True
LEVEL_PROX_THRESHOLD_ATR  = 0.5  # must be within X*ATR of a level to take level-based setups
USE_DAILY_OPEN_AS_LEVEL   = True # use session open as an intraday level

# Instrument specs (NQ E-mini defaults)
POINT_VALUE   = 20.0        # $ per point (E-mini NQ = $20, Micro MNQ = $2)
TICK_SIZE     = 0.25        # NQ tick
TICK_VALUE    = POINT_VALUE * TICK_SIZE  # $5 per tick for NQ

# =============================
# Helpers
# =============================

def to_ny(ts: pd.Series) -> pd.Series:
    """Assume incoming timestamps are UTC; convert to New York tz."""
    utc = pd.to_datetime(ts, utc=True)
    return utc.dt.tz_convert(TZ_LOCAL)

def rsi(close: pd.Series, length: int = 14) -> pd.Series:
    delta = close.diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    roll_up = up.ewm(alpha=1/length, adjust=False).mean()
    roll_down = down.ewm(alpha=1/length, adjust=False).mean()
    rs = roll_up / roll_down.replace(0, np.nan)
    return 100 - (100 / (1 + rs))

def atr(df: pd.DataFrame, length: int = 14) -> pd.Series:
    high, low, close = df["high"], df["low"], df["close"]
    prev_close = close.shift(1)
    tr = pd.concat([
        (high - low),
        (high - prev_close).abs(),
        (low - prev_close).abs()
    ], axis=1).max(axis=1)
    return tr.ewm(alpha=1/length, adjust=False).mean()

def swing_high(series: pd.Series, k: int) -> pd.Series:
    return (series.shift(-i).le(series) for i in range(1, k+1))

def is_swing_high(series: pd.Series, k: int) -> pd.Series:
    left = series.rolling(k+1, center=True).apply(lambda x: x[-1] == x.max(), raw=False)
    # require both sides: check neighbors
    cond = (series == series.rolling(k, center=False).max()).rolling(k+1, center=True).max() == 1
    # Simpler robust version:
    return (series == series.rolling(2*k+1, center=True).max())

def is_swing_low(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).min())

def nearest_level(price: float, levels: List[float]) -> Tuple[float, float]:
    dists = [abs(price - lv) for lv in levels]
    if not dists:
        return np.nan, np.inf
    idx = int(np.argmin(dists))
    return levels[idx], dists[idx]

@dataclass
class Trade:
    symbol: str
    entry_time: pd.Timestamp
    side: str              # "long"/"short"
    entry: float
    stop: float
    target: float
    exit_time: pd.Timestamp
    exit_price: float
    reason: str            # "target" / "stop" / "eod"
    r_mult: float
    points: float
    ticks: int
    pnl_usd: float
    day: str

# =============================
# Data
# =============================

def load_data(conn_str: str,
              table: str,
              symbols: List[str],
              interval: str,
              start: Optional[str],
              end: Optional[str],
              symbol_prefix: Optional[str] = None) -> pd.DataFrame:
    """
    Read-only pull from QuestDB. Expects columns:
      symbol, interval, event_at (timestamp), open, high, low, close, volume
    """
    where = [f"interval = '{interval}'"]
    if symbols and not symbol_prefix:
        syms = ",".join("'" + s.replace("'", "''") + "'" for s in symbols)
        where.append(f"symbol IN ({syms})")
    if symbol_prefix:
        where.append(f"symbol LIKE '{symbol_prefix}%'")
    if start:
        where.append(f"event_at >= TIMESTAMP '{start}'")
    if end:
        where.append(f"event_at < TIMESTAMP '{end}'")

    sql = f"""
        SELECT symbol, interval, event_at, open, high, low, close, volume
        FROM {table}
        WHERE {" AND ".join(where)}
        ORDER BY symbol, event_at
    """

    with psycopg.connect(conn_str) as conn:
        df = pd.read_sql(sql, conn)
    if df.empty:
        return df

    df["ts_ny"] = to_ny(df["event_at"])
    df["date_ny"] = df["ts_ny"].dt.date
    df["time_ny"] = df["ts_ny"].dt.time
    return df

# =============================
# Signals (Divergence + Levels)
# =============================

def build_signals(df: pd.DataFrame) -> pd.DataFrame:
    """
    Adds RSI, ATR, swing points, and divergence markers.
    """
    df = df.copy()
    df["rsi"] = rsi(df["close"], RSI_LEN)
    df["atr"] = atr(df, ATR_LEN)

    # swing highs/lows on price and RSI
    df["sh"] = is_swing_high(df["high"], SWING_K).fillna(False)
    df["sl"] = is_swing_low(df["low"], SWING_K).fillna(False)
    df["rsi_sh"] = is_swing_high(df["rsi"], SWING_K).fillna(False)
    df["rsi_sl"] = is_swing_low(df["rsi"], SWING_K).fillna(False)

    # Track last swing highs/lows to check divergence
    df["last_price_sh"] = np.where(df["sh"], df["high"], np.nan)
    df["last_price_sl"] = np.where(df["sl"], df["low"],  np.nan)
    df["last_rsi_sh"]   = np.where(df["rsi_sh"], df["rsi"], np.nan)
    df["last_rsi_sl"]   = np.where(df["rsi_sl"], df["rsi"], np.nan)

    df["last_price_sh"] = df["last_price_sh"].ffill()
    df["last_price_sl"] = df["last_price_sl"].ffill()
    df["last_rsi_sh"]   = df["last_rsi_sh"].ffill()
    df["last_rsi_sl"]   = df["last_rsi_sl"].ffill()

    # Bearish divergence: price higher high, RSI lower high
    df["bear_div"] = (
        (df["sh"]) &
        (df["high"] > df["last_price_sh"].shift(1)) &
        (df["rsi_sh"]) &
        (df["rsi"] < df["last_rsi_sh"].shift(1))
    )

    # Bullish divergence: price lower low, RSI higher low
    df["bull_div"] = (
        (df["sl"]) &
        (df["low"] < df["last_price_sl"].shift(1)) &
        (df["rsi_sl"]) &
        (df["rsi"] > df["last_rsi_sl"].shift(1))
    )

    return df

def daily_levels(day_df: pd.DataFrame, prev_day_df: Optional[pd.DataFrame]) -> List[float]:
    levels = []
    if prev_day_df is not None and not prev_day_df.empty:
        levels += [prev_day_df["high"].max(), prev_day_df["low"].min()]
    if USE_DAILY_OPEN_AS_LEVEL and not day_df.empty:
        # session open (first bar in day window)
        levels.append(day_df.iloc[0]["open"])
    # Deduplicate & sort
    levels = sorted(list({round(lv, 4) for lv in levels}))
    return levels

# =============================
# Backtest
# =============================

def simulate_intraday(symbol: str, df: pd.DataFrame) -> List[Trade]:
    trades: List[Trade] = []

    # prep features
    df = build_signals(df)

    # intraday groups
    for day, g in df.groupby("date_ny"):
        # RTH window
        day_slice = g[(g["ts_ny"].dt.strftime("%H:%M") >= REG_START) &
                      (g["ts_ny"].dt.strftime("%H:%M") <= REG_END)].copy()
        if day_slice.empty:
            continue

        # previous day for levels
        prev_df = df[df["date_ny"] == (pd.to_datetime(day) - pd.Timedelta(days=1)).date()]
        levels = daily_levels(day_slice, prev_df)
        active_pos = None

        for i in range(1, len(day_slice)):
            row_prev = day_slice.iloc[i-1]
            row = day_slice.iloc[i]

            # manage open trade first
            if active_pos:
                # intra-bar check with minute bars (coarse but deterministic)
                if active_pos["side"] == "long":
                    # stop first, then target or configurable priority
                    if row["low"] <= active_pos["stop"]:
                        # stopped
                        exit_px = active_pos["stop"]
                        r = (exit_px - active_pos["entry"]) / (active_pos["entry"] - active_pos["stop"])
                        pnl_points = exit_px - active_pos["entry"]
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "stop", r, pnl_points))
                        active_pos = None
                    elif row["high"] >= active_pos["target"]:
                        exit_px = active_pos["target"]
                        r = R_MULT_TARGET
                        pnl_points = exit_px - active_pos["entry"]
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "target", r, pnl_points))
                        active_pos = None

                else:  # short
                    if row["high"] >= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        r = (active_pos["entry"] - exit_px) / (active_pos["stop"] - active_pos["entry"])
                        pnl_points = active_pos["entry"] - exit_px
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "stop", r, pnl_points))
                        active_pos = None
                    elif row["low"] <= active_pos["target"]:
                        exit_px = active_pos["target"]
                        r = R_MULT_TARGET
                        pnl_points = active_pos["entry"] - exit_px
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "target", r, pnl_points))
                        active_pos = None

            # if flat, look for signals
            if active_pos is None:
                atr_now = max(1e-6, row["atr"])
                # Level proximity filter (optional)
                near_level = True
                if USE_LEVELS and levels:
                    _, dist = nearest_level(row["close"], levels)
                    near_level = (dist <= LEVEL_PROX_THRESHOLD_ATR * atr_now)

                # Divergence entries with confirmation
                # Bearish divergence -> short
                if row_prev["bear_div"] and near_level:
                    signal_high = row_prev["high"]
                    entry = row["low"] if not CONFIRM_ENTRY else min(row["open"], row["low"])  # conservative
                    entry = min(entry, signal_high - TICK_SIZE)  # below signal bar
                    stop  = entry + ATR_MULT_STOP * atr_now
                    target = entry - R_MULT_TARGET * (stop - entry)
                    active_pos = {"side": "short", "entry_time": row["ts_ny"], "entry": entry,
                                  "stop": stop, "target": target, "reason": "bear_div"}

                # Bullish divergence -> long
                elif row_prev["bull_div"] and near_level:
                    signal_low = row_prev["low"]
                    entry = row["high"] if not CONFIRM_ENTRY else max(row["open"], row["high"])  # conservative
                    entry = max(entry, signal_low + TICK_SIZE)  # above signal bar
                    stop  = entry - ATR_MULT_STOP * atr_now
                    target = entry + R_MULT_TARGET * (entry - stop)
                    active_pos = {"side": "long", "entry_time": row["ts_ny"], "entry": entry,
                                  "stop": stop, "target": target, "reason": "bull_div"}

        # Flat any open position at session end
        if active_pos is not None:
            last = day_slice.iloc[-1]
            exit_px = last["close"]
            if active_pos["side"] == "long":
                pnl_points = exit_px - active_pos["entry"]
                r = pnl_points / (active_pos["entry"] - active_pos["stop"])
            else:
                pnl_points = active_pos["entry"] - exit_px
                r = pnl_points / (active_pos["stop"] - active_pos["entry"])
            trades.append(_finalize_trade(symbol, active_pos, last["ts_ny"], exit_px, "eod", r, pnl_points))
            active_pos = None

    return trades

def _finalize_trade(symbol: str, pos: Dict, exit_time: pd.Timestamp, exit_price: float,
                    reason: str, r_mult: float, pnl_points: float) -> Trade:
    ticks = int(round(pnl_points / TICK_SIZE))
    pnl_usd = pnl_points * POINT_VALUE
    return Trade(
        symbol=symbol,
        entry_time=pos["entry_time"], side=pos["side"],
        entry=pos["entry"], stop=pos["stop"], target=pos["target"],
        exit_time=exit_time, exit_price=exit_price,
        reason=reason, r_mult=r_mult,
        points=pnl_points, ticks=ticks, pnl_usd=pnl_usd,
        day=str(pos["entry_time"].date())
    )

# =============================
# Run
# =============================

def main():
    raw = load_data(CONN_STR, TABLE, SYMBOLS, INTERVAL, START_DATE, END_DATE, SYMBOL_PREFIX)
    if raw.empty:
        print("No data returned. Check TABLE/SYMBOLS/DATE/INTERVAL.")
        return

    results: List[Trade] = []
    for sym, sdf in raw.groupby("symbol"):
        trades = simulate_intraday(sym, sdf.reset_index(drop=True))
        results.extend(trades)

    if not results:
        print("No trades generated with current params.")
        return

    tl = pd.DataFrame([t.__dict__ for t in results]).sort_values(["day", "entry_time"])
    tl["cum_pnl_usd"] = tl["pnl_usd"].cumsum()

    # Summary
    wins = (tl["pnl_usd"] > 0).sum()
    losses = (tl["pnl_usd"] <= 0).sum()
    total = len(tl)
    winrate = wins / total if total else 0.0
    print("\n=== TRADE SUMMARY ===")
    print(f"Trades: {total} | Wins: {wins} | Losses: {losses} | Winrate: {winrate:.1%}")
    print(f"Total PnL: ${tl['pnl_usd'].sum():.2f} | Max DD: ${float((tl['cum_pnl_usd'].cummax()-tl['cum_pnl_usd']).max()):.2f}")

    # Save trade log
    out_csv = SAVE_DIR / "divergence_levels_backtest_trades.csv"
    tl.to_csv(out_csv, index=False)
    print(f"\nSaved trade log → {out_csv.resolve()}")

if __name__ == "__main__":
    main()
```

```python
# === pip (optional, once) ===
# import sys
# !{sys.executable} -m pip install "psycopg[binary]" pandas numpy pytz
!pip install "psycopg[binary]" pandas numpy pytz

import math
from dataclasses import dataclass
from pathlib import Path
from typing import List, Optional, Tuple, Dict

import numpy as np
import pandas as pd
import psycopg
import pytz

# =============================
# DB & RUN SETTINGS (EDIT ME)
# =============================

CONN_STR   = "host=24.128.195.68 port=8812 dbname=qdb user=admin password=quest"
TABLE      = "futures_candles"       # your table per screenshot
INTERVAL   = "1m"
SYMBOLS    = ["NQ"]                   # exact symbols in your table
SYMBOL_PREFIX: Optional[str] = None   # e.g. "NQ" to fetch all NQ*; set None to ignore

# Optional UTC date window
START_DATE: Optional[str] = None      # e.g. "2024-10-01"
END_DATE:   Optional[str] = None      # e.g. "2025-09-01"

# Output
SAVE_DIR = Path(".")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# =============================
# MARKET & STRATEGY PARAMETERS
# =============================

TZ_LOCAL = "America/New_York"

# RTH session for index futures
REG_START = "09:30"
REG_END   = "16:00"

# Indicators / swings
RSI_LEN   = 14
SWING_K   = 5               # swing high/low window (both sides)
ATR_LEN   = 14

# Execution / risk
ATR_MULT_STOP   = 1.0       # stop distance = 1 * ATR
R_MULT_TARGET   = 1.5       # target = 1.5R
CONFIRM_ENTRY   = True      # require next bar confirmation beyond signal bar

# Level-to-level
USE_LEVELS                = True
LEVEL_PROX_THRESHOLD_ATR  = 0.5  # must be within X*ATR of a level
USE_DAILY_OPEN_AS_LEVEL   = True # include session open as a level

# Instrument specs (E-mini NQ defaults)
POINT_VALUE   = 20.0        # $ per point (MNQ = 2.0)
TICK_SIZE     = 0.25
TICK_VALUE    = POINT_VALUE * TICK_SIZE

# =============================
# Helpers
# =============================

def to_ny(ts: pd.Series) -> pd.Series:
    """Assume incoming timestamps are UTC; convert to New York tz."""
    utc = pd.to_datetime(ts, utc=True)
    return utc.dt.tz_convert(TZ_LOCAL)


def rsi(close: pd.Series, length: int = 14) -> pd.Series:
    delta = close.diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    roll_up = up.ewm(alpha=1/length, adjust=False).mean()
    roll_down = down.ewm(alpha=1/length, adjust=False).mean()
    rs = roll_up / roll_down.replace(0, np.nan)
    return 100 - (100 / (1 + rs))


def atr(df: pd.DataFrame, length: int = 14) -> pd.Series:
    high, low, close = df["high"], df["low"], df["close"]
    prev_close = close.shift(1)
    tr = pd.concat([
        (high - low),
        (high - prev_close).abs(),
        (low - prev_close).abs()
    ], axis=1).max(axis=1)
    return tr.ewm(alpha=1/length, adjust=False).mean()


def is_swing_high(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).max())


def is_swing_low(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).min())


def nearest_level(price: float, levels: List[float]) -> Tuple[float, float]:
    if not levels:
        return (math.nan, math.inf)
    dists = [abs(price - lv) for lv in levels]
    idx = int(np.argmin(dists))
    return levels[idx], dists[idx]


@dataclass
class Trade:
    symbol: str
    entry_time: pd.Timestamp
    side: str              # "long"/"short"
    entry: float
    stop: float
    target: float
    exit_time: pd.Timestamp
    exit_price: float
    reason: str            # "target" / "stop" / "eod"
    r_mult: float
    points: float
    ticks: int
    pnl_usd: float
    day: str

# =============================
# Data
# =============================

def load_data(conn_str: str,
              table: str,
              symbols: List[str],
              interval: str,
              start: Optional[str],
              end: Optional[str],
              symbol_prefix: Optional[str] = None) -> pd.DataFrame:
    """
    Read-only pull from QuestDB. Expects columns:
      symbol, interval, event_at (timestamp), open, high, low, close, volume
    """
    where = [f"interval = '{interval}'"]
    if symbols and not symbol_prefix:
        syms = ",".join("'" + s.replace("'", "''") + "'" for s in symbols)
        where.append(f"symbol IN ({syms})")
    if symbol_prefix:
        where.append(f"symbol LIKE '{symbol_prefix}%'")
    if start:
        where.append(f"event_at >= TIMESTAMP '{start}'")
    if end:
        where.append(f"event_at < TIMESTAMP '{end}'")

    sql = f"""
        SELECT symbol, interval, event_at, open, high, low, close, volume
        FROM {table}
        WHERE {" AND ".join(where)}
        ORDER BY symbol, event_at
    """

    with psycopg.connect(conn_str) as conn:
        df = pd.read_sql(sql, conn)

    if df.empty:
        return df

    df["ts_ny"] = to_ny(df["event_at"])
    df["date_ny"] = df["ts_ny"].dt.date
    df["time_ny"] = df["ts_ny"].dt.time
    return df

# =============================
# Signals (Divergence + Levels)
# =============================

def build_signals(df: pd.DataFrame) -> pd.DataFrame:
    """Adds RSI, ATR, swing points, and divergence markers."""
    df = df.copy()
    df["rsi"] = rsi(df["close"], RSI_LEN)
    df["atr"] = atr(df, ATR_LEN)

    # swing highs/lows on price and RSI
    df["sh"] = is_swing_high(df["high"], SWING_K).fillna(False)
    df["sl"] = is_swing_low(df["low"],  SWING_K).fillna(False)
    df["rsi_sh"] = is_swing_high(df["rsi"], SWING_K).fillna(False)
    df["rsi_sl"] = is_swing_low(df["rsi"],  SWING_K).fillna(False)

    # last swing references
    df["last_price_sh"] = np.where(df["sh"], df["high"], np.nan)
    df["last_price_sl"] = np.where(df["sl"], df["low"],  np.nan)
    df["last_rsi_sh"]   = np.where(df["rsi_sh"], df["rsi"], np.nan)
    df["last_rsi_sl"]   = np.where(df["rsi_sl"], df["rsi"], np.nan)

    df["last_price_sh"] = df["last_price_sh"].ffill()
    df["last_price_sl"] = df["last_price_sl"].ffill()
    df["last_rsi_sh"]   = df["last_rsi_sh"].ffill()
    df["last_rsi_sl"]   = df["last_rsi_sl"].ffill()

    # Bearish divergence: price higher high, RSI lower high
    df["bear_div"] = (
        (df["sh"]) &
        (df["high"] > df["last_price_sh"].shift(1)) &
        (df["rsi_sh"]) &
        (df["rsi"] < df["last_rsi_sh"].shift(1))
    )

    # Bullish divergence: price lower low, RSI higher low
    df["bull_div"] = (
        (df["sl"]) &
        (df["low"] < df["last_price_sl"].shift(1)) &
        (df["rsi_sl"]) &
        (df["rsi"] > df["last_rsi_sl"].shift(1))
    )

    return df


def daily_levels(day_df: pd.DataFrame, prev_day_df: Optional[pd.DataFrame]) -> List[float]:
    levels: List[float] = []
    if prev_day_df is not None and not prev_day_df.empty:
        levels += [prev_day_df["high"].max(), prev_day_df["low"].min()]
    if USE_DAILY_OPEN_AS_LEVEL and not day_df.empty:
        levels.append(day_df.iloc[0]["open"])  # session open
    # unique & sorted
    return sorted(list({round(lv, 4) for lv in levels}))

# =============================
# Backtest
# =============================

def _finalize_trade(symbol: str, pos: Dict, exit_time: pd.Timestamp, exit_price: float,
                    reason: str, r_mult: float, pnl_points: float) -> Trade:
    ticks = int(round(pnl_points / TICK_SIZE))
    pnl_usd = pnl_points * POINT_VALUE
    return Trade(
        symbol=symbol,
        entry_time=pos["entry_time"], side=pos["side"],
        entry=pos["entry"], stop=pos["stop"], target=pos["target"],
        exit_time=exit_time, exit_price=exit_price,
        reason=reason, r_mult=r_mult,
        points=pnl_points, ticks=ticks, pnl_usd=pnl_usd,
        day=str(pos["entry_time"].date())
    )


def simulate_intraday(symbol: str, df: pd.DataFrame) -> List[Trade]:
    trades: List[Trade] = []

    # features
    df = build_signals(df)

    for day, g in df.groupby("date_ny"):
        # regular session window
        day_slice = g[(g["ts_ny"].dt.strftime("%H:%M") >= REG_START) &
                      (g["ts_ny"].dt.strftime("%H:%M") <= REG_END)].copy()
        if day_slice.empty:
            continue

        prev_df = df[df["date_ny"] == (pd.to_datetime(day) - pd.Timedelta(days=1)).date()]
        levels = daily_levels(day_slice, prev_df)
        active_pos = None

        for i in range(1, len(day_slice)):
            row_prev = day_slice.iloc[i-1]
            row = day_slice.iloc[i]

            # manage open position
            if active_pos:
                if active_pos["side"] == "long":
                    if row["low"] <= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        r = (exit_px - active_pos["entry"]) / (active_pos["entry"] - active_pos["stop"])
                        pnl_points = exit_px - active_pos["entry"]
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "stop", r, pnl_points))
                        active_pos = None
                    elif row["high"] >= active_pos["target"]:
                        exit_px = active_pos["target"]
                        r = R_MULT_TARGET
                        pnl_points = exit_px - active_pos["entry"]
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "target", r, pnl_points))
                        active_pos = None
                else:  # short
                    if row["high"] >= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        r = (active_pos["entry"] - exit_px) / (active_pos["stop"] - active_pos["entry"])
                        pnl_points = active_pos["entry"] - exit_px
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "stop", r, pnl_points))
                        active_pos = None
                    elif row["low"] <= active_pos["target"]:
                        exit_px = active_pos["target"]
                        r = R_MULT_TARGET
                        pnl_points = active_pos["entry"] - exit_px
                        trades.append(_finalize_trade(symbol, active_pos, row["ts_ny"], exit_px, "target", r, pnl_points))
                        active_pos = None

            # flat: look for new setup
            if active_pos is None:
                atr_now = max(1e-6, row["atr"])
                near_level = True
                if USE_LEVELS and levels:
                    _, dist = nearest_level(row["close"], levels)
                    near_level = (dist <= LEVEL_PROX_THRESHOLD_ATR * atr_now)

                # Bearish divergence -> short
                if row_prev["bear_div"] and near_level:
                    signal_high = row_prev["high"]
                    if CONFIRM_ENTRY:
                        entry = min(row["open"], row["low"])  # need break below
                        entry = min(entry, signal_high - TICK_SIZE)
                    else:
                        entry = signal_high - TICK_SIZE
                    stop  = entry + ATR_MULT_STOP * atr_now
                    target = entry - R_MULT_TARGET * (stop - entry)
                    active_pos = {"side": "short", "entry_time": row["ts_ny"], "entry": float(entry),
                                  "stop": float(stop), "target": float(target), "reason": "bear_div"}

                # Bullish divergence -> long
                elif row_prev["bull_div"] and near_level:
                    signal_low = row_prev["low"]
                    if CONFIRM_ENTRY:
                        entry = max(row["open"], row["high"])  # need break above
                        entry = max(entry, signal_low + TICK_SIZE)
                    else:
                        entry = signal_low + TICK_SIZE
                    stop  = entry - ATR_MULT_STOP * atr_now
                    target = entry + R_MULT_TARGET * (entry - stop)
                    active_pos = {"side": "long", "entry_time": row["ts_ny"], "entry": float(entry),
                                  "stop": float(stop), "target": float(target), "reason": "bull_div"}

        # close any open position at session end
        if active_pos is not None:
            last = day_slice.iloc[-1]
            exit_px = last["close"]
            if active_pos["side"] == "long":
                pnl_points = exit_px - active_pos["entry"]
                r = pnl_points / (active_pos["entry"] - active_pos["stop"])
            else:
                pnl_points = active_pos["entry"] - exit_px
                r = pnl_points / (active_pos["stop"] - active_pos["entry"])
            trades.append(_finalize_trade(symbol, active_pos, last["ts_ny"], float(exit_px), "eod", float(r), float(pnl_points)))
            active_pos = None

    return trades

# =============================
# Run
# =============================

def main():
    raw = load_data(CONN_STR, TABLE, SYMBOLS, INTERVAL, START_DATE, END_DATE, SYMBOL_PREFIX)
    if raw.empty:
        print("No data returned. Check TABLE/SYMBOLS/DATE/INTERVAL.")
        return

    results: List[Trade] = []
    for sym, sdf in raw.groupby("symbol"):
        trades = simulate_intraday(sym, sdf.reset_index(drop=True))
        results.extend(trades)

    if not results:
        print("No trades generated with current params.")
        return

    tl = pd.DataFrame([t.__dict__ for t in results]).sort_values(["day", "entry_time"])  # trade log
    tl["cum_pnl_usd"] = tl["pnl_usd"].cumsum()

    # Summary
    wins = (tl["pnl_usd"] > 0).sum()
    losses = (tl["pnl_usd"] <= 0).sum()
    total = len(tl)
    winrate = wins / total if total else 0.0

    print("\n=== TRADE SUMMARY ===")
    print(f"Trades: {total} | Wins: {wins} | Losses: {losses} | Winrate: {winrate:.1%}")
    print(f"Total PnL: ${tl['pnl_usd'].sum():.2f} | Max DD: ${float((tl['cum_pnl_usd'].cummax()-tl['cum_pnl_usd']).max()):.2f}")

    out_csv = SAVE_DIR / "divergence_levels_backtest_trades.csv"
    tl.to_csv(out_csv, index=False)
    print(f"\nSaved trade log → {out_csv.resolve()}")


if __name__ == "__main__":
    main()
```

```python
# === imports ===
import math
from dataclasses import dataclass
from pathlib import Path
from typing import List, Optional, Tuple, Dict

import numpy as np
import pandas as pd
import pytz
from sqlalchemy import create_engine, text

# =============================
# DB & RUN SETTINGS (EDIT ME)
# =============================

# Use psycopg2 + AUTOCOMMIT for QuestDB PG wire compatibility.
ENGINE_URL = "postgresql+psycopg2://admin:quest@24.128.195.68:8812/qdb"

TABLE      = "futures_candles"   # your table
INTERVAL   = "1m"
SYMBOLS    = ["NQ"]              # set to your symbols present in `symbol`
SYMBOL_PREFIX: Optional[str] = None  # e.g., "NQ" to match NQ*; or None

START_DATE: Optional[str] = None  # e.g., "2024-10-01" (UTC)
END_DATE:   Optional[str] = None  # e.g., "2025-03-01" (UTC)

SAVE_DIR = Path("."); SAVE_DIR.mkdir(parents=True, exist_ok=True)

# =============================
# MARKET & STRATEGY PARAMETERS
# =============================

TZ_LOCAL = "America/New_York"
REG_START = "09:30"
REG_END   = "16:00"

RSI_LEN = 14
SWING_K = 5
ATR_LEN = 14

ATR_MULT_STOP = 1.0
R_MULT_TARGET = 1.5
CONFIRM_ENTRY = True

USE_LEVELS               = True
LEVEL_PROX_THRESHOLD_ATR = 0.5
USE_DAILY_OPEN_AS_LEVEL  = True

# Instrument specs (E-mini NQ defaults)
POINT_VALUE = 20.0     # $/point (MNQ = 2.0)
TICK_SIZE   = 0.25
TICK_VALUE  = POINT_VALUE * TICK_SIZE

# =============================
# Helpers
# =============================

def to_ny(ts: pd.Series) -> pd.Series:
    utc = pd.to_datetime(ts, utc=True)  # treat DB timestamps as UTC
    return utc.dt.tz_convert(TZ_LOCAL)

def rsi(close: pd.Series, length: int = 14) -> pd.Series:
    delta = close.diff()
    up = delta.clip(lower=0)
    down = -delta.clip(upper=0)
    roll_up = up.ewm(alpha=1/length, adjust=False).mean()
    roll_down = down.ewm(alpha=1/length, adjust=False).mean()
    rs = roll_up / roll_down.replace(0, np.nan)
    return 100 - (100 / (1 + rs))

def atr(df: pd.DataFrame, length: int = 14) -> pd.Series:
    high, low, close = df["high"], df["low"], df["close"]
    prev_close = close.shift(1)
    tr = pd.concat([
        (high - low),
        (high - prev_close).abs(),
        (low - prev_close).abs()
    ], axis=1).max(axis=1)
    return tr.ewm(alpha=1/length, adjust=False).mean()

def is_swing_high(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).max())

def is_swing_low(series: pd.Series, k: int) -> pd.Series:
    return (series == series.rolling(2*k+1, center=True).min())

def nearest_level(price: float, levels: List[float]) -> Tuple[float, float]:
    if not levels:
        return (np.nan, np.inf)
    d = [abs(price - lv) for lv in levels]
    i = int(np.argmin(d))
    return (levels[i], d[i])

@dataclass
class Trade:
    symbol: str
    entry_time: pd.Timestamp
    side: str
    entry: float
    stop: float
    target: float
    exit_time: pd.Timestamp
    exit_price: float
    reason: str
    r_mult: float
    points: float
    ticks: int
    pnl_usd: float
    day: str

# =============================
# Data (READ-ONLY via SQLAlchemy + psycopg2 AUTOCOMMIT)
# =============================

def load_data(engine_url: str,
              table: str,
              symbols: List[str],
              interval: str,
              start: Optional[str],
              end: Optional[str],
              symbol_prefix: Optional[str] = None) -> pd.DataFrame:

    where = ["interval = :interval"]
    params = {"interval": interval}

    if symbols and not symbol_prefix:
        # PostgreSQL ANY() binding
        where.append("symbol = ANY(:symbols)")
        params["symbols"] = symbols

    if symbol_prefix:
        where.append("symbol LIKE :prefix")
        params["prefix"] = f"{symbol_prefix}%"

    if start:
        where.append("event_at >= TIMESTAMP :start")
        params["start"] = start
    if end:
        where.append("event_at < TIMESTAMP :end")
        params["end"] = end

    sql = text(f"""
        SELECT symbol, interval, event_at, open, high, low, close, volume
        FROM {table}
        WHERE {' AND '.join(where)}
        ORDER BY symbol, event_at
    """)

    engine = create_engine(
        engine_url,
        isolation_level="AUTOCOMMIT",  # <-- key for QuestDB PG wire
        pool_pre_ping=True,
        future=True,
    )
    with engine.connect() as conn:
        # AUTOCOMMIT prevents SAVEPOINT usage on init/rollback
        df = pd.read_sql_query(sql, conn, params=params)

    if df.empty:
        return df

    df["ts_ny"] = to_ny(df["event_at"])
    df["date_ny"] = df["ts_ny"].dt.date
    return df

# =============================
# Signals (Divergence + Levels)
# =============================

def build_signals(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["rsi"] = rsi(df["close"], RSI_LEN)
    df["atr"] = atr(df, ATR_LEN)

    df["sh"] = is_swing_high(df["high"], SWING_K).fillna(False)
    df["sl"] = is_swing_low(df["low"],  SWING_K).fillna(False)
    df["rsi_sh"] = is_swing_high(df["rsi"], SWING_K).fillna(False)
    df["rsi_sl"] = is_swing_low(df["rsi"],  SWING_K).fillna(False)

    # previous swing values for divergence comparison
    df["last_price_sh"] = np.where(df["sh"], df["high"], np.nan).ffill()
    df["last_price_sl"] = np.where(df["sl"], df["low"],  np.nan).ffill()
    df["last_rsi_sh"]   = np.where(df["rsi_sh"], df["rsi"], np.nan).ffill()
    df["last_rsi_sl"]   = np.where(df["rsi_sl"], df["rsi"], np.nan).ffill()

    # Bearish divergence: higher high in price, lower high in RSI
    df["bear_div"] = (
        df["sh"] & df["rsi_sh"] &
        (df["high"] > df["last_price_sh"].shift(1)) &
        (df["rsi"]  < df["last_rsi_sh"].shift(1))
    )
    # Bullish divergence: lower low in price, higher low in RSI
    df["bull_div"] = (
        df["sl"] & df["rsi_sl"] &
        (df["low"] < df["last_price_sl"].shift(1)) &
        (df["rsi"] > df["last_rsi_sl"].shift(1))
    )
    return df

def daily_levels(day_df: pd.DataFrame, prev_day_df: Optional[pd.DataFrame]) -> List[float]:
    levels = []
    if prev_day_df is not None and not prev_day_df.empty:
        levels += [prev_day_df["high"].max(), prev_day_df["low"].min()]
    if USE_DAILY_OPEN_AS_LEVEL and not day_df.empty:
        levels.append(float(day_df.iloc[0]["open"]))  # session open
    return sorted(list({round(lv, 4) for lv in levels}))

# =============================
# Backtest
# =============================

def simulate_intraday(symbol: str, df: pd.DataFrame) -> List[Trade]:
    trades: List[Trade] = []
    df = build_signals(df)

    for day, g in df.groupby("date_ny"):
        day_slice = g[(g["ts_ny"].dt.strftime("%H:%M") >= REG_START) &
                      (g["ts_ny"].dt.strftime("%H:%M") <= REG_END)].copy()
        if day_slice.empty:
            continue

        prev_df = df[df["date_ny"] == (pd.to_datetime(day) - pd.Timedelta(days=1)).date()]
        levels = daily_levels(day_slice, prev_df)
        active_pos = None

        for i in range(1, len(day_slice)):
            prev = day_slice.iloc[i-1]
            row  = day_slice.iloc[i]

            # manage open position
            if active_pos:
                if active_pos["side"] == "long":
                    if row["low"] <= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        pnl_points = exit_px - active_pos["entry"]
                        r = pnl_points / (active_pos["entry"] - active_pos["stop"])
                        trades.append(_finalize(symbol, active_pos, row["ts_ny"], exit_px, "stop", r, pnl_points))
                        active_pos = None
                    elif row["high"] >= active_pos["target"]:
                        exit_px = active_pos["target"]
                        pnl_points = exit_px - active_pos["entry"]
                        r = R_MULT_TARGET
                        trades.append(_finalize(symbol, active_pos, row["ts_ny"], exit_px, "target", r, pnl_points))
                        active_pos = None
                else:  # short
                    if row["high"] >= active_pos["stop"]:
                        exit_px = active_pos["stop"]
                        pnl_points = active_pos["entry"] - exit_px
                        r = pnl_points / (active_pos["stop"] - active_pos["entry"])
                        trades.append(_finalize(symbol, active_pos, row["ts_ny"], exit_px, "stop", r, pnl_points))
                        active_pos = None
                    elif row["low"] <= active_pos["target"]:
                        exit_px = active_pos["target"]
                        pnl_points = active_pos["entry"] - exit_px
                        r = R_MULT_TARGET
                        trades.append(_finalize(symbol, active_pos, row["ts_ny"], exit_px, "target", r, pnl_points))
                        active_pos = None

            # if flat, look for signals
            if active_pos is None:
                atr_now = max(1e-6, row["atr"])
                near_level = True
                if USE_LEVELS and levels:
                    _, dist = nearest_level(row["close"], levels)
                    near_level = (dist <= LEVEL_PROX_THRESHOLD_ATR * atr_now)

                # bearish divergence → short
                if prev["bear_div"] and near_level:
                    signal_hi = prev["high"]
                    entry = min(row["open"], row["low"]) if CONFIRM_ENTRY else row["low"]
                    entry = min(entry, signal_hi - TICK_SIZE)  # below signal bar
                    stop  = entry + ATR_MULT_STOP * atr_now
                    target = entry - R_MULT_TARGET * (stop - entry)
                    active_pos = {"side": "short", "entry_time": row["ts_ny"], "entry": entry,
                                  "stop": stop, "target": target, "reason": "bear_div"}

                # bullish divergence → long
                elif prev["bull_div"] and near_level:
                    signal_lo = prev["low"]
                    entry = max(row["open"], row["high"]) if CONFIRM_ENTRY else row["high"]
                    entry = max(entry, signal_lo + TICK_SIZE)  # above signal bar
                    stop  = entry - ATR_MULT_STOP * atr_now
                    target = entry + R_MULT_TARGET * (entry - stop)
                    active_pos = {"side": "long", "entry_time": row["ts_ny"], "entry": entry,
                                  "stop": stop, "target": target, "reason": "bull_div"}

        # flatten at session end
        if active_pos is not None:
            last = day_slice.iloc[-1]
            exit_px = last["close"]
            if active_pos["side"] == "long":
                pnl_points = exit_px - active_pos["entry"]
                r = pnl_points / (active_pos["entry"] - active_pos["stop"])
            else:
                pnl_points = active_pos["entry"] - exit_px
                r = pnl_points / (active_pos["stop"] - active_pos["entry"])
            trades.append(_finalize(symbol, active_pos, last["ts_ny"], exit_px, "eod", r, pnl_points))
            active_pos = None

    return trades

def _finalize(symbol: str, pos: Dict, exit_time: pd.Timestamp, exit_price: float,
              reason: str, r_mult: float, pnl_points: float) -> Trade:
    ticks = int(round(pnl_points / TICK_SIZE))
    pnl_usd = pnl_points * POINT_VALUE
    return Trade(
        symbol=symbol,
        entry_time=pos["entry_time"], side=pos["side"],
        entry=pos["entry"], stop=pos["stop"], target=pos["target"],
        exit_time=exit_time, exit_price=exit_price,
        reason=reason, r_mult=r_mult,
        points=pnl_points, ticks=ticks, pnl_usd=pnl_usd,
        day=str(pos["entry_time"].date())
    )

# =============================
# Run
# =============================

def main():
    raw = load_data(ENGINE_URL, TABLE, SYMBOLS, INTERVAL, START_DATE, END_DATE, SYMBOL_PREFIX)
    if raw.empty:
        print("No data returned. Check TABLE/SYMBOLS/DATE/INTERVAL.")
        return

    results: List[Trade] = []
    for sym, sdf in raw.groupby("symbol"):
        trades = simulate_intraday(sym, sdf.reset_index(drop=True))
        results.extend(trades)

    if not results:
        print("No trades generated with current params.")
        return

    tl = pd.DataFrame([t.__dict__ for t in results]).sort_values(["day", "entry_time"])
    tl["cum_pnl_usd"] = tl["pnl_usd"].cumsum()

    wins = (tl["pnl_usd"] > 0).sum()
    losses = (tl["pnl_usd"] <= 0).sum()
    total = len(tl)
    winrate = wins / total if total else 0.0

    print("\n=== TRADE SUMMARY ===")
    print(f"Trades: {total} | Wins: {wins} | Losses: {losses} | Winrate: {winrate:.1%}")
    print(f"Total PnL: ${tl['pnl_usd'].sum():.2f} | Max DD: ${float((tl['cum_pnl_usd'].cummax()-tl['cum_pnl_usd']).max()):.2f}")

    out_csv = SAVE_DIR / "divergence_levels_backtest_trades.csv"
    tl.to_csv(out_csv, index=False)
    print(f"\nSaved trade log → {out_csv.resolve()}")

if __name__ == "__main__":
    main()
```
